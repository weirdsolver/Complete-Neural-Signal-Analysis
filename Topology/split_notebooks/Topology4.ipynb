```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patheffects as pe
from matplotlib.colors import LinearSegmentedColormap
from matplotlib.gridspec import GridSpec
from scipy.spatial.distance import pdist, squareform
from scipy.cluster.hierarchy import linkage, leaves_list

try:
    from scipy.linalg import eigh
    HAVE_SCIPY_EIGH = True
except Exception:
    HAVE_SCIPY_EIGH = False

try:
    from IPython.display import display, Image
    HAVE_IPYTHON = True
except Exception:
    HAVE_IPYTHON = False


base_dir = "/home/a/projects/Complete-Neural-Signal-Analysis"

FAST_OPTION = "ultra"
IAAFT_SURROGATE_TYPE = "independent_coordinate_iaaft"

SAVE_PLOTS_TO_DISK = True
RENDER_ALL_IMAGES_INLINE = True
INLINE_IMAGE_WIDTH = 1050
PLOT_DPI = 150

BLACK = "#000000"
ACCENT = "#00FFFF"
ACCENT_SOFT = "#66FFFF"
ACCENT_DIM = "#008B8B"

CYAN_FILL = "#00D5FF"
CYAN_FILL_SOFT = "#38BDF8"
CYAN_FILL_DARK = "#005A63"

CYAN_SEQUENTIAL_CMAP = LinearSegmentedColormap.from_list(
    "cyan_sequential",
    [BLACK, CYAN_FILL_DARK, CYAN_FILL, ACCENT_SOFT],
    N=256,
)

CYAN_DIVERGING_CMAP = LinearSegmentedColormap.from_list(
    "cyan_diverging",
    [ACCENT_SOFT, BLACK, CYAN_FILL],
    N=256,
)

plt.rcParams.update({
    "figure.facecolor": BLACK,
    "axes.facecolor": BLACK,
    "savefig.facecolor": BLACK,
    "text.color": ACCENT,
    "axes.labelcolor": ACCENT,
    "axes.edgecolor": ACCENT,
    "axes.titlecolor": ACCENT,
    "xtick.color": ACCENT,
    "ytick.color": ACCENT,
    "grid.color": ACCENT_DIM,
    "grid.alpha": 0.25,
    "legend.facecolor": BLACK,
    "legend.edgecolor": ACCENT,
    "legend.labelcolor": ACCENT,
    "font.size": 11,
})


def log(msg):
    print(msg, flush=True)


def style_ax(ax, title=None, xlabel=None, ylabel=None, grid=True):
    ax.set_facecolor(BLACK)

    for spine in ax.spines.values():
        spine.set_color(ACCENT)
        spine.set_linewidth(1.1)

    ax.tick_params(axis="x", colors=ACCENT, labelcolor=ACCENT)
    ax.tick_params(axis="y", colors=ACCENT, labelcolor=ACCENT)

    if title is not None:
        ax.set_title(title, color=ACCENT, pad=12)

    if xlabel is not None:
        ax.set_xlabel(xlabel, color=ACCENT, labelpad=10)

    if ylabel is not None:
        ax.set_ylabel(ylabel, color=ACCENT, labelpad=10)

    if grid:
        ax.grid(True, color=ACCENT_DIM, alpha=0.25, linewidth=0.8)


def force_cyan_text(ax):
    ax.title.set_color(ACCENT)
    ax.xaxis.label.set_color(ACCENT)
    ax.yaxis.label.set_color(ACCENT)

    for tick_label in ax.get_xticklabels():
        tick_label.set_color(ACCENT)

    for tick_label in ax.get_yticklabels():
        tick_label.set_color(ACCENT)


def style_colorbar(cbar, label):
    cbar.set_label(label, color=ACCENT)
    cbar.ax.yaxis.set_tick_params(color=ACCENT)
    plt.setp(cbar.ax.get_yticklabels(), color=ACCENT)
    cbar.outline.set_edgecolor(ACCENT)


def style_legend(legend):
    if legend is not None:
        legend.get_frame().set_facecolor(BLACK)
        legend.get_frame().set_edgecolor(ACCENT)

        for text in legend.get_texts():
            text.set_color(ACCENT)


def save_render_close(fig, path, width=INLINE_IMAGE_WIDTH):
    if SAVE_PLOTS_TO_DISK:
        fig.savefig(path, dpi=PLOT_DPI, bbox_inches="tight", facecolor=BLACK)

    plt.close(fig)

    if RENDER_ALL_IMAGES_INLINE and HAVE_IPYTHON and os.path.exists(path):
        display(Image(filename=path, width=width))


def offdiag_values(M):
    M = np.asarray(M, dtype=float)

    if M.shape[0] <= 1:
        return np.asarray([], dtype=float)

    mask = ~np.eye(M.shape[0], dtype=bool)
    vals = M[mask]
    vals = vals[np.isfinite(vals)]

    return vals


def dynamic_limits(values, lower_pct=2, upper_pct=98, pad_frac=0.08, lower_bound=None, upper_bound=None):
    values = np.asarray(values, dtype=float)
    values = values[np.isfinite(values)]

    if len(values) == 0:
        vmin, vmax = 0.0, 1.0
    else:
        vmin = float(np.percentile(values, lower_pct))
        vmax = float(np.percentile(values, upper_pct))

        if not np.isfinite(vmin):
            vmin = float(np.nanmin(values))

        if not np.isfinite(vmax):
            vmax = float(np.nanmax(values))

        if vmax <= vmin:
            center = float(np.nanmean(values))
            eps = max(1e-6, abs(center) * 0.01, 1e-5)
            vmin = center - eps
            vmax = center + eps

    pad = (vmax - vmin) * pad_frac
    vmin -= pad
    vmax += pad

    if lower_bound is not None:
        vmin = max(lower_bound, vmin)

    if upper_bound is not None:
        vmax = min(upper_bound, vmax)

    if vmax <= vmin:
        vmax = vmin + 1e-6

    return vmin, vmax


def robust_limits(values, lower_pct=2, upper_pct=98, pad_frac=0.08, include_zero=True):
    values = np.asarray(values, dtype=float)
    values = values[np.isfinite(values)]

    if len(values) == 0:
        return 0.0, 1.0

    vmin = float(np.nanpercentile(values, lower_pct))
    vmax = float(np.nanpercentile(values, upper_pct))

    if include_zero:
        vmin = min(vmin, 0.0)
        vmax = max(vmax, 0.0)

    if vmax <= vmin:
        center = float(np.nanmean(values))
        eps = max(abs(center) * 0.05, 1e-6)
        vmin = center - eps
        vmax = center + eps

    pad = (vmax - vmin) * pad_frac

    return vmin - pad, vmax + pad


def percentile_score(values, higher_is_more=True):
    s = pd.Series(values, dtype=float)

    if not higher_is_more:
        s = -s

    if s.nunique(dropna=True) <= 1:
        return np.full(len(s), 50.0, dtype=float)

    return s.rank(method="average", pct=True).to_numpy(dtype=float) * 100.0


def bh_fdr(pvals):
    pvals = np.asarray(pvals, dtype=float)
    qvals = np.full_like(pvals, np.nan, dtype=float)

    valid = np.isfinite(pvals)
    p = pvals[valid]

    if len(p) == 0:
        return qvals

    order = np.argsort(p)
    ranked = p[order]
    m = len(ranked)

    raw_q = ranked * m / np.arange(1, m + 1)
    monotone_q = np.minimum.accumulate(raw_q[::-1])[::-1]
    monotone_q = np.clip(monotone_q, 0.0, 1.0)

    q_valid = np.empty_like(monotone_q)
    q_valid[order] = monotone_q

    qvals[valid] = q_valid

    return qvals


def classical_mds_from_distances(D, n_components=2):
    D = np.asarray(D, dtype=float)
    n = D.shape[0]

    if n < 2:
        return np.zeros((n, n_components), dtype=float)

    D2 = D ** 2
    J = np.eye(n) - np.ones((n, n)) / n
    B = -0.5 * J @ D2 @ J

    if HAVE_SCIPY_EIGH:
        vals, vecs = eigh(B)
    else:
        vals, vecs = np.linalg.eigh(B)

    idx = np.argsort(vals)[::-1]
    vals = vals[idx]
    vecs = vecs[:, idx]

    coords = np.zeros((n, n_components), dtype=float)

    for k in range(min(n_components, n)):
        if vals[k] > 0:
            coords[:, k] = vecs[:, k] * np.sqrt(vals[k])

    return coords


def gaussian_similarity_from_distance(D, sigma=None):
    D = np.asarray(D, dtype=float)
    off = offdiag_values(D)

    if sigma is None:
        sigma = float(np.nanmedian(off)) if len(off) else 1.0

    sigma = max(float(sigma), 1e-8)

    S = np.exp(-(D ** 2) / (2.0 * sigma ** 2))
    np.fill_diagonal(S, 1.0)

    return S, sigma


def signed_log_compress(values):
    values = np.asarray(values, dtype=float)
    return np.sign(values) * np.log1p(np.abs(values))


def find_iaaft_output_dir():
    candidate_dirs = [
        os.path.join(
            base_dir,
            "topology_iaaft_surrogate_persistent_homology",
            f"h1_iaaft_surrogate_ph_{FAST_OPTION}_fps_{IAAFT_SURROGATE_TYPE}_joblib_fft",
        ),
        os.path.join(
            base_dir,
            "topology_iaaft_surrogate_persistent_homology",
            f"h1_iaaft_surrogate_ph_{FAST_OPTION}_fps_{IAAFT_SURROGATE_TYPE}",
        ),
    ]

    for candidate_dir in candidate_dirs:
        summary_path = os.path.join(candidate_dir, "h1_iaaft_surrogate_ph_summary.csv")

        if os.path.exists(summary_path):
            return candidate_dir

    raise FileNotFoundError(
        "Could not find IAAFT surrogate PH output directory. Expected summary CSV in:\n\n"
        + "\n".join(
            os.path.join(d, "h1_iaaft_surrogate_ph_summary.csv")
            for d in candidate_dirs
        )
    )


def compute_regularized_z(summary_df, long_df, primary_metric_names, prefix):
    fixed_z = pd.DataFrame(index=summary_df["channel"].astype(str))
    info_rows = []

    for metric in primary_metric_names:
        obs_col = f"observed_{metric}"
        mean_col = f"{prefix}_mean_{metric}"
        std_col = f"{prefix}_std_{metric}"
        raw_z_col = f"{prefix}_z_{metric}"

        obs = summary_df[obs_col].to_numpy(dtype=float)
        null_mean = summary_df[mean_col].to_numpy(dtype=float)

        if std_col in summary_df.columns:
            null_std = summary_df[std_col].to_numpy(dtype=float)
        else:
            null_std = np.zeros(len(summary_df), dtype=float)

        delta = obs - null_mean

        finite_std = null_std[np.isfinite(null_std) & (null_std > 0)]
        finite_delta = np.abs(delta[np.isfinite(delta)])

        null_vals = np.asarray([], dtype=float)

        if len(long_df) > 0 and metric in long_df.columns:
            null_vals = long_df[metric].to_numpy(dtype=float)
            null_vals = null_vals[np.isfinite(null_vals)]

        floors = []

        if len(finite_std) > 0:
            floors.append(float(np.nanmedian(finite_std)) * 0.10)
            floors.append(float(np.nanpercentile(finite_std, 25)) * 0.25)

        if len(finite_delta) > 0:
            floors.append(float(np.nanmedian(finite_delta)) * 0.05)
            floors.append(float(np.nanpercentile(finite_delta, 75)) * 0.025)

        if len(null_vals) > 1:
            floors.append(float(np.nanstd(null_vals)) * 0.03)
            floors.append(float(np.nanmedian(np.abs(null_vals - np.nanmedian(null_vals)))) * 0.10)

        if np.any(np.isfinite(obs)):
            floors.append(float(np.nanmedian(np.abs(obs[np.isfinite(obs)]))) * 1e-6)

        if np.any(np.isfinite(null_mean)):
            floors.append(float(np.nanmedian(np.abs(null_mean[np.isfinite(null_mean)]))) * 1e-6)

        floors.append(1e-6)
        floors = [v for v in floors if np.isfinite(v) and v > 0]

        std_floor = max(floors) if len(floors) else 1e-6
        z_fixed = delta / np.maximum(null_std, std_floor)

        raw_exploded_count = 0

        if raw_z_col in summary_df.columns:
            raw_z = summary_df[raw_z_col].to_numpy(dtype=float)
            raw_exploded_count = int(np.sum(np.isfinite(raw_z) & (np.abs(raw_z) > 1000)))

        fixed_z[f"{prefix}_z_regularized_{metric}"] = z_fixed

        info_rows.append({
            "metric": metric,
            "std_floor_used": std_floor,
            "raw_exploded_z_count": raw_exploded_count,
            "regularized_z_min": float(np.nanmin(z_fixed)) if np.any(np.isfinite(z_fixed)) else np.nan,
            "regularized_z_max": float(np.nanmax(z_fixed)) if np.any(np.isfinite(z_fixed)) else np.nan,
        })

    return fixed_z, pd.DataFrame(info_rows)


mode_out_dir = find_iaaft_output_dir()

summary_csv_path = os.path.join(mode_out_dir, "h1_iaaft_surrogate_ph_summary.csv")
iaaft_long_csv_path = os.path.join(mode_out_dir, "h1_iaaft_surrogate_ph_iaaft_long_table.csv")
diagnostic_csv_path = os.path.join(mode_out_dir, "h1_iaaft_surrogate_ph_runtime_diagnostics.csv")
signature_csv_path = os.path.join(mode_out_dir, "h1_iaaft_surrogate_ph_regularized_z_signature_matrix.csv")
signature_distance_csv_path = os.path.join(mode_out_dir, "h1_iaaft_surrogate_ph_signature_distance_matrix.csv")
signature_similarity_csv_path = os.path.join(mode_out_dir, "h1_iaaft_surrogate_ph_signature_similarity_matrix.csv")
mds_csv_path = os.path.join(mode_out_dir, "h1_iaaft_surrogate_ph_mds_coordinates.csv")
region_summary_csv_path = os.path.join(mode_out_dir, "h1_iaaft_surrogate_ph_region_summary.csv")
regularized_z_info_csv_path = os.path.join(mode_out_dir, "h1_iaaft_surrogate_ph_regularized_z_info.csv")

summary_df = pd.read_csv(summary_csv_path)
summary_df["channel"] = summary_df["channel"].astype(str)

if os.path.exists(iaaft_long_csv_path):
    iaaft_long_df = pd.read_csv(iaaft_long_csv_path)
else:
    iaaft_long_df = pd.DataFrame()

if os.path.exists(diagnostic_csv_path):
    diagnostic_df = pd.read_csv(diagnostic_csv_path)
else:
    diagnostic_df = summary_df.copy()

primary_metric_names = [
    "total_lifetime",
    "max_lifetime",
    "q90_lifetime",
    "betti_auc",
    "betti_l2",
    "landscape_l2",
    "landscape_max",
]

primary_metric_labels = [
    "Total lifetime",
    "Max lifetime",
    "Q90 lifetime",
    "Betti AUC",
    "Betti L2",
    "Landscape L2",
    "Landscape max",
]

for metric in primary_metric_names:
    for col in [f"observed_{metric}", f"iaaft_mean_{metric}", f"iaaft_p_greater_{metric}"]:
        if col not in summary_df.columns:
            raise ValueError(f"Missing required column in summary CSV: {col}")

    q_col = f"iaaft_q_greater_{metric}"

    if q_col not in summary_df.columns:
        summary_df[q_col] = bh_fdr(summary_df[f"iaaft_p_greater_{metric}"].to_numpy(dtype=float))

    neg_col = f"iaaft_neglog10_q_{metric}"

    if neg_col not in summary_df.columns:
        summary_df[neg_col] = -np.log10(summary_df[q_col].to_numpy(dtype=float) + 1e-12)

if "region" not in summary_df.columns:
    channel_regions = {
        "Fp1": "Frontal", "Fpz": "Frontal", "Fp2": "Frontal",
        "F7": "Frontal", "F3": "Frontal", "Fz": "Frontal", "F4": "Frontal", "F8": "Frontal",
        "FC5": "Frontocentral", "FC1": "Frontocentral", "FC2": "Frontocentral", "FC6": "Frontocentral",
        "M1": "Mastoid", "M2": "Mastoid",
        "T7": "Central", "C3": "Central", "Cz": "Central", "C4": "Central", "T8": "Central",
        "CP5": "Centroparietal", "CP1": "Centroparietal", "CP2": "Centroparietal", "CP6": "Centroparietal",
        "P7": "Parietal", "P3": "Parietal", "Pz": "Parietal", "P4": "Parietal", "P8": "Parietal",
        "POz": "Occipital", "O1": "Occipital", "Oz": "Occipital", "O2": "Occipital",
    }
    summary_df["region"] = summary_df["channel"].map(lambda ch: channel_regions.get(str(ch), "Unknown"))

if "iaaft_corrected_ph_score" not in summary_df.columns:
    if "iaaft_corrected_mean_primary_z" in summary_df.columns:
        summary_df["iaaft_corrected_ph_score"] = percentile_score(summary_df["iaaft_corrected_mean_primary_z"], higher_is_more=True)
    else:
        raw_z_cols = [f"iaaft_z_{metric}" for metric in primary_metric_names if f"iaaft_z_{metric}" in summary_df.columns]
        summary_df["iaaft_corrected_mean_primary_z"] = summary_df[raw_z_cols].mean(axis=1)
        summary_df["iaaft_corrected_ph_score"] = percentile_score(summary_df["iaaft_corrected_mean_primary_z"], higher_is_more=True)

regularized_z_df, regularized_z_info_df = compute_regularized_z(
    summary_df,
    iaaft_long_df,
    primary_metric_names,
    prefix="iaaft",
)

for col in regularized_z_df.columns:
    summary_df[col] = summary_df["channel"].astype(str).map(regularized_z_df[col].to_dict()).to_numpy(dtype=float)

regularized_z_cols = [f"iaaft_z_regularized_{metric}" for metric in primary_metric_names]
q_cols = [f"iaaft_q_greater_{metric}" for metric in primary_metric_names]

summary_df["iaaft_corrected_mean_primary_z"] = summary_df[regularized_z_cols].mean(axis=1)
summary_df["iaaft_corrected_median_primary_z"] = summary_df[regularized_z_cols].median(axis=1)
summary_df["iaaft_corrected_ph_score_fixed"] = percentile_score(summary_df["iaaft_corrected_mean_primary_z"], higher_is_more=True)

if "iaaft_corrected_ph_score" not in summary_df.columns:
    summary_df["iaaft_corrected_ph_score"] = summary_df["iaaft_corrected_ph_score_fixed"]

corrected_signature_df = summary_df.set_index("channel")[regularized_z_cols].copy()
corrected_signature_df = corrected_signature_df.replace([np.inf, -np.inf], np.nan).fillna(0.0)

signature_channels = list(corrected_signature_df.index.astype(str))
signature_X = corrected_signature_df.to_numpy(dtype=float)

if len(signature_channels) >= 2:
    corrected_signature_distance = squareform(pdist(signature_X, metric="euclidean"))
else:
    corrected_signature_distance = np.zeros((len(signature_channels), len(signature_channels)), dtype=float)

corrected_signature_similarity, corrected_similarity_sigma = gaussian_similarity_from_distance(corrected_signature_distance)

if len(signature_channels) >= 3 and np.any(corrected_signature_distance > 0):
    Z_signature = linkage(squareform(corrected_signature_distance, checks=False), method="average", optimal_ordering=True)
    order = leaves_list(Z_signature)
else:
    order = np.arange(len(signature_channels))

ordered_channels = [signature_channels[i] for i in order]
ordered_signature_distance = corrected_signature_distance[np.ix_(order, order)]
ordered_signature_similarity = corrected_signature_similarity[np.ix_(order, order)]
ordered_signature_z = corrected_signature_df.loc[ordered_channels].to_numpy(dtype=float)

display_signature_z = signed_log_compress(ordered_signature_z)

if os.path.exists(mds_csv_path):
    mds_df = pd.read_csv(mds_csv_path)
    mds_df["channel"] = mds_df["channel"].astype(str)
else:
    mds_coords = classical_mds_from_distances(corrected_signature_distance, n_components=2)
    mds_df = pd.DataFrame({
        "channel": signature_channels,
        "region": [
            summary_df.set_index("channel").loc[ch, "region"]
            if ch in summary_df.set_index("channel").index
            else "Unknown"
            for ch in signature_channels
        ],
        "iaaft_corrected_mds_1": mds_coords[:, 0],
        "iaaft_corrected_mds_2": mds_coords[:, 1],
    })

score_map = summary_df.set_index("channel")["iaaft_corrected_ph_score"].to_dict()
mds_df["iaaft_corrected_ph_score"] = [score_map.get(ch, np.nan) for ch in mds_df["channel"]]

region_summary_df = (
    summary_df
    .groupby("region")
    .agg(
        n_channels=("channel", "count"),
        mean_iaaft_corrected_ph_score=("iaaft_corrected_ph_score", "mean"),
        sem_iaaft_corrected_ph_score=("iaaft_corrected_ph_score", lambda x: float(np.std(x, ddof=1) / np.sqrt(len(x))) if len(x) > 1 else 0.0),
        mean_primary_z=("iaaft_corrected_mean_primary_z", "mean"),
    )
    .reset_index()
    .sort_values("mean_iaaft_corrected_ph_score", ascending=False)
)

summary_fixed_path = os.path.join(mode_out_dir, "h1_iaaft_surrogate_ph_summary_fixed_plot_inputs.csv")
signature_fixed_path = os.path.join(mode_out_dir, "h1_iaaft_surrogate_ph_regularized_z_signature_matrix_fixed_plot_inputs.csv")
regularized_z_info_csv_path = os.path.join(mode_out_dir, "h1_iaaft_surrogate_ph_regularized_z_info_fixed_plot_inputs.csv")

summary_df.to_csv(summary_fixed_path, index=False)
corrected_signature_df.to_csv(signature_fixed_path)
regularized_z_info_df.to_csv(regularized_z_info_csv_path, index=False)

print("\nIAAFT PH summary used for regenerated plots:")
if HAVE_IPYTHON:
    display(summary_df)
else:
    print(summary_df)

print("\nRegularized z info:")
if HAVE_IPYTHON:
    display(regularized_z_info_df)
else:
    print(regularized_z_info_df)

log("\n" + "=" * 72)
log("Regenerating ALL IAAFT plots.")
log("Only these are changed/fixed: 02, 03, and 07.")
log(f"Output directory: {mode_out_dir}")
log("=" * 72)

fig, ax = plt.subplots(figsize=(13, 6.2), facecolor=BLACK)
style_ax(
    ax,
    title=f"H1 IAAFT-Surrogate Persistent Homology Score | {FAST_OPTION.upper()}",
    xlabel="Channel",
    ylabel="IAAFT-corrected PH score",
)

rank_df = summary_df.sort_values("iaaft_corrected_ph_score", ascending=False).reset_index(drop=True)
x = np.arange(len(rank_df))

ax.bar(
    x,
    rank_df["iaaft_corrected_ph_score"],
    color=CYAN_FILL,
    edgecolor=ACCENT,
    linewidth=0.9,
    alpha=0.82,
)

ax.axhline(
    float(np.nanmean(rank_df["iaaft_corrected_ph_score"])),
    color=ACCENT_SOFT,
    linestyle="--",
    linewidth=1.3,
    alpha=0.85,
    label=f"Mean={np.nanmean(rank_df['iaaft_corrected_ph_score']):.1f}",
)

for i, row in rank_df.iterrows():
    if i < 8:
        ax.text(
            i,
            row["iaaft_corrected_ph_score"],
            f"{row['iaaft_corrected_ph_score']:.0f}",
            ha="center",
            va="bottom",
            color=ACCENT,
            fontsize=8,
        )

ax.set_xticks(x)
ax.set_xticklabels(rank_df["channel"], rotation=90)
ax.set_ylim(0, 105)

legend = ax.legend(loc="best", framealpha=1.0)
style_legend(legend)
force_cyan_text(ax)

score_path = os.path.join(mode_out_dir, "01_iaaft_corrected_ph_score_ranking.png")
plt.tight_layout()
save_render_close(fig, score_path, width=1150)

fig, ax = plt.subplots(figsize=(12, 8.8), facecolor=BLACK)
style_ax(
    ax,
    title="IAAFT-Corrected H1 Primary Metric Z-Score Heatmap | fixed",
    xlabel="H1 metric",
    ylabel="Channel",
)

z_abs = np.abs(display_signature_z[np.isfinite(display_signature_z)])

if len(z_abs):
    z_vlim = max(float(np.nanpercentile(z_abs, 98)), 1.0)
else:
    z_vlim = 1.0

im = ax.imshow(
    display_signature_z,
    aspect="auto",
    interpolation="nearest",
    cmap=CYAN_DIVERGING_CMAP,
    vmin=-z_vlim,
    vmax=z_vlim,
)

ax.set_yticks(np.arange(len(ordered_channels)))
ax.set_yticklabels(ordered_channels)
ax.set_xticks(np.arange(len(primary_metric_labels)))
ax.set_xticklabels(primary_metric_labels, rotation=45, ha="right")

ax.set_xticks(np.arange(-0.5, len(primary_metric_labels), 1), minor=True)
ax.set_yticks(np.arange(-0.5, len(ordered_channels), 1), minor=True)
ax.grid(which="minor", color=ACCENT_DIM, linewidth=0.55, alpha=0.35)
ax.tick_params(which="minor", bottom=False, left=False)

raw_z_for_text = ordered_signature_z

for i in range(display_signature_z.shape[0]):
    for j in range(display_signature_z.shape[1]):
        val = raw_z_for_text[i, j]

        if np.isfinite(val):
            ax.text(
                j,
                i,
                f"{val:+.1f}",
                ha="center",
                va="center",
                color=BLACK,
                fontsize=5.6,
                alpha=0.72,
                fontweight="bold",
            )

cbar = plt.colorbar(im, ax=ax, shrink=0.85)
style_colorbar(cbar, "Signed log1p regularized z-score")
force_cyan_text(ax)

fig.text(
    0.09,
    0.035,
    "Fixed view: colors use signed log1p(z) so the heatmap does not saturate; cell labels show the regularized z-score.",
    ha="left",
    va="bottom",
    color=ACCENT,
    fontsize=9,
)

z_heatmap_path = os.path.join(mode_out_dir, "02_iaaft_corrected_primary_metric_z_heatmap.png")
z_heatmap_fixed_copy_path = os.path.join(mode_out_dir, "02_iaaft_corrected_primary_metric_z_heatmap_FIXED.png")
plt.tight_layout(rect=[0.0, 0.055, 1.0, 1.0])

if SAVE_PLOTS_TO_DISK:
    fig.savefig(z_heatmap_path, dpi=PLOT_DPI, bbox_inches="tight", facecolor=BLACK)
    fig.savefig(z_heatmap_fixed_copy_path, dpi=PLOT_DPI, bbox_inches="tight", facecolor=BLACK)

plt.close(fig)

if RENDER_ALL_IMAGES_INLINE and HAVE_IPYTHON and os.path.exists(z_heatmap_path):
    display(Image(filename=z_heatmap_path, width=1050))

order_df = summary_df.set_index("channel").loc[ordered_channels].reset_index()

z_matrix_raw = order_df[regularized_z_cols].to_numpy(dtype=float)
z_matrix_display = signed_log_compress(z_matrix_raw)
q_matrix = order_df[q_cols].to_numpy(dtype=float)

sig_005 = q_matrix < 0.05
sig_010 = (q_matrix >= 0.05) & (q_matrix < 0.10)

fig = plt.figure(figsize=(16.8, 10.8), facecolor=BLACK)

ax = fig.add_axes([0.07, 0.18, 0.58, 0.72])
rank_ax = fig.add_axes([0.75, 0.18, 0.20, 0.72])
cbar_ax = fig.add_axes([0.67, 0.18, 0.018, 0.72])

for this_ax in [ax, rank_ax]:
    this_ax.set_facecolor(BLACK)

    for spine in this_ax.spines.values():
        spine.set_color(ACCENT)
        spine.set_linewidth(1.1)

    this_ax.tick_params(axis="x", colors=ACCENT, labelcolor=ACCENT)
    this_ax.tick_params(axis="y", colors=ACCENT, labelcolor=ACCENT)

z_abs = np.abs(z_matrix_display[np.isfinite(z_matrix_display)])
fixed_z_vlim = max(float(np.nanpercentile(z_abs, 98)), 1.0) if len(z_abs) else 1.0

im = ax.imshow(
    z_matrix_display,
    aspect="auto",
    interpolation="nearest",
    cmap=CYAN_DIVERGING_CMAP,
    vmin=-fixed_z_vlim,
    vmax=fixed_z_vlim,
)

ax.set_title(
    "IAAFT-Corrected H1 Significance + Effect Direction | fixed\n"
    "Color = signed log1p regularized z-score, markers = FDR significance",
    color=ACCENT,
    pad=14,
)

ax.set_xlabel("H1 metric", color=ACCENT, labelpad=12)
ax.set_ylabel("Channel", color=ACCENT, labelpad=12)

ax.set_xticks(np.arange(len(primary_metric_labels)))
ax.set_xticklabels(primary_metric_labels, rotation=45, ha="right", color=ACCENT)
ax.set_yticks(np.arange(len(ordered_channels)))
ax.set_yticklabels(ordered_channels, color=ACCENT, fontsize=8)

ax.set_xticks(np.arange(-0.5, len(primary_metric_labels), 1), minor=True)
ax.set_yticks(np.arange(-0.5, len(ordered_channels), 1), minor=True)
ax.grid(which="minor", color=ACCENT_DIM, linewidth=0.55, alpha=0.35)
ax.tick_params(which="minor", bottom=False, left=False)

for i in range(z_matrix_display.shape[0]):
    for j in range(z_matrix_display.shape[1]):
        z_val = z_matrix_raw[i, j]

        if not np.isfinite(z_val):
            continue

        if sig_005[i, j]:
            marker = "●"
            marker_size = 10
            marker_color = ACCENT_SOFT
        elif sig_010[i, j]:
            marker = "○"
            marker_size = 9
            marker_color = ACCENT_SOFT
        else:
            marker = ""

        if marker:
            ax.text(
                j,
                i,
                marker,
                ha="center",
                va="center",
                color=marker_color,
                fontsize=marker_size,
                fontweight="bold",
            )

        if abs(z_val) >= 1.0:
            ax.text(
                j,
                i + 0.28,
                f"{z_val:+.1f}",
                ha="center",
                va="center",
                color=BLACK,
                fontsize=5.4,
                alpha=0.70,
                fontweight="bold",
            )

cbar = plt.colorbar(im, cax=cbar_ax)
cbar.set_label("Signed log1p regularized z-score", color=ACCENT)
cbar.ax.yaxis.set_tick_params(color=ACCENT)
plt.setp(cbar.ax.get_yticklabels(), color=ACCENT)
cbar.outline.set_edgecolor(ACCENT)

score_values = order_df["iaaft_corrected_ph_score"].to_numpy(dtype=float)
y = np.arange(len(ordered_channels))

rank_ax.barh(
    y,
    score_values,
    color=CYAN_FILL,
    edgecolor=ACCENT,
    linewidth=0.7,
    alpha=0.82,
)

rank_ax.set_title("IAAFT-corrected\nPH score", color=ACCENT, pad=14)
rank_ax.set_xlabel("Score", color=ACCENT, labelpad=10)
rank_ax.set_yticks(y)
rank_ax.set_yticklabels(ordered_channels, fontsize=7)
rank_ax.invert_yaxis()
rank_ax.set_xlim(0, 105)

for yy, val in zip(y, score_values):
    if np.isfinite(val) and (val >= 85 or val <= 15):
        rank_ax.text(
            val + 1.0,
            yy,
            f"{val:.0f}",
            va="center",
            ha="left",
            color=ACCENT,
            fontsize=6.8,
            alpha=0.9,
        )

rank_ax.grid(True, color=ACCENT_DIM, alpha=0.25, linewidth=0.8)

n_sig_005_by_metric = sig_005.sum(axis=0)
n_exploded = int(regularized_z_info_df["raw_exploded_z_count"].sum()) if "raw_exploded_z_count" in regularized_z_info_df.columns else 0

footer_lines = []
footer_lines.append("● = FDR q < 0.05    ○ = FDR q < 0.10")
footer_lines.append(
    "q<0.05 counts by metric: "
    + ", ".join([f"{label}: {count}" for label, count in zip(primary_metric_labels, n_sig_005_by_metric)])
)
footer_lines.append(
    f"Fixed view: colors are log-compressed to avoid saturation; raw regularized z-values remain printed in cells. "
    f"{n_exploded} raw z entries were variance-floor stabilized."
)

fig.text(
    0.07,
    0.055,
    "\n".join(footer_lines),
    ha="left",
    va="bottom",
    color=ACCENT,
    fontsize=9,
)

q_heatmap_path = os.path.join(mode_out_dir, "03_iaaft_corrected_primary_metric_q_heatmap.png")
q_heatmap_fixed_copy_path = os.path.join(mode_out_dir, "03_iaaft_corrected_primary_metric_significance_effect_heatmap_fixed.png")

if SAVE_PLOTS_TO_DISK:
    fig.savefig(q_heatmap_path, dpi=PLOT_DPI, bbox_inches="tight", facecolor=BLACK)
    fig.savefig(q_heatmap_fixed_copy_path, dpi=PLOT_DPI, bbox_inches="tight", facecolor=BLACK)

plt.close(fig)

if RENDER_ALL_IMAGES_INLINE and HAVE_IPYTHON and os.path.exists(q_heatmap_path):
    display(Image(filename=q_heatmap_path, width=1150))

sim_vmin, sim_vmax = dynamic_limits(
    offdiag_values(ordered_signature_similarity),
    lower_pct=2,
    upper_pct=98,
    pad_frac=0.10,
    lower_bound=0.0,
    upper_bound=1.0,
)

fig, ax = plt.subplots(figsize=(9.5, 8.5), facecolor=BLACK)
style_ax(
    ax,
    title="IAAFT-Corrected PH Signature Similarity Matrix",
    xlabel="Channel",
    ylabel="Channel",
)

im = ax.imshow(
    ordered_signature_similarity,
    aspect="equal",
    interpolation="nearest",
    cmap=CYAN_SEQUENTIAL_CMAP,
    vmin=sim_vmin,
    vmax=sim_vmax,
)

ax.set_xticks(np.arange(len(ordered_channels)))
ax.set_xticklabels(ordered_channels, rotation=90)
ax.set_yticks(np.arange(len(ordered_channels)))
ax.set_yticklabels(ordered_channels)

cbar = plt.colorbar(im, ax=ax, shrink=0.85)
style_colorbar(cbar, "IAAFT-corrected signature similarity")
force_cyan_text(ax)

signature_similarity_path = os.path.join(mode_out_dir, "04_iaaft_corrected_ph_signature_similarity_heatmap.png")
plt.tight_layout()
save_render_close(fig, signature_similarity_path, width=950)

dist_vmin, dist_vmax = dynamic_limits(
    offdiag_values(ordered_signature_distance),
    lower_pct=2,
    upper_pct=98,
    pad_frac=0.10,
    lower_bound=0.0,
)

fig, ax = plt.subplots(figsize=(9.5, 8.5), facecolor=BLACK)
style_ax(
    ax,
    title="IAAFT-Corrected PH Signature Distance Matrix",
    xlabel="Channel",
    ylabel="Channel",
)

im = ax.imshow(
    ordered_signature_distance,
    aspect="equal",
    interpolation="nearest",
    cmap=CYAN_SEQUENTIAL_CMAP,
    vmin=dist_vmin,
    vmax=dist_vmax,
)

ax.set_xticks(np.arange(len(ordered_channels)))
ax.set_xticklabels(ordered_channels, rotation=90)
ax.set_yticks(np.arange(len(ordered_channels)))
ax.set_yticklabels(ordered_channels)

cbar = plt.colorbar(im, ax=ax, shrink=0.85)
style_colorbar(cbar, "IAAFT-corrected signature distance")
force_cyan_text(ax)

signature_distance_path = os.path.join(mode_out_dir, "05_iaaft_corrected_ph_signature_distance_heatmap.png")
plt.tight_layout()
save_render_close(fig, signature_distance_path, width=950)

fig, ax = plt.subplots(figsize=(8.5, 7.2), facecolor=BLACK)
style_ax(
    ax,
    title="MDS of IAAFT-Corrected PH Signatures",
    xlabel="IAAFT-corrected PH MDS 1",
    ylabel="IAAFT-corrected PH MDS 2",
)

sc = ax.scatter(
    mds_df["iaaft_corrected_mds_1"],
    mds_df["iaaft_corrected_mds_2"],
    c=mds_df["iaaft_corrected_ph_score"],
    s=165,
    cmap=CYAN_SEQUENTIAL_CMAP,
    vmin=0,
    vmax=100,
    edgecolor=ACCENT,
    linewidth=1.2,
    alpha=0.95,
)

for _, row in mds_df.iterrows():
    txt = ax.text(
        row["iaaft_corrected_mds_1"],
        row["iaaft_corrected_mds_2"],
        row["channel"],
        fontsize=8,
        color=ACCENT,
        alpha=0.95,
        zorder=5,
    )
    txt.set_path_effects([pe.withStroke(linewidth=2.0, foreground=BLACK)])

cbar = plt.colorbar(sc, ax=ax, shrink=0.82)
style_colorbar(cbar, "IAAFT-corrected PH score")
force_cyan_text(ax)

mds_path = os.path.join(mode_out_dir, "06_mds_iaaft_corrected_ph_signatures.png")
plt.tight_layout()
save_render_close(fig, mds_path, width=930)

plot_metric = "total_lifetime"

plot_df = summary_df.copy()
plot_df["observed_total_lifetime"] = plot_df[f"observed_{plot_metric}"].to_numpy(dtype=float)
plot_df["iaaft_mean_total_lifetime"] = plot_df[f"iaaft_mean_{plot_metric}"].to_numpy(dtype=float)
plot_df["iaaft_std_total_lifetime"] = plot_df[f"iaaft_std_{plot_metric}"].to_numpy(dtype=float) if f"iaaft_std_{plot_metric}" in plot_df.columns else 0.0
plot_df["delta_total_lifetime"] = plot_df["observed_total_lifetime"] - plot_df["iaaft_mean_total_lifetime"]
plot_df["ratio_total_lifetime"] = plot_df["observed_total_lifetime"] / (plot_df["iaaft_mean_total_lifetime"] + 1e-12)
plot_df["q_total_lifetime"] = plot_df[f"iaaft_q_greater_{plot_metric}"].to_numpy(dtype=float)
plot_df["neglog10_q_total_lifetime"] = -np.log10(plot_df["q_total_lifetime"].to_numpy(dtype=float) + 1e-12)

plot_df = plot_df.replace([np.inf, -np.inf], np.nan)
plot_df = plot_df.sort_values("delta_total_lifetime", ascending=False).reset_index(drop=True)

fig = plt.figure(figsize=(15.8, 8.4), facecolor=BLACK)

paired_ax = fig.add_axes([0.07, 0.14, 0.42, 0.74])
rank_ax = fig.add_axes([0.59, 0.14, 0.34, 0.74])

paired_df = plot_df.sort_values("delta_total_lifetime", ascending=True).reset_index(drop=True)
y = np.arange(len(paired_df))

style_ax(
    paired_ax,
    title="Observed vs IAAFT Mean H1 Total Lifetime | fixed paired view",
    xlabel="H1 total lifetime",
    ylabel="Channel",
)

paired_ax.hlines(
    y,
    paired_df["iaaft_mean_total_lifetime"],
    paired_df["observed_total_lifetime"],
    color=ACCENT_DIM,
    linewidth=1.0,
    alpha=0.55,
)

paired_ax.scatter(
    paired_df["iaaft_mean_total_lifetime"],
    y,
    s=72,
    facecolors=BLACK,
    edgecolors=ACCENT_SOFT,
    linewidth=1.2,
    label="IAAFT mean",
    zorder=4,
)

paired_ax.scatter(
    paired_df["observed_total_lifetime"],
    y,
    s=82,
    color=CYAN_FILL,
    edgecolor=ACCENT,
    linewidth=1.0,
    label="Observed",
    zorder=5,
)

paired_ax.set_yticks(y)
paired_ax.set_yticklabels(paired_df["channel"], fontsize=7)

x_vals = np.concatenate([
    paired_df["iaaft_mean_total_lifetime"].to_numpy(dtype=float),
    paired_df["observed_total_lifetime"].to_numpy(dtype=float),
])
x_lo, x_hi = robust_limits(x_vals, lower_pct=1, upper_pct=99, pad_frac=0.12, include_zero=True)
paired_ax.set_xlim(x_lo, x_hi)

legend = paired_ax.legend(loc="lower right", framealpha=1.0, fontsize=8)
style_legend(legend)

style_ax(
    rank_ax,
    title="Observed − IAAFT Mean",
    xlabel="Δ H1 total lifetime",
    ylabel="Channel",
)

rank_ax.barh(
    y,
    paired_df["delta_total_lifetime"],
    color=CYAN_FILL,
    edgecolor=ACCENT,
    linewidth=0.75,
    alpha=0.82,
)

rank_ax.axvline(
    0,
    color=ACCENT_SOFT,
    linestyle="--",
    linewidth=1.2,
    alpha=0.9,
)

rank_ax.set_yticks(y)
rank_ax.set_yticklabels(paired_df["channel"], fontsize=7)

delta_lo, delta_hi = robust_limits(
    paired_df["delta_total_lifetime"].to_numpy(dtype=float),
    lower_pct=1,
    upper_pct=99,
    pad_frac=0.18,
    include_zero=True,
)
rank_ax.set_xlim(delta_lo, delta_hi)

for yy, (_, row) in zip(y, paired_df.iterrows()):
    delta = float(row["delta_total_lifetime"])
    q_val = float(row["q_total_lifetime"])

    if q_val < 0.05:
        marker = "●"
    elif q_val < 0.10:
        marker = "○"
    else:
        marker = ""

    if marker:
        x_text = delta + (0.015 * (delta_hi - delta_lo) if delta >= 0 else -0.015 * (delta_hi - delta_lo))
        ha = "left" if delta >= 0 else "right"

        rank_ax.text(
            x_text,
            yy,
            marker,
            va="center",
            ha=ha,
            color=ACCENT_SOFT,
            fontsize=9,
            fontweight="bold",
        )

force_cyan_text(paired_ax)
force_cyan_text(rank_ax)

footer = (
    "Fixed view: paired plot avoids the collapsed x≈0 scatter. "
    "Left compares each channel's IAAFT mean to observed H1 total lifetime; right shows signed effect. "
    "● = FDR q < 0.05, ○ = FDR q < 0.10."
)

fig.text(
    0.07,
    0.035,
    footer,
    ha="left",
    va="bottom",
    color=ACCENT,
    fontsize=9,
)

obs_null_path = os.path.join(mode_out_dir, "07_observed_vs_iaaft_mean_total_lifetime.png")
obs_null_copy_path = os.path.join(mode_out_dir, "07_observed_vs_iaaft_mean_total_lifetime_FIXED.png")

if SAVE_PLOTS_TO_DISK:
    fig.savefig(obs_null_path, dpi=PLOT_DPI, bbox_inches="tight", facecolor=BLACK)
    fig.savefig(obs_null_copy_path, dpi=PLOT_DPI, bbox_inches="tight", facecolor=BLACK)

plt.close(fig)

if RENDER_ALL_IMAGES_INLINE and HAVE_IPYTHON and os.path.exists(obs_null_path):
    display(Image(filename=obs_null_path, width=1150))

top_channels = summary_df.sort_values("iaaft_corrected_ph_score", ascending=False).head(min(8, len(summary_df)))["channel"].tolist()

if len(iaaft_long_df) > 0 and "total_lifetime" in iaaft_long_df.columns:
    fig, ax = plt.subplots(figsize=(13, 6.4), facecolor=BLACK)
    style_ax(
        ax,
        title="IAAFT Null Distributions for Top Corrected Channels",
        xlabel="Channel",
        ylabel="H1 total lifetime",
    )

    box_data = []
    box_labels = []
    obs_for_labels = []

    for ch in top_channels:
        vals = iaaft_long_df.loc[
            iaaft_long_df["channel"] == ch,
            "total_lifetime",
        ].to_numpy(dtype=float)

        obs = float(summary_df.loc[summary_df["channel"] == ch, "observed_total_lifetime"].iloc[0])

        box_data.append(vals)
        box_labels.append(ch)
        obs_for_labels.append(obs)

    box = ax.boxplot(
        box_data,
        patch_artist=True,
        labels=box_labels,
        showfliers=False,
        widths=0.55,
    )

    for patch in box["boxes"]:
        patch.set_facecolor(CYAN_FILL)
        patch.set_edgecolor(ACCENT)
        patch.set_alpha(0.32)
        patch.set_linewidth(1.4)

    for key in ["whiskers", "caps", "medians"]:
        for item in box[key]:
            item.set_color(ACCENT)
            item.set_linewidth(1.4)

    for idx, obs in enumerate(obs_for_labels, start=1):
        ax.scatter(
            [idx],
            [obs],
            s=95,
            color=CYAN_FILL,
            edgecolor=ACCENT,
            linewidth=1.2,
            zorder=5,
            label="Observed" if idx == 1 else None,
        )

        ax.text(
            idx,
            obs,
            f"  obs={obs:.2f}",
            color=ACCENT,
            fontsize=8,
            va="center",
            ha="left",
        )

    legend = ax.legend(loc="best", framealpha=1.0)
    style_legend(legend)
    force_cyan_text(ax)

    null_box_path = os.path.join(mode_out_dir, "08_iaaft_null_distributions_top_channels.png")
    plt.tight_layout()
    save_render_close(fig, null_box_path, width=1150)

fig = plt.figure(figsize=(13.8, 8.2), facecolor=BLACK)
gs = GridSpec(
    1,
    2,
    width_ratios=[1.25, 0.85],
    wspace=0.25,
    figure=fig,
)

ax = fig.add_subplot(gs[0, 0])
rank_ax = fig.add_subplot(gs[0, 1])

style_ax(
    ax,
    title="Scalp Map of IAAFT-Corrected PH Score",
    grid=False,
)

scalp_xy = {
    "Fp1": (-0.45, 1.00), "Fpz": (0.00, 1.08), "Fp2": (0.45, 1.00),
    "F7": (-0.95, 0.60), "F3": (-0.45, 0.55), "Fz": (0.00, 0.60),
    "F4": (0.45, 0.55), "F8": (0.95, 0.60),
    "FC5": (-0.75, 0.25), "FC1": (-0.25, 0.25),
    "FC2": (0.25, 0.25), "FC6": (0.75, 0.25),
    "M1": (-1.15, 0.05), "T7": (-1.00, 0.00), "C3": (-0.50, 0.00),
    "Cz": (0.00, 0.00), "C4": (0.50, 0.00), "T8": (1.00, 0.00),
    "M2": (1.15, 0.05),
    "CP5": (-0.75, -0.30), "CP1": (-0.25, -0.30),
    "CP2": (0.25, -0.30), "CP6": (0.75, -0.30),
    "P7": (-0.90, -0.60), "P3": (-0.45, -0.60), "Pz": (0.00, -0.65),
    "P4": (0.45, -0.60), "P8": (0.90, -0.60),
    "POz": (0.00, -0.88),
    "O1": (-0.35, -1.00), "Oz": (0.00, -1.08), "O2": (0.35, -1.00),
}

head = plt.Circle((0, 0), 1.12, fill=False, color=ACCENT, linewidth=2.2)
ax.add_patch(head)
ax.plot([0.00, -0.10, 0.10, 0.00], [1.12, 1.28, 1.28, 1.12], color=ACCENT, linewidth=2.2)

scalp_df = summary_df[summary_df["channel"].isin(scalp_xy)].copy()
scalp_df["x"] = scalp_df["channel"].map(lambda ch: scalp_xy[ch][0])
scalp_df["y"] = scalp_df["channel"].map(lambda ch: scalp_xy[ch][1])

scores = scalp_df["iaaft_corrected_ph_score"].to_numpy(dtype=float)

if np.nanmax(scores) > np.nanmin(scores):
    norm_scores = (scores - np.nanmin(scores)) / (np.nanmax(scores) - np.nanmin(scores) + 1e-12)
else:
    norm_scores = np.full(len(scores), 0.5)

scalp_df["marker_size"] = 90.0 + 520.0 * norm_scores

scatter = ax.scatter(
    scalp_df["x"],
    scalp_df["y"],
    c=scalp_df["iaaft_corrected_ph_score"],
    s=scalp_df["marker_size"],
    cmap=CYAN_SEQUENTIAL_CMAP,
    vmin=0,
    vmax=100,
    edgecolor=ACCENT,
    linewidth=1.25,
    alpha=0.92,
    zorder=4,
)

top_score_channels = scalp_df.sort_values("iaaft_corrected_ph_score", ascending=False).head(8)["channel"].tolist()

for _, row in scalp_df.iterrows():
    ch = row["channel"]
    x0 = float(row["x"])
    y0 = float(row["y"])
    score = float(row["iaaft_corrected_ph_score"])

    txt = ax.text(
        x0,
        y0,
        ch,
        ha="center",
        va="center",
        fontsize=8,
        color=ACCENT,
        fontweight="bold",
        zorder=6,
    )
    txt.set_path_effects([pe.withStroke(linewidth=2.1, foreground=BLACK)])

    if ch in top_score_channels:
        ax.scatter(
            [x0],
            [y0],
            s=float(row["marker_size"]) * 1.75,
            facecolors="none",
            edgecolors=ACCENT_SOFT,
            linewidth=1.8,
            alpha=0.95,
            zorder=5,
        )

        dx = 0.06 if x0 <= 0 else -0.06
        ha = "left" if x0 <= 0 else "right"

        ann = ax.text(
            x0 + dx,
            y0 + 0.055,
            f"{score:.0f}",
            ha=ha,
            va="center",
            fontsize=8,
            color=ACCENT_SOFT,
            fontweight="bold",
            zorder=7,
        )
        ann.set_path_effects([pe.withStroke(linewidth=2.0, foreground=BLACK)])

ax.text(
    0,
    -1.30,
    "Bigger/brighter = stronger H1 structure after IAAFT correction | numbers mark top corrected channels",
    ha="center",
    va="top",
    color=ACCENT,
    fontsize=9,
)

ax.set_xlim(-1.38, 1.38)
ax.set_ylim(-1.40, 1.36)
ax.set_aspect("equal")
ax.set_xticks([])
ax.set_yticks([])

cbar = plt.colorbar(scatter, ax=ax, shrink=0.78, pad=0.03)
style_colorbar(cbar, "IAAFT-corrected PH score")

rank_side_df = scalp_df.sort_values("iaaft_corrected_ph_score", ascending=True).copy()

style_ax(
    rank_ax,
    title="IAAFT-Corrected PH Ranking",
    xlabel="Corrected PH score",
    ylabel="Channel",
)

y = np.arange(len(rank_side_df))

rank_ax.barh(
    y,
    rank_side_df["iaaft_corrected_ph_score"],
    color=CYAN_FILL,
    edgecolor=ACCENT,
    linewidth=0.8,
    alpha=0.82,
)

rank_ax.set_yticks(y)
rank_ax.set_yticklabels(rank_side_df["channel"])
rank_ax.set_xlim(0, 105)

rank_ax.axvline(
    float(np.nanmean(rank_side_df["iaaft_corrected_ph_score"])),
    color=ACCENT_SOFT,
    linestyle="--",
    linewidth=1.2,
    alpha=0.85,
    label=f"Mean={np.nanmean(rank_side_df['iaaft_corrected_ph_score']):.1f}",
)

for yy, (_, row) in zip(y, rank_side_df.iterrows()):
    score = float(row["iaaft_corrected_ph_score"])

    rank_ax.text(
        score + 1.2,
        yy,
        f"{score:.0f}",
        va="center",
        ha="left",
        color=ACCENT,
        fontsize=7,
        alpha=0.82,
    )

legend = rank_ax.legend(loc="lower right", framealpha=1.0, fontsize=8)
style_legend(legend)
force_cyan_text(rank_ax)

top_text = "Top IAAFT-corrected: " + ", ".join(
    [
        f"{row.channel}({row.iaaft_corrected_ph_score:.0f})"
        for row in scalp_df.sort_values("iaaft_corrected_ph_score", ascending=False).head(5).itertuples()
    ]
)

bottom_text = "Lowest IAAFT-corrected: " + ", ".join(
    [
        f"{row.channel}({row.iaaft_corrected_ph_score:.0f})"
        for row in scalp_df.sort_values("iaaft_corrected_ph_score", ascending=False).tail(5).itertuples()
    ]
)

fig.text(
    0.5,
    0.018,
    top_text + "    |    " + bottom_text,
    ha="center",
    va="bottom",
    color=ACCENT,
    fontsize=9,
)

scalp_path = os.path.join(mode_out_dir, "09_scalp_map_iaaft_corrected_ph_score.png")
plt.tight_layout(rect=[0.0, 0.045, 1.0, 1.0])
save_render_close(fig, scalp_path, width=1150)

fig, ax = plt.subplots(figsize=(12, 6.2), facecolor=BLACK)
style_ax(
    ax,
    title="Regional IAAFT-Corrected PH Score",
    xlabel="Region",
    ylabel="Mean IAAFT-corrected PH score",
)

x = np.arange(len(region_summary_df))

ax.bar(
    x,
    region_summary_df["mean_iaaft_corrected_ph_score"],
    yerr=region_summary_df["sem_iaaft_corrected_ph_score"],
    color=CYAN_FILL,
    edgecolor=ACCENT,
    linewidth=0.9,
    alpha=0.82,
    capsize=7,
    error_kw={"elinewidth": 1.5, "capthick": 1.5, "ecolor": ACCENT},
)

for i, row in region_summary_df.iterrows():
    ax.text(
        i,
        row["mean_iaaft_corrected_ph_score"] + row["sem_iaaft_corrected_ph_score"],
        f"{row['mean_iaaft_corrected_ph_score']:.1f}\nn={int(row['n_channels'])}",
        ha="center",
        va="bottom",
        color=ACCENT,
        fontsize=9,
    )

ax.set_xticks(x)
ax.set_xticklabels(region_summary_df["region"], rotation=35, ha="right")
ax.set_ylim(0, 105)
force_cyan_text(ax)

region_path = os.path.join(mode_out_dir, "10_regional_iaaft_corrected_ph_score.png")
plt.tight_layout()
save_render_close(fig, region_path, width=1050)

fig, ax = plt.subplots(figsize=(13, 6.2), facecolor=BLACK)
style_ax(
    ax,
    title="IAAFT-Corrected Primary Metric Z-Profiles by Channel",
    xlabel="Channel",
    ylabel="Mean primary z-score",
)

profile_df = summary_df.sort_values("iaaft_corrected_mean_primary_z", ascending=False).reset_index(drop=True)
x = np.arange(len(profile_df))

ax.bar(
    x,
    profile_df["iaaft_corrected_mean_primary_z"],
    color=CYAN_FILL,
    edgecolor=ACCENT,
    linewidth=0.9,
    alpha=0.82,
)

ax.axhline(0, color=ACCENT_SOFT, linestyle="--", linewidth=1.2)

for i, row in profile_df.iterrows():
    if i < 8:
        ax.text(
            i,
            row["iaaft_corrected_mean_primary_z"],
            f"{row['iaaft_corrected_mean_primary_z']:.1f}",
            ha="center",
            va="bottom" if row["iaaft_corrected_mean_primary_z"] >= 0 else "top",
            color=ACCENT,
            fontsize=8,
        )

ax.set_xticks(x)
ax.set_xticklabels(profile_df["channel"], rotation=90)
force_cyan_text(ax)

mean_z_path = os.path.join(mode_out_dir, "11_mean_primary_iaaft_z_ranking.png")
plt.tight_layout()
save_render_close(fig, mean_z_path, width=1150)

if "elapsed_sec" in summary_df.columns:
    qc_df = summary_df.sort_values("elapsed_sec", ascending=False).reset_index(drop=True)
else:
    qc_df = diagnostic_df.sort_values("elapsed_sec", ascending=False).reset_index(drop=True)

fig, ax = plt.subplots(figsize=(13, 5.8), facecolor=BLACK)
style_ax(
    ax,
    title="IAAFT-Surrogate PH Runtime QC",
    xlabel="Channel",
    ylabel="Elapsed seconds",
)

x = np.arange(len(qc_df))

ax.bar(
    x,
    qc_df["elapsed_sec"],
    color=CYAN_FILL,
    edgecolor=ACCENT,
    linewidth=0.9,
    alpha=0.82,
)

ax.set_xticks(x)
ax.set_xticklabels(qc_df["channel"], rotation=90)

for i, row in qc_df.iterrows():
    if i < 6:
        ax.text(
            i,
            row["elapsed_sec"],
            f"{row['elapsed_sec']:.0f}s",
            ha="center",
            va="bottom",
            color=ACCENT,
            fontsize=8,
        )

force_cyan_text(ax)

runtime_path = os.path.join(mode_out_dir, "12_runtime_qc.png")
plt.tight_layout()
save_render_close(fig, runtime_path, width=1150)

log("\n" + "=" * 72)
log("Finished regenerating all IAAFT plots.")
log("Fixed only:")
log("02_iaaft_corrected_primary_metric_z_heatmap.png")
log("03_iaaft_corrected_primary_metric_q_heatmap.png")
log("07_observed_vs_iaaft_mean_total_lifetime.png")
log("Kept the rest of the plot set the same.")
log(f"Output directory: {mode_out_dir}")
log("Saved plots:")
log("01_iaaft_corrected_ph_score_ranking.png")
log("02_iaaft_corrected_primary_metric_z_heatmap.png  <-- fixed")
log("02_iaaft_corrected_primary_metric_z_heatmap_FIXED.png")
log("03_iaaft_corrected_primary_metric_q_heatmap.png  <-- fixed")
log("03_iaaft_corrected_primary_metric_significance_effect_heatmap_fixed.png")
log("04_iaaft_corrected_ph_signature_similarity_heatmap.png")
log("05_iaaft_corrected_ph_signature_distance_heatmap.png")
log("06_mds_iaaft_corrected_ph_signatures.png")
log("07_observed_vs_iaaft_mean_total_lifetime.png  <-- fixed")
log("07_observed_vs_iaaft_mean_total_lifetime_FIXED.png")
if len(iaaft_long_df) > 0 and "total_lifetime" in iaaft_long_df.columns:
    log("08_iaaft_null_distributions_top_channels.png")
log("09_scalp_map_iaaft_corrected_ph_score.png")
log("10_regional_iaaft_corrected_ph_score.png")
log("11_mean_primary_iaaft_z_ranking.png")
log("12_runtime_qc.png")
log("=" * 72)
```

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patheffects as pe
from matplotlib.colors import LinearSegmentedColormap
from matplotlib.gridspec import GridSpec
from scipy.spatial.distance import pdist, squareform
from scipy.cluster.hierarchy import linkage, leaves_list

try:
    from scipy.linalg import eigh
    HAVE_SCIPY_EIGH = True
except Exception:
    HAVE_SCIPY_EIGH = False

try:
    from IPython.display import display, Image
    HAVE_IPYTHON = True
except Exception:
    HAVE_IPYTHON = False


base_dir = "/home/a/projects/Complete-Neural-Signal-Analysis"

FAST_OPTION = "ultra"
IAAFT_SURROGATE_TYPE = "independent_coordinate_iaaft"

SAVE_PLOTS_TO_DISK = True
RENDER_ALL_IMAGES_INLINE = True
INLINE_IMAGE_WIDTH = 1100
PLOT_DPI = 150

BLACK = "#000000"
ACCENT = "#00FFFF"
ACCENT_SOFT = "#66FFFF"
ACCENT_DIM = "#008B8B"
CYAN_FILL = "#00D5FF"
CYAN_FILL_DARK = "#005A63"

CYAN_SEQUENTIAL_CMAP = LinearSegmentedColormap.from_list(
    "cyan_sequential",
    [BLACK, CYAN_FILL_DARK, CYAN_FILL, ACCENT_SOFT],
    N=256,
)

CYAN_DIVERGING_CMAP = LinearSegmentedColormap.from_list(
    "cyan_diverging",
    [ACCENT_SOFT, BLACK, CYAN_FILL],
    N=256,
)

plt.rcParams.update({
    "figure.facecolor": BLACK,
    "axes.facecolor": BLACK,
    "savefig.facecolor": BLACK,
    "text.color": ACCENT,
    "axes.labelcolor": ACCENT,
    "axes.edgecolor": ACCENT,
    "axes.titlecolor": ACCENT,
    "xtick.color": ACCENT,
    "ytick.color": ACCENT,
    "grid.color": ACCENT_DIM,
    "grid.alpha": 0.25,
    "legend.facecolor": BLACK,
    "legend.edgecolor": ACCENT,
    "legend.labelcolor": ACCENT,
    "font.size": 11,
})


def log(msg):
    print(msg, flush=True)


def style_ax(ax, title=None, xlabel=None, ylabel=None, grid=True):
    ax.set_facecolor(BLACK)

    for spine in ax.spines.values():
        spine.set_color(ACCENT)
        spine.set_linewidth(1.1)

    ax.tick_params(axis="x", colors=ACCENT, labelcolor=ACCENT)
    ax.tick_params(axis="y", colors=ACCENT, labelcolor=ACCENT)

    if title is not None:
        ax.set_title(title, color=ACCENT, pad=12)

    if xlabel is not None:
        ax.set_xlabel(xlabel, color=ACCENT, labelpad=10)

    if ylabel is not None:
        ax.set_ylabel(ylabel, color=ACCENT, labelpad=10)

    if grid:
        ax.grid(True, color=ACCENT_DIM, alpha=0.25, linewidth=0.8)


def force_cyan_text(ax):
    ax.title.set_color(ACCENT)
    ax.xaxis.label.set_color(ACCENT)
    ax.yaxis.label.set_color(ACCENT)

    for tick_label in ax.get_xticklabels():
        tick_label.set_color(ACCENT)

    for tick_label in ax.get_yticklabels():
        tick_label.set_color(ACCENT)


def style_colorbar(cbar, label):
    cbar.set_label(label, color=ACCENT)
    cbar.ax.yaxis.set_tick_params(color=ACCENT)
    plt.setp(cbar.ax.get_yticklabels(), color=ACCENT)
    cbar.outline.set_edgecolor(ACCENT)


def style_legend(legend):
    if legend is not None:
        legend.get_frame().set_facecolor(BLACK)
        legend.get_frame().set_edgecolor(ACCENT)

        for text in legend.get_texts():
            text.set_color(ACCENT)


def save_render_close(fig, path, width=INLINE_IMAGE_WIDTH):
    if SAVE_PLOTS_TO_DISK:
        fig.savefig(path, dpi=PLOT_DPI, bbox_inches="tight", facecolor=BLACK)

    plt.close(fig)

    if RENDER_ALL_IMAGES_INLINE and HAVE_IPYTHON and os.path.exists(path):
        display(Image(filename=path, width=width))


def offdiag_values(M):
    M = np.asarray(M, dtype=float)

    if M.shape[0] <= 1:
        return np.asarray([], dtype=float)

    mask = ~np.eye(M.shape[0], dtype=bool)
    vals = M[mask]
    vals = vals[np.isfinite(vals)]

    return vals


def dynamic_limits(values, lower_pct=2, upper_pct=98, pad_frac=0.08, lower_bound=None, upper_bound=None):
    values = np.asarray(values, dtype=float)
    values = values[np.isfinite(values)]

    if len(values) == 0:
        vmin, vmax = 0.0, 1.0
    else:
        vmin = float(np.percentile(values, lower_pct))
        vmax = float(np.percentile(values, upper_pct))

        if not np.isfinite(vmin):
            vmin = float(np.nanmin(values))

        if not np.isfinite(vmax):
            vmax = float(np.nanmax(values))

        if vmax <= vmin:
            center = float(np.nanmean(values))
            eps = max(1e-6, abs(center) * 0.01, 1e-5)
            vmin = center - eps
            vmax = center + eps

    pad = (vmax - vmin) * pad_frac
    vmin -= pad
    vmax += pad

    if lower_bound is not None:
        vmin = max(lower_bound, vmin)

    if upper_bound is not None:
        vmax = min(upper_bound, vmax)

    if vmax <= vmin:
        vmax = vmin + 1e-6

    return vmin, vmax


def robust_limits(values, lower_pct=2, upper_pct=98, pad_frac=0.08, include_zero=True):
    values = np.asarray(values, dtype=float)
    values = values[np.isfinite(values)]

    if len(values) == 0:
        return 0.0, 1.0

    vmin = float(np.nanpercentile(values, lower_pct))
    vmax = float(np.nanpercentile(values, upper_pct))

    if include_zero:
        vmin = min(vmin, 0.0)
        vmax = max(vmax, 0.0)

    if vmax <= vmin:
        center = float(np.nanmean(values))
        eps = max(abs(center) * 0.05, 1e-6)
        vmin = center - eps
        vmax = center + eps

    pad = (vmax - vmin) * pad_frac

    return vmin - pad, vmax + pad


def bh_fdr(pvals):
    pvals = np.asarray(pvals, dtype=float)
    qvals = np.full_like(pvals, np.nan, dtype=float)

    valid = np.isfinite(pvals)
    p = pvals[valid]

    if len(p) == 0:
        return qvals

    order = np.argsort(p)
    ranked = p[order]
    m = len(ranked)

    raw_q = ranked * m / np.arange(1, m + 1)
    monotone_q = np.minimum.accumulate(raw_q[::-1])[::-1]
    monotone_q = np.clip(monotone_q, 0.0, 1.0)

    q_valid = np.empty_like(monotone_q)
    q_valid[order] = monotone_q

    qvals[valid] = q_valid

    return qvals


def percentile_score(values, higher_is_more=True):
    s = pd.Series(values, dtype=float)

    if not higher_is_more:
        s = -s

    if s.nunique(dropna=True) <= 1:
        return np.full(len(s), 50.0, dtype=float)

    return s.rank(method="average", pct=True).to_numpy(dtype=float) * 100.0


def classical_mds_from_distances(D, n_components=2):
    D = np.asarray(D, dtype=float)
    n = D.shape[0]

    if n < 2:
        return np.zeros((n, n_components), dtype=float)

    D2 = D ** 2
    J = np.eye(n) - np.ones((n, n)) / n
    B = -0.5 * J @ D2 @ J

    if HAVE_SCIPY_EIGH:
        vals, vecs = eigh(B)
    else:
        vals, vecs = np.linalg.eigh(B)

    idx = np.argsort(vals)[::-1]
    vals = vals[idx]
    vecs = vecs[:, idx]

    coords = np.zeros((n, n_components), dtype=float)

    for k in range(min(n_components, n)):
        if vals[k] > 0:
            coords[:, k] = vecs[:, k] * np.sqrt(vals[k])

    return coords


def gaussian_similarity_from_distance(D, sigma=None):
    D = np.asarray(D, dtype=float)
    off = offdiag_values(D)

    if sigma is None:
        sigma = float(np.nanmedian(off)) if len(off) else 1.0

    sigma = max(float(sigma), 1e-8)

    S = np.exp(-(D ** 2) / (2.0 * sigma ** 2))
    np.fill_diagonal(S, 1.0)

    return S, sigma


def robust_metric_contrast_matrix(raw_z_matrix):
    raw_z_matrix = np.asarray(raw_z_matrix, dtype=float)
    out = np.zeros_like(raw_z_matrix, dtype=float)
    info = []

    for j in range(raw_z_matrix.shape[1]):
        col = raw_z_matrix[:, j]
        finite = col[np.isfinite(col)]

        if len(finite) == 0:
            out[:, j] = 0.0
            info.append({"metric_index": j, "center_regularized_z": np.nan, "contrast_scale": np.nan})
            continue

        center = float(np.nanmedian(finite))
        mad = float(np.nanmedian(np.abs(finite - center)))
        scale = 1.4826 * mad

        if not np.isfinite(scale) or scale < 1e-9:
            p10 = float(np.nanpercentile(finite, 10))
            p90 = float(np.nanpercentile(finite, 90))
            scale = (p90 - p10) / 2.563

        if not np.isfinite(scale) or scale < 1e-9:
            scale = float(np.nanstd(finite))

        if not np.isfinite(scale) or scale < 1e-9:
            scale = 1.0

        out[:, j] = (col - center) / scale

        info.append({
            "metric_index": j,
            "center_regularized_z": center,
            "contrast_scale": scale,
        })

    return out, pd.DataFrame(info)


def find_iaaft_output_dir():
    candidate_dirs = [
        os.path.join(
            base_dir,
            "topology_iaaft_surrogate_persistent_homology",
            f"h1_iaaft_surrogate_ph_{FAST_OPTION}_fps_{IAAFT_SURROGATE_TYPE}_joblib_fft",
        ),
        os.path.join(
            base_dir,
            "topology_iaaft_surrogate_persistent_homology",
            f"h1_iaaft_surrogate_ph_{FAST_OPTION}_fps_{IAAFT_SURROGATE_TYPE}",
        ),
    ]

    for candidate_dir in candidate_dirs:
        summary_path = os.path.join(candidate_dir, "h1_iaaft_surrogate_ph_summary.csv")

        if os.path.exists(summary_path):
            return candidate_dir

    raise FileNotFoundError(
        "Could not find IAAFT surrogate PH output directory. Expected summary CSV in:\n\n"
        + "\n".join(
            os.path.join(d, "h1_iaaft_surrogate_ph_summary.csv")
            for d in candidate_dirs
        )
    )


def compute_regularized_z_columns(summary_df, iaaft_long_df, primary_metric_names):
    new_cols = {}

    for metric in primary_metric_names:
        reg_col = f"iaaft_z_regularized_{metric}"

        if reg_col in summary_df.columns:
            continue

        obs_col = f"observed_{metric}"
        mean_col = f"iaaft_mean_{metric}"
        std_col = f"iaaft_std_{metric}"

        if obs_col not in summary_df.columns or mean_col not in summary_df.columns:
            raise ValueError(f"Missing observed/IAAFT mean columns for metric: {metric}")

        obs = summary_df[obs_col].to_numpy(dtype=float)
        null_mean = summary_df[mean_col].to_numpy(dtype=float)

        if std_col in summary_df.columns:
            null_std = summary_df[std_col].to_numpy(dtype=float)
        else:
            null_std = np.zeros(len(summary_df), dtype=float)

        delta = obs - null_mean

        finite_std = null_std[np.isfinite(null_std) & (null_std > 0)]
        finite_delta = np.abs(delta[np.isfinite(delta)])

        floors = []

        if len(finite_std) > 0:
            floors.append(float(np.nanmedian(finite_std)) * 0.10)
            floors.append(float(np.nanpercentile(finite_std, 25)) * 0.25)

        if len(finite_delta) > 0:
            floors.append(float(np.nanmedian(finite_delta)) * 0.05)
            floors.append(float(np.nanpercentile(finite_delta, 75)) * 0.025)

        if len(iaaft_long_df) > 0 and metric in iaaft_long_df.columns:
            null_vals = iaaft_long_df[metric].to_numpy(dtype=float)
            null_vals = null_vals[np.isfinite(null_vals)]

            if len(null_vals) > 1:
                floors.append(float(np.nanstd(null_vals)) * 0.03)
                floors.append(float(np.nanmedian(np.abs(null_vals - np.nanmedian(null_vals)))) * 0.10)

        if np.any(np.isfinite(obs)):
            floors.append(float(np.nanmedian(np.abs(obs[np.isfinite(obs)]))) * 1e-6)

        if np.any(np.isfinite(null_mean)):
            floors.append(float(np.nanmedian(np.abs(null_mean[np.isfinite(null_mean)]))) * 1e-6)

        floors.append(1e-6)
        floors = [v for v in floors if np.isfinite(v) and v > 0]

        std_floor = max(floors) if len(floors) else 1e-6
        new_cols[reg_col] = delta / np.maximum(null_std, std_floor)

    if new_cols:
        summary_df = pd.concat([summary_df, pd.DataFrame(new_cols, index=summary_df.index)], axis=1).copy()
    else:
        summary_df = summary_df.copy()

    return summary_df


mode_out_dir = find_iaaft_output_dir()

summary_csv_path = os.path.join(mode_out_dir, "h1_iaaft_surrogate_ph_summary.csv")
iaaft_long_csv_path = os.path.join(mode_out_dir, "h1_iaaft_surrogate_ph_iaaft_long_table.csv")
diagnostic_csv_path = os.path.join(mode_out_dir, "h1_iaaft_surrogate_ph_runtime_diagnostics.csv")

summary_df = pd.read_csv(summary_csv_path)
summary_df["channel"] = summary_df["channel"].astype(str)

if os.path.exists(iaaft_long_csv_path):
    iaaft_long_df = pd.read_csv(iaaft_long_csv_path)
else:
    iaaft_long_df = pd.DataFrame()

if os.path.exists(diagnostic_csv_path):
    diagnostic_df = pd.read_csv(diagnostic_csv_path)
else:
    diagnostic_df = summary_df.copy()

primary_metric_names = [
    "total_lifetime",
    "max_lifetime",
    "q90_lifetime",
    "betti_auc",
    "betti_l2",
    "landscape_l2",
    "landscape_max",
]

primary_metric_labels = [
    "Total lifetime",
    "Max lifetime",
    "Q90 lifetime",
    "Betti AUC",
    "Betti L2",
    "Landscape L2",
    "Landscape max",
]

summary_df = compute_regularized_z_columns(
    summary_df,
    iaaft_long_df,
    primary_metric_names,
)

new_q_cols = {}

for metric in primary_metric_names:
    q_col = f"iaaft_q_greater_{metric}"
    p_col = f"iaaft_p_greater_{metric}"

    if q_col not in summary_df.columns:
        if p_col not in summary_df.columns:
            raise ValueError(f"Missing both q and p values for metric: {metric}")

        new_q_cols[q_col] = bh_fdr(summary_df[p_col].to_numpy(dtype=float))

    neg_col = f"iaaft_neglog10_q_{metric}"

    if neg_col not in summary_df.columns and q_col in summary_df.columns:
        new_q_cols[neg_col] = -np.log10(summary_df[q_col].to_numpy(dtype=float) + 1e-12)

if new_q_cols:
    summary_df = pd.concat([summary_df, pd.DataFrame(new_q_cols, index=summary_df.index)], axis=1).copy()

regularized_z_cols = [f"iaaft_z_regularized_{metric}" for metric in primary_metric_names]
q_cols = [f"iaaft_q_greater_{metric}" for metric in primary_metric_names]

if "iaaft_corrected_mean_primary_z" not in summary_df.columns:
    summary_df["iaaft_corrected_mean_primary_z"] = summary_df[regularized_z_cols].mean(axis=1)

if "iaaft_corrected_median_primary_z" not in summary_df.columns:
    summary_df["iaaft_corrected_median_primary_z"] = summary_df[regularized_z_cols].median(axis=1)

if "iaaft_corrected_ph_score" not in summary_df.columns:
    summary_df["iaaft_corrected_ph_score"] = percentile_score(
        summary_df["iaaft_corrected_mean_primary_z"],
        higher_is_more=True,
    )

if "iaaft_corrected_mean_primary_neglog10_q" not in summary_df.columns:
    neglogq_cols = [f"iaaft_neglog10_q_{metric}" for metric in primary_metric_names if f"iaaft_neglog10_q_{metric}" in summary_df.columns]
    summary_df["iaaft_corrected_mean_primary_neglog10_q"] = summary_df[neglogq_cols].mean(axis=1) if neglogq_cols else 0.0

if "iaaft_corrected_n_primary_q_lt_0_05" not in summary_df.columns:
    summary_df["iaaft_corrected_n_primary_q_lt_0_05"] = (summary_df[q_cols] < 0.05).sum(axis=1)

if "region" not in summary_df.columns:
    channel_regions = {
        "Fp1": "Frontal", "Fpz": "Frontal", "Fp2": "Frontal",
        "F7": "Frontal", "F3": "Frontal", "Fz": "Frontal", "F4": "Frontal", "F8": "Frontal",
        "FC5": "Frontocentral", "FC1": "Frontocentral", "FC2": "Frontocentral", "FC6": "Frontocentral",
        "M1": "Mastoid", "M2": "Mastoid",
        "T7": "Central", "C3": "Central", "Cz": "Central", "C4": "Central", "T8": "Central",
        "CP5": "Centroparietal", "CP1": "Centroparietal", "CP2": "Centroparietal", "CP6": "Centroparietal",
        "P7": "Parietal", "P3": "Parietal", "Pz": "Parietal", "P4": "Parietal", "P8": "Parietal",
        "POz": "Occipital", "O1": "Occipital", "Oz": "Occipital", "O2": "Occipital",
    }
    summary_df["region"] = summary_df["channel"].map(lambda ch: channel_regions.get(str(ch), "Unknown"))

summary_df = summary_df.sort_values("iaaft_corrected_ph_score", ascending=False).reset_index(drop=True)

signature_channels = summary_df["channel"].astype(str).tolist()
corrected_signature_df = summary_df.set_index("channel")[regularized_z_cols].replace([np.inf, -np.inf], np.nan).fillna(0.0)
signature_X = corrected_signature_df.to_numpy(dtype=float)

if len(signature_channels) >= 2:
    corrected_signature_distance = squareform(pdist(signature_X, metric="euclidean"))
else:
    corrected_signature_distance = np.zeros((len(signature_channels), len(signature_channels)), dtype=float)

corrected_signature_similarity, corrected_similarity_sigma = gaussian_similarity_from_distance(corrected_signature_distance)

if len(signature_channels) >= 3 and np.any(corrected_signature_distance > 0):
    Z_signature = linkage(squareform(corrected_signature_distance, checks=False), method="average", optimal_ordering=True)
    order = leaves_list(Z_signature)
else:
    order = np.arange(len(signature_channels))

ordered_channels = [signature_channels[i] for i in order]
ordered_signature_distance = corrected_signature_distance[np.ix_(order, order)]
ordered_signature_similarity = corrected_signature_similarity[np.ix_(order, order)]
ordered_signature_z = corrected_signature_df.loc[ordered_channels].to_numpy(dtype=float)

relative_matrix, contrast_info_df = robust_metric_contrast_matrix(ordered_signature_z)
relative_matrix_clipped = np.clip(relative_matrix, -3.0, 3.0)

contrast_info_df["metric"] = primary_metric_names

contrast_info_path = os.path.join(mode_out_dir, "iaaft_heatmap_within_metric_contrast_info.csv")
contrast_matrix_path = os.path.join(mode_out_dir, "iaaft_heatmap_within_metric_contrast_matrix.csv")
raw_matrix_path = os.path.join(mode_out_dir, "iaaft_heatmap_regularized_z_raw_matrix.csv")
fixed_summary_path = os.path.join(mode_out_dir, "h1_iaaft_surrogate_ph_summary_fixed_plot_inputs.csv")
fixed_signature_path = os.path.join(mode_out_dir, "h1_iaaft_surrogate_ph_regularized_z_signature_matrix_fixed_plot_inputs.csv")

summary_df.to_csv(fixed_summary_path, index=False)
corrected_signature_df.to_csv(fixed_signature_path)
contrast_info_df.to_csv(contrast_info_path, index=False)
pd.DataFrame(relative_matrix, index=ordered_channels, columns=primary_metric_names).to_csv(contrast_matrix_path)
pd.DataFrame(ordered_signature_z, index=ordered_channels, columns=primary_metric_names).to_csv(raw_matrix_path)

mds_coords = classical_mds_from_distances(corrected_signature_distance, n_components=2)
mds_df = pd.DataFrame({
    "channel": signature_channels,
    "region": [summary_df.set_index("channel").loc[ch, "region"] for ch in signature_channels],
    "iaaft_corrected_mds_1": mds_coords[:, 0],
    "iaaft_corrected_mds_2": mds_coords[:, 1],
    "iaaft_corrected_ph_score": [summary_df.set_index("channel").loc[ch, "iaaft_corrected_ph_score"] for ch in signature_channels],
})

region_summary_df = (
    summary_df
    .groupby("region")
    .agg(
        n_channels=("channel", "count"),
        mean_iaaft_corrected_ph_score=("iaaft_corrected_ph_score", "mean"),
        sem_iaaft_corrected_ph_score=("iaaft_corrected_ph_score", lambda x: float(np.std(x, ddof=1) / np.sqrt(len(x))) if len(x) > 1 else 0.0),
        mean_primary_z=("iaaft_corrected_mean_primary_z", "mean"),
        mean_primary_neglog10_q=("iaaft_corrected_mean_primary_neglog10_q", "mean"),
        mean_n_primary_q_lt_0_05=("iaaft_corrected_n_primary_q_lt_0_05", "mean"),
    )
    .reset_index()
    .sort_values("mean_iaaft_corrected_ph_score", ascending=False)
)

print("\nIAAFT PH summary used for regenerated plots:")
if HAVE_IPYTHON:
    display(summary_df)
else:
    print(summary_df)

print("\nWithin-metric contrast info for fixed heatmaps:")
if HAVE_IPYTHON:
    display(contrast_info_df)
else:
    print(contrast_info_df)

log("\n" + "=" * 72)
log("Regenerating ALL IAAFT plots.")
log("Fixed heatmaps 02 and 03 use within-metric contrast so they do not saturate.")
log("All other plots are included in the same plot set.")
log(f"Output directory: {mode_out_dir}")
log("=" * 72)


fig, ax = plt.subplots(figsize=(13, 6.2), facecolor=BLACK)
style_ax(
    ax,
    title=f"H1 IAAFT-Surrogate Persistent Homology Score | {FAST_OPTION.upper()}",
    xlabel="Channel",
    ylabel="IAAFT-corrected PH score",
)

rank_df = summary_df.sort_values("iaaft_corrected_ph_score", ascending=False).reset_index(drop=True)
x = np.arange(len(rank_df))

ax.bar(
    x,
    rank_df["iaaft_corrected_ph_score"],
    color=CYAN_FILL,
    edgecolor=ACCENT,
    linewidth=0.9,
    alpha=0.82,
)

ax.axhline(
    float(np.nanmean(rank_df["iaaft_corrected_ph_score"])),
    color=ACCENT_SOFT,
    linestyle="--",
    linewidth=1.3,
    alpha=0.85,
    label=f"Mean={np.nanmean(rank_df['iaaft_corrected_ph_score']):.1f}",
)

for i, row in rank_df.iterrows():
    if i < 8:
        ax.text(
            i,
            row["iaaft_corrected_ph_score"],
            f"{row['iaaft_corrected_ph_score']:.0f}",
            ha="center",
            va="bottom",
            color=ACCENT,
            fontsize=8,
        )

ax.set_xticks(x)
ax.set_xticklabels(rank_df["channel"], rotation=90)
ax.set_ylim(0, 105)

legend = ax.legend(loc="best", framealpha=1.0)
style_legend(legend)
force_cyan_text(ax)

score_path = os.path.join(mode_out_dir, "01_iaaft_corrected_ph_score_ranking.png")
plt.tight_layout()
save_render_close(fig, score_path, width=1150)


fig, ax = plt.subplots(figsize=(12.8, 9.2), facecolor=BLACK)

style_ax(
    ax,
    title="IAAFT-Corrected H1 Primary Metric Z-Score Heatmap | fixed contrast",
    xlabel="H1 metric",
    ylabel="Channel",
)

im = ax.imshow(
    relative_matrix_clipped,
    aspect="auto",
    interpolation="nearest",
    cmap=CYAN_DIVERGING_CMAP,
    vmin=-3,
    vmax=3,
)

ax.set_yticks(np.arange(len(ordered_channels)))
ax.set_yticklabels(ordered_channels)
ax.set_xticks(np.arange(len(primary_metric_labels)))
ax.set_xticklabels(primary_metric_labels, rotation=45, ha="right")

ax.set_xticks(np.arange(-0.5, len(primary_metric_labels), 1), minor=True)
ax.set_yticks(np.arange(-0.5, len(ordered_channels), 1), minor=True)
ax.grid(which="minor", color=ACCENT_DIM, linewidth=0.55, alpha=0.35)
ax.tick_params(which="minor", bottom=False, left=False)

for i in range(ordered_signature_z.shape[0]):
    for j in range(ordered_signature_z.shape[1]):
        rel_val = relative_matrix[i, j]
        raw_val = ordered_signature_z[i, j]

        if not np.isfinite(raw_val):
            continue

        show_text = abs(rel_val) >= 1.15 or i < 5 or i >= ordered_signature_z.shape[0] - 3

        if show_text:
            txt = ax.text(
                j,
                i,
                f"{raw_val:+.1f}",
                ha="center",
                va="center",
                color=ACCENT,
                fontsize=5.9,
                alpha=0.95,
                fontweight="bold",
            )
            txt.set_path_effects([pe.withStroke(linewidth=1.4, foreground=BLACK)])

cbar = plt.colorbar(im, ax=ax, shrink=0.85)
style_colorbar(cbar, "Within-metric channel contrast")
force_cyan_text(ax)

fig.text(
    0.09,
    0.035,
    "Fixed view: color shows within-metric channel contrast; printed values are the regularized IAAFT z-scores.",
    ha="left",
    va="bottom",
    color=ACCENT,
    fontsize=9,
)

plot_02_path = os.path.join(mode_out_dir, "02_iaaft_corrected_primary_metric_z_heatmap.png")
plot_02_copy_path = os.path.join(mode_out_dir, "02_iaaft_corrected_primary_metric_z_heatmap_FIXED_CONTRAST.png")
plt.tight_layout(rect=[0.0, 0.055, 1.0, 1.0])

if SAVE_PLOTS_TO_DISK:
    fig.savefig(plot_02_path, dpi=PLOT_DPI, bbox_inches="tight", facecolor=BLACK)
    fig.savefig(plot_02_copy_path, dpi=PLOT_DPI, bbox_inches="tight", facecolor=BLACK)

plt.close(fig)

if RENDER_ALL_IMAGES_INLINE and HAVE_IPYTHON and os.path.exists(plot_02_path):
    display(Image(filename=plot_02_path, width=1050))


order_df = summary_df.set_index("channel").loc[ordered_channels].reset_index().copy()
q_matrix = order_df[q_cols].to_numpy(dtype=float)
sig_005 = q_matrix < 0.05
sig_010 = (q_matrix >= 0.05) & (q_matrix < 0.10)
all_sig_005 = bool(np.all(sig_005))

fig = plt.figure(figsize=(16.8, 10.5), facecolor=BLACK)

ax = fig.add_axes([0.07, 0.18, 0.58, 0.72])
rank_ax = fig.add_axes([0.74, 0.18, 0.20, 0.72])
cbar_ax = fig.add_axes([0.67, 0.18, 0.018, 0.72])

for this_ax in [ax, rank_ax]:
    this_ax.set_facecolor(BLACK)

    for spine in this_ax.spines.values():
        spine.set_color(ACCENT)
        spine.set_linewidth(1.1)

    this_ax.tick_params(axis="x", colors=ACCENT, labelcolor=ACCENT)
    this_ax.tick_params(axis="y", colors=ACCENT, labelcolor=ACCENT)

im = ax.imshow(
    relative_matrix_clipped,
    aspect="auto",
    interpolation="nearest",
    cmap=CYAN_DIVERGING_CMAP,
    vmin=-3,
    vmax=3,
)

ax.set_title(
    "IAAFT-Corrected H1 Significance + Effect Direction | fixed contrast\n"
    "Color = within-metric channel contrast; printed values = regularized IAAFT z-score",
    color=ACCENT,
    pad=14,
)

ax.set_xlabel("H1 metric", color=ACCENT, labelpad=12)
ax.set_ylabel("Channel", color=ACCENT, labelpad=12)

ax.set_xticks(np.arange(len(primary_metric_labels)))
ax.set_xticklabels(primary_metric_labels, rotation=45, ha="right", color=ACCENT)
ax.set_yticks(np.arange(len(ordered_channels)))
ax.set_yticklabels(ordered_channels, color=ACCENT, fontsize=8)

ax.set_xticks(np.arange(-0.5, len(primary_metric_labels), 1), minor=True)
ax.set_yticks(np.arange(-0.5, len(ordered_channels), 1), minor=True)
ax.grid(which="minor", color=ACCENT_DIM, linewidth=0.55, alpha=0.35)
ax.tick_params(which="minor", bottom=False, left=False)

for i in range(ordered_signature_z.shape[0]):
    for j in range(ordered_signature_z.shape[1]):
        rel_val = relative_matrix[i, j]
        raw_val = ordered_signature_z[i, j]

        if not np.isfinite(raw_val):
            continue

        show_text = abs(rel_val) >= 1.15 or i < 5 or i >= ordered_signature_z.shape[0] - 3

        if show_text:
            txt = ax.text(
                j,
                i,
                f"{raw_val:+.1f}",
                ha="center",
                va="center",
                color=ACCENT,
                fontsize=5.3,
                alpha=0.92,
                fontweight="bold",
            )
            txt.set_path_effects([pe.withStroke(linewidth=1.4, foreground=BLACK)])

        if not all_sig_005:
            if sig_005[i, j]:
                marker = "●"
                marker_size = 8
            elif sig_010[i, j]:
                marker = "○"
                marker_size = 8
            else:
                marker = ""

            if marker:
                ax.text(
                    j,
                    i - 0.28,
                    marker,
                    ha="center",
                    va="center",
                    color=ACCENT_SOFT,
                    fontsize=marker_size,
                    fontweight="bold",
                )

cbar = plt.colorbar(im, cax=cbar_ax)
cbar.set_label("Within-metric channel contrast", color=ACCENT)
cbar.ax.yaxis.set_tick_params(color=ACCENT)
plt.setp(cbar.ax.get_yticklabels(), color=ACCENT)
cbar.outline.set_edgecolor(ACCENT)

score_values = order_df["iaaft_corrected_ph_score"].to_numpy(dtype=float)
y = np.arange(len(ordered_channels))

rank_ax.barh(
    y,
    score_values,
    color=CYAN_FILL,
    edgecolor=ACCENT,
    linewidth=0.7,
    alpha=0.82,
)

rank_ax.set_title("IAAFT-corrected\nPH score", color=ACCENT, pad=14)
rank_ax.set_xlabel("Score", color=ACCENT, labelpad=10)
rank_ax.set_yticks([])
rank_ax.invert_yaxis()
rank_ax.set_xlim(0, 105)

for yy, val in zip(y, score_values):
    if np.isfinite(val) and (yy < 7 or yy >= len(y) - 4):
        rank_ax.text(
            min(val + 1.0, 101.5),
            yy,
            f"{val:.0f}",
            va="center",
            ha="left",
            color=ACCENT,
            fontsize=7.5,
            alpha=0.95,
        )

rank_ax.grid(True, color=ACCENT_DIM, alpha=0.25, linewidth=0.8)

n_sig_005_by_metric = sig_005.sum(axis=0)

footer_lines = []

if all_sig_005:
    footer_lines.append("All displayed cells have FDR q < 0.05, so redundant significance dots are omitted.")
else:
    footer_lines.append("● = FDR q < 0.05    ○ = FDR q < 0.10")

footer_lines.append(
    "q<0.05 counts by metric: "
    + ", ".join([f"{label}: {count}" for label, count in zip(primary_metric_labels, n_sig_005_by_metric)])
)
footer_lines.append(
    "Fixed view: previous plot was saturated because every regularized z-score was large and positive. "
    "This plot keeps raw z labels but colors channels relative to other channels within each metric."
)

fig.text(
    0.07,
    0.055,
    "\n".join(footer_lines),
    ha="left",
    va="bottom",
    color=ACCENT,
    fontsize=9,
)

plot_03_path = os.path.join(mode_out_dir, "03_iaaft_corrected_primary_metric_q_heatmap.png")
plot_03_copy_path = os.path.join(mode_out_dir, "03_iaaft_corrected_primary_metric_q_heatmap_FIXED_CONTRAST.png")

if SAVE_PLOTS_TO_DISK:
    fig.savefig(plot_03_path, dpi=PLOT_DPI, bbox_inches="tight", facecolor=BLACK)
    fig.savefig(plot_03_copy_path, dpi=PLOT_DPI, bbox_inches="tight", facecolor=BLACK)

plt.close(fig)

if RENDER_ALL_IMAGES_INLINE and HAVE_IPYTHON and os.path.exists(plot_03_path):
    display(Image(filename=plot_03_path, width=1150))


sim_vmin, sim_vmax = dynamic_limits(
    offdiag_values(ordered_signature_similarity),
    lower_pct=2,
    upper_pct=98,
    pad_frac=0.10,
    lower_bound=0.0,
    upper_bound=1.0,
)

fig, ax = plt.subplots(figsize=(9.5, 8.5), facecolor=BLACK)
style_ax(
    ax,
    title="IAAFT-Corrected PH Signature Similarity Matrix",
    xlabel="Channel",
    ylabel="Channel",
)

im = ax.imshow(
    ordered_signature_similarity,
    aspect="equal",
    interpolation="nearest",
    cmap=CYAN_SEQUENTIAL_CMAP,
    vmin=sim_vmin,
    vmax=sim_vmax,
)

ax.set_xticks(np.arange(len(ordered_channels)))
ax.set_xticklabels(ordered_channels, rotation=90)
ax.set_yticks(np.arange(len(ordered_channels)))
ax.set_yticklabels(ordered_channels)

cbar = plt.colorbar(im, ax=ax, shrink=0.85)
style_colorbar(cbar, "IAAFT-corrected signature similarity")
force_cyan_text(ax)

signature_similarity_path = os.path.join(mode_out_dir, "04_iaaft_corrected_ph_signature_similarity_heatmap.png")
plt.tight_layout()
save_render_close(fig, signature_similarity_path, width=950)


dist_vmin, dist_vmax = dynamic_limits(
    offdiag_values(ordered_signature_distance),
    lower_pct=2,
    upper_pct=98,
    pad_frac=0.10,
    lower_bound=0.0,
)

fig, ax = plt.subplots(figsize=(9.5, 8.5), facecolor=BLACK)
style_ax(
    ax,
    title="IAAFT-Corrected PH Signature Distance Matrix",
    xlabel="Channel",
    ylabel="Channel",
)

im = ax.imshow(
    ordered_signature_distance,
    aspect="equal",
    interpolation="nearest",
    cmap=CYAN_SEQUENTIAL_CMAP,
    vmin=dist_vmin,
    vmax=dist_vmax,
)

ax.set_xticks(np.arange(len(ordered_channels)))
ax.set_xticklabels(ordered_channels, rotation=90)
ax.set_yticks(np.arange(len(ordered_channels)))
ax.set_yticklabels(ordered_channels)

cbar = plt.colorbar(im, ax=ax, shrink=0.85)
style_colorbar(cbar, "IAAFT-corrected signature distance")
force_cyan_text(ax)

signature_distance_path = os.path.join(mode_out_dir, "05_iaaft_corrected_ph_signature_distance_heatmap.png")
plt.tight_layout()
save_render_close(fig, signature_distance_path, width=950)


fig, ax = plt.subplots(figsize=(8.5, 7.2), facecolor=BLACK)
style_ax(
    ax,
    title="MDS of IAAFT-Corrected PH Signatures",
    xlabel="IAAFT-corrected PH MDS 1",
    ylabel="IAAFT-corrected PH MDS 2",
)

sc = ax.scatter(
    mds_df["iaaft_corrected_mds_1"],
    mds_df["iaaft_corrected_mds_2"],
    c=mds_df["iaaft_corrected_ph_score"],
    s=165,
    cmap=CYAN_SEQUENTIAL_CMAP,
    vmin=0,
    vmax=100,
    edgecolor=ACCENT,
    linewidth=1.2,
    alpha=0.95,
)

for _, row in mds_df.iterrows():
    txt = ax.text(
        row["iaaft_corrected_mds_1"],
        row["iaaft_corrected_mds_2"],
        row["channel"],
        fontsize=8,
        color=ACCENT,
        alpha=0.95,
        zorder=5,
    )
    txt.set_path_effects([pe.withStroke(linewidth=2.0, foreground=BLACK)])

cbar = plt.colorbar(sc, ax=ax, shrink=0.82)
style_colorbar(cbar, "IAAFT-corrected PH score")
force_cyan_text(ax)

mds_path = os.path.join(mode_out_dir, "06_mds_iaaft_corrected_ph_signatures.png")
plt.tight_layout()
save_render_close(fig, mds_path, width=930)


plot_metric = "total_lifetime"

plot_df = summary_df.copy()
plot_df["observed_total_lifetime"] = plot_df[f"observed_{plot_metric}"].to_numpy(dtype=float)
plot_df["iaaft_mean_total_lifetime"] = plot_df[f"iaaft_mean_{plot_metric}"].to_numpy(dtype=float)
plot_df["iaaft_std_total_lifetime"] = plot_df[f"iaaft_std_{plot_metric}"].to_numpy(dtype=float) if f"iaaft_std_{plot_metric}" in plot_df.columns else 0.0
plot_df["delta_total_lifetime"] = plot_df["observed_total_lifetime"] - plot_df["iaaft_mean_total_lifetime"]
plot_df["q_total_lifetime"] = plot_df[f"iaaft_q_greater_{plot_metric}"].to_numpy(dtype=float)

plot_df = plot_df.replace([np.inf, -np.inf], np.nan)
paired_df = plot_df.sort_values("delta_total_lifetime", ascending=True).reset_index(drop=True)

fig = plt.figure(figsize=(15.8, 8.4), facecolor=BLACK)

paired_ax = fig.add_axes([0.07, 0.14, 0.42, 0.74])
rank_ax = fig.add_axes([0.59, 0.14, 0.34, 0.74])

y = np.arange(len(paired_df))

style_ax(
    paired_ax,
    title="Observed vs IAAFT Mean H1 Total Lifetime | fixed paired view",
    xlabel="H1 total lifetime",
    ylabel="Channel",
)

paired_ax.hlines(
    y,
    paired_df["iaaft_mean_total_lifetime"],
    paired_df["observed_total_lifetime"],
    color=ACCENT_DIM,
    linewidth=1.0,
    alpha=0.55,
)

paired_ax.scatter(
    paired_df["iaaft_mean_total_lifetime"],
    y,
    s=72,
    facecolors=BLACK,
    edgecolors=ACCENT_SOFT,
    linewidth=1.2,
    label="IAAFT mean",
    zorder=4,
)

paired_ax.scatter(
    paired_df["observed_total_lifetime"],
    y,
    s=82,
    color=CYAN_FILL,
    edgecolor=ACCENT,
    linewidth=1.0,
    label="Observed",
    zorder=5,
)

paired_ax.set_yticks(y)
paired_ax.set_yticklabels(paired_df["channel"], fontsize=7)

x_vals = np.concatenate([
    paired_df["iaaft_mean_total_lifetime"].to_numpy(dtype=float),
    paired_df["observed_total_lifetime"].to_numpy(dtype=float),
])
x_lo, x_hi = robust_limits(x_vals, lower_pct=1, upper_pct=99, pad_frac=0.12, include_zero=True)
paired_ax.set_xlim(x_lo, x_hi)

legend = paired_ax.legend(loc="lower right", framealpha=1.0, fontsize=8)
style_legend(legend)

style_ax(
    rank_ax,
    title="Observed − IAAFT Mean",
    xlabel="Δ H1 total lifetime",
    ylabel="Channel",
)

rank_ax.barh(
    y,
    paired_df["delta_total_lifetime"],
    color=CYAN_FILL,
    edgecolor=ACCENT,
    linewidth=0.75,
    alpha=0.82,
)

rank_ax.axvline(
    0,
    color=ACCENT_SOFT,
    linestyle="--",
    linewidth=1.2,
    alpha=0.9,
)

rank_ax.set_yticks(y)
rank_ax.set_yticklabels(paired_df["channel"], fontsize=7)

delta_lo, delta_hi = robust_limits(
    paired_df["delta_total_lifetime"].to_numpy(dtype=float),
    lower_pct=1,
    upper_pct=99,
    pad_frac=0.18,
    include_zero=True,
)
rank_ax.set_xlim(delta_lo, delta_hi)

for yy, (_, row) in zip(y, paired_df.iterrows()):
    delta = float(row["delta_total_lifetime"])
    q_val = float(row["q_total_lifetime"])

    if q_val < 0.05:
        marker = "●"
    elif q_val < 0.10:
        marker = "○"
    else:
        marker = ""

    if marker:
        x_text = delta + (0.015 * (delta_hi - delta_lo) if delta >= 0 else -0.015 * (delta_hi - delta_lo))
        ha = "left" if delta >= 0 else "right"

        rank_ax.text(
            x_text,
            yy,
            marker,
            va="center",
            ha=ha,
            color=ACCENT_SOFT,
            fontsize=9,
            fontweight="bold",
        )

force_cyan_text(paired_ax)
force_cyan_text(rank_ax)

footer = (
    "Fixed view: paired plot avoids the collapsed x≈0 scatter. "
    "Left compares each channel's IAAFT mean to observed H1 total lifetime; right shows signed effect. "
    "● = FDR q < 0.05, ○ = FDR q < 0.10."
)

fig.text(
    0.07,
    0.035,
    footer,
    ha="left",
    va="bottom",
    color=ACCENT,
    fontsize=9,
)

obs_null_path = os.path.join(mode_out_dir, "07_observed_vs_iaaft_mean_total_lifetime.png")
obs_null_copy_path = os.path.join(mode_out_dir, "07_observed_vs_iaaft_mean_total_lifetime_FIXED.png")

if SAVE_PLOTS_TO_DISK:
    fig.savefig(obs_null_path, dpi=PLOT_DPI, bbox_inches="tight", facecolor=BLACK)
    fig.savefig(obs_null_copy_path, dpi=PLOT_DPI, bbox_inches="tight", facecolor=BLACK)

plt.close(fig)

if RENDER_ALL_IMAGES_INLINE and HAVE_IPYTHON and os.path.exists(obs_null_path):
    display(Image(filename=obs_null_path, width=1150))


top_channels = summary_df.sort_values("iaaft_corrected_ph_score", ascending=False).head(min(8, len(summary_df)))["channel"].tolist()

if len(iaaft_long_df) > 0 and "total_lifetime" in iaaft_long_df.columns:
    fig, ax = plt.subplots(figsize=(13, 6.4), facecolor=BLACK)
    style_ax(
        ax,
        title="IAAFT Null Distributions for Top Corrected Channels",
        xlabel="Channel",
        ylabel="H1 total lifetime",
    )

    box_data = []
    box_labels = []
    obs_for_labels = []

    for ch in top_channels:
        vals = iaaft_long_df.loc[
            iaaft_long_df["channel"] == ch,
            "total_lifetime",
        ].to_numpy(dtype=float)

        obs = float(summary_df.loc[summary_df["channel"] == ch, "observed_total_lifetime"].iloc[0])

        box_data.append(vals)
        box_labels.append(ch)
        obs_for_labels.append(obs)

    box = ax.boxplot(
        box_data,
        patch_artist=True,
        labels=box_labels,
        showfliers=False,
        widths=0.55,
    )

    for patch in box["boxes"]:
        patch.set_facecolor(CYAN_FILL)
        patch.set_edgecolor(ACCENT)
        patch.set_alpha(0.32)
        patch.set_linewidth(1.4)

    for key in ["whiskers", "caps", "medians"]:
        for item in box[key]:
            item.set_color(ACCENT)
            item.set_linewidth(1.4)

    for idx, obs in enumerate(obs_for_labels, start=1):
        ax.scatter(
            [idx],
            [obs],
            s=95,
            color=CYAN_FILL,
            edgecolor=ACCENT,
            linewidth=1.2,
            zorder=5,
            label="Observed" if idx == 1 else None,
        )

        ax.text(
            idx,
            obs,
            f"  obs={obs:.2f}",
            color=ACCENT,
            fontsize=8,
            va="center",
            ha="left",
        )

    legend = ax.legend(loc="best", framealpha=1.0)
    style_legend(legend)
    force_cyan_text(ax)

    null_box_path = os.path.join(mode_out_dir, "08_iaaft_null_distributions_top_channels.png")
    plt.tight_layout()
    save_render_close(fig, null_box_path, width=1150)


fig = plt.figure(figsize=(13.8, 8.2), facecolor=BLACK)
gs = GridSpec(
    1,
    2,
    width_ratios=[1.25, 0.85],
    wspace=0.25,
    figure=fig,
)

ax = fig.add_subplot(gs[0, 0])
rank_ax = fig.add_subplot(gs[0, 1])

style_ax(
    ax,
    title="Scalp Map of IAAFT-Corrected PH Score",
    grid=False,
)

scalp_xy = {
    "Fp1": (-0.45, 1.00), "Fpz": (0.00, 1.08), "Fp2": (0.45, 1.00),
    "F7": (-0.95, 0.60), "F3": (-0.45, 0.55), "Fz": (0.00, 0.60),
    "F4": (0.45, 0.55), "F8": (0.95, 0.60),
    "FC5": (-0.75, 0.25), "FC1": (-0.25, 0.25),
    "FC2": (0.25, 0.25), "FC6": (0.75, 0.25),
    "M1": (-1.15, 0.05), "T7": (-1.00, 0.00), "C3": (-0.50, 0.00),
    "Cz": (0.00, 0.00), "C4": (0.50, 0.00), "T8": (1.00, 0.00),
    "M2": (1.15, 0.05),
    "CP5": (-0.75, -0.30), "CP1": (-0.25, -0.30),
    "CP2": (0.25, -0.30), "CP6": (0.75, -0.30),
    "P7": (-0.90, -0.60), "P3": (-0.45, -0.60), "Pz": (0.00, -0.65),
    "P4": (0.45, -0.60), "P8": (0.90, -0.60),
    "POz": (0.00, -0.88),
    "O1": (-0.35, -1.00), "Oz": (0.00, -1.08), "O2": (0.35, -1.00),
}

head = plt.Circle((0, 0), 1.12, fill=False, color=ACCENT, linewidth=2.2)
ax.add_patch(head)
ax.plot([0.00, -0.10, 0.10, 0.00], [1.12, 1.28, 1.28, 1.12], color=ACCENT, linewidth=2.2)

scalp_df = summary_df[summary_df["channel"].isin(scalp_xy)].copy()
scalp_df["x"] = scalp_df["channel"].map(lambda ch: scalp_xy[ch][0])
scalp_df["y"] = scalp_df["channel"].map(lambda ch: scalp_xy[ch][1])

scores = scalp_df["iaaft_corrected_ph_score"].to_numpy(dtype=float)

if np.nanmax(scores) > np.nanmin(scores):
    norm_scores = (scores - np.nanmin(scores)) / (np.nanmax(scores) - np.nanmin(scores) + 1e-12)
else:
    norm_scores = np.full(len(scores), 0.5)

scalp_df["marker_size"] = 90.0 + 520.0 * norm_scores

scatter = ax.scatter(
    scalp_df["x"],
    scalp_df["y"],
    c=scalp_df["iaaft_corrected_ph_score"],
    s=scalp_df["marker_size"],
    cmap=CYAN_SEQUENTIAL_CMAP,
    vmin=0,
    vmax=100,
    edgecolor=ACCENT,
    linewidth=1.25,
    alpha=0.92,
    zorder=4,
)

top_score_channels = scalp_df.sort_values("iaaft_corrected_ph_score", ascending=False).head(8)["channel"].tolist()

for _, row in scalp_df.iterrows():
    ch = row["channel"]
    x0 = float(row["x"])
    y0 = float(row["y"])
    score = float(row["iaaft_corrected_ph_score"])

    txt = ax.text(
        x0,
        y0,
        ch,
        ha="center",
        va="center",
        fontsize=8,
        color=ACCENT,
        fontweight="bold",
        zorder=6,
    )
    txt.set_path_effects([pe.withStroke(linewidth=2.1, foreground=BLACK)])

    if ch in top_score_channels:
        ax.scatter(
            [x0],
            [y0],
            s=float(row["marker_size"]) * 1.75,
            facecolors="none",
            edgecolors=ACCENT_SOFT,
            linewidth=1.8,
            alpha=0.95,
            zorder=5,
        )

        dx = 0.06 if x0 <= 0 else -0.06
        ha = "left" if x0 <= 0 else "right"

        ann = ax.text(
            x0 + dx,
            y0 + 0.055,
            f"{score:.0f}",
            ha=ha,
            va="center",
            fontsize=8,
            color=ACCENT_SOFT,
            fontweight="bold",
            zorder=7,
        )
        ann.set_path_effects([pe.withStroke(linewidth=2.0, foreground=BLACK)])

ax.text(
    0,
    -1.30,
    "Bigger/brighter = stronger H1 structure after IAAFT correction | numbers mark top corrected channels",
    ha="center",
    va="top",
    color=ACCENT,
    fontsize=9,
)

ax.set_xlim(-1.38, 1.38)
ax.set_ylim(-1.40, 1.36)
ax.set_aspect("equal")
ax.set_xticks([])
ax.set_yticks([])

cbar = plt.colorbar(scatter, ax=ax, shrink=0.78, pad=0.03)
style_colorbar(cbar, "IAAFT-corrected PH score")

rank_side_df = scalp_df.sort_values("iaaft_corrected_ph_score", ascending=True).copy()

style_ax(
    rank_ax,
    title="IAAFT-Corrected PH Ranking",
    xlabel="Corrected PH score",
    ylabel="Channel",
)

y = np.arange(len(rank_side_df))

rank_ax.barh(
    y,
    rank_side_df["iaaft_corrected_ph_score"],
    color=CYAN_FILL,
    edgecolor=ACCENT,
    linewidth=0.8,
    alpha=0.82,
)

rank_ax.set_yticks(y)
rank_ax.set_yticklabels(rank_side_df["channel"])
rank_ax.set_xlim(0, 105)

rank_ax.axvline(
    float(np.nanmean(rank_side_df["iaaft_corrected_ph_score"])),
    color=ACCENT_SOFT,
    linestyle="--",
    linewidth=1.2,
    alpha=0.85,
    label=f"Mean={np.nanmean(rank_side_df['iaaft_corrected_ph_score']):.1f}",
)

for yy, (_, row) in zip(y, rank_side_df.iterrows()):
    score = float(row["iaaft_corrected_ph_score"])

    rank_ax.text(
        score + 1.2,
        yy,
        f"{score:.0f}",
        va="center",
        ha="left",
        color=ACCENT,
        fontsize=7,
        alpha=0.82,
    )

legend = rank_ax.legend(loc="lower right", framealpha=1.0, fontsize=8)
style_legend(legend)
force_cyan_text(rank_ax)

top_text = "Top IAAFT-corrected: " + ", ".join(
    [
        f"{row.channel}({row.iaaft_corrected_ph_score:.0f})"
        for row in scalp_df.sort_values("iaaft_corrected_ph_score", ascending=False).head(5).itertuples()
    ]
)

bottom_text = "Lowest IAAFT-corrected: " + ", ".join(
    [
        f"{row.channel}({row.iaaft_corrected_ph_score:.0f})"
        for row in scalp_df.sort_values("iaaft_corrected_ph_score", ascending=False).tail(5).itertuples()
    ]
)

fig.text(
    0.5,
    0.018,
    top_text + "    |    " + bottom_text,
    ha="center",
    va="bottom",
    color=ACCENT,
    fontsize=9,
)

scalp_path = os.path.join(mode_out_dir, "09_scalp_map_iaaft_corrected_ph_score.png")
plt.tight_layout(rect=[0.0, 0.045, 1.0, 1.0])
save_render_close(fig, scalp_path, width=1150)


fig, ax = plt.subplots(figsize=(12, 6.2), facecolor=BLACK)
style_ax(
    ax,
    title="Regional IAAFT-Corrected PH Score",
    xlabel="Region",
    ylabel="Mean IAAFT-corrected PH score",
)

x = np.arange(len(region_summary_df))

ax.bar(
    x,
    region_summary_df["mean_iaaft_corrected_ph_score"],
    yerr=region_summary_df["sem_iaaft_corrected_ph_score"],
    color=CYAN_FILL,
    edgecolor=ACCENT,
    linewidth=0.9,
    alpha=0.82,
    capsize=7,
    error_kw={"elinewidth": 1.5, "capthick": 1.5, "ecolor": ACCENT},
)

for i, row in region_summary_df.iterrows():
    ax.text(
        i,
        row["mean_iaaft_corrected_ph_score"] + row["sem_iaaft_corrected_ph_score"],
        f"{row['mean_iaaft_corrected_ph_score']:.1f}\nn={int(row['n_channels'])}",
        ha="center",
        va="bottom",
        color=ACCENT,
        fontsize=9,
    )

ax.set_xticks(x)
ax.set_xticklabels(region_summary_df["region"], rotation=35, ha="right")
ax.set_ylim(0, 105)
force_cyan_text(ax)

region_path = os.path.join(mode_out_dir, "10_regional_iaaft_corrected_ph_score.png")
plt.tight_layout()
save_render_close(fig, region_path, width=1050)


fig, ax = plt.subplots(figsize=(13, 6.2), facecolor=BLACK)
style_ax(
    ax,
    title="IAAFT-Corrected Primary Metric Z-Profiles by Channel",
    xlabel="Channel",
    ylabel="Mean primary z-score",
)

profile_df = summary_df.sort_values("iaaft_corrected_mean_primary_z", ascending=False).reset_index(drop=True)
x = np.arange(len(profile_df))

ax.bar(
    x,
    profile_df["iaaft_corrected_mean_primary_z"],
    color=CYAN_FILL,
    edgecolor=ACCENT,
    linewidth=0.9,
    alpha=0.82,
)

ax.axhline(0, color=ACCENT_SOFT, linestyle="--", linewidth=1.2)

for i, row in profile_df.iterrows():
    if i < 8:
        ax.text(
            i,
            row["iaaft_corrected_mean_primary_z"],
            f"{row['iaaft_corrected_mean_primary_z']:.1f}",
            ha="center",
            va="bottom" if row["iaaft_corrected_mean_primary_z"] >= 0 else "top",
            color=ACCENT,
            fontsize=8,
        )

ax.set_xticks(x)
ax.set_xticklabels(profile_df["channel"], rotation=90)
force_cyan_text(ax)

mean_z_path = os.path.join(mode_out_dir, "11_mean_primary_iaaft_z_ranking.png")
plt.tight_layout()
save_render_close(fig, mean_z_path, width=1150)


if "elapsed_sec" in summary_df.columns:
    qc_df = summary_df.sort_values("elapsed_sec", ascending=False).reset_index(drop=True)
else:
    qc_df = diagnostic_df.sort_values("elapsed_sec", ascending=False).reset_index(drop=True)

fig, ax = plt.subplots(figsize=(13, 5.8), facecolor=BLACK)
style_ax(
    ax,
    title="IAAFT-Surrogate PH Runtime QC",
    xlabel="Channel",
    ylabel="Elapsed seconds",
)

x = np.arange(len(qc_df))

ax.bar(
    x,
    qc_df["elapsed_sec"],
    color=CYAN_FILL,
    edgecolor=ACCENT,
    linewidth=0.9,
    alpha=0.82,
)

ax.set_xticks(x)
ax.set_xticklabels(qc_df["channel"], rotation=90)

for i, row in qc_df.iterrows():
    if i < 6:
        ax.text(
            i,
            row["elapsed_sec"],
            f"{row['elapsed_sec']:.0f}s",
            ha="center",
            va="bottom",
            color=ACCENT,
            fontsize=8,
        )

force_cyan_text(ax)

runtime_path = os.path.join(mode_out_dir, "12_runtime_qc.png")
plt.tight_layout()
save_render_close(fig, runtime_path, width=1150)

log("\n" + "=" * 72)
log("Finished regenerating ALL IAAFT plots.")
log("Fixed heatmaps:")
log("02_iaaft_corrected_primary_metric_z_heatmap.png")
log("03_iaaft_corrected_primary_metric_q_heatmap.png")
log("Included all other plots:")
log("01_iaaft_corrected_ph_score_ranking.png")
log("04_iaaft_corrected_ph_signature_similarity_heatmap.png")
log("05_iaaft_corrected_ph_signature_distance_heatmap.png")
log("06_mds_iaaft_corrected_ph_signatures.png")
log("07_observed_vs_iaaft_mean_total_lifetime.png")
if len(iaaft_long_df) > 0 and "total_lifetime" in iaaft_long_df.columns:
    log("08_iaaft_null_distributions_top_channels.png")
log("09_scalp_map_iaaft_corrected_ph_score.png")
log("10_regional_iaaft_corrected_ph_score.png")
log("11_mean_primary_iaaft_z_ranking.png")
log("12_runtime_qc.png")
log(f"Output directory: {mode_out_dir}")
log("=" * 72)
```

```python
import os

os.environ.setdefault("OMP_NUM_THREADS", "1")
os.environ.setdefault("OPENBLAS_NUM_THREADS", "1")
os.environ.setdefault("MKL_NUM_THREADS", "1")
os.environ.setdefault("VECLIB_MAXIMUM_THREADS", "1")
os.environ.setdefault("NUMEXPR_NUM_THREADS", "1")

import time
import numpy as np

if not hasattr(np, "trapz"):
    np.trapz = np.trapezoid

import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patheffects as pe
from matplotlib.colors import LinearSegmentedColormap
from matplotlib.gridspec import GridSpec
from scipy.spatial.distance import pdist, squareform
from scipy.cluster.hierarchy import linkage, leaves_list
from scipy import fft as sp_fft

try:
    from scipy.linalg import eigh
    HAVE_SCIPY_EIGH = True
except Exception:
    HAVE_SCIPY_EIGH = False

try:
    from IPython.display import display, Image
    HAVE_IPYTHON = True
except Exception:
    HAVE_IPYTHON = False

try:
    from joblib import Parallel, delayed
except Exception as exc:
    raise ImportError(
        "joblib is not installed. Run this first:\n\n"
        "!pip install joblib\n\n"
        "Then rerun this colored-noise-surrogate PH cell."
    ) from exc

try:
    from ripser import ripser
except Exception as exc:
    raise ImportError(
        "ripser is not installed. Run this first:\n\n"
        "!pip install ripser persim\n\n"
        "Then rerun this colored-noise-surrogate PH cell."
    ) from exc


base_dir = "/home/a/projects/Complete-Neural-Signal-Analysis"

FAST_OPTION = "ultra"

MAX_CHANNELS_TO_PROCESS = None
SAVE_PLOTS_TO_DISK = True
RENDER_ALL_IMAGES_INLINE = True
INLINE_IMAGE_WIDTH = 1100
PLOT_DPI = 150
VERBOSE_PROGRESS = True

COLORED_NOISE_SURROGATE_TYPE = "independent_coordinate_colored_noise"
USE_GEOMETRY_PRESERVING_SUBSAMPLE = True

CPU_COUNT = os.cpu_count() or 1
JOBLIB_N_JOBS = max(1, min(CPU_COUNT - 1, 8))
JOBLIB_BACKEND = "loky"
FFT_WORKERS_PER_JOB = max(1, min(4, CPU_COUNT // max(1, JOBLIB_N_JOBS)))

MODE_CONFIGS = {
    "ultra": {
        "max_points": 90,
        "ordered_reservoir_points": 5000,
        "max_dims": 3,
        "radius_quantile": 0.22,
        "n_colored_noise_surrogates": 48,
        "beta_clip": (-0.5, 3.5),
        "betti_grid_size": 120,
        "landscape_grid_size": 120,
        "landscape_levels": 4,
        "top_lifetimes": 12,
        "cap_essential_h1": True,
        "plot_dpi": 150,
    },
    "fast": {
        "max_points": 130,
        "ordered_reservoir_points": 8000,
        "max_dims": 3,
        "radius_quantile": 0.24,
        "n_colored_noise_surrogates": 80,
        "beta_clip": (-0.5, 3.5),
        "betti_grid_size": 150,
        "landscape_grid_size": 150,
        "landscape_levels": 4,
        "top_lifetimes": 16,
        "cap_essential_h1": True,
        "plot_dpi": 150,
    },
    "balanced": {
        "max_points": 180,
        "ordered_reservoir_points": 10000,
        "max_dims": 3,
        "radius_quantile": 0.26,
        "n_colored_noise_surrogates": 120,
        "beta_clip": (-0.5, 3.5),
        "betti_grid_size": 180,
        "landscape_grid_size": 180,
        "landscape_levels": 5,
        "top_lifetimes": 20,
        "cap_essential_h1": True,
        "plot_dpi": 150,
    },
    "full": {
        "max_points": 260,
        "ordered_reservoir_points": 14000,
        "max_dims": 3,
        "radius_quantile": 0.28,
        "n_colored_noise_surrogates": 180,
        "beta_clip": (-0.5, 3.5),
        "betti_grid_size": 240,
        "landscape_grid_size": 240,
        "landscape_levels": 6,
        "top_lifetimes": 28,
        "cap_essential_h1": True,
        "plot_dpi": 150,
    },
}

if FAST_OPTION not in MODE_CONFIGS:
    raise ValueError("FAST_OPTION must be one of: ultra, fast, balanced, full")

cfg = MODE_CONFIGS[FAST_OPTION]

MAX_POINTS = cfg["max_points"]
ORDERED_RESERVOIR_POINTS = cfg["ordered_reservoir_points"]
MAX_DIMS = cfg["max_dims"]
RADIUS_QUANTILE = cfg["radius_quantile"]
N_COLORED_NOISE_SURROGATES = cfg["n_colored_noise_surrogates"]
BETA_CLIP = cfg["beta_clip"]
BETTI_GRID_SIZE = cfg["betti_grid_size"]
LANDSCAPE_GRID_SIZE = cfg["landscape_grid_size"]
LANDSCAPE_LEVELS = cfg["landscape_levels"]
TOP_LIFETIMES = cfg["top_lifetimes"]
CAP_ESSENTIAL_H1 = cfg["cap_essential_h1"]
PLOT_DPI = cfg["plot_dpi"]

out_dir = os.path.join(base_dir, "topology_colored_noise_surrogate_persistent_homology")
mode_out_dir = os.path.join(
    out_dir,
    f"h1_colored_noise_surrogate_ph_{FAST_OPTION}_fps_{COLORED_NOISE_SURROGATE_TYPE}_joblib_fft",
)
os.makedirs(mode_out_dir, exist_ok=True)

expert_dir = os.path.join(mode_out_dir, "expert_plots")
os.makedirs(expert_dir, exist_ok=True)

eeg_channels = [
    "Fp1", "Fpz", "Fp2", "F7", "F3", "Fz", "F4", "F8", "FC5", "FC1", "FC2", "FC6",
    "M1", "T7", "C3", "Cz", "C4", "T8", "M2", "CP5", "CP1", "CP2", "CP6",
    "P7", "P3", "Pz", "P4", "P8", "POz", "O1", "Oz", "O2"
]

if MAX_CHANNELS_TO_PROCESS is not None:
    eeg_channels = eeg_channels[:MAX_CHANNELS_TO_PROCESS]

channel_regions = {
    "Fp1": "Frontal", "Fpz": "Frontal", "Fp2": "Frontal",
    "F7": "Frontal", "F3": "Frontal", "Fz": "Frontal", "F4": "Frontal", "F8": "Frontal",
    "FC5": "Frontocentral", "FC1": "Frontocentral", "FC2": "Frontocentral", "FC6": "Frontocentral",
    "M1": "Mastoid", "M2": "Mastoid",
    "T7": "Central", "C3": "Central", "Cz": "Central", "C4": "Central", "T8": "Central",
    "CP5": "Centroparietal", "CP1": "Centroparietal", "CP2": "Centroparietal", "CP6": "Centroparietal",
    "P7": "Parietal", "P3": "Parietal", "Pz": "Parietal", "P4": "Parietal", "P8": "Parietal",
    "POz": "Occipital", "O1": "Occipital", "Oz": "Occipital", "O2": "Occipital",
}

BLACK = "#000000"
ACCENT = "#00FFFF"
ACCENT_SOFT = "#66FFFF"
ACCENT_DIM = "#008B8B"
CYAN_FILL = "#00D5FF"
CYAN_FILL_DARK = "#005A63"

CYAN_SEQUENTIAL_CMAP = LinearSegmentedColormap.from_list(
    "cyan_sequential",
    [BLACK, CYAN_FILL_DARK, CYAN_FILL, ACCENT_SOFT],
    N=256,
)

CYAN_DIVERGING_CMAP = LinearSegmentedColormap.from_list(
    "cyan_diverging",
    [ACCENT_SOFT, BLACK, CYAN_FILL],
    N=256,
)

plt.rcParams.update({
    "figure.facecolor": BLACK,
    "axes.facecolor": BLACK,
    "savefig.facecolor": BLACK,
    "text.color": ACCENT,
    "axes.labelcolor": ACCENT,
    "axes.edgecolor": ACCENT,
    "axes.titlecolor": ACCENT,
    "xtick.color": ACCENT,
    "ytick.color": ACCENT,
    "grid.color": ACCENT_DIM,
    "grid.alpha": 0.25,
    "legend.facecolor": BLACK,
    "legend.edgecolor": ACCENT,
    "legend.labelcolor": ACCENT,
    "font.size": 11,
})


def log(msg):
    if VERBOSE_PROGRESS:
        print(msg, flush=True)


def fmt_time(seconds):
    seconds = float(max(0, seconds))
    if seconds < 60:
        return f"{seconds:.0f}s"
    return f"{seconds / 60:.1f}min"


def style_ax(ax, title=None, xlabel=None, ylabel=None, grid=True):
    ax.set_facecolor(BLACK)
    for spine in ax.spines.values():
        spine.set_color(ACCENT)
        spine.set_linewidth(1.1)
    ax.tick_params(axis="x", colors=ACCENT, labelcolor=ACCENT)
    ax.tick_params(axis="y", colors=ACCENT, labelcolor=ACCENT)
    if title is not None:
        ax.set_title(title, color=ACCENT, pad=12)
    if xlabel is not None:
        ax.set_xlabel(xlabel, color=ACCENT, labelpad=10)
    if ylabel is not None:
        ax.set_ylabel(ylabel, color=ACCENT, labelpad=10)
    if grid:
        ax.grid(True, color=ACCENT_DIM, alpha=0.25, linewidth=0.8)


def force_cyan_text(ax):
    ax.title.set_color(ACCENT)
    ax.xaxis.label.set_color(ACCENT)
    ax.yaxis.label.set_color(ACCENT)
    for tick_label in ax.get_xticklabels():
        tick_label.set_color(ACCENT)
    for tick_label in ax.get_yticklabels():
        tick_label.set_color(ACCENT)


def style_colorbar(cbar, label):
    cbar.set_label(label, color=ACCENT)
    cbar.ax.yaxis.set_tick_params(color=ACCENT)
    plt.setp(cbar.ax.get_yticklabels(), color=ACCENT)
    cbar.outline.set_edgecolor(ACCENT)


def style_legend(legend):
    if legend is not None:
        legend.get_frame().set_facecolor(BLACK)
        legend.get_frame().set_edgecolor(ACCENT)
        for text in legend.get_texts():
            text.set_color(ACCENT)


def save_render_close(fig, path, width=INLINE_IMAGE_WIDTH):
    if SAVE_PLOTS_TO_DISK:
        fig.savefig(path, dpi=PLOT_DPI, bbox_inches="tight", facecolor=BLACK)
    plt.close(fig)
    if RENDER_ALL_IMAGES_INLINE and HAVE_IPYTHON and os.path.exists(path):
        display(Image(filename=path, width=width))


def offdiag_values(M):
    M = np.asarray(M, dtype=float)
    if M.shape[0] <= 1:
        return np.asarray([], dtype=float)
    mask = ~np.eye(M.shape[0], dtype=bool)
    vals = M[mask]
    vals = vals[np.isfinite(vals)]
    return vals


def dynamic_limits(values, lower_pct=2, upper_pct=98, pad_frac=0.08, lower_bound=None, upper_bound=None):
    values = np.asarray(values, dtype=float)
    values = values[np.isfinite(values)]

    if len(values) == 0:
        vmin, vmax = 0.0, 1.0
    else:
        vmin = float(np.percentile(values, lower_pct))
        vmax = float(np.percentile(values, upper_pct))
        if not np.isfinite(vmin):
            vmin = float(np.nanmin(values))
        if not np.isfinite(vmax):
            vmax = float(np.nanmax(values))
        if vmax <= vmin:
            center = float(np.nanmean(values))
            eps = max(1e-6, abs(center) * 0.01, 1e-5)
            vmin = center - eps
            vmax = center + eps

    pad = (vmax - vmin) * pad_frac
    vmin -= pad
    vmax += pad

    if lower_bound is not None:
        vmin = max(lower_bound, vmin)
    if upper_bound is not None:
        vmax = min(upper_bound, vmax)
    if vmax <= vmin:
        vmax = vmin + 1e-6

    return vmin, vmax


def robust_limits(values, lower_pct=2, upper_pct=98, pad_frac=0.08, include_zero=True):
    values = np.asarray(values, dtype=float)
    values = values[np.isfinite(values)]

    if len(values) == 0:
        return 0.0, 1.0

    vmin = float(np.nanpercentile(values, lower_pct))
    vmax = float(np.nanpercentile(values, upper_pct))

    if include_zero:
        vmin = min(vmin, 0.0)
        vmax = max(vmax, 0.0)

    if vmax <= vmin:
        center = float(np.nanmean(values))
        eps = max(abs(center) * 0.05, 1e-6)
        vmin = center - eps
        vmax = center + eps

    pad = (vmax - vmin) * pad_frac
    return vmin - pad, vmax + pad


def safe_entropy(values):
    values = np.asarray(values, dtype=float)
    values = values[np.isfinite(values) & (values > 0)]
    if len(values) == 0:
        return 0.0
    total = float(np.sum(values))
    if total <= 0:
        return 0.0
    p = values / (total + 1e-12)
    return float(-np.sum(p * np.log(p + 1e-12)))


def percentile_score(values, higher_is_more=True):
    s = pd.Series(values, dtype=float)
    if not higher_is_more:
        s = -s
    if s.nunique(dropna=True) <= 1:
        return np.full(len(s), 50.0, dtype=float)
    return s.rank(method="average", pct=True).to_numpy(dtype=float) * 100.0


def bh_fdr(pvals):
    pvals = np.asarray(pvals, dtype=float)
    qvals = np.full_like(pvals, np.nan, dtype=float)

    valid = np.isfinite(pvals)
    p = pvals[valid]

    if len(p) == 0:
        return qvals

    order = np.argsort(p)
    ranked = p[order]
    m = len(ranked)
    raw_q = ranked * m / np.arange(1, m + 1)
    monotone_q = np.minimum.accumulate(raw_q[::-1])[::-1]
    monotone_q = np.clip(monotone_q, 0.0, 1.0)

    q_valid = np.empty_like(monotone_q)
    q_valid[order] = monotone_q
    qvals[valid] = q_valid

    return qvals


def robust_metric_contrast_matrix(raw_z_matrix):
    raw_z_matrix = np.asarray(raw_z_matrix, dtype=float)
    out = np.zeros_like(raw_z_matrix, dtype=float)
    info = []

    for j in range(raw_z_matrix.shape[1]):
        col = raw_z_matrix[:, j]
        finite = col[np.isfinite(col)]

        if len(finite) == 0:
            out[:, j] = 0.0
            info.append({"metric_index": j, "center_regularized_z": np.nan, "contrast_scale": np.nan})
            continue

        center = float(np.nanmedian(finite))
        mad = float(np.nanmedian(np.abs(finite - center)))
        scale = 1.4826 * mad

        if not np.isfinite(scale) or scale < 1e-9:
            p10 = float(np.nanpercentile(finite, 10))
            p90 = float(np.nanpercentile(finite, 90))
            scale = (p90 - p10) / 2.563

        if not np.isfinite(scale) or scale < 1e-9:
            scale = float(np.nanstd(finite))

        if not np.isfinite(scale) or scale < 1e-9:
            scale = 1.0

        out[:, j] = (col - center) / scale

        info.append({
            "metric_index": j,
            "center_regularized_z": center,
            "contrast_scale": scale,
        })

    return out, pd.DataFrame(info)


def estimate_median_pairwise_distance(X, max_pairs=120000, seed=7):
    X = np.asarray(X, dtype=float)
    n = X.shape[0]
    if n < 2:
        return 1.0

    rng = np.random.default_rng(seed)

    if n * (n - 1) // 2 <= max_pairs:
        D = pdist(X)
        D = D[D > 0]
        if len(D) == 0:
            return 1.0
        return float(np.nanmedian(D))

    i = rng.integers(0, n, size=max_pairs)
    j = rng.integers(0, n, size=max_pairs)
    mask = i != j
    i = i[mask]
    j = j[mask]

    if len(i) == 0:
        return 1.0

    d = np.linalg.norm(X[i] - X[j], axis=1)
    d = d[np.isfinite(d) & (d > 0)]

    if len(d) == 0:
        return 1.0

    return float(np.nanmedian(d))


def standardize_matrix_for_geometry(X, max_dims=3, seed=7):
    X = np.asarray(X, dtype=np.float64)
    X = X[np.all(np.isfinite(X), axis=1)]

    if X.shape[0] < 10:
        return None

    X = X - np.mean(X, axis=0, keepdims=True)

    if X.shape[1] > max_dims:
        _U, _S, Vt = np.linalg.svd(X, full_matrices=False)
        X = X @ Vt[:max_dims].T

    X = X / (np.std(X, axis=0, keepdims=True) + 1e-12)
    scale = estimate_median_pairwise_distance(X, seed=seed)
    X = X / (scale + 1e-12)

    return X


def standardize_point_cloud_small(X, max_dims=3):
    X = np.asarray(X, dtype=np.float64)
    X = X[np.all(np.isfinite(X), axis=1)]

    if X.shape[0] < 10:
        return None

    X = X - np.mean(X, axis=0, keepdims=True)

    if X.shape[1] > max_dims:
        _U, _S, Vt = np.linalg.svd(X, full_matrices=False)
        X = X @ Vt[:max_dims].T

    X = X / (np.std(X, axis=0, keepdims=True) + 1e-12)

    D = pdist(X)
    positive = D[D > 0]

    if len(positive) > 0:
        X = X / (np.nanmedian(positive) + 1e-12)

    return X


def farthest_point_sample(X, n_samples, seed=7):
    X = np.asarray(X, dtype=float)
    n = X.shape[0]

    if n <= n_samples:
        return np.arange(n)

    rng = np.random.default_rng(seed)
    center = np.mean(X, axis=0, keepdims=True)
    dist_to_center = np.linalg.norm(X - center, axis=1)
    start_idx = int(np.nanargmax(dist_to_center))

    selected = np.empty(n_samples, dtype=int)
    selected.fill(-1)
    selected[0] = start_idx

    min_dist2 = np.full(n, np.inf, dtype=float)

    for k in range(1, n_samples):
        last = X[selected[k - 1]]
        d2 = np.sum((X - last) ** 2, axis=1)
        min_dist2 = np.minimum(min_dist2, d2)
        min_dist2[selected[:k]] = -np.inf

        next_idx = int(np.nanargmax(min_dist2))

        if not np.isfinite(min_dist2[next_idx]) or min_dist2[next_idx] < 1e-18:
            remaining = np.setdiff1d(np.arange(n), selected[:k], assume_unique=False)

            if len(remaining) == 0:
                selected = selected[:k]
                break

            next_idx = int(rng.choice(remaining))

        selected[k] = next_idx

    selected = selected[selected >= 0]

    return selected


def load_channel_embedding_ordered_reservoir(ch_name, ordered_reservoir_points=5000, max_dims=3, seed=7):
    candidate_paths = [
        os.path.join(base_dir, "embedding_data", "tsne_umap_from_3d", f"umap_embedding_{ch_name}.npy"),
        os.path.join(base_dir, "embedding_data", "tsne_umap_from_3d", f"tsne_embedding_{ch_name}.npy"),
        os.path.join(base_dir, "embedding_data", "3dembedding_data", f"3dembedded_{ch_name}.npy"),
        os.path.join(base_dir, "embedding_data", "2dembedding_data", f"2dembedded_{ch_name}.npy"),
    ]

    for path in candidate_paths:
        if not os.path.exists(path):
            continue

        log(f"    trying: {path}")

        try:
            X = np.load(path, mmap_mode="r", allow_pickle=False)
        except Exception as exc:
            log(f"    skipped non-memmappable or nonnumeric file: {os.path.basename(path)} | {exc}")
            continue

        if X.ndim != 2:
            log(f"    skipped because ndim={X.ndim}")
            continue

        original_shape = tuple(X.shape)

        try:
            if X.shape[0] >= X.shape[1]:
                n_samples = X.shape[0]
                n_features = X.shape[1]

                if n_samples < 10 or n_features < 2:
                    log(f"    skipped bad shape: {original_shape}")
                    continue

                reservoir_n = min(ordered_reservoir_points, n_samples)
                ordered_idx = np.unique(np.round(np.linspace(0, n_samples - 1, reservoir_n)).astype(int))
                X_ordered = np.asarray(X[ordered_idx, :], dtype=np.float64)

            else:
                n_samples = X.shape[1]
                n_features = X.shape[0]

                if n_samples < 10 or n_features < 2:
                    log(f"    skipped bad transposed shape: {original_shape}")
                    continue

                reservoir_n = min(ordered_reservoir_points, n_samples)
                ordered_idx = np.unique(np.round(np.linspace(0, n_samples - 1, reservoir_n)).astype(int))
                X_ordered = np.asarray(X[:, ordered_idx].T, dtype=np.float64)

        except Exception as exc:
            log(f"    failed while ordered-reservoir loading {os.path.basename(path)} | {exc}")
            continue

        X_ordered = X_ordered[np.all(np.isfinite(X_ordered), axis=1)]

        if X_ordered.shape[0] < 10 or X_ordered.shape[1] < 2:
            log(f"    ordered reservoir data unusable: shape={X_ordered.shape}")
            continue

        X_ordered_std = standardize_matrix_for_geometry(
            X_ordered,
            max_dims=max_dims,
            seed=seed,
        )

        if X_ordered_std is None:
            log("    standardized ordered reservoir unusable")
            continue

        return X_ordered_std, path, original_shape, tuple(X_ordered.shape), "ordered_time_thinned_reservoir"

    return None, None, None, None, None


def scipy_fft_rfft(X, axis=0, workers=1):
    try:
        with sp_fft.set_workers(workers):
            return sp_fft.rfft(X, axis=axis, workers=workers)
    except TypeError:
        try:
            with sp_fft.set_workers(workers):
                return sp_fft.rfft(X, axis=axis)
        except Exception:
            return np.fft.rfft(X, axis=axis)
    except Exception:
        return np.fft.rfft(X, axis=axis)


def scipy_fft_irfft(F, n, axis=0, workers=1):
    try:
        with sp_fft.set_workers(workers):
            return sp_fft.irfft(F, n=n, axis=axis, workers=workers)
    except TypeError:
        try:
            with sp_fft.set_workers(workers):
                return sp_fft.irfft(F, n=n, axis=axis)
        except Exception:
            return np.fft.irfft(F, n=n, axis=axis)
    except Exception:
        return np.fft.irfft(F, n=n, axis=axis)


def estimate_colored_noise_beta(x, fft_workers=1, beta_clip=(-0.5, 3.5)):
    x = np.asarray(x, dtype=float)
    x = x[np.isfinite(x)]

    if len(x) < 16:
        return 0.0

    x = x - np.mean(x)

    F = scipy_fft_rfft(x, axis=0, workers=fft_workers)
    power = np.abs(F) ** 2
    freqs = np.fft.rfftfreq(len(x), d=1.0)

    mask = (freqs > 0) & np.isfinite(power) & (power > 1e-18)

    if np.sum(mask) < 8:
        return 0.0

    log_f = np.log(freqs[mask])
    log_p = np.log(power[mask])

    lo = int(0.05 * len(log_f))
    hi = int(0.95 * len(log_f))

    if hi <= lo + 3:
        lo = 0
        hi = len(log_f)

    log_f = log_f[lo:hi]
    log_p = log_p[lo:hi]

    if len(log_f) < 8:
        return 0.0

    slope, intercept = np.polyfit(log_f, log_p, 1)
    beta = float(-slope)

    if not np.isfinite(beta):
        beta = 0.0

    beta = float(np.clip(beta, beta_clip[0], beta_clip[1]))

    return beta


def generate_colored_noise_1d(n, beta, rng, fft_workers=1):
    if n < 8:
        return rng.normal(size=n)

    freqs = np.fft.rfftfreq(n, d=1.0)
    amp = np.ones_like(freqs, dtype=float)

    positive = freqs > 0
    amp[positive] = freqs[positive] ** (-beta / 2.0)
    amp[0] = 0.0

    phases = rng.uniform(0.0, 2.0 * np.pi, size=len(freqs))
    spectrum = amp * np.exp(1j * phases)
    spectrum[0] = 0.0 + 0.0j

    if n % 2 == 0 and len(spectrum) > 1:
        spectrum[-1] = spectrum[-1].real + 0.0j

    y = scipy_fft_irfft(spectrum, n=n, axis=0, workers=fft_workers)
    y = np.asarray(y, dtype=float)

    y = y - np.mean(y)
    y = y / (np.std(y) + 1e-12)

    return y


def colored_noise_ordered_cloud(X_ordered, seed, fft_workers=1):
    rng = np.random.default_rng(seed)

    X_ordered = np.asarray(X_ordered, dtype=float)
    X_ordered = X_ordered[np.all(np.isfinite(X_ordered), axis=1)]

    if X_ordered.shape[0] < 10:
        return None, np.asarray([], dtype=float)

    n, d = X_ordered.shape
    S = np.zeros_like(X_ordered, dtype=float)
    betas = np.zeros(d, dtype=float)

    for dim in range(d):
        beta = estimate_colored_noise_beta(
            X_ordered[:, dim],
            fft_workers=fft_workers,
            beta_clip=BETA_CLIP,
        )
        betas[dim] = beta

        S[:, dim] = generate_colored_noise_1d(
            n=n,
            beta=beta,
            rng=rng,
            fft_workers=fft_workers,
        )

        S[:, dim] = S[:, dim] * (np.std(X_ordered[:, dim]) + 1e-12) + np.mean(X_ordered[:, dim])

    S = standardize_matrix_for_geometry(
        S,
        max_dims=MAX_DIMS,
        seed=int(rng.integers(0, 2**31 - 1)),
    )

    if S is None:
        S = rng.normal(size=X_ordered.shape)
        S = standardize_matrix_for_geometry(
            S,
            max_dims=MAX_DIMS,
            seed=int(rng.integers(0, 2**31 - 1)),
        )

    return S, betas


def sample_geometry_from_ordered_cloud(X_ordered, max_points=90, seed=7):
    if X_ordered is None:
        return None, None

    X_ordered = np.asarray(X_ordered, dtype=float)
    X_ordered = X_ordered[np.all(np.isfinite(X_ordered), axis=1)]

    if X_ordered.shape[0] < 10:
        return None, None

    if USE_GEOMETRY_PRESERVING_SUBSAMPLE:
        idx = farthest_point_sample(
            X_ordered,
            n_samples=min(max_points, X_ordered.shape[0]),
            seed=seed,
        )
        X_small = X_ordered[idx]
        method = "ordered_reservoir_plus_farthest_point"
    else:
        rng = np.random.default_rng(seed)
        sample_n = min(max_points, X_ordered.shape[0])
        idx = np.sort(rng.choice(X_ordered.shape[0], size=sample_n, replace=False))
        X_small = X_ordered[idx]
        method = "ordered_reservoir_random"

    X_small = standardize_point_cloud_small(X_small, max_dims=MAX_DIMS)

    return X_small, method


def estimate_complex_counts(X, radius):
    D = squareform(pdist(X))
    A = (D <= radius) & (D > 0)
    n_edges = int(np.sum(A) // 2)

    if A.shape[0] <= 300:
        A_int = A.astype(np.int64)
        n_triangles = int(np.trace(A_int @ A_int @ A_int) // 6)
    else:
        n_triangles = np.nan

    return n_edges, n_triangles


def run_h1_ripser(X, radius):
    result = ripser(
        X,
        maxdim=1,
        thresh=float(radius),
        coeff=2,
        do_cocycles=False,
    )

    diagrams = result["dgms"]

    h0 = diagrams[0] if len(diagrams) > 0 else np.empty((0, 2), dtype=float)
    h1 = diagrams[1] if len(diagrams) > 1 else np.empty((0, 2), dtype=float)

    return h0, h1


def split_finite_essential(diagram):
    diagram = np.asarray(diagram, dtype=float)

    if diagram.size == 0:
        return np.empty((0, 2), dtype=float), np.empty((0, 2), dtype=float)

    finite = diagram[
        np.isfinite(diagram[:, 0])
        & np.isfinite(diagram[:, 1])
        & (diagram[:, 1] > diagram[:, 0])
    ]

    essential = diagram[
        np.isfinite(diagram[:, 0])
        & np.isinf(diagram[:, 1])
    ]

    return finite, essential


def clean_h1_diagram(finite_intervals, essential_intervals=None, cap_essential=True, cap_value=None):
    cleaned = []
    capped_count = 0

    if finite_intervals is not None:
        for b, d in finite_intervals:
            if np.isfinite(b) and np.isfinite(d) and d > b:
                cleaned.append((float(b), float(d)))

    if cap_essential and essential_intervals is not None and cap_value is not None:
        for b, _d in essential_intervals:
            if np.isfinite(b) and np.isfinite(cap_value) and cap_value > b:
                cleaned.append((float(b), float(cap_value)))
                capped_count += 1

    if len(cleaned) == 0:
        return np.empty((0, 2), dtype=float), capped_count

    return np.asarray(cleaned, dtype=float), capped_count


def betti_curve(diagram, grid):
    diagram = np.asarray(diagram, dtype=float)
    curve = np.zeros(len(grid), dtype=float)

    if diagram.size == 0 or len(diagram) == 0:
        return curve

    births = diagram[:, 0]
    deaths = diagram[:, 1]

    for b, d in zip(births, deaths):
        curve += ((grid >= b) & (grid < d)).astype(float)

    return curve


def persistence_landscapes(diagram, grid, n_levels=4):
    diagram = np.asarray(diagram, dtype=float)
    landscapes = np.zeros((n_levels, len(grid)), dtype=float)

    if diagram.size == 0 or len(diagram) == 0:
        return landscapes

    tents = []

    for b, d in diagram:
        if not np.isfinite(b) or not np.isfinite(d) or d <= b:
            continue

        tent = np.maximum(0.0, np.minimum(grid - b, d - grid))
        tents.append(tent)

    if len(tents) == 0:
        return landscapes

    tent_matrix = np.vstack(tents)

    for idx in range(len(grid)):
        vals = np.sort(tent_matrix[:, idx])[::-1]

        for level in range(min(n_levels, len(vals))):
            landscapes[level, idx] = vals[level]

    return landscapes


def top_lifetime_values(diagram, top_k=12):
    diagram = np.asarray(diagram, dtype=float)
    out = np.zeros(top_k, dtype=float)

    if diagram.size == 0 or len(diagram) == 0:
        return out

    lifetimes = diagram[:, 1] - diagram[:, 0]
    lifetimes = lifetimes[np.isfinite(lifetimes) & (lifetimes > 0)]
    lifetimes = np.sort(lifetimes)[::-1]

    take = min(top_k, len(lifetimes))
    out[:take] = lifetimes[:take]

    return out


def h1_metric_features(diagram, betti_grid, landscape_grid, n_landscape_levels=4, top_k=12):
    diagram = np.asarray(diagram, dtype=float)

    if diagram.size == 0 or len(diagram) == 0:
        lifetimes = np.asarray([], dtype=float)
    else:
        lifetimes = diagram[:, 1] - diagram[:, 0]
        lifetimes = lifetimes[np.isfinite(lifetimes) & (lifetimes > 0)]

    curve = betti_curve(diagram, betti_grid)
    landscapes = persistence_landscapes(diagram, landscape_grid, n_levels=n_landscape_levels)

    if len(lifetimes) == 0:
        lifetime_mean = 0.0
        lifetime_std = 0.0
        lifetime_max = 0.0
        lifetime_median = 0.0
        lifetime_q75 = 0.0
        lifetime_q90 = 0.0
        total_lifetime = 0.0
        lifetime_entropy = 0.0
    else:
        lifetime_mean = float(np.nanmean(lifetimes))
        lifetime_std = float(np.nanstd(lifetimes))
        lifetime_max = float(np.nanmax(lifetimes))
        lifetime_median = float(np.nanmedian(lifetimes))
        lifetime_q75 = float(np.nanpercentile(lifetimes, 75))
        lifetime_q90 = float(np.nanpercentile(lifetimes, 90))
        total_lifetime = float(np.nansum(lifetimes))
        lifetime_entropy = safe_entropy(lifetimes)

    betti_auc = float(np.trapz(curve, betti_grid)) if len(curve) else 0.0
    betti_l2 = float(np.sqrt(np.trapz(curve ** 2, betti_grid))) if len(curve) else 0.0
    betti_max = float(np.nanmax(curve)) if len(curve) else 0.0
    betti_mean = float(np.nanmean(curve)) if len(curve) else 0.0
    betti_nonzero_fraction = float(np.mean(curve > 0)) if len(curve) else 0.0

    landscape_l1 = float(np.sum(np.trapz(np.abs(landscapes), landscape_grid, axis=1))) if landscapes.size else 0.0
    landscape_level_l2 = np.sqrt(np.trapz(landscapes ** 2, landscape_grid, axis=1)) if landscapes.size else np.zeros(n_landscape_levels)
    landscape_l2 = float(np.sqrt(np.sum(landscape_level_l2 ** 2)))
    landscape_max = float(np.nanmax(landscapes)) if landscapes.size else 0.0

    top_vals = top_lifetime_values(diagram, top_k=top_k)

    features = {
        "n_h1_intervals": float(len(lifetimes)),
        "total_lifetime": total_lifetime,
        "max_lifetime": lifetime_max,
        "mean_lifetime": lifetime_mean,
        "median_lifetime": lifetime_median,
        "q75_lifetime": lifetime_q75,
        "q90_lifetime": lifetime_q90,
        "std_lifetime": lifetime_std,
        "lifetime_entropy": lifetime_entropy,
        "betti_auc": betti_auc,
        "betti_l2": betti_l2,
        "betti_max": betti_max,
        "betti_mean": betti_mean,
        "betti_nonzero_fraction": betti_nonzero_fraction,
        "landscape_l1": landscape_l1,
        "landscape_l2": landscape_l2,
        "landscape_max": landscape_max,
    }

    for idx, val in enumerate(top_vals, start=1):
        features[f"top_lifetime_{idx:02d}"] = float(val)

    return features, curve, landscapes


def classical_mds_from_distances(D, n_components=2):
    D = np.asarray(D, dtype=float)
    n = D.shape[0]

    if n < 2:
        return np.zeros((n, n_components), dtype=float)

    D2 = D ** 2
    J = np.eye(n) - np.ones((n, n)) / n
    B = -0.5 * J @ D2 @ J

    if HAVE_SCIPY_EIGH:
        vals, vecs = eigh(B)
    else:
        vals, vecs = np.linalg.eigh(B)

    idx = np.argsort(vals)[::-1]
    vals = vals[idx]
    vecs = vecs[:, idx]

    coords = np.zeros((n, n_components), dtype=float)

    for k in range(min(n_components, n)):
        if vals[k] > 0:
            coords[:, k] = vecs[:, k] * np.sqrt(vals[k])

    return coords


def gaussian_similarity_from_distance(D, sigma=None):
    D = np.asarray(D, dtype=float)
    off = offdiag_values(D)

    if sigma is None:
        sigma = float(np.nanmedian(off)) if len(off) else 1.0

    sigma = max(float(sigma), 1e-8)

    S = np.exp(-(D ** 2) / (2.0 * sigma ** 2))
    np.fill_diagonal(S, 1.0)

    return S, sigma


def compute_regularized_z(summary_df, long_df, primary_metric_names, prefix):
    fixed_z = pd.DataFrame(index=summary_df["channel"].astype(str))
    info_rows = []

    for metric in primary_metric_names:
        obs_col = f"observed_{metric}"
        mean_col = f"{prefix}_mean_{metric}"
        std_col = f"{prefix}_std_{metric}"
        raw_z_col = f"{prefix}_z_{metric}"

        obs = summary_df[obs_col].to_numpy(dtype=float)
        null_mean = summary_df[mean_col].to_numpy(dtype=float)

        if std_col in summary_df.columns:
            null_std = summary_df[std_col].to_numpy(dtype=float)
        else:
            null_std = np.zeros(len(summary_df), dtype=float)

        delta = obs - null_mean

        finite_std = null_std[np.isfinite(null_std) & (null_std > 0)]
        finite_delta = np.abs(delta[np.isfinite(delta)])

        null_vals = np.asarray([], dtype=float)

        if len(long_df) > 0 and metric in long_df.columns:
            null_vals = long_df[metric].to_numpy(dtype=float)
            null_vals = null_vals[np.isfinite(null_vals)]

        floors = []

        if len(finite_std) > 0:
            floors.append(float(np.nanmedian(finite_std)) * 0.10)
            floors.append(float(np.nanpercentile(finite_std, 25)) * 0.25)

        if len(finite_delta) > 0:
            floors.append(float(np.nanmedian(finite_delta)) * 0.05)
            floors.append(float(np.nanpercentile(finite_delta, 75)) * 0.025)

        if len(null_vals) > 1:
            floors.append(float(np.nanstd(null_vals)) * 0.03)
            floors.append(float(np.nanmedian(np.abs(null_vals - np.nanmedian(null_vals)))) * 0.10)

        if np.any(np.isfinite(obs)):
            floors.append(float(np.nanmedian(np.abs(obs[np.isfinite(obs)]))) * 1e-6)

        if np.any(np.isfinite(null_mean)):
            floors.append(float(np.nanmedian(np.abs(null_mean[np.isfinite(null_mean)]))) * 1e-6)

        floors.append(1e-6)
        floors = [v for v in floors if np.isfinite(v) and v > 0]

        std_floor = max(floors) if len(floors) else 1e-6
        z_fixed = delta / np.maximum(null_std, std_floor)

        raw_exploded_count = 0

        if raw_z_col in summary_df.columns:
            raw_z = summary_df[raw_z_col].to_numpy(dtype=float)
            raw_exploded_count = int(np.sum(np.isfinite(raw_z) & (np.abs(raw_z) > 1000)))

        fixed_z[f"{prefix}_z_regularized_{metric}"] = z_fixed

        info_rows.append({
            "metric": metric,
            "std_floor_used": std_floor,
            "raw_exploded_z_count": raw_exploded_count,
            "regularized_z_min": float(np.nanmin(z_fixed)) if np.any(np.isfinite(z_fixed)) else np.nan,
            "regularized_z_max": float(np.nanmax(z_fixed)) if np.any(np.isfinite(z_fixed)) else np.nan,
        })

    return fixed_z, pd.DataFrame(info_rows)


def colored_noise_surrogate_worker(
    surrogate_idx,
    ch_name,
    region,
    X_ordered,
    radius,
    betti_grid,
    landscape_grid,
    seed,
    fft_workers,
):
    rng = np.random.default_rng(seed)

    try:
        S_ordered, betas = colored_noise_ordered_cloud(
            X_ordered,
            seed=seed,
            fft_workers=fft_workers,
        )

        S, s_sampling_method = sample_geometry_from_ordered_cloud(
            S_ordered,
            max_points=MAX_POINTS,
            seed=int(rng.integers(0, 2**31 - 1)),
        )

        if S is None:
            raise RuntimeError("colored-noise sampled cloud was None")

        s_h0_diag, s_h1_diag = run_h1_ripser(S, radius)
        s_h1_finite, s_h1_essential = split_finite_essential(s_h1_diag)

        s_diagram, s_capped = clean_h1_diagram(
            s_h1_finite,
            essential_intervals=s_h1_essential,
            cap_essential=CAP_ESSENTIAL_H1,
            cap_value=radius,
        )

        s_features, _s_betti, _s_landscapes = h1_metric_features(
            s_diagram,
            betti_grid=betti_grid,
            landscape_grid=landscape_grid,
            n_landscape_levels=LANDSCAPE_LEVELS,
            top_k=TOP_LIFETIMES,
        )

        s_n_edges, s_n_triangles = estimate_complex_counts(S, radius)

        beta_mean = float(np.nanmean(betas)) if len(betas) else np.nan
        beta_std = float(np.nanstd(betas)) if len(betas) else np.nan
        beta_min = float(np.nanmin(betas)) if len(betas) else np.nan
        beta_max = float(np.nanmax(betas)) if len(betas) else np.nan

    except Exception:
        s_features = {metric: 0.0 for metric in metric_names}

        for k in range(1, TOP_LIFETIMES + 1):
            s_features[f"top_lifetime_{k:02d}"] = 0.0

        s_capped = 0
        s_n_edges = np.nan
        s_n_triangles = np.nan
        s_sampling_method = "failed"
        S = None

        beta_mean = np.nan
        beta_std = np.nan
        beta_min = np.nan
        beta_max = np.nan

    s_row = {
        "channel": ch_name,
        "region": region,
        "colored_noise_surrogate_id": int(surrogate_idx + 1),
        "colored_noise_surrogate_type": COLORED_NOISE_SURROGATE_TYPE,
        "radius": float(radius),
        "n_points": int(S.shape[0]) if S is not None else np.nan,
        "ordered_reservoir_points": int(X_ordered.shape[0]),
        "sampling_method": s_sampling_method,
        "n_edges_at_radius": int(s_n_edges) if np.isfinite(s_n_edges) else np.nan,
        "n_triangles_at_radius": int(s_n_triangles) if np.isfinite(s_n_triangles) else np.nan,
        "n_h1_capped": int(s_capped),
        "colored_noise_beta_mean": beta_mean,
        "colored_noise_beta_std": beta_std,
        "colored_noise_beta_min": beta_min,
        "colored_noise_beta_max": beta_max,
    }

    for key, val in s_features.items():
        s_row[key] = float(val)

    return s_row, s_features


metric_names = [
    "n_h1_intervals",
    "total_lifetime",
    "max_lifetime",
    "mean_lifetime",
    "median_lifetime",
    "q75_lifetime",
    "q90_lifetime",
    "std_lifetime",
    "lifetime_entropy",
    "betti_auc",
    "betti_l2",
    "betti_max",
    "betti_mean",
    "betti_nonzero_fraction",
    "landscape_l1",
    "landscape_l2",
    "landscape_max",
]

primary_metric_names = [
    "total_lifetime",
    "max_lifetime",
    "q90_lifetime",
    "betti_auc",
    "betti_l2",
    "landscape_l2",
    "landscape_max",
]

primary_metric_labels = [
    "Total lifetime",
    "Max lifetime",
    "Q90 lifetime",
    "Betti AUC",
    "Betti L2",
    "Landscape L2",
    "Landscape max",
]

log("\n" + "=" * 72)
log(f"Running H1 Colored-Noise-Surrogate Persistent Homology | {FAST_OPTION.upper()} | joblib + scipy.fft")
log(f"Channels: {len(eeg_channels)}")
log(f"Colored-noise surrogate type: {COLORED_NOISE_SURROGATE_TYPE}")
log(f"Colored-noise surrogates per channel: {N_COLORED_NOISE_SURROGATES}")
log(f"Beta clip: {BETA_CLIP}")
log(f"joblib backend={JOBLIB_BACKEND}, n_jobs={JOBLIB_N_JOBS}")
log(f"scipy.fft workers per job={FFT_WORKERS_PER_JOB}")
log(f"max_points={MAX_POINTS}, ordered_reservoir_points={ORDERED_RESERVOIR_POINTS}, radius_quantile={RADIUS_QUANTILE}")
log(f"Output: {mode_out_dir}")
log("=" * 72)

overall_t0 = time.perf_counter()

observed_rows = []
colored_noise_rows = []
diagnostic_rows = []

diagram_store = {}
colored_noise_metric_store = {}
betti_curve_store = {}
landscape_store = {}

channels_completed = []
per_channel_times = []

for ch_idx, ch_name in enumerate(eeg_channels, start=1):
    ch_t0 = time.perf_counter()
    rng = np.random.default_rng(31000 + ch_idx)
    prefix = f"[{ch_idx:02d}/{len(eeg_channels)} {ch_name}]"

    log(f"\n{prefix} loading ordered reservoir for colored-noise null")

    X_ordered, source_path, original_shape, ordered_reservoir_shape, ordered_method = load_channel_embedding_ordered_reservoir(
        ch_name,
        ordered_reservoir_points=ORDERED_RESERVOIR_POINTS,
        max_dims=MAX_DIMS,
        seed=32000 + ch_idx,
    )

    if X_ordered is None:
        log(f"{prefix} WARNING: no usable embedding found; skipping")
        continue

    X, sampling_method = sample_geometry_from_ordered_cloud(
        X_ordered,
        max_points=MAX_POINTS,
        seed=33000 + ch_idx,
    )

    if X is None:
        log(f"{prefix} WARNING: invalid observed sampled embedding; skipping")
        continue

    D_flat = pdist(X)
    positive = D_flat[D_flat > 0]

    if len(positive) == 0:
        log(f"{prefix} WARNING: zero-distance observed cloud; skipping")
        continue

    radius = float(np.quantile(positive, RADIUS_QUANTILE))
    betti_grid = np.linspace(0.0, max(radius, 1e-8), BETTI_GRID_SIZE)
    landscape_grid = np.linspace(0.0, max(radius, 1e-8), LANDSCAPE_GRID_SIZE)

    n_edges, n_triangles = estimate_complex_counts(X, radius)

    log(f"{prefix} observed PH | radius={radius:.4f}, points={X.shape[0]}, edges={n_edges}, triangles={n_triangles}")

    h0_diag, h1_diag = run_h1_ripser(X, radius)
    h1_finite, h1_essential = split_finite_essential(h1_diag)

    observed_h1_diagram, n_capped_h1 = clean_h1_diagram(
        h1_finite,
        essential_intervals=h1_essential,
        cap_essential=CAP_ESSENTIAL_H1,
        cap_value=radius,
    )

    observed_features, observed_betti, observed_landscapes = h1_metric_features(
        observed_h1_diagram,
        betti_grid=betti_grid,
        landscape_grid=landscape_grid,
        n_landscape_levels=LANDSCAPE_LEVELS,
        top_k=TOP_LIFETIMES,
    )

    observed_betas = np.asarray([
        estimate_colored_noise_beta(X_ordered[:, dim], fft_workers=FFT_WORKERS_PER_JOB, beta_clip=BETA_CLIP)
        for dim in range(X_ordered.shape[1])
    ], dtype=float)

    observed_row = {
        "channel": ch_name,
        "region": channel_regions.get(ch_name, "Unknown"),
        "source_path": source_path,
        "source_shape": str(original_shape),
        "ordered_reservoir_shape": str(ordered_reservoir_shape),
        "ordered_reservoir_method": ordered_method,
        "sampling_method": sampling_method,
        "n_points": int(X.shape[0]),
        "ordered_reservoir_points_used": int(X_ordered.shape[0]),
        "radius": radius,
        "n_edges_at_radius": int(n_edges),
        "n_triangles_at_radius": int(n_triangles) if np.isfinite(n_triangles) else np.nan,
        "n_h1_finite_observed": int(len(h1_finite)),
        "n_h1_essential_observed": int(len(h1_essential)),
        "n_h1_capped_observed": int(n_capped_h1),
        "observed_colored_noise_beta_mean": float(np.nanmean(observed_betas)) if len(observed_betas) else np.nan,
        "observed_colored_noise_beta_std": float(np.nanstd(observed_betas)) if len(observed_betas) else np.nan,
        "observed_colored_noise_beta_min": float(np.nanmin(observed_betas)) if len(observed_betas) else np.nan,
        "observed_colored_noise_beta_max": float(np.nanmax(observed_betas)) if len(observed_betas) else np.nan,
    }

    for key, val in observed_features.items():
        observed_row[f"observed_{key}"] = float(val)

    surrogate_seeds = [int(rng.integers(0, 2**31 - 1)) for _ in range(N_COLORED_NOISE_SURROGATES)]

    log(f"{prefix} launching {N_COLORED_NOISE_SURROGATES} colored-noise surrogate PH jobs on {JOBLIB_N_JOBS} workers")

    parallel_results = Parallel(
        n_jobs=JOBLIB_N_JOBS,
        backend=JOBLIB_BACKEND,
        verbose=0,
        batch_size=1,
        pre_dispatch=f"{2 * JOBLIB_N_JOBS}",
    )(
        delayed(colored_noise_surrogate_worker)(
            surrogate_idx,
            ch_name,
            channel_regions.get(ch_name, "Unknown"),
            X_ordered,
            radius,
            betti_grid,
            landscape_grid,
            surrogate_seeds[surrogate_idx],
            FFT_WORKERS_PER_JOB,
        )
        for surrogate_idx in range(N_COLORED_NOISE_SURROGATES)
    )

    colored_noise_rows_for_channel = [row for row, _features in parallel_results]
    colored_noise_feature_rows_for_channel = [features for _row, features in parallel_results]

    colored_noise_rows.extend(colored_noise_rows_for_channel)
    colored_noise_metric_df = pd.DataFrame(colored_noise_feature_rows_for_channel)

    for metric in metric_names:
        obs = float(observed_features.get(metric, 0.0))

        if metric not in colored_noise_metric_df.columns:
            s_vals = np.zeros(N_COLORED_NOISE_SURROGATES, dtype=float)
        else:
            s_vals = colored_noise_metric_df[metric].to_numpy(dtype=float)

        s_vals = s_vals[np.isfinite(s_vals)]

        if len(s_vals) == 0:
            s_vals = np.asarray([0.0], dtype=float)

        s_mean = float(np.nanmean(s_vals))
        s_std = float(np.nanstd(s_vals, ddof=1)) if len(s_vals) > 1 else 0.0
        z = float((obs - s_mean) / (s_std + 1e-12))
        p_greater = float((1.0 + np.sum(s_vals >= obs)) / (len(s_vals) + 1.0))
        p_lower = float((1.0 + np.sum(s_vals <= obs)) / (len(s_vals) + 1.0))
        empirical_percentile = float(np.mean(s_vals <= obs) * 100.0)

        observed_row[f"colored_noise_mean_{metric}"] = s_mean
        observed_row[f"colored_noise_std_{metric}"] = s_std
        observed_row[f"colored_noise_z_{metric}"] = z
        observed_row[f"colored_noise_p_greater_{metric}"] = p_greater
        observed_row[f"colored_noise_p_lower_{metric}"] = p_lower
        observed_row[f"colored_noise_empirical_percentile_{metric}"] = empirical_percentile
        observed_row[f"colored_noise_corrected_delta_{metric}"] = obs - s_mean
        observed_row[f"colored_noise_corrected_ratio_{metric}"] = obs / (s_mean + 1e-12)

    primary_z_values = [
        observed_row[f"colored_noise_z_{metric}"]
        for metric in primary_metric_names
        if f"colored_noise_z_{metric}" in observed_row and np.isfinite(observed_row[f"colored_noise_z_{metric}"])
    ]

    primary_neglogp_values = [
        -np.log10(observed_row[f"colored_noise_p_greater_{metric}"] + 1e-12)
        for metric in primary_metric_names
        if f"colored_noise_p_greater_{metric}" in observed_row and np.isfinite(observed_row[f"colored_noise_p_greater_{metric}"])
    ]

    observed_row["colored_noise_corrected_mean_primary_z_raw"] = float(np.mean(primary_z_values)) if len(primary_z_values) else 0.0
    observed_row["colored_noise_corrected_median_primary_z_raw"] = float(np.median(primary_z_values)) if len(primary_z_values) else 0.0
    observed_row["colored_noise_corrected_mean_primary_neglog10_p"] = float(np.mean(primary_neglogp_values)) if len(primary_neglogp_values) else 0.0

    elapsed = time.perf_counter() - ch_t0

    observed_row["elapsed_sec"] = elapsed
    observed_row["n_colored_noise_surrogates"] = N_COLORED_NOISE_SURROGATES
    observed_row["colored_noise_surrogate_type"] = COLORED_NOISE_SURROGATE_TYPE
    observed_row["beta_clip_low"] = BETA_CLIP[0]
    observed_row["beta_clip_high"] = BETA_CLIP[1]
    observed_row["joblib_n_jobs"] = JOBLIB_N_JOBS
    observed_row["joblib_backend"] = JOBLIB_BACKEND
    observed_row["fft_workers_per_job"] = FFT_WORKERS_PER_JOB

    observed_rows.append(observed_row)

    diagram_store[ch_name] = observed_h1_diagram
    colored_noise_metric_store[ch_name] = colored_noise_metric_df
    betti_curve_store[ch_name] = observed_betti
    landscape_store[ch_name] = observed_landscapes

    diagnostic_rows.append({
        "channel": ch_name,
        "region": channel_regions.get(ch_name, "Unknown"),
        "elapsed_sec": elapsed,
        "n_points": int(X.shape[0]),
        "ordered_reservoir_points": int(X_ordered.shape[0]),
        "radius": radius,
        "n_colored_noise_surrogates": N_COLORED_NOISE_SURROGATES,
        "joblib_n_jobs": JOBLIB_N_JOBS,
        "fft_workers_per_job": FFT_WORKERS_PER_JOB,
        "observed_n_h1_intervals": observed_features["n_h1_intervals"],
        "observed_total_lifetime": observed_features["total_lifetime"],
        "observed_max_lifetime": observed_features["max_lifetime"],
        "observed_betti_auc": observed_features["betti_auc"],
        "observed_landscape_l2": observed_features["landscape_l2"],
        "observed_beta_mean": observed_row["observed_colored_noise_beta_mean"],
    })

    channels_completed.append(ch_name)
    per_channel_times.append(elapsed)

    avg_t = float(np.mean(per_channel_times))
    remaining = avg_t * (len(eeg_channels) - ch_idx)
    elapsed_total = time.perf_counter() - overall_t0

    log(
        f"{prefix} done {elapsed:.2f}s | elapsed {fmt_time(elapsed_total)} | ETA {fmt_time(remaining)} | "
        f"obs H1={observed_features['n_h1_intervals']:.0f} | "
        f"obs total={observed_features['total_lifetime']:.4f} | "
        f"raw mean primary colored-noise z={observed_row['colored_noise_corrected_mean_primary_z_raw']:.2f}"
    )

if len(observed_rows) == 0:
    raise RuntimeError("No channels completed. Check embedding paths and ripser installation.")

summary_df = pd.DataFrame(observed_rows)
colored_noise_long_df = pd.DataFrame(colored_noise_rows)
diagnostic_df = pd.DataFrame(diagnostic_rows)

for metric in metric_names:
    p_col = f"colored_noise_p_greater_{metric}"
    q_col = f"colored_noise_q_greater_{metric}"

    if p_col in summary_df.columns:
        summary_df[q_col] = bh_fdr(summary_df[p_col].to_numpy(dtype=float))
        summary_df[f"colored_noise_neglog10_q_{metric}"] = -np.log10(summary_df[q_col].to_numpy(dtype=float) + 1e-12)

primary_q_cols = [f"colored_noise_q_greater_{metric}" for metric in primary_metric_names if f"colored_noise_q_greater_{metric}" in summary_df.columns]
primary_neglogq_cols = [f"colored_noise_neglog10_q_{metric}" for metric in primary_metric_names if f"colored_noise_neglog10_q_{metric}" in summary_df.columns]

regularized_z_df, regularized_z_info_df = compute_regularized_z(
    summary_df,
    colored_noise_long_df,
    primary_metric_names,
    prefix="colored_noise",
)

for col in regularized_z_df.columns:
    summary_df[col] = summary_df["channel"].astype(str).map(regularized_z_df[col].to_dict()).to_numpy(dtype=float)

regularized_z_cols = [f"colored_noise_z_regularized_{metric}" for metric in primary_metric_names]

summary_df["colored_noise_corrected_mean_primary_z"] = summary_df[regularized_z_cols].mean(axis=1)
summary_df["colored_noise_corrected_median_primary_z"] = summary_df[regularized_z_cols].median(axis=1)
summary_df["colored_noise_corrected_mean_primary_neglog10_q"] = summary_df[primary_neglogq_cols].mean(axis=1) if len(primary_neglogq_cols) else 0.0
summary_df["colored_noise_corrected_min_primary_q"] = summary_df[primary_q_cols].min(axis=1) if len(primary_q_cols) else 1.0
summary_df["colored_noise_corrected_n_primary_q_lt_0_05"] = (summary_df[primary_q_cols] < 0.05).sum(axis=1) if len(primary_q_cols) else 0
summary_df["colored_noise_corrected_n_primary_q_lt_0_10"] = (summary_df[primary_q_cols] < 0.10).sum(axis=1) if len(primary_q_cols) else 0

summary_df["colored_noise_corrected_ph_score"] = percentile_score(
    summary_df["colored_noise_corrected_mean_primary_z"],
    higher_is_more=True,
)

summary_df["colored_noise_rejection_score"] = percentile_score(
    summary_df["colored_noise_corrected_mean_primary_neglog10_q"],
    higher_is_more=True,
)

summary_df = summary_df.sort_values("colored_noise_corrected_ph_score", ascending=False).reset_index(drop=True)

corrected_signature_df = summary_df.set_index("channel")[regularized_z_cols].copy()
corrected_signature_df = corrected_signature_df.replace([np.inf, -np.inf], np.nan).fillna(0.0)

signature_channels = list(corrected_signature_df.index.astype(str))
signature_X = corrected_signature_df.to_numpy(dtype=float)

if len(signature_channels) >= 2:
    corrected_signature_distance = squareform(pdist(signature_X, metric="euclidean"))
else:
    corrected_signature_distance = np.zeros((len(signature_channels), len(signature_channels)), dtype=float)

corrected_signature_similarity, corrected_similarity_sigma = gaussian_similarity_from_distance(corrected_signature_distance)

if len(signature_channels) >= 3 and np.any(corrected_signature_distance > 0):
    Z_signature = linkage(squareform(corrected_signature_distance, checks=False), method="average", optimal_ordering=True)
    order = leaves_list(Z_signature)
else:
    order = np.arange(len(signature_channels))

ordered_channels = [signature_channels[i] for i in order]
ordered_signature_distance = corrected_signature_distance[np.ix_(order, order)]
ordered_signature_similarity = corrected_signature_similarity[np.ix_(order, order)]
ordered_signature_z = corrected_signature_df.loc[ordered_channels].to_numpy(dtype=float)

relative_matrix, contrast_info_df = robust_metric_contrast_matrix(ordered_signature_z)
relative_matrix_clipped = np.clip(relative_matrix, -3.0, 3.0)
contrast_info_df["metric"] = primary_metric_names

mds_coords = classical_mds_from_distances(corrected_signature_distance, n_components=2)
mds_df = pd.DataFrame({
    "channel": signature_channels,
    "region": [channel_regions.get(ch, "Unknown") for ch in signature_channels],
    "colored_noise_corrected_mds_1": mds_coords[:, 0],
    "colored_noise_corrected_mds_2": mds_coords[:, 1],
})

score_map = summary_df.set_index("channel")["colored_noise_corrected_ph_score"].to_dict()
reject_map = summary_df.set_index("channel")["colored_noise_rejection_score"].to_dict()

mds_df["colored_noise_corrected_ph_score"] = [score_map.get(ch, np.nan) for ch in signature_channels]
mds_df["colored_noise_rejection_score"] = [reject_map.get(ch, np.nan) for ch in signature_channels]

region_summary_df = (
    summary_df
    .groupby("region")
    .agg(
        n_channels=("channel", "count"),
        mean_colored_noise_corrected_ph_score=("colored_noise_corrected_ph_score", "mean"),
        sem_colored_noise_corrected_ph_score=("colored_noise_corrected_ph_score", lambda x: float(np.std(x, ddof=1) / np.sqrt(len(x))) if len(x) > 1 else 0.0),
        mean_primary_z=("colored_noise_corrected_mean_primary_z", "mean"),
        mean_primary_neglog10_q=("colored_noise_corrected_mean_primary_neglog10_q", "mean"),
        mean_n_primary_q_lt_0_05=("colored_noise_corrected_n_primary_q_lt_0_05", "mean"),
        mean_observed_beta=("observed_colored_noise_beta_mean", "mean"),
    )
    .reset_index()
    .sort_values("mean_colored_noise_corrected_ph_score", ascending=False)
)

summary_csv_path = os.path.join(mode_out_dir, "h1_colored_noise_surrogate_ph_summary.csv")
colored_noise_long_csv_path = os.path.join(mode_out_dir, "h1_colored_noise_surrogate_ph_colored_noise_long_table.csv")
diagnostic_csv_path = os.path.join(mode_out_dir, "h1_colored_noise_surrogate_ph_runtime_diagnostics.csv")
signature_csv_path = os.path.join(mode_out_dir, "h1_colored_noise_surrogate_ph_regularized_z_signature_matrix.csv")
signature_distance_csv_path = os.path.join(mode_out_dir, "h1_colored_noise_surrogate_ph_signature_distance_matrix.csv")
signature_similarity_csv_path = os.path.join(mode_out_dir, "h1_colored_noise_surrogate_ph_signature_similarity_matrix.csv")
mds_csv_path = os.path.join(mode_out_dir, "h1_colored_noise_surrogate_ph_mds_coordinates.csv")
region_summary_csv_path = os.path.join(mode_out_dir, "h1_colored_noise_surrogate_ph_region_summary.csv")
regularized_z_info_csv_path = os.path.join(mode_out_dir, "h1_colored_noise_surrogate_ph_regularized_z_info.csv")
contrast_info_csv_path = os.path.join(mode_out_dir, "h1_colored_noise_surrogate_ph_within_metric_contrast_info.csv")
contrast_matrix_csv_path = os.path.join(mode_out_dir, "h1_colored_noise_surrogate_ph_within_metric_contrast_matrix.csv")
npz_path = os.path.join(mode_out_dir, "h1_colored_noise_surrogate_ph_outputs.npz")

summary_df.to_csv(summary_csv_path, index=False)
colored_noise_long_df.to_csv(colored_noise_long_csv_path, index=False)
diagnostic_df.to_csv(diagnostic_csv_path, index=False)
corrected_signature_df.to_csv(signature_csv_path)
pd.DataFrame(corrected_signature_distance, index=signature_channels, columns=signature_channels).to_csv(signature_distance_csv_path)
pd.DataFrame(corrected_signature_similarity, index=signature_channels, columns=signature_channels).to_csv(signature_similarity_csv_path)
mds_df.to_csv(mds_csv_path, index=False)
region_summary_df.to_csv(region_summary_csv_path, index=False)
regularized_z_info_df.to_csv(regularized_z_info_csv_path, index=False)
contrast_info_df.to_csv(contrast_info_csv_path, index=False)
pd.DataFrame(relative_matrix, index=ordered_channels, columns=primary_metric_names).to_csv(contrast_matrix_csv_path)

save_dict = {
    "channels": np.asarray(signature_channels),
    "primary_metric_names": np.asarray(primary_metric_names),
    "regularized_z_columns": np.asarray(regularized_z_cols),
    "corrected_signature_matrix": corrected_signature_df.to_numpy(dtype=float),
    "within_metric_contrast_matrix": relative_matrix,
    "corrected_signature_distance_matrix": corrected_signature_distance,
    "corrected_signature_similarity_matrix": corrected_signature_similarity,
    "corrected_similarity_sigma": np.asarray([corrected_similarity_sigma]),
    "mds_coordinates": mds_coords,
    "colored_noise_corrected_ph_score": np.asarray([score_map.get(ch, np.nan) for ch in signature_channels], dtype=float),
    "colored_noise_rejection_score": np.asarray([reject_map.get(ch, np.nan) for ch in signature_channels], dtype=float),
    "fast_option": np.asarray([FAST_OPTION]),
    "colored_noise_surrogate_type": np.asarray([COLORED_NOISE_SURROGATE_TYPE]),
    "n_colored_noise_surrogates": np.asarray([N_COLORED_NOISE_SURROGATES]),
    "beta_clip": np.asarray(BETA_CLIP),
    "joblib_n_jobs": np.asarray([JOBLIB_N_JOBS]),
    "joblib_backend": np.asarray([JOBLIB_BACKEND]),
    "fft_workers_per_job": np.asarray([FFT_WORKERS_PER_JOB]),
}

for ch in signature_channels:
    if ch in diagram_store:
        save_dict[f"{ch}_observed_h1_diagram"] = diagram_store[ch]
    if ch in betti_curve_store:
        save_dict[f"{ch}_observed_betti_curve"] = betti_curve_store[ch]
    if ch in landscape_store:
        save_dict[f"{ch}_observed_landscapes"] = landscape_store[ch]

np.savez(npz_path, **save_dict)

print("\nH1 Colored-Noise-Surrogate Persistent Homology Summary:")
if HAVE_IPYTHON:
    display(summary_df)
else:
    print(summary_df)

print("\nColored-Noise-Corrected H1 Regularized Z-Signature Matrix:")
if HAVE_IPYTHON:
    display(corrected_signature_df)
else:
    print(corrected_signature_df)

print("\nRegional Colored-Noise-Corrected PH Summary:")
if HAVE_IPYTHON:
    display(region_summary_df)
else:
    print(region_summary_df)


fig, ax = plt.subplots(figsize=(13, 6.2), facecolor=BLACK)
style_ax(
    ax,
    title=f"H1 Colored-Noise-Surrogate Persistent Homology Score | {FAST_OPTION.upper()}",
    xlabel="Channel",
    ylabel="Colored-noise-corrected PH score",
)

rank_df = summary_df.sort_values("colored_noise_corrected_ph_score", ascending=False).reset_index(drop=True)
x = np.arange(len(rank_df))

ax.bar(
    x,
    rank_df["colored_noise_corrected_ph_score"],
    color=CYAN_FILL,
    edgecolor=ACCENT,
    linewidth=0.9,
    alpha=0.82,
)

ax.axhline(
    float(np.nanmean(rank_df["colored_noise_corrected_ph_score"])),
    color=ACCENT_SOFT,
    linestyle="--",
    linewidth=1.3,
    alpha=0.85,
    label=f"Mean={np.nanmean(rank_df['colored_noise_corrected_ph_score']):.1f}",
)

for i, row in rank_df.iterrows():
    if i < 8:
        ax.text(
            i,
            row["colored_noise_corrected_ph_score"],
            f"{row['colored_noise_corrected_ph_score']:.0f}",
            ha="center",
            va="bottom",
            color=ACCENT,
            fontsize=8,
        )

ax.set_xticks(x)
ax.set_xticklabels(rank_df["channel"], rotation=90)
ax.set_ylim(0, 105)

legend = ax.legend(loc="best", framealpha=1.0)
style_legend(legend)
force_cyan_text(ax)

score_path = os.path.join(mode_out_dir, "01_colored_noise_corrected_ph_score_ranking.png")
plt.tight_layout()
save_render_close(fig, score_path, width=1150)


fig, ax = plt.subplots(figsize=(12.8, 9.2), facecolor=BLACK)
style_ax(
    ax,
    title="Colored-Noise-Corrected H1 Primary Metric Z-Score Heatmap | fixed contrast",
    xlabel="H1 metric",
    ylabel="Channel",
)

im = ax.imshow(
    relative_matrix_clipped,
    aspect="auto",
    interpolation="nearest",
    cmap=CYAN_DIVERGING_CMAP,
    vmin=-3,
    vmax=3,
)

ax.set_yticks(np.arange(len(ordered_channels)))
ax.set_yticklabels(ordered_channels)
ax.set_xticks(np.arange(len(primary_metric_labels)))
ax.set_xticklabels(primary_metric_labels, rotation=45, ha="right")

ax.set_xticks(np.arange(-0.5, len(primary_metric_labels), 1), minor=True)
ax.set_yticks(np.arange(-0.5, len(ordered_channels), 1), minor=True)
ax.grid(which="minor", color=ACCENT_DIM, linewidth=0.55, alpha=0.35)
ax.tick_params(which="minor", bottom=False, left=False)

for i in range(ordered_signature_z.shape[0]):
    for j in range(ordered_signature_z.shape[1]):
        rel_val = relative_matrix[i, j]
        raw_val = ordered_signature_z[i, j]

        if not np.isfinite(raw_val):
            continue

        show_text = abs(rel_val) >= 1.15 or i < 5 or i >= ordered_signature_z.shape[0] - 3

        if show_text:
            txt = ax.text(
                j,
                i,
                f"{raw_val:+.1f}",
                ha="center",
                va="center",
                color=ACCENT,
                fontsize=5.9,
                alpha=0.95,
                fontweight="bold",
            )
            txt.set_path_effects([pe.withStroke(linewidth=1.4, foreground=BLACK)])

cbar = plt.colorbar(im, ax=ax, shrink=0.85)
style_colorbar(cbar, "Within-metric channel contrast")
force_cyan_text(ax)

fig.text(
    0.09,
    0.035,
    "Fixed view: color shows within-metric channel contrast; printed values are regularized colored-noise z-scores.",
    ha="left",
    va="bottom",
    color=ACCENT,
    fontsize=9,
)

z_heatmap_path = os.path.join(mode_out_dir, "02_colored_noise_corrected_primary_metric_z_heatmap.png")
z_heatmap_copy_path = os.path.join(mode_out_dir, "02_colored_noise_corrected_primary_metric_z_heatmap_FIXED_CONTRAST.png")
plt.tight_layout(rect=[0.0, 0.055, 1.0, 1.0])

if SAVE_PLOTS_TO_DISK:
    fig.savefig(z_heatmap_path, dpi=PLOT_DPI, bbox_inches="tight", facecolor=BLACK)
    fig.savefig(z_heatmap_copy_path, dpi=PLOT_DPI, bbox_inches="tight", facecolor=BLACK)

plt.close(fig)

if RENDER_ALL_IMAGES_INLINE and HAVE_IPYTHON and os.path.exists(z_heatmap_path):
    display(Image(filename=z_heatmap_path, width=1050))


order_df = summary_df.set_index("channel").loc[ordered_channels].reset_index()
q_cols = [f"colored_noise_q_greater_{metric}" for metric in primary_metric_names]
q_matrix = order_df[q_cols].to_numpy(dtype=float)

sig_005 = q_matrix < 0.05
sig_010 = (q_matrix >= 0.05) & (q_matrix < 0.10)
all_sig_005 = bool(np.all(sig_005))

fig = plt.figure(figsize=(16.8, 10.5), facecolor=BLACK)

ax = fig.add_axes([0.07, 0.18, 0.58, 0.72])
rank_ax = fig.add_axes([0.74, 0.18, 0.20, 0.72])
cbar_ax = fig.add_axes([0.67, 0.18, 0.018, 0.72])

for this_ax in [ax, rank_ax]:
    this_ax.set_facecolor(BLACK)
    for spine in this_ax.spines.values():
        spine.set_color(ACCENT)
        spine.set_linewidth(1.1)
    this_ax.tick_params(axis="x", colors=ACCENT, labelcolor=ACCENT)
    this_ax.tick_params(axis="y", colors=ACCENT, labelcolor=ACCENT)

im = ax.imshow(
    relative_matrix_clipped,
    aspect="auto",
    interpolation="nearest",
    cmap=CYAN_DIVERGING_CMAP,
    vmin=-3,
    vmax=3,
)

ax.set_title(
    "Colored-Noise-Corrected H1 Significance + Effect Direction | fixed contrast\n"
    "Color = within-metric channel contrast; printed values = regularized colored-noise z-score",
    color=ACCENT,
    pad=14,
)

ax.set_xlabel("H1 metric", color=ACCENT, labelpad=12)
ax.set_ylabel("Channel", color=ACCENT, labelpad=12)

ax.set_xticks(np.arange(len(primary_metric_labels)))
ax.set_xticklabels(primary_metric_labels, rotation=45, ha="right", color=ACCENT)
ax.set_yticks(np.arange(len(ordered_channels)))
ax.set_yticklabels(ordered_channels, color=ACCENT, fontsize=8)

ax.set_xticks(np.arange(-0.5, len(primary_metric_labels), 1), minor=True)
ax.set_yticks(np.arange(-0.5, len(ordered_channels), 1), minor=True)
ax.grid(which="minor", color=ACCENT_DIM, linewidth=0.55, alpha=0.35)
ax.tick_params(which="minor", bottom=False, left=False)

for i in range(ordered_signature_z.shape[0]):
    for j in range(ordered_signature_z.shape[1]):
        rel_val = relative_matrix[i, j]
        raw_val = ordered_signature_z[i, j]

        if not np.isfinite(raw_val):
            continue

        show_text = abs(rel_val) >= 1.15 or i < 5 or i >= ordered_signature_z.shape[0] - 3

        if show_text:
            txt = ax.text(
                j,
                i,
                f"{raw_val:+.1f}",
                ha="center",
                va="center",
                color=ACCENT,
                fontsize=5.3,
                alpha=0.92,
                fontweight="bold",
            )
            txt.set_path_effects([pe.withStroke(linewidth=1.4, foreground=BLACK)])

        if not all_sig_005:
            if sig_005[i, j]:
                marker = "●"
                marker_size = 8
            elif sig_010[i, j]:
                marker = "○"
                marker_size = 8
            else:
                marker = ""

            if marker:
                ax.text(
                    j,
                    i - 0.28,
                    marker,
                    ha="center",
                    va="center",
                    color=ACCENT_SOFT,
                    fontsize=marker_size,
                    fontweight="bold",
                )

cbar = plt.colorbar(im, cax=cbar_ax)
cbar.set_label("Within-metric channel contrast", color=ACCENT)
cbar.ax.yaxis.set_tick_params(color=ACCENT)
plt.setp(cbar.ax.get_yticklabels(), color=ACCENT)
cbar.outline.set_edgecolor(ACCENT)

score_values = order_df["colored_noise_corrected_ph_score"].to_numpy(dtype=float)
y = np.arange(len(ordered_channels))

rank_ax.barh(
    y,
    score_values,
    color=CYAN_FILL,
    edgecolor=ACCENT,
    linewidth=0.7,
    alpha=0.82,
)

rank_ax.set_title("Colored-noise-corrected\nPH score", color=ACCENT, pad=14)
rank_ax.set_xlabel("Score", color=ACCENT, labelpad=10)
rank_ax.set_yticks([])
rank_ax.invert_yaxis()
rank_ax.set_xlim(0, 105)

for yy, val in zip(y, score_values):
    if np.isfinite(val) and (yy < 7 or yy >= len(y) - 4):
        rank_ax.text(
            min(val + 1.0, 101.5),
            yy,
            f"{val:.0f}",
            va="center",
            ha="left",
            color=ACCENT,
            fontsize=7.5,
            alpha=0.95,
        )

rank_ax.grid(True, color=ACCENT_DIM, alpha=0.25, linewidth=0.8)

n_sig_005_by_metric = sig_005.sum(axis=0)

footer_lines = []

if all_sig_005:
    footer_lines.append("All displayed cells have FDR q < 0.05, so redundant significance dots are omitted.")
else:
    footer_lines.append("● = FDR q < 0.05    ○ = FDR q < 0.10")

footer_lines.append(
    "q<0.05 counts by metric: "
    + ", ".join([f"{label}: {count}" for label, count in zip(primary_metric_labels, n_sig_005_by_metric)])
)
footer_lines.append(
    "Fixed view: color compares channels within each metric while printed values retain the regularized colored-noise z-score."
)

fig.text(
    0.07,
    0.055,
    "\n".join(footer_lines),
    ha="left",
    va="bottom",
    color=ACCENT,
    fontsize=9,
)

q_heatmap_path = os.path.join(mode_out_dir, "03_colored_noise_corrected_primary_metric_q_heatmap.png")
q_heatmap_fixed_copy_path = os.path.join(mode_out_dir, "03_colored_noise_corrected_primary_metric_significance_effect_heatmap_fixed.png")

if SAVE_PLOTS_TO_DISK:
    fig.savefig(q_heatmap_path, dpi=PLOT_DPI, bbox_inches="tight", facecolor=BLACK)
    fig.savefig(q_heatmap_fixed_copy_path, dpi=PLOT_DPI, bbox_inches="tight", facecolor=BLACK)

plt.close(fig)

if RENDER_ALL_IMAGES_INLINE and HAVE_IPYTHON and os.path.exists(q_heatmap_path):
    display(Image(filename=q_heatmap_path, width=1150))


sim_vmin, sim_vmax = dynamic_limits(
    offdiag_values(ordered_signature_similarity),
    lower_pct=2,
    upper_pct=98,
    pad_frac=0.10,
    lower_bound=0.0,
    upper_bound=1.0,
)

fig, ax = plt.subplots(figsize=(9.5, 8.5), facecolor=BLACK)
style_ax(
    ax,
    title="Colored-Noise-Corrected PH Signature Similarity Matrix",
    xlabel="Channel",
    ylabel="Channel",
)

im = ax.imshow(
    ordered_signature_similarity,
    aspect="equal",
    interpolation="nearest",
    cmap=CYAN_SEQUENTIAL_CMAP,
    vmin=sim_vmin,
    vmax=sim_vmax,
)

ax.set_xticks(np.arange(len(ordered_channels)))
ax.set_xticklabels(ordered_channels, rotation=90)
ax.set_yticks(np.arange(len(ordered_channels)))
ax.set_yticklabels(ordered_channels)

cbar = plt.colorbar(im, ax=ax, shrink=0.85)
style_colorbar(cbar, "Colored-noise-corrected signature similarity")
force_cyan_text(ax)

signature_similarity_path = os.path.join(mode_out_dir, "04_colored_noise_corrected_ph_signature_similarity_heatmap.png")
plt.tight_layout()
save_render_close(fig, signature_similarity_path, width=950)


dist_vmin, dist_vmax = dynamic_limits(
    offdiag_values(ordered_signature_distance),
    lower_pct=2,
    upper_pct=98,
    pad_frac=0.10,
    lower_bound=0.0,
)

fig, ax = plt.subplots(figsize=(9.5, 8.5), facecolor=BLACK)
style_ax(
    ax,
    title="Colored-Noise-Corrected PH Signature Distance Matrix",
    xlabel="Channel",
    ylabel="Channel",
)

im = ax.imshow(
    ordered_signature_distance,
    aspect="equal",
    interpolation="nearest",
    cmap=CYAN_SEQUENTIAL_CMAP,
    vmin=dist_vmin,
    vmax=dist_vmax,
)

ax.set_xticks(np.arange(len(ordered_channels)))
ax.set_xticklabels(ordered_channels, rotation=90)
ax.set_yticks(np.arange(len(ordered_channels)))
ax.set_yticklabels(ordered_channels)

cbar = plt.colorbar(im, ax=ax, shrink=0.85)
style_colorbar(cbar, "Colored-noise-corrected signature distance")
force_cyan_text(ax)

signature_distance_path = os.path.join(mode_out_dir, "05_colored_noise_corrected_ph_signature_distance_heatmap.png")
plt.tight_layout()
save_render_close(fig, signature_distance_path, width=950)


fig, ax = plt.subplots(figsize=(8.5, 7.2), facecolor=BLACK)
style_ax(
    ax,
    title="MDS of Colored-Noise-Corrected PH Signatures",
    xlabel="Colored-noise-corrected PH MDS 1",
    ylabel="Colored-noise-corrected PH MDS 2",
)

sc = ax.scatter(
    mds_df["colored_noise_corrected_mds_1"],
    mds_df["colored_noise_corrected_mds_2"],
    c=mds_df["colored_noise_corrected_ph_score"],
    s=165,
    cmap=CYAN_SEQUENTIAL_CMAP,
    vmin=0,
    vmax=100,
    edgecolor=ACCENT,
    linewidth=1.2,
    alpha=0.95,
)

for _, row in mds_df.iterrows():
    txt = ax.text(
        row["colored_noise_corrected_mds_1"],
        row["colored_noise_corrected_mds_2"],
        row["channel"],
        fontsize=8,
        color=ACCENT,
        alpha=0.95,
        zorder=5,
    )
    txt.set_path_effects([pe.withStroke(linewidth=2.0, foreground=BLACK)])

cbar = plt.colorbar(sc, ax=ax, shrink=0.82)
style_colorbar(cbar, "Colored-noise-corrected PH score")
force_cyan_text(ax)

mds_path = os.path.join(mode_out_dir, "06_mds_colored_noise_corrected_ph_signatures.png")
plt.tight_layout()
save_render_close(fig, mds_path, width=930)


plot_metric = "total_lifetime"

plot_df = summary_df.copy()
plot_df["observed_total_lifetime"] = plot_df[f"observed_{plot_metric}"].to_numpy(dtype=float)
plot_df["colored_noise_mean_total_lifetime"] = plot_df[f"colored_noise_mean_{plot_metric}"].to_numpy(dtype=float)
plot_df["colored_noise_std_total_lifetime"] = plot_df[f"colored_noise_std_{plot_metric}"].to_numpy(dtype=float) if f"colored_noise_std_{plot_metric}" in plot_df.columns else 0.0
plot_df["delta_total_lifetime"] = plot_df["observed_total_lifetime"] - plot_df["colored_noise_mean_total_lifetime"]
plot_df["q_total_lifetime"] = plot_df[f"colored_noise_q_greater_{plot_metric}"].to_numpy(dtype=float)

plot_df = plot_df.replace([np.inf, -np.inf], np.nan)
paired_df = plot_df.sort_values("delta_total_lifetime", ascending=True).reset_index(drop=True)

fig = plt.figure(figsize=(15.8, 8.4), facecolor=BLACK)

paired_ax = fig.add_axes([0.07, 0.14, 0.42, 0.74])
rank_ax = fig.add_axes([0.59, 0.14, 0.34, 0.74])

y = np.arange(len(paired_df))

style_ax(
    paired_ax,
    title="Observed vs Colored-Noise Mean H1 Total Lifetime | fixed paired view",
    xlabel="H1 total lifetime",
    ylabel="Channel",
)

paired_ax.hlines(
    y,
    paired_df["colored_noise_mean_total_lifetime"],
    paired_df["observed_total_lifetime"],
    color=ACCENT_DIM,
    linewidth=1.0,
    alpha=0.55,
)

paired_ax.scatter(
    paired_df["colored_noise_mean_total_lifetime"],
    y,
    s=72,
    facecolors=BLACK,
    edgecolors=ACCENT_SOFT,
    linewidth=1.2,
    label="Colored-noise mean",
    zorder=4,
)

paired_ax.scatter(
    paired_df["observed_total_lifetime"],
    y,
    s=82,
    color=CYAN_FILL,
    edgecolor=ACCENT,
    linewidth=1.0,
    label="Observed",
    zorder=5,
)

paired_ax.set_yticks(y)
paired_ax.set_yticklabels(paired_df["channel"], fontsize=7)

x_vals = np.concatenate([
    paired_df["colored_noise_mean_total_lifetime"].to_numpy(dtype=float),
    paired_df["observed_total_lifetime"].to_numpy(dtype=float),
])
x_lo, x_hi = robust_limits(x_vals, lower_pct=1, upper_pct=99, pad_frac=0.12, include_zero=True)
paired_ax.set_xlim(x_lo, x_hi)

legend = paired_ax.legend(loc="lower right", framealpha=1.0, fontsize=8)
style_legend(legend)

style_ax(
    rank_ax,
    title="Observed − Colored-Noise Mean",
    xlabel="Δ H1 total lifetime",
    ylabel="Channel",
)

rank_ax.barh(
    y,
    paired_df["delta_total_lifetime"],
    color=CYAN_FILL,
    edgecolor=ACCENT,
    linewidth=0.75,
    alpha=0.82,
)

rank_ax.axvline(
    0,
    color=ACCENT_SOFT,
    linestyle="--",
    linewidth=1.2,
    alpha=0.9,
)

rank_ax.set_yticks(y)
rank_ax.set_yticklabels(paired_df["channel"], fontsize=7)

delta_lo, delta_hi = robust_limits(
    paired_df["delta_total_lifetime"].to_numpy(dtype=float),
    lower_pct=1,
    upper_pct=99,
    pad_frac=0.18,
    include_zero=True,
)
rank_ax.set_xlim(delta_lo, delta_hi)

for yy, (_, row) in zip(y, paired_df.iterrows()):
    delta = float(row["delta_total_lifetime"])
    q_val = float(row["q_total_lifetime"])

    if q_val < 0.05:
        marker = "●"
    elif q_val < 0.10:
        marker = "○"
    else:
        marker = ""

    if marker:
        x_text = delta + (0.015 * (delta_hi - delta_lo) if delta >= 0 else -0.015 * (delta_hi - delta_lo))
        ha = "left" if delta >= 0 else "right"

        rank_ax.text(
            x_text,
            yy,
            marker,
            va="center",
            ha=ha,
            color=ACCENT_SOFT,
            fontsize=9,
            fontweight="bold",
        )

force_cyan_text(paired_ax)
force_cyan_text(rank_ax)

footer = (
    "Fixed view: paired plot avoids collapsed scatter when colored-noise means cluster near one side. "
    "Left compares each channel's colored-noise mean to observed H1 total lifetime; right shows signed effect. "
    "● = FDR q < 0.05, ○ = FDR q < 0.10."
)

fig.text(
    0.07,
    0.035,
    footer,
    ha="left",
    va="bottom",
    color=ACCENT,
    fontsize=9,
)

obs_null_path = os.path.join(mode_out_dir, "07_observed_vs_colored_noise_mean_total_lifetime.png")
obs_null_copy_path = os.path.join(mode_out_dir, "07_observed_vs_colored_noise_mean_total_lifetime_FIXED.png")

if SAVE_PLOTS_TO_DISK:
    fig.savefig(obs_null_path, dpi=PLOT_DPI, bbox_inches="tight", facecolor=BLACK)
    fig.savefig(obs_null_copy_path, dpi=PLOT_DPI, bbox_inches="tight", facecolor=BLACK)

plt.close(fig)

if RENDER_ALL_IMAGES_INLINE and HAVE_IPYTHON and os.path.exists(obs_null_path):
    display(Image(filename=obs_null_path, width=1150))


top_channels = summary_df.sort_values("colored_noise_corrected_ph_score", ascending=False).head(min(8, len(summary_df)))["channel"].tolist()

if len(colored_noise_long_df) > 0 and "total_lifetime" in colored_noise_long_df.columns:
    fig, ax = plt.subplots(figsize=(13, 6.4), facecolor=BLACK)
    style_ax(
        ax,
        title="Colored-Noise Null Distributions for Top Corrected Channels",
        xlabel="Channel",
        ylabel="H1 total lifetime",
    )

    box_data = []
    box_labels = []
    obs_for_labels = []

    for ch in top_channels:
        vals = colored_noise_long_df.loc[
            colored_noise_long_df["channel"] == ch,
            "total_lifetime",
        ].to_numpy(dtype=float)

        obs = float(summary_df.loc[summary_df["channel"] == ch, "observed_total_lifetime"].iloc[0])

        box_data.append(vals)
        box_labels.append(ch)
        obs_for_labels.append(obs)

    box = ax.boxplot(
        box_data,
        patch_artist=True,
        labels=box_labels,
        showfliers=False,
        widths=0.55,
    )

    for patch in box["boxes"]:
        patch.set_facecolor(CYAN_FILL)
        patch.set_edgecolor(ACCENT)
        patch.set_alpha(0.32)
        patch.set_linewidth(1.4)

    for key in ["whiskers", "caps", "medians"]:
        for item in box[key]:
            item.set_color(ACCENT)
            item.set_linewidth(1.4)

    for idx, obs in enumerate(obs_for_labels, start=1):
        ax.scatter(
            [idx],
            [obs],
            s=95,
            color=CYAN_FILL,
            edgecolor=ACCENT,
            linewidth=1.2,
            zorder=5,
            label="Observed" if idx == 1 else None,
        )

        ax.text(
            idx,
            obs,
            f"  obs={obs:.2f}",
            color=ACCENT,
            fontsize=8,
            va="center",
            ha="left",
        )

    legend = ax.legend(loc="best", framealpha=1.0)
    style_legend(legend)
    force_cyan_text(ax)

    null_box_path = os.path.join(mode_out_dir, "08_colored_noise_null_distributions_top_channels.png")
    plt.tight_layout()
    save_render_close(fig, null_box_path, width=1150)


fig = plt.figure(figsize=(13.8, 8.2), facecolor=BLACK)
gs = GridSpec(
    1,
    2,
    width_ratios=[1.25, 0.85],
    wspace=0.25,
    figure=fig,
)

ax = fig.add_subplot(gs[0, 0])
rank_ax = fig.add_subplot(gs[0, 1])

style_ax(
    ax,
    title="Scalp Map of Colored-Noise-Corrected PH Score",
    grid=False,
)

scalp_xy = {
    "Fp1": (-0.45, 1.00), "Fpz": (0.00, 1.08), "Fp2": (0.45, 1.00),
    "F7": (-0.95, 0.60), "F3": (-0.45, 0.55), "Fz": (0.00, 0.60),
    "F4": (0.45, 0.55), "F8": (0.95, 0.60),
    "FC5": (-0.75, 0.25), "FC1": (-0.25, 0.25),
    "FC2": (0.25, 0.25), "FC6": (0.75, 0.25),
    "M1": (-1.15, 0.05), "T7": (-1.00, 0.00), "C3": (-0.50, 0.00),
    "Cz": (0.00, 0.00), "C4": (0.50, 0.00), "T8": (1.00, 0.00),
    "M2": (1.15, 0.05),
    "CP5": (-0.75, -0.30), "CP1": (-0.25, -0.30),
    "CP2": (0.25, -0.30), "CP6": (0.75, -0.30),
    "P7": (-0.90, -0.60), "P3": (-0.45, -0.60), "Pz": (0.00, -0.65),
    "P4": (0.45, -0.60), "P8": (0.90, -0.60),
    "POz": (0.00, -0.88),
    "O1": (-0.35, -1.00), "Oz": (0.00, -1.08), "O2": (0.35, -1.00),
}

head = plt.Circle((0, 0), 1.12, fill=False, color=ACCENT, linewidth=2.2)
ax.add_patch(head)
ax.plot([0.00, -0.10, 0.10, 0.00], [1.12, 1.28, 1.28, 1.12], color=ACCENT, linewidth=2.2)

scalp_df = summary_df[summary_df["channel"].isin(scalp_xy)].copy()
scalp_df["x"] = scalp_df["channel"].map(lambda ch: scalp_xy[ch][0])
scalp_df["y"] = scalp_df["channel"].map(lambda ch: scalp_xy[ch][1])

scores = scalp_df["colored_noise_corrected_ph_score"].to_numpy(dtype=float)

if np.nanmax(scores) > np.nanmin(scores):
    norm_scores = (scores - np.nanmin(scores)) / (np.nanmax(scores) - np.nanmin(scores) + 1e-12)
else:
    norm_scores = np.full(len(scores), 0.5)

scalp_df["marker_size"] = 90.0 + 520.0 * norm_scores

scatter = ax.scatter(
    scalp_df["x"],
    scalp_df["y"],
    c=scalp_df["colored_noise_corrected_ph_score"],
    s=scalp_df["marker_size"],
    cmap=CYAN_SEQUENTIAL_CMAP,
    vmin=0,
    vmax=100,
    edgecolor=ACCENT,
    linewidth=1.25,
    alpha=0.92,
    zorder=4,
)

top_score_channels = scalp_df.sort_values("colored_noise_corrected_ph_score", ascending=False).head(8)["channel"].tolist()

for _, row in scalp_df.iterrows():
    ch = row["channel"]
    x0 = float(row["x"])
    y0 = float(row["y"])
    score = float(row["colored_noise_corrected_ph_score"])

    txt = ax.text(
        x0,
        y0,
        ch,
        ha="center",
        va="center",
        fontsize=8,
        color=ACCENT,
        fontweight="bold",
        zorder=6,
    )
    txt.set_path_effects([pe.withStroke(linewidth=2.1, foreground=BLACK)])

    if ch in top_score_channels:
        ax.scatter(
            [x0],
            [y0],
            s=float(row["marker_size"]) * 1.75,
            facecolors="none",
            edgecolors=ACCENT_SOFT,
            linewidth=1.8,
            alpha=0.95,
            zorder=5,
        )

        dx = 0.06 if x0 <= 0 else -0.06
        ha = "left" if x0 <= 0 else "right"

        ann = ax.text(
            x0 + dx,
            y0 + 0.055,
            f"{score:.0f}",
            ha=ha,
            va="center",
            fontsize=8,
            color=ACCENT_SOFT,
            fontweight="bold",
            zorder=7,
        )
        ann.set_path_effects([pe.withStroke(linewidth=2.0, foreground=BLACK)])

ax.text(
    0,
    -1.30,
    "Bigger/brighter = stronger H1 structure after colored-noise correction | numbers mark top corrected channels",
    ha="center",
    va="top",
    color=ACCENT,
    fontsize=9,
)

ax.set_xlim(-1.38, 1.38)
ax.set_ylim(-1.40, 1.36)
ax.set_aspect("equal")
ax.set_xticks([])
ax.set_yticks([])

cbar = plt.colorbar(scatter, ax=ax, shrink=0.78, pad=0.03)
style_colorbar(cbar, "Colored-noise-corrected PH score")

rank_side_df = scalp_df.sort_values("colored_noise_corrected_ph_score", ascending=True).copy()

style_ax(
    rank_ax,
    title="Colored-Noise-Corrected PH Ranking",
    xlabel="Corrected PH score",
    ylabel="Channel",
)

y = np.arange(len(rank_side_df))

rank_ax.barh(
    y,
    rank_side_df["colored_noise_corrected_ph_score"],
    color=CYAN_FILL,
    edgecolor=ACCENT,
    linewidth=0.8,
    alpha=0.82,
)

rank_ax.set_yticks(y)
rank_ax.set_yticklabels(rank_side_df["channel"])
rank_ax.set_xlim(0, 105)

rank_ax.axvline(
    float(np.nanmean(rank_side_df["colored_noise_corrected_ph_score"])),
    color=ACCENT_SOFT,
    linestyle="--",
    linewidth=1.2,
    alpha=0.85,
    label=f"Mean={np.nanmean(rank_side_df['colored_noise_corrected_ph_score']):.1f}",
)

for yy, (_, row) in zip(y, rank_side_df.iterrows()):
    score = float(row["colored_noise_corrected_ph_score"])

    rank_ax.text(
        score + 1.2,
        yy,
        f"{score:.0f}",
        va="center",
        ha="left",
        color=ACCENT,
        fontsize=7,
        alpha=0.82,
    )

legend = rank_ax.legend(loc="lower right", framealpha=1.0, fontsize=8)
style_legend(legend)
force_cyan_text(rank_ax)

top_text = "Top colored-noise-corrected: " + ", ".join(
    [
        f"{row.channel}({row.colored_noise_corrected_ph_score:.0f})"
        for row in scalp_df.sort_values("colored_noise_corrected_ph_score", ascending=False).head(5).itertuples()
    ]
)

bottom_text = "Lowest colored-noise-corrected: " + ", ".join(
    [
        f"{row.channel}({row.colored_noise_corrected_ph_score:.0f})"
        for row in scalp_df.sort_values("colored_noise_corrected_ph_score", ascending=False).tail(5).itertuples()
    ]
)

fig.text(
    0.5,
    0.018,
    top_text + "    |    " + bottom_text,
    ha="center",
    va="bottom",
    color=ACCENT,
    fontsize=9,
)

scalp_path = os.path.join(mode_out_dir, "09_scalp_map_colored_noise_corrected_ph_score.png")
plt.tight_layout(rect=[0.0, 0.045, 1.0, 1.0])
save_render_close(fig, scalp_path, width=1150)


fig, ax = plt.subplots(figsize=(12, 6.2), facecolor=BLACK)
style_ax(
    ax,
    title="Regional Colored-Noise-Corrected PH Score",
    xlabel="Region",
    ylabel="Mean colored-noise-corrected PH score",
)

x = np.arange(len(region_summary_df))

ax.bar(
    x,
    region_summary_df["mean_colored_noise_corrected_ph_score"],
    yerr=region_summary_df["sem_colored_noise_corrected_ph_score"],
    color=CYAN_FILL,
    edgecolor=ACCENT,
    linewidth=0.9,
    alpha=0.82,
    capsize=7,
    error_kw={"elinewidth": 1.5, "capthick": 1.5, "ecolor": ACCENT},
)

for i, row in region_summary_df.iterrows():
    ax.text(
        i,
        row["mean_colored_noise_corrected_ph_score"] + row["sem_colored_noise_corrected_ph_score"],
        f"{row['mean_colored_noise_corrected_ph_score']:.1f}\nn={int(row['n_channels'])}",
        ha="center",
        va="bottom",
        color=ACCENT,
        fontsize=9,
    )

ax.set_xticks(x)
ax.set_xticklabels(region_summary_df["region"], rotation=35, ha="right")
ax.set_ylim(0, 105)
force_cyan_text(ax)

region_path = os.path.join(mode_out_dir, "10_regional_colored_noise_corrected_ph_score.png")
plt.tight_layout()
save_render_close(fig, region_path, width=1050)


fig, ax = plt.subplots(figsize=(13, 6.2), facecolor=BLACK)
style_ax(
    ax,
    title="Colored-Noise-Corrected Primary Metric Z-Profiles by Channel",
    xlabel="Channel",
    ylabel="Mean primary z-score",
)

profile_df = summary_df.sort_values("colored_noise_corrected_mean_primary_z", ascending=False).reset_index(drop=True)
x = np.arange(len(profile_df))

ax.bar(
    x,
    profile_df["colored_noise_corrected_mean_primary_z"],
    color=CYAN_FILL,
    edgecolor=ACCENT,
    linewidth=0.9,
    alpha=0.82,
)

ax.axhline(0, color=ACCENT_SOFT, linestyle="--", linewidth=1.2)

for i, row in profile_df.iterrows():
    if i < 8:
        ax.text(
            i,
            row["colored_noise_corrected_mean_primary_z"],
            f"{row['colored_noise_corrected_mean_primary_z']:.1f}",
            ha="center",
            va="bottom" if row["colored_noise_corrected_mean_primary_z"] >= 0 else "top",
            color=ACCENT,
            fontsize=8,
        )

ax.set_xticks(x)
ax.set_xticklabels(profile_df["channel"], rotation=90)
force_cyan_text(ax)

mean_z_path = os.path.join(mode_out_dir, "11_mean_primary_colored_noise_z_ranking.png")
plt.tight_layout()
save_render_close(fig, mean_z_path, width=1150)


qc_df = summary_df.sort_values("elapsed_sec", ascending=False).reset_index(drop=True)

fig, ax = plt.subplots(figsize=(13, 5.8), facecolor=BLACK)
style_ax(
    ax,
    title="Colored-Noise-Surrogate PH Runtime QC",
    xlabel="Channel",
    ylabel="Elapsed seconds",
)

x = np.arange(len(qc_df))

ax.bar(
    x,
    qc_df["elapsed_sec"],
    color=CYAN_FILL,
    edgecolor=ACCENT,
    linewidth=0.9,
    alpha=0.82,
)

ax.set_xticks(x)
ax.set_xticklabels(qc_df["channel"], rotation=90)

for i, row in qc_df.iterrows():
    if i < 6:
        ax.text(
            i,
            row["elapsed_sec"],
            f"{row['elapsed_sec']:.0f}s",
            ha="center",
            va="bottom",
            color=ACCENT,
            fontsize=8,
        )

force_cyan_text(ax)

runtime_path = os.path.join(mode_out_dir, "12_runtime_qc.png")
plt.tight_layout()
save_render_close(fig, runtime_path, width=1150)

elapsed_all = time.perf_counter() - overall_t0

log("\n" + "=" * 72)
log(f"Finished H1 Colored-Noise-Surrogate Persistent Homology in {fmt_time(elapsed_all)}")
log(f"Channels completed: {len(channels_completed)}")
log(f"Colored-noise surrogates per channel: {N_COLORED_NOISE_SURROGATES}")
log(f"Beta clip: {BETA_CLIP}")
log(f"joblib backend={JOBLIB_BACKEND}, n_jobs={JOBLIB_N_JOBS}")
log(f"scipy.fft workers/job={FFT_WORKERS_PER_JOB}")
log(f"Output directory: {mode_out_dir}")
log(f"Summary CSV: {summary_csv_path}")
log(f"Colored-noise long table CSV: {colored_noise_long_csv_path}")
log(f"Regularized z signature CSV: {signature_csv_path}")
log(f"Signature similarity CSV: {signature_similarity_csv_path}")
log(f"Signature distance CSV: {signature_distance_csv_path}")
log(f"MDS CSV: {mds_csv_path}")
log(f"Regional summary CSV: {region_summary_csv_path}")
log(f"Regularized z info CSV: {regularized_z_info_csv_path}")
log(f"Within-metric contrast info CSV: {contrast_info_csv_path}")
log(f"NPZ: {npz_path}")
log("Saved plots:")
log("01_colored_noise_corrected_ph_score_ranking.png")
log("02_colored_noise_corrected_primary_metric_z_heatmap.png")
log("02_colored_noise_corrected_primary_metric_z_heatmap_FIXED_CONTRAST.png")
log("03_colored_noise_corrected_primary_metric_q_heatmap.png")
log("03_colored_noise_corrected_primary_metric_significance_effect_heatmap_fixed.png")
log("04_colored_noise_corrected_ph_signature_similarity_heatmap.png")
log("05_colored_noise_corrected_ph_signature_distance_heatmap.png")
log("06_mds_colored_noise_corrected_ph_signatures.png")
log("07_observed_vs_colored_noise_mean_total_lifetime.png")
log("07_observed_vs_colored_noise_mean_total_lifetime_FIXED.png")
if len(colored_noise_long_df) > 0 and "total_lifetime" in colored_noise_long_df.columns:
    log("08_colored_noise_null_distributions_top_channels.png")
log("09_scalp_map_colored_noise_corrected_ph_score.png")
log("10_regional_colored_noise_corrected_ph_score.png")
log("11_mean_primary_colored_noise_z_ranking.png")
log("12_runtime_qc.png")
log("=" * 72)
```

```python
import os
import glob
import time
import numpy as np

if not hasattr(np, "trapz"):
    np.trapz = np.trapezoid

import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patheffects as pe
from matplotlib.colors import LinearSegmentedColormap
from matplotlib.gridspec import GridSpec
from scipy.spatial.distance import pdist, squareform
from scipy.cluster.hierarchy import linkage, leaves_list
from scipy.stats import chi2

try:
    from scipy.linalg import eigh
    HAVE_SCIPY_EIGH = True
except Exception:
    HAVE_SCIPY_EIGH = False

try:
    from IPython.display import display, Image
    HAVE_IPYTHON = True
except Exception:
    HAVE_IPYTHON = False


base_dir = "/home/a/projects/Complete-Neural-Signal-Analysis"

FAST_OPTION = "ultra"

SAVE_PLOTS_TO_DISK = True
RENDER_ALL_IMAGES_INLINE = True
INLINE_IMAGE_WIDTH = 1100
PLOT_DPI = 150
VERBOSE_PROGRESS = True

out_dir = os.path.join(base_dir, "topology_persistence_significance_testing")
mode_out_dir = os.path.join(
    out_dir,
    f"h1_persistence_significance_testing_{FAST_OPTION}_multi_surrogate_consensus",
)
os.makedirs(mode_out_dir, exist_ok=True)

BLACK = "#000000"
ACCENT = "#00FFFF"
ACCENT_SOFT = "#66FFFF"
ACCENT_DIM = "#008B8B"
CYAN_FILL = "#00D5FF"
CYAN_FILL_DARK = "#005A63"

CYAN_SEQUENTIAL_CMAP = LinearSegmentedColormap.from_list(
    "cyan_sequential",
    [BLACK, CYAN_FILL_DARK, CYAN_FILL, ACCENT_SOFT],
    N=256,
)

CYAN_DIVERGING_CMAP = LinearSegmentedColormap.from_list(
    "cyan_diverging",
    [ACCENT_SOFT, BLACK, CYAN_FILL],
    N=256,
)

plt.rcParams.update({
    "figure.facecolor": BLACK,
    "axes.facecolor": BLACK,
    "savefig.facecolor": BLACK,
    "text.color": ACCENT,
    "axes.labelcolor": ACCENT,
    "axes.edgecolor": ACCENT,
    "axes.titlecolor": ACCENT,
    "xtick.color": ACCENT,
    "ytick.color": ACCENT,
    "grid.color": ACCENT_DIM,
    "grid.alpha": 0.25,
    "legend.facecolor": BLACK,
    "legend.edgecolor": ACCENT,
    "legend.labelcolor": ACCENT,
    "font.size": 11,
})


def log(msg):
    if VERBOSE_PROGRESS:
        print(msg, flush=True)


def style_ax(ax, title=None, xlabel=None, ylabel=None, grid=True):
    ax.set_facecolor(BLACK)

    for spine in ax.spines.values():
        spine.set_color(ACCENT)
        spine.set_linewidth(1.1)

    ax.tick_params(axis="x", colors=ACCENT, labelcolor=ACCENT)
    ax.tick_params(axis="y", colors=ACCENT, labelcolor=ACCENT)

    if title is not None:
        ax.set_title(title, color=ACCENT, pad=12)

    if xlabel is not None:
        ax.set_xlabel(xlabel, color=ACCENT, labelpad=10)

    if ylabel is not None:
        ax.set_ylabel(ylabel, color=ACCENT, labelpad=10)

    if grid:
        ax.grid(True, color=ACCENT_DIM, alpha=0.25, linewidth=0.8)


def force_cyan_text(ax):
    ax.title.set_color(ACCENT)
    ax.xaxis.label.set_color(ACCENT)
    ax.yaxis.label.set_color(ACCENT)

    for tick_label in ax.get_xticklabels():
        tick_label.set_color(ACCENT)

    for tick_label in ax.get_yticklabels():
        tick_label.set_color(ACCENT)


def style_colorbar(cbar, label):
    cbar.set_label(label, color=ACCENT)
    cbar.ax.yaxis.set_tick_params(color=ACCENT)
    plt.setp(cbar.ax.get_yticklabels(), color=ACCENT)
    cbar.outline.set_edgecolor(ACCENT)


def style_legend(legend):
    if legend is not None:
        legend.get_frame().set_facecolor(BLACK)
        legend.get_frame().set_edgecolor(ACCENT)

        for text in legend.get_texts():
            text.set_color(ACCENT)


def save_render_close(fig, path, width=INLINE_IMAGE_WIDTH):
    if SAVE_PLOTS_TO_DISK:
        fig.savefig(path, dpi=PLOT_DPI, bbox_inches="tight", facecolor=BLACK)

    plt.close(fig)

    if RENDER_ALL_IMAGES_INLINE and HAVE_IPYTHON and os.path.exists(path):
        display(Image(filename=path, width=width))


def bh_fdr(pvals):
    pvals = np.asarray(pvals, dtype=float)
    qvals = np.full_like(pvals, np.nan, dtype=float)

    valid = np.isfinite(pvals)
    p = pvals[valid]

    if len(p) == 0:
        return qvals

    order = np.argsort(p)
    ranked = p[order]
    m = len(ranked)

    raw_q = ranked * m / np.arange(1, m + 1)
    monotone_q = np.minimum.accumulate(raw_q[::-1])[::-1]
    monotone_q = np.clip(monotone_q, 0.0, 1.0)

    q_valid = np.empty_like(monotone_q)
    q_valid[order] = monotone_q

    qvals[valid] = q_valid

    return qvals


def percentile_score(values, higher_is_more=True):
    s = pd.Series(values, dtype=float)

    if not higher_is_more:
        s = -s

    if s.nunique(dropna=True) <= 1:
        return np.full(len(s), 50.0, dtype=float)

    return s.rank(method="average", pct=True).to_numpy(dtype=float) * 100.0


def offdiag_values(M):
    M = np.asarray(M, dtype=float)

    if M.shape[0] <= 1:
        return np.asarray([], dtype=float)

    mask = ~np.eye(M.shape[0], dtype=bool)
    vals = M[mask]
    vals = vals[np.isfinite(vals)]

    return vals


def dynamic_limits(values, lower_pct=2, upper_pct=98, pad_frac=0.08, lower_bound=None, upper_bound=None):
    values = np.asarray(values, dtype=float)
    values = values[np.isfinite(values)]

    if len(values) == 0:
        vmin, vmax = 0.0, 1.0
    else:
        vmin = float(np.nanpercentile(values, lower_pct))
        vmax = float(np.nanpercentile(values, upper_pct))

        if not np.isfinite(vmin):
            vmin = float(np.nanmin(values))

        if not np.isfinite(vmax):
            vmax = float(np.nanmax(values))

        if vmax <= vmin:
            center = float(np.nanmean(values))
            eps = max(1e-6, abs(center) * 0.01, 1e-5)
            vmin = center - eps
            vmax = center + eps

    pad = (vmax - vmin) * pad_frac
    vmin -= pad
    vmax += pad

    if lower_bound is not None:
        vmin = max(lower_bound, vmin)

    if upper_bound is not None:
        vmax = min(upper_bound, vmax)

    if vmax <= vmin:
        vmax = vmin + 1e-6

    return vmin, vmax


def robust_limits(values, lower_pct=2, upper_pct=98, pad_frac=0.08, include_zero=True):
    values = np.asarray(values, dtype=float)
    values = values[np.isfinite(values)]

    if len(values) == 0:
        return 0.0, 1.0

    vmin = float(np.nanpercentile(values, lower_pct))
    vmax = float(np.nanpercentile(values, upper_pct))

    if include_zero:
        vmin = min(vmin, 0.0)
        vmax = max(vmax, 0.0)

    if vmax <= vmin:
        center = float(np.nanmean(values))
        eps = max(abs(center) * 0.05, 1e-6)
        vmin = center - eps
        vmax = center + eps

    pad = (vmax - vmin) * pad_frac

    return vmin - pad, vmax + pad


def robust_metric_contrast_matrix(raw_matrix):
    raw_matrix = np.asarray(raw_matrix, dtype=float)
    out = np.zeros_like(raw_matrix, dtype=float)
    info = []

    for j in range(raw_matrix.shape[1]):
        col = raw_matrix[:, j]
        finite = col[np.isfinite(col)]

        if len(finite) == 0:
            out[:, j] = 0.0
            info.append({"metric_index": j, "center": np.nan, "scale": np.nan})
            continue

        center = float(np.nanmedian(finite))
        mad = float(np.nanmedian(np.abs(finite - center)))
        scale = 1.4826 * mad

        if not np.isfinite(scale) or scale < 1e-9:
            p10 = float(np.nanpercentile(finite, 10))
            p90 = float(np.nanpercentile(finite, 90))
            scale = (p90 - p10) / 2.563

        if not np.isfinite(scale) or scale < 1e-9:
            scale = float(np.nanstd(finite))

        if not np.isfinite(scale) or scale < 1e-9:
            scale = 1.0

        out[:, j] = (col - center) / scale

        info.append({
            "metric_index": j,
            "center": center,
            "scale": scale,
        })

    return out, pd.DataFrame(info)


def classical_mds_from_distances(D, n_components=2):
    D = np.asarray(D, dtype=float)
    n = D.shape[0]

    if n < 2:
        return np.zeros((n, n_components), dtype=float)

    D2 = D ** 2
    J = np.eye(n) - np.ones((n, n)) / n
    B = -0.5 * J @ D2 @ J

    if HAVE_SCIPY_EIGH:
        vals, vecs = eigh(B)
    else:
        vals, vecs = np.linalg.eigh(B)

    idx = np.argsort(vals)[::-1]
    vals = vals[idx]
    vecs = vecs[:, idx]

    coords = np.zeros((n, n_components), dtype=float)

    for k in range(min(n_components, n)):
        if vals[k] > 0:
            coords[:, k] = vecs[:, k] * np.sqrt(vals[k])

    return coords


def gaussian_similarity_from_distance(D, sigma=None):
    D = np.asarray(D, dtype=float)
    off = offdiag_values(D)

    if sigma is None:
        sigma = float(np.nanmedian(off)) if len(off) else 1.0

    sigma = max(float(sigma), 1e-8)

    S = np.exp(-(D ** 2) / (2.0 * sigma ** 2))
    np.fill_diagonal(S, 1.0)

    return S, sigma


def fisher_combine_pvalues(pvals):
    pvals = np.asarray(pvals, dtype=float)
    pvals = pvals[np.isfinite(pvals)]
    pvals = np.clip(pvals, 1e-300, 1.0)

    if len(pvals) == 0:
        return np.nan, np.nan, 0

    if len(pvals) == 1:
        return float(pvals[0]), float(-2.0 * np.log(pvals[0])), 1

    stat = float(-2.0 * np.sum(np.log(pvals)))
    p_combined = float(chi2.sf(stat, 2 * len(pvals)))

    return p_combined, stat, len(pvals)


def newest_file(patterns, exclude_fragments=()):
    candidates = []

    for pattern in patterns:
        candidates.extend(glob.glob(pattern, recursive=True))

    clean = []

    for path in candidates:
        name = os.path.basename(path).lower()

        if any(fragment.lower() in name for fragment in exclude_fragments):
            continue

        if os.path.isfile(path):
            clean.append(path)

    if len(clean) == 0:
        return None

    clean = sorted(clean, key=lambda p: os.path.getmtime(p), reverse=True)

    return clean[0]


channel_regions = {
    "Fp1": "Frontal", "Fpz": "Frontal", "Fp2": "Frontal",
    "F7": "Frontal", "F3": "Frontal", "Fz": "Frontal", "F4": "Frontal", "F8": "Frontal",
    "FC5": "Frontocentral", "FC1": "Frontocentral", "FC2": "Frontocentral", "FC6": "Frontocentral",
    "M1": "Mastoid", "M2": "Mastoid",
    "T7": "Central", "C3": "Central", "Cz": "Central", "C4": "Central", "T8": "Central",
    "CP5": "Centroparietal", "CP1": "Centroparietal", "CP2": "Centroparietal", "CP6": "Centroparietal",
    "P7": "Parietal", "P3": "Parietal", "Pz": "Parietal", "P4": "Parietal", "P8": "Parietal",
    "POz": "Occipital", "O1": "Occipital", "Oz": "Occipital", "O2": "Occipital",
}

scalp_xy = {
    "Fp1": (-0.45, 1.00), "Fpz": (0.00, 1.08), "Fp2": (0.45, 1.00),
    "F7": (-0.95, 0.60), "F3": (-0.45, 0.55), "Fz": (0.00, 0.60),
    "F4": (0.45, 0.55), "F8": (0.95, 0.60),
    "FC5": (-0.75, 0.25), "FC1": (-0.25, 0.25),
    "FC2": (0.25, 0.25), "FC6": (0.75, 0.25),
    "M1": (-1.15, 0.05), "T7": (-1.00, 0.00), "C3": (-0.50, 0.00),
    "Cz": (0.00, 0.00), "C4": (0.50, 0.00), "T8": (1.00, 0.00),
    "M2": (1.15, 0.05),
    "CP5": (-0.75, -0.30), "CP1": (-0.25, -0.30),
    "CP2": (0.25, -0.30), "CP6": (0.75, -0.30),
    "P7": (-0.90, -0.60), "P3": (-0.45, -0.60), "Pz": (0.00, -0.65),
    "P4": (0.45, -0.60), "P8": (0.90, -0.60),
    "POz": (0.00, -0.88),
    "O1": (-0.35, -1.00), "Oz": (0.00, -1.08), "O2": (0.35, -1.00),
}

primary_metric_names = [
    "total_lifetime",
    "max_lifetime",
    "q90_lifetime",
    "betti_auc",
    "betti_l2",
    "landscape_l2",
    "landscape_max",
]

primary_metric_labels = [
    "Total lifetime",
    "Max lifetime",
    "Q90 lifetime",
    "Betti AUC",
    "Betti L2",
    "Landscape L2",
    "Landscape max",
]

surrogate_specs = [
    {
        "family": "shuffle",
        "label": "Shuffle",
        "prefix_candidates": ["shuffle"],
        "root": os.path.join(base_dir, "topology_shuffle_surrogate_persistent_homology"),
        "summary_patterns": [
            os.path.join(base_dir, "topology_shuffle_surrogate_persistent_homology", "**", "h1_shuffle_surrogate_ph_summary.csv"),
            os.path.join(base_dir, "topology_shuffle_surrogate_persistent_homology", "**", "*shuffle*ph_summary.csv"),
            os.path.join(base_dir, "topology_shuffle_surrogate_persistent_homology", "**", "*summary*.csv"),
        ],
        "long_patterns": [
            os.path.join(base_dir, "topology_shuffle_surrogate_persistent_homology", "**", "*shuffle*long_table.csv"),
            os.path.join(base_dir, "topology_shuffle_surrogate_persistent_homology", "**", "*long_table.csv"),
        ],
    },
    {
        "family": "phase",
        "label": "Phase randomized",
        "prefix_candidates": ["phase"],
        "root": os.path.join(base_dir, "topology_phase_randomized_surrogate_persistent_homology"),
        "summary_patterns": [
            os.path.join(base_dir, "topology_phase_randomized_surrogate_persistent_homology", "**", "h1_phase_randomized_surrogate_ph_summary.csv"),
            os.path.join(base_dir, "topology_phase_randomized_surrogate_persistent_homology", "**", "*phase*ph_summary.csv"),
            os.path.join(base_dir, "topology_phase_randomized_surrogate_persistent_homology", "**", "*summary*.csv"),
        ],
        "long_patterns": [
            os.path.join(base_dir, "topology_phase_randomized_surrogate_persistent_homology", "**", "*phase*long_table.csv"),
            os.path.join(base_dir, "topology_phase_randomized_surrogate_persistent_homology", "**", "*long_table.csv"),
        ],
    },
    {
        "family": "iaaft",
        "label": "IAAFT",
        "prefix_candidates": ["iaaft"],
        "root": os.path.join(base_dir, "topology_iaaft_surrogate_persistent_homology"),
        "summary_patterns": [
            os.path.join(base_dir, "topology_iaaft_surrogate_persistent_homology", "**", "h1_iaaft_surrogate_ph_summary.csv"),
            os.path.join(base_dir, "topology_iaaft_surrogate_persistent_homology", "**", "*iaaft*ph_summary.csv"),
            os.path.join(base_dir, "topology_iaaft_surrogate_persistent_homology", "**", "*summary*.csv"),
        ],
        "long_patterns": [
            os.path.join(base_dir, "topology_iaaft_surrogate_persistent_homology", "**", "*iaaft*long_table.csv"),
            os.path.join(base_dir, "topology_iaaft_surrogate_persistent_homology", "**", "*long_table.csv"),
        ],
    },
    {
        "family": "colored_noise",
        "label": "Colored noise",
        "prefix_candidates": ["colored_noise"],
        "root": os.path.join(base_dir, "topology_colored_noise_surrogate_persistent_homology"),
        "summary_patterns": [
            os.path.join(base_dir, "topology_colored_noise_surrogate_persistent_homology", "**", "h1_colored_noise_surrogate_ph_summary.csv"),
            os.path.join(base_dir, "topology_colored_noise_surrogate_persistent_homology", "**", "*colored_noise*ph_summary.csv"),
            os.path.join(base_dir, "topology_colored_noise_surrogate_persistent_homology", "**", "*summary*.csv"),
        ],
        "long_patterns": [
            os.path.join(base_dir, "topology_colored_noise_surrogate_persistent_homology", "**", "*colored_noise*long_table.csv"),
            os.path.join(base_dir, "topology_colored_noise_surrogate_persistent_homology", "**", "*long_table.csv"),
        ],
    },
]

exclude_summary_fragments = [
    "region",
    "regional",
    "runtime",
    "diagnostic",
    "diagnostics",
    "signature",
    "distance",
    "similarity",
    "mds",
    "contrast",
    "fixed",
    "regularized",
    "matrix",
    "long",
    "plot",
]

loaded_families = []

log("\n" + "=" * 72)
log("Loading surrogate PH outputs for persistence significance testing")
log(f"Output directory: {mode_out_dir}")
log("=" * 72)

for spec in surrogate_specs:
    summary_path = newest_file(
        spec["summary_patterns"],
        exclude_fragments=exclude_summary_fragments,
    )

    long_path = newest_file(
        spec["long_patterns"],
        exclude_fragments=["summary", "region", "runtime", "diagnostic", "signature", "distance", "similarity", "mds", "matrix", "fixed", "contrast"],
    )

    if summary_path is None:
        log(f"  skipped {spec['label']}: no summary CSV found")
        continue

    try:
        summary_df = pd.read_csv(summary_path)
    except Exception as exc:
        log(f"  skipped {spec['label']}: could not read {summary_path} | {exc}")
        continue

    if "channel" not in summary_df.columns:
        log(f"  skipped {spec['label']}: summary has no channel column")
        continue

    summary_df["channel"] = summary_df["channel"].astype(str)

    long_df = pd.DataFrame()

    if long_path is not None:
        try:
            long_df = pd.read_csv(long_path)

            if "channel" in long_df.columns:
                long_df["channel"] = long_df["channel"].astype(str)
            else:
                long_df = pd.DataFrame()

        except Exception as exc:
            log(f"  warning {spec['label']}: could not read long table {long_path} | {exc}")
            long_df = pd.DataFrame()

    actual_prefix = None

    for prefix_candidate in spec["prefix_candidates"]:
        good = False

        for metric in primary_metric_names:
            if f"{prefix_candidate}_mean_{metric}" in summary_df.columns or f"{prefix_candidate}_p_greater_{metric}" in summary_df.columns:
                good = True
                break

        if good:
            actual_prefix = prefix_candidate
            break

    if actual_prefix is None:
        log(f"  skipped {spec['label']}: could not infer metric prefix")
        continue

    loaded_families.append({
        "family": spec["family"],
        "label": spec["label"],
        "prefix": actual_prefix,
        "summary_path": summary_path,
        "long_path": long_path,
        "summary_df": summary_df,
        "long_df": long_df,
    })

    log(f"  loaded {spec['label']}")
    log(f"    summary: {summary_path}")
    log(f"    long:    {long_path if long_path is not None else 'not found / not used'}")
    log(f"    rows:    summary={len(summary_df)}, long={len(long_df)}")

if len(loaded_families) == 0:
    raise RuntimeError(
        "No surrogate PH outputs were found. Run at least one surrogate PH notebook cell first "
        "(shuffle, phase-randomized, IAAFT, or colored-noise)."
    )


def get_long_values(long_df, channel, metric):
    if long_df is None or len(long_df) == 0:
        return np.asarray([], dtype=float)

    if "channel" not in long_df.columns or metric not in long_df.columns:
        return np.asarray([], dtype=float)

    vals = long_df.loc[long_df["channel"].astype(str) == str(channel), metric].to_numpy(dtype=float)
    vals = vals[np.isfinite(vals)]

    return vals


def compute_regularized_family_z(summary_df, long_df, prefix, metric):
    obs_col = f"observed_{metric}"
    mean_col = f"{prefix}_mean_{metric}"
    std_col = f"{prefix}_std_{metric}"

    obs = summary_df[obs_col].to_numpy(dtype=float)
    null_mean = summary_df[mean_col].to_numpy(dtype=float)

    if std_col in summary_df.columns:
        null_std = summary_df[std_col].to_numpy(dtype=float)
    else:
        null_std = np.zeros(len(summary_df), dtype=float)

    delta = obs - null_mean

    finite_std = null_std[np.isfinite(null_std) & (null_std > 0)]
    finite_delta = np.abs(delta[np.isfinite(delta)])

    null_vals_all = np.asarray([], dtype=float)

    if long_df is not None and len(long_df) > 0 and metric in long_df.columns:
        null_vals_all = long_df[metric].to_numpy(dtype=float)
        null_vals_all = null_vals_all[np.isfinite(null_vals_all)]

    floors = []

    if len(finite_std) > 0:
        floors.append(float(np.nanmedian(finite_std)) * 0.10)
        floors.append(float(np.nanpercentile(finite_std, 25)) * 0.25)

    if len(finite_delta) > 0:
        floors.append(float(np.nanmedian(finite_delta)) * 0.05)
        floors.append(float(np.nanpercentile(finite_delta, 75)) * 0.025)

    if len(null_vals_all) > 1:
        floors.append(float(np.nanstd(null_vals_all)) * 0.03)
        floors.append(float(np.nanmedian(np.abs(null_vals_all - np.nanmedian(null_vals_all)))) * 0.10)

    if np.any(np.isfinite(obs)):
        floors.append(float(np.nanmedian(np.abs(obs[np.isfinite(obs)]))) * 1e-6)

    if np.any(np.isfinite(null_mean)):
        floors.append(float(np.nanmedian(np.abs(null_mean[np.isfinite(null_mean)]))) * 1e-6)

    floors.append(1e-6)
    floors = [v for v in floors if np.isfinite(v) and v > 0]

    std_floor = max(floors) if len(floors) else 1e-6
    z = delta / np.maximum(null_std, std_floor)

    return z, std_floor


test_rows = []
family_metric_z_info = []

for family_data in loaded_families:
    family = family_data["family"]
    label = family_data["label"]
    prefix = family_data["prefix"]
    summary_df = family_data["summary_df"].copy()
    long_df = family_data["long_df"]

    for metric in primary_metric_names:
        obs_col = f"observed_{metric}"
        mean_col = f"{prefix}_mean_{metric}"
        std_col = f"{prefix}_std_{metric}"
        p_greater_col = f"{prefix}_p_greater_{metric}"
        p_lower_col = f"{prefix}_p_lower_{metric}"

        if obs_col not in summary_df.columns or mean_col not in summary_df.columns:
            log(f"  warning: {label} missing {metric}; skipping metric")
            continue

        z_regularized, std_floor = compute_regularized_family_z(
            summary_df,
            long_df,
            prefix,
            metric,
        )

        family_metric_z_info.append({
            "family": family,
            "family_label": label,
            "metric": metric,
            "std_floor": std_floor,
        })

        for row_idx, row in summary_df.iterrows():
            ch = str(row["channel"])
            region = row["region"] if "region" in summary_df.columns else channel_regions.get(ch, "Unknown")

            obs = float(row[obs_col])
            null_mean = float(row[mean_col])
            null_std = float(row[std_col]) if std_col in summary_df.columns and np.isfinite(row[std_col]) else np.nan
            delta = obs - null_mean

            null_vals = get_long_values(long_df, ch, metric)

            if len(null_vals) > 0:
                p_greater = float((1.0 + np.sum(null_vals >= obs)) / (len(null_vals) + 1.0))
                p_lower = float((1.0 + np.sum(null_vals <= obs)) / (len(null_vals) + 1.0))
                null_n = int(len(null_vals))
                null_source = "long_table_empirical"
            else:
                p_greater = float(row[p_greater_col]) if p_greater_col in summary_df.columns and np.isfinite(row[p_greater_col]) else np.nan
                p_lower = float(row[p_lower_col]) if p_lower_col in summary_df.columns and np.isfinite(row[p_lower_col]) else np.nan
                null_n = int(row[f"n_{prefix}_surrogates"]) if f"n_{prefix}_surrogates" in summary_df.columns and np.isfinite(row[f"n_{prefix}_surrogates"]) else np.nan
                null_source = "summary_existing"

            p_two = np.nan

            if np.isfinite(p_greater) and np.isfinite(p_lower):
                p_two = float(min(1.0, 2.0 * min(p_greater, p_lower)))

            z_raw_col = f"{prefix}_z_{metric}"

            if z_raw_col in summary_df.columns and np.isfinite(row[z_raw_col]):
                z_raw = float(row[z_raw_col])
            else:
                z_raw = float(delta / (null_std + 1e-12)) if np.isfinite(null_std) else np.nan

            test_rows.append({
                "family": family,
                "family_label": label,
                "prefix": prefix,
                "channel": ch,
                "region": region,
                "metric": metric,
                "metric_label": primary_metric_labels[primary_metric_names.index(metric)],
                "observed": obs,
                "null_mean": null_mean,
                "null_std": null_std,
                "delta": delta,
                "ratio": float(obs / (null_mean + 1e-12)),
                "z_raw": z_raw,
                "z_regularized": float(z_regularized[row_idx]),
                "std_floor_used": std_floor,
                "p_greater": p_greater,
                "p_lower": p_lower,
                "p_two_sided": p_two,
                "null_n": null_n,
                "null_source": null_source,
                "summary_path": family_data["summary_path"],
                "long_path": family_data["long_path"] if family_data["long_path"] is not None else "",
            })

results_df = pd.DataFrame(test_rows)

if len(results_df) == 0:
    raise RuntimeError("No channel-metric-family tests could be computed.")

results_df["q_greater"] = np.nan
results_df["q_two_sided"] = np.nan

for family in results_df["family"].unique():
    mask = results_df["family"] == family
    results_df.loc[mask, "q_greater"] = bh_fdr(results_df.loc[mask, "p_greater"].to_numpy(dtype=float))
    results_df.loc[mask, "q_two_sided"] = bh_fdr(results_df.loc[mask, "p_two_sided"].to_numpy(dtype=float))

results_df["neglog10_p_greater"] = -np.log10(results_df["p_greater"].to_numpy(dtype=float) + 1e-12)
results_df["neglog10_q_greater"] = -np.log10(results_df["q_greater"].to_numpy(dtype=float) + 1e-12)
results_df["is_q05_greater"] = results_df["q_greater"] < 0.05
results_df["is_q10_greater"] = results_df["q_greater"] < 0.10

family_metric_z_info_df = pd.DataFrame(family_metric_z_info)

consensus_rows = []

channels = sorted(results_df["channel"].unique().tolist())

for ch in channels:
    for metric, metric_label in zip(primary_metric_names, primary_metric_labels):
        block = results_df[(results_df["channel"] == ch) & (results_df["metric"] == metric)].copy()

        if len(block) == 0:
            continue

        pvals_greater = block["p_greater"].to_numpy(dtype=float)
        pvals_two = block["p_two_sided"].to_numpy(dtype=float)

        fisher_p_greater, fisher_stat_greater, fisher_k_greater = fisher_combine_pvalues(pvals_greater)
        fisher_p_two, fisher_stat_two, fisher_k_two = fisher_combine_pvalues(pvals_two)

        n_available = int(len(block))
        n_q05 = int(np.sum(block["q_greater"].to_numpy(dtype=float) < 0.05))
        n_q10 = int(np.sum(block["q_greater"].to_numpy(dtype=float) < 0.10))

        region = block["region"].iloc[0]

        consensus_rows.append({
            "channel": ch,
            "region": region,
            "metric": metric,
            "metric_label": metric_label,
            "n_surrogate_families_available": n_available,
            "n_family_q_greater_lt_0_05": n_q05,
            "n_family_q_greater_lt_0_10": n_q10,
            "family_agreement_fraction_q05": n_q05 / max(n_available, 1),
            "family_agreement_fraction_q10": n_q10 / max(n_available, 1),
            "mean_observed": float(np.nanmean(block["observed"])),
            "mean_null_mean": float(np.nanmean(block["null_mean"])),
            "mean_delta": float(np.nanmean(block["delta"])),
            "median_delta": float(np.nanmedian(block["delta"])),
            "mean_ratio": float(np.nanmean(block["ratio"])),
            "mean_z_regularized": float(np.nanmean(block["z_regularized"])),
            "median_z_regularized": float(np.nanmedian(block["z_regularized"])),
            "max_z_regularized": float(np.nanmax(block["z_regularized"])),
            "min_p_greater": float(np.nanmin(block["p_greater"])),
            "min_q_greater": float(np.nanmin(block["q_greater"])),
            "mean_neglog10_q_greater": float(np.nanmean(block["neglog10_q_greater"])),
            "fisher_p_greater": fisher_p_greater,
            "fisher_stat_greater": fisher_stat_greater,
            "fisher_k_greater": fisher_k_greater,
            "fisher_p_two_sided": fisher_p_two,
            "fisher_stat_two_sided": fisher_stat_two,
            "fisher_k_two_sided": fisher_k_two,
            "families_used": ",".join(block["family"].tolist()),
        })

consensus_df = pd.DataFrame(consensus_rows)

consensus_df["fisher_q_greater"] = bh_fdr(consensus_df["fisher_p_greater"].to_numpy(dtype=float))
consensus_df["fisher_q_two_sided"] = bh_fdr(consensus_df["fisher_p_two_sided"].to_numpy(dtype=float))
consensus_df["neglog10_fisher_q_greater"] = -np.log10(consensus_df["fisher_q_greater"].to_numpy(dtype=float) + 1e-12)
consensus_df["neglog10_fisher_p_greater"] = -np.log10(consensus_df["fisher_p_greater"].to_numpy(dtype=float) + 1e-12)

channel_summary_df = (
    consensus_df
    .groupby(["channel", "region"])
    .agg(
        n_metrics=("metric", "count"),
        n_metrics_fisher_q_lt_0_05=("fisher_q_greater", lambda x: int(np.sum(np.asarray(x, dtype=float) < 0.05))),
        n_metrics_fisher_q_lt_0_10=("fisher_q_greater", lambda x: int(np.sum(np.asarray(x, dtype=float) < 0.10))),
        mean_neglog10_fisher_q=("neglog10_fisher_q_greater", "mean"),
        max_neglog10_fisher_q=("neglog10_fisher_q_greater", "max"),
        mean_z_regularized=("mean_z_regularized", "mean"),
        median_z_regularized=("median_z_regularized", "median"),
        mean_family_agreement_q05=("family_agreement_fraction_q05", "mean"),
        max_family_agreement_q05=("family_agreement_fraction_q05", "max"),
        mean_delta=("mean_delta", "mean"),
        mean_ratio=("mean_ratio", "mean"),
        min_fisher_q=("fisher_q_greater", "min"),
    )
    .reset_index()
)

channel_summary_df["significance_component_q"] = percentile_score(channel_summary_df["mean_neglog10_fisher_q"], higher_is_more=True)
channel_summary_df["significance_component_agreement"] = percentile_score(channel_summary_df["mean_family_agreement_q05"], higher_is_more=True)
channel_summary_df["significance_component_z"] = percentile_score(channel_summary_df["mean_z_regularized"], higher_is_more=True)
channel_summary_df["significance_component_metric_count"] = percentile_score(channel_summary_df["n_metrics_fisher_q_lt_0_05"], higher_is_more=True)

channel_summary_df["persistence_significance_score"] = (
    0.42 * channel_summary_df["significance_component_q"]
    + 0.26 * channel_summary_df["significance_component_agreement"]
    + 0.22 * channel_summary_df["significance_component_z"]
    + 0.10 * channel_summary_df["significance_component_metric_count"]
)

channel_summary_df = channel_summary_df.sort_values("persistence_significance_score", ascending=False).reset_index(drop=True)

region_summary_df = (
    channel_summary_df
    .groupby("region")
    .agg(
        n_channels=("channel", "count"),
        mean_persistence_significance_score=("persistence_significance_score", "mean"),
        sem_persistence_significance_score=("persistence_significance_score", lambda x: float(np.std(x, ddof=1) / np.sqrt(len(x))) if len(x) > 1 else 0.0),
        mean_neglog10_fisher_q=("mean_neglog10_fisher_q", "mean"),
        mean_z_regularized=("mean_z_regularized", "mean"),
        mean_family_agreement_q05=("mean_family_agreement_q05", "mean"),
        mean_n_metrics_fisher_q_lt_0_05=("n_metrics_fisher_q_lt_0_05", "mean"),
    )
    .reset_index()
    .sort_values("mean_persistence_significance_score", ascending=False)
)

channel_order = channel_summary_df["channel"].tolist()

signature_feature_frames = []

for metric in primary_metric_names:
    metric_block = consensus_df[consensus_df["metric"] == metric].set_index("channel")
    metric_frame = pd.DataFrame(index=channel_order)

    metric_frame[f"{metric}_mean_z"] = metric_block.reindex(channel_order)["mean_z_regularized"].to_numpy(dtype=float)
    metric_frame[f"{metric}_neglog10_fisher_q"] = metric_block.reindex(channel_order)["neglog10_fisher_q_greater"].to_numpy(dtype=float)
    metric_frame[f"{metric}_agreement_q05"] = metric_block.reindex(channel_order)["family_agreement_fraction_q05"].to_numpy(dtype=float)

    signature_feature_frames.append(metric_frame)

signature_df = pd.concat(signature_feature_frames, axis=1)
signature_df = signature_df.replace([np.inf, -np.inf], np.nan).fillna(0.0)

signature_X = signature_df.to_numpy(dtype=float)
signature_X = signature_X - np.nanmean(signature_X, axis=0, keepdims=True)
signature_X = signature_X / (np.nanstd(signature_X, axis=0, keepdims=True) + 1e-12)
signature_X = np.nan_to_num(signature_X, nan=0.0, posinf=0.0, neginf=0.0)

if len(channel_order) >= 2:
    signature_distance = squareform(pdist(signature_X, metric="euclidean"))
else:
    signature_distance = np.zeros((len(channel_order), len(channel_order)), dtype=float)

signature_similarity, similarity_sigma = gaussian_similarity_from_distance(signature_distance)

if len(channel_order) >= 3 and np.any(signature_distance > 0):
    Z = linkage(squareform(signature_distance, checks=False), method="average", optimal_ordering=True)
    cluster_order_idx = leaves_list(Z)
else:
    cluster_order_idx = np.arange(len(channel_order))

clustered_channels = [channel_order[i] for i in cluster_order_idx]
clustered_similarity = signature_similarity[np.ix_(cluster_order_idx, cluster_order_idx)]
clustered_distance = signature_distance[np.ix_(cluster_order_idx, cluster_order_idx)]

mds_coords = classical_mds_from_distances(signature_distance, n_components=2)

mds_df = pd.DataFrame({
    "channel": channel_order,
    "region": [channel_summary_df.set_index("channel").loc[ch, "region"] for ch in channel_order],
    "persistence_significance_mds_1": mds_coords[:, 0],
    "persistence_significance_mds_2": mds_coords[:, 1],
    "persistence_significance_score": [channel_summary_df.set_index("channel").loc[ch, "persistence_significance_score"] for ch in channel_order],
    "mean_neglog10_fisher_q": [channel_summary_df.set_index("channel").loc[ch, "mean_neglog10_fisher_q"] for ch in channel_order],
})

family_channel_summary_df = (
    results_df
    .groupby(["family", "family_label", "channel", "region"])
    .agg(
        mean_z_regularized=("z_regularized", "mean"),
        mean_neglog10_q=("neglog10_q_greater", "mean"),
        n_metrics_q_lt_0_05=("q_greater", lambda x: int(np.sum(np.asarray(x, dtype=float) < 0.05))),
        min_q=("q_greater", "min"),
    )
    .reset_index()
)

family_channel_summary_df["family_channel_significance_score"] = (
    0.55 * percentile_score(family_channel_summary_df["mean_neglog10_q"], higher_is_more=True)
    + 0.30 * percentile_score(family_channel_summary_df["mean_z_regularized"], higher_is_more=True)
    + 0.15 * percentile_score(family_channel_summary_df["n_metrics_q_lt_0_05"], higher_is_more=True)
)

tests_csv_path = os.path.join(mode_out_dir, "h1_persistence_significance_channel_metric_family_tests.csv")
consensus_csv_path = os.path.join(mode_out_dir, "h1_persistence_significance_consensus_metric_tests.csv")
channel_summary_csv_path = os.path.join(mode_out_dir, "h1_persistence_significance_channel_summary.csv")
region_summary_csv_path = os.path.join(mode_out_dir, "h1_persistence_significance_region_summary.csv")
family_channel_csv_path = os.path.join(mode_out_dir, "h1_persistence_significance_family_channel_summary.csv")
signature_csv_path = os.path.join(mode_out_dir, "h1_persistence_significance_signature_matrix.csv")
signature_distance_csv_path = os.path.join(mode_out_dir, "h1_persistence_significance_signature_distance_matrix.csv")
signature_similarity_csv_path = os.path.join(mode_out_dir, "h1_persistence_significance_signature_similarity_matrix.csv")
mds_csv_path = os.path.join(mode_out_dir, "h1_persistence_significance_mds_coordinates.csv")
z_info_csv_path = os.path.join(mode_out_dir, "h1_persistence_significance_family_metric_z_regularization_info.csv")
npz_path = os.path.join(mode_out_dir, "h1_persistence_significance_outputs.npz")

results_df.to_csv(tests_csv_path, index=False)
consensus_df.to_csv(consensus_csv_path, index=False)
channel_summary_df.to_csv(channel_summary_csv_path, index=False)
region_summary_df.to_csv(region_summary_csv_path, index=False)
family_channel_summary_df.to_csv(family_channel_csv_path, index=False)
signature_df.to_csv(signature_csv_path)
pd.DataFrame(signature_distance, index=channel_order, columns=channel_order).to_csv(signature_distance_csv_path)
pd.DataFrame(signature_similarity, index=channel_order, columns=channel_order).to_csv(signature_similarity_csv_path)
mds_df.to_csv(mds_csv_path, index=False)
family_metric_z_info_df.to_csv(z_info_csv_path, index=False)

np.savez(
    npz_path,
    channels=np.asarray(channel_order),
    primary_metric_names=np.asarray(primary_metric_names),
    signature_matrix=signature_df.to_numpy(dtype=float),
    signature_distance_matrix=signature_distance,
    signature_similarity_matrix=signature_similarity,
    mds_coordinates=mds_coords,
    persistence_significance_score=channel_summary_df.set_index("channel").reindex(channel_order)["persistence_significance_score"].to_numpy(dtype=float),
    mean_neglog10_fisher_q=channel_summary_df.set_index("channel").reindex(channel_order)["mean_neglog10_fisher_q"].to_numpy(dtype=float),
    families_loaded=np.asarray([f["family"] for f in loaded_families]),
    family_labels_loaded=np.asarray([f["label"] for f in loaded_families]),
    similarity_sigma=np.asarray([similarity_sigma]),
)

print("\nPersistence significance tests by channel × metric × surrogate family:")
if HAVE_IPYTHON:
    display(results_df)
else:
    print(results_df)

print("\nConsensus persistence significance by channel × metric:")
if HAVE_IPYTHON:
    display(consensus_df)
else:
    print(consensus_df)

print("\nChannel-level persistence significance summary:")
if HAVE_IPYTHON:
    display(channel_summary_df)
else:
    print(channel_summary_df)

print("\nRegional persistence significance summary:")
if HAVE_IPYTHON:
    display(region_summary_df)
else:
    print(region_summary_df)


fig, ax = plt.subplots(figsize=(13, 6.2), facecolor=BLACK)
style_ax(
    ax,
    title=f"H1 Persistence Significance Score | {FAST_OPTION.upper()} | multi-surrogate consensus",
    xlabel="Channel",
    ylabel="Persistence significance score",
)

rank_df = channel_summary_df.sort_values("persistence_significance_score", ascending=False).reset_index(drop=True)
x = np.arange(len(rank_df))

ax.bar(
    x,
    rank_df["persistence_significance_score"],
    color=CYAN_FILL,
    edgecolor=ACCENT,
    linewidth=0.9,
    alpha=0.82,
)

ax.axhline(
    float(np.nanmean(rank_df["persistence_significance_score"])),
    color=ACCENT_SOFT,
    linestyle="--",
    linewidth=1.3,
    alpha=0.85,
    label=f"Mean={np.nanmean(rank_df['persistence_significance_score']):.1f}",
)

for i, row in rank_df.iterrows():
    if i < 8:
        ax.text(
            i,
            row["persistence_significance_score"],
            f"{row['persistence_significance_score']:.0f}",
            ha="center",
            va="bottom",
            color=ACCENT,
            fontsize=8,
        )

ax.set_xticks(x)
ax.set_xticklabels(rank_df["channel"], rotation=90)
ax.set_ylim(0, 105)

legend = ax.legend(loc="best", framealpha=1.0)
style_legend(legend)
force_cyan_text(ax)

plot_path = os.path.join(mode_out_dir, "01_persistence_significance_score_ranking.png")
plt.tight_layout()
save_render_close(fig, plot_path, width=1150)


consensus_neglogq = (
    consensus_df
    .pivot(index="channel", columns="metric", values="neglog10_fisher_q_greater")
    .reindex(index=channel_order, columns=primary_metric_names)
    .to_numpy(dtype=float)
)

consensus_q = (
    consensus_df
    .pivot(index="channel", columns="metric", values="fisher_q_greater")
    .reindex(index=channel_order, columns=primary_metric_names)
    .to_numpy(dtype=float)
)

agreement_counts = (
    consensus_df
    .pivot(index="channel", columns="metric", values="n_family_q_greater_lt_0_05")
    .reindex(index=channel_order, columns=primary_metric_names)
    .to_numpy(dtype=float)
)

available_counts = (
    consensus_df
    .pivot(index="channel", columns="metric", values="n_surrogate_families_available")
    .reindex(index=channel_order, columns=primary_metric_names)
    .to_numpy(dtype=float)
)

vmin, vmax = dynamic_limits(consensus_neglogq, lower_pct=1, upper_pct=98, pad_frac=0.05, lower_bound=0.0)

fig, ax = plt.subplots(figsize=(12.8, 9.2), facecolor=BLACK)
style_ax(
    ax,
    title="Consensus H1 Persistence Significance Heatmap",
    xlabel="H1 metric",
    ylabel="Channel",
)

im = ax.imshow(
    consensus_neglogq,
    aspect="auto",
    interpolation="nearest",
    cmap=CYAN_SEQUENTIAL_CMAP,
    vmin=vmin,
    vmax=vmax,
)

ax.set_yticks(np.arange(len(channel_order)))
ax.set_yticklabels(channel_order)
ax.set_xticks(np.arange(len(primary_metric_labels)))
ax.set_xticklabels(primary_metric_labels, rotation=45, ha="right")

ax.set_xticks(np.arange(-0.5, len(primary_metric_labels), 1), minor=True)
ax.set_yticks(np.arange(-0.5, len(channel_order), 1), minor=True)
ax.grid(which="minor", color=ACCENT_DIM, linewidth=0.55, alpha=0.35)
ax.tick_params(which="minor", bottom=False, left=False)

for i in range(consensus_neglogq.shape[0]):
    for j in range(consensus_neglogq.shape[1]):
        q_val = consensus_q[i, j]
        agree = agreement_counts[i, j]
        avail = available_counts[i, j]

        if not np.isfinite(q_val):
            continue

        show_text = q_val < 0.10 or i < 6 or i >= consensus_neglogq.shape[0] - 3

        if show_text:
            label = f"{agree:.0f}/{avail:.0f}\nq={q_val:.2g}"
            txt = ax.text(
                j,
                i,
                label,
                ha="center",
                va="center",
                color=ACCENT,
                fontsize=5.3,
                alpha=0.92,
                fontweight="bold",
            )
            txt.set_path_effects([pe.withStroke(linewidth=1.4, foreground=BLACK)])

cbar = plt.colorbar(im, ax=ax, shrink=0.85)
style_colorbar(cbar, "-log10 Fisher FDR q")

force_cyan_text(ax)

fig.text(
    0.09,
    0.035,
    "Cell text: number of surrogate families significant at q<0.05 / available families, plus Fisher-combined FDR q.",
    ha="left",
    va="bottom",
    color=ACCENT,
    fontsize=9,
)

plot_path = os.path.join(mode_out_dir, "02_consensus_persistence_significance_heatmap.png")
plt.tight_layout(rect=[0.0, 0.055, 1.0, 1.0])
save_render_close(fig, plot_path, width=1050)


agreement_fraction = (
    consensus_df
    .pivot(index="channel", columns="metric", values="family_agreement_fraction_q05")
    .reindex(index=channel_order, columns=primary_metric_names)
    .to_numpy(dtype=float)
)

fig, ax = plt.subplots(figsize=(12.8, 9.2), facecolor=BLACK)
style_ax(
    ax,
    title="Surrogate-Family Agreement Heatmap",
    xlabel="H1 metric",
    ylabel="Channel",
)

im = ax.imshow(
    agreement_fraction,
    aspect="auto",
    interpolation="nearest",
    cmap=CYAN_SEQUENTIAL_CMAP,
    vmin=0.0,
    vmax=1.0,
)

ax.set_yticks(np.arange(len(channel_order)))
ax.set_yticklabels(channel_order)
ax.set_xticks(np.arange(len(primary_metric_labels)))
ax.set_xticklabels(primary_metric_labels, rotation=45, ha="right")

ax.set_xticks(np.arange(-0.5, len(primary_metric_labels), 1), minor=True)
ax.set_yticks(np.arange(-0.5, len(channel_order), 1), minor=True)
ax.grid(which="minor", color=ACCENT_DIM, linewidth=0.55, alpha=0.35)
ax.tick_params(which="minor", bottom=False, left=False)

for i in range(agreement_fraction.shape[0]):
    for j in range(agreement_fraction.shape[1]):
        agree = agreement_counts[i, j]
        avail = available_counts[i, j]

        if np.isfinite(agree) and np.isfinite(avail):
            txt = ax.text(
                j,
                i,
                f"{agree:.0f}/{avail:.0f}",
                ha="center",
                va="center",
                color=ACCENT,
                fontsize=5.6,
                alpha=0.92,
                fontweight="bold",
            )
            txt.set_path_effects([pe.withStroke(linewidth=1.4, foreground=BLACK)])

cbar = plt.colorbar(im, ax=ax, shrink=0.85)
style_colorbar(cbar, "Fraction of surrogate families q<0.05")
force_cyan_text(ax)

plot_path = os.path.join(mode_out_dir, "03_surrogate_family_agreement_heatmap.png")
plt.tight_layout()
save_render_close(fig, plot_path, width=1050)


consensus_z = (
    consensus_df
    .pivot(index="channel", columns="metric", values="mean_z_regularized")
    .reindex(index=channel_order, columns=primary_metric_names)
    .to_numpy(dtype=float)
)

z_relative, z_contrast_info_df = robust_metric_contrast_matrix(consensus_z)
z_relative_clipped = np.clip(z_relative, -3.0, 3.0)

fig, ax = plt.subplots(figsize=(12.8, 9.2), facecolor=BLACK)
style_ax(
    ax,
    title="Consensus H1 Persistence Effect Z-Score Heatmap | fixed contrast",
    xlabel="H1 metric",
    ylabel="Channel",
)

im = ax.imshow(
    z_relative_clipped,
    aspect="auto",
    interpolation="nearest",
    cmap=CYAN_DIVERGING_CMAP,
    vmin=-3,
    vmax=3,
)

ax.set_yticks(np.arange(len(channel_order)))
ax.set_yticklabels(channel_order)
ax.set_xticks(np.arange(len(primary_metric_labels)))
ax.set_xticklabels(primary_metric_labels, rotation=45, ha="right")

ax.set_xticks(np.arange(-0.5, len(primary_metric_labels), 1), minor=True)
ax.set_yticks(np.arange(-0.5, len(channel_order), 1), minor=True)
ax.grid(which="minor", color=ACCENT_DIM, linewidth=0.55, alpha=0.35)
ax.tick_params(which="minor", bottom=False, left=False)

for i in range(consensus_z.shape[0]):
    for j in range(consensus_z.shape[1]):
        raw_val = consensus_z[i, j]
        rel_val = z_relative[i, j]

        if not np.isfinite(raw_val):
            continue

        show_text = abs(rel_val) >= 1.15 or i < 5 or i >= consensus_z.shape[0] - 3

        if show_text:
            txt = ax.text(
                j,
                i,
                f"{raw_val:+.1f}",
                ha="center",
                va="center",
                color=ACCENT,
                fontsize=5.8,
                alpha=0.95,
                fontweight="bold",
            )
            txt.set_path_effects([pe.withStroke(linewidth=1.4, foreground=BLACK)])

cbar = plt.colorbar(im, ax=ax, shrink=0.85)
style_colorbar(cbar, "Within-metric channel contrast")

force_cyan_text(ax)

fig.text(
    0.09,
    0.035,
    "Fixed view: color shows within-metric channel contrast; printed values are mean regularized z-scores across surrogate families.",
    ha="left",
    va="bottom",
    color=ACCENT,
    fontsize=9,
)

plot_path = os.path.join(mode_out_dir, "04_consensus_persistence_effect_z_heatmap.png")
plt.tight_layout(rect=[0.0, 0.055, 1.0, 1.0])
save_render_close(fig, plot_path, width=1050)


family_score_matrix = (
    family_channel_summary_df
    .pivot(index="channel", columns="family_label", values="family_channel_significance_score")
    .reindex(index=channel_order)
)

fig, ax = plt.subplots(figsize=(10.8, 9.2), facecolor=BLACK)
style_ax(
    ax,
    title="Persistence Significance Score by Surrogate Family",
    xlabel="Surrogate family",
    ylabel="Channel",
)

family_score_values = family_score_matrix.to_numpy(dtype=float)

im = ax.imshow(
    family_score_values,
    aspect="auto",
    interpolation="nearest",
    cmap=CYAN_SEQUENTIAL_CMAP,
    vmin=0,
    vmax=100,
)

ax.set_yticks(np.arange(len(channel_order)))
ax.set_yticklabels(channel_order)
ax.set_xticks(np.arange(len(family_score_matrix.columns)))
ax.set_xticklabels(list(family_score_matrix.columns), rotation=35, ha="right")

for i in range(family_score_values.shape[0]):
    for j in range(family_score_values.shape[1]):
        val = family_score_values[i, j]

        if np.isfinite(val) and (val >= 80 or val <= 20 or i < 5):
            txt = ax.text(
                j,
                i,
                f"{val:.0f}",
                ha="center",
                va="center",
                color=ACCENT,
                fontsize=6.0,
                alpha=0.95,
                fontweight="bold",
            )
            txt.set_path_effects([pe.withStroke(linewidth=1.4, foreground=BLACK)])

cbar = plt.colorbar(im, ax=ax, shrink=0.85)
style_colorbar(cbar, "Family-specific significance score")
force_cyan_text(ax)

plot_path = os.path.join(mode_out_dir, "05_surrogate_family_channel_score_heatmap.png")
plt.tight_layout()
save_render_close(fig, plot_path, width=1000)


total_lifetime_delta = (
    results_df[results_df["metric"] == "total_lifetime"]
    .pivot(index="channel", columns="family_label", values="delta")
    .reindex(index=channel_order)
)

total_lifetime_delta_matrix = total_lifetime_delta.to_numpy(dtype=float)
delta_relative, delta_contrast_info_df = robust_metric_contrast_matrix(total_lifetime_delta_matrix)
delta_relative_clipped = np.clip(delta_relative, -3.0, 3.0)

fig, ax = plt.subplots(figsize=(10.8, 9.2), facecolor=BLACK)
style_ax(
    ax,
    title="Observed − Surrogate Mean H1 Total Lifetime",
    xlabel="Surrogate family",
    ylabel="Channel",
)

im = ax.imshow(
    delta_relative_clipped,
    aspect="auto",
    interpolation="nearest",
    cmap=CYAN_DIVERGING_CMAP,
    vmin=-3,
    vmax=3,
)

ax.set_yticks(np.arange(len(channel_order)))
ax.set_yticklabels(channel_order)
ax.set_xticks(np.arange(len(total_lifetime_delta.columns)))
ax.set_xticklabels(list(total_lifetime_delta.columns), rotation=35, ha="right")

for i in range(total_lifetime_delta_matrix.shape[0]):
    for j in range(total_lifetime_delta_matrix.shape[1]):
        val = total_lifetime_delta_matrix[i, j]
        rel = delta_relative[i, j]

        if np.isfinite(val) and (abs(rel) >= 1.15 or i < 5 or i >= total_lifetime_delta_matrix.shape[0] - 3):
            txt = ax.text(
                j,
                i,
                f"{val:+.2f}",
                ha="center",
                va="center",
                color=ACCENT,
                fontsize=5.8,
                alpha=0.95,
                fontweight="bold",
            )
            txt.set_path_effects([pe.withStroke(linewidth=1.4, foreground=BLACK)])

cbar = plt.colorbar(im, ax=ax, shrink=0.85)
style_colorbar(cbar, "Within-family delta contrast")
force_cyan_text(ax)

fig.text(
    0.09,
    0.035,
    "Printed values are raw observed minus surrogate-mean H1 total lifetime. Color is contrast-scaled within each surrogate family.",
    ha="left",
    va="bottom",
    color=ACCENT,
    fontsize=9,
)

plot_path = os.path.join(mode_out_dir, "06_total_lifetime_observed_minus_surrogate_heatmap.png")
plt.tight_layout(rect=[0.0, 0.055, 1.0, 1.0])
save_render_close(fig, plot_path, width=1000)


top_tests_df = results_df.sort_values("neglog10_q_greater", ascending=False).head(28).copy()
top_tests_df["label"] = (
    top_tests_df["channel"]
    + " | "
    + top_tests_df["metric_label"]
    + " | "
    + top_tests_df["family_label"]
)

top_tests_df = top_tests_df.sort_values("neglog10_q_greater", ascending=True).reset_index(drop=True)

fig, ax = plt.subplots(figsize=(13, 9.0), facecolor=BLACK)
style_ax(
    ax,
    title="Top Individual Persistence Significance Tests",
    xlabel="-log10 family FDR q",
    ylabel="Channel | metric | surrogate family",
)

y = np.arange(len(top_tests_df))

ax.barh(
    y,
    top_tests_df["neglog10_q_greater"],
    color=CYAN_FILL,
    edgecolor=ACCENT,
    linewidth=0.8,
    alpha=0.82,
)

ax.set_yticks(y)
ax.set_yticklabels(top_tests_df["label"], fontsize=8)

for yy, (_, row) in zip(y, top_tests_df.iterrows()):
    ax.text(
        row["neglog10_q_greater"] + 0.03,
        yy,
        f"z={row['z_regularized']:+.1f}",
        ha="left",
        va="center",
        color=ACCENT,
        fontsize=7,
    )

force_cyan_text(ax)

plot_path = os.path.join(mode_out_dir, "07_top_individual_persistence_significance_tests.png")
plt.tight_layout()
save_render_close(fig, plot_path, width=1150)


sim_vmin, sim_vmax = dynamic_limits(
    offdiag_values(clustered_similarity),
    lower_pct=2,
    upper_pct=98,
    pad_frac=0.10,
    lower_bound=0.0,
    upper_bound=1.0,
)

fig, ax = plt.subplots(figsize=(9.6, 8.8), facecolor=BLACK)
style_ax(
    ax,
    title="Persistence Significance Signature Similarity Matrix",
    xlabel="Channel",
    ylabel="Channel",
)

im = ax.imshow(
    clustered_similarity,
    aspect="equal",
    interpolation="nearest",
    cmap=CYAN_SEQUENTIAL_CMAP,
    vmin=sim_vmin,
    vmax=sim_vmax,
)

ax.set_xticks(np.arange(len(clustered_channels)))
ax.set_xticklabels(clustered_channels, rotation=90)
ax.set_yticks(np.arange(len(clustered_channels)))
ax.set_yticklabels(clustered_channels)

cbar = plt.colorbar(im, ax=ax, shrink=0.85)
style_colorbar(cbar, "Consensus significance signature similarity")
force_cyan_text(ax)

plot_path = os.path.join(mode_out_dir, "08_persistence_significance_signature_similarity_heatmap.png")
plt.tight_layout()
save_render_close(fig, plot_path, width=950)


dist_vmin, dist_vmax = dynamic_limits(
    offdiag_values(clustered_distance),
    lower_pct=2,
    upper_pct=98,
    pad_frac=0.10,
    lower_bound=0.0,
)

fig, ax = plt.subplots(figsize=(9.6, 8.8), facecolor=BLACK)
style_ax(
    ax,
    title="Persistence Significance Signature Distance Matrix",
    xlabel="Channel",
    ylabel="Channel",
)

im = ax.imshow(
    clustered_distance,
    aspect="equal",
    interpolation="nearest",
    cmap=CYAN_SEQUENTIAL_CMAP,
    vmin=dist_vmin,
    vmax=dist_vmax,
)

ax.set_xticks(np.arange(len(clustered_channels)))
ax.set_xticklabels(clustered_channels, rotation=90)
ax.set_yticks(np.arange(len(clustered_channels)))
ax.set_yticklabels(clustered_channels)

cbar = plt.colorbar(im, ax=ax, shrink=0.85)
style_colorbar(cbar, "Consensus significance signature distance")
force_cyan_text(ax)

plot_path = os.path.join(mode_out_dir, "09_persistence_significance_signature_distance_heatmap.png")
plt.tight_layout()
save_render_close(fig, plot_path, width=950)


fig, ax = plt.subplots(figsize=(8.8, 7.4), facecolor=BLACK)
style_ax(
    ax,
    title="MDS of Persistence Significance Signatures",
    xlabel="Persistence significance MDS 1",
    ylabel="Persistence significance MDS 2",
)

sc = ax.scatter(
    mds_df["persistence_significance_mds_1"],
    mds_df["persistence_significance_mds_2"],
    c=mds_df["persistence_significance_score"],
    s=165,
    cmap=CYAN_SEQUENTIAL_CMAP,
    vmin=0,
    vmax=100,
    edgecolor=ACCENT,
    linewidth=1.2,
    alpha=0.95,
)

for _, row in mds_df.iterrows():
    txt = ax.text(
        row["persistence_significance_mds_1"],
        row["persistence_significance_mds_2"],
        row["channel"],
        fontsize=8,
        color=ACCENT,
        alpha=0.95,
        zorder=5,
    )
    txt.set_path_effects([pe.withStroke(linewidth=2.0, foreground=BLACK)])

cbar = plt.colorbar(sc, ax=ax, shrink=0.82)
style_colorbar(cbar, "Persistence significance score")
force_cyan_text(ax)

plot_path = os.path.join(mode_out_dir, "10_mds_persistence_significance_signatures.png")
plt.tight_layout()
save_render_close(fig, plot_path, width=950)


fig = plt.figure(figsize=(13.8, 8.2), facecolor=BLACK)
gs = GridSpec(
    1,
    2,
    width_ratios=[1.25, 0.85],
    wspace=0.25,
    figure=fig,
)

ax = fig.add_subplot(gs[0, 0])
rank_ax = fig.add_subplot(gs[0, 1])

style_ax(
    ax,
    title="Scalp Map of H1 Persistence Significance Score",
    grid=False,
)

head = plt.Circle((0, 0), 1.12, fill=False, color=ACCENT, linewidth=2.2)
ax.add_patch(head)
ax.plot([0.00, -0.10, 0.10, 0.00], [1.12, 1.28, 1.28, 1.12], color=ACCENT, linewidth=2.2)

scalp_df = channel_summary_df[channel_summary_df["channel"].isin(scalp_xy)].copy()
scalp_df["x"] = scalp_df["channel"].map(lambda ch: scalp_xy[ch][0])
scalp_df["y"] = scalp_df["channel"].map(lambda ch: scalp_xy[ch][1])

scores = scalp_df["persistence_significance_score"].to_numpy(dtype=float)

if np.nanmax(scores) > np.nanmin(scores):
    norm_scores = (scores - np.nanmin(scores)) / (np.nanmax(scores) - np.nanmin(scores) + 1e-12)
else:
    norm_scores = np.full(len(scores), 0.5)

scalp_df["marker_size"] = 90.0 + 520.0 * norm_scores

scatter = ax.scatter(
    scalp_df["x"],
    scalp_df["y"],
    c=scalp_df["persistence_significance_score"],
    s=scalp_df["marker_size"],
    cmap=CYAN_SEQUENTIAL_CMAP,
    vmin=0,
    vmax=100,
    edgecolor=ACCENT,
    linewidth=1.25,
    alpha=0.92,
    zorder=4,
)

top_score_channels = scalp_df.sort_values("persistence_significance_score", ascending=False).head(8)["channel"].tolist()

for _, row in scalp_df.iterrows():
    ch = row["channel"]
    x0 = float(row["x"])
    y0 = float(row["y"])
    score = float(row["persistence_significance_score"])

    txt = ax.text(
        x0,
        y0,
        ch,
        ha="center",
        va="center",
        fontsize=8,
        color=ACCENT,
        fontweight="bold",
        zorder=6,
    )
    txt.set_path_effects([pe.withStroke(linewidth=2.1, foreground=BLACK)])

    if ch in top_score_channels:
        ax.scatter(
            [x0],
            [y0],
            s=float(row["marker_size"]) * 1.75,
            facecolors="none",
            edgecolors=ACCENT_SOFT,
            linewidth=1.8,
            alpha=0.95,
            zorder=5,
        )

        dx = 0.06 if x0 <= 0 else -0.06
        ha = "left" if x0 <= 0 else "right"

        ann = ax.text(
            x0 + dx,
            y0 + 0.055,
            f"{score:.0f}",
            ha=ha,
            va="center",
            fontsize=8,
            color=ACCENT_SOFT,
            fontweight="bold",
            zorder=7,
        )
        ann.set_path_effects([pe.withStroke(linewidth=2.0, foreground=BLACK)])

ax.text(
    0,
    -1.30,
    "Bigger/brighter = stronger consensus persistence significance across surrogate nulls",
    ha="center",
    va="top",
    color=ACCENT,
    fontsize=9,
)

ax.set_xlim(-1.38, 1.38)
ax.set_ylim(-1.40, 1.36)
ax.set_aspect("equal")
ax.set_xticks([])
ax.set_yticks([])

cbar = plt.colorbar(scatter, ax=ax, shrink=0.78, pad=0.03)
style_colorbar(cbar, "Persistence significance score")

rank_side_df = scalp_df.sort_values("persistence_significance_score", ascending=True).copy()

style_ax(
    rank_ax,
    title="Persistence Significance Ranking",
    xlabel="Significance score",
    ylabel="Channel",
)

y = np.arange(len(rank_side_df))

rank_ax.barh(
    y,
    rank_side_df["persistence_significance_score"],
    color=CYAN_FILL,
    edgecolor=ACCENT,
    linewidth=0.8,
    alpha=0.82,
)

rank_ax.set_yticks(y)
rank_ax.set_yticklabels(rank_side_df["channel"])
rank_ax.set_xlim(0, 105)

rank_ax.axvline(
    float(np.nanmean(rank_side_df["persistence_significance_score"])),
    color=ACCENT_SOFT,
    linestyle="--",
    linewidth=1.2,
    alpha=0.85,
    label=f"Mean={np.nanmean(rank_side_df['persistence_significance_score']):.1f}",
)

for yy, (_, row) in zip(y, rank_side_df.iterrows()):
    score = float(row["persistence_significance_score"])

    rank_ax.text(
        score + 1.2,
        yy,
        f"{score:.0f}",
        va="center",
        ha="left",
        color=ACCENT,
        fontsize=7,
        alpha=0.82,
    )

legend = rank_ax.legend(loc="lower right", framealpha=1.0, fontsize=8)
style_legend(legend)
force_cyan_text(rank_ax)

top_text = "Top consensus: " + ", ".join(
    [
        f"{row.channel}({row.persistence_significance_score:.0f})"
        for row in scalp_df.sort_values("persistence_significance_score", ascending=False).head(5).itertuples()
    ]
)

bottom_text = "Lowest consensus: " + ", ".join(
    [
        f"{row.channel}({row.persistence_significance_score:.0f})"
        for row in scalp_df.sort_values("persistence_significance_score", ascending=False).tail(5).itertuples()
    ]
)

fig.text(
    0.5,
    0.018,
    top_text + "    |    " + bottom_text,
    ha="center",
    va="bottom",
    color=ACCENT,
    fontsize=9,
)

plot_path = os.path.join(mode_out_dir, "11_scalp_map_persistence_significance_score.png")
plt.tight_layout(rect=[0.0, 0.045, 1.0, 1.0])
save_render_close(fig, plot_path, width=1150)


fig, ax = plt.subplots(figsize=(12, 6.2), facecolor=BLACK)
style_ax(
    ax,
    title="Regional H1 Persistence Significance Score",
    xlabel="Region",
    ylabel="Mean persistence significance score",
)

x = np.arange(len(region_summary_df))

ax.bar(
    x,
    region_summary_df["mean_persistence_significance_score"],
    yerr=region_summary_df["sem_persistence_significance_score"],
    color=CYAN_FILL,
    edgecolor=ACCENT,
    linewidth=0.9,
    alpha=0.82,
    capsize=7,
    error_kw={"elinewidth": 1.5, "capthick": 1.5, "ecolor": ACCENT},
)

for i, row in region_summary_df.iterrows():
    ax.text(
        i,
        row["mean_persistence_significance_score"] + row["sem_persistence_significance_score"],
        f"{row['mean_persistence_significance_score']:.1f}\nn={int(row['n_channels'])}",
        ha="center",
        va="bottom",
        color=ACCENT,
        fontsize=9,
    )

ax.set_xticks(x)
ax.set_xticklabels(region_summary_df["region"], rotation=35, ha="right")
ax.set_ylim(0, 105)
force_cyan_text(ax)

plot_path = os.path.join(mode_out_dir, "12_regional_persistence_significance_score.png")
plt.tight_layout()
save_render_close(fig, plot_path, width=1050)


coverage_rows = []

for family_data in loaded_families:
    family = family_data["family"]
    label = family_data["label"]
    summary_df = family_data["summary_df"]
    long_df = family_data["long_df"]

    elapsed_mean = float(np.nanmean(summary_df["elapsed_sec"])) if "elapsed_sec" in summary_df.columns else np.nan
    elapsed_sum = float(np.nansum(summary_df["elapsed_sec"])) if "elapsed_sec" in summary_df.columns else np.nan

    coverage_rows.append({
        "family": family,
        "family_label": label,
        "n_summary_channels": int(summary_df["channel"].nunique()),
        "n_long_rows": int(len(long_df)),
        "mean_elapsed_sec": elapsed_mean,
        "total_elapsed_sec": elapsed_sum,
    })

coverage_df = pd.DataFrame(coverage_rows)
coverage_csv_path = os.path.join(mode_out_dir, "h1_persistence_significance_input_coverage.csv")
coverage_df.to_csv(coverage_csv_path, index=False)

fig, ax = plt.subplots(figsize=(11.5, 6.2), facecolor=BLACK)
style_ax(
    ax,
    title="Persistence Significance Input Coverage",
    xlabel="Surrogate family",
    ylabel="Count",
)

x = np.arange(len(coverage_df))
width = 0.36

ax.bar(
    x - width / 2,
    coverage_df["n_summary_channels"],
    width=width,
    color=CYAN_FILL,
    edgecolor=ACCENT,
    linewidth=0.9,
    alpha=0.82,
    label="Summary channels",
)

ax.bar(
    x + width / 2,
    coverage_df["n_long_rows"],
    width=width,
    color=BLACK,
    edgecolor=ACCENT_SOFT,
    linewidth=1.4,
    alpha=0.92,
    label="Long-table rows",
)

ax.set_xticks(x)
ax.set_xticklabels(coverage_df["family_label"], rotation=25, ha="right")

for i, row in coverage_df.iterrows():
    ax.text(
        i - width / 2,
        row["n_summary_channels"],
        f"{int(row['n_summary_channels'])}",
        ha="center",
        va="bottom",
        color=ACCENT,
        fontsize=9,
    )

    ax.text(
        i + width / 2,
        row["n_long_rows"],
        f"{int(row['n_long_rows'])}",
        ha="center",
        va="bottom",
        color=ACCENT,
        fontsize=9,
    )

legend = ax.legend(loc="best", framealpha=1.0)
style_legend(legend)
force_cyan_text(ax)

plot_path = os.path.join(mode_out_dir, "13_surrogate_input_coverage_qc.png")
plt.tight_layout()
save_render_close(fig, plot_path, width=1000)


runtime_available = coverage_df["mean_elapsed_sec"].notna().any()

if runtime_available:
    fig, ax = plt.subplots(figsize=(11.5, 6.2), facecolor=BLACK)
    style_ax(
        ax,
        title="Surrogate PH Runtime Summary",
        xlabel="Surrogate family",
        ylabel="Elapsed seconds",
    )

    x = np.arange(len(coverage_df))

    ax.bar(
        x,
        coverage_df["mean_elapsed_sec"],
        color=CYAN_FILL,
        edgecolor=ACCENT,
        linewidth=0.9,
        alpha=0.82,
        label="Mean per channel",
    )

    for i, row in coverage_df.iterrows():
        if np.isfinite(row["mean_elapsed_sec"]):
            ax.text(
                i,
                row["mean_elapsed_sec"],
                f"{row['mean_elapsed_sec']:.1f}s",
                ha="center",
                va="bottom",
                color=ACCENT,
                fontsize=9,
            )

    ax.set_xticks(x)
    ax.set_xticklabels(coverage_df["family_label"], rotation=25, ha="right")

    legend = ax.legend(loc="best", framealpha=1.0)
    style_legend(legend)
    force_cyan_text(ax)

    plot_path = os.path.join(mode_out_dir, "14_surrogate_runtime_qc.png")
    plt.tight_layout()
    save_render_close(fig, plot_path, width=1000)


log("\n" + "=" * 72)
log("Finished H1 Persistence Significance Testing")
log(f"Surrogate families loaded: {', '.join([f['label'] for f in loaded_families])}")
log(f"Output directory: {mode_out_dir}")
log("Saved CSVs:")
log(f"  {tests_csv_path}")
log(f"  {consensus_csv_path}")
log(f"  {channel_summary_csv_path}")
log(f"  {region_summary_csv_path}")
log(f"  {family_channel_csv_path}")
log(f"  {signature_csv_path}")
log(f"  {signature_similarity_csv_path}")
log(f"  {signature_distance_csv_path}")
log(f"  {mds_csv_path}")
log(f"  {coverage_csv_path}")
log(f"Saved NPZ: {npz_path}")
log("Saved plots:")
log("01_persistence_significance_score_ranking.png")
log("02_consensus_persistence_significance_heatmap.png")
log("03_surrogate_family_agreement_heatmap.png")
log("04_consensus_persistence_effect_z_heatmap.png")
log("05_surrogate_family_channel_score_heatmap.png")
log("06_total_lifetime_observed_minus_surrogate_heatmap.png")
log("07_top_individual_persistence_significance_tests.png")
log("08_persistence_significance_signature_similarity_heatmap.png")
log("09_persistence_significance_signature_distance_heatmap.png")
log("10_mds_persistence_significance_signatures.png")
log("11_scalp_map_persistence_significance_score.png")
log("12_regional_persistence_significance_score.png")
log("13_surrogate_input_coverage_qc.png")
if runtime_available:
    log("14_surrogate_runtime_qc.png")
log("=" * 72)
```
