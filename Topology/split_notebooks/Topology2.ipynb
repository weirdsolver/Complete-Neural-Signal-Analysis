```python
import os
import sys
import time
import numpy as np

# NumPy compatibility.
if not hasattr(np, "trapz"):
    np.trapz = np.trapezoid

import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.colors import LinearSegmentedColormap

os.environ["OMP_NUM_THREADS"] = "2"
os.environ["OPENBLAS_NUM_THREADS"] = "2"
os.environ["MKL_NUM_THREADS"] = "2"
os.environ["VECLIB_MAXIMUM_THREADS"] = "2"
os.environ["NUMEXPR_NUM_THREADS"] = "2"

from scipy.spatial.distance import pdist, squareform
from scipy.cluster.hierarchy import linkage, leaves_list

try:
    from scipy.linalg import eigh
    HAVE_SCIPY_EIGH = True
except Exception:
    HAVE_SCIPY_EIGH = False

try:
    from IPython.display import display, Image
    HAVE_IPYTHON = True
except Exception:
    HAVE_IPYTHON = False

try:
    from ripser import ripser
except Exception as exc:
    raise ImportError(
        "ripser is not installed. Run this in a separate notebook cell first:\n\n"
        "!pip install ripser persim\n\n"
        "Then rerun this Bottleneck Distance cell."
    ) from exc

try:
    from persim import bottleneck
except Exception as exc:
    raise ImportError(
        "persim is not installed. Run this in a separate notebook cell first:\n\n"
        "!pip install persim\n\n"
        "Then rerun this Bottleneck Distance cell."
    ) from exc


base_dir = "/home/a/projects/Complete-Neural-Signal-Analysis"
out_dir = os.path.join(base_dir, "topology_bottleneck_distances")
os.makedirs(out_dir, exist_ok=True)

FAST_OPTION = "ultra"  # options: "ultra", "fast", "balanced", "full"

MAX_CHANNELS_TO_PROCESS = None
RENDER_ALL_IMAGES_INLINE = True
SAVE_PLOTS_TO_DISK = True
MAKE_EXPERT_PLOTS = True
INLINE_IMAGE_WIDTH = 950
VERBOSE_PROGRESS = True

MODE_CONFIGS = {
    "ultra": {
        "max_points": 55,
        "max_dims": 3,
        "radius_quantile": 0.12,
        "cap_essential_h1": True,
        "plot_dpi": 120,
    },
    "fast": {
        "max_points": 90,
        "max_dims": 3,
        "radius_quantile": 0.15,
        "cap_essential_h1": True,
        "plot_dpi": 130,
    },
    "balanced": {
        "max_points": 130,
        "max_dims": 3,
        "radius_quantile": 0.18,
        "cap_essential_h1": True,
        "plot_dpi": 140,
    },
    "full": {
        "max_points": 180,
        "max_dims": 3,
        "radius_quantile": 0.22,
        "cap_essential_h1": True,
        "plot_dpi": 150,
    },
}

if FAST_OPTION not in MODE_CONFIGS:
    raise ValueError("FAST_OPTION must be one of: ultra, fast, balanced, full")

cfg = MODE_CONFIGS[FAST_OPTION]

MAX_POINTS = cfg["max_points"]
MAX_DIMS = cfg["max_dims"]
RADIUS_QUANTILE = cfg["radius_quantile"]
CAP_ESSENTIAL_H1 = cfg["cap_essential_h1"]
PLOT_DPI = cfg["plot_dpi"]

mode_out_dir = os.path.join(out_dir, f"h1_bottleneck_distances_{FAST_OPTION}_ripser_presampled")
os.makedirs(mode_out_dir, exist_ok=True)

eeg_channels = [
    "Fp1", "Fpz", "Fp2", "F7", "F3", "Fz", "F4", "F8", "FC5", "FC1", "FC2", "FC6",
    "M1", "T7", "C3", "Cz", "C4", "T8", "M2", "CP5", "CP1", "CP2", "CP6",
    "P7", "P3", "Pz", "P4", "P8", "POz", "O1", "Oz", "O2"
]

if MAX_CHANNELS_TO_PROCESS is not None:
    eeg_channels = eeg_channels[:MAX_CHANNELS_TO_PROCESS]

ACCENT = "#00FFFF"
ACCENT_SOFT = "#66FFFF"
ACCENT_DIM = "#008B8B"
BLACK = "#000000"

CYAN_SEQUENTIAL_CMAP = LinearSegmentedColormap.from_list(
    "cyan_sequential",
    [BLACK, ACCENT_DIM, ACCENT, ACCENT_SOFT],
    N=256,
)

plt.rcParams.update({
    "figure.facecolor": BLACK,
    "axes.facecolor": BLACK,
    "savefig.facecolor": BLACK,
    "text.color": ACCENT,
    "axes.labelcolor": ACCENT,
    "axes.edgecolor": ACCENT,
    "axes.titlecolor": ACCENT,
    "xtick.color": ACCENT,
    "ytick.color": ACCENT,
    "grid.color": ACCENT_DIM,
    "grid.alpha": 0.25,
    "legend.facecolor": BLACK,
    "legend.edgecolor": ACCENT,
    "legend.labelcolor": ACCENT,
    "font.size": 11,
})


def log(msg):
    if VERBOSE_PROGRESS:
        print(msg, flush=True)


def style_ax(ax, title=None, xlabel=None, ylabel=None, grid=True):
    ax.set_facecolor(BLACK)

    for spine in ax.spines.values():
        spine.set_color(ACCENT)
        spine.set_linewidth(1.1)

    ax.tick_params(axis="x", colors=ACCENT, labelcolor=ACCENT)
    ax.tick_params(axis="y", colors=ACCENT, labelcolor=ACCENT)

    if title is not None:
        ax.set_title(title, color=ACCENT, pad=12)

    if xlabel is not None:
        ax.set_xlabel(xlabel, color=ACCENT, labelpad=10)

    if ylabel is not None:
        ax.set_ylabel(ylabel, color=ACCENT, labelpad=10)

    if grid:
        ax.grid(True, color=ACCENT_DIM, alpha=0.25, linewidth=0.8)


def style_colorbar(cbar, label):
    cbar.set_label(label, color=ACCENT)
    cbar.ax.yaxis.set_tick_params(color=ACCENT)
    plt.setp(cbar.ax.get_yticklabels(), color=ACCENT)
    cbar.outline.set_edgecolor(ACCENT)


def save_render_close(fig, path, width=INLINE_IMAGE_WIDTH):
    if SAVE_PLOTS_TO_DISK:
        fig.savefig(path, dpi=PLOT_DPI, bbox_inches="tight", facecolor=BLACK)

    plt.close(fig)

    if RENDER_ALL_IMAGES_INLINE and HAVE_IPYTHON and os.path.exists(path):
        display(Image(filename=path, width=width))


def load_channel_embedding_sampled(ch_name, max_points=55, seed=7):
    candidate_paths = [
        os.path.join(base_dir, "embedding_data", "tsne_umap_from_3d", f"umap_embedding_{ch_name}.npy"),
        os.path.join(base_dir, "embedding_data", "tsne_umap_from_3d", f"tsne_embedding_{ch_name}.npy"),
        os.path.join(base_dir, "embedding_data", "3dembedding_data", f"3dembedded_{ch_name}.npy"),
        os.path.join(base_dir, "embedding_data", "2dembedding_data", f"2dembedded_{ch_name}.npy"),
    ]

    rng = np.random.default_rng(seed)

    for path in candidate_paths:
        if not os.path.exists(path):
            continue

        log(f"    trying: {path}")

        try:
            X = np.load(path, mmap_mode="r", allow_pickle=False)
        except Exception as exc:
            log(f"    skipped non-memmappable or nonnumeric file: {os.path.basename(path)} | {exc}")
            continue

        if X.ndim != 2:
            log(f"    skipped because ndim={X.ndim}")
            continue

        original_shape = tuple(X.shape)

        try:
            if X.shape[0] >= X.shape[1]:
                n_samples = X.shape[0]
                n_features = X.shape[1]

                if n_samples < 10 or n_features < 2:
                    log(f"    skipped bad shape: {original_shape}")
                    continue

                sample_n = min(max_points, n_samples)
                idx = np.sort(rng.choice(n_samples, size=sample_n, replace=False))
                X_small = np.asarray(X[idx, :], dtype=np.float64)

            else:
                n_samples = X.shape[1]
                n_features = X.shape[0]

                if n_samples < 10 or n_features < 2:
                    log(f"    skipped bad transposed shape: {original_shape}")
                    continue

                sample_n = min(max_points, n_samples)
                idx = np.sort(rng.choice(n_samples, size=sample_n, replace=False))
                X_small = np.asarray(X[:, idx].T, dtype=np.float64)

        except Exception as exc:
            log(f"    failed while sampling {os.path.basename(path)} | {exc}")
            continue

        X_small = X_small[np.all(np.isfinite(X_small), axis=1)]

        if X_small.shape[0] >= 10 and X_small.shape[1] >= 2:
            return X_small, path, original_shape

        log(f"    sampled data unusable: sampled_shape={X_small.shape}")

    return None, None, None


def standardize_point_cloud_small(X, max_dims=3):
    X = np.asarray(X, dtype=np.float64)
    X = X[np.all(np.isfinite(X), axis=1)]

    if X.shape[0] < 10:
        return None

    X = X - np.mean(X, axis=0, keepdims=True)

    if X.shape[1] > max_dims:
        _U, _S, Vt = np.linalg.svd(X, full_matrices=False)
        X = X @ Vt[:max_dims].T

    X = X / (np.std(X, axis=0, keepdims=True) + 1e-12)

    D = pdist(X)
    positive = D[D > 0]

    if len(positive) > 0:
        X = X / (np.nanmedian(positive) + 1e-12)

    return X


def estimate_complex_counts(X, radius):
    D = squareform(pdist(X))
    A = (D <= radius) & (D > 0)

    n_edges = int(np.sum(A) // 2)

    if A.shape[0] <= 300:
        A_int = A.astype(np.int64)
        n_triangles = int(np.trace(A_int @ A_int @ A_int) // 6)
    else:
        n_triangles = np.nan

    return n_edges, n_triangles


def run_h1_ripser(X, radius):
    result = ripser(
        X,
        maxdim=1,
        thresh=float(radius),
        coeff=2,
        do_cocycles=False,
    )

    diagrams = result["dgms"]

    h0 = diagrams[0] if len(diagrams) > 0 else np.empty((0, 2), dtype=float)
    h1 = diagrams[1] if len(diagrams) > 1 else np.empty((0, 2), dtype=float)

    return h0, h1


def split_finite_essential(diagram):
    diagram = np.asarray(diagram, dtype=float)

    if diagram.size == 0:
        return np.empty((0, 2), dtype=float), np.empty((0, 2), dtype=float)

    finite = diagram[
        np.isfinite(diagram[:, 0])
        & np.isfinite(diagram[:, 1])
        & (diagram[:, 1] > diagram[:, 0])
    ]

    essential = diagram[
        np.isfinite(diagram[:, 0])
        & np.isinf(diagram[:, 1])
    ]

    return finite, essential


def clean_diagram_for_bottleneck(
    finite_intervals,
    essential_intervals=None,
    cap_essential=True,
    cap_value=None,
):
    cleaned = []
    capped_count = 0

    if finite_intervals is not None:
        for b, d in finite_intervals:
            if np.isfinite(b) and np.isfinite(d) and d > b:
                cleaned.append((float(b), float(d)))

    if cap_essential and essential_intervals is not None and cap_value is not None:
        for b, _d in essential_intervals:
            if np.isfinite(b) and np.isfinite(cap_value) and cap_value > b:
                cleaned.append((float(b), float(cap_value)))
                capped_count += 1

    if len(cleaned) == 0:
        return np.empty((0, 2), dtype=float), capped_count

    return np.asarray(cleaned, dtype=float), capped_count


def diagram_features(diagram):
    diagram = np.asarray(diagram, dtype=float)

    if diagram.size == 0 or len(diagram) == 0:
        return {
            "n_intervals_diagram": 0,
            "mean_lifetime": np.nan,
            "max_lifetime": np.nan,
            "total_lifetime": 0.0,
            "mean_birth": np.nan,
            "mean_death": np.nan,
            "max_death": np.nan,
        }

    lifetimes = diagram[:, 1] - diagram[:, 0]

    return {
        "n_intervals_diagram": int(len(diagram)),
        "mean_lifetime": float(np.nanmean(lifetimes)),
        "max_lifetime": float(np.nanmax(lifetimes)),
        "total_lifetime": float(np.nansum(lifetimes)),
        "mean_birth": float(np.nanmean(diagram[:, 0])),
        "mean_death": float(np.nanmean(diagram[:, 1])),
        "max_death": float(np.nanmax(diagram[:, 1])),
    }


def empty_aware_bottleneck(D1, D2):
    D1 = np.asarray(D1, dtype=float)
    D2 = np.asarray(D2, dtype=float)

    if D1.size == 0:
        D1 = np.empty((0, 2), dtype=float)

    if D2.size == 0:
        D2 = np.empty((0, 2), dtype=float)

    n1 = len(D1)
    n2 = len(D2)

    if n1 == 0 and n2 == 0:
        return 0.0

    if n1 == 0:
        return float(np.nanmax((D2[:, 1] - D2[:, 0]) / 2.0))

    if n2 == 0:
        return float(np.nanmax((D1[:, 1] - D1[:, 0]) / 2.0))

    return float(bottleneck(D1, D2, matching=False))


def classical_mds_from_distances(D, n_components=2):
    D = np.asarray(D, dtype=float)
    n = D.shape[0]

    if n < 2:
        return np.zeros((n, n_components), dtype=float)

    D2 = D ** 2
    J = np.eye(n) - np.ones((n, n)) / n
    B = -0.5 * J @ D2 @ J

    if HAVE_SCIPY_EIGH:
        vals, vecs = eigh(B)
    else:
        vals, vecs = np.linalg.eigh(B)

    idx = np.argsort(vals)[::-1]
    vals = vals[idx]
    vecs = vecs[:, idx]

    coords = np.zeros((n, n_components), dtype=float)

    for k in range(min(n_components, n)):
        if vals[k] > 0:
            coords[:, k] = vecs[:, k] * np.sqrt(vals[k])

    return coords


log("\n" + "=" * 72)
log(f"Running H1 Bottleneck Distances with ripser backend | {FAST_OPTION.upper()}")
log(f"Channels: {len(eeg_channels)}")
log(f"max_points={MAX_POINTS}, radius_quantile={RADIUS_QUANTILE}")
log(f"cap_essential_h1={CAP_ESSENTIAL_H1}")
log(f"expert_plots={MAKE_EXPERT_PLOTS}, render_all_images={RENDER_ALL_IMAGES_INLINE}")
log(f"Output: {mode_out_dir}")
log("=" * 72)

diagnostics = {}
diagram_store = {}
raw_interval_store = {}
loaded_channels = []

overall_t0 = time.time()

for ch_idx, ch_name in enumerate(eeg_channels, start=1):
    t0 = time.time()
    prefix = f"[{ch_idx:02d}/{len(eeg_channels)} {ch_name}]"

    log(f"\n{prefix} loading + pre-subsampling to {MAX_POINTS} points")

    X_small, source_path, original_shape = load_channel_embedding_sampled(
        ch_name,
        max_points=MAX_POINTS,
        seed=400 + ch_idx,
    )

    if X_small is None:
        log(f"{prefix} WARNING: no usable embedding found; skipping")
        continue

    log(f"{prefix} source shape={original_shape}, sampled shape={X_small.shape}")

    X = standardize_point_cloud_small(X_small, max_dims=MAX_DIMS)

    if X is None:
        log(f"{prefix} WARNING: invalid sampled embedding; skipping")
        continue

    log(f"{prefix} standardized shape={X.shape}; computing radius")

    D_flat = pdist(X)
    positive = D_flat[D_flat > 0]

    if len(positive) == 0:
        log(f"{prefix} WARNING: zero-distance cloud; skipping")
        continue

    radius = float(np.quantile(positive, RADIUS_QUANTILE))

    log(f"{prefix} estimating threshold complex size")
    n_edges, n_triangles = estimate_complex_counts(X, radius)

    log(f"{prefix} ripser H0/H1")
    h0_diag, h1_diag = run_h1_ripser(X, radius)

    h0_finite, h0_essential = split_finite_essential(h0_diag)
    h1_finite, h1_essential = split_finite_essential(h1_diag)

    h1_diagram, n_capped_h1 = clean_diagram_for_bottleneck(
        h1_finite,
        essential_intervals=h1_essential,
        cap_essential=CAP_ESSENTIAL_H1,
        cap_value=radius,
    )

    elapsed = time.time() - t0

    diagram_store[ch_name] = h1_diagram

    raw_interval_store[ch_name] = {
        "H0_finite": h0_finite,
        "H0_essential": h0_essential,
        "H1_finite": h1_finite,
        "H1_essential": h1_essential,
        "H1_bottleneck_diagram": h1_diagram,
    }

    diagnostics[ch_name] = {
        "channel": ch_name,
        "source_path": source_path,
        "source_shape": str(original_shape),
        "backend": "ripser_presampled",
        "n_points": int(X.shape[0]),
        "n_edges_at_radius": int(n_edges),
        "n_triangles_at_radius": int(n_triangles) if np.isfinite(n_triangles) else np.nan,
        "radius": radius,
        "n_h0_finite": int(len(h0_finite)),
        "n_h1_finite": int(len(h1_finite)),
        "n_h1_essential": int(len(h1_essential)),
        "n_h1_capped_for_bottleneck": int(n_capped_h1),
        "elapsed_sec": elapsed,
        **diagram_features(h1_diagram),
    }

    loaded_channels.append(ch_name)

    log(
        f"{prefix} done {elapsed:.2f}s | "
        f"edges={n_edges}, triangles={n_triangles}, "
        f"H1 finite={len(h1_finite)}, H1 bottleneck diagram={len(h1_diagram)}"
    )

if len(loaded_channels) == 0:
    raise RuntimeError("No usable embeddings found. Check embedding_data paths.")

log("\nComputing pairwise bottleneck distances")

n_ch = len(loaded_channels)
bottleneck_mat = np.zeros((n_ch, n_ch), dtype=float)

pair_rows = []
pair_t0 = time.time()

for i in range(n_ch):
    ch_i = loaded_channels[i]

    for j in range(i + 1, n_ch):
        ch_j = loaded_channels[j]

        dist = empty_aware_bottleneck(
            diagram_store[ch_i],
            diagram_store[ch_j],
        )

        bottleneck_mat[i, j] = dist
        bottleneck_mat[j, i] = dist

        pair_rows.append({
            "channel_a": ch_i,
            "channel_b": ch_j,
            "bottleneck_distance": dist,
        })

    log(f"  completed bottleneck row {i + 1}/{n_ch}: {ch_i}")

pair_elapsed = time.time() - pair_t0

distance_df = pd.DataFrame(
    bottleneck_mat,
    index=loaded_channels,
    columns=loaded_channels,
)

pairwise_df = pd.DataFrame(pair_rows).sort_values("bottleneck_distance", ascending=False).reset_index(drop=True)

summary_rows = []

for idx, ch_name in enumerate(loaded_channels):
    d = bottleneck_mat[idx].copy()
    nonself = d[np.arange(len(d)) != idx]

    summary_rows.append({
        **diagnostics[ch_name],
        "mean_bottleneck_to_others": float(np.nanmean(nonself)) if len(nonself) else 0.0,
        "median_bottleneck_to_others": float(np.nanmedian(nonself)) if len(nonself) else 0.0,
        "max_bottleneck_to_others": float(np.nanmax(nonself)) if len(nonself) else 0.0,
        "min_bottleneck_to_others": float(np.nanmin(nonself)) if len(nonself) else 0.0,
    })

summary_df = (
    pd.DataFrame(summary_rows)
    .sort_values("mean_bottleneck_to_others", ascending=False)
    .reset_index(drop=True)
)

summary_csv_path = os.path.join(mode_out_dir, "h1_bottleneck_channel_summary.csv")
distance_csv_path = os.path.join(mode_out_dir, "h1_bottleneck_distance_matrix.csv")
pairwise_csv_path = os.path.join(mode_out_dir, "h1_bottleneck_pairwise_distances.csv")

summary_df.to_csv(summary_csv_path, index=False)
distance_df.to_csv(distance_csv_path)
pairwise_df.to_csv(pairwise_csv_path, index=False)

npz_path = os.path.join(mode_out_dir, "h1_bottleneck_distances_and_diagrams.npz")

save_dict = {
    "channels": np.asarray(loaded_channels),
    "bottleneck_distance_matrix": bottleneck_mat,
    "fast_option": np.asarray([FAST_OPTION]),
    "backend": np.asarray(["ripser_presampled"]),
}

for ch_name in loaded_channels:
    save_dict[f"{ch_name}_H1_bottleneck_diagram"] = diagram_store[ch_name]

    for key, val in raw_interval_store[ch_name].items():
        save_dict[f"{ch_name}_{key}"] = val

np.savez(npz_path, **save_dict)

print("\nH1 Bottleneck Channel Summary:")
if HAVE_IPYTHON:
    display(summary_df)
else:
    print(summary_df)

print("\nLargest H1 bottleneck distances:")
if HAVE_IPYTHON:
    display(pairwise_df.head(12))
else:
    print(pairwise_df.head(12))

log(f"\nSaved summary CSV: {summary_csv_path}")
log(f"Saved distance matrix CSV: {distance_csv_path}")
log(f"Saved pairwise CSV: {pairwise_csv_path}")
log(f"Saved NPZ: {npz_path}")
log(f"Pairwise bottleneck elapsed: {pair_elapsed:.2f}s")

# ============================================================
# PLOTS
# ============================================================

plot_df = summary_df.copy().sort_values("mean_bottleneck_to_others", ascending=False).reset_index(drop=True)

fig, ax = plt.subplots(figsize=(13, 5.8), facecolor=BLACK)
style_ax(
    ax,
    title=f"Mean H1 Bottleneck Distance to Other Channels | {FAST_OPTION.upper()}",
    xlabel="Channel",
    ylabel="Mean bottleneck distance",
)

x_pos = np.arange(len(plot_df))
y = plot_df["mean_bottleneck_to_others"].to_numpy(dtype=float)

ax.bar(x_pos, y, color=ACCENT, edgecolor=ACCENT, alpha=0.78)

if np.any(np.isfinite(y)):
    ax.axhline(
        np.nanmean(y),
        color=ACCENT_SOFT,
        linestyle="--",
        linewidth=1.4,
        alpha=0.8,
        label=f"Mean={np.nanmean(y):.5f}",
    )

ax.set_xticks(x_pos)
ax.set_xticklabels(plot_df["channel"], rotation=90)
ax.legend(loc="best", framealpha=1.0)

mean_dist_path = os.path.join(mode_out_dir, "01_mean_h1_bottleneck_distance_by_channel.png")
plt.tight_layout()
save_render_close(fig, mean_dist_path)

if n_ch >= 3 and np.any(bottleneck_mat > 0):
    condensed = bottleneck_mat[np.triu_indices(n_ch, k=1)]
    Z = linkage(condensed, method="average")
    order = leaves_list(Z)
else:
    order = np.arange(n_ch)

ordered_dist = bottleneck_mat[np.ix_(order, order)]
ordered_channels = [loaded_channels[i] for i in order]

fig, ax = plt.subplots(figsize=(9, 8), facecolor=BLACK)
style_ax(
    ax,
    title="Pairwise H1 Bottleneck Distance Matrix",
    xlabel="Channel",
    ylabel="Channel",
)

im = ax.imshow(
    ordered_dist,
    aspect="equal",
    interpolation="nearest",
    cmap=CYAN_SEQUENTIAL_CMAP,
)

ax.set_xticks(np.arange(n_ch))
ax.set_xticklabels(ordered_channels, rotation=90)
ax.set_yticks(np.arange(n_ch))
ax.set_yticklabels(ordered_channels)

cbar = plt.colorbar(im, ax=ax, shrink=0.85)
style_colorbar(cbar, "Bottleneck distance")

dist_heatmap_path = os.path.join(mode_out_dir, "02_h1_bottleneck_distance_heatmap.png")
plt.tight_layout()
save_render_close(fig, dist_heatmap_path)

mds_coords = classical_mds_from_distances(bottleneck_mat, n_components=2)

mds_df = pd.DataFrame({
    "channel": loaded_channels,
    "mds_1": mds_coords[:, 0],
    "mds_2": mds_coords[:, 1],
    "mean_bottleneck_to_others": [
        summary_df.loc[summary_df["channel"] == ch, "mean_bottleneck_to_others"].values[0]
        for ch in loaded_channels
    ],
    "n_intervals_diagram": [
        summary_df.loc[summary_df["channel"] == ch, "n_intervals_diagram"].values[0]
        for ch in loaded_channels
    ],
})

mds_csv_path = os.path.join(mode_out_dir, "h1_bottleneck_mds_coordinates.csv")
mds_df.to_csv(mds_csv_path, index=False)

fig, ax = plt.subplots(figsize=(8, 7), facecolor=BLACK)
style_ax(
    ax,
    title="Classical MDS of H1 Bottleneck Distances",
    xlabel="MDS 1",
    ylabel="MDS 2",
)

sc = ax.scatter(
    mds_df["mds_1"],
    mds_df["mds_2"],
    c=mds_df["mean_bottleneck_to_others"],
    s=150,
    cmap=CYAN_SEQUENTIAL_CMAP,
    edgecolor=ACCENT,
    linewidth=1.1,
)

for _, row in mds_df.iterrows():
    ax.text(
        row["mds_1"],
        row["mds_2"],
        row["channel"],
        fontsize=8,
        color=ACCENT,
        alpha=0.9,
    )

cbar = plt.colorbar(sc, ax=ax, shrink=0.85)
style_colorbar(cbar, "Mean bottleneck distance")

mds_path = os.path.join(mode_out_dir, "03_h1_bottleneck_mds_embedding.png")
plt.tight_layout()
save_render_close(fig, mds_path)

diagram_channels = summary_df.head(min(12, len(summary_df)))["channel"].tolist()
n_cols = 4
n_rows = int(np.ceil(len(diagram_channels) / n_cols))

fig, axes = plt.subplots(
    n_rows,
    n_cols,
    figsize=(4 * n_cols, 3.2 * n_rows),
    facecolor=BLACK,
)

axes = np.asarray(axes).reshape(-1)

all_diagram_points = np.vstack([
    diagram_store[ch] for ch in loaded_channels if len(diagram_store[ch]) > 0
]) if any(len(diagram_store[ch]) > 0 for ch in loaded_channels) else np.empty((0, 2))

if len(all_diagram_points) > 0:
    diag_max = float(np.nanmax(all_diagram_points[:, 1]))
else:
    diag_max = 1.0

diag_max = max(diag_max, 1e-6)

for ax_idx, ch_name in enumerate(diagram_channels):
    ax = axes[ax_idx]
    D = diagram_store[ch_name]

    mean_dist = summary_df.loc[
        summary_df["channel"] == ch_name,
        "mean_bottleneck_to_others",
    ].values[0]

    style_ax(
        ax,
        title=f"{ch_name}: mean dB={mean_dist:.3f}, n={len(D)}",
        xlabel="Birth",
        ylabel="Death",
    )

    ax.plot([0, diag_max], [0, diag_max], color=ACCENT_DIM, linestyle="--", linewidth=1.0)

    if len(D) > 0:
        ax.scatter(
            D[:, 0],
            D[:, 1],
            s=50,
            facecolors=BLACK,
            edgecolors=ACCENT,
            linewidths=1.2,
        )

    ax.set_xlim(0, diag_max * 1.05)
    ax.set_ylim(0, diag_max * 1.05)

for j in range(len(diagram_channels), len(axes)):
    axes[j].axis("off")

diagrams_path = os.path.join(mode_out_dir, "04_top_channels_h1_persistence_diagrams.png")
plt.tight_layout()
save_render_close(fig, diagrams_path, width=1100)

fig, ax = plt.subplots(figsize=(12, 6), facecolor=BLACK)
style_ax(
    ax,
    title="Largest Pairwise H1 Bottleneck Distances",
    xlabel="Channel pair",
    ylabel="Bottleneck distance",
)

top_pairs = pairwise_df.head(min(15, len(pairwise_df))).copy()
pair_labels = top_pairs["channel_a"] + "-" + top_pairs["channel_b"]

x_pair = np.arange(len(top_pairs))

ax.bar(
    x_pair,
    top_pairs["bottleneck_distance"],
    color=ACCENT,
    edgecolor=ACCENT,
    alpha=0.78,
)

ax.set_xticks(x_pair)
ax.set_xticklabels(pair_labels, rotation=60, ha="right")

top_pairs_path = os.path.join(mode_out_dir, "05_largest_pairwise_h1_bottleneck_distances.png")
plt.tight_layout()
save_render_close(fig, top_pairs_path)

if MAKE_EXPERT_PLOTS:
    expert_dir = os.path.join(mode_out_dir, "expert_plots")
    os.makedirs(expert_dir, exist_ok=True)

    mean_bottleneck_map = dict(zip(summary_df["channel"], summary_df["mean_bottleneck_to_others"]))

    scalp_xy = {
        "Fp1": (-0.45, 1.00), "Fpz": (0.00, 1.08), "Fp2": (0.45, 1.00),
        "F7": (-0.95, 0.60), "F3": (-0.45, 0.55), "Fz": (0.00, 0.60),
        "F4": (0.45, 0.55), "F8": (0.95, 0.60),
        "FC5": (-0.75, 0.25), "FC1": (-0.25, 0.25),
        "FC2": (0.25, 0.25), "FC6": (0.75, 0.25),
        "M1": (-1.15, 0.05), "T7": (-1.00, 0.00), "C3": (-0.50, 0.00),
        "Cz": (0.00, 0.00), "C4": (0.50, 0.00), "T8": (1.00, 0.00),
        "M2": (1.15, 0.05),
        "CP5": (-0.75, -0.30), "CP1": (-0.25, -0.30),
        "CP2": (0.25, -0.30), "CP6": (0.75, -0.30),
        "P7": (-0.90, -0.60), "P3": (-0.45, -0.60), "Pz": (0.00, -0.65),
        "P4": (0.45, -0.60), "P8": (0.90, -0.60),
        "POz": (0.00, -0.88),
        "O1": (-0.35, -1.00), "Oz": (0.00, -1.08), "O2": (0.35, -1.00),
    }

    xs, ys, vals, labels = [], [], [], []

    for ch in eeg_channels:
        if ch in scalp_xy and ch in mean_bottleneck_map and np.isfinite(mean_bottleneck_map[ch]):
            xs.append(scalp_xy[ch][0])
            ys.append(scalp_xy[ch][1])
            vals.append(mean_bottleneck_map[ch])
            labels.append(ch)

    fig, ax = plt.subplots(figsize=(7, 7), facecolor=BLACK)
    style_ax(
        ax,
        title=f"Scalp Map of Mean H1 Bottleneck Distance | {FAST_OPTION.upper()}",
        grid=False,
    )

    head = plt.Circle((0, 0), 1.12, fill=False, color=ACCENT, linewidth=2)
    ax.add_patch(head)
    ax.plot([0.00, -0.10, 0.10, 0.00], [1.12, 1.28, 1.28, 1.12], color=ACCENT, linewidth=2)

    sc = ax.scatter(
        xs,
        ys,
        c=vals,
        s=260,
        cmap=CYAN_SEQUENTIAL_CMAP,
        edgecolor=ACCENT,
        linewidth=1.2,
    )

    for x0, y0, label in zip(xs, ys, labels):
        ax.text(
            x0,
            y0,
            label,
            ha="center",
            va="center",
            fontsize=8,
            color=BLACK,
            fontweight="bold",
        )

    cbar = plt.colorbar(sc, ax=ax, shrink=0.75)
    style_colorbar(cbar, "Mean bottleneck distance")

    ax.set_xlim(-1.35, 1.35)
    ax.set_ylim(-1.35, 1.35)
    ax.set_aspect("equal")
    ax.set_xticks([])
    ax.set_yticks([])

    scalp_path = os.path.join(expert_dir, "06_mean_h1_bottleneck_distance_scalp_map.png")
    plt.tight_layout()
    save_render_close(fig, scalp_path, width=760)

    regions = {
        "Frontal": ["Fp1", "Fpz", "Fp2", "F7", "F3", "Fz", "F4", "F8"],
        "Frontocentral": ["FC5", "FC1", "FC2", "FC6"],
        "Central": ["T7", "C3", "Cz", "C4", "T8"],
        "Centroparietal": ["CP5", "CP1", "CP2", "CP6"],
        "Parietal": ["P7", "P3", "Pz", "P4", "P8"],
        "Occipital": ["POz", "O1", "Oz", "O2"],
        "Mastoid": ["M1", "M2"],
    }

    region_rows = []
    region_points = []

    for region_name, chans in regions.items():
        vals_region = []

        for ch in chans:
            if ch in mean_bottleneck_map and np.isfinite(mean_bottleneck_map[ch]):
                vals_region.append(mean_bottleneck_map[ch])
                region_points.append({
                    "region": region_name,
                    "channel": ch,
                    "mean_bottleneck_to_others": mean_bottleneck_map[ch],
                })

        if len(vals_region) == 0:
            continue

        vals_region = np.asarray(vals_region, dtype=float)

        region_rows.append({
            "region": region_name,
            "mean_region_bottleneck": float(np.mean(vals_region)),
            "std_region_bottleneck": float(np.std(vals_region, ddof=1)) if len(vals_region) > 1 else 0.0,
            "sem_region_bottleneck": (
                float(np.std(vals_region, ddof=1) / np.sqrt(len(vals_region)))
                if len(vals_region) > 1 else 0.0
            ),
            "n_channels": int(len(vals_region)),
        })

    region_df = pd.DataFrame(region_rows)
    region_points_df = pd.DataFrame(region_points)

    region_csv_path = os.path.join(expert_dir, "regional_h1_bottleneck_summary.csv")
    region_df.to_csv(region_csv_path, index=False)

    if len(region_df) > 0:
        fig, ax = plt.subplots(figsize=(12, 6.5), facecolor=BLACK)
        style_ax(
            ax,
            title=f"Regional Mean H1 Bottleneck Distance +/- SEM | {FAST_OPTION.upper()}",
            xlabel="Region",
            ylabel="Mean bottleneck distance",
        )

        x_region = np.arange(len(region_df))

        ax.bar(
            x_region,
            region_df["mean_region_bottleneck"],
            yerr=region_df["sem_region_bottleneck"],
            color=ACCENT,
            edgecolor=ACCENT,
            alpha=0.82,
            width=0.68,
            capsize=8,
            error_kw={
                "elinewidth": 2.2,
                "capthick": 2.2,
                "ecolor": ACCENT,
                "alpha": 1.0,
            },
            label="Mean +/- SEM",
        )

        rng = np.random.default_rng(7)

        for i, region_name in enumerate(region_df["region"]):
            vals = region_points_df.loc[
                region_points_df["region"] == region_name,
                "mean_bottleneck_to_others",
            ].to_numpy(dtype=float)

            if len(vals) == 0:
                continue

            jitter = rng.uniform(-0.13, 0.13, size=len(vals))

            ax.scatter(
                np.full(len(vals), i) + jitter,
                vals,
                s=42,
                facecolors=BLACK,
                edgecolors=ACCENT,
                linewidths=1.2,
                alpha=0.95,
                zorder=4,
            )

        for i, row in region_df.iterrows():
            y_val = row["mean_region_bottleneck"]
            sem = row["sem_region_bottleneck"]

            ax.text(
                i,
                y_val + sem,
                f"{y_val:.3f}\n+/-{sem:.3f}\nn={int(row['n_channels'])}",
                ha="center",
                va="bottom",
                fontsize=8,
                color=ACCENT,
            )

        ax.set_xticks(x_region)
        ax.set_xticklabels(
            region_df["region"],
            rotation=35,
            ha="right",
            rotation_mode="anchor",
        )
        ax.margins(x=0.04)
        ax.legend(loc="best", framealpha=1.0)

        region_path = os.path.join(expert_dir, "07_regional_h1_bottleneck_summary.png")
        plt.tight_layout(pad=2.0)
        save_render_close(fig, region_path)

    fig, ax = plt.subplots(figsize=(8, 6), facecolor=BLACK)
    style_ax(
        ax,
        title="Diagram Size vs Mean H1 Bottleneck Distance",
        xlabel="H1 diagram intervals",
        ylabel="Mean bottleneck distance to others",
    )

    sc = ax.scatter(
        summary_df["n_intervals_diagram"],
        summary_df["mean_bottleneck_to_others"],
        c=summary_df["max_lifetime"],
        s=140,
        cmap=CYAN_SEQUENTIAL_CMAP,
        edgecolor=ACCENT,
        linewidth=1.0,
    )

    for _, row in summary_df.iterrows():
        ax.text(
            row["n_intervals_diagram"],
            row["mean_bottleneck_to_others"],
            row["channel"],
            fontsize=8,
            color=ACCENT,
            alpha=0.85,
        )

    cbar = plt.colorbar(sc, ax=ax, shrink=0.85)
    style_colorbar(cbar, "Max H1 lifetime")

    diagram_size_path = os.path.join(expert_dir, "08_diagram_size_vs_mean_bottleneck.png")
    plt.tight_layout()
    save_render_close(fig, diagram_size_path)

    fig, ax = plt.subplots(figsize=(8, 6), facecolor=BLACK)
    style_ax(
        ax,
        title="Threshold Triangles vs Mean H1 Bottleneck Distance",
        xlabel="Triangles at radius",
        ylabel="Mean bottleneck distance to others",
    )

    sc = ax.scatter(
        summary_df["n_triangles_at_radius"],
        summary_df["mean_bottleneck_to_others"],
        c=summary_df["n_intervals_diagram"],
        s=140,
        cmap=CYAN_SEQUENTIAL_CMAP,
        edgecolor=ACCENT,
        linewidth=1.0,
    )

    for _, row in summary_df.iterrows():
        ax.text(
            row["n_triangles_at_radius"],
            row["mean_bottleneck_to_others"],
            row["channel"],
            fontsize=8,
            color=ACCENT,
            alpha=0.85,
        )

    cbar = plt.colorbar(sc, ax=ax, shrink=0.85)
    style_colorbar(cbar, "H1 diagram intervals")

    triangle_path = os.path.join(expert_dir, "09_threshold_triangles_vs_mean_bottleneck.png")
    plt.tight_layout()
    save_render_close(fig, triangle_path)

    qc_df = summary_df.copy().sort_values("elapsed_sec", ascending=False).reset_index(drop=True)

    fig, ax = plt.subplots(figsize=(12, 5.8), facecolor=BLACK)
    style_ax(
        ax,
        title="Bottleneck Diagram Runtime QC",
        xlabel="Channel",
        ylabel="Elapsed seconds",
    )

    x_qc = np.arange(len(qc_df))

    ax.bar(
        x_qc,
        qc_df["elapsed_sec"],
        color=ACCENT,
        edgecolor=ACCENT,
        alpha=0.82,
    )

    ax.set_xticks(x_qc)
    ax.set_xticklabels(qc_df["channel"], rotation=90)

    qc_path = os.path.join(expert_dir, "10_runtime_qc.png")
    plt.tight_layout()
    save_render_close(fig, qc_path)

    log(f"\nExpert plots saved and rendered from: {expert_dir}")
    log(f"Regional summary CSV: {region_csv_path}")

elapsed_total = time.time() - overall_t0

log("\n" + "=" * 72)
log(f"Finished H1 Bottleneck Distances in {elapsed_total:.2f}s")
log(f"Mode: {FAST_OPTION}")
log(f"Backend: ripser_presampled + persim.bottleneck")
log(f"Channels completed: {len(loaded_channels)}")
log(f"Summary CSV: {summary_csv_path}")
log(f"Distance matrix CSV: {distance_csv_path}")
log(f"Pairwise CSV: {pairwise_csv_path}")
log(f"MDS CSV: {mds_csv_path}")
log(f"NPZ: {npz_path}")
log(f"Output directory: {mode_out_dir}")
log("=" * 72)
```

```python
import os
import sys
import time
import numpy as np

if not hasattr(np, "trapz"):
    np.trapz = np.trapezoid

import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.colors import LinearSegmentedColormap

os.environ["OMP_NUM_THREADS"] = "2"
os.environ["OPENBLAS_NUM_THREADS"] = "2"
os.environ["MKL_NUM_THREADS"] = "2"
os.environ["VECLIB_MAXIMUM_THREADS"] = "2"
os.environ["NUMEXPR_NUM_THREADS"] = "2"

from scipy.spatial.distance import pdist, squareform
from scipy.cluster.hierarchy import linkage, leaves_list

try:
    from scipy.linalg import eigh
    HAVE_SCIPY_EIGH = True
except Exception:
    HAVE_SCIPY_EIGH = False

try:
    from IPython.display import display, Image
    HAVE_IPYTHON = True
except Exception:
    HAVE_IPYTHON = False

try:
    from ripser import ripser
except Exception as exc:
    raise ImportError(
        "ripser is not installed. Run this in a separate notebook cell first:\n\n"
        "!pip install ripser persim\n\n"
        "Then rerun this Wasserstein Distance cell."
    ) from exc

try:
    from persim import wasserstein
except Exception as exc:
    raise ImportError(
        "persim is not installed. Run this in a separate notebook cell first:\n\n"
        "!pip install persim\n\n"
        "Then rerun this Wasserstein Distance cell."
    ) from exc


base_dir = "/home/a/projects/Complete-Neural-Signal-Analysis"
out_dir = os.path.join(base_dir, "topology_wasserstein_distances")
os.makedirs(out_dir, exist_ok=True)

FAST_OPTION = "ultra"  # options: "ultra", "fast", "balanced", "full"

MAX_CHANNELS_TO_PROCESS = None
RENDER_ALL_IMAGES_INLINE = True
SAVE_PLOTS_TO_DISK = True
MAKE_EXPERT_PLOTS = True
INLINE_IMAGE_WIDTH = 950
VERBOSE_PROGRESS = True

MODE_CONFIGS = {
    "ultra": {
        "max_points": 55,
        "max_dims": 3,
        "radius_quantile": 0.12,
        "cap_essential_h1": True,
        "plot_dpi": 120,
    },
    "fast": {
        "max_points": 90,
        "max_dims": 3,
        "radius_quantile": 0.15,
        "cap_essential_h1": True,
        "plot_dpi": 130,
    },
    "balanced": {
        "max_points": 130,
        "max_dims": 3,
        "radius_quantile": 0.18,
        "cap_essential_h1": True,
        "plot_dpi": 140,
    },
    "full": {
        "max_points": 180,
        "max_dims": 3,
        "radius_quantile": 0.22,
        "cap_essential_h1": True,
        "plot_dpi": 150,
    },
}

if FAST_OPTION not in MODE_CONFIGS:
    raise ValueError("FAST_OPTION must be one of: ultra, fast, balanced, full")

cfg = MODE_CONFIGS[FAST_OPTION]

MAX_POINTS = cfg["max_points"]
MAX_DIMS = cfg["max_dims"]
RADIUS_QUANTILE = cfg["radius_quantile"]
CAP_ESSENTIAL_H1 = cfg["cap_essential_h1"]
PLOT_DPI = cfg["plot_dpi"]

mode_out_dir = os.path.join(out_dir, f"h1_wasserstein_distances_{FAST_OPTION}_ripser_presampled")
os.makedirs(mode_out_dir, exist_ok=True)

eeg_channels = [
    "Fp1", "Fpz", "Fp2", "F7", "F3", "Fz", "F4", "F8", "FC5", "FC1", "FC2", "FC6",
    "M1", "T7", "C3", "Cz", "C4", "T8", "M2", "CP5", "CP1", "CP2", "CP6",
    "P7", "P3", "Pz", "P4", "P8", "POz", "O1", "Oz", "O2"
]

if MAX_CHANNELS_TO_PROCESS is not None:
    eeg_channels = eeg_channels[:MAX_CHANNELS_TO_PROCESS]

ACCENT = "#00FFFF"
ACCENT_SOFT = "#66FFFF"
ACCENT_DIM = "#008B8B"
BLACK = "#000000"

CYAN_SEQUENTIAL_CMAP = LinearSegmentedColormap.from_list(
    "cyan_sequential",
    [BLACK, ACCENT_DIM, ACCENT, ACCENT_SOFT],
    N=256,
)

plt.rcParams.update({
    "figure.facecolor": BLACK,
    "axes.facecolor": BLACK,
    "savefig.facecolor": BLACK,
    "text.color": ACCENT,
    "axes.labelcolor": ACCENT,
    "axes.edgecolor": ACCENT,
    "axes.titlecolor": ACCENT,
    "xtick.color": ACCENT,
    "ytick.color": ACCENT,
    "grid.color": ACCENT_DIM,
    "grid.alpha": 0.25,
    "legend.facecolor": BLACK,
    "legend.edgecolor": ACCENT,
    "legend.labelcolor": ACCENT,
    "font.size": 11,
})


def log(msg):
    if VERBOSE_PROGRESS:
        print(msg, flush=True)


def style_ax(ax, title=None, xlabel=None, ylabel=None, grid=True):
    ax.set_facecolor(BLACK)

    for spine in ax.spines.values():
        spine.set_color(ACCENT)
        spine.set_linewidth(1.1)

    ax.tick_params(axis="x", colors=ACCENT, labelcolor=ACCENT)
    ax.tick_params(axis="y", colors=ACCENT, labelcolor=ACCENT)

    if title is not None:
        ax.set_title(title, color=ACCENT, pad=12)

    if xlabel is not None:
        ax.set_xlabel(xlabel, color=ACCENT, labelpad=10)

    if ylabel is not None:
        ax.set_ylabel(ylabel, color=ACCENT, labelpad=10)

    if grid:
        ax.grid(True, color=ACCENT_DIM, alpha=0.25, linewidth=0.8)


def style_colorbar(cbar, label):
    cbar.set_label(label, color=ACCENT)
    cbar.ax.yaxis.set_tick_params(color=ACCENT)
    plt.setp(cbar.ax.get_yticklabels(), color=ACCENT)
    cbar.outline.set_edgecolor(ACCENT)


def save_render_close(fig, path, width=INLINE_IMAGE_WIDTH):
    if SAVE_PLOTS_TO_DISK:
        fig.savefig(path, dpi=PLOT_DPI, bbox_inches="tight", facecolor=BLACK)

    plt.close(fig)

    if RENDER_ALL_IMAGES_INLINE and HAVE_IPYTHON and os.path.exists(path):
        display(Image(filename=path, width=width))


def load_channel_embedding_sampled(ch_name, max_points=55, seed=7):
    candidate_paths = [
        os.path.join(base_dir, "embedding_data", "tsne_umap_from_3d", f"umap_embedding_{ch_name}.npy"),
        os.path.join(base_dir, "embedding_data", "tsne_umap_from_3d", f"tsne_embedding_{ch_name}.npy"),
        os.path.join(base_dir, "embedding_data", "3dembedding_data", f"3dembedded_{ch_name}.npy"),
        os.path.join(base_dir, "embedding_data", "2dembedding_data", f"2dembedded_{ch_name}.npy"),
    ]

    rng = np.random.default_rng(seed)

    for path in candidate_paths:
        if not os.path.exists(path):
            continue

        log(f"    trying: {path}")

        try:
            X = np.load(path, mmap_mode="r", allow_pickle=False)
        except Exception as exc:
            log(f"    skipped non-memmappable or nonnumeric file: {os.path.basename(path)} | {exc}")
            continue

        if X.ndim != 2:
            log(f"    skipped because ndim={X.ndim}")
            continue

        original_shape = tuple(X.shape)

        try:
            if X.shape[0] >= X.shape[1]:
                n_samples = X.shape[0]
                n_features = X.shape[1]

                if n_samples < 10 or n_features < 2:
                    log(f"    skipped bad shape: {original_shape}")
                    continue

                sample_n = min(max_points, n_samples)
                idx = np.sort(rng.choice(n_samples, size=sample_n, replace=False))
                X_small = np.asarray(X[idx, :], dtype=np.float64)

            else:
                n_samples = X.shape[1]
                n_features = X.shape[0]

                if n_samples < 10 or n_features < 2:
                    log(f"    skipped bad transposed shape: {original_shape}")
                    continue

                sample_n = min(max_points, n_samples)
                idx = np.sort(rng.choice(n_samples, size=sample_n, replace=False))
                X_small = np.asarray(X[:, idx].T, dtype=np.float64)

        except Exception as exc:
            log(f"    failed while sampling {os.path.basename(path)} | {exc}")
            continue

        X_small = X_small[np.all(np.isfinite(X_small), axis=1)]

        if X_small.shape[0] >= 10 and X_small.shape[1] >= 2:
            return X_small, path, original_shape

        log(f"    sampled data unusable: sampled_shape={X_small.shape}")

    return None, None, None


def standardize_point_cloud_small(X, max_dims=3):
    X = np.asarray(X, dtype=np.float64)
    X = X[np.all(np.isfinite(X), axis=1)]

    if X.shape[0] < 10:
        return None

    X = X - np.mean(X, axis=0, keepdims=True)

    if X.shape[1] > max_dims:
        _U, _S, Vt = np.linalg.svd(X, full_matrices=False)
        X = X @ Vt[:max_dims].T

    X = X / (np.std(X, axis=0, keepdims=True) + 1e-12)

    D = pdist(X)
    positive = D[D > 0]

    if len(positive) > 0:
        X = X / (np.nanmedian(positive) + 1e-12)

    return X


def estimate_complex_counts(X, radius):
    D = squareform(pdist(X))
    A = (D <= radius) & (D > 0)

    n_edges = int(np.sum(A) // 2)

    if A.shape[0] <= 300:
        A_int = A.astype(np.int64)
        n_triangles = int(np.trace(A_int @ A_int @ A_int) // 6)
    else:
        n_triangles = np.nan

    return n_edges, n_triangles


def run_h1_ripser(X, radius):
    result = ripser(
        X,
        maxdim=1,
        thresh=float(radius),
        coeff=2,
        do_cocycles=False,
    )

    diagrams = result["dgms"]

    h0 = diagrams[0] if len(diagrams) > 0 else np.empty((0, 2), dtype=float)
    h1 = diagrams[1] if len(diagrams) > 1 else np.empty((0, 2), dtype=float)

    return h0, h1


def split_finite_essential(diagram):
    diagram = np.asarray(diagram, dtype=float)

    if diagram.size == 0:
        return np.empty((0, 2), dtype=float), np.empty((0, 2), dtype=float)

    finite = diagram[
        np.isfinite(diagram[:, 0])
        & np.isfinite(diagram[:, 1])
        & (diagram[:, 1] > diagram[:, 0])
    ]

    essential = diagram[
        np.isfinite(diagram[:, 0])
        & np.isinf(diagram[:, 1])
    ]

    return finite, essential


def clean_diagram_for_wasserstein(
    finite_intervals,
    essential_intervals=None,
    cap_essential=True,
    cap_value=None,
):
    cleaned = []
    capped_count = 0

    if finite_intervals is not None:
        for b, d in finite_intervals:
            if np.isfinite(b) and np.isfinite(d) and d > b:
                cleaned.append((float(b), float(d)))

    if cap_essential and essential_intervals is not None and cap_value is not None:
        for b, _d in essential_intervals:
            if np.isfinite(b) and np.isfinite(cap_value) and cap_value > b:
                cleaned.append((float(b), float(cap_value)))
                capped_count += 1

    if len(cleaned) == 0:
        return np.empty((0, 2), dtype=float), capped_count

    return np.asarray(cleaned, dtype=float), capped_count


def diagram_features(diagram):
    diagram = np.asarray(diagram, dtype=float)

    if diagram.size == 0 or len(diagram) == 0:
        return {
            "n_intervals_diagram": 0,
            "mean_lifetime": np.nan,
            "max_lifetime": np.nan,
            "total_lifetime": 0.0,
            "mean_birth": np.nan,
            "mean_death": np.nan,
            "max_death": np.nan,
        }

    lifetimes = diagram[:, 1] - diagram[:, 0]

    return {
        "n_intervals_diagram": int(len(diagram)),
        "mean_lifetime": float(np.nanmean(lifetimes)),
        "max_lifetime": float(np.nanmax(lifetimes)),
        "total_lifetime": float(np.nansum(lifetimes)),
        "mean_birth": float(np.nanmean(diagram[:, 0])),
        "mean_death": float(np.nanmean(diagram[:, 1])),
        "max_death": float(np.nanmax(diagram[:, 1])),
    }


def empty_aware_wasserstein(D1, D2):
    D1 = np.asarray(D1, dtype=float)
    D2 = np.asarray(D2, dtype=float)

    if D1.size == 0:
        D1 = np.empty((0, 2), dtype=float)

    if D2.size == 0:
        D2 = np.empty((0, 2), dtype=float)

    n1 = len(D1)
    n2 = len(D2)

    if n1 == 0 and n2 == 0:
        return 0.0

    if n1 == 0:
        return float(np.nansum((D2[:, 1] - D2[:, 0]) / 2.0))

    if n2 == 0:
        return float(np.nansum((D1[:, 1] - D1[:, 0]) / 2.0))

    return float(wasserstein(D1, D2, matching=False))


def classical_mds_from_distances(D, n_components=2):
    D = np.asarray(D, dtype=float)
    n = D.shape[0]

    if n < 2:
        return np.zeros((n, n_components), dtype=float)

    D2 = D ** 2
    J = np.eye(n) - np.ones((n, n)) / n
    B = -0.5 * J @ D2 @ J

    if HAVE_SCIPY_EIGH:
        vals, vecs = eigh(B)
    else:
        vals, vecs = np.linalg.eigh(B)

    idx = np.argsort(vals)[::-1]
    vals = vals[idx]
    vecs = vecs[:, idx]

    coords = np.zeros((n, n_components), dtype=float)

    for k in range(min(n_components, n)):
        if vals[k] > 0:
            coords[:, k] = vecs[:, k] * np.sqrt(vals[k])

    return coords


log("\n" + "=" * 72)
log(f"Running H1 Wasserstein Distances with ripser backend | {FAST_OPTION.upper()}")
log(f"Channels: {len(eeg_channels)}")
log(f"max_points={MAX_POINTS}, radius_quantile={RADIUS_QUANTILE}")
log(f"cap_essential_h1={CAP_ESSENTIAL_H1}")
log(f"expert_plots={MAKE_EXPERT_PLOTS}, render_all_images={RENDER_ALL_IMAGES_INLINE}")
log(f"Output: {mode_out_dir}")
log("=" * 72)

diagnostics = {}
diagram_store = {}
raw_interval_store = {}
loaded_channels = []

overall_t0 = time.time()

for ch_idx, ch_name in enumerate(eeg_channels, start=1):
    t0 = time.time()
    prefix = f"[{ch_idx:02d}/{len(eeg_channels)} {ch_name}]"

    log(f"\n{prefix} loading + pre-subsampling to {MAX_POINTS} points")

    X_small, source_path, original_shape = load_channel_embedding_sampled(
        ch_name,
        max_points=MAX_POINTS,
        seed=500 + ch_idx,
    )

    if X_small is None:
        log(f"{prefix} WARNING: no usable embedding found; skipping")
        continue

    log(f"{prefix} source shape={original_shape}, sampled shape={X_small.shape}")

    X = standardize_point_cloud_small(X_small, max_dims=MAX_DIMS)

    if X is None:
        log(f"{prefix} WARNING: invalid sampled embedding; skipping")
        continue

    log(f"{prefix} standardized shape={X.shape}; computing radius")

    D_flat = pdist(X)
    positive = D_flat[D_flat > 0]

    if len(positive) == 0:
        log(f"{prefix} WARNING: zero-distance cloud; skipping")
        continue

    radius = float(np.quantile(positive, RADIUS_QUANTILE))

    log(f"{prefix} estimating threshold complex size")
    n_edges, n_triangles = estimate_complex_counts(X, radius)

    log(f"{prefix} ripser H0/H1")
    h0_diag, h1_diag = run_h1_ripser(X, radius)

    h0_finite, h0_essential = split_finite_essential(h0_diag)
    h1_finite, h1_essential = split_finite_essential(h1_diag)

    h1_diagram, n_capped_h1 = clean_diagram_for_wasserstein(
        h1_finite,
        essential_intervals=h1_essential,
        cap_essential=CAP_ESSENTIAL_H1,
        cap_value=radius,
    )

    elapsed = time.time() - t0

    diagram_store[ch_name] = h1_diagram

    raw_interval_store[ch_name] = {
        "H0_finite": h0_finite,
        "H0_essential": h0_essential,
        "H1_finite": h1_finite,
        "H1_essential": h1_essential,
        "H1_wasserstein_diagram": h1_diagram,
    }

    diagnostics[ch_name] = {
        "channel": ch_name,
        "source_path": source_path,
        "source_shape": str(original_shape),
        "backend": "ripser_presampled",
        "n_points": int(X.shape[0]),
        "n_edges_at_radius": int(n_edges),
        "n_triangles_at_radius": int(n_triangles) if np.isfinite(n_triangles) else np.nan,
        "radius": radius,
        "n_h0_finite": int(len(h0_finite)),
        "n_h1_finite": int(len(h1_finite)),
        "n_h1_essential": int(len(h1_essential)),
        "n_h1_capped_for_wasserstein": int(n_capped_h1),
        "elapsed_sec": elapsed,
        **diagram_features(h1_diagram),
    }

    loaded_channels.append(ch_name)

    log(
        f"{prefix} done {elapsed:.2f}s | "
        f"edges={n_edges}, triangles={n_triangles}, "
        f"H1 finite={len(h1_finite)}, H1 wasserstein diagram={len(h1_diagram)}"
    )

if len(loaded_channels) == 0:
    raise RuntimeError("No usable embeddings found. Check embedding_data paths.")

log("\nComputing pairwise Wasserstein distances")

n_ch = len(loaded_channels)
wasserstein_mat = np.zeros((n_ch, n_ch), dtype=float)

pair_rows = []
pair_t0 = time.time()

for i in range(n_ch):
    ch_i = loaded_channels[i]

    for j in range(i + 1, n_ch):
        ch_j = loaded_channels[j]

        dist = empty_aware_wasserstein(
            diagram_store[ch_i],
            diagram_store[ch_j],
        )

        wasserstein_mat[i, j] = dist
        wasserstein_mat[j, i] = dist

        pair_rows.append({
            "channel_a": ch_i,
            "channel_b": ch_j,
            "wasserstein_distance": dist,
        })

    log(f"  completed Wasserstein row {i + 1}/{n_ch}: {ch_i}")

pair_elapsed = time.time() - pair_t0

distance_df = pd.DataFrame(
    wasserstein_mat,
    index=loaded_channels,
    columns=loaded_channels,
)

pairwise_df = pd.DataFrame(pair_rows).sort_values("wasserstein_distance", ascending=False).reset_index(drop=True)

summary_rows = []

for idx, ch_name in enumerate(loaded_channels):
    d = wasserstein_mat[idx].copy()
    nonself = d[np.arange(len(d)) != idx]

    summary_rows.append({
        **diagnostics[ch_name],
        "mean_wasserstein_to_others": float(np.nanmean(nonself)) if len(nonself) else 0.0,
        "median_wasserstein_to_others": float(np.nanmedian(nonself)) if len(nonself) else 0.0,
        "max_wasserstein_to_others": float(np.nanmax(nonself)) if len(nonself) else 0.0,
        "min_wasserstein_to_others": float(np.nanmin(nonself)) if len(nonself) else 0.0,
    })

summary_df = (
    pd.DataFrame(summary_rows)
    .sort_values("mean_wasserstein_to_others", ascending=False)
    .reset_index(drop=True)
)

summary_csv_path = os.path.join(mode_out_dir, "h1_wasserstein_channel_summary.csv")
distance_csv_path = os.path.join(mode_out_dir, "h1_wasserstein_distance_matrix.csv")
pairwise_csv_path = os.path.join(mode_out_dir, "h1_wasserstein_pairwise_distances.csv")

summary_df.to_csv(summary_csv_path, index=False)
distance_df.to_csv(distance_csv_path)
pairwise_df.to_csv(pairwise_csv_path, index=False)

npz_path = os.path.join(mode_out_dir, "h1_wasserstein_distances_and_diagrams.npz")

save_dict = {
    "channels": np.asarray(loaded_channels),
    "wasserstein_distance_matrix": wasserstein_mat,
    "fast_option": np.asarray([FAST_OPTION]),
    "backend": np.asarray(["ripser_presampled"]),
}

for ch_name in loaded_channels:
    save_dict[f"{ch_name}_H1_wasserstein_diagram"] = diagram_store[ch_name]

    for key, val in raw_interval_store[ch_name].items():
        save_dict[f"{ch_name}_{key}"] = val

np.savez(npz_path, **save_dict)

print("\nH1 Wasserstein Channel Summary:")
if HAVE_IPYTHON:
    display(summary_df)
else:
    print(summary_df)

print("\nLargest H1 Wasserstein distances:")
if HAVE_IPYTHON:
    display(pairwise_df.head(12))
else:
    print(pairwise_df.head(12))

log(f"\nSaved summary CSV: {summary_csv_path}")
log(f"Saved distance matrix CSV: {distance_csv_path}")
log(f"Saved pairwise CSV: {pairwise_csv_path}")
log(f"Saved NPZ: {npz_path}")
log(f"Pairwise Wasserstein elapsed: {pair_elapsed:.2f}s")

plot_df = summary_df.copy().sort_values("mean_wasserstein_to_others", ascending=False).reset_index(drop=True)

fig, ax = plt.subplots(figsize=(13, 5.8), facecolor=BLACK)
style_ax(
    ax,
    title=f"Mean H1 Wasserstein Distance to Other Channels | {FAST_OPTION.upper()}",
    xlabel="Channel",
    ylabel="Mean Wasserstein distance",
)

x_pos = np.arange(len(plot_df))
y = plot_df["mean_wasserstein_to_others"].to_numpy(dtype=float)

ax.bar(x_pos, y, color=ACCENT, edgecolor=ACCENT, alpha=0.78)

if np.any(np.isfinite(y)):
    ax.axhline(
        np.nanmean(y),
        color=ACCENT_SOFT,
        linestyle="--",
        linewidth=1.4,
        alpha=0.8,
        label=f"Mean={np.nanmean(y):.5f}",
    )

ax.set_xticks(x_pos)
ax.set_xticklabels(plot_df["channel"], rotation=90)
ax.legend(loc="best", framealpha=1.0)

mean_dist_path = os.path.join(mode_out_dir, "01_mean_h1_wasserstein_distance_by_channel.png")
plt.tight_layout()
save_render_close(fig, mean_dist_path)

if n_ch >= 3 and np.any(wasserstein_mat > 0):
    condensed = wasserstein_mat[np.triu_indices(n_ch, k=1)]
    Z = linkage(condensed, method="average")
    order = leaves_list(Z)
else:
    order = np.arange(n_ch)

ordered_dist = wasserstein_mat[np.ix_(order, order)]
ordered_channels = [loaded_channels[i] for i in order]

fig, ax = plt.subplots(figsize=(9, 8), facecolor=BLACK)
style_ax(
    ax,
    title="Pairwise H1 Wasserstein Distance Matrix",
    xlabel="Channel",
    ylabel="Channel",
)

im = ax.imshow(
    ordered_dist,
    aspect="equal",
    interpolation="nearest",
    cmap=CYAN_SEQUENTIAL_CMAP,
)

ax.set_xticks(np.arange(n_ch))
ax.set_xticklabels(ordered_channels, rotation=90)
ax.set_yticks(np.arange(n_ch))
ax.set_yticklabels(ordered_channels)

cbar = plt.colorbar(im, ax=ax, shrink=0.85)
style_colorbar(cbar, "Wasserstein distance")

dist_heatmap_path = os.path.join(mode_out_dir, "02_h1_wasserstein_distance_heatmap.png")
plt.tight_layout()
save_render_close(fig, dist_heatmap_path)

mds_coords = classical_mds_from_distances(wasserstein_mat, n_components=2)

mds_df = pd.DataFrame({
    "channel": loaded_channels,
    "mds_1": mds_coords[:, 0],
    "mds_2": mds_coords[:, 1],
    "mean_wasserstein_to_others": [
        summary_df.loc[summary_df["channel"] == ch, "mean_wasserstein_to_others"].values[0]
        for ch in loaded_channels
    ],
    "n_intervals_diagram": [
        summary_df.loc[summary_df["channel"] == ch, "n_intervals_diagram"].values[0]
        for ch in loaded_channels
    ],
})

mds_csv_path = os.path.join(mode_out_dir, "h1_wasserstein_mds_coordinates.csv")
mds_df.to_csv(mds_csv_path, index=False)

fig, ax = plt.subplots(figsize=(8, 7), facecolor=BLACK)
style_ax(
    ax,
    title="Classical MDS of H1 Wasserstein Distances",
    xlabel="MDS 1",
    ylabel="MDS 2",
)

sc = ax.scatter(
    mds_df["mds_1"],
    mds_df["mds_2"],
    c=mds_df["mean_wasserstein_to_others"],
    s=150,
    cmap=CYAN_SEQUENTIAL_CMAP,
    edgecolor=ACCENT,
    linewidth=1.1,
)

for _, row in mds_df.iterrows():
    ax.text(
        row["mds_1"],
        row["mds_2"],
        row["channel"],
        fontsize=8,
        color=ACCENT,
        alpha=0.9,
    )

cbar = plt.colorbar(sc, ax=ax, shrink=0.85)
style_colorbar(cbar, "Mean Wasserstein distance")

mds_path = os.path.join(mode_out_dir, "03_h1_wasserstein_mds_embedding.png")
plt.tight_layout()
save_render_close(fig, mds_path)

diagram_channels = summary_df.head(min(12, len(summary_df)))["channel"].tolist()
n_cols = 4
n_rows = int(np.ceil(len(diagram_channels) / n_cols))

fig, axes = plt.subplots(
    n_rows,
    n_cols,
    figsize=(4 * n_cols, 3.2 * n_rows),
    facecolor=BLACK,
)

axes = np.asarray(axes).reshape(-1)

all_diagram_points = np.vstack([
    diagram_store[ch] for ch in loaded_channels if len(diagram_store[ch]) > 0
]) if any(len(diagram_store[ch]) > 0 for ch in loaded_channels) else np.empty((0, 2))

if len(all_diagram_points) > 0:
    diag_max = float(np.nanmax(all_diagram_points[:, 1]))
else:
    diag_max = 1.0

diag_max = max(diag_max, 1e-6)

for ax_idx, ch_name in enumerate(diagram_channels):
    ax = axes[ax_idx]
    D = diagram_store[ch_name]

    mean_dist = summary_df.loc[
        summary_df["channel"] == ch_name,
        "mean_wasserstein_to_others",
    ].values[0]

    style_ax(
        ax,
        title=f"{ch_name}: mean W={mean_dist:.3f}, n={len(D)}",
        xlabel="Birth",
        ylabel="Death",
    )

    ax.plot([0, diag_max], [0, diag_max], color=ACCENT_DIM, linestyle="--", linewidth=1.0)

    if len(D) > 0:
        ax.scatter(
            D[:, 0],
            D[:, 1],
            s=50,
            facecolors=BLACK,
            edgecolors=ACCENT,
            linewidths=1.2,
        )

    ax.set_xlim(0, diag_max * 1.05)
    ax.set_ylim(0, diag_max * 1.05)

for j in range(len(diagram_channels), len(axes)):
    axes[j].axis("off")

diagrams_path = os.path.join(mode_out_dir, "04_top_channels_h1_persistence_diagrams.png")
plt.tight_layout()
save_render_close(fig, diagrams_path, width=1100)

fig, ax = plt.subplots(figsize=(12, 6), facecolor=BLACK)
style_ax(
    ax,
    title="Largest Pairwise H1 Wasserstein Distances",
    xlabel="Channel pair",
    ylabel="Wasserstein distance",
)

top_pairs = pairwise_df.head(min(15, len(pairwise_df))).copy()
pair_labels = top_pairs["channel_a"] + "-" + top_pairs["channel_b"]

x_pair = np.arange(len(top_pairs))

ax.bar(
    x_pair,
    top_pairs["wasserstein_distance"],
    color=ACCENT,
    edgecolor=ACCENT,
    alpha=0.78,
)

ax.set_xticks(x_pair)
ax.set_xticklabels(pair_labels, rotation=60, ha="right")

top_pairs_path = os.path.join(mode_out_dir, "05_largest_pairwise_h1_wasserstein_distances.png")
plt.tight_layout()
save_render_close(fig, top_pairs_path)

if MAKE_EXPERT_PLOTS:
    expert_dir = os.path.join(mode_out_dir, "expert_plots")
    os.makedirs(expert_dir, exist_ok=True)

    mean_wasserstein_map = dict(zip(summary_df["channel"], summary_df["mean_wasserstein_to_others"]))

    scalp_xy = {
        "Fp1": (-0.45, 1.00), "Fpz": (0.00, 1.08), "Fp2": (0.45, 1.00),
        "F7": (-0.95, 0.60), "F3": (-0.45, 0.55), "Fz": (0.00, 0.60),
        "F4": (0.45, 0.55), "F8": (0.95, 0.60),
        "FC5": (-0.75, 0.25), "FC1": (-0.25, 0.25),
        "FC2": (0.25, 0.25), "FC6": (0.75, 0.25),
        "M1": (-1.15, 0.05), "T7": (-1.00, 0.00), "C3": (-0.50, 0.00),
        "Cz": (0.00, 0.00), "C4": (0.50, 0.00), "T8": (1.00, 0.00),
        "M2": (1.15, 0.05),
        "CP5": (-0.75, -0.30), "CP1": (-0.25, -0.30),
        "CP2": (0.25, -0.30), "CP6": (0.75, -0.30),
        "P7": (-0.90, -0.60), "P3": (-0.45, -0.60), "Pz": (0.00, -0.65),
        "P4": (0.45, -0.60), "P8": (0.90, -0.60),
        "POz": (0.00, -0.88),
        "O1": (-0.35, -1.00), "Oz": (0.00, -1.08), "O2": (0.35, -1.00),
    }

    xs, ys, vals, labels = [], [], [], []

    for ch in eeg_channels:
        if ch in scalp_xy and ch in mean_wasserstein_map and np.isfinite(mean_wasserstein_map[ch]):
            xs.append(scalp_xy[ch][0])
            ys.append(scalp_xy[ch][1])
            vals.append(mean_wasserstein_map[ch])
            labels.append(ch)

    fig, ax = plt.subplots(figsize=(7, 7), facecolor=BLACK)
    style_ax(
        ax,
        title=f"Scalp Map of Mean H1 Wasserstein Distance | {FAST_OPTION.upper()}",
        grid=False,
    )

    head = plt.Circle((0, 0), 1.12, fill=False, color=ACCENT, linewidth=2)
    ax.add_patch(head)
    ax.plot([0.00, -0.10, 0.10, 0.00], [1.12, 1.28, 1.28, 1.12], color=ACCENT, linewidth=2)

    sc = ax.scatter(
        xs,
        ys,
        c=vals,
        s=260,
        cmap=CYAN_SEQUENTIAL_CMAP,
        edgecolor=ACCENT,
        linewidth=1.2,
    )

    for x0, y0, label in zip(xs, ys, labels):
        ax.text(
            x0,
            y0,
            label,
            ha="center",
            va="center",
            fontsize=8,
            color=BLACK,
            fontweight="bold",
        )

    cbar = plt.colorbar(sc, ax=ax, shrink=0.75)
    style_colorbar(cbar, "Mean Wasserstein distance")

    ax.set_xlim(-1.35, 1.35)
    ax.set_ylim(-1.35, 1.35)
    ax.set_aspect("equal")
    ax.set_xticks([])
    ax.set_yticks([])

    scalp_path = os.path.join(expert_dir, "06_mean_h1_wasserstein_distance_scalp_map.png")
    plt.tight_layout()
    save_render_close(fig, scalp_path, width=760)

    regions = {
        "Frontal": ["Fp1", "Fpz", "Fp2", "F7", "F3", "Fz", "F4", "F8"],
        "Frontocentral": ["FC5", "FC1", "FC2", "FC6"],
        "Central": ["T7", "C3", "Cz", "C4", "T8"],
        "Centroparietal": ["CP5", "CP1", "CP2", "CP6"],
        "Parietal": ["P7", "P3", "Pz", "P4", "P8"],
        "Occipital": ["POz", "O1", "Oz", "O2"],
        "Mastoid": ["M1", "M2"],
    }

    region_rows = []
    region_points = []

    for region_name, chans in regions.items():
        vals_region = []

        for ch in chans:
            if ch in mean_wasserstein_map and np.isfinite(mean_wasserstein_map[ch]):
                vals_region.append(mean_wasserstein_map[ch])
                region_points.append({
                    "region": region_name,
                    "channel": ch,
                    "mean_wasserstein_to_others": mean_wasserstein_map[ch],
                })

        if len(vals_region) == 0:
            continue

        vals_region = np.asarray(vals_region, dtype=float)

        region_rows.append({
            "region": region_name,
            "mean_region_wasserstein": float(np.mean(vals_region)),
            "std_region_wasserstein": float(np.std(vals_region, ddof=1)) if len(vals_region) > 1 else 0.0,
            "sem_region_wasserstein": (
                float(np.std(vals_region, ddof=1) / np.sqrt(len(vals_region)))
                if len(vals_region) > 1 else 0.0
            ),
            "n_channels": int(len(vals_region)),
        })

    region_df = pd.DataFrame(region_rows)
    region_points_df = pd.DataFrame(region_points)

    region_csv_path = os.path.join(expert_dir, "regional_h1_wasserstein_summary.csv")
    region_df.to_csv(region_csv_path, index=False)

    if len(region_df) > 0:
        fig, ax = plt.subplots(figsize=(12, 6.5), facecolor=BLACK)
        style_ax(
            ax,
            title=f"Regional Mean H1 Wasserstein Distance +/- SEM | {FAST_OPTION.upper()}",
            xlabel="Region",
            ylabel="Mean Wasserstein distance",
        )

        x_region = np.arange(len(region_df))

        ax.bar(
            x_region,
            region_df["mean_region_wasserstein"],
            yerr=region_df["sem_region_wasserstein"],
            color=ACCENT,
            edgecolor=ACCENT,
            alpha=0.82,
            width=0.68,
            capsize=8,
            error_kw={
                "elinewidth": 2.2,
                "capthick": 2.2,
                "ecolor": ACCENT,
                "alpha": 1.0,
            },
            label="Mean +/- SEM",
        )

        rng = np.random.default_rng(7)

        for i, region_name in enumerate(region_df["region"]):
            vals = region_points_df.loc[
                region_points_df["region"] == region_name,
                "mean_wasserstein_to_others",
            ].to_numpy(dtype=float)

            if len(vals) == 0:
                continue

            jitter = rng.uniform(-0.13, 0.13, size=len(vals))

            ax.scatter(
                np.full(len(vals), i) + jitter,
                vals,
                s=42,
                facecolors=BLACK,
                edgecolors=ACCENT,
                linewidths=1.2,
                alpha=0.95,
                zorder=4,
            )

        for i, row in region_df.iterrows():
            y_val = row["mean_region_wasserstein"]
            sem = row["sem_region_wasserstein"]

            ax.text(
                i,
                y_val + sem,
                f"{y_val:.3f}\n+/-{sem:.3f}\nn={int(row['n_channels'])}",
                ha="center",
                va="bottom",
                fontsize=8,
                color=ACCENT,
            )

        ax.set_xticks(x_region)
        ax.set_xticklabels(
            region_df["region"],
            rotation=35,
            ha="right",
            rotation_mode="anchor",
        )
        ax.margins(x=0.04)
        ax.legend(loc="best", framealpha=1.0)

        region_path = os.path.join(expert_dir, "07_regional_h1_wasserstein_summary.png")
        plt.tight_layout(pad=2.0)
        save_render_close(fig, region_path)

    fig, ax = plt.subplots(figsize=(8, 6), facecolor=BLACK)
    style_ax(
        ax,
        title="Diagram Size vs Mean H1 Wasserstein Distance",
        xlabel="H1 diagram intervals",
        ylabel="Mean Wasserstein distance to others",
    )

    sc = ax.scatter(
        summary_df["n_intervals_diagram"],
        summary_df["mean_wasserstein_to_others"],
        c=summary_df["max_lifetime"],
        s=140,
        cmap=CYAN_SEQUENTIAL_CMAP,
        edgecolor=ACCENT,
        linewidth=1.0,
    )

    for _, row in summary_df.iterrows():
        ax.text(
            row["n_intervals_diagram"],
            row["mean_wasserstein_to_others"],
            row["channel"],
            fontsize=8,
            color=ACCENT,
            alpha=0.85,
        )

    cbar = plt.colorbar(sc, ax=ax, shrink=0.85)
    style_colorbar(cbar, "Max H1 lifetime")

    diagram_size_path = os.path.join(expert_dir, "08_diagram_size_vs_mean_wasserstein.png")
    plt.tight_layout()
    save_render_close(fig, diagram_size_path)

    fig, ax = plt.subplots(figsize=(8, 6), facecolor=BLACK)
    style_ax(
        ax,
        title="Threshold Triangles vs Mean H1 Wasserstein Distance",
        xlabel="Triangles at radius",
        ylabel="Mean Wasserstein distance to others",
    )

    sc = ax.scatter(
        summary_df["n_triangles_at_radius"],
        summary_df["mean_wasserstein_to_others"],
        c=summary_df["n_intervals_diagram"],
        s=140,
        cmap=CYAN_SEQUENTIAL_CMAP,
        edgecolor=ACCENT,
        linewidth=1.0,
    )

    for _, row in summary_df.iterrows():
        ax.text(
            row["n_triangles_at_radius"],
            row["mean_wasserstein_to_others"],
            row["channel"],
            fontsize=8,
            color=ACCENT,
            alpha=0.85,
        )

    cbar = plt.colorbar(sc, ax=ax, shrink=0.85)
    style_colorbar(cbar, "H1 diagram intervals")

    triangle_path = os.path.join(expert_dir, "09_threshold_triangles_vs_mean_wasserstein.png")
    plt.tight_layout()
    save_render_close(fig, triangle_path)

    qc_df = summary_df.copy().sort_values("elapsed_sec", ascending=False).reset_index(drop=True)

    fig, ax = plt.subplots(figsize=(12, 5.8), facecolor=BLACK)
    style_ax(
        ax,
        title="Wasserstein Diagram Runtime QC",
        xlabel="Channel",
        ylabel="Elapsed seconds",
    )

    x_qc = np.arange(len(qc_df))

    ax.bar(
        x_qc,
        qc_df["elapsed_sec"],
        color=ACCENT,
        edgecolor=ACCENT,
        alpha=0.82,
    )

    ax.set_xticks(x_qc)
    ax.set_xticklabels(qc_df["channel"], rotation=90)

    qc_path = os.path.join(expert_dir, "10_runtime_qc.png")
    plt.tight_layout()
    save_render_close(fig, qc_path)

    log(f"\nExpert plots saved and rendered from: {expert_dir}")
    log(f"Regional summary CSV: {region_csv_path}")

elapsed_total = time.time() - overall_t0

log("\n" + "=" * 72)
log(f"Finished H1 Wasserstein Distances in {elapsed_total:.2f}s")
log(f"Mode: {FAST_OPTION}")
log(f"Backend: ripser_presampled + persim.wasserstein")
log(f"Channels completed: {len(loaded_channels)}")
log(f"Summary CSV: {summary_csv_path}")
log(f"Distance matrix CSV: {distance_csv_path}")
log(f"Pairwise CSV: {pairwise_csv_path}")
log(f"MDS CSV: {mds_csv_path}")
log(f"NPZ: {npz_path}")
log(f"Output directory: {mode_out_dir}")
log("=" * 72)
```

```python
import os
import sys
import time
import numpy as np

if not hasattr(np, "trapz"):
    np.trapz = np.trapezoid

import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.colors import LinearSegmentedColormap

os.environ["OMP_NUM_THREADS"] = "2"
os.environ["OPENBLAS_NUM_THREADS"] = "2"
os.environ["MKL_NUM_THREADS"] = "2"
os.environ["VECLIB_MAXIMUM_THREADS"] = "2"
os.environ["NUMEXPR_NUM_THREADS"] = "2"

from scipy.spatial.distance import pdist, squareform
from scipy.cluster.hierarchy import linkage, leaves_list

try:
    from scipy.linalg import eigh
    HAVE_SCIPY_EIGH = True
except Exception:
    HAVE_SCIPY_EIGH = False

try:
    from IPython.display import display, Image
    HAVE_IPYTHON = True
except Exception:
    HAVE_IPYTHON = False

try:
    from ripser import ripser
except Exception as exc:
    raise ImportError(
        "ripser is not installed. Run this in a separate notebook cell first:\n\n"
        "!pip install ripser persim\n\n"
        "Then rerun this Sliced Wasserstein Distance cell."
    ) from exc


base_dir = "/home/a/projects/Complete-Neural-Signal-Analysis"
out_dir = os.path.join(base_dir, "topology_sliced_wasserstein_distances")
os.makedirs(out_dir, exist_ok=True)

FAST_OPTION = "ultra"  # options: "ultra", "fast", "balanced", "full"

MAX_CHANNELS_TO_PROCESS = None
RENDER_ALL_IMAGES_INLINE = True
SAVE_PLOTS_TO_DISK = True
MAKE_EXPERT_PLOTS = True
INLINE_IMAGE_WIDTH = 950
VERBOSE_PROGRESS = True

MODE_CONFIGS = {
    "ultra": {
        "max_points": 55,
        "max_dims": 3,
        "radius_quantile": 0.12,
        "n_sliced_directions": 24,
        "cap_essential_h1": True,
        "plot_dpi": 120,
    },
    "fast": {
        "max_points": 90,
        "max_dims": 3,
        "radius_quantile": 0.15,
        "n_sliced_directions": 32,
        "cap_essential_h1": True,
        "plot_dpi": 130,
    },
    "balanced": {
        "max_points": 130,
        "max_dims": 3,
        "radius_quantile": 0.18,
        "n_sliced_directions": 48,
        "cap_essential_h1": True,
        "plot_dpi": 140,
    },
    "full": {
        "max_points": 180,
        "max_dims": 3,
        "radius_quantile": 0.22,
        "n_sliced_directions": 64,
        "cap_essential_h1": True,
        "plot_dpi": 150,
    },
}

if FAST_OPTION not in MODE_CONFIGS:
    raise ValueError("FAST_OPTION must be one of: ultra, fast, balanced, full")

cfg = MODE_CONFIGS[FAST_OPTION]

MAX_POINTS = cfg["max_points"]
MAX_DIMS = cfg["max_dims"]
RADIUS_QUANTILE = cfg["radius_quantile"]
N_SLICED_DIRECTIONS = cfg["n_sliced_directions"]
CAP_ESSENTIAL_H1 = cfg["cap_essential_h1"]
PLOT_DPI = cfg["plot_dpi"]

mode_out_dir = os.path.join(out_dir, f"h1_sliced_wasserstein_distances_{FAST_OPTION}_ripser_presampled")
os.makedirs(mode_out_dir, exist_ok=True)

eeg_channels = [
    "Fp1", "Fpz", "Fp2", "F7", "F3", "Fz", "F4", "F8", "FC5", "FC1", "FC2", "FC6",
    "M1", "T7", "C3", "Cz", "C4", "T8", "M2", "CP5", "CP1", "CP2", "CP6",
    "P7", "P3", "Pz", "P4", "P8", "POz", "O1", "Oz", "O2"
]

if MAX_CHANNELS_TO_PROCESS is not None:
    eeg_channels = eeg_channels[:MAX_CHANNELS_TO_PROCESS]

ACCENT = "#00FFFF"
ACCENT_SOFT = "#66FFFF"
ACCENT_DIM = "#008B8B"
BLACK = "#000000"

CYAN_SEQUENTIAL_CMAP = LinearSegmentedColormap.from_list(
    "cyan_sequential",
    [BLACK, ACCENT_DIM, ACCENT, ACCENT_SOFT],
    N=256,
)

plt.rcParams.update({
    "figure.facecolor": BLACK,
    "axes.facecolor": BLACK,
    "savefig.facecolor": BLACK,
    "text.color": ACCENT,
    "axes.labelcolor": ACCENT,
    "axes.edgecolor": ACCENT,
    "axes.titlecolor": ACCENT,
    "xtick.color": ACCENT,
    "ytick.color": ACCENT,
    "grid.color": ACCENT_DIM,
    "grid.alpha": 0.25,
    "legend.facecolor": BLACK,
    "legend.edgecolor": ACCENT,
    "legend.labelcolor": ACCENT,
    "font.size": 11,
})


def log(msg):
    if VERBOSE_PROGRESS:
        print(msg, flush=True)


def style_ax(ax, title=None, xlabel=None, ylabel=None, grid=True):
    ax.set_facecolor(BLACK)

    for spine in ax.spines.values():
        spine.set_color(ACCENT)
        spine.set_linewidth(1.1)

    ax.tick_params(axis="x", colors=ACCENT, labelcolor=ACCENT)
    ax.tick_params(axis="y", colors=ACCENT, labelcolor=ACCENT)

    if title is not None:
        ax.set_title(title, color=ACCENT, pad=12)

    if xlabel is not None:
        ax.set_xlabel(xlabel, color=ACCENT, labelpad=10)

    if ylabel is not None:
        ax.set_ylabel(ylabel, color=ACCENT, labelpad=10)

    if grid:
        ax.grid(True, color=ACCENT_DIM, alpha=0.25, linewidth=0.8)


def style_colorbar(cbar, label):
    cbar.set_label(label, color=ACCENT)
    cbar.ax.yaxis.set_tick_params(color=ACCENT)
    plt.setp(cbar.ax.get_yticklabels(), color=ACCENT)
    cbar.outline.set_edgecolor(ACCENT)


def save_render_close(fig, path, width=INLINE_IMAGE_WIDTH):
    if SAVE_PLOTS_TO_DISK:
        fig.savefig(path, dpi=PLOT_DPI, bbox_inches="tight", facecolor=BLACK)

    plt.close(fig)

    if RENDER_ALL_IMAGES_INLINE and HAVE_IPYTHON and os.path.exists(path):
        display(Image(filename=path, width=width))


def load_channel_embedding_sampled(ch_name, max_points=55, seed=7):
    candidate_paths = [
        os.path.join(base_dir, "embedding_data", "tsne_umap_from_3d", f"umap_embedding_{ch_name}.npy"),
        os.path.join(base_dir, "embedding_data", "tsne_umap_from_3d", f"tsne_embedding_{ch_name}.npy"),
        os.path.join(base_dir, "embedding_data", "3dembedding_data", f"3dembedded_{ch_name}.npy"),
        os.path.join(base_dir, "embedding_data", "2dembedding_data", f"2dembedded_{ch_name}.npy"),
    ]

    rng = np.random.default_rng(seed)

    for path in candidate_paths:
        if not os.path.exists(path):
            continue

        log(f"    trying: {path}")

        try:
            X = np.load(path, mmap_mode="r", allow_pickle=False)
        except Exception as exc:
            log(f"    skipped non-memmappable or nonnumeric file: {os.path.basename(path)} | {exc}")
            continue

        if X.ndim != 2:
            log(f"    skipped because ndim={X.ndim}")
            continue

        original_shape = tuple(X.shape)

        try:
            if X.shape[0] >= X.shape[1]:
                n_samples = X.shape[0]
                n_features = X.shape[1]

                if n_samples < 10 or n_features < 2:
                    log(f"    skipped bad shape: {original_shape}")
                    continue

                sample_n = min(max_points, n_samples)
                idx = np.sort(rng.choice(n_samples, size=sample_n, replace=False))
                X_small = np.asarray(X[idx, :], dtype=np.float64)

            else:
                n_samples = X.shape[1]
                n_features = X.shape[0]

                if n_samples < 10 or n_features < 2:
                    log(f"    skipped bad transposed shape: {original_shape}")
                    continue

                sample_n = min(max_points, n_samples)
                idx = np.sort(rng.choice(n_samples, size=sample_n, replace=False))
                X_small = np.asarray(X[:, idx].T, dtype=np.float64)

        except Exception as exc:
            log(f"    failed while sampling {os.path.basename(path)} | {exc}")
            continue

        X_small = X_small[np.all(np.isfinite(X_small), axis=1)]

        if X_small.shape[0] >= 10 and X_small.shape[1] >= 2:
            return X_small, path, original_shape

        log(f"    sampled data unusable: sampled_shape={X_small.shape}")

    return None, None, None


def standardize_point_cloud_small(X, max_dims=3):
    X = np.asarray(X, dtype=np.float64)
    X = X[np.all(np.isfinite(X), axis=1)]

    if X.shape[0] < 10:
        return None

    X = X - np.mean(X, axis=0, keepdims=True)

    if X.shape[1] > max_dims:
        _U, _S, Vt = np.linalg.svd(X, full_matrices=False)
        X = X @ Vt[:max_dims].T

    X = X / (np.std(X, axis=0, keepdims=True) + 1e-12)

    D = pdist(X)
    positive = D[D > 0]

    if len(positive) > 0:
        X = X / (np.nanmedian(positive) + 1e-12)

    return X


def estimate_complex_counts(X, radius):
    D = squareform(pdist(X))
    A = (D <= radius) & (D > 0)

    n_edges = int(np.sum(A) // 2)

    if A.shape[0] <= 300:
        A_int = A.astype(np.int64)
        n_triangles = int(np.trace(A_int @ A_int @ A_int) // 6)
    else:
        n_triangles = np.nan

    return n_edges, n_triangles


def run_h1_ripser(X, radius):
    result = ripser(
        X,
        maxdim=1,
        thresh=float(radius),
        coeff=2,
        do_cocycles=False,
    )

    diagrams = result["dgms"]

    h0 = diagrams[0] if len(diagrams) > 0 else np.empty((0, 2), dtype=float)
    h1 = diagrams[1] if len(diagrams) > 1 else np.empty((0, 2), dtype=float)

    return h0, h1


def split_finite_essential(diagram):
    diagram = np.asarray(diagram, dtype=float)

    if diagram.size == 0:
        return np.empty((0, 2), dtype=float), np.empty((0, 2), dtype=float)

    finite = diagram[
        np.isfinite(diagram[:, 0])
        & np.isfinite(diagram[:, 1])
        & (diagram[:, 1] > diagram[:, 0])
    ]

    essential = diagram[
        np.isfinite(diagram[:, 0])
        & np.isinf(diagram[:, 1])
    ]

    return finite, essential


def clean_diagram_for_sliced_wasserstein(
    finite_intervals,
    essential_intervals=None,
    cap_essential=True,
    cap_value=None,
):
    cleaned = []
    capped_count = 0

    if finite_intervals is not None:
        for b, d in finite_intervals:
            if np.isfinite(b) and np.isfinite(d) and d > b:
                cleaned.append((float(b), float(d)))

    if cap_essential and essential_intervals is not None and cap_value is not None:
        for b, _d in essential_intervals:
            if np.isfinite(b) and np.isfinite(cap_value) and cap_value > b:
                cleaned.append((float(b), float(cap_value)))
                capped_count += 1

    if len(cleaned) == 0:
        return np.empty((0, 2), dtype=float), capped_count

    return np.asarray(cleaned, dtype=float), capped_count


def diagram_features(diagram):
    diagram = np.asarray(diagram, dtype=float)

    if diagram.size == 0 or len(diagram) == 0:
        return {
            "n_intervals_diagram": 0,
            "mean_lifetime": np.nan,
            "max_lifetime": np.nan,
            "total_lifetime": 0.0,
            "mean_birth": np.nan,
            "mean_death": np.nan,
            "max_death": np.nan,
        }

    lifetimes = diagram[:, 1] - diagram[:, 0]

    return {
        "n_intervals_diagram": int(len(diagram)),
        "mean_lifetime": float(np.nanmean(lifetimes)),
        "max_lifetime": float(np.nanmax(lifetimes)),
        "total_lifetime": float(np.nansum(lifetimes)),
        "mean_birth": float(np.nanmean(diagram[:, 0])),
        "mean_death": float(np.nanmean(diagram[:, 1])),
        "max_death": float(np.nanmax(diagram[:, 1])),
    }


def diagonal_projection_points(D):
    D = np.asarray(D, dtype=float)

    if D.size == 0:
        return np.empty((0, 2), dtype=float)

    mid = 0.5 * (D[:, 0] + D[:, 1])
    return np.column_stack([mid, mid])


def sliced_wasserstein_distance(D1, D2, n_directions=32, p=1):
    """
    Approximate sliced Wasserstein distance between two persistence diagrams.

    Uses diagonal augmentation:
    - D1 is compared against D2 plus diagonal projections of D1.
    - D2 is compared against D1 plus diagonal projections of D2.
    For each projection direction, it computes 1D Wasserstein distance
    between sorted projected coordinates, then averages over directions.
    """
    D1 = np.asarray(D1, dtype=float)
    D2 = np.asarray(D2, dtype=float)

    if D1.size == 0:
        D1 = np.empty((0, 2), dtype=float)

    if D2.size == 0:
        D2 = np.empty((0, 2), dtype=float)

    if len(D1) == 0 and len(D2) == 0:
        return 0.0

    D1_diag = diagonal_projection_points(D1)
    D2_diag = diagonal_projection_points(D2)

    A = np.vstack([D1, D2_diag]) if len(D2_diag) else D1.copy()
    B = np.vstack([D2, D1_diag]) if len(D1_diag) else D2.copy()

    if len(A) == 0 and len(B) == 0:
        return 0.0

    if len(A) != len(B):
        raise RuntimeError(
            f"Internal sliced Wasserstein augmentation failed: len(A)={len(A)}, len(B)={len(B)}"
        )

    # Directions from 0 to pi. Using endpoint=False avoids duplicate direction.
    thetas = np.linspace(0.0, np.pi, n_directions, endpoint=False)

    distances = []

    for theta in thetas:
        direction = np.array([np.cos(theta), np.sin(theta)], dtype=float)

        proj_a = np.sort(A @ direction)
        proj_b = np.sort(B @ direction)

        diff = np.abs(proj_a - proj_b)

        if p == 1:
            distances.append(float(np.mean(diff)))
        else:
            distances.append(float(np.mean(diff ** p) ** (1.0 / p)))

    return float(np.mean(distances))


def classical_mds_from_distances(D, n_components=2):
    D = np.asarray(D, dtype=float)
    n = D.shape[0]

    if n < 2:
        return np.zeros((n, n_components), dtype=float)

    D2 = D ** 2
    J = np.eye(n) - np.ones((n, n)) / n
    B = -0.5 * J @ D2 @ J

    if HAVE_SCIPY_EIGH:
        vals, vecs = eigh(B)
    else:
        vals, vecs = np.linalg.eigh(B)

    idx = np.argsort(vals)[::-1]
    vals = vals[idx]
    vecs = vecs[:, idx]

    coords = np.zeros((n, n_components), dtype=float)

    for k in range(min(n_components, n)):
        if vals[k] > 0:
            coords[:, k] = vecs[:, k] * np.sqrt(vals[k])

    return coords


log("\n" + "=" * 72)
log(f"Running H1 Sliced Wasserstein Distances with ripser backend | {FAST_OPTION.upper()}")
log(f"Channels: {len(eeg_channels)}")
log(f"max_points={MAX_POINTS}, radius_quantile={RADIUS_QUANTILE}")
log(f"n_sliced_directions={N_SLICED_DIRECTIONS}")
log(f"cap_essential_h1={CAP_ESSENTIAL_H1}")
log(f"expert_plots={MAKE_EXPERT_PLOTS}, render_all_images={RENDER_ALL_IMAGES_INLINE}")
log(f"Output: {mode_out_dir}")
log("=" * 72)

diagnostics = {}
diagram_store = {}
raw_interval_store = {}
loaded_channels = []

overall_t0 = time.time()

for ch_idx, ch_name in enumerate(eeg_channels, start=1):
    t0 = time.time()
    prefix = f"[{ch_idx:02d}/{len(eeg_channels)} {ch_name}]"

    log(f"\n{prefix} loading + pre-subsampling to {MAX_POINTS} points")

    X_small, source_path, original_shape = load_channel_embedding_sampled(
        ch_name,
        max_points=MAX_POINTS,
        seed=600 + ch_idx,
    )

    if X_small is None:
        log(f"{prefix} WARNING: no usable embedding found; skipping")
        continue

    log(f"{prefix} source shape={original_shape}, sampled shape={X_small.shape}")

    X = standardize_point_cloud_small(X_small, max_dims=MAX_DIMS)

    if X is None:
        log(f"{prefix} WARNING: invalid sampled embedding; skipping")
        continue

    log(f"{prefix} standardized shape={X.shape}; computing radius")

    D_flat = pdist(X)
    positive = D_flat[D_flat > 0]

    if len(positive) == 0:
        log(f"{prefix} WARNING: zero-distance cloud; skipping")
        continue

    radius = float(np.quantile(positive, RADIUS_QUANTILE))

    log(f"{prefix} estimating threshold complex size")
    n_edges, n_triangles = estimate_complex_counts(X, radius)

    log(f"{prefix} ripser H0/H1")
    h0_diag, h1_diag = run_h1_ripser(X, radius)

    h0_finite, h0_essential = split_finite_essential(h0_diag)
    h1_finite, h1_essential = split_finite_essential(h1_diag)

    h1_diagram, n_capped_h1 = clean_diagram_for_sliced_wasserstein(
        h1_finite,
        essential_intervals=h1_essential,
        cap_essential=CAP_ESSENTIAL_H1,
        cap_value=radius,
    )

    elapsed = time.time() - t0

    diagram_store[ch_name] = h1_diagram

    raw_interval_store[ch_name] = {
        "H0_finite": h0_finite,
        "H0_essential": h0_essential,
        "H1_finite": h1_finite,
        "H1_essential": h1_essential,
        "H1_sliced_wasserstein_diagram": h1_diagram,
    }

    diagnostics[ch_name] = {
        "channel": ch_name,
        "source_path": source_path,
        "source_shape": str(original_shape),
        "backend": "ripser_presampled",
        "n_points": int(X.shape[0]),
        "n_edges_at_radius": int(n_edges),
        "n_triangles_at_radius": int(n_triangles) if np.isfinite(n_triangles) else np.nan,
        "radius": radius,
        "n_h0_finite": int(len(h0_finite)),
        "n_h1_finite": int(len(h1_finite)),
        "n_h1_essential": int(len(h1_essential)),
        "n_h1_capped_for_sliced_wasserstein": int(n_capped_h1),
        "elapsed_sec": elapsed,
        **diagram_features(h1_diagram),
    }

    loaded_channels.append(ch_name)

    log(
        f"{prefix} done {elapsed:.2f}s | "
        f"edges={n_edges}, triangles={n_triangles}, "
        f"H1 finite={len(h1_finite)}, H1 sliced-W diagram={len(h1_diagram)}"
    )

if len(loaded_channels) == 0:
    raise RuntimeError("No usable embeddings found. Check embedding_data paths.")

log("\nComputing pairwise sliced Wasserstein distances")

n_ch = len(loaded_channels)
sliced_wasserstein_mat = np.zeros((n_ch, n_ch), dtype=float)

pair_rows = []
pair_t0 = time.time()

for i in range(n_ch):
    ch_i = loaded_channels[i]

    for j in range(i + 1, n_ch):
        ch_j = loaded_channels[j]

        dist = sliced_wasserstein_distance(
            diagram_store[ch_i],
            diagram_store[ch_j],
            n_directions=N_SLICED_DIRECTIONS,
            p=1,
        )

        sliced_wasserstein_mat[i, j] = dist
        sliced_wasserstein_mat[j, i] = dist

        pair_rows.append({
            "channel_a": ch_i,
            "channel_b": ch_j,
            "sliced_wasserstein_distance": dist,
            "n_sliced_directions": N_SLICED_DIRECTIONS,
        })

    log(f"  completed sliced Wasserstein row {i + 1}/{n_ch}: {ch_i}")

pair_elapsed = time.time() - pair_t0

distance_df = pd.DataFrame(
    sliced_wasserstein_mat,
    index=loaded_channels,
    columns=loaded_channels,
)

pairwise_df = (
    pd.DataFrame(pair_rows)
    .sort_values("sliced_wasserstein_distance", ascending=False)
    .reset_index(drop=True)
)

summary_rows = []

for idx, ch_name in enumerate(loaded_channels):
    d = sliced_wasserstein_mat[idx].copy()
    nonself = d[np.arange(len(d)) != idx]

    summary_rows.append({
        **diagnostics[ch_name],
        "mean_sliced_wasserstein_to_others": float(np.nanmean(nonself)) if len(nonself) else 0.0,
        "median_sliced_wasserstein_to_others": float(np.nanmedian(nonself)) if len(nonself) else 0.0,
        "max_sliced_wasserstein_to_others": float(np.nanmax(nonself)) if len(nonself) else 0.0,
        "min_sliced_wasserstein_to_others": float(np.nanmin(nonself)) if len(nonself) else 0.0,
    })

summary_df = (
    pd.DataFrame(summary_rows)
    .sort_values("mean_sliced_wasserstein_to_others", ascending=False)
    .reset_index(drop=True)
)

summary_csv_path = os.path.join(mode_out_dir, "h1_sliced_wasserstein_channel_summary.csv")
distance_csv_path = os.path.join(mode_out_dir, "h1_sliced_wasserstein_distance_matrix.csv")
pairwise_csv_path = os.path.join(mode_out_dir, "h1_sliced_wasserstein_pairwise_distances.csv")

summary_df.to_csv(summary_csv_path, index=False)
distance_df.to_csv(distance_csv_path)
pairwise_df.to_csv(pairwise_csv_path, index=False)

npz_path = os.path.join(mode_out_dir, "h1_sliced_wasserstein_distances_and_diagrams.npz")

save_dict = {
    "channels": np.asarray(loaded_channels),
    "sliced_wasserstein_distance_matrix": sliced_wasserstein_mat,
    "fast_option": np.asarray([FAST_OPTION]),
    "backend": np.asarray(["ripser_presampled"]),
    "n_sliced_directions": np.asarray([N_SLICED_DIRECTIONS]),
}

for ch_name in loaded_channels:
    save_dict[f"{ch_name}_H1_sliced_wasserstein_diagram"] = diagram_store[ch_name]

    for key, val in raw_interval_store[ch_name].items():
        save_dict[f"{ch_name}_{key}"] = val

np.savez(npz_path, **save_dict)

print("\nH1 Sliced Wasserstein Channel Summary:")
if HAVE_IPYTHON:
    display(summary_df)
else:
    print(summary_df)

print("\nLargest H1 Sliced Wasserstein distances:")
if HAVE_IPYTHON:
    display(pairwise_df.head(12))
else:
    print(pairwise_df.head(12))

log(f"\nSaved summary CSV: {summary_csv_path}")
log(f"Saved distance matrix CSV: {distance_csv_path}")
log(f"Saved pairwise CSV: {pairwise_csv_path}")
log(f"Saved NPZ: {npz_path}")
log(f"Pairwise sliced Wasserstein elapsed: {pair_elapsed:.2f}s")

plot_df = summary_df.copy().sort_values("mean_sliced_wasserstein_to_others", ascending=False).reset_index(drop=True)

fig, ax = plt.subplots(figsize=(13, 5.8), facecolor=BLACK)
style_ax(
    ax,
    title=f"Mean H1 Sliced Wasserstein Distance to Other Channels | {FAST_OPTION.upper()}",
    xlabel="Channel",
    ylabel="Mean sliced Wasserstein distance",
)

x_pos = np.arange(len(plot_df))
y = plot_df["mean_sliced_wasserstein_to_others"].to_numpy(dtype=float)

ax.bar(x_pos, y, color=ACCENT, edgecolor=ACCENT, alpha=0.78)

if np.any(np.isfinite(y)):
    ax.axhline(
        np.nanmean(y),
        color=ACCENT_SOFT,
        linestyle="--",
        linewidth=1.4,
        alpha=0.8,
        label=f"Mean={np.nanmean(y):.5f}",
    )

ax.set_xticks(x_pos)
ax.set_xticklabels(plot_df["channel"], rotation=90)
ax.legend(loc="best", framealpha=1.0)

mean_dist_path = os.path.join(mode_out_dir, "01_mean_h1_sliced_wasserstein_distance_by_channel.png")
plt.tight_layout()
save_render_close(fig, mean_dist_path)

if n_ch >= 3 and np.any(sliced_wasserstein_mat > 0):
    condensed = sliced_wasserstein_mat[np.triu_indices(n_ch, k=1)]
    Z = linkage(condensed, method="average")
    order = leaves_list(Z)
else:
    order = np.arange(n_ch)

ordered_dist = sliced_wasserstein_mat[np.ix_(order, order)]
ordered_channels = [loaded_channels[i] for i in order]

fig, ax = plt.subplots(figsize=(9, 8), facecolor=BLACK)
style_ax(
    ax,
    title="Pairwise H1 Sliced Wasserstein Distance Matrix",
    xlabel="Channel",
    ylabel="Channel",
)

im = ax.imshow(
    ordered_dist,
    aspect="equal",
    interpolation="nearest",
    cmap=CYAN_SEQUENTIAL_CMAP,
)

ax.set_xticks(np.arange(n_ch))
ax.set_xticklabels(ordered_channels, rotation=90)
ax.set_yticks(np.arange(n_ch))
ax.set_yticklabels(ordered_channels)

cbar = plt.colorbar(im, ax=ax, shrink=0.85)
style_colorbar(cbar, "Sliced Wasserstein distance")

dist_heatmap_path = os.path.join(mode_out_dir, "02_h1_sliced_wasserstein_distance_heatmap.png")
plt.tight_layout()
save_render_close(fig, dist_heatmap_path)

mds_coords = classical_mds_from_distances(sliced_wasserstein_mat, n_components=2)

mds_df = pd.DataFrame({
    "channel": loaded_channels,
    "mds_1": mds_coords[:, 0],
    "mds_2": mds_coords[:, 1],
    "mean_sliced_wasserstein_to_others": [
        summary_df.loc[summary_df["channel"] == ch, "mean_sliced_wasserstein_to_others"].values[0]
        for ch in loaded_channels
    ],
    "n_intervals_diagram": [
        summary_df.loc[summary_df["channel"] == ch, "n_intervals_diagram"].values[0]
        for ch in loaded_channels
    ],
})

mds_csv_path = os.path.join(mode_out_dir, "h1_sliced_wasserstein_mds_coordinates.csv")
mds_df.to_csv(mds_csv_path, index=False)

fig, ax = plt.subplots(figsize=(8, 7), facecolor=BLACK)
style_ax(
    ax,
    title="Classical MDS of H1 Sliced Wasserstein Distances",
    xlabel="MDS 1",
    ylabel="MDS 2",
)

sc = ax.scatter(
    mds_df["mds_1"],
    mds_df["mds_2"],
    c=mds_df["mean_sliced_wasserstein_to_others"],
    s=150,
    cmap=CYAN_SEQUENTIAL_CMAP,
    edgecolor=ACCENT,
    linewidth=1.1,
)

for _, row in mds_df.iterrows():
    ax.text(
        row["mds_1"],
        row["mds_2"],
        row["channel"],
        fontsize=8,
        color=ACCENT,
        alpha=0.9,
    )

cbar = plt.colorbar(sc, ax=ax, shrink=0.85)
style_colorbar(cbar, "Mean sliced Wasserstein distance")

mds_path = os.path.join(mode_out_dir, "03_h1_sliced_wasserstein_mds_embedding.png")
plt.tight_layout()
save_render_close(fig, mds_path)

diagram_channels = summary_df.head(min(12, len(summary_df)))["channel"].tolist()
n_cols = 4
n_rows = int(np.ceil(len(diagram_channels) / n_cols))

fig, axes = plt.subplots(
    n_rows,
    n_cols,
    figsize=(4 * n_cols, 3.2 * n_rows),
    facecolor=BLACK,
)

axes = np.asarray(axes).reshape(-1)

all_diagram_points = np.vstack([
    diagram_store[ch] for ch in loaded_channels if len(diagram_store[ch]) > 0
]) if any(len(diagram_store[ch]) > 0 for ch in loaded_channels) else np.empty((0, 2))

if len(all_diagram_points) > 0:
    diag_max = float(np.nanmax(all_diagram_points[:, 1]))
else:
    diag_max = 1.0

diag_max = max(diag_max, 1e-6)

for ax_idx, ch_name in enumerate(diagram_channels):
    ax = axes[ax_idx]
    D = diagram_store[ch_name]

    mean_dist = summary_df.loc[
        summary_df["channel"] == ch_name,
        "mean_sliced_wasserstein_to_others",
    ].values[0]

    style_ax(
        ax,
        title=f"{ch_name}: mean SW={mean_dist:.3f}, n={len(D)}",
        xlabel="Birth",
        ylabel="Death",
    )

    ax.plot([0, diag_max], [0, diag_max], color=ACCENT_DIM, linestyle="--", linewidth=1.0)

    if len(D) > 0:
        ax.scatter(
            D[:, 0],
            D[:, 1],
            s=50,
            facecolors=BLACK,
            edgecolors=ACCENT,
            linewidths=1.2,
        )

    ax.set_xlim(0, diag_max * 1.05)
    ax.set_ylim(0, diag_max * 1.05)

for j in range(len(diagram_channels), len(axes)):
    axes[j].axis("off")

diagrams_path = os.path.join(mode_out_dir, "04_top_channels_h1_persistence_diagrams.png")
plt.tight_layout()
save_render_close(fig, diagrams_path, width=1100)

fig, ax = plt.subplots(figsize=(12, 6), facecolor=BLACK)
style_ax(
    ax,
    title="Largest Pairwise H1 Sliced Wasserstein Distances",
    xlabel="Channel pair",
    ylabel="Sliced Wasserstein distance",
)

top_pairs = pairwise_df.head(min(15, len(pairwise_df))).copy()
pair_labels = top_pairs["channel_a"] + "-" + top_pairs["channel_b"]

x_pair = np.arange(len(top_pairs))

ax.bar(
    x_pair,
    top_pairs["sliced_wasserstein_distance"],
    color=ACCENT,
    edgecolor=ACCENT,
    alpha=0.78,
)

ax.set_xticks(x_pair)
ax.set_xticklabels(pair_labels, rotation=60, ha="right")

top_pairs_path = os.path.join(mode_out_dir, "05_largest_pairwise_h1_sliced_wasserstein_distances.png")
plt.tight_layout()
save_render_close(fig, top_pairs_path)

if MAKE_EXPERT_PLOTS:
    expert_dir = os.path.join(mode_out_dir, "expert_plots")
    os.makedirs(expert_dir, exist_ok=True)

    mean_sw_map = dict(zip(summary_df["channel"], summary_df["mean_sliced_wasserstein_to_others"]))

    scalp_xy = {
        "Fp1": (-0.45, 1.00), "Fpz": (0.00, 1.08), "Fp2": (0.45, 1.00),
        "F7": (-0.95, 0.60), "F3": (-0.45, 0.55), "Fz": (0.00, 0.60),
        "F4": (0.45, 0.55), "F8": (0.95, 0.60),
        "FC5": (-0.75, 0.25), "FC1": (-0.25, 0.25),
        "FC2": (0.25, 0.25), "FC6": (0.75, 0.25),
        "M1": (-1.15, 0.05), "T7": (-1.00, 0.00), "C3": (-0.50, 0.00),
        "Cz": (0.00, 0.00), "C4": (0.50, 0.00), "T8": (1.00, 0.00),
        "M2": (1.15, 0.05),
        "CP5": (-0.75, -0.30), "CP1": (-0.25, -0.30),
        "CP2": (0.25, -0.30), "CP6": (0.75, -0.30),
        "P7": (-0.90, -0.60), "P3": (-0.45, -0.60), "Pz": (0.00, -0.65),
        "P4": (0.45, -0.60), "P8": (0.90, -0.60),
        "POz": (0.00, -0.88),
        "O1": (-0.35, -1.00), "Oz": (0.00, -1.08), "O2": (0.35, -1.00),
    }

    xs, ys, vals, labels = [], [], [], []

    for ch in eeg_channels:
        if ch in scalp_xy and ch in mean_sw_map and np.isfinite(mean_sw_map[ch]):
            xs.append(scalp_xy[ch][0])
            ys.append(scalp_xy[ch][1])
            vals.append(mean_sw_map[ch])
            labels.append(ch)

    fig, ax = plt.subplots(figsize=(7, 7), facecolor=BLACK)
    style_ax(
        ax,
        title=f"Scalp Map of Mean H1 Sliced Wasserstein Distance | {FAST_OPTION.upper()}",
        grid=False,
    )

    head = plt.Circle((0, 0), 1.12, fill=False, color=ACCENT, linewidth=2)
    ax.add_patch(head)
    ax.plot([0.00, -0.10, 0.10, 0.00], [1.12, 1.28, 1.28, 1.12], color=ACCENT, linewidth=2)

    sc = ax.scatter(
        xs,
        ys,
        c=vals,
        s=260,
        cmap=CYAN_SEQUENTIAL_CMAP,
        edgecolor=ACCENT,
        linewidth=1.2,
    )

    for x0, y0, label in zip(xs, ys, labels):
        ax.text(
            x0,
            y0,
            label,
            ha="center",
            va="center",
            fontsize=8,
            color=BLACK,
            fontweight="bold",
        )

    cbar = plt.colorbar(sc, ax=ax, shrink=0.75)
    style_colorbar(cbar, "Mean sliced Wasserstein distance")

    ax.set_xlim(-1.35, 1.35)
    ax.set_ylim(-1.35, 1.35)
    ax.set_aspect("equal")
    ax.set_xticks([])
    ax.set_yticks([])

    scalp_path = os.path.join(expert_dir, "06_mean_h1_sliced_wasserstein_distance_scalp_map.png")
    plt.tight_layout()
    save_render_close(fig, scalp_path, width=760)

    regions = {
        "Frontal": ["Fp1", "Fpz", "Fp2", "F7", "F3", "Fz", "F4", "F8"],
        "Frontocentral": ["FC5", "FC1", "FC2", "FC6"],
        "Central": ["T7", "C3", "Cz", "C4", "T8"],
        "Centroparietal": ["CP5", "CP1", "CP2", "CP6"],
        "Parietal": ["P7", "P3", "Pz", "P4", "P8"],
        "Occipital": ["POz", "O1", "Oz", "O2"],
        "Mastoid": ["M1", "M2"],
    }

    region_rows = []
    region_points = []

    for region_name, chans in regions.items():
        vals_region = []

        for ch in chans:
            if ch in mean_sw_map and np.isfinite(mean_sw_map[ch]):
                vals_region.append(mean_sw_map[ch])
                region_points.append({
                    "region": region_name,
                    "channel": ch,
                    "mean_sliced_wasserstein_to_others": mean_sw_map[ch],
                })

        if len(vals_region) == 0:
            continue

        vals_region = np.asarray(vals_region, dtype=float)

        region_rows.append({
            "region": region_name,
            "mean_region_sliced_wasserstein": float(np.mean(vals_region)),
            "std_region_sliced_wasserstein": float(np.std(vals_region, ddof=1)) if len(vals_region) > 1 else 0.0,
            "sem_region_sliced_wasserstein": (
                float(np.std(vals_region, ddof=1) / np.sqrt(len(vals_region)))
                if len(vals_region) > 1 else 0.0
            ),
            "n_channels": int(len(vals_region)),
        })

    region_df = pd.DataFrame(region_rows)
    region_points_df = pd.DataFrame(region_points)

    region_csv_path = os.path.join(expert_dir, "regional_h1_sliced_wasserstein_summary.csv")
    region_df.to_csv(region_csv_path, index=False)

    if len(region_df) > 0:
        fig, ax = plt.subplots(figsize=(12, 6.5), facecolor=BLACK)
        style_ax(
            ax,
            title=f"Regional Mean H1 Sliced Wasserstein Distance +/- SEM | {FAST_OPTION.upper()}",
            xlabel="Region",
            ylabel="Mean sliced Wasserstein distance",
        )

        x_region = np.arange(len(region_df))

        ax.bar(
            x_region,
            region_df["mean_region_sliced_wasserstein"],
            yerr=region_df["sem_region_sliced_wasserstein"],
            color=ACCENT,
            edgecolor=ACCENT,
            alpha=0.82,
            width=0.68,
            capsize=8,
            error_kw={
                "elinewidth": 2.2,
                "capthick": 2.2,
                "ecolor": ACCENT,
                "alpha": 1.0,
            },
            label="Mean +/- SEM",
        )

        rng = np.random.default_rng(7)

        for i, region_name in enumerate(region_df["region"]):
            vals = region_points_df.loc[
                region_points_df["region"] == region_name,
                "mean_sliced_wasserstein_to_others",
            ].to_numpy(dtype=float)

            if len(vals) == 0:
                continue

            jitter = rng.uniform(-0.13, 0.13, size=len(vals))

            ax.scatter(
                np.full(len(vals), i) + jitter,
                vals,
                s=42,
                facecolors=BLACK,
                edgecolors=ACCENT,
                linewidths=1.2,
                alpha=0.95,
                zorder=4,
            )

        for i, row in region_df.iterrows():
            y_val = row["mean_region_sliced_wasserstein"]
            sem = row["sem_region_sliced_wasserstein"]

            ax.text(
                i,
                y_val + sem,
                f"{y_val:.3f}\n+/-{sem:.3f}\nn={int(row['n_channels'])}",
                ha="center",
                va="bottom",
                fontsize=8,
                color=ACCENT,
            )

        ax.set_xticks(x_region)
        ax.set_xticklabels(
            region_df["region"],
            rotation=35,
            ha="right",
            rotation_mode="anchor",
        )
        ax.margins(x=0.04)
        ax.legend(loc="best", framealpha=1.0)

        region_path = os.path.join(expert_dir, "07_regional_h1_sliced_wasserstein_summary.png")
        plt.tight_layout(pad=2.0)
        save_render_close(fig, region_path)

    fig, ax = plt.subplots(figsize=(8, 6), facecolor=BLACK)
    style_ax(
        ax,
        title="Diagram Size vs Mean H1 Sliced Wasserstein Distance",
        xlabel="H1 diagram intervals",
        ylabel="Mean sliced Wasserstein distance to others",
    )

    sc = ax.scatter(
        summary_df["n_intervals_diagram"],
        summary_df["mean_sliced_wasserstein_to_others"],
        c=summary_df["max_lifetime"],
        s=140,
        cmap=CYAN_SEQUENTIAL_CMAP,
        edgecolor=ACCENT,
        linewidth=1.0,
    )

    for _, row in summary_df.iterrows():
        ax.text(
            row["n_intervals_diagram"],
            row["mean_sliced_wasserstein_to_others"],
            row["channel"],
            fontsize=8,
            color=ACCENT,
            alpha=0.85,
        )

    cbar = plt.colorbar(sc, ax=ax, shrink=0.85)
    style_colorbar(cbar, "Max H1 lifetime")

    diagram_size_path = os.path.join(expert_dir, "08_diagram_size_vs_mean_sliced_wasserstein.png")
    plt.tight_layout()
    save_render_close(fig, diagram_size_path)

    fig, ax = plt.subplots(figsize=(8, 6), facecolor=BLACK)
    style_ax(
        ax,
        title="Threshold Triangles vs Mean H1 Sliced Wasserstein Distance",
        xlabel="Triangles at radius",
        ylabel="Mean sliced Wasserstein distance to others",
    )

    sc = ax.scatter(
        summary_df["n_triangles_at_radius"],
        summary_df["mean_sliced_wasserstein_to_others"],
        c=summary_df["n_intervals_diagram"],
        s=140,
        cmap=CYAN_SEQUENTIAL_CMAP,
        edgecolor=ACCENT,
        linewidth=1.0,
    )

    for _, row in summary_df.iterrows():
        ax.text(
            row["n_triangles_at_radius"],
            row["mean_sliced_wasserstein_to_others"],
            row["channel"],
            fontsize=8,
            color=ACCENT,
            alpha=0.85,
        )

    cbar = plt.colorbar(sc, ax=ax, shrink=0.85)
    style_colorbar(cbar, "H1 diagram intervals")

    triangle_path = os.path.join(expert_dir, "09_threshold_triangles_vs_mean_sliced_wasserstein.png")
    plt.tight_layout()
    save_render_close(fig, triangle_path)

    qc_df = summary_df.copy().sort_values("elapsed_sec", ascending=False).reset_index(drop=True)

    fig, ax = plt.subplots(figsize=(12, 5.8), facecolor=BLACK)
    style_ax(
        ax,
        title="Sliced Wasserstein Diagram Runtime QC",
        xlabel="Channel",
        ylabel="Elapsed seconds",
    )

    x_qc = np.arange(len(qc_df))

    ax.bar(
        x_qc,
        qc_df["elapsed_sec"],
        color=ACCENT,
        edgecolor=ACCENT,
        alpha=0.82,
    )

    ax.set_xticks(x_qc)
    ax.set_xticklabels(qc_df["channel"], rotation=90)

    qc_path = os.path.join(expert_dir, "10_runtime_qc.png")
    plt.tight_layout()
    save_render_close(fig, qc_path)

    log(f"\nExpert plots saved and rendered from: {expert_dir}")
    log(f"Regional summary CSV: {region_csv_path}")

elapsed_total = time.time() - overall_t0

log("\n" + "=" * 72)
log(f"Finished H1 Sliced Wasserstein Distances in {elapsed_total:.2f}s")
log(f"Mode: {FAST_OPTION}")
log(f"Backend: ripser_presampled + internal sliced Wasserstein")
log(f"Channels completed: {len(loaded_channels)}")
log(f"Summary CSV: {summary_csv_path}")
log(f"Distance matrix CSV: {distance_csv_path}")
log(f"Pairwise CSV: {pairwise_csv_path}")
log(f"MDS CSV: {mds_csv_path}")
log(f"NPZ: {npz_path}")
log(f"Output directory: {mode_out_dir}")
log("=" * 72)
```

```python
import os
import sys
import time
import numpy as np

if not hasattr(np, "trapz"):
    np.trapz = np.trapezoid

import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.colors import LinearSegmentedColormap

os.environ["OMP_NUM_THREADS"] = "2"
os.environ["OPENBLAS_NUM_THREADS"] = "2"
os.environ["MKL_NUM_THREADS"] = "2"
os.environ["VECLIB_MAXIMUM_THREADS"] = "2"
os.environ["NUMEXPR_NUM_THREADS"] = "2"

from scipy.spatial.distance import pdist, squareform
from scipy.cluster.hierarchy import linkage, leaves_list

try:
    from scipy.linalg import eigh
    HAVE_SCIPY_EIGH = True
except Exception:
    HAVE_SCIPY_EIGH = False

try:
    from IPython.display import display, Image
    HAVE_IPYTHON = True
except Exception:
    HAVE_IPYTHON = False

try:
    from ripser import ripser
except Exception as exc:
    raise ImportError(
        "ripser is not installed. Run this in a separate notebook cell first:\n\n"
        "!pip install ripser persim\n\n"
        "Then rerun this Persistence-Kernel Similarity Matrix cell."
    ) from exc


base_dir = "/home/a/projects/Complete-Neural-Signal-Analysis"
out_dir = os.path.join(base_dir, "topology_persistence_kernel_similarity")
os.makedirs(out_dir, exist_ok=True)

FAST_OPTION = "ultra"  # options: "ultra", "fast", "balanced", "full"

MAX_CHANNELS_TO_PROCESS = None
RENDER_ALL_IMAGES_INLINE = True
SAVE_PLOTS_TO_DISK = True
MAKE_EXPERT_PLOTS = True
INLINE_IMAGE_WIDTH = 950
VERBOSE_PROGRESS = True

USE_GEOMETRY_PRESERVING_SUBSAMPLE = True
AUTO_KERNEL_BANDWIDTH = True

MODE_CONFIGS = {
    "ultra": {
        "max_points": 90,
        "reservoir_points": 5000,
        "max_dims": 3,
        "radius_quantile": 0.22,
        "kernel_bandwidth_frac": 0.035,
        "kernel_auto_quantile": 0.25,
        "kernel_auto_multiplier": 0.60,
        "cap_essential_h1": True,
        "plot_dpi": 120,
    },
    "fast": {
        "max_points": 130,
        "reservoir_points": 8000,
        "max_dims": 3,
        "radius_quantile": 0.24,
        "kernel_bandwidth_frac": 0.030,
        "kernel_auto_quantile": 0.25,
        "kernel_auto_multiplier": 0.55,
        "cap_essential_h1": True,
        "plot_dpi": 130,
    },
    "balanced": {
        "max_points": 180,
        "reservoir_points": 10000,
        "max_dims": 3,
        "radius_quantile": 0.26,
        "kernel_bandwidth_frac": 0.026,
        "kernel_auto_quantile": 0.22,
        "kernel_auto_multiplier": 0.50,
        "cap_essential_h1": True,
        "plot_dpi": 140,
    },
    "full": {
        "max_points": 260,
        "reservoir_points": 14000,
        "max_dims": 3,
        "radius_quantile": 0.28,
        "kernel_bandwidth_frac": 0.022,
        "kernel_auto_quantile": 0.20,
        "kernel_auto_multiplier": 0.45,
        "cap_essential_h1": True,
        "plot_dpi": 150,
    },
}

if FAST_OPTION not in MODE_CONFIGS:
    raise ValueError("FAST_OPTION must be one of: ultra, fast, balanced, full")

cfg = MODE_CONFIGS[FAST_OPTION]

MAX_POINTS = cfg["max_points"]
RESERVOIR_POINTS = cfg["reservoir_points"]
MAX_DIMS = cfg["max_dims"]
RADIUS_QUANTILE = cfg["radius_quantile"]
KERNEL_BANDWIDTH_FRAC = cfg["kernel_bandwidth_frac"]
KERNEL_AUTO_QUANTILE = cfg["kernel_auto_quantile"]
KERNEL_AUTO_MULTIPLIER = cfg["kernel_auto_multiplier"]
CAP_ESSENTIAL_H1 = cfg["cap_essential_h1"]
PLOT_DPI = cfg["plot_dpi"]

mode_out_dir = os.path.join(
    out_dir,
    f"h1_persistence_kernel_similarity_{FAST_OPTION}_fps_adaptive_kernel",
)
os.makedirs(mode_out_dir, exist_ok=True)

eeg_channels = [
    "Fp1", "Fpz", "Fp2", "F7", "F3", "Fz", "F4", "F8", "FC5", "FC1", "FC2", "FC6",
    "M1", "T7", "C3", "Cz", "C4", "T8", "M2", "CP5", "CP1", "CP2", "CP6",
    "P7", "P3", "Pz", "P4", "P8", "POz", "O1", "Oz", "O2"
]

if MAX_CHANNELS_TO_PROCESS is not None:
    eeg_channels = eeg_channels[:MAX_CHANNELS_TO_PROCESS]

ACCENT = "#00FFFF"
ACCENT_SOFT = "#66FFFF"
ACCENT_DIM = "#008B8B"
BLACK = "#000000"

CYAN_SEQUENTIAL_CMAP = LinearSegmentedColormap.from_list(
    "cyan_sequential",
    [BLACK, ACCENT_DIM, ACCENT, ACCENT_SOFT],
    N=256,
)

CYAN_DIVERGING_CMAP = LinearSegmentedColormap.from_list(
    "cyan_diverging",
    [ACCENT_DIM, BLACK, ACCENT_SOFT],
    N=256,
)

plt.rcParams.update({
    "figure.facecolor": BLACK,
    "axes.facecolor": BLACK,
    "savefig.facecolor": BLACK,
    "text.color": ACCENT,
    "axes.labelcolor": ACCENT,
    "axes.edgecolor": ACCENT,
    "axes.titlecolor": ACCENT,
    "xtick.color": ACCENT,
    "ytick.color": ACCENT,
    "grid.color": ACCENT_DIM,
    "grid.alpha": 0.25,
    "legend.facecolor": BLACK,
    "legend.edgecolor": ACCENT,
    "legend.labelcolor": ACCENT,
    "font.size": 11,
})


def log(msg):
    if VERBOSE_PROGRESS:
        print(msg, flush=True)


def style_ax(ax, title=None, xlabel=None, ylabel=None, grid=True):
    ax.set_facecolor(BLACK)

    for spine in ax.spines.values():
        spine.set_color(ACCENT)
        spine.set_linewidth(1.1)

    ax.tick_params(axis="x", colors=ACCENT, labelcolor=ACCENT)
    ax.tick_params(axis="y", colors=ACCENT, labelcolor=ACCENT)

    if title is not None:
        ax.set_title(title, color=ACCENT, pad=12)

    if xlabel is not None:
        ax.set_xlabel(xlabel, color=ACCENT, labelpad=10)

    if ylabel is not None:
        ax.set_ylabel(ylabel, color=ACCENT, labelpad=10)

    if grid:
        ax.grid(True, color=ACCENT_DIM, alpha=0.25, linewidth=0.8)


def style_colorbar(cbar, label):
    cbar.set_label(label, color=ACCENT)
    cbar.ax.yaxis.set_tick_params(color=ACCENT)
    plt.setp(cbar.ax.get_yticklabels(), color=ACCENT)
    cbar.outline.set_edgecolor(ACCENT)


def save_render_close(fig, path, width=INLINE_IMAGE_WIDTH):
    if SAVE_PLOTS_TO_DISK:
        fig.savefig(path, dpi=PLOT_DPI, bbox_inches="tight", facecolor=BLACK)

    plt.close(fig)

    if RENDER_ALL_IMAGES_INLINE and HAVE_IPYTHON and os.path.exists(path):
        display(Image(filename=path, width=width))


def dynamic_limits(values, lower_pct=2, upper_pct=98, pad_frac=0.08, lower_bound=None, upper_bound=None):
    values = np.asarray(values, dtype=float)
    values = values[np.isfinite(values)]

    if len(values) == 0:
        vmin, vmax = 0.0, 1.0
    else:
        vmin = float(np.percentile(values, lower_pct))
        vmax = float(np.percentile(values, upper_pct))

        if not np.isfinite(vmin):
            vmin = float(np.nanmin(values))

        if not np.isfinite(vmax):
            vmax = float(np.nanmax(values))

        if vmax <= vmin:
            center = float(np.nanmean(values))
            eps = max(1e-6, abs(center) * 0.01, 1e-5)
            vmin = center - eps
            vmax = center + eps

    pad = (vmax - vmin) * pad_frac
    vmin -= pad
    vmax += pad

    if lower_bound is not None:
        vmin = max(lower_bound, vmin)

    if upper_bound is not None:
        vmax = min(upper_bound, vmax)

    if vmax <= vmin:
        vmax = vmin + 1e-6

    return vmin, vmax


def offdiag_values(M):
    M = np.asarray(M, dtype=float)

    if M.shape[0] <= 1:
        return np.asarray([], dtype=float)

    mask = ~np.eye(M.shape[0], dtype=bool)
    vals = M[mask]
    vals = vals[np.isfinite(vals)]

    return vals


def save_matrix_plot(
    matrix,
    labels,
    title,
    colorbar_label,
    path,
    vmin=None,
    vmax=None,
    width=INLINE_IMAGE_WIDTH,
):
    fig, ax = plt.subplots(figsize=(9, 8), facecolor=BLACK)
    style_ax(
        ax,
        title=title,
        xlabel="Channel",
        ylabel="Channel",
    )

    im = ax.imshow(
        matrix,
        aspect="equal",
        interpolation="nearest",
        cmap=CYAN_SEQUENTIAL_CMAP,
        vmin=vmin,
        vmax=vmax,
    )

    ax.set_xticks(np.arange(len(labels)))
    ax.set_xticklabels(labels, rotation=90)
    ax.set_yticks(np.arange(len(labels)))
    ax.set_yticklabels(labels)

    cbar = plt.colorbar(im, ax=ax, shrink=0.85)
    style_colorbar(cbar, colorbar_label)

    plt.tight_layout()
    save_render_close(fig, path, width=width)


def estimate_median_pairwise_distance(X, max_pairs=120000, seed=7):
    X = np.asarray(X, dtype=float)
    n = X.shape[0]

    if n < 2:
        return 1.0

    rng = np.random.default_rng(seed)

    if n * (n - 1) // 2 <= max_pairs:
        D = pdist(X)
        D = D[D > 0]

        if len(D) == 0:
            return 1.0

        return float(np.nanmedian(D))

    i = rng.integers(0, n, size=max_pairs)
    j = rng.integers(0, n, size=max_pairs)

    mask = i != j
    i = i[mask]
    j = j[mask]

    if len(i) == 0:
        return 1.0

    d = np.linalg.norm(X[i] - X[j], axis=1)
    d = d[np.isfinite(d) & (d > 0)]

    if len(d) == 0:
        return 1.0

    return float(np.nanmedian(d))


def standardize_matrix_for_geometry(X, max_dims=3, seed=7):
    X = np.asarray(X, dtype=np.float64)
    X = X[np.all(np.isfinite(X), axis=1)]

    if X.shape[0] < 10:
        return None

    X = X - np.mean(X, axis=0, keepdims=True)

    if X.shape[1] > max_dims:
        _U, _S, Vt = np.linalg.svd(X, full_matrices=False)
        X = X @ Vt[:max_dims].T

    X = X / (np.std(X, axis=0, keepdims=True) + 1e-12)

    scale = estimate_median_pairwise_distance(X, seed=seed)

    X = X / (scale + 1e-12)

    return X


def farthest_point_sample(X, n_samples, seed=7):
    X = np.asarray(X, dtype=float)
    n = X.shape[0]

    if n <= n_samples:
        return np.arange(n)

    rng = np.random.default_rng(seed)

    center = np.mean(X, axis=0, keepdims=True)
    dist_to_center = np.linalg.norm(X - center, axis=1)

    start_idx = int(np.nanargmax(dist_to_center))

    selected = np.empty(n_samples, dtype=int)
    selected[0] = start_idx

    min_dist2 = np.full(n, np.inf, dtype=float)

    for k in range(1, n_samples):
        last = X[selected[k - 1]]
        d2 = np.sum((X - last) ** 2, axis=1)
        min_dist2 = np.minimum(min_dist2, d2)
        min_dist2[selected[:k]] = -np.inf

        next_idx = int(np.nanargmax(min_dist2))

        if not np.isfinite(min_dist2[next_idx]) or min_dist2[next_idx] < 1e-18:
            remaining = np.setdiff1d(np.arange(n), selected[:k], assume_unique=False)

            if len(remaining) == 0:
                selected = selected[:k]
                break

            next_idx = int(rng.choice(remaining))

        selected[k] = next_idx

    selected = selected[selected >= 0]

    return selected


def load_channel_embedding_geometry_sampled(ch_name, max_points=90, reservoir_points=5000, max_dims=3, seed=7):
    candidate_paths = [
        os.path.join(base_dir, "embedding_data", "tsne_umap_from_3d", f"umap_embedding_{ch_name}.npy"),
        os.path.join(base_dir, "embedding_data", "tsne_umap_from_3d", f"tsne_embedding_{ch_name}.npy"),
        os.path.join(base_dir, "embedding_data", "3dembedding_data", f"3dembedded_{ch_name}.npy"),
        os.path.join(base_dir, "embedding_data", "2dembedding_data", f"2dembedded_{ch_name}.npy"),
    ]

    rng = np.random.default_rng(seed)

    for path in candidate_paths:
        if not os.path.exists(path):
            continue

        log(f"    trying: {path}")

        try:
            X = np.load(path, mmap_mode="r", allow_pickle=False)
        except Exception as exc:
            log(f"    skipped non-memmappable or nonnumeric file: {os.path.basename(path)} | {exc}")
            continue

        if X.ndim != 2:
            log(f"    skipped because ndim={X.ndim}")
            continue

        original_shape = tuple(X.shape)

        try:
            if X.shape[0] >= X.shape[1]:
                n_samples = X.shape[0]
                n_features = X.shape[1]

                if n_samples < 10 or n_features < 2:
                    log(f"    skipped bad shape: {original_shape}")
                    continue

                reservoir_n = min(max(reservoir_points, max_points), n_samples)

                if n_samples > reservoir_n:
                    reservoir_idx = np.sort(rng.choice(n_samples, size=reservoir_n, replace=False))
                else:
                    reservoir_idx = np.arange(n_samples)

                X_reservoir = np.asarray(X[reservoir_idx, :], dtype=np.float64)

            else:
                n_samples = X.shape[1]
                n_features = X.shape[0]

                if n_samples < 10 or n_features < 2:
                    log(f"    skipped bad transposed shape: {original_shape}")
                    continue

                reservoir_n = min(max(reservoir_points, max_points), n_samples)

                if n_samples > reservoir_n:
                    reservoir_idx = np.sort(rng.choice(n_samples, size=reservoir_n, replace=False))
                else:
                    reservoir_idx = np.arange(n_samples)

                X_reservoir = np.asarray(X[:, reservoir_idx].T, dtype=np.float64)

        except Exception as exc:
            log(f"    failed while reservoir-sampling {os.path.basename(path)} | {exc}")
            continue

        X_reservoir = X_reservoir[np.all(np.isfinite(X_reservoir), axis=1)]

        if X_reservoir.shape[0] < 10 or X_reservoir.shape[1] < 2:
            log(f"    reservoir data unusable: reservoir_shape={X_reservoir.shape}")
            continue

        X_reservoir_std = standardize_matrix_for_geometry(
            X_reservoir,
            max_dims=max_dims,
            seed=seed,
        )

        if X_reservoir_std is None:
            log(f"    standardized reservoir unusable")
            continue

        if USE_GEOMETRY_PRESERVING_SUBSAMPLE:
            fps_idx = farthest_point_sample(
                X_reservoir_std,
                n_samples=min(max_points, X_reservoir_std.shape[0]),
                seed=seed,
            )
            X_small = X_reservoir_std[fps_idx]
            method = "reservoir_plus_farthest_point"
        else:
            sample_n = min(max_points, X_reservoir_std.shape[0])
            idx = np.sort(rng.choice(X_reservoir_std.shape[0], size=sample_n, replace=False))
            X_small = X_reservoir_std[idx]
            method = "reservoir_random"

        X_small = X_small[np.all(np.isfinite(X_small), axis=1)]

        if X_small.shape[0] >= 10 and X_small.shape[1] >= 2:
            return X_small, path, original_shape, tuple(X_reservoir.shape), method

        log(f"    sampled data unusable: sampled_shape={X_small.shape}")

    return None, None, None, None, None


def standardize_point_cloud_small(X, max_dims=3):
    X = np.asarray(X, dtype=np.float64)
    X = X[np.all(np.isfinite(X), axis=1)]

    if X.shape[0] < 10:
        return None

    X = X - np.mean(X, axis=0, keepdims=True)

    if X.shape[1] > max_dims:
        _U, _S, Vt = np.linalg.svd(X, full_matrices=False)
        X = X @ Vt[:max_dims].T

    X = X / (np.std(X, axis=0, keepdims=True) + 1e-12)

    D = pdist(X)
    positive = D[D > 0]

    if len(positive) > 0:
        X = X / (np.nanmedian(positive) + 1e-12)

    return X


def estimate_complex_counts(X, radius):
    D = squareform(pdist(X))
    A = (D <= radius) & (D > 0)

    n_edges = int(np.sum(A) // 2)

    if A.shape[0] <= 300:
        A_int = A.astype(np.int64)
        n_triangles = int(np.trace(A_int @ A_int @ A_int) // 6)
    else:
        n_triangles = np.nan

    return n_edges, n_triangles


def run_h1_ripser(X, radius):
    result = ripser(
        X,
        maxdim=1,
        thresh=float(radius),
        coeff=2,
        do_cocycles=False,
    )

    diagrams = result["dgms"]

    h0 = diagrams[0] if len(diagrams) > 0 else np.empty((0, 2), dtype=float)
    h1 = diagrams[1] if len(diagrams) > 1 else np.empty((0, 2), dtype=float)

    return h0, h1


def split_finite_essential(diagram):
    diagram = np.asarray(diagram, dtype=float)

    if diagram.size == 0:
        return np.empty((0, 2), dtype=float), np.empty((0, 2), dtype=float)

    finite = diagram[
        np.isfinite(diagram[:, 0])
        & np.isfinite(diagram[:, 1])
        & (diagram[:, 1] > diagram[:, 0])
    ]

    essential = diagram[
        np.isfinite(diagram[:, 0])
        & np.isinf(diagram[:, 1])
    ]

    return finite, essential


def clean_diagram_for_kernel(
    finite_intervals,
    essential_intervals=None,
    cap_essential=True,
    cap_value=None,
):
    cleaned = []
    capped_count = 0

    if finite_intervals is not None:
        for b, d in finite_intervals:
            if np.isfinite(b) and np.isfinite(d) and d > b:
                cleaned.append((float(b), float(d)))

    if cap_essential and essential_intervals is not None and cap_value is not None:
        for b, _d in essential_intervals:
            if np.isfinite(b) and np.isfinite(cap_value) and cap_value > b:
                cleaned.append((float(b), float(cap_value)))
                capped_count += 1

    if len(cleaned) == 0:
        return np.empty((0, 2), dtype=float), capped_count

    return np.asarray(cleaned, dtype=float), capped_count


def diagram_features(diagram):
    diagram = np.asarray(diagram, dtype=float)

    if diagram.size == 0 or len(diagram) == 0:
        return {
            "n_intervals_diagram": 0,
            "mean_lifetime": np.nan,
            "max_lifetime": np.nan,
            "total_lifetime": 0.0,
            "mean_birth": np.nan,
            "mean_death": np.nan,
            "max_death": np.nan,
        }

    lifetimes = diagram[:, 1] - diagram[:, 0]

    return {
        "n_intervals_diagram": int(len(diagram)),
        "mean_lifetime": float(np.nanmean(lifetimes)),
        "max_lifetime": float(np.nanmax(lifetimes)),
        "total_lifetime": float(np.nansum(lifetimes)),
        "mean_birth": float(np.nanmean(diagram[:, 0])),
        "mean_death": float(np.nanmean(diagram[:, 1])),
        "max_death": float(np.nanmax(diagram[:, 1])),
    }


def persistence_scale_space_kernel(D1, D2, kernel_time):
    D1 = np.asarray(D1, dtype=float)
    D2 = np.asarray(D2, dtype=float)

    if D1.size == 0:
        D1 = np.empty((0, 2), dtype=float)

    if D2.size == 0:
        D2 = np.empty((0, 2), dtype=float)

    if len(D1) == 0 or len(D2) == 0:
        return 0.0

    D1 = D1[
        np.isfinite(D1[:, 0])
        & np.isfinite(D1[:, 1])
        & (D1[:, 1] > D1[:, 0])
    ]

    D2 = D2[
        np.isfinite(D2[:, 0])
        & np.isfinite(D2[:, 1])
        & (D2[:, 1] > D2[:, 0])
    ]

    if len(D1) == 0 or len(D2) == 0:
        return 0.0

    kernel_time = float(max(kernel_time, 1e-12))

    D2_reflected = D2[:, ::-1]

    diff_direct = D1[:, None, :] - D2[None, :, :]
    diff_reflect = D1[:, None, :] - D2_reflected[None, :, :]

    dist2_direct = np.sum(diff_direct ** 2, axis=2)
    dist2_reflect = np.sum(diff_reflect ** 2, axis=2)

    denom = 8.0 * kernel_time
    prefactor = 1.0 / (8.0 * np.pi * kernel_time)

    K = prefactor * np.sum(
        np.exp(-dist2_direct / denom) - np.exp(-dist2_reflect / denom)
    )

    if not np.isfinite(K):
        return 0.0

    return float(K)


def normalize_kernel_matrix(K):
    K = np.asarray(K, dtype=float)
    K = 0.5 * (K + K.T)

    diag = np.diag(K).copy()
    K_norm = np.zeros_like(K, dtype=float)

    for i in range(K.shape[0]):
        for j in range(K.shape[1]):
            if diag[i] > 0 and diag[j] > 0:
                K_norm[i, j] = K[i, j] / np.sqrt(diag[i] * diag[j])
            elif i == j:
                K_norm[i, j] = 1.0
            else:
                K_norm[i, j] = 0.0

    K_norm = 0.5 * (K_norm + K_norm.T)
    K_norm = np.clip(K_norm, -1.0, 1.0)
    np.fill_diagonal(K_norm, 1.0)

    return K_norm


def kernel_distance_from_normalized_similarity(K_norm):
    K_norm = np.asarray(K_norm, dtype=float)
    D2 = np.maximum(0.0, 2.0 - 2.0 * K_norm)
    D = np.sqrt(D2)
    np.fill_diagonal(D, 0.0)
    return D


def choose_adaptive_kernel_bandwidth(diagram_store, fallback_scale):
    all_points = []

    for D in diagram_store.values():
        D = np.asarray(D, dtype=float)

        if D.size == 0 or len(D) == 0:
            continue

        D = D[
            np.isfinite(D[:, 0])
            & np.isfinite(D[:, 1])
            & (D[:, 1] > D[:, 0])
        ]

        if len(D) > 0:
            all_points.append(D)

    fallback_bandwidth = max(1e-6, KERNEL_BANDWIDTH_FRAC * fallback_scale)

    if len(all_points) == 0:
        return fallback_bandwidth, fallback_bandwidth ** 2, "fallback_no_points"

    all_points = np.vstack(all_points)

    if len(all_points) < 2:
        return fallback_bandwidth, fallback_bandwidth ** 2, "fallback_too_few_points"

    D = pdist(all_points)
    D = D[np.isfinite(D) & (D > 0)]

    if len(D) == 0:
        return fallback_bandwidth, fallback_bandwidth ** 2, "fallback_zero_distances"

    auto_bandwidth = float(np.quantile(D, KERNEL_AUTO_QUANTILE) * KERNEL_AUTO_MULTIPLIER)

    if not np.isfinite(auto_bandwidth) or auto_bandwidth <= 0:
        return fallback_bandwidth, fallback_bandwidth ** 2, "fallback_bad_auto"

    if AUTO_KERNEL_BANDWIDTH:
        chosen = min(fallback_bandwidth, auto_bandwidth)
        chosen = max(chosen, fallback_scale * 0.0025, 1e-6)
        method = "adaptive_min_of_fractional_and_diagram_quantile"
    else:
        chosen = fallback_bandwidth
        method = "fixed_fractional"

    return chosen, chosen ** 2, method


def classical_mds_from_distances(D, n_components=2):
    D = np.asarray(D, dtype=float)
    n = D.shape[0]

    if n < 2:
        return np.zeros((n, n_components), dtype=float)

    D2 = D ** 2
    J = np.eye(n) - np.ones((n, n)) / n
    B = -0.5 * J @ D2 @ J

    if HAVE_SCIPY_EIGH:
        vals, vecs = eigh(B)
    else:
        vals, vecs = np.linalg.eigh(B)

    idx = np.argsort(vals)[::-1]
    vals = vals[idx]
    vecs = vecs[:, idx]

    coords = np.zeros((n, n_components), dtype=float)

    for k in range(min(n_components, n)):
        if vals[k] > 0:
            coords[:, k] = vecs[:, k] * np.sqrt(vals[k])

    return coords


log("\n" + "=" * 72)
log(f"Running H1 Persistence-Kernel Similarity Matrix | {FAST_OPTION.upper()}")
log(f"Channels: {len(eeg_channels)}")
log(f"max_points={MAX_POINTS}, reservoir_points={RESERVOIR_POINTS}, radius_quantile={RADIUS_QUANTILE}")
log(f"kernel_bandwidth_frac={KERNEL_BANDWIDTH_FRAC}, auto_kernel_bandwidth={AUTO_KERNEL_BANDWIDTH}")
log(f"geometry_preserving_subsample={USE_GEOMETRY_PRESERVING_SUBSAMPLE}")
log(f"cap_essential_h1={CAP_ESSENTIAL_H1}")
log(f"expert_plots={MAKE_EXPERT_PLOTS}, render_all_images={RENDER_ALL_IMAGES_INLINE}")
log(f"Output: {mode_out_dir}")
log("=" * 72)

diagnostics = {}
diagram_store = {}
raw_interval_store = {}
loaded_channels = []

global_max_death = None
global_max_lifetime = None

overall_t0 = time.time()

for ch_idx, ch_name in enumerate(eeg_channels, start=1):
    t0 = time.time()
    prefix = f"[{ch_idx:02d}/{len(eeg_channels)} {ch_name}]"

    log(f"\n{prefix} loading reservoir + geometry-preserving subsampling")

    X_small, source_path, original_shape, reservoir_shape, sampling_method = load_channel_embedding_geometry_sampled(
        ch_name,
        max_points=MAX_POINTS,
        reservoir_points=RESERVOIR_POINTS,
        max_dims=MAX_DIMS,
        seed=700 + ch_idx,
    )

    if X_small is None:
        log(f"{prefix} WARNING: no usable embedding found; skipping")
        continue

    log(
        f"{prefix} source shape={original_shape}, reservoir shape={reservoir_shape}, "
        f"sampled shape={X_small.shape}, method={sampling_method}"
    )

    X = standardize_point_cloud_small(X_small, max_dims=MAX_DIMS)

    if X is None:
        log(f"{prefix} WARNING: invalid sampled embedding; skipping")
        continue

    log(f"{prefix} standardized shape={X.shape}; computing radius")

    D_flat = pdist(X)
    positive = D_flat[D_flat > 0]

    if len(positive) == 0:
        log(f"{prefix} WARNING: zero-distance cloud; skipping")
        continue

    radius = float(np.quantile(positive, RADIUS_QUANTILE))

    log(f"{prefix} estimating threshold complex size")
    n_edges, n_triangles = estimate_complex_counts(X, radius)

    log(f"{prefix} ripser H0/H1")
    h0_diag, h1_diag = run_h1_ripser(X, radius)

    h0_finite, h0_essential = split_finite_essential(h0_diag)
    h1_finite, h1_essential = split_finite_essential(h1_diag)

    h1_diagram, n_capped_h1 = clean_diagram_for_kernel(
        h1_finite,
        essential_intervals=h1_essential,
        cap_essential=CAP_ESSENTIAL_H1,
        cap_value=radius,
    )

    if len(h1_diagram) > 0:
        max_death = float(np.nanmax(h1_diagram[:, 1]))
        max_lifetime = float(np.nanmax(h1_diagram[:, 1] - h1_diagram[:, 0]))

        global_max_death = max_death if global_max_death is None else max(global_max_death, max_death)
        global_max_lifetime = max_lifetime if global_max_lifetime is None else max(global_max_lifetime, max_lifetime)

    elapsed = time.time() - t0

    diagram_store[ch_name] = h1_diagram

    raw_interval_store[ch_name] = {
        "H0_finite": h0_finite,
        "H0_essential": h0_essential,
        "H1_finite": h1_finite,
        "H1_essential": h1_essential,
        "H1_kernel_diagram": h1_diagram,
    }

    diagnostics[ch_name] = {
        "channel": ch_name,
        "source_path": source_path,
        "source_shape": str(original_shape),
        "reservoir_shape": str(reservoir_shape),
        "sampling_method": sampling_method,
        "backend": "ripser_presampled_fps",
        "n_points": int(X.shape[0]),
        "n_edges_at_radius": int(n_edges),
        "n_triangles_at_radius": int(n_triangles) if np.isfinite(n_triangles) else np.nan,
        "radius": radius,
        "n_h0_finite": int(len(h0_finite)),
        "n_h1_finite": int(len(h1_finite)),
        "n_h1_essential": int(len(h1_essential)),
        "n_h1_capped_for_kernel": int(n_capped_h1),
        "elapsed_sec": elapsed,
        **diagram_features(h1_diagram),
    }

    loaded_channels.append(ch_name)

    log(
        f"{prefix} done {elapsed:.2f}s | "
        f"edges={n_edges}, triangles={n_triangles}, "
        f"H1 finite={len(h1_finite)}, H1 kernel diagram={len(h1_diagram)}"
    )

if len(loaded_channels) == 0:
    raise RuntimeError("No usable embeddings found. Check embedding_data paths.")

if global_max_death is None or not np.isfinite(global_max_death) or global_max_death <= 0:
    global_max_death = 1.0

if global_max_lifetime is None or not np.isfinite(global_max_lifetime) or global_max_lifetime <= 0:
    global_max_lifetime = 1.0

diagram_scale = max(global_max_death, global_max_lifetime, 1e-6)

kernel_bandwidth, kernel_time, kernel_bandwidth_method = choose_adaptive_kernel_bandwidth(
    diagram_store,
    fallback_scale=diagram_scale,
)

log("\nComputing persistence scale-space kernel matrix")
log(f"global_max_death={global_max_death:.6f}")
log(f"global_max_lifetime={global_max_lifetime:.6f}")
log(f"diagram_scale={diagram_scale:.6f}")
log(f"kernel_bandwidth_length={kernel_bandwidth:.8f}")
log(f"kernel_time={kernel_time:.10f}")
log(f"kernel_bandwidth_method={kernel_bandwidth_method}")

n_ch = len(loaded_channels)
kernel_mat = np.zeros((n_ch, n_ch), dtype=float)

pair_rows = []
pair_t0 = time.time()

for i in range(n_ch):
    ch_i = loaded_channels[i]

    for j in range(i, n_ch):
        ch_j = loaded_channels[j]

        sim = persistence_scale_space_kernel(
            diagram_store[ch_i],
            diagram_store[ch_j],
            kernel_time=kernel_time,
        )

        kernel_mat[i, j] = sim
        kernel_mat[j, i] = sim

        if i != j:
            pair_rows.append({
                "channel_a": ch_i,
                "channel_b": ch_j,
                "raw_kernel_similarity": sim,
            })

    log(f"  completed kernel row {i + 1}/{n_ch}: {ch_i}")

pair_elapsed = time.time() - pair_t0

kernel_norm_mat = normalize_kernel_matrix(kernel_mat)
kernel_distance_mat = kernel_distance_from_normalized_similarity(kernel_norm_mat)

raw_kernel_df = pd.DataFrame(
    kernel_mat,
    index=loaded_channels,
    columns=loaded_channels,
)

norm_kernel_df = pd.DataFrame(
    kernel_norm_mat,
    index=loaded_channels,
    columns=loaded_channels,
)

kernel_distance_df = pd.DataFrame(
    kernel_distance_mat,
    index=loaded_channels,
    columns=loaded_channels,
)

pairwise_df = pd.DataFrame(pair_rows)

if len(pairwise_df) > 0:
    pairwise_df["normalized_kernel_similarity"] = [
        kernel_norm_mat[loaded_channels.index(a), loaded_channels.index(b)]
        for a, b in zip(pairwise_df["channel_a"], pairwise_df["channel_b"])
    ]
    pairwise_df["kernel_distance"] = [
        kernel_distance_mat[loaded_channels.index(a), loaded_channels.index(b)]
        for a, b in zip(pairwise_df["channel_a"], pairwise_df["channel_b"])
    ]
    pairwise_df = pairwise_df.sort_values("normalized_kernel_similarity", ascending=False).reset_index(drop=True)

summary_rows = []

for idx, ch_name in enumerate(loaded_channels):
    sim = kernel_norm_mat[idx].copy()
    dist = kernel_distance_mat[idx].copy()

    mask = np.arange(len(sim)) != idx
    sim_nonself = sim[mask]
    dist_nonself = dist[mask]

    summary_rows.append({
        **diagnostics[ch_name],
        "self_raw_kernel": float(kernel_mat[idx, idx]),
        "mean_kernel_similarity_to_others": float(np.nanmean(sim_nonself)) if len(sim_nonself) else 1.0,
        "median_kernel_similarity_to_others": float(np.nanmedian(sim_nonself)) if len(sim_nonself) else 1.0,
        "max_kernel_similarity_to_others": float(np.nanmax(sim_nonself)) if len(sim_nonself) else 1.0,
        "min_kernel_similarity_to_others": float(np.nanmin(sim_nonself)) if len(sim_nonself) else 1.0,
        "std_kernel_similarity_to_others": float(np.nanstd(sim_nonself)) if len(sim_nonself) else 0.0,
        "mean_kernel_distance_to_others": float(np.nanmean(dist_nonself)) if len(dist_nonself) else 0.0,
        "median_kernel_distance_to_others": float(np.nanmedian(dist_nonself)) if len(dist_nonself) else 0.0,
        "max_kernel_distance_to_others": float(np.nanmax(dist_nonself)) if len(dist_nonself) else 0.0,
        "min_kernel_distance_to_others": float(np.nanmin(dist_nonself)) if len(dist_nonself) else 0.0,
        "std_kernel_distance_to_others": float(np.nanstd(dist_nonself)) if len(dist_nonself) else 0.0,
    })

summary_df = (
    pd.DataFrame(summary_rows)
    .sort_values("mean_kernel_similarity_to_others", ascending=False)
    .reset_index(drop=True)
)

offdiag_sim = offdiag_values(kernel_norm_mat)
offdiag_dist = offdiag_values(kernel_distance_mat)
offdiag_raw = offdiag_values(kernel_mat)

similarity_range = float(np.nanmax(offdiag_sim) - np.nanmin(offdiag_sim)) if len(offdiag_sim) else 0.0
distance_range = float(np.nanmax(offdiag_dist) - np.nanmin(offdiag_dist)) if len(offdiag_dist) else 0.0

log("\nKernel variability QC")
if len(offdiag_sim):
    log(
        f"normalized off-diagonal similarity: "
        f"min={np.nanmin(offdiag_sim):.6f}, "
        f"max={np.nanmax(offdiag_sim):.6f}, "
        f"mean={np.nanmean(offdiag_sim):.6f}, "
        f"std={np.nanstd(offdiag_sim):.6f}, "
        f"range={similarity_range:.6f}"
    )

if len(offdiag_dist):
    log(
        f"kernel distance: "
        f"min={np.nanmin(offdiag_dist):.6f}, "
        f"max={np.nanmax(offdiag_dist):.6f}, "
        f"mean={np.nanmean(offdiag_dist):.6f}, "
        f"std={np.nanstd(offdiag_dist):.6f}, "
        f"range={distance_range:.6f}"
    )

if len(offdiag_sim) and np.nanstd(offdiag_sim) < 1e-4:
    log(
        "WARNING: normalized similarities are still nearly constant. "
        "Try lowering kernel_bandwidth_frac or increasing max_points / radius_quantile."
    )

summary_csv_path = os.path.join(mode_out_dir, "h1_persistence_kernel_channel_summary.csv")
raw_kernel_csv_path = os.path.join(mode_out_dir, "h1_persistence_kernel_raw_matrix.csv")
norm_kernel_csv_path = os.path.join(mode_out_dir, "h1_persistence_kernel_normalized_similarity_matrix.csv")
kernel_distance_csv_path = os.path.join(mode_out_dir, "h1_persistence_kernel_distance_matrix.csv")
pairwise_csv_path = os.path.join(mode_out_dir, "h1_persistence_kernel_pairwise_similarities.csv")

summary_df.to_csv(summary_csv_path, index=False)
raw_kernel_df.to_csv(raw_kernel_csv_path)
norm_kernel_df.to_csv(norm_kernel_csv_path)
kernel_distance_df.to_csv(kernel_distance_csv_path)
pairwise_df.to_csv(pairwise_csv_path, index=False)

npz_path = os.path.join(mode_out_dir, "h1_persistence_kernel_similarity_and_diagrams.npz")

save_dict = {
    "channels": np.asarray(loaded_channels),
    "raw_kernel_matrix": kernel_mat,
    "normalized_kernel_similarity_matrix": kernel_norm_mat,
    "kernel_distance_matrix": kernel_distance_mat,
    "kernel_bandwidth_length": np.asarray([kernel_bandwidth]),
    "kernel_time": np.asarray([kernel_time]),
    "kernel_bandwidth_frac": np.asarray([KERNEL_BANDWIDTH_FRAC]),
    "kernel_bandwidth_method": np.asarray([kernel_bandwidth_method]),
    "fast_option": np.asarray([FAST_OPTION]),
    "backend": np.asarray(["ripser_presampled_fps + adaptive persistence scale-space kernel"]),
}

for ch_name in loaded_channels:
    save_dict[f"{ch_name}_H1_kernel_diagram"] = diagram_store[ch_name]

    for key, val in raw_interval_store[ch_name].items():
        save_dict[f"{ch_name}_{key}"] = val

np.savez(npz_path, **save_dict)

print("\nH1 Persistence-Kernel Channel Summary:")
if HAVE_IPYTHON:
    display(summary_df)
else:
    print(summary_df)

print("\nMost similar H1 diagram pairs by normalized persistence kernel:")
if HAVE_IPYTHON:
    display(pairwise_df.head(12))
else:
    print(pairwise_df.head(12))

print("\nLeast similar H1 diagram pairs by normalized persistence kernel:")
least_similar_df = pairwise_df.tail(12).sort_values("normalized_kernel_similarity", ascending=True)
if HAVE_IPYTHON:
    display(least_similar_df)
else:
    print(least_similar_df)

log(f"\nSaved summary CSV: {summary_csv_path}")
log(f"Saved raw kernel matrix CSV: {raw_kernel_csv_path}")
log(f"Saved normalized kernel similarity matrix CSV: {norm_kernel_csv_path}")
log(f"Saved kernel distance matrix CSV: {kernel_distance_csv_path}")
log(f"Saved pairwise similarities CSV: {pairwise_csv_path}")
log(f"Saved NPZ: {npz_path}")
log(f"Pairwise kernel elapsed: {pair_elapsed:.2f}s")

plot_df = summary_df.copy().sort_values("mean_kernel_similarity_to_others", ascending=False).reset_index(drop=True)

fig, ax = plt.subplots(figsize=(13, 5.8), facecolor=BLACK)
style_ax(
    ax,
    title=f"Mean H1 Persistence-Kernel Similarity to Other Channels | {FAST_OPTION.upper()}",
    xlabel="Channel",
    ylabel="Mean normalized kernel similarity",
)

x_pos = np.arange(len(plot_df))
y = plot_df["mean_kernel_similarity_to_others"].to_numpy(dtype=float)

ax.bar(x_pos, y, color=ACCENT, edgecolor=ACCENT, alpha=0.78)

if np.any(np.isfinite(y)):
    ax.axhline(
        np.nanmean(y),
        color=ACCENT_SOFT,
        linestyle="--",
        linewidth=1.4,
        alpha=0.8,
        label=f"Mean={np.nanmean(y):.5f}",
    )

bar_vmin, bar_vmax = dynamic_limits(y, lower_pct=0, upper_pct=100, pad_frac=0.25, lower_bound=0.0, upper_bound=1.0)
ax.set_ylim(bar_vmin, bar_vmax)

ax.set_xticks(x_pos)
ax.set_xticklabels(plot_df["channel"], rotation=90)
ax.legend(loc="best", framealpha=1.0)

mean_sim_path = os.path.join(mode_out_dir, "01_mean_h1_persistence_kernel_similarity_by_channel_zoomed.png")
plt.tight_layout()
save_render_close(fig, mean_sim_path)

if n_ch >= 3 and np.any(kernel_distance_mat > 0):
    condensed = kernel_distance_mat[np.triu_indices(n_ch, k=1)]
    Z = linkage(condensed, method="average")
    order = leaves_list(Z)
else:
    order = np.arange(n_ch)

ordered_sim = kernel_norm_mat[np.ix_(order, order)]
ordered_raw = kernel_mat[np.ix_(order, order)]
ordered_dist = kernel_distance_mat[np.ix_(order, order)]
ordered_channels = [loaded_channels[i] for i in order]

sim_vmin, sim_vmax = dynamic_limits(
    offdiag_values(ordered_sim),
    lower_pct=2,
    upper_pct=98,
    pad_frac=0.10,
    lower_bound=-1.0,
    upper_bound=1.0,
)

save_matrix_plot(
    ordered_sim,
    ordered_channels,
    "Normalized H1 Persistence-Kernel Similarity Matrix | Zoomed Color Scale",
    "Normalized kernel similarity",
    os.path.join(mode_out_dir, "02_h1_persistence_kernel_normalized_similarity_heatmap_zoomed.png"),
    vmin=sim_vmin,
    vmax=sim_vmax,
)

raw_vmin, raw_vmax = dynamic_limits(
    offdiag_values(ordered_raw),
    lower_pct=2,
    upper_pct=98,
    pad_frac=0.10,
)

save_matrix_plot(
    ordered_raw,
    ordered_channels,
    "Raw H1 Persistence Scale-Space Kernel Matrix | Zoomed Color Scale",
    "Raw kernel value",
    os.path.join(mode_out_dir, "03_h1_persistence_kernel_raw_heatmap_zoomed.png"),
    vmin=raw_vmin,
    vmax=raw_vmax,
)

dist_vmin, dist_vmax = dynamic_limits(
    offdiag_values(ordered_dist),
    lower_pct=2,
    upper_pct=98,
    pad_frac=0.10,
    lower_bound=0.0,
)

save_matrix_plot(
    ordered_dist,
    ordered_channels,
    "H1 Persistence-Kernel Distance Matrix | Zoomed Color Scale",
    "Kernel distance",
    os.path.join(mode_out_dir, "04_h1_persistence_kernel_distance_heatmap_zoomed.png"),
    vmin=dist_vmin,
    vmax=dist_vmax,
)

mds_coords = classical_mds_from_distances(kernel_distance_mat, n_components=2)

mds_df = pd.DataFrame({
    "channel": loaded_channels,
    "mds_1": mds_coords[:, 0],
    "mds_2": mds_coords[:, 1],
    "mean_kernel_similarity_to_others": [
        summary_df.loc[summary_df["channel"] == ch, "mean_kernel_similarity_to_others"].values[0]
        for ch in loaded_channels
    ],
    "mean_kernel_distance_to_others": [
        summary_df.loc[summary_df["channel"] == ch, "mean_kernel_distance_to_others"].values[0]
        for ch in loaded_channels
    ],
    "n_intervals_diagram": [
        summary_df.loc[summary_df["channel"] == ch, "n_intervals_diagram"].values[0]
        for ch in loaded_channels
    ],
})

mds_csv_path = os.path.join(mode_out_dir, "h1_persistence_kernel_mds_coordinates.csv")
mds_df.to_csv(mds_csv_path, index=False)

fig, ax = plt.subplots(figsize=(8, 7), facecolor=BLACK)
style_ax(
    ax,
    title="Classical MDS of H1 Persistence-Kernel Distances",
    xlabel="MDS 1",
    ylabel="MDS 2",
)

mds_vmin, mds_vmax = dynamic_limits(
    mds_df["mean_kernel_similarity_to_others"],
    lower_pct=2,
    upper_pct=98,
    pad_frac=0.08,
    lower_bound=0.0,
    upper_bound=1.0,
)

sc = ax.scatter(
    mds_df["mds_1"],
    mds_df["mds_2"],
    c=mds_df["mean_kernel_similarity_to_others"],
    s=150,
    cmap=CYAN_SEQUENTIAL_CMAP,
    edgecolor=ACCENT,
    linewidth=1.1,
    vmin=mds_vmin,
    vmax=mds_vmax,
)

for _, row in mds_df.iterrows():
    ax.text(
        row["mds_1"],
        row["mds_2"],
        row["channel"],
        fontsize=8,
        color=ACCENT,
        alpha=0.9,
    )

cbar = plt.colorbar(sc, ax=ax, shrink=0.85)
style_colorbar(cbar, "Mean normalized similarity")

mds_path = os.path.join(mode_out_dir, "05_h1_persistence_kernel_mds_embedding.png")
plt.tight_layout()
save_render_close(fig, mds_path)

diagram_channels = summary_df.head(min(12, len(summary_df)))["channel"].tolist()
n_cols = 4
n_rows = int(np.ceil(len(diagram_channels) / n_cols))

fig, axes = plt.subplots(
    n_rows,
    n_cols,
    figsize=(4 * n_cols, 3.2 * n_rows),
    facecolor=BLACK,
)

axes = np.asarray(axes).reshape(-1)

all_diagram_points = np.vstack([
    diagram_store[ch] for ch in loaded_channels if len(diagram_store[ch]) > 0
]) if any(len(diagram_store[ch]) > 0 for ch in loaded_channels) else np.empty((0, 2))

if len(all_diagram_points) > 0:
    diag_max = float(np.nanmax(all_diagram_points[:, 1]))
else:
    diag_max = 1.0

diag_max = max(diag_max, 1e-6)

for ax_idx, ch_name in enumerate(diagram_channels):
    ax = axes[ax_idx]
    D = diagram_store[ch_name]

    mean_sim = summary_df.loc[
        summary_df["channel"] == ch_name,
        "mean_kernel_similarity_to_others",
    ].values[0]

    style_ax(
        ax,
        title=f"{ch_name}: mean K={mean_sim:.3f}, n={len(D)}",
        xlabel="Birth",
        ylabel="Death",
    )

    ax.plot([0, diag_max], [0, diag_max], color=ACCENT_DIM, linestyle="--", linewidth=1.0)

    if len(D) > 0:
        ax.scatter(
            D[:, 0],
            D[:, 1],
            s=50,
            facecolors=BLACK,
            edgecolors=ACCENT,
            linewidths=1.2,
        )

    ax.set_xlim(0, diag_max * 1.05)
    ax.set_ylim(0, diag_max * 1.05)

for j in range(len(diagram_channels), len(axes)):
    axes[j].axis("off")

diagrams_path = os.path.join(mode_out_dir, "06_top_channels_h1_persistence_diagrams_by_kernel_similarity.png")
plt.tight_layout()
save_render_close(fig, diagrams_path, width=1100)

fig, ax = plt.subplots(figsize=(12, 6), facecolor=BLACK)
style_ax(
    ax,
    title="Most Similar Channel Pairs by H1 Persistence Kernel",
    xlabel="Channel pair",
    ylabel="Normalized kernel similarity",
)

top_pairs = pairwise_df.head(min(15, len(pairwise_df))).copy()
pair_labels = top_pairs["channel_a"] + "-" + top_pairs["channel_b"]

x_pair = np.arange(len(top_pairs))

ax.bar(
    x_pair,
    top_pairs["normalized_kernel_similarity"],
    color=ACCENT,
    edgecolor=ACCENT,
    alpha=0.78,
)

top_vmin, top_vmax = dynamic_limits(
    top_pairs["normalized_kernel_similarity"],
    lower_pct=0,
    upper_pct=100,
    pad_frac=0.25,
    lower_bound=0.0,
    upper_bound=1.0,
)
ax.set_ylim(top_vmin, top_vmax)

ax.set_xticks(x_pair)
ax.set_xticklabels(pair_labels, rotation=60, ha="right")

top_pairs_path = os.path.join(mode_out_dir, "07_most_similar_channel_pairs_h1_persistence_kernel_zoomed.png")
plt.tight_layout()
save_render_close(fig, top_pairs_path)

fig, ax = plt.subplots(figsize=(12, 6), facecolor=BLACK)
style_ax(
    ax,
    title="Least Similar Channel Pairs by H1 Persistence Kernel",
    xlabel="Channel pair",
    ylabel="Normalized kernel similarity",
)

bottom_pairs = pairwise_df.tail(min(15, len(pairwise_df))).copy().sort_values(
    "normalized_kernel_similarity",
    ascending=True,
)
bottom_pair_labels = bottom_pairs["channel_a"] + "-" + bottom_pairs["channel_b"]

x_pair = np.arange(len(bottom_pairs))

ax.bar(
    x_pair,
    bottom_pairs["normalized_kernel_similarity"],
    color=ACCENT,
    edgecolor=ACCENT,
    alpha=0.78,
)

bottom_vmin, bottom_vmax = dynamic_limits(
    bottom_pairs["normalized_kernel_similarity"],
    lower_pct=0,
    upper_pct=100,
    pad_frac=0.25,
    lower_bound=0.0,
    upper_bound=1.0,
)
ax.set_ylim(bottom_vmin, bottom_vmax)

ax.set_xticks(x_pair)
ax.set_xticklabels(bottom_pair_labels, rotation=60, ha="right")

bottom_pairs_path = os.path.join(mode_out_dir, "08_least_similar_channel_pairs_h1_persistence_kernel_zoomed.png")
plt.tight_layout()
save_render_close(fig, bottom_pairs_path)

fig, ax = plt.subplots(figsize=(8, 5.8), facecolor=BLACK)
style_ax(
    ax,
    title="Off-Diagonal Normalized Kernel Similarity Distribution",
    xlabel="Normalized kernel similarity",
    ylabel="Count",
)

if len(offdiag_sim):
    ax.hist(
        offdiag_sim,
        bins=24,
        color=ACCENT,
        edgecolor=ACCENT,
        alpha=0.75,
    )

hist_path = os.path.join(mode_out_dir, "09_offdiagonal_kernel_similarity_distribution.png")
plt.tight_layout()
save_render_close(fig, hist_path)

if MAKE_EXPERT_PLOTS:
    expert_dir = os.path.join(mode_out_dir, "expert_plots")
    os.makedirs(expert_dir, exist_ok=True)

    mean_kernel_map = dict(zip(summary_df["channel"], summary_df["mean_kernel_similarity_to_others"]))
    mean_kernel_distance_map = dict(zip(summary_df["channel"], summary_df["mean_kernel_distance_to_others"]))

    scalp_xy = {
        "Fp1": (-0.45, 1.00), "Fpz": (0.00, 1.08), "Fp2": (0.45, 1.00),
        "F7": (-0.95, 0.60), "F3": (-0.45, 0.55), "Fz": (0.00, 0.60),
        "F4": (0.45, 0.55), "F8": (0.95, 0.60),
        "FC5": (-0.75, 0.25), "FC1": (-0.25, 0.25),
        "FC2": (0.25, 0.25), "FC6": (0.75, 0.25),
        "M1": (-1.15, 0.05), "T7": (-1.00, 0.00), "C3": (-0.50, 0.00),
        "Cz": (0.00, 0.00), "C4": (0.50, 0.00), "T8": (1.00, 0.00),
        "M2": (1.15, 0.05),
        "CP5": (-0.75, -0.30), "CP1": (-0.25, -0.30),
        "CP2": (0.25, -0.30), "CP6": (0.75, -0.30),
        "P7": (-0.90, -0.60), "P3": (-0.45, -0.60), "Pz": (0.00, -0.65),
        "P4": (0.45, -0.60), "P8": (0.90, -0.60),
        "POz": (0.00, -0.88),
        "O1": (-0.35, -1.00), "Oz": (0.00, -1.08), "O2": (0.35, -1.00),
    }

    xs, ys, vals, labels = [], [], [], []

    for ch in eeg_channels:
        if ch in scalp_xy and ch in mean_kernel_map and np.isfinite(mean_kernel_map[ch]):
            xs.append(scalp_xy[ch][0])
            ys.append(scalp_xy[ch][1])
            vals.append(mean_kernel_map[ch])
            labels.append(ch)

    scalp_vmin, scalp_vmax = dynamic_limits(
        vals,
        lower_pct=2,
        upper_pct=98,
        pad_frac=0.10,
        lower_bound=0.0,
        upper_bound=1.0,
    )

    fig, ax = plt.subplots(figsize=(7, 7), facecolor=BLACK)
    style_ax(
        ax,
        title=f"Scalp Map of Mean H1 Persistence-Kernel Similarity | {FAST_OPTION.upper()}",
        grid=False,
    )

    head = plt.Circle((0, 0), 1.12, fill=False, color=ACCENT, linewidth=2)
    ax.add_patch(head)
    ax.plot([0.00, -0.10, 0.10, 0.00], [1.12, 1.28, 1.28, 1.12], color=ACCENT, linewidth=2)

    sc = ax.scatter(
        xs,
        ys,
        c=vals,
        s=260,
        cmap=CYAN_SEQUENTIAL_CMAP,
        edgecolor=ACCENT,
        linewidth=1.2,
        vmin=scalp_vmin,
        vmax=scalp_vmax,
    )

    for x0, y0, label in zip(xs, ys, labels):
        ax.text(
            x0,
            y0,
            label,
            ha="center",
            va="center",
            fontsize=8,
            color=BLACK,
            fontweight="bold",
        )

    cbar = plt.colorbar(sc, ax=ax, shrink=0.75)
    style_colorbar(cbar, "Mean normalized similarity")

    ax.set_xlim(-1.35, 1.35)
    ax.set_ylim(-1.35, 1.35)
    ax.set_aspect("equal")
    ax.set_xticks([])
    ax.set_yticks([])

    scalp_path = os.path.join(expert_dir, "10_mean_h1_persistence_kernel_similarity_scalp_map_zoomed.png")
    plt.tight_layout()
    save_render_close(fig, scalp_path, width=760)

    regions = {
        "Frontal": ["Fp1", "Fpz", "Fp2", "F7", "F3", "Fz", "F4", "F8"],
        "Frontocentral": ["FC5", "FC1", "FC2", "FC6"],
        "Central": ["T7", "C3", "Cz", "C4", "T8"],
        "Centroparietal": ["CP5", "CP1", "CP2", "CP6"],
        "Parietal": ["P7", "P3", "Pz", "P4", "P8"],
        "Occipital": ["POz", "O1", "Oz", "O2"],
        "Mastoid": ["M1", "M2"],
    }

    region_rows = []
    region_points = []

    for region_name, chans in regions.items():
        vals_region = []
        dist_vals_region = []

        for ch in chans:
            if ch in mean_kernel_map and np.isfinite(mean_kernel_map[ch]):
                vals_region.append(mean_kernel_map[ch])
                dist_vals_region.append(mean_kernel_distance_map[ch])
                region_points.append({
                    "region": region_name,
                    "channel": ch,
                    "mean_kernel_similarity_to_others": mean_kernel_map[ch],
                    "mean_kernel_distance_to_others": mean_kernel_distance_map[ch],
                })

        if len(vals_region) == 0:
            continue

        vals_region = np.asarray(vals_region, dtype=float)
        dist_vals_region = np.asarray(dist_vals_region, dtype=float)

        region_rows.append({
            "region": region_name,
            "mean_region_kernel_similarity": float(np.mean(vals_region)),
            "std_region_kernel_similarity": float(np.std(vals_region, ddof=1)) if len(vals_region) > 1 else 0.0,
            "sem_region_kernel_similarity": (
                float(np.std(vals_region, ddof=1) / np.sqrt(len(vals_region)))
                if len(vals_region) > 1 else 0.0
            ),
            "mean_region_kernel_distance": float(np.mean(dist_vals_region)),
            "n_channels": int(len(vals_region)),
        })

    region_df = pd.DataFrame(region_rows)
    region_points_df = pd.DataFrame(region_points)

    region_csv_path = os.path.join(expert_dir, "regional_h1_persistence_kernel_summary.csv")
    region_df.to_csv(region_csv_path, index=False)

    if len(region_df) > 0:
        fig, ax = plt.subplots(figsize=(12, 6.5), facecolor=BLACK)
        style_ax(
            ax,
            title=f"Regional Mean H1 Persistence-Kernel Similarity +/- SEM | {FAST_OPTION.upper()}",
            xlabel="Region",
            ylabel="Mean normalized kernel similarity",
        )

        x_region = np.arange(len(region_df))

        ax.bar(
            x_region,
            region_df["mean_region_kernel_similarity"],
            yerr=region_df["sem_region_kernel_similarity"],
            color=ACCENT,
            edgecolor=ACCENT,
            alpha=0.82,
            width=0.68,
            capsize=8,
            error_kw={
                "elinewidth": 2.2,
                "capthick": 2.2,
                "ecolor": ACCENT,
                "alpha": 1.0,
            },
            label="Mean +/- SEM",
        )

        rng = np.random.default_rng(7)

        for i, region_name in enumerate(region_df["region"]):
            vals_region_plot = region_points_df.loc[
                region_points_df["region"] == region_name,
                "mean_kernel_similarity_to_others",
            ].to_numpy(dtype=float)

            if len(vals_region_plot) == 0:
                continue

            jitter = rng.uniform(-0.13, 0.13, size=len(vals_region_plot))

            ax.scatter(
                np.full(len(vals_region_plot), i) + jitter,
                vals_region_plot,
                s=42,
                facecolors=BLACK,
                edgecolors=ACCENT,
                linewidths=1.2,
                alpha=0.95,
                zorder=4,
            )

        for i, row in region_df.iterrows():
            y_val = row["mean_region_kernel_similarity"]
            sem = row["sem_region_kernel_similarity"]

            ax.text(
                i,
                y_val + sem,
                f"{y_val:.3f}\n+/-{sem:.3f}\nn={int(row['n_channels'])}",
                ha="center",
                va="bottom",
                fontsize=8,
                color=ACCENT,
            )

        region_y = region_df["mean_region_kernel_similarity"].to_numpy(dtype=float)
        reg_vmin, reg_vmax = dynamic_limits(
            region_y,
            lower_pct=0,
            upper_pct=100,
            pad_frac=0.28,
            lower_bound=0.0,
            upper_bound=1.0,
        )
        ax.set_ylim(reg_vmin, reg_vmax)

        ax.set_xticks(x_region)
        ax.set_xticklabels(
            region_df["region"],
            rotation=35,
            ha="right",
            rotation_mode="anchor",
        )
        ax.margins(x=0.04)
        ax.legend(loc="best", framealpha=1.0)

        region_path = os.path.join(expert_dir, "11_regional_h1_persistence_kernel_similarity_summary_zoomed.png")
        plt.tight_layout(pad=2.0)
        save_render_close(fig, region_path)

    fig, ax = plt.subplots(figsize=(8, 6), facecolor=BLACK)
    style_ax(
        ax,
        title="Diagram Size vs Mean H1 Persistence-Kernel Similarity",
        xlabel="H1 diagram intervals",
        ylabel="Mean normalized kernel similarity",
    )

    sc = ax.scatter(
        summary_df["n_intervals_diagram"],
        summary_df["mean_kernel_similarity_to_others"],
        c=summary_df["max_lifetime"],
        s=140,
        cmap=CYAN_SEQUENTIAL_CMAP,
        edgecolor=ACCENT,
        linewidth=1.0,
    )

    for _, row in summary_df.iterrows():
        ax.text(
            row["n_intervals_diagram"],
            row["mean_kernel_similarity_to_others"],
            row["channel"],
            fontsize=8,
            color=ACCENT,
            alpha=0.85,
        )

    cbar = plt.colorbar(sc, ax=ax, shrink=0.85)
    style_colorbar(cbar, "Max H1 lifetime")

    scatter_vmin, scatter_vmax = dynamic_limits(
        summary_df["mean_kernel_similarity_to_others"],
        lower_pct=2,
        upper_pct=98,
        pad_frac=0.15,
        lower_bound=0.0,
        upper_bound=1.0,
    )
    ax.set_ylim(scatter_vmin, scatter_vmax)

    diagram_size_path = os.path.join(expert_dir, "12_diagram_size_vs_mean_kernel_similarity_zoomed.png")
    plt.tight_layout()
    save_render_close(fig, diagram_size_path)

    fig, ax = plt.subplots(figsize=(8, 6), facecolor=BLACK)
    style_ax(
        ax,
        title="Threshold Triangles vs Mean H1 Persistence-Kernel Similarity",
        xlabel="Triangles at radius",
        ylabel="Mean normalized kernel similarity",
    )

    sc = ax.scatter(
        summary_df["n_triangles_at_radius"],
        summary_df["mean_kernel_similarity_to_others"],
        c=summary_df["n_intervals_diagram"],
        s=140,
        cmap=CYAN_SEQUENTIAL_CMAP,
        edgecolor=ACCENT,
        linewidth=1.0,
    )

    for _, row in summary_df.iterrows():
        ax.text(
            row["n_triangles_at_radius"],
            row["mean_kernel_similarity_to_others"],
            row["channel"],
            fontsize=8,
            color=ACCENT,
            alpha=0.85,
        )

    cbar = plt.colorbar(sc, ax=ax, shrink=0.85)
    style_colorbar(cbar, "H1 diagram intervals")

    ax.set_ylim(scatter_vmin, scatter_vmax)

    triangle_path = os.path.join(expert_dir, "13_threshold_triangles_vs_mean_kernel_similarity_zoomed.png")
    plt.tight_layout()
    save_render_close(fig, triangle_path)

    fig, ax = plt.subplots(figsize=(8, 6), facecolor=BLACK)
    style_ax(
        ax,
        title="Kernel Similarity Spread per Channel",
        xlabel="Mean normalized kernel similarity",
        ylabel="Similarity standard deviation",
    )

    sc = ax.scatter(
        summary_df["mean_kernel_similarity_to_others"],
        summary_df["std_kernel_similarity_to_others"],
        c=summary_df["n_intervals_diagram"],
        s=140,
        cmap=CYAN_SEQUENTIAL_CMAP,
        edgecolor=ACCENT,
        linewidth=1.0,
    )

    for _, row in summary_df.iterrows():
        ax.text(
            row["mean_kernel_similarity_to_others"],
            row["std_kernel_similarity_to_others"],
            row["channel"],
            fontsize=8,
            color=ACCENT,
            alpha=0.85,
        )

    cbar = plt.colorbar(sc, ax=ax, shrink=0.85)
    style_colorbar(cbar, "H1 diagram intervals")

    spread_path = os.path.join(expert_dir, "14_kernel_similarity_spread_per_channel.png")
    plt.tight_layout()
    save_render_close(fig, spread_path)

    qc_df = summary_df.copy().sort_values("elapsed_sec", ascending=False).reset_index(drop=True)

    fig, ax = plt.subplots(figsize=(12, 5.8), facecolor=BLACK)
    style_ax(
        ax,
        title="Persistence-Kernel Diagram Runtime QC",
        xlabel="Channel",
        ylabel="Elapsed seconds",
    )

    x_qc = np.arange(len(qc_df))

    ax.bar(
        x_qc,
        qc_df["elapsed_sec"],
        color=ACCENT,
        edgecolor=ACCENT,
        alpha=0.82,
    )

    ax.set_xticks(x_qc)
    ax.set_xticklabels(qc_df["channel"], rotation=90)

    qc_path = os.path.join(expert_dir, "15_runtime_qc.png")
    plt.tight_layout()
    save_render_close(fig, qc_path)

    log(f"\nExpert plots saved and rendered from: {expert_dir}")
    log(f"Regional summary CSV: {region_csv_path}")

elapsed_total = time.time() - overall_t0

log("\n" + "=" * 72)
log(f"Finished H1 Persistence-Kernel Similarity Matrix in {elapsed_total:.2f}s")
log(f"Mode: {FAST_OPTION}")
log(f"Backend: ripser_presampled_fps + adaptive persistence scale-space kernel")
log(f"Kernel bandwidth length: {kernel_bandwidth:.8f}")
log(f"Kernel time: {kernel_time:.10f}")
log(f"Kernel bandwidth method: {kernel_bandwidth_method}")
log(f"Channels completed: {len(loaded_channels)}")
log(f"Summary CSV: {summary_csv_path}")
log(f"Raw kernel CSV: {raw_kernel_csv_path}")
log(f"Normalized similarity CSV: {norm_kernel_csv_path}")
log(f"Kernel distance CSV: {kernel_distance_csv_path}")
log(f"Pairwise CSV: {pairwise_csv_path}")
log(f"MDS CSV: {mds_csv_path}")
log(f"NPZ: {npz_path}")
log(f"Output directory: {mode_out_dir}")
log("=" * 72)
```

```python
import os
import sys
import time
import numpy as np

if not hasattr(np, "trapz"):
    np.trapz = np.trapezoid

import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.colors import LinearSegmentedColormap

os.environ["OMP_NUM_THREADS"] = "2"
os.environ["OPENBLAS_NUM_THREADS"] = "2"
os.environ["MKL_NUM_THREADS"] = "2"
os.environ["VECLIB_MAXIMUM_THREADS"] = "2"
os.environ["NUMEXPR_NUM_THREADS"] = "2"

from scipy.spatial.distance import pdist, squareform
from scipy.cluster.hierarchy import linkage, leaves_list

try:
    from scipy.linalg import eigh
    HAVE_SCIPY_EIGH = True
except Exception:
    HAVE_SCIPY_EIGH = False

try:
    from IPython.display import display, Image
    HAVE_IPYTHON = True
except Exception:
    HAVE_IPYTHON = False

try:
    from ripser import ripser
except Exception as exc:
    raise ImportError(
        "ripser is not installed. Run this in a separate notebook cell first:\n\n"
        "!pip install ripser persim\n\n"
        "Then rerun this Persistence Scale-Space Kernel cell."
    ) from exc


base_dir = "/home/a/projects/Complete-Neural-Signal-Analysis"
out_dir = os.path.join(base_dir, "topology_persistence_scale_space_kernel")
os.makedirs(out_dir, exist_ok=True)

FAST_OPTION = "ultra"  # options: "ultra", "fast", "balanced", "full"

MAX_CHANNELS_TO_PROCESS = None
RENDER_ALL_IMAGES_INLINE = True
SAVE_PLOTS_TO_DISK = True
MAKE_EXPERT_PLOTS = True
INLINE_IMAGE_WIDTH = 950
VERBOSE_PROGRESS = True

USE_GEOMETRY_PRESERVING_SUBSAMPLE = True
AUTO_KERNEL_BANDWIDTH = True

MODE_CONFIGS = {
    "ultra": {
        "max_points": 90,
        "reservoir_points": 5000,
        "max_dims": 3,
        "radius_quantile": 0.22,
        "kernel_bandwidth_frac": 0.035,
        "kernel_auto_quantile": 0.25,
        "kernel_auto_multiplier": 0.60,
        "cap_essential_h1": True,
        "plot_dpi": 120,
    },
    "fast": {
        "max_points": 130,
        "reservoir_points": 8000,
        "max_dims": 3,
        "radius_quantile": 0.24,
        "kernel_bandwidth_frac": 0.030,
        "kernel_auto_quantile": 0.25,
        "kernel_auto_multiplier": 0.55,
        "cap_essential_h1": True,
        "plot_dpi": 130,
    },
    "balanced": {
        "max_points": 180,
        "reservoir_points": 10000,
        "max_dims": 3,
        "radius_quantile": 0.26,
        "kernel_bandwidth_frac": 0.026,
        "kernel_auto_quantile": 0.22,
        "kernel_auto_multiplier": 0.50,
        "cap_essential_h1": True,
        "plot_dpi": 140,
    },
    "full": {
        "max_points": 260,
        "reservoir_points": 14000,
        "max_dims": 3,
        "radius_quantile": 0.28,
        "kernel_bandwidth_frac": 0.022,
        "kernel_auto_quantile": 0.20,
        "kernel_auto_multiplier": 0.45,
        "cap_essential_h1": True,
        "plot_dpi": 150,
    },
}

if FAST_OPTION not in MODE_CONFIGS:
    raise ValueError("FAST_OPTION must be one of: ultra, fast, balanced, full")

cfg = MODE_CONFIGS[FAST_OPTION]

MAX_POINTS = cfg["max_points"]
RESERVOIR_POINTS = cfg["reservoir_points"]
MAX_DIMS = cfg["max_dims"]
RADIUS_QUANTILE = cfg["radius_quantile"]
KERNEL_BANDWIDTH_FRAC = cfg["kernel_bandwidth_frac"]
KERNEL_AUTO_QUANTILE = cfg["kernel_auto_quantile"]
KERNEL_AUTO_MULTIPLIER = cfg["kernel_auto_multiplier"]
CAP_ESSENTIAL_H1 = cfg["cap_essential_h1"]
PLOT_DPI = cfg["plot_dpi"]

mode_out_dir = os.path.join(
    out_dir,
    f"h1_persistence_scale_space_kernel_{FAST_OPTION}_fps_adaptive",
)
os.makedirs(mode_out_dir, exist_ok=True)

eeg_channels = [
    "Fp1", "Fpz", "Fp2", "F7", "F3", "Fz", "F4", "F8", "FC5", "FC1", "FC2", "FC6",
    "M1", "T7", "C3", "Cz", "C4", "T8", "M2", "CP5", "CP1", "CP2", "CP6",
    "P7", "P3", "Pz", "P4", "P8", "POz", "O1", "Oz", "O2"
]

if MAX_CHANNELS_TO_PROCESS is not None:
    eeg_channels = eeg_channels[:MAX_CHANNELS_TO_PROCESS]

ACCENT = "#00FFFF"
ACCENT_SOFT = "#66FFFF"
ACCENT_DIM = "#008B8B"
BLACK = "#000000"

CYAN_SEQUENTIAL_CMAP = LinearSegmentedColormap.from_list(
    "cyan_sequential",
    [BLACK, ACCENT_DIM, ACCENT, ACCENT_SOFT],
    N=256,
)

CYAN_DIVERGING_CMAP = LinearSegmentedColormap.from_list(
    "cyan_diverging",
    [ACCENT_DIM, BLACK, ACCENT_SOFT],
    N=256,
)

plt.rcParams.update({
    "figure.facecolor": BLACK,
    "axes.facecolor": BLACK,
    "savefig.facecolor": BLACK,
    "text.color": ACCENT,
    "axes.labelcolor": ACCENT,
    "axes.edgecolor": ACCENT,
    "axes.titlecolor": ACCENT,
    "xtick.color": ACCENT,
    "ytick.color": ACCENT,
    "grid.color": ACCENT_DIM,
    "grid.alpha": 0.25,
    "legend.facecolor": BLACK,
    "legend.edgecolor": ACCENT,
    "legend.labelcolor": ACCENT,
    "font.size": 11,
})


def log(msg):
    if VERBOSE_PROGRESS:
        print(msg, flush=True)


def style_ax(ax, title=None, xlabel=None, ylabel=None, grid=True):
    ax.set_facecolor(BLACK)

    for spine in ax.spines.values():
        spine.set_color(ACCENT)
        spine.set_linewidth(1.1)

    ax.tick_params(axis="x", colors=ACCENT, labelcolor=ACCENT)
    ax.tick_params(axis="y", colors=ACCENT, labelcolor=ACCENT)

    if title is not None:
        ax.set_title(title, color=ACCENT, pad=12)

    if xlabel is not None:
        ax.set_xlabel(xlabel, color=ACCENT, labelpad=10)

    if ylabel is not None:
        ax.set_ylabel(ylabel, color=ACCENT, labelpad=10)

    if grid:
        ax.grid(True, color=ACCENT_DIM, alpha=0.25, linewidth=0.8)


def style_colorbar(cbar, label):
    cbar.set_label(label, color=ACCENT)
    cbar.ax.yaxis.set_tick_params(color=ACCENT)
    plt.setp(cbar.ax.get_yticklabels(), color=ACCENT)
    cbar.outline.set_edgecolor(ACCENT)


def save_render_close(fig, path, width=INLINE_IMAGE_WIDTH):
    if SAVE_PLOTS_TO_DISK:
        fig.savefig(path, dpi=PLOT_DPI, bbox_inches="tight", facecolor=BLACK)

    plt.close(fig)

    if RENDER_ALL_IMAGES_INLINE and HAVE_IPYTHON and os.path.exists(path):
        display(Image(filename=path, width=width))


def dynamic_limits(values, lower_pct=2, upper_pct=98, pad_frac=0.08, lower_bound=None, upper_bound=None):
    values = np.asarray(values, dtype=float)
    values = values[np.isfinite(values)]

    if len(values) == 0:
        vmin, vmax = 0.0, 1.0
    else:
        vmin = float(np.percentile(values, lower_pct))
        vmax = float(np.percentile(values, upper_pct))

        if not np.isfinite(vmin):
            vmin = float(np.nanmin(values))

        if not np.isfinite(vmax):
            vmax = float(np.nanmax(values))

        if vmax <= vmin:
            center = float(np.nanmean(values))
            eps = max(1e-6, abs(center) * 0.01, 1e-5)
            vmin = center - eps
            vmax = center + eps

    pad = (vmax - vmin) * pad_frac
    vmin -= pad
    vmax += pad

    if lower_bound is not None:
        vmin = max(lower_bound, vmin)

    if upper_bound is not None:
        vmax = min(upper_bound, vmax)

    if vmax <= vmin:
        vmax = vmin + 1e-6

    return vmin, vmax


def offdiag_values(M):
    M = np.asarray(M, dtype=float)

    if M.shape[0] <= 1:
        return np.asarray([], dtype=float)

    mask = ~np.eye(M.shape[0], dtype=bool)
    vals = M[mask]
    vals = vals[np.isfinite(vals)]

    return vals


def save_matrix_plot(
    matrix,
    labels,
    title,
    colorbar_label,
    path,
    vmin=None,
    vmax=None,
    width=INLINE_IMAGE_WIDTH,
    cmap=None,
):
    fig, ax = plt.subplots(figsize=(9, 8), facecolor=BLACK)
    style_ax(
        ax,
        title=title,
        xlabel="Channel",
        ylabel="Channel",
    )

    if cmap is None:
        cmap = CYAN_SEQUENTIAL_CMAP

    im = ax.imshow(
        matrix,
        aspect="equal",
        interpolation="nearest",
        cmap=cmap,
        vmin=vmin,
        vmax=vmax,
    )

    ax.set_xticks(np.arange(len(labels)))
    ax.set_xticklabels(labels, rotation=90)
    ax.set_yticks(np.arange(len(labels)))
    ax.set_yticklabels(labels)

    cbar = plt.colorbar(im, ax=ax, shrink=0.85)
    style_colorbar(cbar, colorbar_label)

    plt.tight_layout()
    save_render_close(fig, path, width=width)


def estimate_median_pairwise_distance(X, max_pairs=120000, seed=7):
    X = np.asarray(X, dtype=float)
    n = X.shape[0]

    if n < 2:
        return 1.0

    rng = np.random.default_rng(seed)

    if n * (n - 1) // 2 <= max_pairs:
        D = pdist(X)
        D = D[D > 0]

        if len(D) == 0:
            return 1.0

        return float(np.nanmedian(D))

    i = rng.integers(0, n, size=max_pairs)
    j = rng.integers(0, n, size=max_pairs)

    mask = i != j
    i = i[mask]
    j = j[mask]

    if len(i) == 0:
        return 1.0

    d = np.linalg.norm(X[i] - X[j], axis=1)
    d = d[np.isfinite(d) & (d > 0)]

    if len(d) == 0:
        return 1.0

    return float(np.nanmedian(d))


def standardize_matrix_for_geometry(X, max_dims=3, seed=7):
    X = np.asarray(X, dtype=np.float64)
    X = X[np.all(np.isfinite(X), axis=1)]

    if X.shape[0] < 10:
        return None

    X = X - np.mean(X, axis=0, keepdims=True)

    if X.shape[1] > max_dims:
        _U, _S, Vt = np.linalg.svd(X, full_matrices=False)
        X = X @ Vt[:max_dims].T

    X = X / (np.std(X, axis=0, keepdims=True) + 1e-12)

    scale = estimate_median_pairwise_distance(X, seed=seed)
    X = X / (scale + 1e-12)

    return X


def farthest_point_sample(X, n_samples, seed=7):
    X = np.asarray(X, dtype=float)
    n = X.shape[0]

    if n <= n_samples:
        return np.arange(n)

    rng = np.random.default_rng(seed)

    center = np.mean(X, axis=0, keepdims=True)
    dist_to_center = np.linalg.norm(X - center, axis=1)

    start_idx = int(np.nanargmax(dist_to_center))

    selected = np.empty(n_samples, dtype=int)
    selected.fill(-1)
    selected[0] = start_idx

    min_dist2 = np.full(n, np.inf, dtype=float)

    for k in range(1, n_samples):
        last = X[selected[k - 1]]
        d2 = np.sum((X - last) ** 2, axis=1)
        min_dist2 = np.minimum(min_dist2, d2)
        min_dist2[selected[:k]] = -np.inf

        next_idx = int(np.nanargmax(min_dist2))

        if not np.isfinite(min_dist2[next_idx]) or min_dist2[next_idx] < 1e-18:
            remaining = np.setdiff1d(np.arange(n), selected[:k], assume_unique=False)

            if len(remaining) == 0:
                selected = selected[:k]
                break

            next_idx = int(rng.choice(remaining))

        selected[k] = next_idx

    selected = selected[selected >= 0]

    return selected


def load_channel_embedding_geometry_sampled(ch_name, max_points=90, reservoir_points=5000, max_dims=3, seed=7):
    candidate_paths = [
        os.path.join(base_dir, "embedding_data", "tsne_umap_from_3d", f"umap_embedding_{ch_name}.npy"),
        os.path.join(base_dir, "embedding_data", "tsne_umap_from_3d", f"tsne_embedding_{ch_name}.npy"),
        os.path.join(base_dir, "embedding_data", "3dembedding_data", f"3dembedded_{ch_name}.npy"),
        os.path.join(base_dir, "embedding_data", "2dembedding_data", f"2dembedded_{ch_name}.npy"),
    ]

    rng = np.random.default_rng(seed)

    for path in candidate_paths:
        if not os.path.exists(path):
            continue

        log(f"    trying: {path}")

        try:
            X = np.load(path, mmap_mode="r", allow_pickle=False)
        except Exception as exc:
            log(f"    skipped non-memmappable or nonnumeric file: {os.path.basename(path)} | {exc}")
            continue

        if X.ndim != 2:
            log(f"    skipped because ndim={X.ndim}")
            continue

        original_shape = tuple(X.shape)

        try:
            if X.shape[0] >= X.shape[1]:
                n_samples = X.shape[0]
                n_features = X.shape[1]

                if n_samples < 10 or n_features < 2:
                    log(f"    skipped bad shape: {original_shape}")
                    continue

                reservoir_n = min(max(reservoir_points, max_points), n_samples)

                if n_samples > reservoir_n:
                    reservoir_idx = np.sort(rng.choice(n_samples, size=reservoir_n, replace=False))
                else:
                    reservoir_idx = np.arange(n_samples)

                X_reservoir = np.asarray(X[reservoir_idx, :], dtype=np.float64)

            else:
                n_samples = X.shape[1]
                n_features = X.shape[0]

                if n_samples < 10 or n_features < 2:
                    log(f"    skipped bad transposed shape: {original_shape}")
                    continue

                reservoir_n = min(max(reservoir_points, max_points), n_samples)

                if n_samples > reservoir_n:
                    reservoir_idx = np.sort(rng.choice(n_samples, size=reservoir_n, replace=False))
                else:
                    reservoir_idx = np.arange(n_samples)

                X_reservoir = np.asarray(X[:, reservoir_idx].T, dtype=np.float64)

        except Exception as exc:
            log(f"    failed while reservoir-sampling {os.path.basename(path)} | {exc}")
            continue

        X_reservoir = X_reservoir[np.all(np.isfinite(X_reservoir), axis=1)]

        if X_reservoir.shape[0] < 10 or X_reservoir.shape[1] < 2:
            log(f"    reservoir data unusable: reservoir_shape={X_reservoir.shape}")
            continue

        X_reservoir_std = standardize_matrix_for_geometry(
            X_reservoir,
            max_dims=max_dims,
            seed=seed,
        )

        if X_reservoir_std is None:
            log("    standardized reservoir unusable")
            continue

        if USE_GEOMETRY_PRESERVING_SUBSAMPLE:
            fps_idx = farthest_point_sample(
                X_reservoir_std,
                n_samples=min(max_points, X_reservoir_std.shape[0]),
                seed=seed,
            )
            X_small = X_reservoir_std[fps_idx]
            method = "reservoir_plus_farthest_point"
        else:
            sample_n = min(max_points, X_reservoir_std.shape[0])
            idx = np.sort(rng.choice(X_reservoir_std.shape[0], size=sample_n, replace=False))
            X_small = X_reservoir_std[idx]
            method = "reservoir_random"

        X_small = X_small[np.all(np.isfinite(X_small), axis=1)]

        if X_small.shape[0] >= 10 and X_small.shape[1] >= 2:
            return X_small, path, original_shape, tuple(X_reservoir.shape), method

        log(f"    sampled data unusable: sampled_shape={X_small.shape}")

    return None, None, None, None, None


def standardize_point_cloud_small(X, max_dims=3):
    X = np.asarray(X, dtype=np.float64)
    X = X[np.all(np.isfinite(X), axis=1)]

    if X.shape[0] < 10:
        return None

    X = X - np.mean(X, axis=0, keepdims=True)

    if X.shape[1] > max_dims:
        _U, _S, Vt = np.linalg.svd(X, full_matrices=False)
        X = X @ Vt[:max_dims].T

    X = X / (np.std(X, axis=0, keepdims=True) + 1e-12)

    D = pdist(X)
    positive = D[D > 0]

    if len(positive) > 0:
        X = X / (np.nanmedian(positive) + 1e-12)

    return X


def estimate_complex_counts(X, radius):
    D = squareform(pdist(X))
    A = (D <= radius) & (D > 0)

    n_edges = int(np.sum(A) // 2)

    if A.shape[0] <= 300:
        A_int = A.astype(np.int64)
        n_triangles = int(np.trace(A_int @ A_int @ A_int) // 6)
    else:
        n_triangles = np.nan

    return n_edges, n_triangles


def run_h1_ripser(X, radius):
    result = ripser(
        X,
        maxdim=1,
        thresh=float(radius),
        coeff=2,
        do_cocycles=False,
    )

    diagrams = result["dgms"]

    h0 = diagrams[0] if len(diagrams) > 0 else np.empty((0, 2), dtype=float)
    h1 = diagrams[1] if len(diagrams) > 1 else np.empty((0, 2), dtype=float)

    return h0, h1


def split_finite_essential(diagram):
    diagram = np.asarray(diagram, dtype=float)

    if diagram.size == 0:
        return np.empty((0, 2), dtype=float), np.empty((0, 2), dtype=float)

    finite = diagram[
        np.isfinite(diagram[:, 0])
        & np.isfinite(diagram[:, 1])
        & (diagram[:, 1] > diagram[:, 0])
    ]

    essential = diagram[
        np.isfinite(diagram[:, 0])
        & np.isinf(diagram[:, 1])
    ]

    return finite, essential


def clean_diagram_for_scale_space(
    finite_intervals,
    essential_intervals=None,
    cap_essential=True,
    cap_value=None,
):
    cleaned = []
    capped_count = 0

    if finite_intervals is not None:
        for b, d in finite_intervals:
            if np.isfinite(b) and np.isfinite(d) and d > b:
                cleaned.append((float(b), float(d)))

    if cap_essential and essential_intervals is not None and cap_value is not None:
        for b, _d in essential_intervals:
            if np.isfinite(b) and np.isfinite(cap_value) and cap_value > b:
                cleaned.append((float(b), float(cap_value)))
                capped_count += 1

    if len(cleaned) == 0:
        return np.empty((0, 2), dtype=float), capped_count

    return np.asarray(cleaned, dtype=float), capped_count


def diagram_features(diagram):
    diagram = np.asarray(diagram, dtype=float)

    if diagram.size == 0 or len(diagram) == 0:
        return {
            "n_intervals_diagram": 0,
            "mean_lifetime": np.nan,
            "max_lifetime": np.nan,
            "total_lifetime": 0.0,
            "mean_birth": np.nan,
            "mean_death": np.nan,
            "max_death": np.nan,
        }

    lifetimes = diagram[:, 1] - diagram[:, 0]

    return {
        "n_intervals_diagram": int(len(diagram)),
        "mean_lifetime": float(np.nanmean(lifetimes)),
        "max_lifetime": float(np.nanmax(lifetimes)),
        "total_lifetime": float(np.nansum(lifetimes)),
        "mean_birth": float(np.nanmean(diagram[:, 0])),
        "mean_death": float(np.nanmean(diagram[:, 1])),
        "max_death": float(np.nanmax(diagram[:, 1])),
    }


def persistence_scale_space_kernel(D1, D2, kernel_time):
    D1 = np.asarray(D1, dtype=float)
    D2 = np.asarray(D2, dtype=float)

    if D1.size == 0:
        D1 = np.empty((0, 2), dtype=float)

    if D2.size == 0:
        D2 = np.empty((0, 2), dtype=float)

    if len(D1) == 0 or len(D2) == 0:
        return 0.0

    D1 = D1[
        np.isfinite(D1[:, 0])
        & np.isfinite(D1[:, 1])
        & (D1[:, 1] > D1[:, 0])
    ]

    D2 = D2[
        np.isfinite(D2[:, 0])
        & np.isfinite(D2[:, 1])
        & (D2[:, 1] > D2[:, 0])
    ]

    if len(D1) == 0 or len(D2) == 0:
        return 0.0

    kernel_time = float(max(kernel_time, 1e-12))

    D2_reflected = D2[:, ::-1]

    diff_direct = D1[:, None, :] - D2[None, :, :]
    diff_reflect = D1[:, None, :] - D2_reflected[None, :, :]

    dist2_direct = np.sum(diff_direct ** 2, axis=2)
    dist2_reflect = np.sum(diff_reflect ** 2, axis=2)

    denom = 8.0 * kernel_time
    prefactor = 1.0 / (8.0 * np.pi * kernel_time)

    K = prefactor * np.sum(
        np.exp(-dist2_direct / denom) - np.exp(-dist2_reflect / denom)
    )

    if not np.isfinite(K):
        return 0.0

    return float(K)


def normalize_kernel_matrix(K):
    K = np.asarray(K, dtype=float)
    K = 0.5 * (K + K.T)

    diag = np.diag(K).copy()
    K_norm = np.zeros_like(K, dtype=float)

    for i in range(K.shape[0]):
        for j in range(K.shape[1]):
            if diag[i] > 0 and diag[j] > 0:
                K_norm[i, j] = K[i, j] / np.sqrt(diag[i] * diag[j])
            elif i == j:
                K_norm[i, j] = 1.0
            else:
                K_norm[i, j] = 0.0

    K_norm = 0.5 * (K_norm + K_norm.T)
    K_norm = np.clip(K_norm, -1.0, 1.0)
    np.fill_diagonal(K_norm, 1.0)

    return K_norm


def kernel_distance_from_normalized_similarity(K_norm):
    K_norm = np.asarray(K_norm, dtype=float)
    D2 = np.maximum(0.0, 2.0 - 2.0 * K_norm)
    D = np.sqrt(D2)
    np.fill_diagonal(D, 0.0)
    return D


def choose_adaptive_kernel_bandwidth(diagram_store, fallback_scale):
    all_points = []

    for D in diagram_store.values():
        D = np.asarray(D, dtype=float)

        if D.size == 0 or len(D) == 0:
            continue

        D = D[
            np.isfinite(D[:, 0])
            & np.isfinite(D[:, 1])
            & (D[:, 1] > D[:, 0])
        ]

        if len(D) > 0:
            all_points.append(D)

    fallback_bandwidth = max(1e-6, KERNEL_BANDWIDTH_FRAC * fallback_scale)

    if len(all_points) == 0:
        return fallback_bandwidth, fallback_bandwidth ** 2, "fallback_no_points"

    all_points = np.vstack(all_points)

    if len(all_points) < 2:
        return fallback_bandwidth, fallback_bandwidth ** 2, "fallback_too_few_points"

    D = pdist(all_points)
    D = D[np.isfinite(D) & (D > 0)]

    if len(D) == 0:
        return fallback_bandwidth, fallback_bandwidth ** 2, "fallback_zero_distances"

    auto_bandwidth = float(np.quantile(D, KERNEL_AUTO_QUANTILE) * KERNEL_AUTO_MULTIPLIER)

    if not np.isfinite(auto_bandwidth) or auto_bandwidth <= 0:
        return fallback_bandwidth, fallback_bandwidth ** 2, "fallback_bad_auto"

    if AUTO_KERNEL_BANDWIDTH:
        chosen = min(fallback_bandwidth, auto_bandwidth)
        chosen = max(chosen, fallback_scale * 0.0025, 1e-6)
        method = "adaptive_min_of_fractional_and_diagram_quantile"
    else:
        chosen = fallback_bandwidth
        method = "fixed_fractional"

    return chosen, chosen ** 2, method


def classical_mds_from_distances(D, n_components=2):
    D = np.asarray(D, dtype=float)
    n = D.shape[0]

    if n < 2:
        return np.zeros((n, n_components), dtype=float)

    D2 = D ** 2
    J = np.eye(n) - np.ones((n, n)) / n
    B = -0.5 * J @ D2 @ J

    if HAVE_SCIPY_EIGH:
        vals, vecs = eigh(B)
    else:
        vals, vecs = np.linalg.eigh(B)

    idx = np.argsort(vals)[::-1]
    vals = vals[idx]
    vecs = vecs[:, idx]

    coords = np.zeros((n, n_components), dtype=float)

    for k in range(min(n_components, n)):
        if vals[k] > 0:
            coords[:, k] = vecs[:, k] * np.sqrt(vals[k])

    return coords


log("\n" + "=" * 72)
log(f"Running H1 Persistence Scale-Space Kernel | {FAST_OPTION.upper()}")
log(f"Channels: {len(eeg_channels)}")
log(f"max_points={MAX_POINTS}, reservoir_points={RESERVOIR_POINTS}, radius_quantile={RADIUS_QUANTILE}")
log(f"kernel_bandwidth_frac={KERNEL_BANDWIDTH_FRAC}, auto_kernel_bandwidth={AUTO_KERNEL_BANDWIDTH}")
log(f"geometry_preserving_subsample={USE_GEOMETRY_PRESERVING_SUBSAMPLE}")
log(f"cap_essential_h1={CAP_ESSENTIAL_H1}")
log(f"expert_plots={MAKE_EXPERT_PLOTS}, render_all_images={RENDER_ALL_IMAGES_INLINE}")
log(f"Output: {mode_out_dir}")
log("=" * 72)

diagnostics = {}
diagram_store = {}
raw_interval_store = {}
loaded_channels = []

global_max_death = None
global_max_lifetime = None

overall_t0 = time.time()

for ch_idx, ch_name in enumerate(eeg_channels, start=1):
    t0 = time.time()
    prefix = f"[{ch_idx:02d}/{len(eeg_channels)} {ch_name}]"

    log(f"\n{prefix} loading reservoir + geometry-preserving subsampling")

    X_small, source_path, original_shape, reservoir_shape, sampling_method = load_channel_embedding_geometry_sampled(
        ch_name,
        max_points=MAX_POINTS,
        reservoir_points=RESERVOIR_POINTS,
        max_dims=MAX_DIMS,
        seed=800 + ch_idx,
    )

    if X_small is None:
        log(f"{prefix} WARNING: no usable embedding found; skipping")
        continue

    log(
        f"{prefix} source shape={original_shape}, reservoir shape={reservoir_shape}, "
        f"sampled shape={X_small.shape}, method={sampling_method}"
    )

    X = standardize_point_cloud_small(X_small, max_dims=MAX_DIMS)

    if X is None:
        log(f"{prefix} WARNING: invalid sampled embedding; skipping")
        continue

    log(f"{prefix} standardized shape={X.shape}; computing radius")

    D_flat = pdist(X)
    positive = D_flat[D_flat > 0]

    if len(positive) == 0:
        log(f"{prefix} WARNING: zero-distance cloud; skipping")
        continue

    radius = float(np.quantile(positive, RADIUS_QUANTILE))

    log(f"{prefix} estimating threshold complex size")
    n_edges, n_triangles = estimate_complex_counts(X, radius)

    log(f"{prefix} ripser H0/H1")
    h0_diag, h1_diag = run_h1_ripser(X, radius)

    h0_finite, h0_essential = split_finite_essential(h0_diag)
    h1_finite, h1_essential = split_finite_essential(h1_diag)

    h1_diagram, n_capped_h1 = clean_diagram_for_scale_space(
        h1_finite,
        essential_intervals=h1_essential,
        cap_essential=CAP_ESSENTIAL_H1,
        cap_value=radius,
    )

    if len(h1_diagram) > 0:
        max_death = float(np.nanmax(h1_diagram[:, 1]))
        max_lifetime = float(np.nanmax(h1_diagram[:, 1] - h1_diagram[:, 0]))

        global_max_death = max_death if global_max_death is None else max(global_max_death, max_death)
        global_max_lifetime = max_lifetime if global_max_lifetime is None else max(global_max_lifetime, max_lifetime)

    elapsed = time.time() - t0

    diagram_store[ch_name] = h1_diagram

    raw_interval_store[ch_name] = {
        "H0_finite": h0_finite,
        "H0_essential": h0_essential,
        "H1_finite": h1_finite,
        "H1_essential": h1_essential,
        "H1_scale_space_diagram": h1_diagram,
    }

    diagnostics[ch_name] = {
        "channel": ch_name,
        "source_path": source_path,
        "source_shape": str(original_shape),
        "reservoir_shape": str(reservoir_shape),
        "sampling_method": sampling_method,
        "backend": "ripser_presampled_fps",
        "n_points": int(X.shape[0]),
        "n_edges_at_radius": int(n_edges),
        "n_triangles_at_radius": int(n_triangles) if np.isfinite(n_triangles) else np.nan,
        "radius": radius,
        "n_h0_finite": int(len(h0_finite)),
        "n_h1_finite": int(len(h1_finite)),
        "n_h1_essential": int(len(h1_essential)),
        "n_h1_capped_for_scale_space": int(n_capped_h1),
        "elapsed_sec": elapsed,
        **diagram_features(h1_diagram),
    }

    loaded_channels.append(ch_name)

    log(
        f"{prefix} done {elapsed:.2f}s | "
        f"edges={n_edges}, triangles={n_triangles}, "
        f"H1 finite={len(h1_finite)}, H1 scale-space diagram={len(h1_diagram)}"
    )

if len(loaded_channels) == 0:
    raise RuntimeError("No usable embeddings found. Check embedding_data paths.")

if global_max_death is None or not np.isfinite(global_max_death) or global_max_death <= 0:
    global_max_death = 1.0

if global_max_lifetime is None or not np.isfinite(global_max_lifetime) or global_max_lifetime <= 0:
    global_max_lifetime = 1.0

diagram_scale = max(global_max_death, global_max_lifetime, 1e-6)

kernel_bandwidth, kernel_time, kernel_bandwidth_method = choose_adaptive_kernel_bandwidth(
    diagram_store,
    fallback_scale=diagram_scale,
)

log("\nComputing persistence scale-space kernel matrix")
log(f"global_max_death={global_max_death:.6f}")
log(f"global_max_lifetime={global_max_lifetime:.6f}")
log(f"diagram_scale={diagram_scale:.6f}")
log(f"kernel_bandwidth_length={kernel_bandwidth:.8f}")
log(f"kernel_time={kernel_time:.10f}")
log(f"kernel_bandwidth_method={kernel_bandwidth_method}")

n_ch = len(loaded_channels)
scale_space_kernel_mat = np.zeros((n_ch, n_ch), dtype=float)

pair_rows = []
pair_t0 = time.time()

for i in range(n_ch):
    ch_i = loaded_channels[i]

    for j in range(i, n_ch):
        ch_j = loaded_channels[j]

        sim = persistence_scale_space_kernel(
            diagram_store[ch_i],
            diagram_store[ch_j],
            kernel_time=kernel_time,
        )

        scale_space_kernel_mat[i, j] = sim
        scale_space_kernel_mat[j, i] = sim

        if i != j:
            pair_rows.append({
                "channel_a": ch_i,
                "channel_b": ch_j,
                "raw_scale_space_kernel": sim,
            })

    log(f"  completed scale-space kernel row {i + 1}/{n_ch}: {ch_i}")

pair_elapsed = time.time() - pair_t0

normalized_kernel_mat = normalize_kernel_matrix(scale_space_kernel_mat)
kernel_distance_mat = kernel_distance_from_normalized_similarity(normalized_kernel_mat)

raw_kernel_df = pd.DataFrame(
    scale_space_kernel_mat,
    index=loaded_channels,
    columns=loaded_channels,
)

normalized_kernel_df = pd.DataFrame(
    normalized_kernel_mat,
    index=loaded_channels,
    columns=loaded_channels,
)

kernel_distance_df = pd.DataFrame(
    kernel_distance_mat,
    index=loaded_channels,
    columns=loaded_channels,
)

pairwise_df = pd.DataFrame(pair_rows)

if len(pairwise_df) > 0:
    pairwise_df["normalized_scale_space_similarity"] = [
        normalized_kernel_mat[loaded_channels.index(a), loaded_channels.index(b)]
        for a, b in zip(pairwise_df["channel_a"], pairwise_df["channel_b"])
    ]
    pairwise_df["scale_space_kernel_distance"] = [
        kernel_distance_mat[loaded_channels.index(a), loaded_channels.index(b)]
        for a, b in zip(pairwise_df["channel_a"], pairwise_df["channel_b"])
    ]
    pairwise_df = pairwise_df.sort_values("normalized_scale_space_similarity", ascending=False).reset_index(drop=True)

summary_rows = []

for idx, ch_name in enumerate(loaded_channels):
    sim = normalized_kernel_mat[idx].copy()
    dist = kernel_distance_mat[idx].copy()

    mask = np.arange(len(sim)) != idx
    sim_nonself = sim[mask]
    dist_nonself = dist[mask]

    summary_rows.append({
        **diagnostics[ch_name],
        "self_raw_scale_space_kernel": float(scale_space_kernel_mat[idx, idx]),
        "mean_scale_space_similarity_to_others": float(np.nanmean(sim_nonself)) if len(sim_nonself) else 1.0,
        "median_scale_space_similarity_to_others": float(np.nanmedian(sim_nonself)) if len(sim_nonself) else 1.0,
        "max_scale_space_similarity_to_others": float(np.nanmax(sim_nonself)) if len(sim_nonself) else 1.0,
        "min_scale_space_similarity_to_others": float(np.nanmin(sim_nonself)) if len(sim_nonself) else 1.0,
        "std_scale_space_similarity_to_others": float(np.nanstd(sim_nonself)) if len(sim_nonself) else 0.0,
        "mean_scale_space_distance_to_others": float(np.nanmean(dist_nonself)) if len(dist_nonself) else 0.0,
        "median_scale_space_distance_to_others": float(np.nanmedian(dist_nonself)) if len(dist_nonself) else 0.0,
        "max_scale_space_distance_to_others": float(np.nanmax(dist_nonself)) if len(dist_nonself) else 0.0,
        "min_scale_space_distance_to_others": float(np.nanmin(dist_nonself)) if len(dist_nonself) else 0.0,
        "std_scale_space_distance_to_others": float(np.nanstd(dist_nonself)) if len(dist_nonself) else 0.0,
    })

summary_df = (
    pd.DataFrame(summary_rows)
    .sort_values("mean_scale_space_similarity_to_others", ascending=False)
    .reset_index(drop=True)
)

offdiag_sim = offdiag_values(normalized_kernel_mat)
offdiag_dist = offdiag_values(kernel_distance_mat)
offdiag_raw = offdiag_values(scale_space_kernel_mat)

similarity_range = float(np.nanmax(offdiag_sim) - np.nanmin(offdiag_sim)) if len(offdiag_sim) else 0.0
distance_range = float(np.nanmax(offdiag_dist) - np.nanmin(offdiag_dist)) if len(offdiag_dist) else 0.0

log("\nScale-space kernel variability QC")
if len(offdiag_sim):
    log(
        f"normalized off-diagonal similarity: "
        f"min={np.nanmin(offdiag_sim):.6f}, "
        f"max={np.nanmax(offdiag_sim):.6f}, "
        f"mean={np.nanmean(offdiag_sim):.6f}, "
        f"std={np.nanstd(offdiag_sim):.6f}, "
        f"range={similarity_range:.6f}"
    )

if len(offdiag_dist):
    log(
        f"scale-space kernel distance: "
        f"min={np.nanmin(offdiag_dist):.6f}, "
        f"max={np.nanmax(offdiag_dist):.6f}, "
        f"mean={np.nanmean(offdiag_dist):.6f}, "
        f"std={np.nanstd(offdiag_dist):.6f}, "
        f"range={distance_range:.6f}"
    )

if len(offdiag_sim) and np.nanstd(offdiag_sim) < 1e-4:
    log(
        "WARNING: normalized similarities are still nearly constant. "
        "Try lowering kernel_bandwidth_frac, increasing max_points, or increasing radius_quantile."
    )

summary_csv_path = os.path.join(mode_out_dir, "h1_persistence_scale_space_kernel_channel_summary.csv")
raw_kernel_csv_path = os.path.join(mode_out_dir, "h1_persistence_scale_space_raw_kernel_matrix.csv")
normalized_kernel_csv_path = os.path.join(mode_out_dir, "h1_persistence_scale_space_normalized_similarity_matrix.csv")
kernel_distance_csv_path = os.path.join(mode_out_dir, "h1_persistence_scale_space_kernel_distance_matrix.csv")
pairwise_csv_path = os.path.join(mode_out_dir, "h1_persistence_scale_space_pairwise_similarities.csv")

summary_df.to_csv(summary_csv_path, index=False)
raw_kernel_df.to_csv(raw_kernel_csv_path)
normalized_kernel_df.to_csv(normalized_kernel_csv_path)
kernel_distance_df.to_csv(kernel_distance_csv_path)
pairwise_df.to_csv(pairwise_csv_path, index=False)

npz_path = os.path.join(mode_out_dir, "h1_persistence_scale_space_kernel_and_diagrams.npz")

save_dict = {
    "channels": np.asarray(loaded_channels),
    "raw_scale_space_kernel_matrix": scale_space_kernel_mat,
    "normalized_scale_space_similarity_matrix": normalized_kernel_mat,
    "scale_space_kernel_distance_matrix": kernel_distance_mat,
    "kernel_bandwidth_length": np.asarray([kernel_bandwidth]),
    "kernel_time": np.asarray([kernel_time]),
    "kernel_bandwidth_frac": np.asarray([KERNEL_BANDWIDTH_FRAC]),
    "kernel_bandwidth_method": np.asarray([kernel_bandwidth_method]),
    "fast_option": np.asarray([FAST_OPTION]),
    "backend": np.asarray(["ripser_presampled_fps + adaptive persistence scale-space kernel"]),
}

for ch_name in loaded_channels:
    save_dict[f"{ch_name}_H1_scale_space_diagram"] = diagram_store[ch_name]

    for key, val in raw_interval_store[ch_name].items():
        save_dict[f"{ch_name}_{key}"] = val

np.savez(npz_path, **save_dict)

print("\nH1 Persistence Scale-Space Kernel Channel Summary:")
if HAVE_IPYTHON:
    display(summary_df)
else:
    print(summary_df)

print("\nMost similar H1 diagram pairs by normalized persistence scale-space kernel:")
if HAVE_IPYTHON:
    display(pairwise_df.head(12))
else:
    print(pairwise_df.head(12))

print("\nLeast similar H1 diagram pairs by normalized persistence scale-space kernel:")
least_similar_df = pairwise_df.tail(12).sort_values("normalized_scale_space_similarity", ascending=True)
if HAVE_IPYTHON:
    display(least_similar_df)
else:
    print(least_similar_df)

log(f"\nSaved summary CSV: {summary_csv_path}")
log(f"Saved raw scale-space kernel matrix CSV: {raw_kernel_csv_path}")
log(f"Saved normalized scale-space similarity matrix CSV: {normalized_kernel_csv_path}")
log(f"Saved scale-space kernel distance matrix CSV: {kernel_distance_csv_path}")
log(f"Saved pairwise similarities CSV: {pairwise_csv_path}")
log(f"Saved NPZ: {npz_path}")
log(f"Pairwise scale-space kernel elapsed: {pair_elapsed:.2f}s")

plot_df = summary_df.copy().sort_values("mean_scale_space_similarity_to_others", ascending=False).reset_index(drop=True)

fig, ax = plt.subplots(figsize=(13, 5.8), facecolor=BLACK)
style_ax(
    ax,
    title=f"Mean H1 Persistence Scale-Space Similarity to Other Channels | {FAST_OPTION.upper()}",
    xlabel="Channel",
    ylabel="Mean normalized scale-space similarity",
)

x_pos = np.arange(len(plot_df))
y = plot_df["mean_scale_space_similarity_to_others"].to_numpy(dtype=float)

ax.bar(x_pos, y, color=ACCENT, edgecolor=ACCENT, alpha=0.78)

if np.any(np.isfinite(y)):
    ax.axhline(
        np.nanmean(y),
        color=ACCENT_SOFT,
        linestyle="--",
        linewidth=1.4,
        alpha=0.8,
        label=f"Mean={np.nanmean(y):.5f}",
    )

bar_vmin, bar_vmax = dynamic_limits(y, lower_pct=0, upper_pct=100, pad_frac=0.25, lower_bound=0.0, upper_bound=1.0)
ax.set_ylim(bar_vmin, bar_vmax)

ax.set_xticks(x_pos)
ax.set_xticklabels(plot_df["channel"], rotation=90)
ax.legend(loc="best", framealpha=1.0)

mean_sim_path = os.path.join(mode_out_dir, "01_mean_h1_persistence_scale_space_similarity_by_channel_zoomed.png")
plt.tight_layout()
save_render_close(fig, mean_sim_path)

if n_ch >= 3 and np.any(kernel_distance_mat > 0):
    condensed = kernel_distance_mat[np.triu_indices(n_ch, k=1)]
    Z = linkage(condensed, method="average")
    order = leaves_list(Z)
else:
    order = np.arange(n_ch)

ordered_sim = normalized_kernel_mat[np.ix_(order, order)]
ordered_raw = scale_space_kernel_mat[np.ix_(order, order)]
ordered_dist = kernel_distance_mat[np.ix_(order, order)]
ordered_channels = [loaded_channels[i] for i in order]

sim_vmin, sim_vmax = dynamic_limits(
    offdiag_values(ordered_sim),
    lower_pct=2,
    upper_pct=98,
    pad_frac=0.10,
    lower_bound=-1.0,
    upper_bound=1.0,
)

save_matrix_plot(
    ordered_sim,
    ordered_channels,
    "Normalized H1 Persistence Scale-Space Similarity Matrix | Zoomed Color Scale",
    "Normalized scale-space similarity",
    os.path.join(mode_out_dir, "02_h1_persistence_scale_space_normalized_similarity_heatmap_zoomed.png"),
    vmin=sim_vmin,
    vmax=sim_vmax,
)

raw_vmin, raw_vmax = dynamic_limits(
    offdiag_values(ordered_raw),
    lower_pct=2,
    upper_pct=98,
    pad_frac=0.10,
)

save_matrix_plot(
    ordered_raw,
    ordered_channels,
    "Raw H1 Persistence Scale-Space Kernel Matrix | Zoomed Color Scale",
    "Raw scale-space kernel value",
    os.path.join(mode_out_dir, "03_h1_persistence_scale_space_raw_kernel_heatmap_zoomed.png"),
    vmin=raw_vmin,
    vmax=raw_vmax,
)

dist_vmin, dist_vmax = dynamic_limits(
    offdiag_values(ordered_dist),
    lower_pct=2,
    upper_pct=98,
    pad_frac=0.10,
    lower_bound=0.0,
)

save_matrix_plot(
    ordered_dist,
    ordered_channels,
    "H1 Persistence Scale-Space Kernel Distance Matrix | Zoomed Color Scale",
    "Scale-space kernel distance",
    os.path.join(mode_out_dir, "04_h1_persistence_scale_space_kernel_distance_heatmap_zoomed.png"),
    vmin=dist_vmin,
    vmax=dist_vmax,
)

mds_coords = classical_mds_from_distances(kernel_distance_mat, n_components=2)

mds_df = pd.DataFrame({
    "channel": loaded_channels,
    "mds_1": mds_coords[:, 0],
    "mds_2": mds_coords[:, 1],
    "mean_scale_space_similarity_to_others": [
        summary_df.loc[summary_df["channel"] == ch, "mean_scale_space_similarity_to_others"].values[0]
        for ch in loaded_channels
    ],
    "mean_scale_space_distance_to_others": [
        summary_df.loc[summary_df["channel"] == ch, "mean_scale_space_distance_to_others"].values[0]
        for ch in loaded_channels
    ],
    "n_intervals_diagram": [
        summary_df.loc[summary_df["channel"] == ch, "n_intervals_diagram"].values[0]
        for ch in loaded_channels
    ],
})

mds_csv_path = os.path.join(mode_out_dir, "h1_persistence_scale_space_kernel_mds_coordinates.csv")
mds_df.to_csv(mds_csv_path, index=False)

fig, ax = plt.subplots(figsize=(8, 7), facecolor=BLACK)
style_ax(
    ax,
    title="Classical MDS of H1 Persistence Scale-Space Kernel Distances",
    xlabel="MDS 1",
    ylabel="MDS 2",
)

mds_vmin, mds_vmax = dynamic_limits(
    mds_df["mean_scale_space_similarity_to_others"],
    lower_pct=2,
    upper_pct=98,
    pad_frac=0.08,
    lower_bound=0.0,
    upper_bound=1.0,
)

sc = ax.scatter(
    mds_df["mds_1"],
    mds_df["mds_2"],
    c=mds_df["mean_scale_space_similarity_to_others"],
    s=150,
    cmap=CYAN_SEQUENTIAL_CMAP,
    edgecolor=ACCENT,
    linewidth=1.1,
    vmin=mds_vmin,
    vmax=mds_vmax,
)

for _, row in mds_df.iterrows():
    ax.text(
        row["mds_1"],
        row["mds_2"],
        row["channel"],
        fontsize=8,
        color=ACCENT,
        alpha=0.9,
    )

cbar = plt.colorbar(sc, ax=ax, shrink=0.85)
style_colorbar(cbar, "Mean normalized similarity")

mds_path = os.path.join(mode_out_dir, "05_h1_persistence_scale_space_kernel_mds_embedding.png")
plt.tight_layout()
save_render_close(fig, mds_path)

diagram_channels = summary_df.head(min(12, len(summary_df)))["channel"].tolist()
n_cols = 4
n_rows = int(np.ceil(len(diagram_channels) / n_cols))

fig, axes = plt.subplots(
    n_rows,
    n_cols,
    figsize=(4 * n_cols, 3.2 * n_rows),
    facecolor=BLACK,
)

axes = np.asarray(axes).reshape(-1)

all_diagram_points = np.vstack([
    diagram_store[ch] for ch in loaded_channels if len(diagram_store[ch]) > 0
]) if any(len(diagram_store[ch]) > 0 for ch in loaded_channels) else np.empty((0, 2))

if len(all_diagram_points) > 0:
    diag_max = float(np.nanmax(all_diagram_points[:, 1]))
else:
    diag_max = 1.0

diag_max = max(diag_max, 1e-6)

for ax_idx, ch_name in enumerate(diagram_channels):
    ax = axes[ax_idx]
    D = diagram_store[ch_name]

    mean_sim = summary_df.loc[
        summary_df["channel"] == ch_name,
        "mean_scale_space_similarity_to_others",
    ].values[0]

    style_ax(
        ax,
        title=f"{ch_name}: mean SSK={mean_sim:.3f}, n={len(D)}",
        xlabel="Birth",
        ylabel="Death",
    )

    ax.plot([0, diag_max], [0, diag_max], color=ACCENT_DIM, linestyle="--", linewidth=1.0)

    if len(D) > 0:
        ax.scatter(
            D[:, 0],
            D[:, 1],
            s=50,
            facecolors=BLACK,
            edgecolors=ACCENT,
            linewidths=1.2,
        )

    ax.set_xlim(0, diag_max * 1.05)
    ax.set_ylim(0, diag_max * 1.05)

for j in range(len(diagram_channels), len(axes)):
    axes[j].axis("off")

diagrams_path = os.path.join(mode_out_dir, "06_top_channels_h1_persistence_diagrams_by_scale_space_similarity.png")
plt.tight_layout()
save_render_close(fig, diagrams_path, width=1100)

fig, ax = plt.subplots(figsize=(12, 6), facecolor=BLACK)
style_ax(
    ax,
    title="Most Similar Channel Pairs by H1 Persistence Scale-Space Kernel",
    xlabel="Channel pair",
    ylabel="Normalized scale-space similarity",
)

top_pairs = pairwise_df.head(min(15, len(pairwise_df))).copy()
pair_labels = top_pairs["channel_a"] + "-" + top_pairs["channel_b"]

x_pair = np.arange(len(top_pairs))

ax.bar(
    x_pair,
    top_pairs["normalized_scale_space_similarity"],
    color=ACCENT,
    edgecolor=ACCENT,
    alpha=0.78,
)

top_vmin, top_vmax = dynamic_limits(
    top_pairs["normalized_scale_space_similarity"],
    lower_pct=0,
    upper_pct=100,
    pad_frac=0.25,
    lower_bound=0.0,
    upper_bound=1.0,
)
ax.set_ylim(top_vmin, top_vmax)

ax.set_xticks(x_pair)
ax.set_xticklabels(pair_labels, rotation=60, ha="right")

top_pairs_path = os.path.join(mode_out_dir, "07_most_similar_channel_pairs_h1_persistence_scale_space_kernel_zoomed.png")
plt.tight_layout()
save_render_close(fig, top_pairs_path)

fig, ax = plt.subplots(figsize=(12, 6), facecolor=BLACK)
style_ax(
    ax,
    title="Least Similar Channel Pairs by H1 Persistence Scale-Space Kernel",
    xlabel="Channel pair",
    ylabel="Normalized scale-space similarity",
)

bottom_pairs = pairwise_df.tail(min(15, len(pairwise_df))).copy().sort_values(
    "normalized_scale_space_similarity",
    ascending=True,
)
bottom_pair_labels = bottom_pairs["channel_a"] + "-" + bottom_pairs["channel_b"]

x_pair = np.arange(len(bottom_pairs))

ax.bar(
    x_pair,
    bottom_pairs["normalized_scale_space_similarity"],
    color=ACCENT,
    edgecolor=ACCENT,
    alpha=0.78,
)

bottom_vmin, bottom_vmax = dynamic_limits(
    bottom_pairs["normalized_scale_space_similarity"],
    lower_pct=0,
    upper_pct=100,
    pad_frac=0.25,
    lower_bound=0.0,
    upper_bound=1.0,
)
ax.set_ylim(bottom_vmin, bottom_vmax)

ax.set_xticks(x_pair)
ax.set_xticklabels(bottom_pair_labels, rotation=60, ha="right")

bottom_pairs_path = os.path.join(mode_out_dir, "08_least_similar_channel_pairs_h1_persistence_scale_space_kernel_zoomed.png")
plt.tight_layout()
save_render_close(fig, bottom_pairs_path)

fig, ax = plt.subplots(figsize=(8, 5.8), facecolor=BLACK)
style_ax(
    ax,
    title="Off-Diagonal Normalized Scale-Space Similarity Distribution",
    xlabel="Normalized scale-space similarity",
    ylabel="Count",
)

if len(offdiag_sim):
    ax.hist(
        offdiag_sim,
        bins=24,
        color=ACCENT,
        edgecolor=ACCENT,
        alpha=0.75,
    )

hist_path = os.path.join(mode_out_dir, "09_offdiagonal_scale_space_similarity_distribution.png")
plt.tight_layout()
save_render_close(fig, hist_path)

if MAKE_EXPERT_PLOTS:
    expert_dir = os.path.join(mode_out_dir, "expert_plots")
    os.makedirs(expert_dir, exist_ok=True)

    mean_similarity_map = dict(zip(summary_df["channel"], summary_df["mean_scale_space_similarity_to_others"]))
    mean_distance_map = dict(zip(summary_df["channel"], summary_df["mean_scale_space_distance_to_others"]))

    scalp_xy = {
        "Fp1": (-0.45, 1.00), "Fpz": (0.00, 1.08), "Fp2": (0.45, 1.00),
        "F7": (-0.95, 0.60), "F3": (-0.45, 0.55), "Fz": (0.00, 0.60),
        "F4": (0.45, 0.55), "F8": (0.95, 0.60),
        "FC5": (-0.75, 0.25), "FC1": (-0.25, 0.25),
        "FC2": (0.25, 0.25), "FC6": (0.75, 0.25),
        "M1": (-1.15, 0.05), "T7": (-1.00, 0.00), "C3": (-0.50, 0.00),
        "Cz": (0.00, 0.00), "C4": (0.50, 0.00), "T8": (1.00, 0.00),
        "M2": (1.15, 0.05),
        "CP5": (-0.75, -0.30), "CP1": (-0.25, -0.30),
        "CP2": (0.25, -0.30), "CP6": (0.75, -0.30),
        "P7": (-0.90, -0.60), "P3": (-0.45, -0.60), "Pz": (0.00, -0.65),
        "P4": (0.45, -0.60), "P8": (0.90, -0.60),
        "POz": (0.00, -0.88),
        "O1": (-0.35, -1.00), "Oz": (0.00, -1.08), "O2": (0.35, -1.00),
    }

    xs, ys, vals, labels = [], [], [], []

    for ch in eeg_channels:
        if ch in scalp_xy and ch in mean_similarity_map and np.isfinite(mean_similarity_map[ch]):
            xs.append(scalp_xy[ch][0])
            ys.append(scalp_xy[ch][1])
            vals.append(mean_similarity_map[ch])
            labels.append(ch)

    scalp_vmin, scalp_vmax = dynamic_limits(
        vals,
        lower_pct=2,
        upper_pct=98,
        pad_frac=0.10,
        lower_bound=0.0,
        upper_bound=1.0,
    )

    fig, ax = plt.subplots(figsize=(7, 7), facecolor=BLACK)
    style_ax(
        ax,
        title=f"Scalp Map of Mean H1 Persistence Scale-Space Similarity | {FAST_OPTION.upper()}",
        grid=False,
    )

    head = plt.Circle((0, 0), 1.12, fill=False, color=ACCENT, linewidth=2)
    ax.add_patch(head)
    ax.plot([0.00, -0.10, 0.10, 0.00], [1.12, 1.28, 1.28, 1.12], color=ACCENT, linewidth=2)

    sc = ax.scatter(
        xs,
        ys,
        c=vals,
        s=260,
        cmap=CYAN_SEQUENTIAL_CMAP,
        edgecolor=ACCENT,
        linewidth=1.2,
        vmin=scalp_vmin,
        vmax=scalp_vmax,
    )

    for x0, y0, label in zip(xs, ys, labels):
        ax.text(
            x0,
            y0,
            label,
            ha="center",
            va="center",
            fontsize=8,
            color=BLACK,
            fontweight="bold",
        )

    cbar = plt.colorbar(sc, ax=ax, shrink=0.75)
    style_colorbar(cbar, "Mean normalized similarity")

    ax.set_xlim(-1.35, 1.35)
    ax.set_ylim(-1.35, 1.35)
    ax.set_aspect("equal")
    ax.set_xticks([])
    ax.set_yticks([])

    scalp_path = os.path.join(expert_dir, "10_mean_h1_persistence_scale_space_similarity_scalp_map_zoomed.png")
    plt.tight_layout()
    save_render_close(fig, scalp_path, width=760)

    regions = {
        "Frontal": ["Fp1", "Fpz", "Fp2", "F7", "F3", "Fz", "F4", "F8"],
        "Frontocentral": ["FC5", "FC1", "FC2", "FC6"],
        "Central": ["T7", "C3", "Cz", "C4", "T8"],
        "Centroparietal": ["CP5", "CP1", "CP2", "CP6"],
        "Parietal": ["P7", "P3", "Pz", "P4", "P8"],
        "Occipital": ["POz", "O1", "Oz", "O2"],
        "Mastoid": ["M1", "M2"],
    }

    region_rows = []
    region_points = []

    for region_name, chans in regions.items():
        vals_region = []
        dist_vals_region = []

        for ch in chans:
            if ch in mean_similarity_map and np.isfinite(mean_similarity_map[ch]):
                vals_region.append(mean_similarity_map[ch])
                dist_vals_region.append(mean_distance_map[ch])
                region_points.append({
                    "region": region_name,
                    "channel": ch,
                    "mean_scale_space_similarity_to_others": mean_similarity_map[ch],
                    "mean_scale_space_distance_to_others": mean_distance_map[ch],
                })

        if len(vals_region) == 0:
            continue

        vals_region = np.asarray(vals_region, dtype=float)
        dist_vals_region = np.asarray(dist_vals_region, dtype=float)

        region_rows.append({
            "region": region_name,
            "mean_region_scale_space_similarity": float(np.mean(vals_region)),
            "std_region_scale_space_similarity": float(np.std(vals_region, ddof=1)) if len(vals_region) > 1 else 0.0,
            "sem_region_scale_space_similarity": (
                float(np.std(vals_region, ddof=1) / np.sqrt(len(vals_region)))
                if len(vals_region) > 1 else 0.0
            ),
            "mean_region_scale_space_distance": float(np.mean(dist_vals_region)),
            "n_channels": int(len(vals_region)),
        })

    region_df = pd.DataFrame(region_rows)
    region_points_df = pd.DataFrame(region_points)

    region_csv_path = os.path.join(expert_dir, "regional_h1_persistence_scale_space_kernel_summary.csv")
    region_df.to_csv(region_csv_path, index=False)

    if len(region_df) > 0:
        fig, ax = plt.subplots(figsize=(12, 6.5), facecolor=BLACK)
        style_ax(
            ax,
            title=f"Regional Mean H1 Persistence Scale-Space Similarity +/- SEM | {FAST_OPTION.upper()}",
            xlabel="Region",
            ylabel="Mean normalized scale-space similarity",
        )

        x_region = np.arange(len(region_df))

        ax.bar(
            x_region,
            region_df["mean_region_scale_space_similarity"],
            yerr=region_df["sem_region_scale_space_similarity"],
            color=ACCENT,
            edgecolor=ACCENT,
            alpha=0.82,
            width=0.68,
            capsize=8,
            error_kw={
                "elinewidth": 2.2,
                "capthick": 2.2,
                "ecolor": ACCENT,
                "alpha": 1.0,
            },
            label="Mean +/- SEM",
        )

        rng = np.random.default_rng(7)

        for i, region_name in enumerate(region_df["region"]):
            vals_region_plot = region_points_df.loc[
                region_points_df["region"] == region_name,
                "mean_scale_space_similarity_to_others",
            ].to_numpy(dtype=float)

            if len(vals_region_plot) == 0:
                continue

            jitter = rng.uniform(-0.13, 0.13, size=len(vals_region_plot))

            ax.scatter(
                np.full(len(vals_region_plot), i) + jitter,
                vals_region_plot,
                s=42,
                facecolors=BLACK,
                edgecolors=ACCENT,
                linewidths=1.2,
                alpha=0.95,
                zorder=4,
            )

        for i, row in region_df.iterrows():
            y_val = row["mean_region_scale_space_similarity"]
            sem = row["sem_region_scale_space_similarity"]

            ax.text(
                i,
                y_val + sem,
                f"{y_val:.3f}\n+/-{sem:.3f}\nn={int(row['n_channels'])}",
                ha="center",
                va="bottom",
                fontsize=8,
                color=ACCENT,
            )

        region_y = region_df["mean_region_scale_space_similarity"].to_numpy(dtype=float)
        reg_vmin, reg_vmax = dynamic_limits(
            region_y,
            lower_pct=0,
            upper_pct=100,
            pad_frac=0.28,
            lower_bound=0.0,
            upper_bound=1.0,
        )
        ax.set_ylim(reg_vmin, reg_vmax)

        ax.set_xticks(x_region)
        ax.set_xticklabels(
            region_df["region"],
            rotation=35,
            ha="right",
            rotation_mode="anchor",
        )
        ax.margins(x=0.04)
        ax.legend(loc="best", framealpha=1.0)

        region_path = os.path.join(expert_dir, "11_regional_h1_persistence_scale_space_similarity_summary_zoomed.png")
        plt.tight_layout(pad=2.0)
        save_render_close(fig, region_path)

    fig, ax = plt.subplots(figsize=(8, 6), facecolor=BLACK)
    style_ax(
        ax,
        title="Diagram Size vs Mean H1 Persistence Scale-Space Similarity",
        xlabel="H1 diagram intervals",
        ylabel="Mean normalized scale-space similarity",
    )

    sc = ax.scatter(
        summary_df["n_intervals_diagram"],
        summary_df["mean_scale_space_similarity_to_others"],
        c=summary_df["max_lifetime"],
        s=140,
        cmap=CYAN_SEQUENTIAL_CMAP,
        edgecolor=ACCENT,
        linewidth=1.0,
    )

    for _, row in summary_df.iterrows():
        ax.text(
            row["n_intervals_diagram"],
            row["mean_scale_space_similarity_to_others"],
            row["channel"],
            fontsize=8,
            color=ACCENT,
            alpha=0.85,
        )

    cbar = plt.colorbar(sc, ax=ax, shrink=0.85)
    style_colorbar(cbar, "Max H1 lifetime")

    scatter_vmin, scatter_vmax = dynamic_limits(
        summary_df["mean_scale_space_similarity_to_others"],
        lower_pct=2,
        upper_pct=98,
        pad_frac=0.15,
        lower_bound=0.0,
        upper_bound=1.0,
    )
    ax.set_ylim(scatter_vmin, scatter_vmax)

    diagram_size_path = os.path.join(expert_dir, "12_diagram_size_vs_mean_scale_space_similarity_zoomed.png")
    plt.tight_layout()
    save_render_close(fig, diagram_size_path)

    fig, ax = plt.subplots(figsize=(8, 6), facecolor=BLACK)
    style_ax(
        ax,
        title="Threshold Triangles vs Mean H1 Persistence Scale-Space Similarity",
        xlabel="Triangles at radius",
        ylabel="Mean normalized scale-space similarity",
    )

    sc = ax.scatter(
        summary_df["n_triangles_at_radius"],
        summary_df["mean_scale_space_similarity_to_others"],
        c=summary_df["n_intervals_diagram"],
        s=140,
        cmap=CYAN_SEQUENTIAL_CMAP,
        edgecolor=ACCENT,
        linewidth=1.0,
    )

    for _, row in summary_df.iterrows():
        ax.text(
            row["n_triangles_at_radius"],
            row["mean_scale_space_similarity_to_others"],
            row["channel"],
            fontsize=8,
            color=ACCENT,
            alpha=0.85,
        )

    cbar = plt.colorbar(sc, ax=ax, shrink=0.85)
    style_colorbar(cbar, "H1 diagram intervals")

    ax.set_ylim(scatter_vmin, scatter_vmax)

    triangle_path = os.path.join(expert_dir, "13_threshold_triangles_vs_mean_scale_space_similarity_zoomed.png")
    plt.tight_layout()
    save_render_close(fig, triangle_path)

    fig, ax = plt.subplots(figsize=(8, 6), facecolor=BLACK)
    style_ax(
        ax,
        title="Scale-Space Similarity Spread per Channel",
        xlabel="Mean normalized scale-space similarity",
        ylabel="Similarity standard deviation",
    )

    sc = ax.scatter(
        summary_df["mean_scale_space_similarity_to_others"],
        summary_df["std_scale_space_similarity_to_others"],
        c=summary_df["n_intervals_diagram"],
        s=140,
        cmap=CYAN_SEQUENTIAL_CMAP,
        edgecolor=ACCENT,
        linewidth=1.0,
    )

    for _, row in summary_df.iterrows():
        ax.text(
            row["mean_scale_space_similarity_to_others"],
            row["std_scale_space_similarity_to_others"],
            row["channel"],
            fontsize=8,
            color=ACCENT,
            alpha=0.85,
        )

    cbar = plt.colorbar(sc, ax=ax, shrink=0.85)
    style_colorbar(cbar, "H1 diagram intervals")

    spread_path = os.path.join(expert_dir, "14_scale_space_similarity_spread_per_channel.png")
    plt.tight_layout()
    save_render_close(fig, spread_path)

    qc_df = summary_df.copy().sort_values("elapsed_sec", ascending=False).reset_index(drop=True)

    fig, ax = plt.subplots(figsize=(12, 5.8), facecolor=BLACK)
    style_ax(
        ax,
        title="Persistence Scale-Space Kernel Runtime QC",
        xlabel="Channel",
        ylabel="Elapsed seconds",
    )

    x_qc = np.arange(len(qc_df))

    ax.bar(
        x_qc,
        qc_df["elapsed_sec"],
        color=ACCENT,
        edgecolor=ACCENT,
        alpha=0.82,
    )

    ax.set_xticks(x_qc)
    ax.set_xticklabels(qc_df["channel"], rotation=90)

    qc_path = os.path.join(expert_dir, "15_runtime_qc.png")
    plt.tight_layout()
    save_render_close(fig, qc_path)

    log(f"\nExpert plots saved and rendered from: {expert_dir}")
    log(f"Regional summary CSV: {region_csv_path}")

elapsed_total = time.time() - overall_t0

log("\n" + "=" * 72)
log(f"Finished H1 Persistence Scale-Space Kernel in {elapsed_total:.2f}s")
log(f"Mode: {FAST_OPTION}")
log(f"Backend: ripser_presampled_fps + adaptive persistence scale-space kernel")
log(f"Kernel bandwidth length: {kernel_bandwidth:.8f}")
log(f"Kernel time: {kernel_time:.10f}")
log(f"Kernel bandwidth method: {kernel_bandwidth_method}")
log(f"Channels completed: {len(loaded_channels)}")
log(f"Summary CSV: {summary_csv_path}")
log(f"Raw scale-space kernel CSV: {raw_kernel_csv_path}")
log(f"Normalized similarity CSV: {normalized_kernel_csv_path}")
log(f"Scale-space kernel distance CSV: {kernel_distance_csv_path}")
log(f"Pairwise CSV: {pairwise_csv_path}")
log(f"MDS CSV: {mds_csv_path}")
log(f"NPZ: {npz_path}")
log(f"Output directory: {mode_out_dir}")
log("=" * 72)
```

```python
import os
import sys
import time
import numpy as np

if not hasattr(np, "trapz"):
    np.trapz = np.trapezoid

import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.colors import LinearSegmentedColormap

os.environ["OMP_NUM_THREADS"] = "2"
os.environ["OPENBLAS_NUM_THREADS"] = "2"
os.environ["MKL_NUM_THREADS"] = "2"
os.environ["VECLIB_MAXIMUM_THREADS"] = "2"
os.environ["NUMEXPR_NUM_THREADS"] = "2"

from scipy.spatial.distance import pdist, squareform
from scipy.cluster.hierarchy import linkage, leaves_list

try:
    from scipy.linalg import eigh
    HAVE_SCIPY_EIGH = True
except Exception:
    HAVE_SCIPY_EIGH = False

try:
    from IPython.display import display, Image
    HAVE_IPYTHON = True
except Exception:
    HAVE_IPYTHON = False

try:
    from ripser import ripser
except Exception as exc:
    raise ImportError(
        "ripser is not installed. Run this in a separate notebook cell first:\n\n"
        "!pip install ripser persim\n\n"
        "Then rerun this Persistence Weighted Gaussian Kernel cell."
    ) from exc


base_dir = "/home/a/projects/Complete-Neural-Signal-Analysis"
out_dir = os.path.join(base_dir, "topology_persistence_weighted_gaussian_kernel")
os.makedirs(out_dir, exist_ok=True)

FAST_OPTION = "ultra"  # options: "ultra", "fast", "balanced", "full"

MAX_CHANNELS_TO_PROCESS = None
RENDER_ALL_IMAGES_INLINE = True
SAVE_PLOTS_TO_DISK = True
MAKE_EXPERT_PLOTS = True
INLINE_IMAGE_WIDTH = 950
VERBOSE_PROGRESS = True

USE_GEOMETRY_PRESERVING_SUBSAMPLE = True
AUTO_KERNEL_BANDWIDTH = True
AUTO_WEIGHT_SCALE = True

MODE_CONFIGS = {
    "ultra": {
        "max_points": 90,
        "reservoir_points": 5000,
        "max_dims": 3,
        "radius_quantile": 0.22,
        "gaussian_bandwidth_frac": 0.035,
        "gaussian_auto_quantile": 0.25,
        "gaussian_auto_multiplier": 0.60,
        "weight_power": 2.0,
        "weight_scale_frac": 0.20,
        "weight_auto_quantile": 0.50,
        "cap_essential_h1": True,
        "plot_dpi": 120,
    },
    "fast": {
        "max_points": 130,
        "reservoir_points": 8000,
        "max_dims": 3,
        "radius_quantile": 0.24,
        "gaussian_bandwidth_frac": 0.030,
        "gaussian_auto_quantile": 0.25,
        "gaussian_auto_multiplier": 0.55,
        "weight_power": 2.0,
        "weight_scale_frac": 0.18,
        "weight_auto_quantile": 0.50,
        "cap_essential_h1": True,
        "plot_dpi": 130,
    },
    "balanced": {
        "max_points": 180,
        "reservoir_points": 10000,
        "max_dims": 3,
        "radius_quantile": 0.26,
        "gaussian_bandwidth_frac": 0.026,
        "gaussian_auto_quantile": 0.22,
        "gaussian_auto_multiplier": 0.50,
        "weight_power": 2.0,
        "weight_scale_frac": 0.16,
        "weight_auto_quantile": 0.45,
        "cap_essential_h1": True,
        "plot_dpi": 140,
    },
    "full": {
        "max_points": 260,
        "reservoir_points": 14000,
        "max_dims": 3,
        "radius_quantile": 0.28,
        "gaussian_bandwidth_frac": 0.022,
        "gaussian_auto_quantile": 0.20,
        "gaussian_auto_multiplier": 0.45,
        "weight_power": 2.0,
        "weight_scale_frac": 0.14,
        "weight_auto_quantile": 0.40,
        "cap_essential_h1": True,
        "plot_dpi": 150,
    },
}

if FAST_OPTION not in MODE_CONFIGS:
    raise ValueError("FAST_OPTION must be one of: ultra, fast, balanced, full")

cfg = MODE_CONFIGS[FAST_OPTION]

MAX_POINTS = cfg["max_points"]
RESERVOIR_POINTS = cfg["reservoir_points"]
MAX_DIMS = cfg["max_dims"]
RADIUS_QUANTILE = cfg["radius_quantile"]
GAUSSIAN_BANDWIDTH_FRAC = cfg["gaussian_bandwidth_frac"]
GAUSSIAN_AUTO_QUANTILE = cfg["gaussian_auto_quantile"]
GAUSSIAN_AUTO_MULTIPLIER = cfg["gaussian_auto_multiplier"]
WEIGHT_POWER = cfg["weight_power"]
WEIGHT_SCALE_FRAC = cfg["weight_scale_frac"]
WEIGHT_AUTO_QUANTILE = cfg["weight_auto_quantile"]
CAP_ESSENTIAL_H1 = cfg["cap_essential_h1"]
PLOT_DPI = cfg["plot_dpi"]

mode_out_dir = os.path.join(
    out_dir,
    f"h1_persistence_weighted_gaussian_kernel_{FAST_OPTION}_fps_adaptive",
)
os.makedirs(mode_out_dir, exist_ok=True)

eeg_channels = [
    "Fp1", "Fpz", "Fp2", "F7", "F3", "Fz", "F4", "F8", "FC5", "FC1", "FC2", "FC6",
    "M1", "T7", "C3", "Cz", "C4", "T8", "M2", "CP5", "CP1", "CP2", "CP6",
    "P7", "P3", "Pz", "P4", "P8", "POz", "O1", "Oz", "O2"
]

if MAX_CHANNELS_TO_PROCESS is not None:
    eeg_channels = eeg_channels[:MAX_CHANNELS_TO_PROCESS]

ACCENT = "#00FFFF"
ACCENT_SOFT = "#66FFFF"
ACCENT_DIM = "#008B8B"
BLACK = "#000000"

CYAN_SEQUENTIAL_CMAP = LinearSegmentedColormap.from_list(
    "cyan_sequential",
    [BLACK, ACCENT_DIM, ACCENT, ACCENT_SOFT],
    N=256,
)

plt.rcParams.update({
    "figure.facecolor": BLACK,
    "axes.facecolor": BLACK,
    "savefig.facecolor": BLACK,
    "text.color": ACCENT,
    "axes.labelcolor": ACCENT,
    "axes.edgecolor": ACCENT,
    "axes.titlecolor": ACCENT,
    "xtick.color": ACCENT,
    "ytick.color": ACCENT,
    "grid.color": ACCENT_DIM,
    "grid.alpha": 0.25,
    "legend.facecolor": BLACK,
    "legend.edgecolor": ACCENT,
    "legend.labelcolor": ACCENT,
    "font.size": 11,
})


def log(msg):
    if VERBOSE_PROGRESS:
        print(msg, flush=True)


def style_ax(ax, title=None, xlabel=None, ylabel=None, grid=True):
    ax.set_facecolor(BLACK)

    for spine in ax.spines.values():
        spine.set_color(ACCENT)
        spine.set_linewidth(1.1)

    ax.tick_params(axis="x", colors=ACCENT, labelcolor=ACCENT)
    ax.tick_params(axis="y", colors=ACCENT, labelcolor=ACCENT)

    if title is not None:
        ax.set_title(title, color=ACCENT, pad=12)

    if xlabel is not None:
        ax.set_xlabel(xlabel, color=ACCENT, labelpad=10)

    if ylabel is not None:
        ax.set_ylabel(ylabel, color=ACCENT, labelpad=10)

    if grid:
        ax.grid(True, color=ACCENT_DIM, alpha=0.25, linewidth=0.8)


def style_colorbar(cbar, label):
    cbar.set_label(label, color=ACCENT)
    cbar.ax.yaxis.set_tick_params(color=ACCENT)
    plt.setp(cbar.ax.get_yticklabels(), color=ACCENT)
    cbar.outline.set_edgecolor(ACCENT)


def save_render_close(fig, path, width=INLINE_IMAGE_WIDTH):
    if SAVE_PLOTS_TO_DISK:
        fig.savefig(path, dpi=PLOT_DPI, bbox_inches="tight", facecolor=BLACK)

    plt.close(fig)

    if RENDER_ALL_IMAGES_INLINE and HAVE_IPYTHON and os.path.exists(path):
        display(Image(filename=path, width=width))


def dynamic_limits(values, lower_pct=2, upper_pct=98, pad_frac=0.08, lower_bound=None, upper_bound=None):
    values = np.asarray(values, dtype=float)
    values = values[np.isfinite(values)]

    if len(values) == 0:
        vmin, vmax = 0.0, 1.0
    else:
        vmin = float(np.percentile(values, lower_pct))
        vmax = float(np.percentile(values, upper_pct))

        if not np.isfinite(vmin):
            vmin = float(np.nanmin(values))

        if not np.isfinite(vmax):
            vmax = float(np.nanmax(values))

        if vmax <= vmin:
            center = float(np.nanmean(values))
            eps = max(1e-6, abs(center) * 0.01, 1e-5)
            vmin = center - eps
            vmax = center + eps

    pad = (vmax - vmin) * pad_frac
    vmin -= pad
    vmax += pad

    if lower_bound is not None:
        vmin = max(lower_bound, vmin)

    if upper_bound is not None:
        vmax = min(upper_bound, vmax)

    if vmax <= vmin:
        vmax = vmin + 1e-6

    return vmin, vmax


def offdiag_values(M):
    M = np.asarray(M, dtype=float)

    if M.shape[0] <= 1:
        return np.asarray([], dtype=float)

    mask = ~np.eye(M.shape[0], dtype=bool)
    vals = M[mask]
    vals = vals[np.isfinite(vals)]

    return vals


def save_matrix_plot(matrix, labels, title, colorbar_label, path, vmin=None, vmax=None, width=INLINE_IMAGE_WIDTH):
    fig, ax = plt.subplots(figsize=(9, 8), facecolor=BLACK)
    style_ax(
        ax,
        title=title,
        xlabel="Channel",
        ylabel="Channel",
    )

    im = ax.imshow(
        matrix,
        aspect="equal",
        interpolation="nearest",
        cmap=CYAN_SEQUENTIAL_CMAP,
        vmin=vmin,
        vmax=vmax,
    )

    ax.set_xticks(np.arange(len(labels)))
    ax.set_xticklabels(labels, rotation=90)
    ax.set_yticks(np.arange(len(labels)))
    ax.set_yticklabels(labels)

    cbar = plt.colorbar(im, ax=ax, shrink=0.85)
    style_colorbar(cbar, colorbar_label)

    plt.tight_layout()
    save_render_close(fig, path, width=width)


def estimate_median_pairwise_distance(X, max_pairs=120000, seed=7):
    X = np.asarray(X, dtype=float)
    n = X.shape[0]

    if n < 2:
        return 1.0

    rng = np.random.default_rng(seed)

    if n * (n - 1) // 2 <= max_pairs:
        D = pdist(X)
        D = D[D > 0]

        if len(D) == 0:
            return 1.0

        return float(np.nanmedian(D))

    i = rng.integers(0, n, size=max_pairs)
    j = rng.integers(0, n, size=max_pairs)

    mask = i != j
    i = i[mask]
    j = j[mask]

    if len(i) == 0:
        return 1.0

    d = np.linalg.norm(X[i] - X[j], axis=1)
    d = d[np.isfinite(d) & (d > 0)]

    if len(d) == 0:
        return 1.0

    return float(np.nanmedian(d))


def standardize_matrix_for_geometry(X, max_dims=3, seed=7):
    X = np.asarray(X, dtype=np.float64)
    X = X[np.all(np.isfinite(X), axis=1)]

    if X.shape[0] < 10:
        return None

    X = X - np.mean(X, axis=0, keepdims=True)

    if X.shape[1] > max_dims:
        _U, _S, Vt = np.linalg.svd(X, full_matrices=False)
        X = X @ Vt[:max_dims].T

    X = X / (np.std(X, axis=0, keepdims=True) + 1e-12)

    scale = estimate_median_pairwise_distance(X, seed=seed)
    X = X / (scale + 1e-12)

    return X


def farthest_point_sample(X, n_samples, seed=7):
    X = np.asarray(X, dtype=float)
    n = X.shape[0]

    if n <= n_samples:
        return np.arange(n)

    rng = np.random.default_rng(seed)

    center = np.mean(X, axis=0, keepdims=True)
    dist_to_center = np.linalg.norm(X - center, axis=1)

    start_idx = int(np.nanargmax(dist_to_center))

    selected = np.empty(n_samples, dtype=int)
    selected.fill(-1)
    selected[0] = start_idx

    min_dist2 = np.full(n, np.inf, dtype=float)

    for k in range(1, n_samples):
        last = X[selected[k - 1]]
        d2 = np.sum((X - last) ** 2, axis=1)
        min_dist2 = np.minimum(min_dist2, d2)
        min_dist2[selected[:k]] = -np.inf

        next_idx = int(np.nanargmax(min_dist2))

        if not np.isfinite(min_dist2[next_idx]) or min_dist2[next_idx] < 1e-18:
            remaining = np.setdiff1d(np.arange(n), selected[:k], assume_unique=False)

            if len(remaining) == 0:
                selected = selected[:k]
                break

            next_idx = int(rng.choice(remaining))

        selected[k] = next_idx

    selected = selected[selected >= 0]

    return selected


def load_channel_embedding_geometry_sampled(ch_name, max_points=90, reservoir_points=5000, max_dims=3, seed=7):
    candidate_paths = [
        os.path.join(base_dir, "embedding_data", "tsne_umap_from_3d", f"umap_embedding_{ch_name}.npy"),
        os.path.join(base_dir, "embedding_data", "tsne_umap_from_3d", f"tsne_embedding_{ch_name}.npy"),
        os.path.join(base_dir, "embedding_data", "3dembedding_data", f"3dembedded_{ch_name}.npy"),
        os.path.join(base_dir, "embedding_data", "2dembedding_data", f"2dembedded_{ch_name}.npy"),
    ]

    rng = np.random.default_rng(seed)

    for path in candidate_paths:
        if not os.path.exists(path):
            continue

        log(f"    trying: {path}")

        try:
            X = np.load(path, mmap_mode="r", allow_pickle=False)
        except Exception as exc:
            log(f"    skipped non-memmappable or nonnumeric file: {os.path.basename(path)} | {exc}")
            continue

        if X.ndim != 2:
            log(f"    skipped because ndim={X.ndim}")
            continue

        original_shape = tuple(X.shape)

        try:
            if X.shape[0] >= X.shape[1]:
                n_samples = X.shape[0]
                n_features = X.shape[1]

                if n_samples < 10 or n_features < 2:
                    log(f"    skipped bad shape: {original_shape}")
                    continue

                reservoir_n = min(max(reservoir_points, max_points), n_samples)

                if n_samples > reservoir_n:
                    reservoir_idx = np.sort(rng.choice(n_samples, size=reservoir_n, replace=False))
                else:
                    reservoir_idx = np.arange(n_samples)

                X_reservoir = np.asarray(X[reservoir_idx, :], dtype=np.float64)

            else:
                n_samples = X.shape[1]
                n_features = X.shape[0]

                if n_samples < 10 or n_features < 2:
                    log(f"    skipped bad transposed shape: {original_shape}")
                    continue

                reservoir_n = min(max(reservoir_points, max_points), n_samples)

                if n_samples > reservoir_n:
                    reservoir_idx = np.sort(rng.choice(n_samples, size=reservoir_n, replace=False))
                else:
                    reservoir_idx = np.arange(n_samples)

                X_reservoir = np.asarray(X[:, reservoir_idx].T, dtype=np.float64)

        except Exception as exc:
            log(f"    failed while reservoir-sampling {os.path.basename(path)} | {exc}")
            continue

        X_reservoir = X_reservoir[np.all(np.isfinite(X_reservoir), axis=1)]

        if X_reservoir.shape[0] < 10 or X_reservoir.shape[1] < 2:
            log(f"    reservoir data unusable: reservoir_shape={X_reservoir.shape}")
            continue

        X_reservoir_std = standardize_matrix_for_geometry(
            X_reservoir,
            max_dims=max_dims,
            seed=seed,
        )

        if X_reservoir_std is None:
            log("    standardized reservoir unusable")
            continue

        if USE_GEOMETRY_PRESERVING_SUBSAMPLE:
            fps_idx = farthest_point_sample(
                X_reservoir_std,
                n_samples=min(max_points, X_reservoir_std.shape[0]),
                seed=seed,
            )
            X_small = X_reservoir_std[fps_idx]
            method = "reservoir_plus_farthest_point"
        else:
            sample_n = min(max_points, X_reservoir_std.shape[0])
            idx = np.sort(rng.choice(X_reservoir_std.shape[0], size=sample_n, replace=False))
            X_small = X_reservoir_std[idx]
            method = "reservoir_random"

        X_small = X_small[np.all(np.isfinite(X_small), axis=1)]

        if X_small.shape[0] >= 10 and X_small.shape[1] >= 2:
            return X_small, path, original_shape, tuple(X_reservoir.shape), method

        log(f"    sampled data unusable: sampled_shape={X_small.shape}")

    return None, None, None, None, None


def standardize_point_cloud_small(X, max_dims=3):
    X = np.asarray(X, dtype=np.float64)
    X = X[np.all(np.isfinite(X), axis=1)]

    if X.shape[0] < 10:
        return None

    X = X - np.mean(X, axis=0, keepdims=True)

    if X.shape[1] > max_dims:
        _U, _S, Vt = np.linalg.svd(X, full_matrices=False)
        X = X @ Vt[:max_dims].T

    X = X / (np.std(X, axis=0, keepdims=True) + 1e-12)

    D = pdist(X)
    positive = D[D > 0]

    if len(positive) > 0:
        X = X / (np.nanmedian(positive) + 1e-12)

    return X


def estimate_complex_counts(X, radius):
    D = squareform(pdist(X))
    A = (D <= radius) & (D > 0)

    n_edges = int(np.sum(A) // 2)

    if A.shape[0] <= 300:
        A_int = A.astype(np.int64)
        n_triangles = int(np.trace(A_int @ A_int @ A_int) // 6)
    else:
        n_triangles = np.nan

    return n_edges, n_triangles


def run_h1_ripser(X, radius):
    result = ripser(
        X,
        maxdim=1,
        thresh=float(radius),
        coeff=2,
        do_cocycles=False,
    )

    diagrams = result["dgms"]

    h0 = diagrams[0] if len(diagrams) > 0 else np.empty((0, 2), dtype=float)
    h1 = diagrams[1] if len(diagrams) > 1 else np.empty((0, 2), dtype=float)

    return h0, h1


def split_finite_essential(diagram):
    diagram = np.asarray(diagram, dtype=float)

    if diagram.size == 0:
        return np.empty((0, 2), dtype=float), np.empty((0, 2), dtype=float)

    finite = diagram[
        np.isfinite(diagram[:, 0])
        & np.isfinite(diagram[:, 1])
        & (diagram[:, 1] > diagram[:, 0])
    ]

    essential = diagram[
        np.isfinite(diagram[:, 0])
        & np.isinf(diagram[:, 1])
    ]

    return finite, essential


def clean_diagram_for_weighted_gaussian(
    finite_intervals,
    essential_intervals=None,
    cap_essential=True,
    cap_value=None,
):
    cleaned = []
    capped_count = 0

    if finite_intervals is not None:
        for b, d in finite_intervals:
            if np.isfinite(b) and np.isfinite(d) and d > b:
                cleaned.append((float(b), float(d)))

    if cap_essential and essential_intervals is not None and cap_value is not None:
        for b, _d in essential_intervals:
            if np.isfinite(b) and np.isfinite(cap_value) and cap_value > b:
                cleaned.append((float(b), float(cap_value)))
                capped_count += 1

    if len(cleaned) == 0:
        return np.empty((0, 2), dtype=float), capped_count

    return np.asarray(cleaned, dtype=float), capped_count


def diagram_features(diagram):
    diagram = np.asarray(diagram, dtype=float)

    if diagram.size == 0 or len(diagram) == 0:
        return {
            "n_intervals_diagram": 0,
            "mean_lifetime": np.nan,
            "max_lifetime": np.nan,
            "total_lifetime": 0.0,
            "mean_birth": np.nan,
            "mean_death": np.nan,
            "max_death": np.nan,
        }

    lifetimes = diagram[:, 1] - diagram[:, 0]

    return {
        "n_intervals_diagram": int(len(diagram)),
        "mean_lifetime": float(np.nanmean(lifetimes)),
        "max_lifetime": float(np.nanmax(lifetimes)),
        "total_lifetime": float(np.nansum(lifetimes)),
        "mean_birth": float(np.nanmean(diagram[:, 0])),
        "mean_death": float(np.nanmean(diagram[:, 1])),
        "max_death": float(np.nanmax(diagram[:, 1])),
    }


def choose_adaptive_gaussian_bandwidth(diagram_store, fallback_scale):
    all_points = []

    for D in diagram_store.values():
        D = np.asarray(D, dtype=float)

        if D.size == 0 or len(D) == 0:
            continue

        D = D[
            np.isfinite(D[:, 0])
            & np.isfinite(D[:, 1])
            & (D[:, 1] > D[:, 0])
        ]

        if len(D) > 0:
            all_points.append(D)

    fallback_bandwidth = max(1e-6, GAUSSIAN_BANDWIDTH_FRAC * fallback_scale)

    if len(all_points) == 0:
        return fallback_bandwidth, "fallback_no_points"

    all_points = np.vstack(all_points)

    if len(all_points) < 2:
        return fallback_bandwidth, "fallback_too_few_points"

    D = pdist(all_points)
    D = D[np.isfinite(D) & (D > 0)]

    if len(D) == 0:
        return fallback_bandwidth, "fallback_zero_distances"

    auto_bandwidth = float(np.quantile(D, GAUSSIAN_AUTO_QUANTILE) * GAUSSIAN_AUTO_MULTIPLIER)

    if not np.isfinite(auto_bandwidth) or auto_bandwidth <= 0:
        return fallback_bandwidth, "fallback_bad_auto"

    if AUTO_KERNEL_BANDWIDTH:
        chosen = min(fallback_bandwidth, auto_bandwidth)
        chosen = max(chosen, fallback_scale * 0.0025, 1e-6)
        method = "adaptive_min_of_fractional_and_diagram_quantile"
    else:
        chosen = fallback_bandwidth
        method = "fixed_fractional"

    return chosen, method


def choose_adaptive_weight_scale(diagram_store, fallback_scale):
    lifetimes = []

    for D in diagram_store.values():
        D = np.asarray(D, dtype=float)

        if D.size == 0 or len(D) == 0:
            continue

        D = D[
            np.isfinite(D[:, 0])
            & np.isfinite(D[:, 1])
            & (D[:, 1] > D[:, 0])
        ]

        if len(D) > 0:
            lifetimes.extend((D[:, 1] - D[:, 0]).tolist())

    fallback_weight_scale = max(1e-6, WEIGHT_SCALE_FRAC * fallback_scale)

    lifetimes = np.asarray(lifetimes, dtype=float)
    lifetimes = lifetimes[np.isfinite(lifetimes) & (lifetimes > 0)]

    if len(lifetimes) == 0:
        return fallback_weight_scale, "fallback_no_lifetimes"

    auto_scale = float(np.quantile(lifetimes, WEIGHT_AUTO_QUANTILE))

    if not np.isfinite(auto_scale) or auto_scale <= 0:
        return fallback_weight_scale, "fallback_bad_auto"

    if AUTO_WEIGHT_SCALE:
        chosen = max(auto_scale, fallback_scale * 0.0025, 1e-6)
        method = "adaptive_lifetime_quantile"
    else:
        chosen = fallback_weight_scale
        method = "fixed_fractional"

    return chosen, method


def persistence_weights(diagram, weight_scale, weight_power=2.0):
    D = np.asarray(diagram, dtype=float)

    if D.size == 0 or len(D) == 0:
        return np.asarray([], dtype=float)

    lifetimes = D[:, 1] - D[:, 0]
    lifetimes = np.maximum(lifetimes, 0.0)

    weight_scale = max(float(weight_scale), 1e-12)
    weight_power = float(weight_power)

    numerator = lifetimes ** weight_power
    denominator = numerator + weight_scale ** weight_power + 1e-12

    weights = numerator / denominator

    weights = np.asarray(weights, dtype=float)
    weights[~np.isfinite(weights)] = 0.0

    return weights


def persistence_weighted_gaussian_kernel(D1, D2, gaussian_bandwidth, weight_scale, weight_power=2.0):
    D1 = np.asarray(D1, dtype=float)
    D2 = np.asarray(D2, dtype=float)

    if D1.size == 0:
        D1 = np.empty((0, 2), dtype=float)

    if D2.size == 0:
        D2 = np.empty((0, 2), dtype=float)

    if len(D1) == 0 or len(D2) == 0:
        return 0.0

    D1 = D1[
        np.isfinite(D1[:, 0])
        & np.isfinite(D1[:, 1])
        & (D1[:, 1] > D1[:, 0])
    ]

    D2 = D2[
        np.isfinite(D2[:, 0])
        & np.isfinite(D2[:, 1])
        & (D2[:, 1] > D2[:, 0])
    ]

    if len(D1) == 0 or len(D2) == 0:
        return 0.0

    w1 = persistence_weights(D1, weight_scale=weight_scale, weight_power=weight_power)
    w2 = persistence_weights(D2, weight_scale=weight_scale, weight_power=weight_power)

    if len(w1) == 0 or len(w2) == 0:
        return 0.0

    gaussian_bandwidth = max(float(gaussian_bandwidth), 1e-12)
    denom = 2.0 * gaussian_bandwidth ** 2

    diff = D1[:, None, :] - D2[None, :, :]
    dist2 = np.sum(diff ** 2, axis=2)

    K = np.sum((w1[:, None] * w2[None, :]) * np.exp(-dist2 / denom))

    if not np.isfinite(K):
        return 0.0

    return float(max(0.0, K))


def normalize_kernel_matrix(K):
    K = np.asarray(K, dtype=float)
    K = 0.5 * (K + K.T)

    diag = np.diag(K).copy()
    K_norm = np.zeros_like(K, dtype=float)

    for i in range(K.shape[0]):
        for j in range(K.shape[1]):
            if diag[i] > 0 and diag[j] > 0:
                K_norm[i, j] = K[i, j] / np.sqrt(diag[i] * diag[j])
            elif i == j:
                K_norm[i, j] = 1.0
            else:
                K_norm[i, j] = 0.0

    K_norm = 0.5 * (K_norm + K_norm.T)
    K_norm = np.clip(K_norm, 0.0, 1.0)
    np.fill_diagonal(K_norm, 1.0)

    return K_norm


def kernel_distance_from_normalized_similarity(K_norm):
    K_norm = np.asarray(K_norm, dtype=float)
    D2 = np.maximum(0.0, 2.0 - 2.0 * K_norm)
    D = np.sqrt(D2)
    np.fill_diagonal(D, 0.0)
    return D


def classical_mds_from_distances(D, n_components=2):
    D = np.asarray(D, dtype=float)
    n = D.shape[0]

    if n < 2:
        return np.zeros((n, n_components), dtype=float)

    D2 = D ** 2
    J = np.eye(n) - np.ones((n, n)) / n
    B = -0.5 * J @ D2 @ J

    if HAVE_SCIPY_EIGH:
        vals, vecs = eigh(B)
    else:
        vals, vecs = np.linalg.eigh(B)

    idx = np.argsort(vals)[::-1]
    vals = vals[idx]
    vecs = vecs[:, idx]

    coords = np.zeros((n, n_components), dtype=float)

    for k in range(min(n_components, n)):
        if vals[k] > 0:
            coords[:, k] = vecs[:, k] * np.sqrt(vals[k])

    return coords


log("\n" + "=" * 72)
log(f"Running H1 Persistence Weighted Gaussian Kernel | {FAST_OPTION.upper()}")
log(f"Channels: {len(eeg_channels)}")
log(f"max_points={MAX_POINTS}, reservoir_points={RESERVOIR_POINTS}, radius_quantile={RADIUS_QUANTILE}")
log(f"gaussian_bandwidth_frac={GAUSSIAN_BANDWIDTH_FRAC}, auto_kernel_bandwidth={AUTO_KERNEL_BANDWIDTH}")
log(f"weight_power={WEIGHT_POWER}, auto_weight_scale={AUTO_WEIGHT_SCALE}")
log(f"geometry_preserving_subsample={USE_GEOMETRY_PRESERVING_SUBSAMPLE}")
log(f"cap_essential_h1={CAP_ESSENTIAL_H1}")
log(f"expert_plots={MAKE_EXPERT_PLOTS}, render_all_images={RENDER_ALL_IMAGES_INLINE}")
log(f"Output: {mode_out_dir}")
log("=" * 72)

diagnostics = {}
diagram_store = {}
weight_store = {}
raw_interval_store = {}
loaded_channels = []

global_max_death = None
global_max_lifetime = None

overall_t0 = time.time()

for ch_idx, ch_name in enumerate(eeg_channels, start=1):
    t0 = time.time()
    prefix = f"[{ch_idx:02d}/{len(eeg_channels)} {ch_name}]"

    log(f"\n{prefix} loading reservoir + geometry-preserving subsampling")

    X_small, source_path, original_shape, reservoir_shape, sampling_method = load_channel_embedding_geometry_sampled(
        ch_name,
        max_points=MAX_POINTS,
        reservoir_points=RESERVOIR_POINTS,
        max_dims=MAX_DIMS,
        seed=900 + ch_idx,
    )

    if X_small is None:
        log(f"{prefix} WARNING: no usable embedding found; skipping")
        continue

    log(
        f"{prefix} source shape={original_shape}, reservoir shape={reservoir_shape}, "
        f"sampled shape={X_small.shape}, method={sampling_method}"
    )

    X = standardize_point_cloud_small(X_small, max_dims=MAX_DIMS)

    if X is None:
        log(f"{prefix} WARNING: invalid sampled embedding; skipping")
        continue

    log(f"{prefix} standardized shape={X.shape}; computing radius")

    D_flat = pdist(X)
    positive = D_flat[D_flat > 0]

    if len(positive) == 0:
        log(f"{prefix} WARNING: zero-distance cloud; skipping")
        continue

    radius = float(np.quantile(positive, RADIUS_QUANTILE))

    log(f"{prefix} estimating threshold complex size")
    n_edges, n_triangles = estimate_complex_counts(X, radius)

    log(f"{prefix} ripser H0/H1")
    h0_diag, h1_diag = run_h1_ripser(X, radius)

    h0_finite, h0_essential = split_finite_essential(h0_diag)
    h1_finite, h1_essential = split_finite_essential(h1_diag)

    h1_diagram, n_capped_h1 = clean_diagram_for_weighted_gaussian(
        h1_finite,
        essential_intervals=h1_essential,
        cap_essential=CAP_ESSENTIAL_H1,
        cap_value=radius,
    )

    if len(h1_diagram) > 0:
        max_death = float(np.nanmax(h1_diagram[:, 1]))
        max_lifetime = float(np.nanmax(h1_diagram[:, 1] - h1_diagram[:, 0]))

        global_max_death = max_death if global_max_death is None else max(global_max_death, max_death)
        global_max_lifetime = max_lifetime if global_max_lifetime is None else max(global_max_lifetime, max_lifetime)

    elapsed = time.time() - t0

    diagram_store[ch_name] = h1_diagram

    raw_interval_store[ch_name] = {
        "H0_finite": h0_finite,
        "H0_essential": h0_essential,
        "H1_finite": h1_finite,
        "H1_essential": h1_essential,
        "H1_weighted_gaussian_diagram": h1_diagram,
    }

    diagnostics[ch_name] = {
        "channel": ch_name,
        "source_path": source_path,
        "source_shape": str(original_shape),
        "reservoir_shape": str(reservoir_shape),
        "sampling_method": sampling_method,
        "backend": "ripser_presampled_fps",
        "n_points": int(X.shape[0]),
        "n_edges_at_radius": int(n_edges),
        "n_triangles_at_radius": int(n_triangles) if np.isfinite(n_triangles) else np.nan,
        "radius": radius,
        "n_h0_finite": int(len(h0_finite)),
        "n_h1_finite": int(len(h1_finite)),
        "n_h1_essential": int(len(h1_essential)),
        "n_h1_capped_for_weighted_gaussian": int(n_capped_h1),
        "elapsed_sec": elapsed,
        **diagram_features(h1_diagram),
    }

    loaded_channels.append(ch_name)

    log(
        f"{prefix} done {elapsed:.2f}s | "
        f"edges={n_edges}, triangles={n_triangles}, "
        f"H1 finite={len(h1_finite)}, H1 weighted-Gaussian diagram={len(h1_diagram)}"
    )

if len(loaded_channels) == 0:
    raise RuntimeError("No usable embeddings found. Check embedding_data paths.")

if global_max_death is None or not np.isfinite(global_max_death) or global_max_death <= 0:
    global_max_death = 1.0

if global_max_lifetime is None or not np.isfinite(global_max_lifetime) or global_max_lifetime <= 0:
    global_max_lifetime = 1.0

diagram_scale = max(global_max_death, global_max_lifetime, 1e-6)

gaussian_bandwidth, gaussian_bandwidth_method = choose_adaptive_gaussian_bandwidth(
    diagram_store,
    fallback_scale=diagram_scale,
)

weight_scale, weight_scale_method = choose_adaptive_weight_scale(
    diagram_store,
    fallback_scale=diagram_scale,
)

for ch_name in loaded_channels:
    weight_store[ch_name] = persistence_weights(
        diagram_store[ch_name],
        weight_scale=weight_scale,
        weight_power=WEIGHT_POWER,
    )

    weights = weight_store[ch_name]

    if len(weights) > 0:
        diagnostics[ch_name]["mean_persistence_weight"] = float(np.nanmean(weights))
        diagnostics[ch_name]["max_persistence_weight"] = float(np.nanmax(weights))
        diagnostics[ch_name]["total_persistence_weight"] = float(np.nansum(weights))
    else:
        diagnostics[ch_name]["mean_persistence_weight"] = np.nan
        diagnostics[ch_name]["max_persistence_weight"] = np.nan
        diagnostics[ch_name]["total_persistence_weight"] = 0.0

log("\nComputing persistence weighted Gaussian kernel matrix")
log(f"global_max_death={global_max_death:.6f}")
log(f"global_max_lifetime={global_max_lifetime:.6f}")
log(f"diagram_scale={diagram_scale:.6f}")
log(f"gaussian_bandwidth={gaussian_bandwidth:.8f}")
log(f"gaussian_bandwidth_method={gaussian_bandwidth_method}")
log(f"weight_scale={weight_scale:.8f}")
log(f"weight_scale_method={weight_scale_method}")
log(f"weight_power={WEIGHT_POWER:.3f}")

n_ch = len(loaded_channels)
weighted_gaussian_kernel_mat = np.zeros((n_ch, n_ch), dtype=float)

pair_rows = []
pair_t0 = time.time()

for i in range(n_ch):
    ch_i = loaded_channels[i]

    for j in range(i, n_ch):
        ch_j = loaded_channels[j]

        sim = persistence_weighted_gaussian_kernel(
            diagram_store[ch_i],
            diagram_store[ch_j],
            gaussian_bandwidth=gaussian_bandwidth,
            weight_scale=weight_scale,
            weight_power=WEIGHT_POWER,
        )

        weighted_gaussian_kernel_mat[i, j] = sim
        weighted_gaussian_kernel_mat[j, i] = sim

        if i != j:
            pair_rows.append({
                "channel_a": ch_i,
                "channel_b": ch_j,
                "raw_weighted_gaussian_kernel": sim,
            })

    log(f"  completed weighted Gaussian kernel row {i + 1}/{n_ch}: {ch_i}")

pair_elapsed = time.time() - pair_t0

normalized_kernel_mat = normalize_kernel_matrix(weighted_gaussian_kernel_mat)
kernel_distance_mat = kernel_distance_from_normalized_similarity(normalized_kernel_mat)

raw_kernel_df = pd.DataFrame(
    weighted_gaussian_kernel_mat,
    index=loaded_channels,
    columns=loaded_channels,
)

normalized_kernel_df = pd.DataFrame(
    normalized_kernel_mat,
    index=loaded_channels,
    columns=loaded_channels,
)

kernel_distance_df = pd.DataFrame(
    kernel_distance_mat,
    index=loaded_channels,
    columns=loaded_channels,
)

pairwise_df = pd.DataFrame(pair_rows)

if len(pairwise_df) > 0:
    pairwise_df["normalized_weighted_gaussian_similarity"] = [
        normalized_kernel_mat[loaded_channels.index(a), loaded_channels.index(b)]
        for a, b in zip(pairwise_df["channel_a"], pairwise_df["channel_b"])
    ]
    pairwise_df["weighted_gaussian_kernel_distance"] = [
        kernel_distance_mat[loaded_channels.index(a), loaded_channels.index(b)]
        for a, b in zip(pairwise_df["channel_a"], pairwise_df["channel_b"])
    ]
    pairwise_df = pairwise_df.sort_values("normalized_weighted_gaussian_similarity", ascending=False).reset_index(drop=True)

summary_rows = []

for idx, ch_name in enumerate(loaded_channels):
    sim = normalized_kernel_mat[idx].copy()
    dist = kernel_distance_mat[idx].copy()

    mask = np.arange(len(sim)) != idx
    sim_nonself = sim[mask]
    dist_nonself = dist[mask]

    summary_rows.append({
        **diagnostics[ch_name],
        "self_raw_weighted_gaussian_kernel": float(weighted_gaussian_kernel_mat[idx, idx]),
        "mean_weighted_gaussian_similarity_to_others": float(np.nanmean(sim_nonself)) if len(sim_nonself) else 1.0,
        "median_weighted_gaussian_similarity_to_others": float(np.nanmedian(sim_nonself)) if len(sim_nonself) else 1.0,
        "max_weighted_gaussian_similarity_to_others": float(np.nanmax(sim_nonself)) if len(sim_nonself) else 1.0,
        "min_weighted_gaussian_similarity_to_others": float(np.nanmin(sim_nonself)) if len(sim_nonself) else 1.0,
        "std_weighted_gaussian_similarity_to_others": float(np.nanstd(sim_nonself)) if len(sim_nonself) else 0.0,
        "mean_weighted_gaussian_distance_to_others": float(np.nanmean(dist_nonself)) if len(dist_nonself) else 0.0,
        "median_weighted_gaussian_distance_to_others": float(np.nanmedian(dist_nonself)) if len(dist_nonself) else 0.0,
        "max_weighted_gaussian_distance_to_others": float(np.nanmax(dist_nonself)) if len(dist_nonself) else 0.0,
        "min_weighted_gaussian_distance_to_others": float(np.nanmin(dist_nonself)) if len(dist_nonself) else 0.0,
        "std_weighted_gaussian_distance_to_others": float(np.nanstd(dist_nonself)) if len(dist_nonself) else 0.0,
    })

summary_df = (
    pd.DataFrame(summary_rows)
    .sort_values("mean_weighted_gaussian_similarity_to_others", ascending=False)
    .reset_index(drop=True)
)

offdiag_sim = offdiag_values(normalized_kernel_mat)
offdiag_dist = offdiag_values(kernel_distance_mat)
offdiag_raw = offdiag_values(weighted_gaussian_kernel_mat)

similarity_range = float(np.nanmax(offdiag_sim) - np.nanmin(offdiag_sim)) if len(offdiag_sim) else 0.0
distance_range = float(np.nanmax(offdiag_dist) - np.nanmin(offdiag_dist)) if len(offdiag_dist) else 0.0

log("\nWeighted Gaussian kernel variability QC")
if len(offdiag_sim):
    log(
        f"normalized off-diagonal similarity: "
        f"min={np.nanmin(offdiag_sim):.6f}, "
        f"max={np.nanmax(offdiag_sim):.6f}, "
        f"mean={np.nanmean(offdiag_sim):.6f}, "
        f"std={np.nanstd(offdiag_sim):.6f}, "
        f"range={similarity_range:.6f}"
    )

if len(offdiag_dist):
    log(
        f"weighted Gaussian kernel distance: "
        f"min={np.nanmin(offdiag_dist):.6f}, "
        f"max={np.nanmax(offdiag_dist):.6f}, "
        f"mean={np.nanmean(offdiag_dist):.6f}, "
        f"std={np.nanstd(offdiag_dist):.6f}, "
        f"range={distance_range:.6f}"
    )

if len(offdiag_sim) and np.nanstd(offdiag_sim) < 1e-4:
    log(
        "WARNING: normalized similarities are still nearly constant. "
        "Try lowering gaussian_bandwidth_frac, increasing max_points, or increasing radius_quantile."
    )

summary_csv_path = os.path.join(mode_out_dir, "h1_persistence_weighted_gaussian_kernel_channel_summary.csv")
raw_kernel_csv_path = os.path.join(mode_out_dir, "h1_persistence_weighted_gaussian_raw_kernel_matrix.csv")
normalized_kernel_csv_path = os.path.join(mode_out_dir, "h1_persistence_weighted_gaussian_normalized_similarity_matrix.csv")
kernel_distance_csv_path = os.path.join(mode_out_dir, "h1_persistence_weighted_gaussian_kernel_distance_matrix.csv")
pairwise_csv_path = os.path.join(mode_out_dir, "h1_persistence_weighted_gaussian_pairwise_similarities.csv")

summary_df.to_csv(summary_csv_path, index=False)
raw_kernel_df.to_csv(raw_kernel_csv_path)
normalized_kernel_df.to_csv(normalized_kernel_csv_path)
kernel_distance_df.to_csv(kernel_distance_csv_path)
pairwise_df.to_csv(pairwise_csv_path, index=False)

npz_path = os.path.join(mode_out_dir, "h1_persistence_weighted_gaussian_kernel_and_diagrams.npz")

save_dict = {
    "channels": np.asarray(loaded_channels),
    "raw_weighted_gaussian_kernel_matrix": weighted_gaussian_kernel_mat,
    "normalized_weighted_gaussian_similarity_matrix": normalized_kernel_mat,
    "weighted_gaussian_kernel_distance_matrix": kernel_distance_mat,
    "gaussian_bandwidth": np.asarray([gaussian_bandwidth]),
    "gaussian_bandwidth_method": np.asarray([gaussian_bandwidth_method]),
    "weight_scale": np.asarray([weight_scale]),
    "weight_scale_method": np.asarray([weight_scale_method]),
    "weight_power": np.asarray([WEIGHT_POWER]),
    "fast_option": np.asarray([FAST_OPTION]),
    "backend": np.asarray(["ripser_presampled_fps + adaptive persistence weighted gaussian kernel"]),
}

for ch_name in loaded_channels:
    save_dict[f"{ch_name}_H1_weighted_gaussian_diagram"] = diagram_store[ch_name]
    save_dict[f"{ch_name}_H1_weighted_gaussian_weights"] = weight_store[ch_name]

    for key, val in raw_interval_store[ch_name].items():
        save_dict[f"{ch_name}_{key}"] = val

np.savez(npz_path, **save_dict)

print("\nH1 Persistence Weighted Gaussian Kernel Channel Summary:")
if HAVE_IPYTHON:
    display(summary_df)
else:
    print(summary_df)

print("\nMost similar H1 diagram pairs by normalized persistence weighted Gaussian kernel:")
if HAVE_IPYTHON:
    display(pairwise_df.head(12))
else:
    print(pairwise_df.head(12))

print("\nLeast similar H1 diagram pairs by normalized persistence weighted Gaussian kernel:")
least_similar_df = pairwise_df.tail(12).sort_values("normalized_weighted_gaussian_similarity", ascending=True)
if HAVE_IPYTHON:
    display(least_similar_df)
else:
    print(least_similar_df)

log(f"\nSaved summary CSV: {summary_csv_path}")
log(f"Saved raw weighted Gaussian kernel matrix CSV: {raw_kernel_csv_path}")
log(f"Saved normalized weighted Gaussian similarity matrix CSV: {normalized_kernel_csv_path}")
log(f"Saved weighted Gaussian kernel distance matrix CSV: {kernel_distance_csv_path}")
log(f"Saved pairwise similarities CSV: {pairwise_csv_path}")
log(f"Saved NPZ: {npz_path}")
log(f"Pairwise weighted Gaussian kernel elapsed: {pair_elapsed:.2f}s")

plot_df = summary_df.copy().sort_values("mean_weighted_gaussian_similarity_to_others", ascending=False).reset_index(drop=True)

fig, ax = plt.subplots(figsize=(13, 5.8), facecolor=BLACK)
style_ax(
    ax,
    title=f"Mean H1 Persistence Weighted Gaussian Similarity to Other Channels | {FAST_OPTION.upper()}",
    xlabel="Channel",
    ylabel="Mean normalized weighted Gaussian similarity",
)

x_pos = np.arange(len(plot_df))
y = plot_df["mean_weighted_gaussian_similarity_to_others"].to_numpy(dtype=float)

ax.bar(x_pos, y, color=ACCENT, edgecolor=ACCENT, alpha=0.78)

if np.any(np.isfinite(y)):
    ax.axhline(
        np.nanmean(y),
        color=ACCENT_SOFT,
        linestyle="--",
        linewidth=1.4,
        alpha=0.8,
        label=f"Mean={np.nanmean(y):.5f}",
    )

bar_vmin, bar_vmax = dynamic_limits(y, lower_pct=0, upper_pct=100, pad_frac=0.25, lower_bound=0.0, upper_bound=1.0)
ax.set_ylim(bar_vmin, bar_vmax)

ax.set_xticks(x_pos)
ax.set_xticklabels(plot_df["channel"], rotation=90)
ax.legend(loc="best", framealpha=1.0)

mean_sim_path = os.path.join(mode_out_dir, "01_mean_h1_persistence_weighted_gaussian_similarity_by_channel_zoomed.png")
plt.tight_layout()
save_render_close(fig, mean_sim_path)

if n_ch >= 3 and np.any(kernel_distance_mat > 0):
    condensed = kernel_distance_mat[np.triu_indices(n_ch, k=1)]
    Z = linkage(condensed, method="average")
    order = leaves_list(Z)
else:
    order = np.arange(n_ch)

ordered_sim = normalized_kernel_mat[np.ix_(order, order)]
ordered_raw = weighted_gaussian_kernel_mat[np.ix_(order, order)]
ordered_dist = kernel_distance_mat[np.ix_(order, order)]
ordered_channels = [loaded_channels[i] for i in order]

sim_vmin, sim_vmax = dynamic_limits(
    offdiag_values(ordered_sim),
    lower_pct=2,
    upper_pct=98,
    pad_frac=0.10,
    lower_bound=0.0,
    upper_bound=1.0,
)

save_matrix_plot(
    ordered_sim,
    ordered_channels,
    "Normalized H1 Persistence Weighted Gaussian Similarity Matrix | Zoomed Color Scale",
    "Normalized weighted Gaussian similarity",
    os.path.join(mode_out_dir, "02_h1_persistence_weighted_gaussian_normalized_similarity_heatmap_zoomed.png"),
    vmin=sim_vmin,
    vmax=sim_vmax,
)

raw_vmin, raw_vmax = dynamic_limits(
    offdiag_values(ordered_raw),
    lower_pct=2,
    upper_pct=98,
    pad_frac=0.10,
)

save_matrix_plot(
    ordered_raw,
    ordered_channels,
    "Raw H1 Persistence Weighted Gaussian Kernel Matrix | Zoomed Color Scale",
    "Raw weighted Gaussian kernel value",
    os.path.join(mode_out_dir, "03_h1_persistence_weighted_gaussian_raw_kernel_heatmap_zoomed.png"),
    vmin=raw_vmin,
    vmax=raw_vmax,
)

dist_vmin, dist_vmax = dynamic_limits(
    offdiag_values(ordered_dist),
    lower_pct=2,
    upper_pct=98,
    pad_frac=0.10,
    lower_bound=0.0,
)

save_matrix_plot(
    ordered_dist,
    ordered_channels,
    "H1 Persistence Weighted Gaussian Kernel Distance Matrix | Zoomed Color Scale",
    "Weighted Gaussian kernel distance",
    os.path.join(mode_out_dir, "04_h1_persistence_weighted_gaussian_kernel_distance_heatmap_zoomed.png"),
    vmin=dist_vmin,
    vmax=dist_vmax,
)

mds_coords = classical_mds_from_distances(kernel_distance_mat, n_components=2)

mds_df = pd.DataFrame({
    "channel": loaded_channels,
    "mds_1": mds_coords[:, 0],
    "mds_2": mds_coords[:, 1],
    "mean_weighted_gaussian_similarity_to_others": [
        summary_df.loc[summary_df["channel"] == ch, "mean_weighted_gaussian_similarity_to_others"].values[0]
        for ch in loaded_channels
    ],
    "mean_weighted_gaussian_distance_to_others": [
        summary_df.loc[summary_df["channel"] == ch, "mean_weighted_gaussian_distance_to_others"].values[0]
        for ch in loaded_channels
    ],
    "n_intervals_diagram": [
        summary_df.loc[summary_df["channel"] == ch, "n_intervals_diagram"].values[0]
        for ch in loaded_channels
    ],
})

mds_csv_path = os.path.join(mode_out_dir, "h1_persistence_weighted_gaussian_kernel_mds_coordinates.csv")
mds_df.to_csv(mds_csv_path, index=False)

fig, ax = plt.subplots(figsize=(8, 7), facecolor=BLACK)
style_ax(
    ax,
    title="Classical MDS of H1 Persistence Weighted Gaussian Kernel Distances",
    xlabel="MDS 1",
    ylabel="MDS 2",
)

mds_vmin, mds_vmax = dynamic_limits(
    mds_df["mean_weighted_gaussian_similarity_to_others"],
    lower_pct=2,
    upper_pct=98,
    pad_frac=0.08,
    lower_bound=0.0,
    upper_bound=1.0,
)

sc = ax.scatter(
    mds_df["mds_1"],
    mds_df["mds_2"],
    c=mds_df["mean_weighted_gaussian_similarity_to_others"],
    s=150,
    cmap=CYAN_SEQUENTIAL_CMAP,
    edgecolor=ACCENT,
    linewidth=1.1,
    vmin=mds_vmin,
    vmax=mds_vmax,
)

for _, row in mds_df.iterrows():
    ax.text(
        row["mds_1"],
        row["mds_2"],
        row["channel"],
        fontsize=8,
        color=ACCENT,
        alpha=0.9,
    )

cbar = plt.colorbar(sc, ax=ax, shrink=0.85)
style_colorbar(cbar, "Mean normalized similarity")

mds_path = os.path.join(mode_out_dir, "05_h1_persistence_weighted_gaussian_kernel_mds_embedding.png")
plt.tight_layout()
save_render_close(fig, mds_path)

diagram_channels = summary_df.head(min(12, len(summary_df)))["channel"].tolist()
n_cols = 4
n_rows = int(np.ceil(len(diagram_channels) / n_cols))

fig, axes = plt.subplots(
    n_rows,
    n_cols,
    figsize=(4 * n_cols, 3.2 * n_rows),
    facecolor=BLACK,
)

axes = np.asarray(axes).reshape(-1)

all_diagram_points = np.vstack([
    diagram_store[ch] for ch in loaded_channels if len(diagram_store[ch]) > 0
]) if any(len(diagram_store[ch]) > 0 for ch in loaded_channels) else np.empty((0, 2))

if len(all_diagram_points) > 0:
    diag_max = float(np.nanmax(all_diagram_points[:, 1]))
else:
    diag_max = 1.0

diag_max = max(diag_max, 1e-6)

for ax_idx, ch_name in enumerate(diagram_channels):
    ax = axes[ax_idx]
    D = diagram_store[ch_name]
    W = weight_store[ch_name]

    mean_sim = summary_df.loc[
        summary_df["channel"] == ch_name,
        "mean_weighted_gaussian_similarity_to_others",
    ].values[0]

    style_ax(
        ax,
        title=f"{ch_name}: mean WGK={mean_sim:.3f}, n={len(D)}",
        xlabel="Birth",
        ylabel="Death",
    )

    ax.plot([0, diag_max], [0, diag_max], color=ACCENT_DIM, linestyle="--", linewidth=1.0)

    if len(D) > 0:
        sizes = 35 + 130 * W if len(W) == len(D) else 55

        ax.scatter(
            D[:, 0],
            D[:, 1],
            s=sizes,
            facecolors=BLACK,
            edgecolors=ACCENT,
            linewidths=1.2,
        )

    ax.set_xlim(0, diag_max * 1.05)
    ax.set_ylim(0, diag_max * 1.05)

for j in range(len(diagram_channels), len(axes)):
    axes[j].axis("off")

diagrams_path = os.path.join(mode_out_dir, "06_top_channels_h1_persistence_diagrams_weighted_by_persistence.png")
plt.tight_layout()
save_render_close(fig, diagrams_path, width=1100)

fig, ax = plt.subplots(figsize=(12, 6), facecolor=BLACK)
style_ax(
    ax,
    title="Most Similar Channel Pairs by H1 Persistence Weighted Gaussian Kernel",
    xlabel="Channel pair",
    ylabel="Normalized weighted Gaussian similarity",
)

top_pairs = pairwise_df.head(min(15, len(pairwise_df))).copy()
pair_labels = top_pairs["channel_a"] + "-" + top_pairs["channel_b"]

x_pair = np.arange(len(top_pairs))

ax.bar(
    x_pair,
    top_pairs["normalized_weighted_gaussian_similarity"],
    color=ACCENT,
    edgecolor=ACCENT,
    alpha=0.78,
)

top_vmin, top_vmax = dynamic_limits(
    top_pairs["normalized_weighted_gaussian_similarity"],
    lower_pct=0,
    upper_pct=100,
    pad_frac=0.25,
    lower_bound=0.0,
    upper_bound=1.0,
)
ax.set_ylim(top_vmin, top_vmax)

ax.set_xticks(x_pair)
ax.set_xticklabels(pair_labels, rotation=60, ha="right")

top_pairs_path = os.path.join(mode_out_dir, "07_most_similar_channel_pairs_h1_persistence_weighted_gaussian_kernel_zoomed.png")
plt.tight_layout()
save_render_close(fig, top_pairs_path)

fig, ax = plt.subplots(figsize=(12, 6), facecolor=BLACK)
style_ax(
    ax,
    title="Least Similar Channel Pairs by H1 Persistence Weighted Gaussian Kernel",
    xlabel="Channel pair",
    ylabel="Normalized weighted Gaussian similarity",
)

bottom_pairs = pairwise_df.tail(min(15, len(pairwise_df))).copy().sort_values(
    "normalized_weighted_gaussian_similarity",
    ascending=True,
)
bottom_pair_labels = bottom_pairs["channel_a"] + "-" + bottom_pairs["channel_b"]

x_pair = np.arange(len(bottom_pairs))

ax.bar(
    x_pair,
    bottom_pairs["normalized_weighted_gaussian_similarity"],
    color=ACCENT,
    edgecolor=ACCENT,
    alpha=0.78,
)

bottom_vmin, bottom_vmax = dynamic_limits(
    bottom_pairs["normalized_weighted_gaussian_similarity"],
    lower_pct=0,
    upper_pct=100,
    pad_frac=0.25,
    lower_bound=0.0,
    upper_bound=1.0,
)
ax.set_ylim(bottom_vmin, bottom_vmax)

ax.set_xticks(x_pair)
ax.set_xticklabels(bottom_pair_labels, rotation=60, ha="right")

bottom_pairs_path = os.path.join(mode_out_dir, "08_least_similar_channel_pairs_h1_persistence_weighted_gaussian_kernel_zoomed.png")
plt.tight_layout()
save_render_close(fig, bottom_pairs_path)

fig, ax = plt.subplots(figsize=(8, 5.8), facecolor=BLACK)
style_ax(
    ax,
    title="Off-Diagonal Normalized Weighted Gaussian Similarity Distribution",
    xlabel="Normalized weighted Gaussian similarity",
    ylabel="Count",
)

if len(offdiag_sim):
    ax.hist(
        offdiag_sim,
        bins=24,
        color=ACCENT,
        edgecolor=ACCENT,
        alpha=0.75,
    )

hist_path = os.path.join(mode_out_dir, "09_offdiagonal_weighted_gaussian_similarity_distribution.png")
plt.tight_layout()
save_render_close(fig, hist_path)

if MAKE_EXPERT_PLOTS:
    expert_dir = os.path.join(mode_out_dir, "expert_plots")
    os.makedirs(expert_dir, exist_ok=True)

    mean_similarity_map = dict(zip(summary_df["channel"], summary_df["mean_weighted_gaussian_similarity_to_others"]))
    mean_distance_map = dict(zip(summary_df["channel"], summary_df["mean_weighted_gaussian_distance_to_others"]))

    scalp_xy = {
        "Fp1": (-0.45, 1.00), "Fpz": (0.00, 1.08), "Fp2": (0.45, 1.00),
        "F7": (-0.95, 0.60), "F3": (-0.45, 0.55), "Fz": (0.00, 0.60),
        "F4": (0.45, 0.55), "F8": (0.95, 0.60),
        "FC5": (-0.75, 0.25), "FC1": (-0.25, 0.25),
        "FC2": (0.25, 0.25), "FC6": (0.75, 0.25),
        "M1": (-1.15, 0.05), "T7": (-1.00, 0.00), "C3": (-0.50, 0.00),
        "Cz": (0.00, 0.00), "C4": (0.50, 0.00), "T8": (1.00, 0.00),
        "M2": (1.15, 0.05),
        "CP5": (-0.75, -0.30), "CP1": (-0.25, -0.30),
        "CP2": (0.25, -0.30), "CP6": (0.75, -0.30),
        "P7": (-0.90, -0.60), "P3": (-0.45, -0.60), "Pz": (0.00, -0.65),
        "P4": (0.45, -0.60), "P8": (0.90, -0.60),
        "POz": (0.00, -0.88),
        "O1": (-0.35, -1.00), "Oz": (0.00, -1.08), "O2": (0.35, -1.00),
    }

    xs, ys, vals, labels = [], [], [], []

    for ch in eeg_channels:
        if ch in scalp_xy and ch in mean_similarity_map and np.isfinite(mean_similarity_map[ch]):
            xs.append(scalp_xy[ch][0])
            ys.append(scalp_xy[ch][1])
            vals.append(mean_similarity_map[ch])
            labels.append(ch)

    scalp_vmin, scalp_vmax = dynamic_limits(
        vals,
        lower_pct=2,
        upper_pct=98,
        pad_frac=0.10,
        lower_bound=0.0,
        upper_bound=1.0,
    )

    fig, ax = plt.subplots(figsize=(7, 7), facecolor=BLACK)
    style_ax(
        ax,
        title=f"Scalp Map of Mean H1 Persistence Weighted Gaussian Similarity | {FAST_OPTION.upper()}",
        grid=False,
    )

    head = plt.Circle((0, 0), 1.12, fill=False, color=ACCENT, linewidth=2)
    ax.add_patch(head)
    ax.plot([0.00, -0.10, 0.10, 0.00], [1.12, 1.28, 1.28, 1.12], color=ACCENT, linewidth=2)

    sc = ax.scatter(
        xs,
        ys,
        c=vals,
        s=260,
        cmap=CYAN_SEQUENTIAL_CMAP,
        edgecolor=ACCENT,
        linewidth=1.2,
        vmin=scalp_vmin,
        vmax=scalp_vmax,
    )

    for x0, y0, label in zip(xs, ys, labels):
        ax.text(
            x0,
            y0,
            label,
            ha="center",
            va="center",
            fontsize=8,
            color=BLACK,
            fontweight="bold",
        )

    cbar = plt.colorbar(sc, ax=ax, shrink=0.75)
    style_colorbar(cbar, "Mean normalized similarity")

    ax.set_xlim(-1.35, 1.35)
    ax.set_ylim(-1.35, 1.35)
    ax.set_aspect("equal")
    ax.set_xticks([])
    ax.set_yticks([])

    scalp_path = os.path.join(expert_dir, "10_mean_h1_persistence_weighted_gaussian_similarity_scalp_map_zoomed.png")
    plt.tight_layout()
    save_render_close(fig, scalp_path, width=760)

    regions = {
        "Frontal": ["Fp1", "Fpz", "Fp2", "F7", "F3", "Fz", "F4", "F8"],
        "Frontocentral": ["FC5", "FC1", "FC2", "FC6"],
        "Central": ["T7", "C3", "Cz", "C4", "T8"],
        "Centroparietal": ["CP5", "CP1", "CP2", "CP6"],
        "Parietal": ["P7", "P3", "Pz", "P4", "P8"],
        "Occipital": ["POz", "O1", "Oz", "O2"],
        "Mastoid": ["M1", "M2"],
    }

    region_rows = []
    region_points = []

    for region_name, chans in regions.items():
        vals_region = []
        dist_vals_region = []

        for ch in chans:
            if ch in mean_similarity_map and np.isfinite(mean_similarity_map[ch]):
                vals_region.append(mean_similarity_map[ch])
                dist_vals_region.append(mean_distance_map[ch])
                region_points.append({
                    "region": region_name,
                    "channel": ch,
                    "mean_weighted_gaussian_similarity_to_others": mean_similarity_map[ch],
                    "mean_weighted_gaussian_distance_to_others": mean_distance_map[ch],
                })

        if len(vals_region) == 0:
            continue

        vals_region = np.asarray(vals_region, dtype=float)
        dist_vals_region = np.asarray(dist_vals_region, dtype=float)

        region_rows.append({
            "region": region_name,
            "mean_region_weighted_gaussian_similarity": float(np.mean(vals_region)),
            "std_region_weighted_gaussian_similarity": float(np.std(vals_region, ddof=1)) if len(vals_region) > 1 else 0.0,
            "sem_region_weighted_gaussian_similarity": (
                float(np.std(vals_region, ddof=1) / np.sqrt(len(vals_region)))
                if len(vals_region) > 1 else 0.0
            ),
            "mean_region_weighted_gaussian_distance": float(np.mean(dist_vals_region)),
            "n_channels": int(len(vals_region)),
        })

    region_df = pd.DataFrame(region_rows)
    region_points_df = pd.DataFrame(region_points)

    region_csv_path = os.path.join(expert_dir, "regional_h1_persistence_weighted_gaussian_kernel_summary.csv")
    region_df.to_csv(region_csv_path, index=False)

    if len(region_df) > 0:
        fig, ax = plt.subplots(figsize=(12, 6.5), facecolor=BLACK)
        style_ax(
            ax,
            title=f"Regional Mean H1 Persistence Weighted Gaussian Similarity +/- SEM | {FAST_OPTION.upper()}",
            xlabel="Region",
            ylabel="Mean normalized weighted Gaussian similarity",
        )

        x_region = np.arange(len(region_df))

        ax.bar(
            x_region,
            region_df["mean_region_weighted_gaussian_similarity"],
            yerr=region_df["sem_region_weighted_gaussian_similarity"],
            color=ACCENT,
            edgecolor=ACCENT,
            alpha=0.82,
            width=0.68,
            capsize=8,
            error_kw={
                "elinewidth": 2.2,
                "capthick": 2.2,
                "ecolor": ACCENT,
                "alpha": 1.0,
            },
            label="Mean +/- SEM",
        )

        rng = np.random.default_rng(7)

        for i, region_name in enumerate(region_df["region"]):
            vals_region_plot = region_points_df.loc[
                region_points_df["region"] == region_name,
                "mean_weighted_gaussian_similarity_to_others",
            ].to_numpy(dtype=float)

            if len(vals_region_plot) == 0:
                continue

            jitter = rng.uniform(-0.13, 0.13, size=len(vals_region_plot))

            ax.scatter(
                np.full(len(vals_region_plot), i) + jitter,
                vals_region_plot,
                s=42,
                facecolors=BLACK,
                edgecolors=ACCENT,
                linewidths=1.2,
                alpha=0.95,
                zorder=4,
            )

        for i, row in region_df.iterrows():
            y_val = row["mean_region_weighted_gaussian_similarity"]
            sem = row["sem_region_weighted_gaussian_similarity"]

            ax.text(
                i,
                y_val + sem,
                f"{y_val:.3f}\n+/-{sem:.3f}\nn={int(row['n_channels'])}",
                ha="center",
                va="bottom",
                fontsize=8,
                color=ACCENT,
            )

        region_y = region_df["mean_region_weighted_gaussian_similarity"].to_numpy(dtype=float)
        reg_vmin, reg_vmax = dynamic_limits(
            region_y,
            lower_pct=0,
            upper_pct=100,
            pad_frac=0.28,
            lower_bound=0.0,
            upper_bound=1.0,
        )
        ax.set_ylim(reg_vmin, reg_vmax)

        ax.set_xticks(x_region)
        ax.set_xticklabels(
            region_df["region"],
            rotation=35,
            ha="right",
            rotation_mode="anchor",
        )
        ax.margins(x=0.04)
        ax.legend(loc="best", framealpha=1.0)

        region_path = os.path.join(expert_dir, "11_regional_h1_persistence_weighted_gaussian_similarity_summary_zoomed.png")
        plt.tight_layout(pad=2.0)
        save_render_close(fig, region_path)

    fig, ax = plt.subplots(figsize=(8, 6), facecolor=BLACK)
    style_ax(
        ax,
        title="Diagram Size vs Mean H1 Persistence Weighted Gaussian Similarity",
        xlabel="H1 diagram intervals",
        ylabel="Mean normalized weighted Gaussian similarity",
    )

    sc = ax.scatter(
        summary_df["n_intervals_diagram"],
        summary_df["mean_weighted_gaussian_similarity_to_others"],
        c=summary_df["max_lifetime"],
        s=140,
        cmap=CYAN_SEQUENTIAL_CMAP,
        edgecolor=ACCENT,
        linewidth=1.0,
    )

    for _, row in summary_df.iterrows():
        ax.text(
            row["n_intervals_diagram"],
            row["mean_weighted_gaussian_similarity_to_others"],
            row["channel"],
            fontsize=8,
            color=ACCENT,
            alpha=0.85,
        )

    cbar = plt.colorbar(sc, ax=ax, shrink=0.85)
    style_colorbar(cbar, "Max H1 lifetime")

    scatter_vmin, scatter_vmax = dynamic_limits(
        summary_df["mean_weighted_gaussian_similarity_to_others"],
        lower_pct=2,
        upper_pct=98,
        pad_frac=0.15,
        lower_bound=0.0,
        upper_bound=1.0,
    )
    ax.set_ylim(scatter_vmin, scatter_vmax)

    diagram_size_path = os.path.join(expert_dir, "12_diagram_size_vs_mean_weighted_gaussian_similarity_zoomed.png")
    plt.tight_layout()
    save_render_close(fig, diagram_size_path)

    fig, ax = plt.subplots(figsize=(8, 6), facecolor=BLACK)
    style_ax(
        ax,
        title="Threshold Triangles vs Mean H1 Persistence Weighted Gaussian Similarity",
        xlabel="Triangles at radius",
        ylabel="Mean normalized weighted Gaussian similarity",
    )

    sc = ax.scatter(
        summary_df["n_triangles_at_radius"],
        summary_df["mean_weighted_gaussian_similarity_to_others"],
        c=summary_df["n_intervals_diagram"],
        s=140,
        cmap=CYAN_SEQUENTIAL_CMAP,
        edgecolor=ACCENT,
        linewidth=1.0,
    )

    for _, row in summary_df.iterrows():
        ax.text(
            row["n_triangles_at_radius"],
            row["mean_weighted_gaussian_similarity_to_others"],
            row["channel"],
            fontsize=8,
            color=ACCENT,
            alpha=0.85,
        )

    cbar = plt.colorbar(sc, ax=ax, shrink=0.85)
    style_colorbar(cbar, "H1 diagram intervals")

    ax.set_ylim(scatter_vmin, scatter_vmax)

    triangle_path = os.path.join(expert_dir, "13_threshold_triangles_vs_mean_weighted_gaussian_similarity_zoomed.png")
    plt.tight_layout()
    save_render_close(fig, triangle_path)

    fig, ax = plt.subplots(figsize=(8, 6), facecolor=BLACK)
    style_ax(
        ax,
        title="Weighted Gaussian Similarity Spread per Channel",
        xlabel="Mean normalized weighted Gaussian similarity",
        ylabel="Similarity standard deviation",
    )

    sc = ax.scatter(
        summary_df["mean_weighted_gaussian_similarity_to_others"],
        summary_df["std_weighted_gaussian_similarity_to_others"],
        c=summary_df["n_intervals_diagram"],
        s=140,
        cmap=CYAN_SEQUENTIAL_CMAP,
        edgecolor=ACCENT,
        linewidth=1.0,
    )

    for _, row in summary_df.iterrows():
        ax.text(
            row["mean_weighted_gaussian_similarity_to_others"],
            row["std_weighted_gaussian_similarity_to_others"],
            row["channel"],
            fontsize=8,
            color=ACCENT,
            alpha=0.85,
        )

    cbar = plt.colorbar(sc, ax=ax, shrink=0.85)
    style_colorbar(cbar, "H1 diagram intervals")

    spread_path = os.path.join(expert_dir, "14_weighted_gaussian_similarity_spread_per_channel.png")
    plt.tight_layout()
    save_render_close(fig, spread_path)

    fig, ax = plt.subplots(figsize=(8, 6), facecolor=BLACK)
    style_ax(
        ax,
        title="Persistence Weight Mass vs Mean Weighted Gaussian Similarity",
        xlabel="Total persistence weight",
        ylabel="Mean normalized weighted Gaussian similarity",
    )

    sc = ax.scatter(
        summary_df["total_persistence_weight"],
        summary_df["mean_weighted_gaussian_similarity_to_others"],
        c=summary_df["n_intervals_diagram"],
        s=140,
        cmap=CYAN_SEQUENTIAL_CMAP,
        edgecolor=ACCENT,
        linewidth=1.0,
    )

    for _, row in summary_df.iterrows():
        ax.text(
            row["total_persistence_weight"],
            row["mean_weighted_gaussian_similarity_to_others"],
            row["channel"],
            fontsize=8,
            color=ACCENT,
            alpha=0.85,
        )

    cbar = plt.colorbar(sc, ax=ax, shrink=0.85)
    style_colorbar(cbar, "H1 diagram intervals")

    weight_mass_path = os.path.join(expert_dir, "15_weight_mass_vs_mean_weighted_gaussian_similarity.png")
    plt.tight_layout()
    save_render_close(fig, weight_mass_path)

    qc_df = summary_df.copy().sort_values("elapsed_sec", ascending=False).reset_index(drop=True)

    fig, ax = plt.subplots(figsize=(12, 5.8), facecolor=BLACK)
    style_ax(
        ax,
        title="Persistence Weighted Gaussian Kernel Runtime QC",
        xlabel="Channel",
        ylabel="Elapsed seconds",
    )

    x_qc = np.arange(len(qc_df))

    ax.bar(
        x_qc,
        qc_df["elapsed_sec"],
        color=ACCENT,
        edgecolor=ACCENT,
        alpha=0.82,
    )

    ax.set_xticks(x_qc)
    ax.set_xticklabels(qc_df["channel"], rotation=90)

    qc_path = os.path.join(expert_dir, "16_runtime_qc.png")
    plt.tight_layout()
    save_render_close(fig, qc_path)

    log(f"\nExpert plots saved and rendered from: {expert_dir}")
    log(f"Regional summary CSV: {region_csv_path}")

elapsed_total = time.time() - overall_t0

log("\n" + "=" * 72)
log(f"Finished H1 Persistence Weighted Gaussian Kernel in {elapsed_total:.2f}s")
log(f"Mode: {FAST_OPTION}")
log(f"Backend: ripser_presampled_fps + adaptive persistence weighted Gaussian kernel")
log(f"Gaussian bandwidth: {gaussian_bandwidth:.8f}")
log(f"Gaussian bandwidth method: {gaussian_bandwidth_method}")
log(f"Weight scale: {weight_scale:.8f}")
log(f"Weight scale method: {weight_scale_method}")
log(f"Weight power: {WEIGHT_POWER:.3f}")
log(f"Channels completed: {len(loaded_channels)}")
log(f"Summary CSV: {summary_csv_path}")
log(f"Raw weighted Gaussian kernel CSV: {raw_kernel_csv_path}")
log(f"Normalized similarity CSV: {normalized_kernel_csv_path}")
log(f"Weighted Gaussian kernel distance CSV: {kernel_distance_csv_path}")
log(f"Pairwise CSV: {pairwise_csv_path}")
log(f"MDS CSV: {mds_csv_path}")
log(f"NPZ: {npz_path}")
log(f"Output directory: {mode_out_dir}")
log("=" * 72)
```

```python
import os
import sys
import time
import numpy as np

if not hasattr(np, "trapz"):
    np.trapz = np.trapezoid

import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.colors import LinearSegmentedColormap

os.environ["OMP_NUM_THREADS"] = "2"
os.environ["OPENBLAS_NUM_THREADS"] = "2"
os.environ["MKL_NUM_THREADS"] = "2"
os.environ["VECLIB_MAXIMUM_THREADS"] = "2"
os.environ["NUMEXPR_NUM_THREADS"] = "2"

from scipy.spatial.distance import pdist, squareform
from scipy.cluster.hierarchy import linkage, leaves_list

try:
    from scipy.linalg import eigh
    HAVE_SCIPY_EIGH = True
except Exception:
    HAVE_SCIPY_EIGH = False

try:
    from IPython.display import display, Image
    HAVE_IPYTHON = True
except Exception:
    HAVE_IPYTHON = False

try:
    from ripser import ripser
except Exception as exc:
    raise ImportError(
        "ripser is not installed. Run this in a separate notebook cell first:\n\n"
        "!pip install ripser persim\n\n"
        "Then rerun this Sliced Wasserstein Persistence Kernel cell."
    ) from exc


base_dir = "/home/a/projects/Complete-Neural-Signal-Analysis"
out_dir = os.path.join(base_dir, "topology_sliced_wasserstein_persistence_kernel")
os.makedirs(out_dir, exist_ok=True)

FAST_OPTION = "ultra"  # options: "ultra", "fast", "balanced", "full"

MAX_CHANNELS_TO_PROCESS = None
RENDER_ALL_IMAGES_INLINE = True
SAVE_PLOTS_TO_DISK = True
MAKE_EXPERT_PLOTS = True
INLINE_IMAGE_WIDTH = 950
VERBOSE_PROGRESS = True

USE_GEOMETRY_PRESERVING_SUBSAMPLE = True
AUTO_KERNEL_BANDWIDTH = True

MODE_CONFIGS = {
    "ultra": {
        "max_points": 90,
        "reservoir_points": 5000,
        "max_dims": 3,
        "radius_quantile": 0.22,
        "n_sliced_directions": 32,
        "sw_kernel_sigma_frac": 0.35,
        "sw_kernel_auto_quantile": 0.50,
        "sw_kernel_auto_multiplier": 1.00,
        "cap_essential_h1": True,
        "plot_dpi": 120,
    },
    "fast": {
        "max_points": 130,
        "reservoir_points": 8000,
        "max_dims": 3,
        "radius_quantile": 0.24,
        "n_sliced_directions": 48,
        "sw_kernel_sigma_frac": 0.32,
        "sw_kernel_auto_quantile": 0.50,
        "sw_kernel_auto_multiplier": 0.95,
        "cap_essential_h1": True,
        "plot_dpi": 130,
    },
    "balanced": {
        "max_points": 180,
        "reservoir_points": 10000,
        "max_dims": 3,
        "radius_quantile": 0.26,
        "n_sliced_directions": 64,
        "sw_kernel_sigma_frac": 0.30,
        "sw_kernel_auto_quantile": 0.45,
        "sw_kernel_auto_multiplier": 0.90,
        "cap_essential_h1": True,
        "plot_dpi": 140,
    },
    "full": {
        "max_points": 260,
        "reservoir_points": 14000,
        "max_dims": 3,
        "radius_quantile": 0.28,
        "n_sliced_directions": 96,
        "sw_kernel_sigma_frac": 0.28,
        "sw_kernel_auto_quantile": 0.40,
        "sw_kernel_auto_multiplier": 0.85,
        "cap_essential_h1": True,
        "plot_dpi": 150,
    },
}

if FAST_OPTION not in MODE_CONFIGS:
    raise ValueError("FAST_OPTION must be one of: ultra, fast, balanced, full")

cfg = MODE_CONFIGS[FAST_OPTION]

MAX_POINTS = cfg["max_points"]
RESERVOIR_POINTS = cfg["reservoir_points"]
MAX_DIMS = cfg["max_dims"]
RADIUS_QUANTILE = cfg["radius_quantile"]
N_SLICED_DIRECTIONS = cfg["n_sliced_directions"]
SW_KERNEL_SIGMA_FRAC = cfg["sw_kernel_sigma_frac"]
SW_KERNEL_AUTO_QUANTILE = cfg["sw_kernel_auto_quantile"]
SW_KERNEL_AUTO_MULTIPLIER = cfg["sw_kernel_auto_multiplier"]
CAP_ESSENTIAL_H1 = cfg["cap_essential_h1"]
PLOT_DPI = cfg["plot_dpi"]

mode_out_dir = os.path.join(
    out_dir,
    f"h1_sliced_wasserstein_persistence_kernel_{FAST_OPTION}_fps_adaptive",
)
os.makedirs(mode_out_dir, exist_ok=True)

eeg_channels = [
    "Fp1", "Fpz", "Fp2", "F7", "F3", "Fz", "F4", "F8", "FC5", "FC1", "FC2", "FC6",
    "M1", "T7", "C3", "Cz", "C4", "T8", "M2", "CP5", "CP1", "CP2", "CP6",
    "P7", "P3", "Pz", "P4", "P8", "POz", "O1", "Oz", "O2"
]

if MAX_CHANNELS_TO_PROCESS is not None:
    eeg_channels = eeg_channels[:MAX_CHANNELS_TO_PROCESS]

ACCENT = "#00FFFF"
ACCENT_SOFT = "#66FFFF"
ACCENT_DIM = "#008B8B"
BLACK = "#000000"

CYAN_SEQUENTIAL_CMAP = LinearSegmentedColormap.from_list(
    "cyan_sequential",
    [BLACK, ACCENT_DIM, ACCENT, ACCENT_SOFT],
    N=256,
)

plt.rcParams.update({
    "figure.facecolor": BLACK,
    "axes.facecolor": BLACK,
    "savefig.facecolor": BLACK,
    "text.color": ACCENT,
    "axes.labelcolor": ACCENT,
    "axes.edgecolor": ACCENT,
    "axes.titlecolor": ACCENT,
    "xtick.color": ACCENT,
    "ytick.color": ACCENT,
    "grid.color": ACCENT_DIM,
    "grid.alpha": 0.25,
    "legend.facecolor": BLACK,
    "legend.edgecolor": ACCENT,
    "legend.labelcolor": ACCENT,
    "font.size": 11,
})


def log(msg):
    if VERBOSE_PROGRESS:
        print(msg, flush=True)


def style_ax(ax, title=None, xlabel=None, ylabel=None, grid=True):
    ax.set_facecolor(BLACK)

    for spine in ax.spines.values():
        spine.set_color(ACCENT)
        spine.set_linewidth(1.1)

    ax.tick_params(axis="x", colors=ACCENT, labelcolor=ACCENT)
    ax.tick_params(axis="y", colors=ACCENT, labelcolor=ACCENT)

    if title is not None:
        ax.set_title(title, color=ACCENT, pad=12)

    if xlabel is not None:
        ax.set_xlabel(xlabel, color=ACCENT, labelpad=10)

    if ylabel is not None:
        ax.set_ylabel(ylabel, color=ACCENT, labelpad=10)

    if grid:
        ax.grid(True, color=ACCENT_DIM, alpha=0.25, linewidth=0.8)


def style_colorbar(cbar, label):
    cbar.set_label(label, color=ACCENT)
    cbar.ax.yaxis.set_tick_params(color=ACCENT)
    plt.setp(cbar.ax.get_yticklabels(), color=ACCENT)
    cbar.outline.set_edgecolor(ACCENT)


def save_render_close(fig, path, width=INLINE_IMAGE_WIDTH):
    if SAVE_PLOTS_TO_DISK:
        fig.savefig(path, dpi=PLOT_DPI, bbox_inches="tight", facecolor=BLACK)

    plt.close(fig)

    if RENDER_ALL_IMAGES_INLINE and HAVE_IPYTHON and os.path.exists(path):
        display(Image(filename=path, width=width))


def dynamic_limits(values, lower_pct=2, upper_pct=98, pad_frac=0.08, lower_bound=None, upper_bound=None):
    values = np.asarray(values, dtype=float)
    values = values[np.isfinite(values)]

    if len(values) == 0:
        vmin, vmax = 0.0, 1.0
    else:
        vmin = float(np.percentile(values, lower_pct))
        vmax = float(np.percentile(values, upper_pct))

        if not np.isfinite(vmin):
            vmin = float(np.nanmin(values))

        if not np.isfinite(vmax):
            vmax = float(np.nanmax(values))

        if vmax <= vmin:
            center = float(np.nanmean(values))
            eps = max(1e-6, abs(center) * 0.01, 1e-5)
            vmin = center - eps
            vmax = center + eps

    pad = (vmax - vmin) * pad_frac
    vmin -= pad
    vmax += pad

    if lower_bound is not None:
        vmin = max(lower_bound, vmin)

    if upper_bound is not None:
        vmax = min(upper_bound, vmax)

    if vmax <= vmin:
        vmax = vmin + 1e-6

    return vmin, vmax


def offdiag_values(M):
    M = np.asarray(M, dtype=float)

    if M.shape[0] <= 1:
        return np.asarray([], dtype=float)

    mask = ~np.eye(M.shape[0], dtype=bool)
    vals = M[mask]
    vals = vals[np.isfinite(vals)]

    return vals


def save_matrix_plot(matrix, labels, title, colorbar_label, path, vmin=None, vmax=None, width=INLINE_IMAGE_WIDTH):
    fig, ax = plt.subplots(figsize=(9, 8), facecolor=BLACK)
    style_ax(
        ax,
        title=title,
        xlabel="Channel",
        ylabel="Channel",
    )

    im = ax.imshow(
        matrix,
        aspect="equal",
        interpolation="nearest",
        cmap=CYAN_SEQUENTIAL_CMAP,
        vmin=vmin,
        vmax=vmax,
    )

    ax.set_xticks(np.arange(len(labels)))
    ax.set_xticklabels(labels, rotation=90)
    ax.set_yticks(np.arange(len(labels)))
    ax.set_yticklabels(labels)

    cbar = plt.colorbar(im, ax=ax, shrink=0.85)
    style_colorbar(cbar, colorbar_label)

    plt.tight_layout()
    save_render_close(fig, path, width=width)


def estimate_median_pairwise_distance(X, max_pairs=120000, seed=7):
    X = np.asarray(X, dtype=float)
    n = X.shape[0]

    if n < 2:
        return 1.0

    rng = np.random.default_rng(seed)

    if n * (n - 1) // 2 <= max_pairs:
        D = pdist(X)
        D = D[D > 0]

        if len(D) == 0:
            return 1.0

        return float(np.nanmedian(D))

    i = rng.integers(0, n, size=max_pairs)
    j = rng.integers(0, n, size=max_pairs)

    mask = i != j
    i = i[mask]
    j = j[mask]

    if len(i) == 0:
        return 1.0

    d = np.linalg.norm(X[i] - X[j], axis=1)
    d = d[np.isfinite(d) & (d > 0)]

    if len(d) == 0:
        return 1.0

    return float(np.nanmedian(d))


def standardize_matrix_for_geometry(X, max_dims=3, seed=7):
    X = np.asarray(X, dtype=np.float64)
    X = X[np.all(np.isfinite(X), axis=1)]

    if X.shape[0] < 10:
        return None

    X = X - np.mean(X, axis=0, keepdims=True)

    if X.shape[1] > max_dims:
        _U, _S, Vt = np.linalg.svd(X, full_matrices=False)
        X = X @ Vt[:max_dims].T

    X = X / (np.std(X, axis=0, keepdims=True) + 1e-12)

    scale = estimate_median_pairwise_distance(X, seed=seed)
    X = X / (scale + 1e-12)

    return X


def farthest_point_sample(X, n_samples, seed=7):
    X = np.asarray(X, dtype=float)
    n = X.shape[0]

    if n <= n_samples:
        return np.arange(n)

    rng = np.random.default_rng(seed)

    center = np.mean(X, axis=0, keepdims=True)
    dist_to_center = np.linalg.norm(X - center, axis=1)

    start_idx = int(np.nanargmax(dist_to_center))

    selected = np.empty(n_samples, dtype=int)
    selected.fill(-1)
    selected[0] = start_idx

    min_dist2 = np.full(n, np.inf, dtype=float)

    for k in range(1, n_samples):
        last = X[selected[k - 1]]
        d2 = np.sum((X - last) ** 2, axis=1)
        min_dist2 = np.minimum(min_dist2, d2)
        min_dist2[selected[:k]] = -np.inf

        next_idx = int(np.nanargmax(min_dist2))

        if not np.isfinite(min_dist2[next_idx]) or min_dist2[next_idx] < 1e-18:
            remaining = np.setdiff1d(np.arange(n), selected[:k], assume_unique=False)

            if len(remaining) == 0:
                selected = selected[:k]
                break

            next_idx = int(rng.choice(remaining))

        selected[k] = next_idx

    selected = selected[selected >= 0]

    return selected


def load_channel_embedding_geometry_sampled(ch_name, max_points=90, reservoir_points=5000, max_dims=3, seed=7):
    candidate_paths = [
        os.path.join(base_dir, "embedding_data", "tsne_umap_from_3d", f"umap_embedding_{ch_name}.npy"),
        os.path.join(base_dir, "embedding_data", "tsne_umap_from_3d", f"tsne_embedding_{ch_name}.npy"),
        os.path.join(base_dir, "embedding_data", "3dembedding_data", f"3dembedded_{ch_name}.npy"),
        os.path.join(base_dir, "embedding_data", "2dembedding_data", f"2dembedded_{ch_name}.npy"),
    ]

    rng = np.random.default_rng(seed)

    for path in candidate_paths:
        if not os.path.exists(path):
            continue

        log(f"    trying: {path}")

        try:
            X = np.load(path, mmap_mode="r", allow_pickle=False)
        except Exception as exc:
            log(f"    skipped non-memmappable or nonnumeric file: {os.path.basename(path)} | {exc}")
            continue

        if X.ndim != 2:
            log(f"    skipped because ndim={X.ndim}")
            continue

        original_shape = tuple(X.shape)

        try:
            if X.shape[0] >= X.shape[1]:
                n_samples = X.shape[0]
                n_features = X.shape[1]

                if n_samples < 10 or n_features < 2:
                    log(f"    skipped bad shape: {original_shape}")
                    continue

                reservoir_n = min(max(reservoir_points, max_points), n_samples)

                if n_samples > reservoir_n:
                    reservoir_idx = np.sort(rng.choice(n_samples, size=reservoir_n, replace=False))
                else:
                    reservoir_idx = np.arange(n_samples)

                X_reservoir = np.asarray(X[reservoir_idx, :], dtype=np.float64)

            else:
                n_samples = X.shape[1]
                n_features = X.shape[0]

                if n_samples < 10 or n_features < 2:
                    log(f"    skipped bad transposed shape: {original_shape}")
                    continue

                reservoir_n = min(max(reservoir_points, max_points), n_samples)

                if n_samples > reservoir_n:
                    reservoir_idx = np.sort(rng.choice(n_samples, size=reservoir_n, replace=False))
                else:
                    reservoir_idx = np.arange(n_samples)

                X_reservoir = np.asarray(X[:, reservoir_idx].T, dtype=np.float64)

        except Exception as exc:
            log(f"    failed while reservoir-sampling {os.path.basename(path)} | {exc}")
            continue

        X_reservoir = X_reservoir[np.all(np.isfinite(X_reservoir), axis=1)]

        if X_reservoir.shape[0] < 10 or X_reservoir.shape[1] < 2:
            log(f"    reservoir data unusable: reservoir_shape={X_reservoir.shape}")
            continue

        X_reservoir_std = standardize_matrix_for_geometry(
            X_reservoir,
            max_dims=max_dims,
            seed=seed,
        )

        if X_reservoir_std is None:
            log("    standardized reservoir unusable")
            continue

        if USE_GEOMETRY_PRESERVING_SUBSAMPLE:
            fps_idx = farthest_point_sample(
                X_reservoir_std,
                n_samples=min(max_points, X_reservoir_std.shape[0]),
                seed=seed,
            )
            X_small = X_reservoir_std[fps_idx]
            method = "reservoir_plus_farthest_point"
        else:
            sample_n = min(max_points, X_reservoir_std.shape[0])
            idx = np.sort(rng.choice(X_reservoir_std.shape[0], size=sample_n, replace=False))
            X_small = X_reservoir_std[idx]
            method = "reservoir_random"

        X_small = X_small[np.all(np.isfinite(X_small), axis=1)]

        if X_small.shape[0] >= 10 and X_small.shape[1] >= 2:
            return X_small, path, original_shape, tuple(X_reservoir.shape), method

        log(f"    sampled data unusable: sampled_shape={X_small.shape}")

    return None, None, None, None, None


def standardize_point_cloud_small(X, max_dims=3):
    X = np.asarray(X, dtype=np.float64)
    X = X[np.all(np.isfinite(X), axis=1)]

    if X.shape[0] < 10:
        return None

    X = X - np.mean(X, axis=0, keepdims=True)

    if X.shape[1] > max_dims:
        _U, _S, Vt = np.linalg.svd(X, full_matrices=False)
        X = X @ Vt[:max_dims].T

    X = X / (np.std(X, axis=0, keepdims=True) + 1e-12)

    D = pdist(X)
    positive = D[D > 0]

    if len(positive) > 0:
        X = X / (np.nanmedian(positive) + 1e-12)

    return X


def estimate_complex_counts(X, radius):
    D = squareform(pdist(X))
    A = (D <= radius) & (D > 0)

    n_edges = int(np.sum(A) // 2)

    if A.shape[0] <= 300:
        A_int = A.astype(np.int64)
        n_triangles = int(np.trace(A_int @ A_int @ A_int) // 6)
    else:
        n_triangles = np.nan

    return n_edges, n_triangles


def run_h1_ripser(X, radius):
    result = ripser(
        X,
        maxdim=1,
        thresh=float(radius),
        coeff=2,
        do_cocycles=False,
    )

    diagrams = result["dgms"]

    h0 = diagrams[0] if len(diagrams) > 0 else np.empty((0, 2), dtype=float)
    h1 = diagrams[1] if len(diagrams) > 1 else np.empty((0, 2), dtype=float)

    return h0, h1


def split_finite_essential(diagram):
    diagram = np.asarray(diagram, dtype=float)

    if diagram.size == 0:
        return np.empty((0, 2), dtype=float), np.empty((0, 2), dtype=float)

    finite = diagram[
        np.isfinite(diagram[:, 0])
        & np.isfinite(diagram[:, 1])
        & (diagram[:, 1] > diagram[:, 0])
    ]

    essential = diagram[
        np.isfinite(diagram[:, 0])
        & np.isinf(diagram[:, 1])
    ]

    return finite, essential


def clean_diagram_for_sliced_wasserstein_kernel(
    finite_intervals,
    essential_intervals=None,
    cap_essential=True,
    cap_value=None,
):
    cleaned = []
    capped_count = 0

    if finite_intervals is not None:
        for b, d in finite_intervals:
            if np.isfinite(b) and np.isfinite(d) and d > b:
                cleaned.append((float(b), float(d)))

    if cap_essential and essential_intervals is not None and cap_value is not None:
        for b, _d in essential_intervals:
            if np.isfinite(b) and np.isfinite(cap_value) and cap_value > b:
                cleaned.append((float(b), float(cap_value)))
                capped_count += 1

    if len(cleaned) == 0:
        return np.empty((0, 2), dtype=float), capped_count

    return np.asarray(cleaned, dtype=float), capped_count


def diagram_features(diagram):
    diagram = np.asarray(diagram, dtype=float)

    if diagram.size == 0 or len(diagram) == 0:
        return {
            "n_intervals_diagram": 0,
            "mean_lifetime": np.nan,
            "max_lifetime": np.nan,
            "total_lifetime": 0.0,
            "mean_birth": np.nan,
            "mean_death": np.nan,
            "max_death": np.nan,
        }

    lifetimes = diagram[:, 1] - diagram[:, 0]

    return {
        "n_intervals_diagram": int(len(diagram)),
        "mean_lifetime": float(np.nanmean(lifetimes)),
        "max_lifetime": float(np.nanmax(lifetimes)),
        "total_lifetime": float(np.nansum(lifetimes)),
        "mean_birth": float(np.nanmean(diagram[:, 0])),
        "mean_death": float(np.nanmean(diagram[:, 1])),
        "max_death": float(np.nanmax(diagram[:, 1])),
    }


def diagonal_projection_points(D):
    D = np.asarray(D, dtype=float)

    if D.size == 0:
        return np.empty((0, 2), dtype=float)

    mid = 0.5 * (D[:, 0] + D[:, 1])
    return np.column_stack([mid, mid])


def sliced_wasserstein_distance(D1, D2, n_directions=32, p=1):
    D1 = np.asarray(D1, dtype=float)
    D2 = np.asarray(D2, dtype=float)

    if D1.size == 0:
        D1 = np.empty((0, 2), dtype=float)

    if D2.size == 0:
        D2 = np.empty((0, 2), dtype=float)

    if len(D1) == 0 and len(D2) == 0:
        return 0.0

    D1 = D1[
        np.isfinite(D1[:, 0])
        & np.isfinite(D1[:, 1])
        & (D1[:, 1] > D1[:, 0])
    ] if len(D1) else np.empty((0, 2), dtype=float)

    D2 = D2[
        np.isfinite(D2[:, 0])
        & np.isfinite(D2[:, 1])
        & (D2[:, 1] > D2[:, 0])
    ] if len(D2) else np.empty((0, 2), dtype=float)

    if len(D1) == 0 and len(D2) == 0:
        return 0.0

    D1_diag = diagonal_projection_points(D1)
    D2_diag = diagonal_projection_points(D2)

    A = np.vstack([D1, D2_diag]) if len(D2_diag) else D1.copy()
    B = np.vstack([D2, D1_diag]) if len(D1_diag) else D2.copy()

    if len(A) == 0 and len(B) == 0:
        return 0.0

    if len(A) != len(B):
        raise RuntimeError(
            f"Internal sliced Wasserstein augmentation failed: len(A)={len(A)}, len(B)={len(B)}"
        )

    thetas = np.linspace(0.0, np.pi, n_directions, endpoint=False)

    distances = []

    for theta in thetas:
        direction = np.array([np.cos(theta), np.sin(theta)], dtype=float)

        proj_a = np.sort(A @ direction)
        proj_b = np.sort(B @ direction)

        diff = np.abs(proj_a - proj_b)

        if p == 1:
            distances.append(float(np.mean(diff)))
        else:
            distances.append(float(np.mean(diff ** p) ** (1.0 / p)))

    return float(np.mean(distances))


def choose_sliced_wasserstein_kernel_sigma(sw_distance_matrix, fallback_scale):
    vals = offdiag_values(sw_distance_matrix)
    vals = vals[np.isfinite(vals) & (vals > 0)]

    fallback_sigma = max(1e-6, SW_KERNEL_SIGMA_FRAC * fallback_scale)

    if len(vals) == 0:
        return fallback_sigma, "fallback_no_positive_sw_distances"

    auto_sigma = float(np.quantile(vals, SW_KERNEL_AUTO_QUANTILE) * SW_KERNEL_AUTO_MULTIPLIER)

    if not np.isfinite(auto_sigma) or auto_sigma <= 0:
        return fallback_sigma, "fallback_bad_auto_sigma"

    if AUTO_KERNEL_BANDWIDTH:
        chosen = auto_sigma
        chosen = max(chosen, fallback_scale * 0.0025, 1e-6)
        method = "adaptive_pairwise_sliced_wasserstein_quantile"
    else:
        chosen = fallback_sigma
        method = "fixed_fractional"

    return chosen, method


def sliced_wasserstein_persistence_kernel_from_distance(sw_distance, sigma):
    sigma = max(float(sigma), 1e-12)
    return float(np.exp(-(float(sw_distance) ** 2) / (2.0 * sigma ** 2)))


def kernel_distance_from_similarity(K):
    K = np.asarray(K, dtype=float)
    K = 0.5 * (K + K.T)
    K = np.clip(K, 0.0, 1.0)

    D2 = np.maximum(0.0, 2.0 - 2.0 * K)
    D = np.sqrt(D2)
    np.fill_diagonal(D, 0.0)

    return D


def classical_mds_from_distances(D, n_components=2):
    D = np.asarray(D, dtype=float)
    n = D.shape[0]

    if n < 2:
        return np.zeros((n, n_components), dtype=float)

    D2 = D ** 2
    J = np.eye(n) - np.ones((n, n)) / n
    B = -0.5 * J @ D2 @ J

    if HAVE_SCIPY_EIGH:
        vals, vecs = eigh(B)
    else:
        vals, vecs = np.linalg.eigh(B)

    idx = np.argsort(vals)[::-1]
    vals = vals[idx]
    vecs = vecs[:, idx]

    coords = np.zeros((n, n_components), dtype=float)

    for k in range(min(n_components, n)):
        if vals[k] > 0:
            coords[:, k] = vecs[:, k] * np.sqrt(vals[k])

    return coords


log("\n" + "=" * 72)
log(f"Running H1 Sliced Wasserstein Persistence Kernel | {FAST_OPTION.upper()}")
log(f"Channels: {len(eeg_channels)}")
log(f"max_points={MAX_POINTS}, reservoir_points={RESERVOIR_POINTS}, radius_quantile={RADIUS_QUANTILE}")
log(f"n_sliced_directions={N_SLICED_DIRECTIONS}")
log(f"sw_kernel_sigma_frac={SW_KERNEL_SIGMA_FRAC}, auto_kernel_bandwidth={AUTO_KERNEL_BANDWIDTH}")
log(f"geometry_preserving_subsample={USE_GEOMETRY_PRESERVING_SUBSAMPLE}")
log(f"cap_essential_h1={CAP_ESSENTIAL_H1}")
log(f"expert_plots={MAKE_EXPERT_PLOTS}, render_all_images={RENDER_ALL_IMAGES_INLINE}")
log(f"Output: {mode_out_dir}")
log("=" * 72)

diagnostics = {}
diagram_store = {}
raw_interval_store = {}
loaded_channels = []

global_max_death = None
global_max_lifetime = None

overall_t0 = time.time()

for ch_idx, ch_name in enumerate(eeg_channels, start=1):
    t0 = time.time()
    prefix = f"[{ch_idx:02d}/{len(eeg_channels)} {ch_name}]"

    log(f"\n{prefix} loading reservoir + geometry-preserving subsampling")

    X_small, source_path, original_shape, reservoir_shape, sampling_method = load_channel_embedding_geometry_sampled(
        ch_name,
        max_points=MAX_POINTS,
        reservoir_points=RESERVOIR_POINTS,
        max_dims=MAX_DIMS,
        seed=1000 + ch_idx,
    )

    if X_small is None:
        log(f"{prefix} WARNING: no usable embedding found; skipping")
        continue

    log(
        f"{prefix} source shape={original_shape}, reservoir shape={reservoir_shape}, "
        f"sampled shape={X_small.shape}, method={sampling_method}"
    )

    X = standardize_point_cloud_small(X_small, max_dims=MAX_DIMS)

    if X is None:
        log(f"{prefix} WARNING: invalid sampled embedding; skipping")
        continue

    log(f"{prefix} standardized shape={X.shape}; computing radius")

    D_flat = pdist(X)
    positive = D_flat[D_flat > 0]

    if len(positive) == 0:
        log(f"{prefix} WARNING: zero-distance cloud; skipping")
        continue

    radius = float(np.quantile(positive, RADIUS_QUANTILE))

    log(f"{prefix} estimating threshold complex size")
    n_edges, n_triangles = estimate_complex_counts(X, radius)

    log(f"{prefix} ripser H0/H1")
    h0_diag, h1_diag = run_h1_ripser(X, radius)

    h0_finite, h0_essential = split_finite_essential(h0_diag)
    h1_finite, h1_essential = split_finite_essential(h1_diag)

    h1_diagram, n_capped_h1 = clean_diagram_for_sliced_wasserstein_kernel(
        h1_finite,
        essential_intervals=h1_essential,
        cap_essential=CAP_ESSENTIAL_H1,
        cap_value=radius,
    )

    if len(h1_diagram) > 0:
        max_death = float(np.nanmax(h1_diagram[:, 1]))
        max_lifetime = float(np.nanmax(h1_diagram[:, 1] - h1_diagram[:, 0]))

        global_max_death = max_death if global_max_death is None else max(global_max_death, max_death)
        global_max_lifetime = max_lifetime if global_max_lifetime is None else max(global_max_lifetime, max_lifetime)

    elapsed = time.time() - t0

    diagram_store[ch_name] = h1_diagram

    raw_interval_store[ch_name] = {
        "H0_finite": h0_finite,
        "H0_essential": h0_essential,
        "H1_finite": h1_finite,
        "H1_essential": h1_essential,
        "H1_sliced_wasserstein_kernel_diagram": h1_diagram,
    }

    diagnostics[ch_name] = {
        "channel": ch_name,
        "source_path": source_path,
        "source_shape": str(original_shape),
        "reservoir_shape": str(reservoir_shape),
        "sampling_method": sampling_method,
        "backend": "ripser_presampled_fps",
        "n_points": int(X.shape[0]),
        "n_edges_at_radius": int(n_edges),
        "n_triangles_at_radius": int(n_triangles) if np.isfinite(n_triangles) else np.nan,
        "radius": radius,
        "n_h0_finite": int(len(h0_finite)),
        "n_h1_finite": int(len(h1_finite)),
        "n_h1_essential": int(len(h1_essential)),
        "n_h1_capped_for_sliced_wasserstein_kernel": int(n_capped_h1),
        "elapsed_sec": elapsed,
        **diagram_features(h1_diagram),
    }

    loaded_channels.append(ch_name)

    log(
        f"{prefix} done {elapsed:.2f}s | "
        f"edges={n_edges}, triangles={n_triangles}, "
        f"H1 finite={len(h1_finite)}, H1 SW-kernel diagram={len(h1_diagram)}"
    )

if len(loaded_channels) == 0:
    raise RuntimeError("No usable embeddings found. Check embedding_data paths.")

if global_max_death is None or not np.isfinite(global_max_death) or global_max_death <= 0:
    global_max_death = 1.0

if global_max_lifetime is None or not np.isfinite(global_max_lifetime) or global_max_lifetime <= 0:
    global_max_lifetime = 1.0

diagram_scale = max(global_max_death, global_max_lifetime, 1e-6)

log("\nComputing pairwise sliced Wasserstein distance matrix")
log(f"global_max_death={global_max_death:.6f}")
log(f"global_max_lifetime={global_max_lifetime:.6f}")
log(f"diagram_scale={diagram_scale:.6f}")

n_ch = len(loaded_channels)
sw_distance_mat = np.zeros((n_ch, n_ch), dtype=float)

pair_distance_rows = []
pair_t0 = time.time()

for i in range(n_ch):
    ch_i = loaded_channels[i]

    for j in range(i + 1, n_ch):
        ch_j = loaded_channels[j]

        dist = sliced_wasserstein_distance(
            diagram_store[ch_i],
            diagram_store[ch_j],
            n_directions=N_SLICED_DIRECTIONS,
            p=1,
        )

        sw_distance_mat[i, j] = dist
        sw_distance_mat[j, i] = dist

        pair_distance_rows.append({
            "channel_a": ch_i,
            "channel_b": ch_j,
            "sliced_wasserstein_distance": dist,
        })

    log(f"  completed SW distance row {i + 1}/{n_ch}: {ch_i}")

pair_distance_elapsed = time.time() - pair_t0

sw_kernel_sigma, sw_kernel_sigma_method = choose_sliced_wasserstein_kernel_sigma(
    sw_distance_mat,
    fallback_scale=diagram_scale,
)

log("\nComputing sliced Wasserstein persistence kernel matrix")
log(f"sw_kernel_sigma={sw_kernel_sigma:.8f}")
log(f"sw_kernel_sigma_method={sw_kernel_sigma_method}")

sw_kernel_mat = np.zeros((n_ch, n_ch), dtype=float)

pair_rows = []

for i in range(n_ch):
    for j in range(i, n_ch):
        sim = sliced_wasserstein_persistence_kernel_from_distance(
            sw_distance_mat[i, j],
            sigma=sw_kernel_sigma,
        )

        sw_kernel_mat[i, j] = sim
        sw_kernel_mat[j, i] = sim

        if i != j:
            pair_rows.append({
                "channel_a": loaded_channels[i],
                "channel_b": loaded_channels[j],
                "sliced_wasserstein_distance": sw_distance_mat[i, j],
                "sliced_wasserstein_kernel_similarity": sim,
            })

np.fill_diagonal(sw_kernel_mat, 1.0)

sw_kernel_distance_mat = kernel_distance_from_similarity(sw_kernel_mat)

sw_distance_df = pd.DataFrame(
    sw_distance_mat,
    index=loaded_channels,
    columns=loaded_channels,
)

sw_kernel_df = pd.DataFrame(
    sw_kernel_mat,
    index=loaded_channels,
    columns=loaded_channels,
)

sw_kernel_distance_df = pd.DataFrame(
    sw_kernel_distance_mat,
    index=loaded_channels,
    columns=loaded_channels,
)

pairwise_df = (
    pd.DataFrame(pair_rows)
    .sort_values("sliced_wasserstein_kernel_similarity", ascending=False)
    .reset_index(drop=True)
)

summary_rows = []

for idx, ch_name in enumerate(loaded_channels):
    sim = sw_kernel_mat[idx].copy()
    dist = sw_kernel_distance_mat[idx].copy()
    sw_dist = sw_distance_mat[idx].copy()

    mask = np.arange(len(sim)) != idx
    sim_nonself = sim[mask]
    dist_nonself = dist[mask]
    sw_dist_nonself = sw_dist[mask]

    summary_rows.append({
        **diagnostics[ch_name],
        "mean_sliced_wasserstein_distance_to_others": float(np.nanmean(sw_dist_nonself)) if len(sw_dist_nonself) else 0.0,
        "median_sliced_wasserstein_distance_to_others": float(np.nanmedian(sw_dist_nonself)) if len(sw_dist_nonself) else 0.0,
        "max_sliced_wasserstein_distance_to_others": float(np.nanmax(sw_dist_nonself)) if len(sw_dist_nonself) else 0.0,
        "min_sliced_wasserstein_distance_to_others": float(np.nanmin(sw_dist_nonself)) if len(sw_dist_nonself) else 0.0,
        "mean_sw_kernel_similarity_to_others": float(np.nanmean(sim_nonself)) if len(sim_nonself) else 1.0,
        "median_sw_kernel_similarity_to_others": float(np.nanmedian(sim_nonself)) if len(sim_nonself) else 1.0,
        "max_sw_kernel_similarity_to_others": float(np.nanmax(sim_nonself)) if len(sim_nonself) else 1.0,
        "min_sw_kernel_similarity_to_others": float(np.nanmin(sim_nonself)) if len(sim_nonself) else 1.0,
        "std_sw_kernel_similarity_to_others": float(np.nanstd(sim_nonself)) if len(sim_nonself) else 0.0,
        "mean_sw_kernel_distance_to_others": float(np.nanmean(dist_nonself)) if len(dist_nonself) else 0.0,
        "median_sw_kernel_distance_to_others": float(np.nanmedian(dist_nonself)) if len(dist_nonself) else 0.0,
        "max_sw_kernel_distance_to_others": float(np.nanmax(dist_nonself)) if len(dist_nonself) else 0.0,
        "min_sw_kernel_distance_to_others": float(np.nanmin(dist_nonself)) if len(dist_nonself) else 0.0,
        "std_sw_kernel_distance_to_others": float(np.nanstd(dist_nonself)) if len(dist_nonself) else 0.0,
    })

summary_df = (
    pd.DataFrame(summary_rows)
    .sort_values("mean_sw_kernel_similarity_to_others", ascending=False)
    .reset_index(drop=True)
)

offdiag_sim = offdiag_values(sw_kernel_mat)
offdiag_kernel_dist = offdiag_values(sw_kernel_distance_mat)
offdiag_sw_dist = offdiag_values(sw_distance_mat)

similarity_range = float(np.nanmax(offdiag_sim) - np.nanmin(offdiag_sim)) if len(offdiag_sim) else 0.0
kernel_distance_range = float(np.nanmax(offdiag_kernel_dist) - np.nanmin(offdiag_kernel_dist)) if len(offdiag_kernel_dist) else 0.0
sw_distance_range = float(np.nanmax(offdiag_sw_dist) - np.nanmin(offdiag_sw_dist)) if len(offdiag_sw_dist) else 0.0

log("\nSliced Wasserstein persistence kernel variability QC")
if len(offdiag_sw_dist):
    log(
        f"off-diagonal SW distance: "
        f"min={np.nanmin(offdiag_sw_dist):.6f}, "
        f"max={np.nanmax(offdiag_sw_dist):.6f}, "
        f"mean={np.nanmean(offdiag_sw_dist):.6f}, "
        f"std={np.nanstd(offdiag_sw_dist):.6f}, "
        f"range={sw_distance_range:.6f}"
    )

if len(offdiag_sim):
    log(
        f"off-diagonal SW kernel similarity: "
        f"min={np.nanmin(offdiag_sim):.6f}, "
        f"max={np.nanmax(offdiag_sim):.6f}, "
        f"mean={np.nanmean(offdiag_sim):.6f}, "
        f"std={np.nanstd(offdiag_sim):.6f}, "
        f"range={similarity_range:.6f}"
    )

if len(offdiag_kernel_dist):
    log(
        f"SW kernel distance: "
        f"min={np.nanmin(offdiag_kernel_dist):.6f}, "
        f"max={np.nanmax(offdiag_kernel_dist):.6f}, "
        f"mean={np.nanmean(offdiag_kernel_dist):.6f}, "
        f"std={np.nanstd(offdiag_kernel_dist):.6f}, "
        f"range={kernel_distance_range:.6f}"
    )

if len(offdiag_sim) and np.nanstd(offdiag_sim) < 1e-4:
    log(
        "WARNING: sliced Wasserstein kernel similarities are nearly constant. "
        "Try lowering sw_kernel_sigma_frac, increasing max_points, or increasing radius_quantile."
    )

summary_csv_path = os.path.join(mode_out_dir, "h1_sliced_wasserstein_persistence_kernel_channel_summary.csv")
sw_distance_csv_path = os.path.join(mode_out_dir, "h1_sliced_wasserstein_distance_matrix.csv")
sw_kernel_csv_path = os.path.join(mode_out_dir, "h1_sliced_wasserstein_persistence_kernel_similarity_matrix.csv")
sw_kernel_distance_csv_path = os.path.join(mode_out_dir, "h1_sliced_wasserstein_persistence_kernel_distance_matrix.csv")
pairwise_csv_path = os.path.join(mode_out_dir, "h1_sliced_wasserstein_persistence_kernel_pairwise_similarities.csv")

summary_df.to_csv(summary_csv_path, index=False)
sw_distance_df.to_csv(sw_distance_csv_path)
sw_kernel_df.to_csv(sw_kernel_csv_path)
sw_kernel_distance_df.to_csv(sw_kernel_distance_csv_path)
pairwise_df.to_csv(pairwise_csv_path, index=False)

npz_path = os.path.join(mode_out_dir, "h1_sliced_wasserstein_persistence_kernel_and_diagrams.npz")

save_dict = {
    "channels": np.asarray(loaded_channels),
    "sliced_wasserstein_distance_matrix": sw_distance_mat,
    "sliced_wasserstein_persistence_kernel_matrix": sw_kernel_mat,
    "sliced_wasserstein_persistence_kernel_distance_matrix": sw_kernel_distance_mat,
    "sw_kernel_sigma": np.asarray([sw_kernel_sigma]),
    "sw_kernel_sigma_method": np.asarray([sw_kernel_sigma_method]),
    "n_sliced_directions": np.asarray([N_SLICED_DIRECTIONS]),
    "fast_option": np.asarray([FAST_OPTION]),
    "backend": np.asarray(["ripser_presampled_fps + sliced wasserstein persistence kernel"]),
}

for ch_name in loaded_channels:
    save_dict[f"{ch_name}_H1_sliced_wasserstein_kernel_diagram"] = diagram_store[ch_name]

    for key, val in raw_interval_store[ch_name].items():
        save_dict[f"{ch_name}_{key}"] = val

np.savez(npz_path, **save_dict)

print("\nH1 Sliced Wasserstein Persistence Kernel Channel Summary:")
if HAVE_IPYTHON:
    display(summary_df)
else:
    print(summary_df)

print("\nMost similar H1 diagram pairs by sliced Wasserstein persistence kernel:")
if HAVE_IPYTHON:
    display(pairwise_df.head(12))
else:
    print(pairwise_df.head(12))

print("\nLeast similar H1 diagram pairs by sliced Wasserstein persistence kernel:")
least_similar_df = pairwise_df.tail(12).sort_values("sliced_wasserstein_kernel_similarity", ascending=True)
if HAVE_IPYTHON:
    display(least_similar_df)
else:
    print(least_similar_df)

log(f"\nSaved summary CSV: {summary_csv_path}")
log(f"Saved SW distance matrix CSV: {sw_distance_csv_path}")
log(f"Saved SW persistence kernel matrix CSV: {sw_kernel_csv_path}")
log(f"Saved SW persistence kernel distance matrix CSV: {sw_kernel_distance_csv_path}")
log(f"Saved pairwise similarities CSV: {pairwise_csv_path}")
log(f"Saved NPZ: {npz_path}")
log(f"Pairwise SW distance elapsed: {pair_distance_elapsed:.2f}s")

plot_df = summary_df.copy().sort_values("mean_sw_kernel_similarity_to_others", ascending=False).reset_index(drop=True)

fig, ax = plt.subplots(figsize=(13, 5.8), facecolor=BLACK)
style_ax(
    ax,
    title=f"Mean H1 Sliced Wasserstein Kernel Similarity to Other Channels | {FAST_OPTION.upper()}",
    xlabel="Channel",
    ylabel="Mean SW kernel similarity",
)

x_pos = np.arange(len(plot_df))
y = plot_df["mean_sw_kernel_similarity_to_others"].to_numpy(dtype=float)

ax.bar(x_pos, y, color=ACCENT, edgecolor=ACCENT, alpha=0.78)

if np.any(np.isfinite(y)):
    ax.axhline(
        np.nanmean(y),
        color=ACCENT_SOFT,
        linestyle="--",
        linewidth=1.4,
        alpha=0.8,
        label=f"Mean={np.nanmean(y):.5f}",
    )

bar_vmin, bar_vmax = dynamic_limits(y, lower_pct=0, upper_pct=100, pad_frac=0.25, lower_bound=0.0, upper_bound=1.0)
ax.set_ylim(bar_vmin, bar_vmax)

ax.set_xticks(x_pos)
ax.set_xticklabels(plot_df["channel"], rotation=90)
ax.legend(loc="best", framealpha=1.0)

mean_sim_path = os.path.join(mode_out_dir, "01_mean_h1_sliced_wasserstein_kernel_similarity_by_channel_zoomed.png")
plt.tight_layout()
save_render_close(fig, mean_sim_path)

if n_ch >= 3 and np.any(sw_kernel_distance_mat > 0):
    condensed = sw_kernel_distance_mat[np.triu_indices(n_ch, k=1)]
    Z = linkage(condensed, method="average")
    order = leaves_list(Z)
else:
    order = np.arange(n_ch)

ordered_sw_dist = sw_distance_mat[np.ix_(order, order)]
ordered_kernel = sw_kernel_mat[np.ix_(order, order)]
ordered_kernel_dist = sw_kernel_distance_mat[np.ix_(order, order)]
ordered_channels = [loaded_channels[i] for i in order]

sim_vmin, sim_vmax = dynamic_limits(
    offdiag_values(ordered_kernel),
    lower_pct=2,
    upper_pct=98,
    pad_frac=0.10,
    lower_bound=0.0,
    upper_bound=1.0,
)

save_matrix_plot(
    ordered_kernel,
    ordered_channels,
    "H1 Sliced Wasserstein Persistence Kernel Similarity Matrix | Zoomed Color Scale",
    "SW kernel similarity",
    os.path.join(mode_out_dir, "02_h1_sliced_wasserstein_persistence_kernel_similarity_heatmap_zoomed.png"),
    vmin=sim_vmin,
    vmax=sim_vmax,
)

sw_dist_vmin, sw_dist_vmax = dynamic_limits(
    offdiag_values(ordered_sw_dist),
    lower_pct=2,
    upper_pct=98,
    pad_frac=0.10,
    lower_bound=0.0,
)

save_matrix_plot(
    ordered_sw_dist,
    ordered_channels,
    "H1 Sliced Wasserstein Distance Matrix | Zoomed Color Scale",
    "Sliced Wasserstein distance",
    os.path.join(mode_out_dir, "03_h1_sliced_wasserstein_distance_heatmap_zoomed.png"),
    vmin=sw_dist_vmin,
    vmax=sw_dist_vmax,
)

kernel_dist_vmin, kernel_dist_vmax = dynamic_limits(
    offdiag_values(ordered_kernel_dist),
    lower_pct=2,
    upper_pct=98,
    pad_frac=0.10,
    lower_bound=0.0,
)

save_matrix_plot(
    ordered_kernel_dist,
    ordered_channels,
    "H1 Sliced Wasserstein Persistence Kernel Distance Matrix | Zoomed Color Scale",
    "SW kernel distance",
    os.path.join(mode_out_dir, "04_h1_sliced_wasserstein_persistence_kernel_distance_heatmap_zoomed.png"),
    vmin=kernel_dist_vmin,
    vmax=kernel_dist_vmax,
)

mds_coords = classical_mds_from_distances(sw_kernel_distance_mat, n_components=2)

mds_df = pd.DataFrame({
    "channel": loaded_channels,
    "mds_1": mds_coords[:, 0],
    "mds_2": mds_coords[:, 1],
    "mean_sw_kernel_similarity_to_others": [
        summary_df.loc[summary_df["channel"] == ch, "mean_sw_kernel_similarity_to_others"].values[0]
        for ch in loaded_channels
    ],
    "mean_sw_kernel_distance_to_others": [
        summary_df.loc[summary_df["channel"] == ch, "mean_sw_kernel_distance_to_others"].values[0]
        for ch in loaded_channels
    ],
    "mean_sliced_wasserstein_distance_to_others": [
        summary_df.loc[summary_df["channel"] == ch, "mean_sliced_wasserstein_distance_to_others"].values[0]
        for ch in loaded_channels
    ],
    "n_intervals_diagram": [
        summary_df.loc[summary_df["channel"] == ch, "n_intervals_diagram"].values[0]
        for ch in loaded_channels
    ],
})

mds_csv_path = os.path.join(mode_out_dir, "h1_sliced_wasserstein_persistence_kernel_mds_coordinates.csv")
mds_df.to_csv(mds_csv_path, index=False)

fig, ax = plt.subplots(figsize=(8, 7), facecolor=BLACK)
style_ax(
    ax,
    title="Classical MDS of H1 Sliced Wasserstein Kernel Distances",
    xlabel="MDS 1",
    ylabel="MDS 2",
)

mds_vmin, mds_vmax = dynamic_limits(
    mds_df["mean_sw_kernel_similarity_to_others"],
    lower_pct=2,
    upper_pct=98,
    pad_frac=0.08,
    lower_bound=0.0,
    upper_bound=1.0,
)

sc = ax.scatter(
    mds_df["mds_1"],
    mds_df["mds_2"],
    c=mds_df["mean_sw_kernel_similarity_to_others"],
    s=150,
    cmap=CYAN_SEQUENTIAL_CMAP,
    edgecolor=ACCENT,
    linewidth=1.1,
    vmin=mds_vmin,
    vmax=mds_vmax,
)

for _, row in mds_df.iterrows():
    ax.text(
        row["mds_1"],
        row["mds_2"],
        row["channel"],
        fontsize=8,
        color=ACCENT,
        alpha=0.9,
    )

cbar = plt.colorbar(sc, ax=ax, shrink=0.85)
style_colorbar(cbar, "Mean SW kernel similarity")

mds_path = os.path.join(mode_out_dir, "05_h1_sliced_wasserstein_persistence_kernel_mds_embedding.png")
plt.tight_layout()
save_render_close(fig, mds_path)

diagram_channels = summary_df.head(min(12, len(summary_df)))["channel"].tolist()
n_cols = 4
n_rows = int(np.ceil(len(diagram_channels) / n_cols))

fig, axes = plt.subplots(
    n_rows,
    n_cols,
    figsize=(4 * n_cols, 3.2 * n_rows),
    facecolor=BLACK,
)

axes = np.asarray(axes).reshape(-1)

all_diagram_points = np.vstack([
    diagram_store[ch] for ch in loaded_channels if len(diagram_store[ch]) > 0
]) if any(len(diagram_store[ch]) > 0 for ch in loaded_channels) else np.empty((0, 2))

if len(all_diagram_points) > 0:
    diag_max = float(np.nanmax(all_diagram_points[:, 1]))
else:
    diag_max = 1.0

diag_max = max(diag_max, 1e-6)

for ax_idx, ch_name in enumerate(diagram_channels):
    ax = axes[ax_idx]
    D = diagram_store[ch_name]

    mean_sim = summary_df.loc[
        summary_df["channel"] == ch_name,
        "mean_sw_kernel_similarity_to_others",
    ].values[0]

    style_ax(
        ax,
        title=f"{ch_name}: mean SWK={mean_sim:.3f}, n={len(D)}",
        xlabel="Birth",
        ylabel="Death",
    )

    ax.plot([0, diag_max], [0, diag_max], color=ACCENT_DIM, linestyle="--", linewidth=1.0)

    if len(D) > 0:
        ax.scatter(
            D[:, 0],
            D[:, 1],
            s=55,
            facecolors=BLACK,
            edgecolors=ACCENT,
            linewidths=1.2,
        )

    ax.set_xlim(0, diag_max * 1.05)
    ax.set_ylim(0, diag_max * 1.05)

for j in range(len(diagram_channels), len(axes)):
    axes[j].axis("off")

diagrams_path = os.path.join(mode_out_dir, "06_top_channels_h1_persistence_diagrams_by_sliced_wasserstein_kernel.png")
plt.tight_layout()
save_render_close(fig, diagrams_path, width=1100)

fig, ax = plt.subplots(figsize=(12, 6), facecolor=BLACK)
style_ax(
    ax,
    title="Most Similar Channel Pairs by H1 Sliced Wasserstein Kernel",
    xlabel="Channel pair",
    ylabel="SW kernel similarity",
)

top_pairs = pairwise_df.head(min(15, len(pairwise_df))).copy()
pair_labels = top_pairs["channel_a"] + "-" + top_pairs["channel_b"]

x_pair = np.arange(len(top_pairs))

ax.bar(
    x_pair,
    top_pairs["sliced_wasserstein_kernel_similarity"],
    color=ACCENT,
    edgecolor=ACCENT,
    alpha=0.78,
)

top_vmin, top_vmax = dynamic_limits(
    top_pairs["sliced_wasserstein_kernel_similarity"],
    lower_pct=0,
    upper_pct=100,
    pad_frac=0.25,
    lower_bound=0.0,
    upper_bound=1.0,
)
ax.set_ylim(top_vmin, top_vmax)

ax.set_xticks(x_pair)
ax.set_xticklabels(pair_labels, rotation=60, ha="right")

top_pairs_path = os.path.join(mode_out_dir, "07_most_similar_channel_pairs_h1_sliced_wasserstein_kernel_zoomed.png")
plt.tight_layout()
save_render_close(fig, top_pairs_path)

fig, ax = plt.subplots(figsize=(12, 6), facecolor=BLACK)
style_ax(
    ax,
    title="Least Similar Channel Pairs by H1 Sliced Wasserstein Kernel",
    xlabel="Channel pair",
    ylabel="SW kernel similarity",
)

bottom_pairs = pairwise_df.tail(min(15, len(pairwise_df))).copy().sort_values(
    "sliced_wasserstein_kernel_similarity",
    ascending=True,
)
bottom_pair_labels = bottom_pairs["channel_a"] + "-" + bottom_pairs["channel_b"]

x_pair = np.arange(len(bottom_pairs))

ax.bar(
    x_pair,
    bottom_pairs["sliced_wasserstein_kernel_similarity"],
    color=ACCENT,
    edgecolor=ACCENT,
    alpha=0.78,
)

bottom_vmin, bottom_vmax = dynamic_limits(
    bottom_pairs["sliced_wasserstein_kernel_similarity"],
    lower_pct=0,
    upper_pct=100,
    pad_frac=0.25,
    lower_bound=0.0,
    upper_bound=1.0,
)
ax.set_ylim(bottom_vmin, bottom_vmax)

ax.set_xticks(x_pair)
ax.set_xticklabels(bottom_pair_labels, rotation=60, ha="right")

bottom_pairs_path = os.path.join(mode_out_dir, "08_least_similar_channel_pairs_h1_sliced_wasserstein_kernel_zoomed.png")
plt.tight_layout()
save_render_close(fig, bottom_pairs_path)

fig, ax = plt.subplots(figsize=(8, 5.8), facecolor=BLACK)
style_ax(
    ax,
    title="Off-Diagonal Sliced Wasserstein Kernel Similarity Distribution",
    xlabel="SW kernel similarity",
    ylabel="Count",
)

if len(offdiag_sim):
    ax.hist(
        offdiag_sim,
        bins=24,
        color=ACCENT,
        edgecolor=ACCENT,
        alpha=0.75,
    )

hist_path = os.path.join(mode_out_dir, "09_offdiagonal_sliced_wasserstein_kernel_similarity_distribution.png")
plt.tight_layout()
save_render_close(fig, hist_path)

if MAKE_EXPERT_PLOTS:
    expert_dir = os.path.join(mode_out_dir, "expert_plots")
    os.makedirs(expert_dir, exist_ok=True)

    mean_similarity_map = dict(zip(summary_df["channel"], summary_df["mean_sw_kernel_similarity_to_others"]))
    mean_distance_map = dict(zip(summary_df["channel"], summary_df["mean_sw_kernel_distance_to_others"]))
    mean_sw_distance_map = dict(zip(summary_df["channel"], summary_df["mean_sliced_wasserstein_distance_to_others"]))

    scalp_xy = {
        "Fp1": (-0.45, 1.00), "Fpz": (0.00, 1.08), "Fp2": (0.45, 1.00),
        "F7": (-0.95, 0.60), "F3": (-0.45, 0.55), "Fz": (0.00, 0.60),
        "F4": (0.45, 0.55), "F8": (0.95, 0.60),
        "FC5": (-0.75, 0.25), "FC1": (-0.25, 0.25),
        "FC2": (0.25, 0.25), "FC6": (0.75, 0.25),
        "M1": (-1.15, 0.05), "T7": (-1.00, 0.00), "C3": (-0.50, 0.00),
        "Cz": (0.00, 0.00), "C4": (0.50, 0.00), "T8": (1.00, 0.00),
        "M2": (1.15, 0.05),
        "CP5": (-0.75, -0.30), "CP1": (-0.25, -0.30),
        "CP2": (0.25, -0.30), "CP6": (0.75, -0.30),
        "P7": (-0.90, -0.60), "P3": (-0.45, -0.60), "Pz": (0.00, -0.65),
        "P4": (0.45, -0.60), "P8": (0.90, -0.60),
        "POz": (0.00, -0.88),
        "O1": (-0.35, -1.00), "Oz": (0.00, -1.08), "O2": (0.35, -1.00),
    }

    xs, ys, vals, labels = [], [], [], []

    for ch in eeg_channels:
        if ch in scalp_xy and ch in mean_similarity_map and np.isfinite(mean_similarity_map[ch]):
            xs.append(scalp_xy[ch][0])
            ys.append(scalp_xy[ch][1])
            vals.append(mean_similarity_map[ch])
            labels.append(ch)

    scalp_vmin, scalp_vmax = dynamic_limits(
        vals,
        lower_pct=2,
        upper_pct=98,
        pad_frac=0.10,
        lower_bound=0.0,
        upper_bound=1.0,
    )

    fig, ax = plt.subplots(figsize=(7, 7), facecolor=BLACK)
    style_ax(
        ax,
        title=f"Scalp Map of Mean H1 Sliced Wasserstein Kernel Similarity | {FAST_OPTION.upper()}",
        grid=False,
    )

    head = plt.Circle((0, 0), 1.12, fill=False, color=ACCENT, linewidth=2)
    ax.add_patch(head)
    ax.plot([0.00, -0.10, 0.10, 0.00], [1.12, 1.28, 1.28, 1.12], color=ACCENT, linewidth=2)

    sc = ax.scatter(
        xs,
        ys,
        c=vals,
        s=260,
        cmap=CYAN_SEQUENTIAL_CMAP,
        edgecolor=ACCENT,
        linewidth=1.2,
        vmin=scalp_vmin,
        vmax=scalp_vmax,
    )

    for x0, y0, label in zip(xs, ys, labels):
        ax.text(
            x0,
            y0,
            label,
            ha="center",
            va="center",
            fontsize=8,
            color=BLACK,
            fontweight="bold",
        )

    cbar = plt.colorbar(sc, ax=ax, shrink=0.75)
    style_colorbar(cbar, "Mean SW kernel similarity")

    ax.set_xlim(-1.35, 1.35)
    ax.set_ylim(-1.35, 1.35)
    ax.set_aspect("equal")
    ax.set_xticks([])
    ax.set_yticks([])

    scalp_path = os.path.join(expert_dir, "10_mean_h1_sliced_wasserstein_kernel_similarity_scalp_map_zoomed.png")
    plt.tight_layout()
    save_render_close(fig, scalp_path, width=760)

    regions = {
        "Frontal": ["Fp1", "Fpz", "Fp2", "F7", "F3", "Fz", "F4", "F8"],
        "Frontocentral": ["FC5", "FC1", "FC2", "FC6"],
        "Central": ["T7", "C3", "Cz", "C4", "T8"],
        "Centroparietal": ["CP5", "CP1", "CP2", "CP6"],
        "Parietal": ["P7", "P3", "Pz", "P4", "P8"],
        "Occipital": ["POz", "O1", "Oz", "O2"],
        "Mastoid": ["M1", "M2"],
    }

    region_rows = []
    region_points = []

    for region_name, chans in regions.items():
        vals_region = []
        dist_vals_region = []
        sw_vals_region = []

        for ch in chans:
            if ch in mean_similarity_map and np.isfinite(mean_similarity_map[ch]):
                vals_region.append(mean_similarity_map[ch])
                dist_vals_region.append(mean_distance_map[ch])
                sw_vals_region.append(mean_sw_distance_map[ch])
                region_points.append({
                    "region": region_name,
                    "channel": ch,
                    "mean_sw_kernel_similarity_to_others": mean_similarity_map[ch],
                    "mean_sw_kernel_distance_to_others": mean_distance_map[ch],
                    "mean_sliced_wasserstein_distance_to_others": mean_sw_distance_map[ch],
                })

        if len(vals_region) == 0:
            continue

        vals_region = np.asarray(vals_region, dtype=float)
        dist_vals_region = np.asarray(dist_vals_region, dtype=float)
        sw_vals_region = np.asarray(sw_vals_region, dtype=float)

        region_rows.append({
            "region": region_name,
            "mean_region_sw_kernel_similarity": float(np.mean(vals_region)),
            "std_region_sw_kernel_similarity": float(np.std(vals_region, ddof=1)) if len(vals_region) > 1 else 0.0,
            "sem_region_sw_kernel_similarity": (
                float(np.std(vals_region, ddof=1) / np.sqrt(len(vals_region)))
                if len(vals_region) > 1 else 0.0
            ),
            "mean_region_sw_kernel_distance": float(np.mean(dist_vals_region)),
            "mean_region_sliced_wasserstein_distance": float(np.mean(sw_vals_region)),
            "n_channels": int(len(vals_region)),
        })

    region_df = pd.DataFrame(region_rows)
    region_points_df = pd.DataFrame(region_points)

    region_csv_path = os.path.join(expert_dir, "regional_h1_sliced_wasserstein_persistence_kernel_summary.csv")
    region_df.to_csv(region_csv_path, index=False)

    if len(region_df) > 0:
        fig, ax = plt.subplots(figsize=(12, 6.5), facecolor=BLACK)
        style_ax(
            ax,
            title=f"Regional Mean H1 Sliced Wasserstein Kernel Similarity +/- SEM | {FAST_OPTION.upper()}",
            xlabel="Region",
            ylabel="Mean SW kernel similarity",
        )

        x_region = np.arange(len(region_df))

        ax.bar(
            x_region,
            region_df["mean_region_sw_kernel_similarity"],
            yerr=region_df["sem_region_sw_kernel_similarity"],
            color=ACCENT,
            edgecolor=ACCENT,
            alpha=0.82,
            width=0.68,
            capsize=8,
            error_kw={
                "elinewidth": 2.2,
                "capthick": 2.2,
                "ecolor": ACCENT,
                "alpha": 1.0,
            },
            label="Mean +/- SEM",
        )

        rng = np.random.default_rng(7)

        for i, region_name in enumerate(region_df["region"]):
            vals_region_plot = region_points_df.loc[
                region_points_df["region"] == region_name,
                "mean_sw_kernel_similarity_to_others",
            ].to_numpy(dtype=float)

            if len(vals_region_plot) == 0:
                continue

            jitter = rng.uniform(-0.13, 0.13, size=len(vals_region_plot))

            ax.scatter(
                np.full(len(vals_region_plot), i) + jitter,
                vals_region_plot,
                s=42,
                facecolors=BLACK,
                edgecolors=ACCENT,
                linewidths=1.2,
                alpha=0.95,
                zorder=4,
            )

        for i, row in region_df.iterrows():
            y_val = row["mean_region_sw_kernel_similarity"]
            sem = row["sem_region_sw_kernel_similarity"]

            ax.text(
                i,
                y_val + sem,
                f"{y_val:.3f}\n+/-{sem:.3f}\nn={int(row['n_channels'])}",
                ha="center",
                va="bottom",
                fontsize=8,
                color=ACCENT,
            )

        region_y = region_df["mean_region_sw_kernel_similarity"].to_numpy(dtype=float)
        reg_vmin, reg_vmax = dynamic_limits(
            region_y,
            lower_pct=0,
            upper_pct=100,
            pad_frac=0.28,
            lower_bound=0.0,
            upper_bound=1.0,
        )
        ax.set_ylim(reg_vmin, reg_vmax)

        ax.set_xticks(x_region)
        ax.set_xticklabels(
            region_df["region"],
            rotation=35,
            ha="right",
            rotation_mode="anchor",
        )
        ax.margins(x=0.04)
        ax.legend(loc="best", framealpha=1.0)

        region_path = os.path.join(expert_dir, "11_regional_h1_sliced_wasserstein_kernel_similarity_summary_zoomed.png")
        plt.tight_layout(pad=2.0)
        save_render_close(fig, region_path)

    fig, ax = plt.subplots(figsize=(8, 6), facecolor=BLACK)
    style_ax(
        ax,
        title="Diagram Size vs Mean H1 Sliced Wasserstein Kernel Similarity",
        xlabel="H1 diagram intervals",
        ylabel="Mean SW kernel similarity",
    )

    sc = ax.scatter(
        summary_df["n_intervals_diagram"],
        summary_df["mean_sw_kernel_similarity_to_others"],
        c=summary_df["max_lifetime"],
        s=140,
        cmap=CYAN_SEQUENTIAL_CMAP,
        edgecolor=ACCENT,
        linewidth=1.0,
    )

    for _, row in summary_df.iterrows():
        ax.text(
            row["n_intervals_diagram"],
            row["mean_sw_kernel_similarity_to_others"],
            row["channel"],
            fontsize=8,
            color=ACCENT,
            alpha=0.85,
        )

    cbar = plt.colorbar(sc, ax=ax, shrink=0.85)
    style_colorbar(cbar, "Max H1 lifetime")

    scatter_vmin, scatter_vmax = dynamic_limits(
        summary_df["mean_sw_kernel_similarity_to_others"],
        lower_pct=2,
        upper_pct=98,
        pad_frac=0.15,
        lower_bound=0.0,
        upper_bound=1.0,
    )
    ax.set_ylim(scatter_vmin, scatter_vmax)

    diagram_size_path = os.path.join(expert_dir, "12_diagram_size_vs_mean_sliced_wasserstein_kernel_similarity_zoomed.png")
    plt.tight_layout()
    save_render_close(fig, diagram_size_path)

    fig, ax = plt.subplots(figsize=(8, 6), facecolor=BLACK)
    style_ax(
        ax,
        title="Mean SW Distance vs Mean SW Kernel Similarity",
        xlabel="Mean sliced Wasserstein distance to others",
        ylabel="Mean SW kernel similarity",
    )

    sc = ax.scatter(
        summary_df["mean_sliced_wasserstein_distance_to_others"],
        summary_df["mean_sw_kernel_similarity_to_others"],
        c=summary_df["n_intervals_diagram"],
        s=140,
        cmap=CYAN_SEQUENTIAL_CMAP,
        edgecolor=ACCENT,
        linewidth=1.0,
    )

    for _, row in summary_df.iterrows():
        ax.text(
            row["mean_sliced_wasserstein_distance_to_others"],
            row["mean_sw_kernel_similarity_to_others"],
            row["channel"],
            fontsize=8,
            color=ACCENT,
            alpha=0.85,
        )

    cbar = plt.colorbar(sc, ax=ax, shrink=0.85)
    style_colorbar(cbar, "H1 diagram intervals")

    distance_vs_kernel_path = os.path.join(expert_dir, "13_mean_sw_distance_vs_mean_sw_kernel_similarity.png")
    plt.tight_layout()
    save_render_close(fig, distance_vs_kernel_path)

    fig, ax = plt.subplots(figsize=(8, 6), facecolor=BLACK)
    style_ax(
        ax,
        title="Sliced Wasserstein Kernel Similarity Spread per Channel",
        xlabel="Mean SW kernel similarity",
        ylabel="Similarity standard deviation",
    )

    sc = ax.scatter(
        summary_df["mean_sw_kernel_similarity_to_others"],
        summary_df["std_sw_kernel_similarity_to_others"],
        c=summary_df["n_intervals_diagram"],
        s=140,
        cmap=CYAN_SEQUENTIAL_CMAP,
        edgecolor=ACCENT,
        linewidth=1.0,
    )

    for _, row in summary_df.iterrows():
        ax.text(
            row["mean_sw_kernel_similarity_to_others"],
            row["std_sw_kernel_similarity_to_others"],
            row["channel"],
            fontsize=8,
            color=ACCENT,
            alpha=0.85,
        )

    cbar = plt.colorbar(sc, ax=ax, shrink=0.85)
    style_colorbar(cbar, "H1 diagram intervals")

    spread_path = os.path.join(expert_dir, "14_sliced_wasserstein_kernel_similarity_spread_per_channel.png")
    plt.tight_layout()
    save_render_close(fig, spread_path)

    qc_df = summary_df.copy().sort_values("elapsed_sec", ascending=False).reset_index(drop=True)

    fig, ax = plt.subplots(figsize=(12, 5.8), facecolor=BLACK)
    style_ax(
        ax,
        title="Sliced Wasserstein Persistence Kernel Runtime QC",
        xlabel="Channel",
        ylabel="Elapsed seconds",
    )

    x_qc = np.arange(len(qc_df))

    ax.bar(
        x_qc,
        qc_df["elapsed_sec"],
        color=ACCENT,
        edgecolor=ACCENT,
        alpha=0.82,
    )

    ax.set_xticks(x_qc)
    ax.set_xticklabels(qc_df["channel"], rotation=90)

    qc_path = os.path.join(expert_dir, "15_runtime_qc.png")
    plt.tight_layout()
    save_render_close(fig, qc_path)

    log(f"\nExpert plots saved and rendered from: {expert_dir}")
    log(f"Regional summary CSV: {region_csv_path}")

elapsed_total = time.time() - overall_t0

log("\n" + "=" * 72)
log(f"Finished H1 Sliced Wasserstein Persistence Kernel in {elapsed_total:.2f}s")
log(f"Mode: {FAST_OPTION}")
log(f"Backend: ripser_presampled_fps + sliced Wasserstein persistence kernel")
log(f"Sliced directions: {N_SLICED_DIRECTIONS}")
log(f"SW kernel sigma: {sw_kernel_sigma:.8f}")
log(f"SW kernel sigma method: {sw_kernel_sigma_method}")
log(f"Channels completed: {len(loaded_channels)}")
log(f"Summary CSV: {summary_csv_path}")
log(f"SW distance matrix CSV: {sw_distance_csv_path}")
log(f"SW persistence kernel CSV: {sw_kernel_csv_path}")
log(f"SW persistence kernel distance CSV: {sw_kernel_distance_csv_path}")
log(f"Pairwise CSV: {pairwise_csv_path}")
log(f"MDS CSV: {mds_csv_path}")
log(f"NPZ: {npz_path}")
log(f"Output directory: {mode_out_dir}")
log("=" * 72)
```
