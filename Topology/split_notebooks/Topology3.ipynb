```python
import os
import sys
import time
import numpy as np

if not hasattr(np, "trapz"):
    np.trapz = np.trapezoid

import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.colors import LinearSegmentedColormap

os.environ["OMP_NUM_THREADS"] = "2"
os.environ["OPENBLAS_NUM_THREADS"] = "2"
os.environ["MKL_NUM_THREADS"] = "2"
os.environ["VECLIB_MAXIMUM_THREADS"] = "2"
os.environ["NUMEXPR_NUM_THREADS"] = "2"

from scipy.spatial.distance import pdist, squareform
from scipy.cluster.hierarchy import linkage, leaves_list

try:
    from scipy.linalg import eigh
    HAVE_SCIPY_EIGH = True
except Exception:
    HAVE_SCIPY_EIGH = False

try:
    from IPython.display import display, Image
    HAVE_IPYTHON = True
except Exception:
    HAVE_IPYTHON = False

try:
    from ripser import ripser
except Exception as exc:
    raise ImportError(
        "ripser is not installed. Run this in a separate notebook cell first:\n\n"
        "!pip install ripser persim\n\n"
        "Then rerun this Persistence Diagram Vectorization cell."
    ) from exc


base_dir = "/home/a/projects/Complete-Neural-Signal-Analysis"
out_dir = os.path.join(base_dir, "topology_persistence_diagram_vectorization")
os.makedirs(out_dir, exist_ok=True)

FAST_OPTION = "ultra"  # options: "ultra", "fast", "balanced", "full"

MAX_CHANNELS_TO_PROCESS = None
RENDER_ALL_IMAGES_INLINE = True
SAVE_PLOTS_TO_DISK = True
MAKE_EXPERT_PLOTS = True
INLINE_IMAGE_WIDTH = 950
VERBOSE_PROGRESS = True

USE_GEOMETRY_PRESERVING_SUBSAMPLE = True
STANDARDIZE_FEATURES_FOR_DISTANCE = True

MODE_CONFIGS = {
    "ultra": {
        "max_points": 90,
        "reservoir_points": 5000,
        "max_dims": 3,
        "radius_quantile": 0.22,
        "birth_bins": 20,
        "persistence_bins": 20,
        "betti_grid_size": 80,
        "landscape_grid_size": 80,
        "landscape_levels": 3,
        "top_lifetimes": 12,
        "image_sigma_frac": 0.035,
        "image_weight_power": 1.0,
        "cap_essential_h1": True,
        "plot_dpi": 120,
    },
    "fast": {
        "max_points": 130,
        "reservoir_points": 8000,
        "max_dims": 3,
        "radius_quantile": 0.24,
        "birth_bins": 24,
        "persistence_bins": 24,
        "betti_grid_size": 110,
        "landscape_grid_size": 110,
        "landscape_levels": 4,
        "top_lifetimes": 16,
        "image_sigma_frac": 0.030,
        "image_weight_power": 1.0,
        "cap_essential_h1": True,
        "plot_dpi": 130,
    },
    "balanced": {
        "max_points": 180,
        "reservoir_points": 10000,
        "max_dims": 3,
        "radius_quantile": 0.26,
        "birth_bins": 30,
        "persistence_bins": 30,
        "betti_grid_size": 150,
        "landscape_grid_size": 150,
        "landscape_levels": 5,
        "top_lifetimes": 20,
        "image_sigma_frac": 0.027,
        "image_weight_power": 1.0,
        "cap_essential_h1": True,
        "plot_dpi": 140,
    },
    "full": {
        "max_points": 260,
        "reservoir_points": 14000,
        "max_dims": 3,
        "radius_quantile": 0.28,
        "birth_bins": 36,
        "persistence_bins": 36,
        "betti_grid_size": 220,
        "landscape_grid_size": 220,
        "landscape_levels": 6,
        "top_lifetimes": 28,
        "image_sigma_frac": 0.024,
        "image_weight_power": 1.0,
        "cap_essential_h1": True,
        "plot_dpi": 150,
    },
}

if FAST_OPTION not in MODE_CONFIGS:
    raise ValueError("FAST_OPTION must be one of: ultra, fast, balanced, full")

cfg = MODE_CONFIGS[FAST_OPTION]

MAX_POINTS = cfg["max_points"]
RESERVOIR_POINTS = cfg["reservoir_points"]
MAX_DIMS = cfg["max_dims"]
RADIUS_QUANTILE = cfg["radius_quantile"]
BIRTH_BINS = cfg["birth_bins"]
PERSISTENCE_BINS = cfg["persistence_bins"]
BETTI_GRID_SIZE = cfg["betti_grid_size"]
LANDSCAPE_GRID_SIZE = cfg["landscape_grid_size"]
LANDSCAPE_LEVELS = cfg["landscape_levels"]
TOP_LIFETIMES = cfg["top_lifetimes"]
IMAGE_SIGMA_FRAC = cfg["image_sigma_frac"]
IMAGE_WEIGHT_POWER = cfg["image_weight_power"]
CAP_ESSENTIAL_H1 = cfg["cap_essential_h1"]
PLOT_DPI = cfg["plot_dpi"]

mode_out_dir = os.path.join(
    out_dir,
    f"h1_persistence_diagram_vectorization_{FAST_OPTION}_fps",
)
os.makedirs(mode_out_dir, exist_ok=True)

eeg_channels = [
    "Fp1", "Fpz", "Fp2", "F7", "F3", "Fz", "F4", "F8", "FC5", "FC1", "FC2", "FC6",
    "M1", "T7", "C3", "Cz", "C4", "T8", "M2", "CP5", "CP1", "CP2", "CP6",
    "P7", "P3", "Pz", "P4", "P8", "POz", "O1", "Oz", "O2"
]

if MAX_CHANNELS_TO_PROCESS is not None:
    eeg_channels = eeg_channels[:MAX_CHANNELS_TO_PROCESS]

ACCENT = "#00FFFF"
ACCENT_SOFT = "#66FFFF"
ACCENT_DIM = "#008B8B"
BLACK = "#000000"

CYAN_SEQUENTIAL_CMAP = LinearSegmentedColormap.from_list(
    "cyan_sequential",
    [BLACK, ACCENT_DIM, ACCENT, ACCENT_SOFT],
    N=256,
)

plt.rcParams.update({
    "figure.facecolor": BLACK,
    "axes.facecolor": BLACK,
    "savefig.facecolor": BLACK,
    "text.color": ACCENT,
    "axes.labelcolor": ACCENT,
    "axes.edgecolor": ACCENT,
    "axes.titlecolor": ACCENT,
    "xtick.color": ACCENT,
    "ytick.color": ACCENT,
    "grid.color": ACCENT_DIM,
    "grid.alpha": 0.25,
    "legend.facecolor": BLACK,
    "legend.edgecolor": ACCENT,
    "legend.labelcolor": ACCENT,
    "font.size": 11,
})


def log(msg):
    if VERBOSE_PROGRESS:
        print(msg, flush=True)


def style_ax(ax, title=None, xlabel=None, ylabel=None, grid=True):
    ax.set_facecolor(BLACK)

    for spine in ax.spines.values():
        spine.set_color(ACCENT)
        spine.set_linewidth(1.1)

    ax.tick_params(axis="x", colors=ACCENT, labelcolor=ACCENT)
    ax.tick_params(axis="y", colors=ACCENT, labelcolor=ACCENT)

    if title is not None:
        ax.set_title(title, color=ACCENT, pad=12)

    if xlabel is not None:
        ax.set_xlabel(xlabel, color=ACCENT, labelpad=10)

    if ylabel is not None:
        ax.set_ylabel(ylabel, color=ACCENT, labelpad=10)

    if grid:
        ax.grid(True, color=ACCENT_DIM, alpha=0.25, linewidth=0.8)


def style_colorbar(cbar, label):
    cbar.set_label(label, color=ACCENT)
    cbar.ax.yaxis.set_tick_params(color=ACCENT)
    plt.setp(cbar.ax.get_yticklabels(), color=ACCENT)
    cbar.outline.set_edgecolor(ACCENT)


def save_render_close(fig, path, width=INLINE_IMAGE_WIDTH):
    if SAVE_PLOTS_TO_DISK:
        fig.savefig(path, dpi=PLOT_DPI, bbox_inches="tight", facecolor=BLACK)

    plt.close(fig)

    if RENDER_ALL_IMAGES_INLINE and HAVE_IPYTHON and os.path.exists(path):
        display(Image(filename=path, width=width))


def dynamic_limits(values, lower_pct=2, upper_pct=98, pad_frac=0.08, lower_bound=None, upper_bound=None):
    values = np.asarray(values, dtype=float)
    values = values[np.isfinite(values)]

    if len(values) == 0:
        vmin, vmax = 0.0, 1.0
    else:
        vmin = float(np.percentile(values, lower_pct))
        vmax = float(np.percentile(values, upper_pct))

        if not np.isfinite(vmin):
            vmin = float(np.nanmin(values))

        if not np.isfinite(vmax):
            vmax = float(np.nanmax(values))

        if vmax <= vmin:
            center = float(np.nanmean(values))
            eps = max(1e-6, abs(center) * 0.01, 1e-5)
            vmin = center - eps
            vmax = center + eps

    pad = (vmax - vmin) * pad_frac
    vmin -= pad
    vmax += pad

    if lower_bound is not None:
        vmin = max(lower_bound, vmin)

    if upper_bound is not None:
        vmax = min(upper_bound, vmax)

    if vmax <= vmin:
        vmax = vmin + 1e-6

    return vmin, vmax


def offdiag_values(M):
    M = np.asarray(M, dtype=float)

    if M.shape[0] <= 1:
        return np.asarray([], dtype=float)

    mask = ~np.eye(M.shape[0], dtype=bool)
    vals = M[mask]
    vals = vals[np.isfinite(vals)]

    return vals


def save_matrix_plot(matrix, labels, title, colorbar_label, path, vmin=None, vmax=None, width=INLINE_IMAGE_WIDTH):
    fig, ax = plt.subplots(figsize=(9, 8), facecolor=BLACK)
    style_ax(
        ax,
        title=title,
        xlabel="Channel",
        ylabel="Channel",
    )

    im = ax.imshow(
        matrix,
        aspect="equal",
        interpolation="nearest",
        cmap=CYAN_SEQUENTIAL_CMAP,
        vmin=vmin,
        vmax=vmax,
    )

    ax.set_xticks(np.arange(len(labels)))
    ax.set_xticklabels(labels, rotation=90)
    ax.set_yticks(np.arange(len(labels)))
    ax.set_yticklabels(labels)

    cbar = plt.colorbar(im, ax=ax, shrink=0.85)
    style_colorbar(cbar, colorbar_label)

    plt.tight_layout()
    save_render_close(fig, path, width=width)


def estimate_median_pairwise_distance(X, max_pairs=120000, seed=7):
    X = np.asarray(X, dtype=float)
    n = X.shape[0]

    if n < 2:
        return 1.0

    rng = np.random.default_rng(seed)

    if n * (n - 1) // 2 <= max_pairs:
        D = pdist(X)
        D = D[D > 0]

        if len(D) == 0:
            return 1.0

        return float(np.nanmedian(D))

    i = rng.integers(0, n, size=max_pairs)
    j = rng.integers(0, n, size=max_pairs)

    mask = i != j
    i = i[mask]
    j = j[mask]

    if len(i) == 0:
        return 1.0

    d = np.linalg.norm(X[i] - X[j], axis=1)
    d = d[np.isfinite(d) & (d > 0)]

    if len(d) == 0:
        return 1.0

    return float(np.nanmedian(d))


def standardize_matrix_for_geometry(X, max_dims=3, seed=7):
    X = np.asarray(X, dtype=np.float64)
    X = X[np.all(np.isfinite(X), axis=1)]

    if X.shape[0] < 10:
        return None

    X = X - np.mean(X, axis=0, keepdims=True)

    if X.shape[1] > max_dims:
        _U, _S, Vt = np.linalg.svd(X, full_matrices=False)
        X = X @ Vt[:max_dims].T

    X = X / (np.std(X, axis=0, keepdims=True) + 1e-12)

    scale = estimate_median_pairwise_distance(X, seed=seed)
    X = X / (scale + 1e-12)

    return X


def farthest_point_sample(X, n_samples, seed=7):
    X = np.asarray(X, dtype=float)
    n = X.shape[0]

    if n <= n_samples:
        return np.arange(n)

    rng = np.random.default_rng(seed)

    center = np.mean(X, axis=0, keepdims=True)
    dist_to_center = np.linalg.norm(X - center, axis=1)

    start_idx = int(np.nanargmax(dist_to_center))

    selected = np.empty(n_samples, dtype=int)
    selected.fill(-1)
    selected[0] = start_idx

    min_dist2 = np.full(n, np.inf, dtype=float)

    for k in range(1, n_samples):
        last = X[selected[k - 1]]
        d2 = np.sum((X - last) ** 2, axis=1)
        min_dist2 = np.minimum(min_dist2, d2)
        min_dist2[selected[:k]] = -np.inf

        next_idx = int(np.nanargmax(min_dist2))

        if not np.isfinite(min_dist2[next_idx]) or min_dist2[next_idx] < 1e-18:
            remaining = np.setdiff1d(np.arange(n), selected[:k], assume_unique=False)

            if len(remaining) == 0:
                selected = selected[:k]
                break

            next_idx = int(rng.choice(remaining))

        selected[k] = next_idx

    selected = selected[selected >= 0]

    return selected


def load_channel_embedding_geometry_sampled(ch_name, max_points=90, reservoir_points=5000, max_dims=3, seed=7):
    candidate_paths = [
        os.path.join(base_dir, "embedding_data", "tsne_umap_from_3d", f"umap_embedding_{ch_name}.npy"),
        os.path.join(base_dir, "embedding_data", "tsne_umap_from_3d", f"tsne_embedding_{ch_name}.npy"),
        os.path.join(base_dir, "embedding_data", "3dembedding_data", f"3dembedded_{ch_name}.npy"),
        os.path.join(base_dir, "embedding_data", "2dembedding_data", f"2dembedded_{ch_name}.npy"),
    ]

    rng = np.random.default_rng(seed)

    for path in candidate_paths:
        if not os.path.exists(path):
            continue

        log(f"    trying: {path}")

        try:
            X = np.load(path, mmap_mode="r", allow_pickle=False)
        except Exception as exc:
            log(f"    skipped non-memmappable or nonnumeric file: {os.path.basename(path)} | {exc}")
            continue

        if X.ndim != 2:
            log(f"    skipped because ndim={X.ndim}")
            continue

        original_shape = tuple(X.shape)

        try:
            if X.shape[0] >= X.shape[1]:
                n_samples = X.shape[0]
                n_features = X.shape[1]

                if n_samples < 10 or n_features < 2:
                    log(f"    skipped bad shape: {original_shape}")
                    continue

                reservoir_n = min(max(reservoir_points, max_points), n_samples)

                if n_samples > reservoir_n:
                    reservoir_idx = np.sort(rng.choice(n_samples, size=reservoir_n, replace=False))
                else:
                    reservoir_idx = np.arange(n_samples)

                X_reservoir = np.asarray(X[reservoir_idx, :], dtype=np.float64)

            else:
                n_samples = X.shape[1]
                n_features = X.shape[0]

                if n_samples < 10 or n_features < 2:
                    log(f"    skipped bad transposed shape: {original_shape}")
                    continue

                reservoir_n = min(max(reservoir_points, max_points), n_samples)

                if n_samples > reservoir_n:
                    reservoir_idx = np.sort(rng.choice(n_samples, size=reservoir_n, replace=False))
                else:
                    reservoir_idx = np.arange(n_samples)

                X_reservoir = np.asarray(X[:, reservoir_idx].T, dtype=np.float64)

        except Exception as exc:
            log(f"    failed while reservoir-sampling {os.path.basename(path)} | {exc}")
            continue

        X_reservoir = X_reservoir[np.all(np.isfinite(X_reservoir), axis=1)]

        if X_reservoir.shape[0] < 10 or X_reservoir.shape[1] < 2:
            log(f"    reservoir data unusable: reservoir_shape={X_reservoir.shape}")
            continue

        X_reservoir_std = standardize_matrix_for_geometry(
            X_reservoir,
            max_dims=max_dims,
            seed=seed,
        )

        if X_reservoir_std is None:
            log("    standardized reservoir unusable")
            continue

        if USE_GEOMETRY_PRESERVING_SUBSAMPLE:
            fps_idx = farthest_point_sample(
                X_reservoir_std,
                n_samples=min(max_points, X_reservoir_std.shape[0]),
                seed=seed,
            )
            X_small = X_reservoir_std[fps_idx]
            method = "reservoir_plus_farthest_point"
        else:
            sample_n = min(max_points, X_reservoir_std.shape[0])
            idx = np.sort(rng.choice(X_reservoir_std.shape[0], size=sample_n, replace=False))
            X_small = X_reservoir_std[idx]
            method = "reservoir_random"

        X_small = X_small[np.all(np.isfinite(X_small), axis=1)]

        if X_small.shape[0] >= 10 and X_small.shape[1] >= 2:
            return X_small, path, original_shape, tuple(X_reservoir.shape), method

        log(f"    sampled data unusable: sampled_shape={X_small.shape}")

    return None, None, None, None, None


def standardize_point_cloud_small(X, max_dims=3):
    X = np.asarray(X, dtype=np.float64)
    X = X[np.all(np.isfinite(X), axis=1)]

    if X.shape[0] < 10:
        return None

    X = X - np.mean(X, axis=0, keepdims=True)

    if X.shape[1] > max_dims:
        _U, _S, Vt = np.linalg.svd(X, full_matrices=False)
        X = X @ Vt[:max_dims].T

    X = X / (np.std(X, axis=0, keepdims=True) + 1e-12)

    D = pdist(X)
    positive = D[D > 0]

    if len(positive) > 0:
        X = X / (np.nanmedian(positive) + 1e-12)

    return X


def estimate_complex_counts(X, radius):
    D = squareform(pdist(X))
    A = (D <= radius) & (D > 0)

    n_edges = int(np.sum(A) // 2)

    if A.shape[0] <= 300:
        A_int = A.astype(np.int64)
        n_triangles = int(np.trace(A_int @ A_int @ A_int) // 6)
    else:
        n_triangles = np.nan

    return n_edges, n_triangles


def run_h1_ripser(X, radius):
    result = ripser(
        X,
        maxdim=1,
        thresh=float(radius),
        coeff=2,
        do_cocycles=False,
    )

    diagrams = result["dgms"]

    h0 = diagrams[0] if len(diagrams) > 0 else np.empty((0, 2), dtype=float)
    h1 = diagrams[1] if len(diagrams) > 1 else np.empty((0, 2), dtype=float)

    return h0, h1


def split_finite_essential(diagram):
    diagram = np.asarray(diagram, dtype=float)

    if diagram.size == 0:
        return np.empty((0, 2), dtype=float), np.empty((0, 2), dtype=float)

    finite = diagram[
        np.isfinite(diagram[:, 0])
        & np.isfinite(diagram[:, 1])
        & (diagram[:, 1] > diagram[:, 0])
    ]

    essential = diagram[
        np.isfinite(diagram[:, 0])
        & np.isinf(diagram[:, 1])
    ]

    return finite, essential


def clean_diagram_for_vectorization(
    finite_intervals,
    essential_intervals=None,
    cap_essential=True,
    cap_value=None,
):
    cleaned = []
    capped_count = 0

    if finite_intervals is not None:
        for b, d in finite_intervals:
            if np.isfinite(b) and np.isfinite(d) and d > b:
                cleaned.append((float(b), float(d)))

    if cap_essential and essential_intervals is not None and cap_value is not None:
        for b, _d in essential_intervals:
            if np.isfinite(b) and np.isfinite(cap_value) and cap_value > b:
                cleaned.append((float(b), float(cap_value)))
                capped_count += 1

    if len(cleaned) == 0:
        return np.empty((0, 2), dtype=float), capped_count

    return np.asarray(cleaned, dtype=float), capped_count


def diagram_to_birth_persistence(diagram):
    diagram = np.asarray(diagram, dtype=float)

    if diagram.size == 0 or len(diagram) == 0:
        return np.empty((0, 2), dtype=float)

    bp = np.column_stack([diagram[:, 0], diagram[:, 1] - diagram[:, 0]])
    bp = bp[
        np.isfinite(bp[:, 0])
        & np.isfinite(bp[:, 1])
        & (bp[:, 1] > 0)
    ]

    return bp


def diagram_features(diagram):
    diagram = np.asarray(diagram, dtype=float)

    if diagram.size == 0 or len(diagram) == 0:
        return {
            "n_intervals_diagram": 0,
            "mean_lifetime": np.nan,
            "std_lifetime": np.nan,
            "max_lifetime": np.nan,
            "total_lifetime": 0.0,
            "mean_birth": np.nan,
            "std_birth": np.nan,
            "mean_death": np.nan,
            "std_death": np.nan,
            "max_death": np.nan,
            "persistence_entropy": 0.0,
        }

    lifetimes = diagram[:, 1] - diagram[:, 0]
    lifetime_sum = float(np.nansum(lifetimes))

    if lifetime_sum > 0:
        p = lifetimes / (lifetime_sum + 1e-12)
        p = p[p > 0]
        entropy = float(-np.sum(p * np.log(p + 1e-12)))
    else:
        entropy = 0.0

    return {
        "n_intervals_diagram": int(len(diagram)),
        "mean_lifetime": float(np.nanmean(lifetimes)),
        "std_lifetime": float(np.nanstd(lifetimes)),
        "max_lifetime": float(np.nanmax(lifetimes)),
        "total_lifetime": lifetime_sum,
        "mean_birth": float(np.nanmean(diagram[:, 0])),
        "std_birth": float(np.nanstd(diagram[:, 0])),
        "mean_death": float(np.nanmean(diagram[:, 1])),
        "std_death": float(np.nanstd(diagram[:, 1])),
        "max_death": float(np.nanmax(diagram[:, 1])),
        "persistence_entropy": entropy,
    }


def persistence_image_vector(bp, birth_grid, persistence_grid, sigma_birth, sigma_persistence, weight_power=1.0):
    bp = np.asarray(bp, dtype=float)

    image = np.zeros((len(persistence_grid), len(birth_grid)), dtype=float)

    if bp.size == 0 or len(bp) == 0:
        return image.ravel()

    births = bp[:, 0]
    persistences = bp[:, 1]

    weights = persistences ** weight_power

    for birth, persistence, weight in zip(births, persistences, weights):
        bx = np.exp(-0.5 * ((birth_grid - birth) / (sigma_birth + 1e-12)) ** 2)
        py = np.exp(-0.5 * ((persistence_grid - persistence) / (sigma_persistence + 1e-12)) ** 2)
        image += weight * np.outer(py, bx)

    return image.ravel()


def persistence_histogram_vector(bp, birth_edges, persistence_edges):
    bp = np.asarray(bp, dtype=float)

    if bp.size == 0 or len(bp) == 0:
        return np.zeros((len(birth_edges) - 1) * (len(persistence_edges) - 1), dtype=float)

    H, _, _ = np.histogram2d(
        bp[:, 0],
        bp[:, 1],
        bins=[birth_edges, persistence_edges],
    )

    return H.T.ravel()


def betti_curve_vector(diagram, grid):
    diagram = np.asarray(diagram, dtype=float)
    curve = np.zeros(len(grid), dtype=float)

    if diagram.size == 0 or len(diagram) == 0:
        return curve

    births = diagram[:, 0]
    deaths = diagram[:, 1]

    for b, d in zip(births, deaths):
        curve += ((grid >= b) & (grid < d)).astype(float)

    return curve


def persistence_landscape_vector(diagram, grid, n_levels=3):
    diagram = np.asarray(diagram, dtype=float)

    landscapes = np.zeros((n_levels, len(grid)), dtype=float)

    if diagram.size == 0 or len(diagram) == 0:
        return landscapes.ravel()

    tents = []

    for b, d in diagram:
        if not np.isfinite(b) or not np.isfinite(d) or d <= b:
            continue

        tent = np.maximum(0.0, np.minimum(grid - b, d - grid))
        tents.append(tent)

    if len(tents) == 0:
        return landscapes.ravel()

    tent_matrix = np.vstack(tents)

    for idx in range(len(grid)):
        vals = np.sort(tent_matrix[:, idx])[::-1]

        for level in range(min(n_levels, len(vals))):
            landscapes[level, idx] = vals[level]

    return landscapes.ravel()


def sorted_lifetime_vector(diagram, top_k=12):
    diagram = np.asarray(diagram, dtype=float)
    v = np.zeros(top_k, dtype=float)

    if diagram.size == 0 or len(diagram) == 0:
        return v

    lifetimes = diagram[:, 1] - diagram[:, 0]
    lifetimes = lifetimes[np.isfinite(lifetimes) & (lifetimes > 0)]
    lifetimes = np.sort(lifetimes)[::-1]

    take = min(top_k, len(lifetimes))
    v[:take] = lifetimes[:take]

    return v


def moment_vector(diagram):
    diagram = np.asarray(diagram, dtype=float)

    if diagram.size == 0 or len(diagram) == 0:
        return np.zeros(14, dtype=float)

    births = diagram[:, 0]
    deaths = diagram[:, 1]
    lifetimes = deaths - births
    midpoints = 0.5 * (births + deaths)

    def safe_stats(x):
        x = np.asarray(x, dtype=float)
        x = x[np.isfinite(x)]

        if len(x) == 0:
            return [0.0, 0.0, 0.0, 0.0]

        return [
            float(np.mean(x)),
            float(np.std(x)),
            float(np.min(x)),
            float(np.max(x)),
        ]

    lifetime_sum = float(np.sum(lifetimes)) if len(lifetimes) else 0.0

    if lifetime_sum > 0:
        p = lifetimes / (lifetime_sum + 1e-12)
        entropy = float(-np.sum(p[p > 0] * np.log(p[p > 0] + 1e-12)))
    else:
        entropy = 0.0

    out = []
    out.extend(safe_stats(births))
    out.extend(safe_stats(deaths))
    out.extend(safe_stats(lifetimes))
    out.extend(safe_stats(midpoints))
    out.extend([
        float(len(diagram)),
        lifetime_sum,
        entropy,
    ])

    return np.asarray(out, dtype=float)


def build_feature_names():
    names = []

    for yi in range(PERSISTENCE_BINS):
        for xi in range(BIRTH_BINS):
            names.append(f"hist_birth_{xi:02d}_pers_{yi:02d}")

    for yi in range(PERSISTENCE_BINS):
        for xi in range(BIRTH_BINS):
            names.append(f"pimg_birth_{xi:02d}_pers_{yi:02d}")

    for i in range(BETTI_GRID_SIZE):
        names.append(f"betti_{i:03d}")

    for level in range(LANDSCAPE_LEVELS):
        for i in range(LANDSCAPE_GRID_SIZE):
            names.append(f"landscape_l{level + 1}_{i:03d}")

    for i in range(TOP_LIFETIMES):
        names.append(f"top_lifetime_{i + 1:02d}")

    moment_names = [
        "birth_mean", "birth_std", "birth_min", "birth_max",
        "death_mean", "death_std", "death_min", "death_max",
        "lifetime_mean", "lifetime_std", "lifetime_min", "lifetime_max",
        "midpoint_mean", "midpoint_std", "midpoint_min", "midpoint_max",
        "n_intervals", "total_lifetime", "persistence_entropy",
    ]

    names.extend(moment_names)

    return names


def build_channel_vector(
    diagram,
    bp,
    birth_edges,
    persistence_edges,
    birth_grid,
    persistence_grid,
    filtration_grid,
    landscape_grid,
    sigma_birth,
    sigma_persistence,
):
    hist_v = persistence_histogram_vector(bp, birth_edges, persistence_edges)

    image_v = persistence_image_vector(
        bp,
        birth_grid,
        persistence_grid,
        sigma_birth=sigma_birth,
        sigma_persistence=sigma_persistence,
        weight_power=IMAGE_WEIGHT_POWER,
    )

    betti_v = betti_curve_vector(diagram, filtration_grid)

    landscape_v = persistence_landscape_vector(
        diagram,
        landscape_grid,
        n_levels=LANDSCAPE_LEVELS,
    )

    top_life_v = sorted_lifetime_vector(
        diagram,
        top_k=TOP_LIFETIMES,
    )

    moment_v = moment_vector(diagram)

    return np.concatenate([
        hist_v,
        image_v,
        betti_v,
        landscape_v,
        top_life_v,
        moment_v,
    ])


def standardize_features(X):
    X = np.asarray(X, dtype=float)
    X = np.nan_to_num(X, nan=0.0, posinf=0.0, neginf=0.0)

    mu = np.mean(X, axis=0, keepdims=True)
    sd = np.std(X, axis=0, keepdims=True)

    Xz = (X - mu) / (sd + 1e-12)
    Xz = np.nan_to_num(Xz, nan=0.0, posinf=0.0, neginf=0.0)

    return Xz, mu.ravel(), sd.ravel()


def pairwise_euclidean_matrix(X):
    if len(X) == 0:
        return np.empty((0, 0), dtype=float)

    return squareform(pdist(X, metric="euclidean"))


def cosine_similarity_matrix(X):
    X = np.asarray(X, dtype=float)

    if len(X) == 0:
        return np.empty((0, 0), dtype=float)

    norms = np.linalg.norm(X, axis=1, keepdims=True)
    Xn = X / (norms + 1e-12)

    C = Xn @ Xn.T
    C = np.clip(C, -1.0, 1.0)

    return C


def cosine_distance_from_similarity(C):
    C = np.asarray(C, dtype=float)
    D = 1.0 - C
    D = np.maximum(D, 0.0)
    np.fill_diagonal(D, 0.0)

    return D


def classical_mds_from_distances(D, n_components=2):
    D = np.asarray(D, dtype=float)
    n = D.shape[0]

    if n < 2:
        return np.zeros((n, n_components), dtype=float)

    D2 = D ** 2
    J = np.eye(n) - np.ones((n, n)) / n
    B = -0.5 * J @ D2 @ J

    if HAVE_SCIPY_EIGH:
        vals, vecs = eigh(B)
    else:
        vals, vecs = np.linalg.eigh(B)

    idx = np.argsort(vals)[::-1]
    vals = vals[idx]
    vecs = vecs[:, idx]

    coords = np.zeros((n, n_components), dtype=float)

    for k in range(min(n_components, n)):
        if vals[k] > 0:
            coords[:, k] = vecs[:, k] * np.sqrt(vals[k])

    return coords


def pca_from_feature_matrix(X, n_components=2):
    X = np.asarray(X, dtype=float)

    if X.shape[0] == 0:
        return np.empty((0, n_components)), np.zeros(n_components), np.zeros((n_components, X.shape[1]))

    Xc = X - np.mean(X, axis=0, keepdims=True)

    U, S, Vt = np.linalg.svd(Xc, full_matrices=False)

    n_take = min(n_components, U.shape[1])

    coords = np.zeros((X.shape[0], n_components), dtype=float)
    coords[:, :n_take] = U[:, :n_take] * S[:n_take]

    var = S ** 2
    total = np.sum(var)

    explained = np.zeros(n_components, dtype=float)

    if total > 0:
        explained[:n_take] = var[:n_take] / total

    components = np.zeros((n_components, X.shape[1]), dtype=float)
    components[:n_take] = Vt[:n_take]

    return coords, explained, components


log("\n" + "=" * 72)
log(f"Running H1 Persistence Diagram Vectorization | {FAST_OPTION.upper()}")
log(f"Channels: {len(eeg_channels)}")
log(f"max_points={MAX_POINTS}, reservoir_points={RESERVOIR_POINTS}, radius_quantile={RADIUS_QUANTILE}")
log(f"birth_bins={BIRTH_BINS}, persistence_bins={PERSISTENCE_BINS}")
log(f"betti_grid_size={BETTI_GRID_SIZE}, landscape_grid_size={LANDSCAPE_GRID_SIZE}, landscape_levels={LANDSCAPE_LEVELS}")
log(f"top_lifetimes={TOP_LIFETIMES}, image_sigma_frac={IMAGE_SIGMA_FRAC}")
log(f"geometry_preserving_subsample={USE_GEOMETRY_PRESERVING_SUBSAMPLE}")
log(f"standardize_features_for_distance={STANDARDIZE_FEATURES_FOR_DISTANCE}")
log(f"cap_essential_h1={CAP_ESSENTIAL_H1}")
log(f"expert_plots={MAKE_EXPERT_PLOTS}, render_all_images={RENDER_ALL_IMAGES_INLINE}")
log(f"Output: {mode_out_dir}")
log("=" * 72)

diagnostics = {}
diagram_store = {}
bp_store = {}
raw_interval_store = {}
loaded_channels = []

global_birth_max = None
global_persistence_max = None
global_filtration_max = None

overall_t0 = time.time()

for ch_idx, ch_name in enumerate(eeg_channels, start=1):
    t0 = time.time()
    prefix = f"[{ch_idx:02d}/{len(eeg_channels)} {ch_name}]"

    log(f"\n{prefix} loading reservoir + geometry-preserving subsampling")

    X_small, source_path, original_shape, reservoir_shape, sampling_method = load_channel_embedding_geometry_sampled(
        ch_name,
        max_points=MAX_POINTS,
        reservoir_points=RESERVOIR_POINTS,
        max_dims=MAX_DIMS,
        seed=1100 + ch_idx,
    )

    if X_small is None:
        log(f"{prefix} WARNING: no usable embedding found; skipping")
        continue

    log(
        f"{prefix} source shape={original_shape}, reservoir shape={reservoir_shape}, "
        f"sampled shape={X_small.shape}, method={sampling_method}"
    )

    X = standardize_point_cloud_small(X_small, max_dims=MAX_DIMS)

    if X is None:
        log(f"{prefix} WARNING: invalid sampled embedding; skipping")
        continue

    log(f"{prefix} standardized shape={X.shape}; computing radius")

    D_flat = pdist(X)
    positive = D_flat[D_flat > 0]

    if len(positive) == 0:
        log(f"{prefix} WARNING: zero-distance cloud; skipping")
        continue

    radius = float(np.quantile(positive, RADIUS_QUANTILE))

    log(f"{prefix} estimating threshold complex size")
    n_edges, n_triangles = estimate_complex_counts(X, radius)

    log(f"{prefix} ripser H0/H1")
    h0_diag, h1_diag = run_h1_ripser(X, radius)

    h0_finite, h0_essential = split_finite_essential(h0_diag)
    h1_finite, h1_essential = split_finite_essential(h1_diag)

    h1_diagram, n_capped_h1 = clean_diagram_for_vectorization(
        h1_finite,
        essential_intervals=h1_essential,
        cap_essential=CAP_ESSENTIAL_H1,
        cap_value=radius,
    )

    bp = diagram_to_birth_persistence(h1_diagram)

    if len(bp) > 0:
        birth_max = float(np.nanmax(bp[:, 0]))
        persistence_max = float(np.nanmax(bp[:, 1]))
        death_max = float(np.nanmax(h1_diagram[:, 1]))

        global_birth_max = birth_max if global_birth_max is None else max(global_birth_max, birth_max)
        global_persistence_max = persistence_max if global_persistence_max is None else max(global_persistence_max, persistence_max)
        global_filtration_max = death_max if global_filtration_max is None else max(global_filtration_max, death_max)

    elapsed = time.time() - t0

    diagram_store[ch_name] = h1_diagram
    bp_store[ch_name] = bp

    raw_interval_store[ch_name] = {
        "H0_finite": h0_finite,
        "H0_essential": h0_essential,
        "H1_finite": h1_finite,
        "H1_essential": h1_essential,
        "H1_vectorization_diagram": h1_diagram,
        "H1_birth_persistence": bp,
    }

    diagnostics[ch_name] = {
        "channel": ch_name,
        "source_path": source_path,
        "source_shape": str(original_shape),
        "reservoir_shape": str(reservoir_shape),
        "sampling_method": sampling_method,
        "backend": "ripser_presampled_fps",
        "n_points": int(X.shape[0]),
        "n_edges_at_radius": int(n_edges),
        "n_triangles_at_radius": int(n_triangles) if np.isfinite(n_triangles) else np.nan,
        "radius": radius,
        "n_h0_finite": int(len(h0_finite)),
        "n_h1_finite": int(len(h1_finite)),
        "n_h1_essential": int(len(h1_essential)),
        "n_h1_capped_for_vectorization": int(n_capped_h1),
        "n_h1_vectorized_intervals": int(len(h1_diagram)),
        "elapsed_sec": elapsed,
        **diagram_features(h1_diagram),
    }

    loaded_channels.append(ch_name)

    log(
        f"{prefix} done {elapsed:.2f}s | "
        f"edges={n_edges}, triangles={n_triangles}, "
        f"H1 finite={len(h1_finite)}, H1 vectorized diagram={len(h1_diagram)}"
    )

if len(loaded_channels) == 0:
    raise RuntimeError("No usable embeddings found. Check embedding_data paths.")

if global_birth_max is None or not np.isfinite(global_birth_max) or global_birth_max <= 0:
    global_birth_max = 1.0

if global_persistence_max is None or not np.isfinite(global_persistence_max) or global_persistence_max <= 0:
    global_persistence_max = 1.0

if global_filtration_max is None or not np.isfinite(global_filtration_max) or global_filtration_max <= 0:
    global_filtration_max = 1.0

birth_max_for_grid = global_birth_max * 1.0001
persistence_max_for_grid = global_persistence_max * 1.0001
filtration_max_for_grid = global_filtration_max * 1.0001

birth_edges = np.linspace(0.0, birth_max_for_grid, BIRTH_BINS + 1)
persistence_edges = np.linspace(0.0, persistence_max_for_grid, PERSISTENCE_BINS + 1)

birth_grid = 0.5 * (birth_edges[:-1] + birth_edges[1:])
persistence_grid = 0.5 * (persistence_edges[:-1] + persistence_edges[1:])

filtration_grid = np.linspace(0.0, filtration_max_for_grid, BETTI_GRID_SIZE)
landscape_grid = np.linspace(0.0, filtration_max_for_grid, LANDSCAPE_GRID_SIZE)

sigma_birth = max(1e-8, IMAGE_SIGMA_FRAC * birth_max_for_grid)
sigma_persistence = max(1e-8, IMAGE_SIGMA_FRAC * persistence_max_for_grid)

log("\nBuilding fixed-length persistence diagram vectors")
log(f"birth range: 0 to {birth_max_for_grid:.6f}")
log(f"persistence range: 0 to {persistence_max_for_grid:.6f}")
log(f"filtration range: 0 to {filtration_max_for_grid:.6f}")
log(f"image sigma birth={sigma_birth:.8f}, persistence={sigma_persistence:.8f}")

feature_names = build_feature_names()
feature_rows = []

for ch_name in loaded_channels:
    v = build_channel_vector(
        diagram_store[ch_name],
        bp_store[ch_name],
        birth_edges=birth_edges,
        persistence_edges=persistence_edges,
        birth_grid=birth_grid,
        persistence_grid=persistence_grid,
        filtration_grid=filtration_grid,
        landscape_grid=landscape_grid,
        sigma_birth=sigma_birth,
        sigma_persistence=sigma_persistence,
    )

    feature_rows.append(v)

feature_matrix = np.asarray(feature_rows, dtype=float)
feature_matrix = np.nan_to_num(feature_matrix, nan=0.0, posinf=0.0, neginf=0.0)

if feature_matrix.shape[1] != len(feature_names):
    raise RuntimeError(
        f"Feature name mismatch: matrix has {feature_matrix.shape[1]} columns, "
        f"feature_names has {len(feature_names)} entries."
    )

feature_matrix_z, feature_mean, feature_std = standardize_features(feature_matrix)

matrix_for_distance = feature_matrix_z if STANDARDIZE_FEATURES_FOR_DISTANCE else feature_matrix

euclidean_distance_mat = pairwise_euclidean_matrix(matrix_for_distance)
cosine_similarity_mat = cosine_similarity_matrix(matrix_for_distance)
cosine_distance_mat = cosine_distance_from_similarity(cosine_similarity_mat)

feature_norms = np.linalg.norm(matrix_for_distance, axis=1)
raw_feature_norms = np.linalg.norm(feature_matrix, axis=1)

summary_rows = []

for idx, ch_name in enumerate(loaded_channels):
    euc = euclidean_distance_mat[idx].copy()
    cos = cosine_similarity_mat[idx].copy()

    mask = np.arange(len(loaded_channels)) != idx

    summary_rows.append({
        **diagnostics[ch_name],
        "raw_feature_norm": float(raw_feature_norms[idx]),
        "standardized_feature_norm": float(feature_norms[idx]),
        "mean_euclidean_vector_distance_to_others": float(np.nanmean(euc[mask])) if np.any(mask) else 0.0,
        "median_euclidean_vector_distance_to_others": float(np.nanmedian(euc[mask])) if np.any(mask) else 0.0,
        "max_euclidean_vector_distance_to_others": float(np.nanmax(euc[mask])) if np.any(mask) else 0.0,
        "min_euclidean_vector_distance_to_others": float(np.nanmin(euc[mask])) if np.any(mask) else 0.0,
        "mean_cosine_similarity_to_others": float(np.nanmean(cos[mask])) if np.any(mask) else 1.0,
        "median_cosine_similarity_to_others": float(np.nanmedian(cos[mask])) if np.any(mask) else 1.0,
        "max_cosine_similarity_to_others": float(np.nanmax(cos[mask])) if np.any(mask) else 1.0,
        "min_cosine_similarity_to_others": float(np.nanmin(cos[mask])) if np.any(mask) else 1.0,
    })

summary_df = (
    pd.DataFrame(summary_rows)
    .sort_values("mean_euclidean_vector_distance_to_others", ascending=False)
    .reset_index(drop=True)
)

feature_df = pd.DataFrame(feature_matrix, index=loaded_channels, columns=feature_names)
feature_z_df = pd.DataFrame(feature_matrix_z, index=loaded_channels, columns=feature_names)

euclidean_df = pd.DataFrame(euclidean_distance_mat, index=loaded_channels, columns=loaded_channels)
cosine_similarity_df = pd.DataFrame(cosine_similarity_mat, index=loaded_channels, columns=loaded_channels)
cosine_distance_df = pd.DataFrame(cosine_distance_mat, index=loaded_channels, columns=loaded_channels)

pair_rows = []

for i in range(len(loaded_channels)):
    for j in range(i + 1, len(loaded_channels)):
        pair_rows.append({
            "channel_a": loaded_channels[i],
            "channel_b": loaded_channels[j],
            "euclidean_vector_distance": float(euclidean_distance_mat[i, j]),
            "cosine_similarity": float(cosine_similarity_mat[i, j]),
            "cosine_distance": float(cosine_distance_mat[i, j]),
        })

pairwise_df = (
    pd.DataFrame(pair_rows)
    .sort_values("euclidean_vector_distance", ascending=False)
    .reset_index(drop=True)
)

feature_variance = np.var(feature_matrix_z, axis=0)
feature_abs_mean = np.mean(np.abs(feature_matrix_z), axis=0)

feature_importance_df = pd.DataFrame({
    "feature_name": feature_names,
    "standardized_variance": feature_variance,
    "mean_abs_standardized_value": feature_abs_mean,
}).sort_values("standardized_variance", ascending=False).reset_index(drop=True)

pca_coords, pca_explained, pca_components = pca_from_feature_matrix(matrix_for_distance, n_components=2)
mds_coords = classical_mds_from_distances(euclidean_distance_mat, n_components=2)

pca_df = pd.DataFrame({
    "channel": loaded_channels,
    "pc1": pca_coords[:, 0],
    "pc2": pca_coords[:, 1],
    "feature_norm": feature_norms,
    "mean_euclidean_vector_distance_to_others": [
        summary_df.loc[summary_df["channel"] == ch, "mean_euclidean_vector_distance_to_others"].values[0]
        for ch in loaded_channels
    ],
    "mean_cosine_similarity_to_others": [
        summary_df.loc[summary_df["channel"] == ch, "mean_cosine_similarity_to_others"].values[0]
        for ch in loaded_channels
    ],
    "n_intervals_diagram": [
        summary_df.loc[summary_df["channel"] == ch, "n_intervals_diagram"].values[0]
        for ch in loaded_channels
    ],
})

mds_df = pd.DataFrame({
    "channel": loaded_channels,
    "mds_1": mds_coords[:, 0],
    "mds_2": mds_coords[:, 1],
    "feature_norm": feature_norms,
    "mean_euclidean_vector_distance_to_others": [
        summary_df.loc[summary_df["channel"] == ch, "mean_euclidean_vector_distance_to_others"].values[0]
        for ch in loaded_channels
    ],
    "n_intervals_diagram": [
        summary_df.loc[summary_df["channel"] == ch, "n_intervals_diagram"].values[0]
        for ch in loaded_channels
    ],
})

summary_csv_path = os.path.join(mode_out_dir, "h1_persistence_diagram_vectorization_channel_summary.csv")
feature_csv_path = os.path.join(mode_out_dir, "h1_persistence_diagram_feature_matrix.csv")
feature_z_csv_path = os.path.join(mode_out_dir, "h1_persistence_diagram_feature_matrix_standardized.csv")
feature_importance_csv_path = os.path.join(mode_out_dir, "h1_persistence_diagram_feature_variance_ranking.csv")
euclidean_csv_path = os.path.join(mode_out_dir, "h1_persistence_diagram_vector_euclidean_distance_matrix.csv")
cosine_similarity_csv_path = os.path.join(mode_out_dir, "h1_persistence_diagram_vector_cosine_similarity_matrix.csv")
cosine_distance_csv_path = os.path.join(mode_out_dir, "h1_persistence_diagram_vector_cosine_distance_matrix.csv")
pairwise_csv_path = os.path.join(mode_out_dir, "h1_persistence_diagram_vector_pairwise_distances.csv")
pca_csv_path = os.path.join(mode_out_dir, "h1_persistence_diagram_vector_pca_coordinates.csv")
mds_csv_path = os.path.join(mode_out_dir, "h1_persistence_diagram_vector_mds_coordinates.csv")

summary_df.to_csv(summary_csv_path, index=False)
feature_df.to_csv(feature_csv_path)
feature_z_df.to_csv(feature_z_csv_path)
feature_importance_df.to_csv(feature_importance_csv_path, index=False)
euclidean_df.to_csv(euclidean_csv_path)
cosine_similarity_df.to_csv(cosine_similarity_csv_path)
cosine_distance_df.to_csv(cosine_distance_csv_path)
pairwise_df.to_csv(pairwise_csv_path, index=False)
pca_df.to_csv(pca_csv_path, index=False)
mds_df.to_csv(mds_csv_path, index=False)

npz_path = os.path.join(mode_out_dir, "h1_persistence_diagram_vectorization_outputs.npz")

save_dict = {
    "channels": np.asarray(loaded_channels),
    "feature_names": np.asarray(feature_names),
    "feature_matrix": feature_matrix,
    "feature_matrix_standardized": feature_matrix_z,
    "feature_mean": feature_mean,
    "feature_std": feature_std,
    "euclidean_distance_matrix": euclidean_distance_mat,
    "cosine_similarity_matrix": cosine_similarity_mat,
    "cosine_distance_matrix": cosine_distance_mat,
    "pca_coordinates": pca_coords,
    "pca_explained_variance_ratio": pca_explained,
    "pca_components": pca_components,
    "mds_coordinates": mds_coords,
    "birth_edges": birth_edges,
    "persistence_edges": persistence_edges,
    "birth_grid": birth_grid,
    "persistence_grid": persistence_grid,
    "filtration_grid": filtration_grid,
    "landscape_grid": landscape_grid,
    "sigma_birth": np.asarray([sigma_birth]),
    "sigma_persistence": np.asarray([sigma_persistence]),
    "fast_option": np.asarray([FAST_OPTION]),
    "backend": np.asarray(["ripser_presampled_fps + persistence diagram vectorization"]),
}

for ch_name in loaded_channels:
    save_dict[f"{ch_name}_H1_vectorization_diagram"] = diagram_store[ch_name]
    save_dict[f"{ch_name}_H1_birth_persistence"] = bp_store[ch_name]

    for key, val in raw_interval_store[ch_name].items():
        save_dict[f"{ch_name}_{key}"] = val

np.savez(npz_path, **save_dict)

print("\nH1 Persistence Diagram Vectorization Channel Summary:")
if HAVE_IPYTHON:
    display(summary_df)
else:
    print(summary_df)

print("\nLargest vector distances:")
if HAVE_IPYTHON:
    display(pairwise_df.head(12))
else:
    print(pairwise_df.head(12))

print("\nTop variable vector features:")
if HAVE_IPYTHON:
    display(feature_importance_df.head(20))
else:
    print(feature_importance_df.head(20))

log(f"\nSaved summary CSV: {summary_csv_path}")
log(f"Saved raw feature matrix CSV: {feature_csv_path}")
log(f"Saved standardized feature matrix CSV: {feature_z_csv_path}")
log(f"Saved feature variance ranking CSV: {feature_importance_csv_path}")
log(f"Saved Euclidean distance matrix CSV: {euclidean_csv_path}")
log(f"Saved cosine similarity matrix CSV: {cosine_similarity_csv_path}")
log(f"Saved cosine distance matrix CSV: {cosine_distance_csv_path}")
log(f"Saved pairwise distances CSV: {pairwise_csv_path}")
log(f"Saved PCA CSV: {pca_csv_path}")
log(f"Saved MDS CSV: {mds_csv_path}")
log(f"Saved NPZ: {npz_path}")

plot_df = summary_df.copy().sort_values("mean_euclidean_vector_distance_to_others", ascending=False).reset_index(drop=True)

fig, ax = plt.subplots(figsize=(13, 5.8), facecolor=BLACK)
style_ax(
    ax,
    title=f"Mean Persistence-Vector Euclidean Distance to Other Channels | {FAST_OPTION.upper()}",
    xlabel="Channel",
    ylabel="Mean Euclidean vector distance",
)

x_pos = np.arange(len(plot_df))
y = plot_df["mean_euclidean_vector_distance_to_others"].to_numpy(dtype=float)

ax.bar(x_pos, y, color=ACCENT, edgecolor=ACCENT, alpha=0.78)

if np.any(np.isfinite(y)):
    ax.axhline(
        np.nanmean(y),
        color=ACCENT_SOFT,
        linestyle="--",
        linewidth=1.4,
        alpha=0.8,
        label=f"Mean={np.nanmean(y):.5f}",
    )

ax.set_xticks(x_pos)
ax.set_xticklabels(plot_df["channel"], rotation=90)
ax.legend(loc="best", framealpha=1.0)

mean_distance_path = os.path.join(mode_out_dir, "01_mean_persistence_vector_distance_by_channel.png")
plt.tight_layout()
save_render_close(fig, mean_distance_path)

if len(loaded_channels) >= 3 and np.any(euclidean_distance_mat > 0):
    condensed = euclidean_distance_mat[np.triu_indices(len(loaded_channels), k=1)]
    Z = linkage(condensed, method="average")
    order = leaves_list(Z)
else:
    order = np.arange(len(loaded_channels))

ordered_channels = [loaded_channels[i] for i in order]
ordered_euclidean = euclidean_distance_mat[np.ix_(order, order)]
ordered_cosine = cosine_similarity_mat[np.ix_(order, order)]

euc_vmin, euc_vmax = dynamic_limits(
    offdiag_values(ordered_euclidean),
    lower_pct=2,
    upper_pct=98,
    pad_frac=0.10,
    lower_bound=0.0,
)

save_matrix_plot(
    ordered_euclidean,
    ordered_channels,
    "H1 Persistence Diagram Vector Euclidean Distance Matrix",
    "Euclidean vector distance",
    os.path.join(mode_out_dir, "02_h1_persistence_vector_euclidean_distance_heatmap.png"),
    vmin=euc_vmin,
    vmax=euc_vmax,
)

cos_vmin, cos_vmax = dynamic_limits(
    offdiag_values(ordered_cosine),
    lower_pct=2,
    upper_pct=98,
    pad_frac=0.10,
    lower_bound=-1.0,
    upper_bound=1.0,
)

save_matrix_plot(
    ordered_cosine,
    ordered_channels,
    "H1 Persistence Diagram Vector Cosine Similarity Matrix",
    "Cosine similarity",
    os.path.join(mode_out_dir, "03_h1_persistence_vector_cosine_similarity_heatmap.png"),
    vmin=cos_vmin,
    vmax=cos_vmax,
)

fig, ax = plt.subplots(figsize=(8, 7), facecolor=BLACK)
style_ax(
    ax,
    title=f"PCA of H1 Persistence Diagram Vectors | PC1={pca_explained[0]*100:.1f}%, PC2={pca_explained[1]*100:.1f}%",
    xlabel="PC1",
    ylabel="PC2",
)

pca_color = pca_df["mean_euclidean_vector_distance_to_others"].to_numpy(dtype=float)
pca_vmin, pca_vmax = dynamic_limits(pca_color, lower_pct=2, upper_pct=98, pad_frac=0.08, lower_bound=0.0)

sc = ax.scatter(
    pca_df["pc1"],
    pca_df["pc2"],
    c=pca_color,
    s=150,
    cmap=CYAN_SEQUENTIAL_CMAP,
    edgecolor=ACCENT,
    linewidth=1.1,
    vmin=pca_vmin,
    vmax=pca_vmax,
)

for _, row in pca_df.iterrows():
    ax.text(
        row["pc1"],
        row["pc2"],
        row["channel"],
        fontsize=8,
        color=ACCENT,
        alpha=0.9,
    )

cbar = plt.colorbar(sc, ax=ax, shrink=0.85)
style_colorbar(cbar, "Mean vector distance")

pca_path = os.path.join(mode_out_dir, "04_h1_persistence_vector_pca_embedding.png")
plt.tight_layout()
save_render_close(fig, pca_path)

fig, ax = plt.subplots(figsize=(8, 7), facecolor=BLACK)
style_ax(
    ax,
    title="Classical MDS of H1 Persistence Diagram Vector Distances",
    xlabel="MDS 1",
    ylabel="MDS 2",
)

mds_color = mds_df["mean_euclidean_vector_distance_to_others"].to_numpy(dtype=float)
mds_vmin, mds_vmax = dynamic_limits(mds_color, lower_pct=2, upper_pct=98, pad_frac=0.08, lower_bound=0.0)

sc = ax.scatter(
    mds_df["mds_1"],
    mds_df["mds_2"],
    c=mds_color,
    s=150,
    cmap=CYAN_SEQUENTIAL_CMAP,
    edgecolor=ACCENT,
    linewidth=1.1,
    vmin=mds_vmin,
    vmax=mds_vmax,
)

for _, row in mds_df.iterrows():
    ax.text(
        row["mds_1"],
        row["mds_2"],
        row["channel"],
        fontsize=8,
        color=ACCENT,
        alpha=0.9,
    )

cbar = plt.colorbar(sc, ax=ax, shrink=0.85)
style_colorbar(cbar, "Mean vector distance")

mds_path = os.path.join(mode_out_dir, "05_h1_persistence_vector_mds_embedding.png")
plt.tight_layout()
save_render_close(fig, mds_path)

top_n_features = min(40, len(feature_importance_df))
top_features = feature_importance_df.head(top_n_features)["feature_name"].tolist()

top_feature_matrix = feature_z_df.loc[loaded_channels, top_features].to_numpy(dtype=float)

fig, ax = plt.subplots(figsize=(14, 8), facecolor=BLACK)
style_ax(
    ax,
    title=f"Top {top_n_features} Variable Persistence-Vector Features",
    xlabel="Feature",
    ylabel="Channel",
)

vlim = np.nanpercentile(np.abs(top_feature_matrix), 98) if np.any(np.isfinite(top_feature_matrix)) else 1.0
vlim = max(float(vlim), 1e-6)

im = ax.imshow(
    top_feature_matrix,
    aspect="auto",
    interpolation="nearest",
    cmap=CYAN_SEQUENTIAL_CMAP,
    vmin=-vlim,
    vmax=vlim,
)

ax.set_yticks(np.arange(len(loaded_channels)))
ax.set_yticklabels(loaded_channels)
ax.set_xticks(np.arange(top_n_features))
ax.set_xticklabels(top_features, rotation=90, fontsize=7)

cbar = plt.colorbar(im, ax=ax, shrink=0.85)
style_colorbar(cbar, "Standardized feature value")

top_feature_path = os.path.join(mode_out_dir, "06_top_variable_persistence_vector_features_heatmap.png")
plt.tight_layout()
save_render_close(fig, top_feature_path, width=1200)

fig, ax = plt.subplots(figsize=(12, 6), facecolor=BLACK)
style_ax(
    ax,
    title="Top Persistence-Vector Feature Variances",
    xlabel="Feature",
    ylabel="Variance after standardization",
)

variance_plot = feature_importance_df.head(min(25, len(feature_importance_df))).copy()
x_var = np.arange(len(variance_plot))

ax.bar(
    x_var,
    variance_plot["standardized_variance"],
    color=ACCENT,
    edgecolor=ACCENT,
    alpha=0.78,
)

ax.set_xticks(x_var)
ax.set_xticklabels(variance_plot["feature_name"], rotation=75, ha="right", fontsize=8)

variance_path = os.path.join(mode_out_dir, "07_top_persistence_vector_feature_variances.png")
plt.tight_layout()
save_render_close(fig, variance_path)

diagram_channels = summary_df.head(min(12, len(summary_df)))["channel"].tolist()
n_cols = 4
n_rows = int(np.ceil(len(diagram_channels) / n_cols))

fig, axes = plt.subplots(
    n_rows,
    n_cols,
    figsize=(4 * n_cols, 3.2 * n_rows),
    facecolor=BLACK,
)

axes = np.asarray(axes).reshape(-1)

all_diagram_points = np.vstack([
    diagram_store[ch] for ch in loaded_channels if len(diagram_store[ch]) > 0
]) if any(len(diagram_store[ch]) > 0 for ch in loaded_channels) else np.empty((0, 2))

if len(all_diagram_points) > 0:
    diag_max = float(np.nanmax(all_diagram_points[:, 1]))
else:
    diag_max = 1.0

diag_max = max(diag_max, 1e-6)

for ax_idx, ch_name in enumerate(diagram_channels):
    ax = axes[ax_idx]
    D = diagram_store[ch_name]

    mean_dist = summary_df.loc[
        summary_df["channel"] == ch_name,
        "mean_euclidean_vector_distance_to_others",
    ].values[0]

    style_ax(
        ax,
        title=f"{ch_name}: mean d={mean_dist:.2f}, n={len(D)}",
        xlabel="Birth",
        ylabel="Death",
    )

    ax.plot([0, diag_max], [0, diag_max], color=ACCENT_DIM, linestyle="--", linewidth=1.0)

    if len(D) > 0:
        ax.scatter(
            D[:, 0],
            D[:, 1],
            s=55,
            facecolors=BLACK,
            edgecolors=ACCENT,
            linewidths=1.2,
        )

    ax.set_xlim(0, diag_max * 1.05)
    ax.set_ylim(0, diag_max * 1.05)

for j in range(len(diagram_channels), len(axes)):
    axes[j].axis("off")

diagrams_path = os.path.join(mode_out_dir, "08_top_channels_h1_persistence_diagrams_by_vector_distance.png")
plt.tight_layout()
save_render_close(fig, diagrams_path, width=1100)

fig, ax = plt.subplots(figsize=(12, 6), facecolor=BLACK)
style_ax(
    ax,
    title="Largest Pairwise H1 Persistence-Vector Distances",
    xlabel="Channel pair",
    ylabel="Euclidean vector distance",
)

top_pairs = pairwise_df.head(min(15, len(pairwise_df))).copy()
pair_labels = top_pairs["channel_a"] + "-" + top_pairs["channel_b"]

x_pair = np.arange(len(top_pairs))

ax.bar(
    x_pair,
    top_pairs["euclidean_vector_distance"],
    color=ACCENT,
    edgecolor=ACCENT,
    alpha=0.78,
)

ax.set_xticks(x_pair)
ax.set_xticklabels(pair_labels, rotation=60, ha="right")

top_pairs_path = os.path.join(mode_out_dir, "09_largest_pairwise_persistence_vector_distances.png")
plt.tight_layout()
save_render_close(fig, top_pairs_path)

fig, ax = plt.subplots(figsize=(8, 6), facecolor=BLACK)
style_ax(
    ax,
    title="Feature Norm vs Mean Persistence-Vector Distance",
    xlabel="Standardized feature norm",
    ylabel="Mean Euclidean vector distance to others",
)

sc = ax.scatter(
    summary_df["standardized_feature_norm"],
    summary_df["mean_euclidean_vector_distance_to_others"],
    c=summary_df["n_intervals_diagram"],
    s=140,
    cmap=CYAN_SEQUENTIAL_CMAP,
    edgecolor=ACCENT,
    linewidth=1.0,
)

for _, row in summary_df.iterrows():
    ax.text(
        row["standardized_feature_norm"],
        row["mean_euclidean_vector_distance_to_others"],
        row["channel"],
        fontsize=8,
        color=ACCENT,
        alpha=0.85,
    )

cbar = plt.colorbar(sc, ax=ax, shrink=0.85)
style_colorbar(cbar, "H1 diagram intervals")

norm_path = os.path.join(mode_out_dir, "10_feature_norm_vs_mean_vector_distance.png")
plt.tight_layout()
save_render_close(fig, norm_path)

if MAKE_EXPERT_PLOTS:
    expert_dir = os.path.join(mode_out_dir, "expert_plots")
    os.makedirs(expert_dir, exist_ok=True)

    mean_vector_distance_map = dict(zip(summary_df["channel"], summary_df["mean_euclidean_vector_distance_to_others"]))

    scalp_xy = {
        "Fp1": (-0.45, 1.00), "Fpz": (0.00, 1.08), "Fp2": (0.45, 1.00),
        "F7": (-0.95, 0.60), "F3": (-0.45, 0.55), "Fz": (0.00, 0.60),
        "F4": (0.45, 0.55), "F8": (0.95, 0.60),
        "FC5": (-0.75, 0.25), "FC1": (-0.25, 0.25),
        "FC2": (0.25, 0.25), "FC6": (0.75, 0.25),
        "M1": (-1.15, 0.05), "T7": (-1.00, 0.00), "C3": (-0.50, 0.00),
        "Cz": (0.00, 0.00), "C4": (0.50, 0.00), "T8": (1.00, 0.00),
        "M2": (1.15, 0.05),
        "CP5": (-0.75, -0.30), "CP1": (-0.25, -0.30),
        "CP2": (0.25, -0.30), "CP6": (0.75, -0.30),
        "P7": (-0.90, -0.60), "P3": (-0.45, -0.60), "Pz": (0.00, -0.65),
        "P4": (0.45, -0.60), "P8": (0.90, -0.60),
        "POz": (0.00, -0.88),
        "O1": (-0.35, -1.00), "Oz": (0.00, -1.08), "O2": (0.35, -1.00),
    }

    xs, ys, vals, labels = [], [], [], []

    for ch in eeg_channels:
        if ch in scalp_xy and ch in mean_vector_distance_map and np.isfinite(mean_vector_distance_map[ch]):
            xs.append(scalp_xy[ch][0])
            ys.append(scalp_xy[ch][1])
            vals.append(mean_vector_distance_map[ch])
            labels.append(ch)

    scalp_vmin, scalp_vmax = dynamic_limits(
        vals,
        lower_pct=2,
        upper_pct=98,
        pad_frac=0.10,
        lower_bound=0.0,
    )

    fig, ax = plt.subplots(figsize=(7, 7), facecolor=BLACK)
    style_ax(
        ax,
        title=f"Scalp Map of Mean H1 Persistence-Vector Distance | {FAST_OPTION.upper()}",
        grid=False,
    )

    head = plt.Circle((0, 0), 1.12, fill=False, color=ACCENT, linewidth=2)
    ax.add_patch(head)
    ax.plot([0.00, -0.10, 0.10, 0.00], [1.12, 1.28, 1.28, 1.12], color=ACCENT, linewidth=2)

    sc = ax.scatter(
        xs,
        ys,
        c=vals,
        s=260,
        cmap=CYAN_SEQUENTIAL_CMAP,
        edgecolor=ACCENT,
        linewidth=1.2,
        vmin=scalp_vmin,
        vmax=scalp_vmax,
    )

    for x0, y0, label in zip(xs, ys, labels):
        ax.text(
            x0,
            y0,
            label,
            ha="center",
            va="center",
            fontsize=8,
            color=BLACK,
            fontweight="bold",
        )

    cbar = plt.colorbar(sc, ax=ax, shrink=0.75)
    style_colorbar(cbar, "Mean vector distance")

    ax.set_xlim(-1.35, 1.35)
    ax.set_ylim(-1.35, 1.35)
    ax.set_aspect("equal")
    ax.set_xticks([])
    ax.set_yticks([])

    scalp_path = os.path.join(expert_dir, "11_mean_h1_persistence_vector_distance_scalp_map.png")
    plt.tight_layout()
    save_render_close(fig, scalp_path, width=760)

    regions = {
        "Frontal": ["Fp1", "Fpz", "Fp2", "F7", "F3", "Fz", "F4", "F8"],
        "Frontocentral": ["FC5", "FC1", "FC2", "FC6"],
        "Central": ["T7", "C3", "Cz", "C4", "T8"],
        "Centroparietal": ["CP5", "CP1", "CP2", "CP6"],
        "Parietal": ["P7", "P3", "Pz", "P4", "P8"],
        "Occipital": ["POz", "O1", "Oz", "O2"],
        "Mastoid": ["M1", "M2"],
    }

    region_rows = []
    region_points = []

    for region_name, chans in regions.items():
        vals_region = []

        for ch in chans:
            if ch in mean_vector_distance_map and np.isfinite(mean_vector_distance_map[ch]):
                vals_region.append(mean_vector_distance_map[ch])
                region_points.append({
                    "region": region_name,
                    "channel": ch,
                    "mean_euclidean_vector_distance_to_others": mean_vector_distance_map[ch],
                })

        if len(vals_region) == 0:
            continue

        vals_region = np.asarray(vals_region, dtype=float)

        region_rows.append({
            "region": region_name,
            "mean_region_vector_distance": float(np.mean(vals_region)),
            "std_region_vector_distance": float(np.std(vals_region, ddof=1)) if len(vals_region) > 1 else 0.0,
            "sem_region_vector_distance": (
                float(np.std(vals_region, ddof=1) / np.sqrt(len(vals_region)))
                if len(vals_region) > 1 else 0.0
            ),
            "n_channels": int(len(vals_region)),
        })

    region_df = pd.DataFrame(region_rows)
    region_points_df = pd.DataFrame(region_points)

    region_csv_path = os.path.join(expert_dir, "regional_h1_persistence_diagram_vectorization_summary.csv")
    region_df.to_csv(region_csv_path, index=False)

    if len(region_df) > 0:
        fig, ax = plt.subplots(figsize=(12, 6.5), facecolor=BLACK)
        style_ax(
            ax,
            title=f"Regional Mean H1 Persistence-Vector Distance +/- SEM | {FAST_OPTION.upper()}",
            xlabel="Region",
            ylabel="Mean Euclidean vector distance",
        )

        x_region = np.arange(len(region_df))

        ax.bar(
            x_region,
            region_df["mean_region_vector_distance"],
            yerr=region_df["sem_region_vector_distance"],
            color=ACCENT,
            edgecolor=ACCENT,
            alpha=0.82,
            width=0.68,
            capsize=8,
            error_kw={
                "elinewidth": 2.2,
                "capthick": 2.2,
                "ecolor": ACCENT,
                "alpha": 1.0,
            },
            label="Mean +/- SEM",
        )

        rng = np.random.default_rng(7)

        for i, region_name in enumerate(region_df["region"]):
            vals_region_plot = region_points_df.loc[
                region_points_df["region"] == region_name,
                "mean_euclidean_vector_distance_to_others",
            ].to_numpy(dtype=float)

            if len(vals_region_plot) == 0:
                continue

            jitter = rng.uniform(-0.13, 0.13, size=len(vals_region_plot))

            ax.scatter(
                np.full(len(vals_region_plot), i) + jitter,
                vals_region_plot,
                s=42,
                facecolors=BLACK,
                edgecolors=ACCENT,
                linewidths=1.2,
                alpha=0.95,
                zorder=4,
            )

        for i, row in region_df.iterrows():
            y_val = row["mean_region_vector_distance"]
            sem = row["sem_region_vector_distance"]

            ax.text(
                i,
                y_val + sem,
                f"{y_val:.3f}\n+/-{sem:.3f}\nn={int(row['n_channels'])}",
                ha="center",
                va="bottom",
                fontsize=8,
                color=ACCENT,
            )

        ax.set_xticks(x_region)
        ax.set_xticklabels(
            region_df["region"],
            rotation=35,
            ha="right",
            rotation_mode="anchor",
        )
        ax.margins(x=0.04)
        ax.legend(loc="best", framealpha=1.0)

        region_path = os.path.join(expert_dir, "12_regional_h1_persistence_vector_distance_summary.png")
        plt.tight_layout(pad=2.0)
        save_render_close(fig, region_path)

    fig, ax = plt.subplots(figsize=(8, 6), facecolor=BLACK)
    style_ax(
        ax,
        title="Diagram Size vs Mean Persistence-Vector Distance",
        xlabel="H1 diagram intervals",
        ylabel="Mean Euclidean vector distance",
    )

    sc = ax.scatter(
        summary_df["n_intervals_diagram"],
        summary_df["mean_euclidean_vector_distance_to_others"],
        c=summary_df["max_lifetime"],
        s=140,
        cmap=CYAN_SEQUENTIAL_CMAP,
        edgecolor=ACCENT,
        linewidth=1.0,
    )

    for _, row in summary_df.iterrows():
        ax.text(
            row["n_intervals_diagram"],
            row["mean_euclidean_vector_distance_to_others"],
            row["channel"],
            fontsize=8,
            color=ACCENT,
            alpha=0.85,
        )

    cbar = plt.colorbar(sc, ax=ax, shrink=0.85)
    style_colorbar(cbar, "Max H1 lifetime")

    diagram_size_path = os.path.join(expert_dir, "13_diagram_size_vs_mean_vector_distance.png")
    plt.tight_layout()
    save_render_close(fig, diagram_size_path)

    fig, ax = plt.subplots(figsize=(8, 6), facecolor=BLACK)
    style_ax(
        ax,
        title="Threshold Triangles vs Mean Persistence-Vector Distance",
        xlabel="Triangles at radius",
        ylabel="Mean Euclidean vector distance",
    )

    sc = ax.scatter(
        summary_df["n_triangles_at_radius"],
        summary_df["mean_euclidean_vector_distance_to_others"],
        c=summary_df["n_intervals_diagram"],
        s=140,
        cmap=CYAN_SEQUENTIAL_CMAP,
        edgecolor=ACCENT,
        linewidth=1.0,
    )

    for _, row in summary_df.iterrows():
        ax.text(
            row["n_triangles_at_radius"],
            row["mean_euclidean_vector_distance_to_others"],
            row["channel"],
            fontsize=8,
            color=ACCENT,
            alpha=0.85,
        )

    cbar = plt.colorbar(sc, ax=ax, shrink=0.85)
    style_colorbar(cbar, "H1 diagram intervals")

    triangle_path = os.path.join(expert_dir, "14_threshold_triangles_vs_mean_vector_distance.png")
    plt.tight_layout()
    save_render_close(fig, triangle_path)

    qc_df = summary_df.copy().sort_values("elapsed_sec", ascending=False).reset_index(drop=True)

    fig, ax = plt.subplots(figsize=(12, 5.8), facecolor=BLACK)
    style_ax(
        ax,
        title="Persistence Diagram Vectorization Runtime QC",
        xlabel="Channel",
        ylabel="Elapsed seconds",
    )

    x_qc = np.arange(len(qc_df))

    ax.bar(
        x_qc,
        qc_df["elapsed_sec"],
        color=ACCENT,
        edgecolor=ACCENT,
        alpha=0.82,
    )

    ax.set_xticks(x_qc)
    ax.set_xticklabels(qc_df["channel"], rotation=90)

    qc_path = os.path.join(expert_dir, "15_runtime_qc.png")
    plt.tight_layout()
    save_render_close(fig, qc_path)

    log(f"\nExpert plots saved and rendered from: {expert_dir}")
    log(f"Regional summary CSV: {region_csv_path}")

elapsed_total = time.time() - overall_t0

log("\n" + "=" * 72)
log(f"Finished H1 Persistence Diagram Vectorization in {elapsed_total:.2f}s")
log(f"Mode: {FAST_OPTION}")
log(f"Backend: ripser_presampled_fps + fixed-length persistence-vector features")
log(f"Channels completed: {len(loaded_channels)}")
log(f"Feature matrix shape: {feature_matrix.shape}")
log(f"PCA explained variance ratio: PC1={pca_explained[0]:.4f}, PC2={pca_explained[1]:.4f}")
log(f"Summary CSV: {summary_csv_path}")
log(f"Feature matrix CSV: {feature_csv_path}")
log(f"Standardized feature matrix CSV: {feature_z_csv_path}")
log(f"Feature ranking CSV: {feature_importance_csv_path}")
log(f"Euclidean distance CSV: {euclidean_csv_path}")
log(f"Cosine similarity CSV: {cosine_similarity_csv_path}")
log(f"Pairwise CSV: {pairwise_csv_path}")
log(f"PCA CSV: {pca_csv_path}")
log(f"MDS CSV: {mds_csv_path}")
log(f"NPZ: {npz_path}")
log(f"Output directory: {mode_out_dir}")
log("=" * 72)
```

```python
import os
import sys
import time
import numpy as np

if not hasattr(np, "trapz"):
    np.trapz = np.trapezoid

import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.colors import LinearSegmentedColormap

os.environ["OMP_NUM_THREADS"] = "2"
os.environ["OPENBLAS_NUM_THREADS"] = "2"
os.environ["MKL_NUM_THREADS"] = "2"
os.environ["VECLIB_MAXIMUM_THREADS"] = "2"
os.environ["NUMEXPR_NUM_THREADS"] = "2"

from scipy.spatial.distance import pdist, squareform
from scipy.cluster.hierarchy import linkage, leaves_list

try:
    from scipy.linalg import eigh
    HAVE_SCIPY_EIGH = True
except Exception:
    HAVE_SCIPY_EIGH = False

try:
    from IPython.display import display, Image
    HAVE_IPYTHON = True
except Exception:
    HAVE_IPYTHON = False

try:
    from ripser import ripser
except Exception as exc:
    raise ImportError(
        "ripser is not installed. Run this in a separate notebook cell first:\n\n"
        "!pip install ripser persim\n\n"
        "Then rerun this Topology-Based Channel Fingerprints cell."
    ) from exc


base_dir = "/home/a/projects/Complete-Neural-Signal-Analysis"
out_dir = os.path.join(base_dir, "topology_channel_fingerprints")
os.makedirs(out_dir, exist_ok=True)

FAST_OPTION = "ultra"  # options: "ultra", "fast", "balanced", "full"

MAX_CHANNELS_TO_PROCESS = None
RENDER_ALL_IMAGES_INLINE = True
SAVE_PLOTS_TO_DISK = True
MAKE_EXPERT_PLOTS = True
INLINE_IMAGE_WIDTH = 950
VERBOSE_PROGRESS = True

USE_GEOMETRY_PRESERVING_SUBSAMPLE = True

MODE_CONFIGS = {
    "ultra": {
        "max_points": 90,
        "reservoir_points": 5000,
        "max_dims": 3,
        "radius_quantile": 0.22,
        "betti_grid_size": 100,
        "landscape_grid_size": 100,
        "landscape_levels": 4,
        "image_birth_bins": 28,
        "image_persistence_bins": 28,
        "image_sigma_frac": 0.035,
        "image_weight_power": 1.0,
        "top_lifetimes": 10,
        "cap_essential_h1": True,
        "plot_dpi": 120,
    },
    "fast": {
        "max_points": 130,
        "reservoir_points": 8000,
        "max_dims": 3,
        "radius_quantile": 0.24,
        "betti_grid_size": 130,
        "landscape_grid_size": 130,
        "landscape_levels": 4,
        "image_birth_bins": 36,
        "image_persistence_bins": 36,
        "image_sigma_frac": 0.030,
        "image_weight_power": 1.0,
        "top_lifetimes": 14,
        "cap_essential_h1": True,
        "plot_dpi": 130,
    },
    "balanced": {
        "max_points": 180,
        "reservoir_points": 10000,
        "max_dims": 3,
        "radius_quantile": 0.26,
        "betti_grid_size": 170,
        "landscape_grid_size": 170,
        "landscape_levels": 5,
        "image_birth_bins": 44,
        "image_persistence_bins": 44,
        "image_sigma_frac": 0.027,
        "image_weight_power": 1.0,
        "top_lifetimes": 18,
        "cap_essential_h1": True,
        "plot_dpi": 140,
    },
    "full": {
        "max_points": 260,
        "reservoir_points": 14000,
        "max_dims": 3,
        "radius_quantile": 0.28,
        "betti_grid_size": 240,
        "landscape_grid_size": 240,
        "landscape_levels": 6,
        "image_birth_bins": 56,
        "image_persistence_bins": 56,
        "image_sigma_frac": 0.024,
        "image_weight_power": 1.0,
        "top_lifetimes": 24,
        "cap_essential_h1": True,
        "plot_dpi": 150,
    },
}

if FAST_OPTION not in MODE_CONFIGS:
    raise ValueError("FAST_OPTION must be one of: ultra, fast, balanced, full")

cfg = MODE_CONFIGS[FAST_OPTION]

MAX_POINTS = cfg["max_points"]
RESERVOIR_POINTS = cfg["reservoir_points"]
MAX_DIMS = cfg["max_dims"]
RADIUS_QUANTILE = cfg["radius_quantile"]
BETTI_GRID_SIZE = cfg["betti_grid_size"]
LANDSCAPE_GRID_SIZE = cfg["landscape_grid_size"]
LANDSCAPE_LEVELS = cfg["landscape_levels"]
IMAGE_BIRTH_BINS = cfg["image_birth_bins"]
IMAGE_PERSISTENCE_BINS = cfg["image_persistence_bins"]
IMAGE_SIGMA_FRAC = cfg["image_sigma_frac"]
IMAGE_WEIGHT_POWER = cfg["image_weight_power"]
TOP_LIFETIMES = cfg["top_lifetimes"]
CAP_ESSENTIAL_H1 = cfg["cap_essential_h1"]
PLOT_DPI = cfg["plot_dpi"]

mode_out_dir = os.path.join(
    out_dir,
    f"h1_topology_channel_fingerprints_{FAST_OPTION}_fps",
)
os.makedirs(mode_out_dir, exist_ok=True)

eeg_channels = [
    "Fp1", "Fpz", "Fp2", "F7", "F3", "Fz", "F4", "F8", "FC5", "FC1", "FC2", "FC6",
    "M1", "T7", "C3", "Cz", "C4", "T8", "M2", "CP5", "CP1", "CP2", "CP6",
    "P7", "P3", "Pz", "P4", "P8", "POz", "O1", "Oz", "O2"
]

if MAX_CHANNELS_TO_PROCESS is not None:
    eeg_channels = eeg_channels[:MAX_CHANNELS_TO_PROCESS]

channel_regions = {
    "Fp1": "Frontal", "Fpz": "Frontal", "Fp2": "Frontal",
    "F7": "Frontal", "F3": "Frontal", "Fz": "Frontal", "F4": "Frontal", "F8": "Frontal",
    "FC5": "Frontocentral", "FC1": "Frontocentral", "FC2": "Frontocentral", "FC6": "Frontocentral",
    "M1": "Mastoid", "M2": "Mastoid",
    "T7": "Central", "C3": "Central", "Cz": "Central", "C4": "Central", "T8": "Central",
    "CP5": "Centroparietal", "CP1": "Centroparietal", "CP2": "Centroparietal", "CP6": "Centroparietal",
    "P7": "Parietal", "P3": "Parietal", "Pz": "Parietal", "P4": "Parietal", "P8": "Parietal",
    "POz": "Occipital", "O1": "Occipital", "Oz": "Occipital", "O2": "Occipital",
}

ACCENT = "#00FFFF"
ACCENT_SOFT = "#66FFFF"
ACCENT_DIM = "#008B8B"
BLACK = "#000000"

CYAN_SEQUENTIAL_CMAP = LinearSegmentedColormap.from_list(
    "cyan_sequential",
    [BLACK, ACCENT_DIM, ACCENT, ACCENT_SOFT],
    N=256,
)

plt.rcParams.update({
    "figure.facecolor": BLACK,
    "axes.facecolor": BLACK,
    "savefig.facecolor": BLACK,
    "text.color": ACCENT,
    "axes.labelcolor": ACCENT,
    "axes.edgecolor": ACCENT,
    "axes.titlecolor": ACCENT,
    "xtick.color": ACCENT,
    "ytick.color": ACCENT,
    "grid.color": ACCENT_DIM,
    "grid.alpha": 0.25,
    "legend.facecolor": BLACK,
    "legend.edgecolor": ACCENT,
    "legend.labelcolor": ACCENT,
    "font.size": 11,
})


def log(msg):
    if VERBOSE_PROGRESS:
        print(msg, flush=True)


def style_ax(ax, title=None, xlabel=None, ylabel=None, grid=True):
    ax.set_facecolor(BLACK)

    for spine in ax.spines.values():
        spine.set_color(ACCENT)
        spine.set_linewidth(1.1)

    ax.tick_params(axis="x", colors=ACCENT, labelcolor=ACCENT)
    ax.tick_params(axis="y", colors=ACCENT, labelcolor=ACCENT)

    if title is not None:
        ax.set_title(title, color=ACCENT, pad=12)

    if xlabel is not None:
        ax.set_xlabel(xlabel, color=ACCENT, labelpad=10)

    if ylabel is not None:
        ax.set_ylabel(ylabel, color=ACCENT, labelpad=10)

    if grid:
        ax.grid(True, color=ACCENT_DIM, alpha=0.25, linewidth=0.8)


def style_colorbar(cbar, label):
    cbar.set_label(label, color=ACCENT)
    cbar.ax.yaxis.set_tick_params(color=ACCENT)
    plt.setp(cbar.ax.get_yticklabels(), color=ACCENT)
    cbar.outline.set_edgecolor(ACCENT)


def save_render_close(fig, path, width=INLINE_IMAGE_WIDTH):
    if SAVE_PLOTS_TO_DISK:
        fig.savefig(path, dpi=PLOT_DPI, bbox_inches="tight", facecolor=BLACK)

    plt.close(fig)

    if RENDER_ALL_IMAGES_INLINE and HAVE_IPYTHON and os.path.exists(path):
        display(Image(filename=path, width=width))


def dynamic_limits(values, lower_pct=2, upper_pct=98, pad_frac=0.08, lower_bound=None, upper_bound=None):
    values = np.asarray(values, dtype=float)
    values = values[np.isfinite(values)]

    if len(values) == 0:
        vmin, vmax = 0.0, 1.0
    else:
        vmin = float(np.percentile(values, lower_pct))
        vmax = float(np.percentile(values, upper_pct))

        if not np.isfinite(vmin):
            vmin = float(np.nanmin(values))

        if not np.isfinite(vmax):
            vmax = float(np.nanmax(values))

        if vmax <= vmin:
            center = float(np.nanmean(values))
            eps = max(1e-6, abs(center) * 0.01, 1e-5)
            vmin = center - eps
            vmax = center + eps

    pad = (vmax - vmin) * pad_frac
    vmin -= pad
    vmax += pad

    if lower_bound is not None:
        vmin = max(lower_bound, vmin)

    if upper_bound is not None:
        vmax = min(upper_bound, vmax)

    if vmax <= vmin:
        vmax = vmin + 1e-6

    return vmin, vmax


def offdiag_values(M):
    M = np.asarray(M, dtype=float)

    if M.shape[0] <= 1:
        return np.asarray([], dtype=float)

    mask = ~np.eye(M.shape[0], dtype=bool)
    vals = M[mask]
    vals = vals[np.isfinite(vals)]

    return vals


def save_matrix_plot(matrix, labels, title, colorbar_label, path, vmin=None, vmax=None, width=INLINE_IMAGE_WIDTH):
    fig, ax = plt.subplots(figsize=(9, 8), facecolor=BLACK)
    style_ax(
        ax,
        title=title,
        xlabel="Channel",
        ylabel="Channel",
    )

    im = ax.imshow(
        matrix,
        aspect="equal",
        interpolation="nearest",
        cmap=CYAN_SEQUENTIAL_CMAP,
        vmin=vmin,
        vmax=vmax,
    )

    ax.set_xticks(np.arange(len(labels)))
    ax.set_xticklabels(labels, rotation=90)
    ax.set_yticks(np.arange(len(labels)))
    ax.set_yticklabels(labels)

    cbar = plt.colorbar(im, ax=ax, shrink=0.85)
    style_colorbar(cbar, colorbar_label)

    plt.tight_layout()
    save_render_close(fig, path, width=width)


def estimate_median_pairwise_distance(X, max_pairs=120000, seed=7):
    X = np.asarray(X, dtype=float)
    n = X.shape[0]

    if n < 2:
        return 1.0

    rng = np.random.default_rng(seed)

    if n * (n - 1) // 2 <= max_pairs:
        D = pdist(X)
        D = D[D > 0]

        if len(D) == 0:
            return 1.0

        return float(np.nanmedian(D))

    i = rng.integers(0, n, size=max_pairs)
    j = rng.integers(0, n, size=max_pairs)

    mask = i != j
    i = i[mask]
    j = j[mask]

    if len(i) == 0:
        return 1.0

    d = np.linalg.norm(X[i] - X[j], axis=1)
    d = d[np.isfinite(d) & (d > 0)]

    if len(d) == 0:
        return 1.0

    return float(np.nanmedian(d))


def standardize_matrix_for_geometry(X, max_dims=3, seed=7):
    X = np.asarray(X, dtype=np.float64)
    X = X[np.all(np.isfinite(X), axis=1)]

    if X.shape[0] < 10:
        return None

    X = X - np.mean(X, axis=0, keepdims=True)

    if X.shape[1] > max_dims:
        _U, _S, Vt = np.linalg.svd(X, full_matrices=False)
        X = X @ Vt[:max_dims].T

    X = X / (np.std(X, axis=0, keepdims=True) + 1e-12)

    scale = estimate_median_pairwise_distance(X, seed=seed)
    X = X / (scale + 1e-12)

    return X


def farthest_point_sample(X, n_samples, seed=7):
    X = np.asarray(X, dtype=float)
    n = X.shape[0]

    if n <= n_samples:
        return np.arange(n)

    rng = np.random.default_rng(seed)

    center = np.mean(X, axis=0, keepdims=True)
    dist_to_center = np.linalg.norm(X - center, axis=1)

    start_idx = int(np.nanargmax(dist_to_center))

    selected = np.empty(n_samples, dtype=int)
    selected.fill(-1)
    selected[0] = start_idx

    min_dist2 = np.full(n, np.inf, dtype=float)

    for k in range(1, n_samples):
        last = X[selected[k - 1]]
        d2 = np.sum((X - last) ** 2, axis=1)
        min_dist2 = np.minimum(min_dist2, d2)
        min_dist2[selected[:k]] = -np.inf

        next_idx = int(np.nanargmax(min_dist2))

        if not np.isfinite(min_dist2[next_idx]) or min_dist2[next_idx] < 1e-18:
            remaining = np.setdiff1d(np.arange(n), selected[:k], assume_unique=False)

            if len(remaining) == 0:
                selected = selected[:k]
                break

            next_idx = int(rng.choice(remaining))

        selected[k] = next_idx

    selected = selected[selected >= 0]

    return selected


def load_channel_embedding_geometry_sampled(ch_name, max_points=90, reservoir_points=5000, max_dims=3, seed=7):
    candidate_paths = [
        os.path.join(base_dir, "embedding_data", "tsne_umap_from_3d", f"umap_embedding_{ch_name}.npy"),
        os.path.join(base_dir, "embedding_data", "tsne_umap_from_3d", f"tsne_embedding_{ch_name}.npy"),
        os.path.join(base_dir, "embedding_data", "3dembedding_data", f"3dembedded_{ch_name}.npy"),
        os.path.join(base_dir, "embedding_data", "2dembedding_data", f"2dembedded_{ch_name}.npy"),
    ]

    rng = np.random.default_rng(seed)

    for path in candidate_paths:
        if not os.path.exists(path):
            continue

        log(f"    trying: {path}")

        try:
            X = np.load(path, mmap_mode="r", allow_pickle=False)
        except Exception as exc:
            log(f"    skipped non-memmappable or nonnumeric file: {os.path.basename(path)} | {exc}")
            continue

        if X.ndim != 2:
            log(f"    skipped because ndim={X.ndim}")
            continue

        original_shape = tuple(X.shape)

        try:
            if X.shape[0] >= X.shape[1]:
                n_samples = X.shape[0]
                n_features = X.shape[1]

                if n_samples < 10 or n_features < 2:
                    log(f"    skipped bad shape: {original_shape}")
                    continue

                reservoir_n = min(max(reservoir_points, max_points), n_samples)

                if n_samples > reservoir_n:
                    reservoir_idx = np.sort(rng.choice(n_samples, size=reservoir_n, replace=False))
                else:
                    reservoir_idx = np.arange(n_samples)

                X_reservoir = np.asarray(X[reservoir_idx, :], dtype=np.float64)

            else:
                n_samples = X.shape[1]
                n_features = X.shape[0]

                if n_samples < 10 or n_features < 2:
                    log(f"    skipped bad transposed shape: {original_shape}")
                    continue

                reservoir_n = min(max(reservoir_points, max_points), n_samples)

                if n_samples > reservoir_n:
                    reservoir_idx = np.sort(rng.choice(n_samples, size=reservoir_n, replace=False))
                else:
                    reservoir_idx = np.arange(n_samples)

                X_reservoir = np.asarray(X[:, reservoir_idx].T, dtype=np.float64)

        except Exception as exc:
            log(f"    failed while reservoir-sampling {os.path.basename(path)} | {exc}")
            continue

        X_reservoir = X_reservoir[np.all(np.isfinite(X_reservoir), axis=1)]

        if X_reservoir.shape[0] < 10 or X_reservoir.shape[1] < 2:
            log(f"    reservoir data unusable: reservoir_shape={X_reservoir.shape}")
            continue

        X_reservoir_std = standardize_matrix_for_geometry(
            X_reservoir,
            max_dims=max_dims,
            seed=seed,
        )

        if X_reservoir_std is None:
            log("    standardized reservoir unusable")
            continue

        if USE_GEOMETRY_PRESERVING_SUBSAMPLE:
            fps_idx = farthest_point_sample(
                X_reservoir_std,
                n_samples=min(max_points, X_reservoir_std.shape[0]),
                seed=seed,
            )
            X_small = X_reservoir_std[fps_idx]
            method = "reservoir_plus_farthest_point"
        else:
            sample_n = min(max_points, X_reservoir_std.shape[0])
            idx = np.sort(rng.choice(X_reservoir_std.shape[0], size=sample_n, replace=False))
            X_small = X_reservoir_std[idx]
            method = "reservoir_random"

        X_small = X_small[np.all(np.isfinite(X_small), axis=1)]

        if X_small.shape[0] >= 10 and X_small.shape[1] >= 2:
            return X_small, path, original_shape, tuple(X_reservoir.shape), method

        log(f"    sampled data unusable: sampled_shape={X_small.shape}")

    return None, None, None, None, None


def standardize_point_cloud_small(X, max_dims=3):
    X = np.asarray(X, dtype=np.float64)
    X = X[np.all(np.isfinite(X), axis=1)]

    if X.shape[0] < 10:
        return None

    X = X - np.mean(X, axis=0, keepdims=True)

    if X.shape[1] > max_dims:
        _U, _S, Vt = np.linalg.svd(X, full_matrices=False)
        X = X @ Vt[:max_dims].T

    X = X / (np.std(X, axis=0, keepdims=True) + 1e-12)

    D = pdist(X)
    positive = D[D > 0]

    if len(positive) > 0:
        X = X / (np.nanmedian(positive) + 1e-12)

    return X


def estimate_complex_counts(X, radius):
    D = squareform(pdist(X))
    A = (D <= radius) & (D > 0)

    n_edges = int(np.sum(A) // 2)

    if A.shape[0] <= 300:
        A_int = A.astype(np.int64)
        n_triangles = int(np.trace(A_int @ A_int @ A_int) // 6)
    else:
        n_triangles = np.nan

    return n_edges, n_triangles


def run_h1_ripser(X, radius):
    result = ripser(
        X,
        maxdim=1,
        thresh=float(radius),
        coeff=2,
        do_cocycles=False,
    )

    diagrams = result["dgms"]

    h0 = diagrams[0] if len(diagrams) > 0 else np.empty((0, 2), dtype=float)
    h1 = diagrams[1] if len(diagrams) > 1 else np.empty((0, 2), dtype=float)

    return h0, h1


def split_finite_essential(diagram):
    diagram = np.asarray(diagram, dtype=float)

    if diagram.size == 0:
        return np.empty((0, 2), dtype=float), np.empty((0, 2), dtype=float)

    finite = diagram[
        np.isfinite(diagram[:, 0])
        & np.isfinite(diagram[:, 1])
        & (diagram[:, 1] > diagram[:, 0])
    ]

    essential = diagram[
        np.isfinite(diagram[:, 0])
        & np.isinf(diagram[:, 1])
    ]

    return finite, essential


def clean_diagram_for_fingerprint(
    finite_intervals,
    essential_intervals=None,
    cap_essential=True,
    cap_value=None,
):
    cleaned = []
    capped_count = 0

    if finite_intervals is not None:
        for b, d in finite_intervals:
            if np.isfinite(b) and np.isfinite(d) and d > b:
                cleaned.append((float(b), float(d)))

    if cap_essential and essential_intervals is not None and cap_value is not None:
        for b, _d in essential_intervals:
            if np.isfinite(b) and np.isfinite(cap_value) and cap_value > b:
                cleaned.append((float(b), float(cap_value)))
                capped_count += 1

    if len(cleaned) == 0:
        return np.empty((0, 2), dtype=float), capped_count

    return np.asarray(cleaned, dtype=float), capped_count


def diagram_to_birth_persistence(diagram):
    diagram = np.asarray(diagram, dtype=float)

    if diagram.size == 0 or len(diagram) == 0:
        return np.empty((0, 2), dtype=float)

    bp = np.column_stack([diagram[:, 0], diagram[:, 1] - diagram[:, 0]])
    bp = bp[
        np.isfinite(bp[:, 0])
        & np.isfinite(bp[:, 1])
        & (bp[:, 1] > 0)
    ]

    return bp


def safe_entropy(values):
    values = np.asarray(values, dtype=float)
    values = values[np.isfinite(values) & (values > 0)]

    if len(values) == 0:
        return 0.0

    total = float(np.sum(values))

    if total <= 0:
        return 0.0

    p = values / (total + 1e-12)

    return float(-np.sum(p * np.log(p + 1e-12)))


def safe_gini(values):
    values = np.asarray(values, dtype=float)
    values = values[np.isfinite(values)]

    if len(values) == 0:
        return 0.0

    values = np.maximum(values, 0.0)

    if np.sum(values) <= 0:
        return 0.0

    sorted_vals = np.sort(values)
    n = len(sorted_vals)
    index = np.arange(1, n + 1)

    return float((2.0 * np.sum(index * sorted_vals) / (n * np.sum(sorted_vals))) - ((n + 1.0) / n))


def safe_skew(values):
    values = np.asarray(values, dtype=float)
    values = values[np.isfinite(values)]

    if len(values) < 3:
        return 0.0

    mu = np.mean(values)
    sd = np.std(values)

    if sd <= 0:
        return 0.0

    return float(np.mean(((values - mu) / sd) ** 3))


def safe_kurtosis(values):
    values = np.asarray(values, dtype=float)
    values = values[np.isfinite(values)]

    if len(values) < 4:
        return 0.0

    mu = np.mean(values)
    sd = np.std(values)

    if sd <= 0:
        return 0.0

    return float(np.mean(((values - mu) / sd) ** 4) - 3.0)


def diagram_core_features(diagram):
    diagram = np.asarray(diagram, dtype=float)

    if diagram.size == 0 or len(diagram) == 0:
        return {
            "n_intervals": 0.0,
            "birth_mean": 0.0,
            "birth_std": 0.0,
            "birth_min": 0.0,
            "birth_max": 0.0,
            "death_mean": 0.0,
            "death_std": 0.0,
            "death_min": 0.0,
            "death_max": 0.0,
            "lifetime_mean": 0.0,
            "lifetime_std": 0.0,
            "lifetime_min": 0.0,
            "lifetime_q25": 0.0,
            "lifetime_median": 0.0,
            "lifetime_q75": 0.0,
            "lifetime_q90": 0.0,
            "lifetime_max": 0.0,
            "total_lifetime": 0.0,
            "lifetime_entropy": 0.0,
            "lifetime_gini": 0.0,
            "lifetime_skew": 0.0,
            "lifetime_kurtosis": 0.0,
            "midpoint_mean": 0.0,
            "midpoint_std": 0.0,
        }

    births = diagram[:, 0]
    deaths = diagram[:, 1]
    lifetimes = deaths - births
    midpoints = 0.5 * (births + deaths)

    return {
        "n_intervals": float(len(diagram)),
        "birth_mean": float(np.nanmean(births)),
        "birth_std": float(np.nanstd(births)),
        "birth_min": float(np.nanmin(births)),
        "birth_max": float(np.nanmax(births)),
        "death_mean": float(np.nanmean(deaths)),
        "death_std": float(np.nanstd(deaths)),
        "death_min": float(np.nanmin(deaths)),
        "death_max": float(np.nanmax(deaths)),
        "lifetime_mean": float(np.nanmean(lifetimes)),
        "lifetime_std": float(np.nanstd(lifetimes)),
        "lifetime_min": float(np.nanmin(lifetimes)),
        "lifetime_q25": float(np.nanpercentile(lifetimes, 25)),
        "lifetime_median": float(np.nanmedian(lifetimes)),
        "lifetime_q75": float(np.nanpercentile(lifetimes, 75)),
        "lifetime_q90": float(np.nanpercentile(lifetimes, 90)),
        "lifetime_max": float(np.nanmax(lifetimes)),
        "total_lifetime": float(np.nansum(lifetimes)),
        "lifetime_entropy": safe_entropy(lifetimes),
        "lifetime_gini": safe_gini(lifetimes),
        "lifetime_skew": safe_skew(lifetimes),
        "lifetime_kurtosis": safe_kurtosis(lifetimes),
        "midpoint_mean": float(np.nanmean(midpoints)),
        "midpoint_std": float(np.nanstd(midpoints)),
    }


def betti_curve(diagram, grid):
    diagram = np.asarray(diagram, dtype=float)
    curve = np.zeros(len(grid), dtype=float)

    if diagram.size == 0 or len(diagram) == 0:
        return curve

    births = diagram[:, 0]
    deaths = diagram[:, 1]

    for b, d in zip(births, deaths):
        curve += ((grid >= b) & (grid < d)).astype(float)

    return curve


def betti_features(curve, grid):
    curve = np.asarray(curve, dtype=float)

    if len(curve) == 0:
        return {
            "betti_max": 0.0,
            "betti_mean": 0.0,
            "betti_auc": 0.0,
            "betti_std": 0.0,
            "betti_nonzero_fraction": 0.0,
            "betti_peak_count": 0.0,
        }

    if len(curve) >= 3:
        peak_count = float(np.sum((curve[1:-1] >= curve[:-2]) & (curve[1:-1] >= curve[2:]) & (curve[1:-1] > 0)))
    else:
        peak_count = 0.0

    return {
        "betti_max": float(np.nanmax(curve)),
        "betti_mean": float(np.nanmean(curve)),
        "betti_auc": float(np.trapz(curve, grid)),
        "betti_std": float(np.nanstd(curve)),
        "betti_nonzero_fraction": float(np.mean(curve > 0)),
        "betti_peak_count": peak_count,
    }


def persistence_landscapes(diagram, grid, n_levels=4):
    diagram = np.asarray(diagram, dtype=float)
    landscapes = np.zeros((n_levels, len(grid)), dtype=float)

    if diagram.size == 0 or len(diagram) == 0:
        return landscapes

    tents = []

    for b, d in diagram:
        if not np.isfinite(b) or not np.isfinite(d) or d <= b:
            continue

        tent = np.maximum(0.0, np.minimum(grid - b, d - grid))
        tents.append(tent)

    if len(tents) == 0:
        return landscapes

    tent_matrix = np.vstack(tents)

    for idx in range(len(grid)):
        vals = np.sort(tent_matrix[:, idx])[::-1]

        for level in range(min(n_levels, len(vals))):
            landscapes[level, idx] = vals[level]

    return landscapes


def landscape_features(landscapes, grid):
    landscapes = np.asarray(landscapes, dtype=float)

    if landscapes.size == 0:
        out = {
            "landscape_l1": 0.0,
            "landscape_l2": 0.0,
            "landscape_max": 0.0,
            "landscape_entropy": 0.0,
        }

        for level in range(LANDSCAPE_LEVELS):
            out[f"landscape_l2_level_{level + 1}"] = 0.0

        return out

    level_l1 = np.trapz(np.abs(landscapes), grid, axis=1)
    level_l2 = np.sqrt(np.trapz(landscapes ** 2, grid, axis=1))

    out = {
        "landscape_l1": float(np.sum(level_l1)),
        "landscape_l2": float(np.sqrt(np.sum(level_l2 ** 2))),
        "landscape_max": float(np.nanmax(landscapes)),
        "landscape_entropy": safe_entropy(level_l2),
    }

    for level in range(LANDSCAPE_LEVELS):
        out[f"landscape_l2_level_{level + 1}"] = float(level_l2[level]) if level < len(level_l2) else 0.0

    return out


def persistence_image(bp, birth_grid, persistence_grid, sigma_birth, sigma_persistence, weight_power=1.0):
    bp = np.asarray(bp, dtype=float)
    image = np.zeros((len(persistence_grid), len(birth_grid)), dtype=float)

    if bp.size == 0 or len(bp) == 0:
        return image

    births = bp[:, 0]
    persistences = bp[:, 1]
    weights = persistences ** weight_power

    for birth, persistence, weight in zip(births, persistences, weights):
        bx = np.exp(-0.5 * ((birth_grid - birth) / (sigma_birth + 1e-12)) ** 2)
        py = np.exp(-0.5 * ((persistence_grid - persistence) / (sigma_persistence + 1e-12)) ** 2)
        image += weight * np.outer(py, bx)

    return image


def image_features(image, birth_grid, persistence_grid):
    image = np.asarray(image, dtype=float)

    mass = float(np.nansum(image))
    l2 = float(np.sqrt(np.nansum(image ** 2)))
    max_pixel = float(np.nanmax(image)) if image.size else 0.0
    entropy = safe_entropy(image.ravel())

    if mass > 0:
        birth_mesh, persistence_mesh = np.meshgrid(birth_grid, persistence_grid)
        centroid_birth = float(np.nansum(image * birth_mesh) / (mass + 1e-12))
        centroid_persistence = float(np.nansum(image * persistence_mesh) / (mass + 1e-12))
        spread_birth = float(np.sqrt(np.nansum(image * (birth_mesh - centroid_birth) ** 2) / (mass + 1e-12)))
        spread_persistence = float(np.sqrt(np.nansum(image * (persistence_mesh - centroid_persistence) ** 2) / (mass + 1e-12)))
    else:
        centroid_birth = 0.0
        centroid_persistence = 0.0
        spread_birth = 0.0
        spread_persistence = 0.0

    return {
        "image_mass": mass,
        "image_l2": l2,
        "image_max": max_pixel,
        "image_entropy": entropy,
        "image_centroid_birth": centroid_birth,
        "image_centroid_persistence": centroid_persistence,
        "image_spread_birth": spread_birth,
        "image_spread_persistence": spread_persistence,
    }


def top_lifetime_features(diagram, top_k=10):
    diagram = np.asarray(diagram, dtype=float)

    out = {}

    for i in range(top_k):
        out[f"top_lifetime_{i + 1:02d}"] = 0.0

    if diagram.size == 0 or len(diagram) == 0:
        return out

    lifetimes = diagram[:, 1] - diagram[:, 0]
    lifetimes = lifetimes[np.isfinite(lifetimes) & (lifetimes > 0)]
    lifetimes = np.sort(lifetimes)[::-1]

    take = min(top_k, len(lifetimes))

    for i in range(take):
        out[f"top_lifetime_{i + 1:02d}"] = float(lifetimes[i])

    return out


def standardize_features(df):
    X = df.to_numpy(dtype=float)
    X = np.nan_to_num(X, nan=0.0, posinf=0.0, neginf=0.0)

    mu = np.mean(X, axis=0, keepdims=True)
    sd = np.std(X, axis=0, keepdims=True)

    Xz = (X - mu) / (sd + 1e-12)
    Xz = np.nan_to_num(Xz, nan=0.0, posinf=0.0, neginf=0.0)

    return pd.DataFrame(Xz, index=df.index, columns=df.columns), mu.ravel(), sd.ravel()


def pairwise_euclidean_matrix(X):
    X = np.asarray(X, dtype=float)

    if len(X) == 0:
        return np.empty((0, 0), dtype=float)

    return squareform(pdist(X, metric="euclidean"))


def cosine_similarity_matrix(X):
    X = np.asarray(X, dtype=float)

    if len(X) == 0:
        return np.empty((0, 0), dtype=float)

    norms = np.linalg.norm(X, axis=1, keepdims=True)
    Xn = X / (norms + 1e-12)

    C = Xn @ Xn.T
    C = np.clip(C, -1.0, 1.0)

    return C


def percentile_score(values, higher_is_more=True):
    s = pd.Series(values, dtype=float)

    if not higher_is_more:
        s = -s

    if s.nunique(dropna=True) <= 1:
        return np.full(len(s), 50.0, dtype=float)

    return (s.rank(method="average", pct=True).to_numpy(dtype=float) * 100.0)


def mean_z_score(z_df, feature_names):
    valid = [f for f in feature_names if f in z_df.columns]

    if len(valid) == 0:
        return np.zeros(len(z_df), dtype=float)

    return z_df[valid].mean(axis=1).to_numpy(dtype=float)


def pca_from_feature_matrix(X, n_components=2):
    X = np.asarray(X, dtype=float)

    if X.shape[0] == 0:
        return np.empty((0, n_components)), np.zeros(n_components), np.zeros((n_components, X.shape[1]))

    Xc = X - np.mean(X, axis=0, keepdims=True)

    U, S, Vt = np.linalg.svd(Xc, full_matrices=False)

    n_take = min(n_components, U.shape[1])

    coords = np.zeros((X.shape[0], n_components), dtype=float)
    coords[:, :n_take] = U[:, :n_take] * S[:n_take]

    var = S ** 2
    total = np.sum(var)

    explained = np.zeros(n_components, dtype=float)

    if total > 0:
        explained[:n_take] = var[:n_take] / total

    components = np.zeros((n_components, X.shape[1]), dtype=float)
    components[:n_take] = Vt[:n_take]

    return coords, explained, components


def classical_mds_from_distances(D, n_components=2):
    D = np.asarray(D, dtype=float)
    n = D.shape[0]

    if n < 2:
        return np.zeros((n, n_components), dtype=float)

    D2 = D ** 2
    J = np.eye(n) - np.ones((n, n)) / n
    B = -0.5 * J @ D2 @ J

    if HAVE_SCIPY_EIGH:
        vals, vecs = eigh(B)
    else:
        vals, vecs = np.linalg.eigh(B)

    idx = np.argsort(vals)[::-1]
    vals = vals[idx]
    vecs = vecs[:, idx]

    coords = np.zeros((n, n_components), dtype=float)

    for k in range(min(n_components, n)):
        if vals[k] > 0:
            coords[:, k] = vecs[:, k] * np.sqrt(vals[k])

    return coords


log("\n" + "=" * 72)
log(f"Running H1 Topology-Based Channel Fingerprints | {FAST_OPTION.upper()}")
log(f"Channels: {len(eeg_channels)}")
log(f"max_points={MAX_POINTS}, reservoir_points={RESERVOIR_POINTS}, radius_quantile={RADIUS_QUANTILE}")
log(f"betti_grid_size={BETTI_GRID_SIZE}, landscape_grid_size={LANDSCAPE_GRID_SIZE}, landscape_levels={LANDSCAPE_LEVELS}")
log(f"image_bins={IMAGE_BIRTH_BINS}x{IMAGE_PERSISTENCE_BINS}, image_sigma_frac={IMAGE_SIGMA_FRAC}")
log(f"top_lifetimes={TOP_LIFETIMES}, geometry_preserving_subsample={USE_GEOMETRY_PRESERVING_SUBSAMPLE}")
log(f"cap_essential_h1={CAP_ESSENTIAL_H1}")
log(f"expert_plots={MAKE_EXPERT_PLOTS}, render_all_images={RENDER_ALL_IMAGES_INLINE}")
log(f"Output: {mode_out_dir}")
log("=" * 72)

diagnostics = {}
diagram_store = {}
bp_store = {}
raw_interval_store = {}
loaded_channels = []

global_birth_max = None
global_persistence_max = None
global_filtration_max = None

overall_t0 = time.time()

for ch_idx, ch_name in enumerate(eeg_channels, start=1):
    t0 = time.time()
    prefix = f"[{ch_idx:02d}/{len(eeg_channels)} {ch_name}]"

    log(f"\n{prefix} loading reservoir + geometry-preserving subsampling")

    X_small, source_path, original_shape, reservoir_shape, sampling_method = load_channel_embedding_geometry_sampled(
        ch_name,
        max_points=MAX_POINTS,
        reservoir_points=RESERVOIR_POINTS,
        max_dims=MAX_DIMS,
        seed=1200 + ch_idx,
    )

    if X_small is None:
        log(f"{prefix} WARNING: no usable embedding found; skipping")
        continue

    log(
        f"{prefix} source shape={original_shape}, reservoir shape={reservoir_shape}, "
        f"sampled shape={X_small.shape}, method={sampling_method}"
    )

    X = standardize_point_cloud_small(X_small, max_dims=MAX_DIMS)

    if X is None:
        log(f"{prefix} WARNING: invalid sampled embedding; skipping")
        continue

    D_flat = pdist(X)
    positive = D_flat[D_flat > 0]

    if len(positive) == 0:
        log(f"{prefix} WARNING: zero-distance cloud; skipping")
        continue

    radius = float(np.quantile(positive, RADIUS_QUANTILE))

    log(f"{prefix} estimating complex + ripser H0/H1")
    n_edges, n_triangles = estimate_complex_counts(X, radius)
    h0_diag, h1_diag = run_h1_ripser(X, radius)

    h0_finite, h0_essential = split_finite_essential(h0_diag)
    h1_finite, h1_essential = split_finite_essential(h1_diag)

    h1_diagram, n_capped_h1 = clean_diagram_for_fingerprint(
        h1_finite,
        essential_intervals=h1_essential,
        cap_essential=CAP_ESSENTIAL_H1,
        cap_value=radius,
    )

    bp = diagram_to_birth_persistence(h1_diagram)

    if len(bp) > 0:
        birth_max = float(np.nanmax(bp[:, 0]))
        persistence_max = float(np.nanmax(bp[:, 1]))
        death_max = float(np.nanmax(h1_diagram[:, 1]))

        global_birth_max = birth_max if global_birth_max is None else max(global_birth_max, birth_max)
        global_persistence_max = persistence_max if global_persistence_max is None else max(global_persistence_max, persistence_max)
        global_filtration_max = death_max if global_filtration_max is None else max(global_filtration_max, death_max)

    elapsed = time.time() - t0

    diagram_store[ch_name] = h1_diagram
    bp_store[ch_name] = bp

    raw_interval_store[ch_name] = {
        "H0_finite": h0_finite,
        "H0_essential": h0_essential,
        "H1_finite": h1_finite,
        "H1_essential": h1_essential,
        "H1_fingerprint_diagram": h1_diagram,
        "H1_birth_persistence": bp,
    }

    diagnostics[ch_name] = {
        "channel": ch_name,
        "region": channel_regions.get(ch_name, "Unknown"),
        "source_path": source_path,
        "source_shape": str(original_shape),
        "reservoir_shape": str(reservoir_shape),
        "sampling_method": sampling_method,
        "backend": "ripser_presampled_fps",
        "n_points": int(X.shape[0]),
        "n_edges_at_radius": int(n_edges),
        "n_triangles_at_radius": int(n_triangles) if np.isfinite(n_triangles) else np.nan,
        "radius": radius,
        "n_h0_finite": int(len(h0_finite)),
        "n_h1_finite": int(len(h1_finite)),
        "n_h1_essential": int(len(h1_essential)),
        "n_h1_capped_for_fingerprint": int(n_capped_h1),
        "elapsed_sec": elapsed,
    }

    loaded_channels.append(ch_name)

    log(
        f"{prefix} done {elapsed:.2f}s | "
        f"edges={n_edges}, triangles={n_triangles}, "
        f"H1 finite={len(h1_finite)}, H1 fingerprint diagram={len(h1_diagram)}"
    )

if len(loaded_channels) == 0:
    raise RuntimeError("No usable embeddings found. Check embedding_data paths.")

if global_birth_max is None or not np.isfinite(global_birth_max) or global_birth_max <= 0:
    global_birth_max = 1.0

if global_persistence_max is None or not np.isfinite(global_persistence_max) or global_persistence_max <= 0:
    global_persistence_max = 1.0

if global_filtration_max is None or not np.isfinite(global_filtration_max) or global_filtration_max <= 0:
    global_filtration_max = 1.0

birth_max_for_grid = global_birth_max * 1.0001
persistence_max_for_grid = global_persistence_max * 1.0001
filtration_max_for_grid = global_filtration_max * 1.0001

birth_edges = np.linspace(0.0, birth_max_for_grid, IMAGE_BIRTH_BINS + 1)
persistence_edges = np.linspace(0.0, persistence_max_for_grid, IMAGE_PERSISTENCE_BINS + 1)
birth_grid = 0.5 * (birth_edges[:-1] + birth_edges[1:])
persistence_grid = 0.5 * (persistence_edges[:-1] + persistence_edges[1:])

betti_grid = np.linspace(0.0, filtration_max_for_grid, BETTI_GRID_SIZE)
landscape_grid = np.linspace(0.0, filtration_max_for_grid, LANDSCAPE_GRID_SIZE)

sigma_birth = max(1e-8, IMAGE_SIGMA_FRAC * birth_max_for_grid)
sigma_persistence = max(1e-8, IMAGE_SIGMA_FRAC * persistence_max_for_grid)

log("\nBuilding topology fingerprint features")
log(f"birth range: 0 to {birth_max_for_grid:.6f}")
log(f"persistence range: 0 to {persistence_max_for_grid:.6f}")
log(f"filtration range: 0 to {filtration_max_for_grid:.6f}")
log(f"persistence image sigma birth={sigma_birth:.8f}, persistence={sigma_persistence:.8f}")

fingerprint_rows = []
betti_store = {}
landscape_store = {}
image_store = {}

for ch_name in loaded_channels:
    diagram = diagram_store[ch_name]
    bp = bp_store[ch_name]

    curve = betti_curve(diagram, betti_grid)
    landscapes = persistence_landscapes(diagram, landscape_grid, n_levels=LANDSCAPE_LEVELS)
    image = persistence_image(
        bp,
        birth_grid=birth_grid,
        persistence_grid=persistence_grid,
        sigma_birth=sigma_birth,
        sigma_persistence=sigma_persistence,
        weight_power=IMAGE_WEIGHT_POWER,
    )

    betti_store[ch_name] = curve
    landscape_store[ch_name] = landscapes
    image_store[ch_name] = image

    row = {}
    row.update(diagram_core_features(diagram))
    row.update(betti_features(curve, betti_grid))
    row.update(landscape_features(landscapes, landscape_grid))
    row.update(image_features(image, birth_grid, persistence_grid))
    row.update(top_lifetime_features(diagram, top_k=TOP_LIFETIMES))

    row["radius"] = diagnostics[ch_name]["radius"]
    row["n_edges_at_radius"] = diagnostics[ch_name]["n_edges_at_radius"]
    row["n_triangles_at_radius"] = diagnostics[ch_name]["n_triangles_at_radius"]
    row["n_points"] = diagnostics[ch_name]["n_points"]

    fingerprint_rows.append(row)

fingerprint_df = pd.DataFrame(fingerprint_rows, index=loaded_channels)
fingerprint_df = fingerprint_df.replace([np.inf, -np.inf], np.nan).fillna(0.0)

fingerprint_z_df, fingerprint_mean, fingerprint_std = standardize_features(fingerprint_df)

fingerprint_matrix = fingerprint_df.to_numpy(dtype=float)
fingerprint_z_matrix = fingerprint_z_df.to_numpy(dtype=float)

fingerprint_distance_mat = pairwise_euclidean_matrix(fingerprint_z_matrix)
fingerprint_cosine_similarity_mat = cosine_similarity_matrix(fingerprint_z_matrix)
fingerprint_cosine_distance_mat = np.maximum(0.0, 1.0 - fingerprint_cosine_similarity_mat)
np.fill_diagonal(fingerprint_cosine_distance_mat, 0.0)

fingerprint_norm = np.linalg.norm(fingerprint_z_matrix, axis=1)

complexity_features = [
    "n_intervals",
    "total_lifetime",
    "lifetime_entropy",
    "lifetime_q90",
    "lifetime_max",
    "betti_auc",
    "betti_max",
    "landscape_l2",
    "image_mass",
    "image_l2",
    "n_triangles_at_radius",
]

loop_strength_features = [
    "total_lifetime",
    "lifetime_mean",
    "lifetime_q75",
    "lifetime_q90",
    "lifetime_max",
    "landscape_l2",
    "landscape_max",
    "image_l2",
    "top_lifetime_01",
]

activity_features = [
    "n_intervals",
    "betti_mean",
    "betti_auc",
    "betti_nonzero_fraction",
    "landscape_l1",
    "image_mass",
]

spread_features = [
    "birth_std",
    "death_std",
    "lifetime_std",
    "midpoint_std",
    "image_spread_birth",
    "image_spread_persistence",
]

structure_density_features = [
    "n_edges_at_radius",
    "n_triangles_at_radius",
    "n_points",
    "betti_max",
    "betti_peak_count",
]

complexity_z = mean_z_score(fingerprint_z_df, complexity_features)
loop_strength_z = mean_z_score(fingerprint_z_df, loop_strength_features)
activity_z = mean_z_score(fingerprint_z_df, activity_features)
spread_z = mean_z_score(fingerprint_z_df, spread_features)
structure_density_z = mean_z_score(fingerprint_z_df, structure_density_features)

mean_distance_to_others = []
median_distance_to_others = []
max_distance_to_others = []
min_distance_to_others = []
mean_cosine_to_others = []

for idx in range(len(loaded_channels)):
    mask = np.arange(len(loaded_channels)) != idx

    if np.any(mask):
        mean_distance_to_others.append(float(np.nanmean(fingerprint_distance_mat[idx, mask])))
        median_distance_to_others.append(float(np.nanmedian(fingerprint_distance_mat[idx, mask])))
        max_distance_to_others.append(float(np.nanmax(fingerprint_distance_mat[idx, mask])))
        min_distance_to_others.append(float(np.nanmin(fingerprint_distance_mat[idx, mask])))
        mean_cosine_to_others.append(float(np.nanmean(fingerprint_cosine_similarity_mat[idx, mask])))
    else:
        mean_distance_to_others.append(0.0)
        median_distance_to_others.append(0.0)
        max_distance_to_others.append(0.0)
        min_distance_to_others.append(0.0)
        mean_cosine_to_others.append(1.0)

uniqueness_score = percentile_score(mean_distance_to_others, higher_is_more=True)
complexity_score = percentile_score(complexity_z, higher_is_more=True)
loop_strength_score = percentile_score(loop_strength_z, higher_is_more=True)
activity_score = percentile_score(activity_z, higher_is_more=True)
spread_score = percentile_score(spread_z, higher_is_more=True)
structure_density_score = percentile_score(structure_density_z, higher_is_more=True)

fingerprint_score_df = pd.DataFrame({
    "channel": loaded_channels,
    "region": [channel_regions.get(ch, "Unknown") for ch in loaded_channels],
    "topology_complexity_z": complexity_z,
    "topology_complexity_score": complexity_score,
    "loop_strength_z": loop_strength_z,
    "loop_strength_score": loop_strength_score,
    "topological_activity_z": activity_z,
    "topological_activity_score": activity_score,
    "topological_spread_z": spread_z,
    "topological_spread_score": spread_score,
    "structure_density_z": structure_density_z,
    "structure_density_score": structure_density_score,
    "topological_uniqueness_score": uniqueness_score,
    "fingerprint_norm": fingerprint_norm,
    "mean_fingerprint_distance_to_others": mean_distance_to_others,
    "median_fingerprint_distance_to_others": median_distance_to_others,
    "max_fingerprint_distance_to_others": max_distance_to_others,
    "min_fingerprint_distance_to_others": min_distance_to_others,
    "mean_fingerprint_cosine_similarity_to_others": mean_cosine_to_others,
})

summary_rows = []

for ch_name in loaded_channels:
    base = dict(diagnostics[ch_name])
    fp = fingerprint_df.loc[ch_name].to_dict()
    scores = fingerprint_score_df.loc[fingerprint_score_df["channel"] == ch_name].iloc[0].to_dict()

    combined = {}
    combined.update(base)
    combined.update(fp)
    combined.update(scores)

    summary_rows.append(combined)

summary_df = (
    pd.DataFrame(summary_rows)
    .sort_values("topology_complexity_score", ascending=False)
    .reset_index(drop=True)
)

pair_rows = []

for i in range(len(loaded_channels)):
    for j in range(i + 1, len(loaded_channels)):
        pair_rows.append({
            "channel_a": loaded_channels[i],
            "channel_b": loaded_channels[j],
            "fingerprint_euclidean_distance": float(fingerprint_distance_mat[i, j]),
            "fingerprint_cosine_similarity": float(fingerprint_cosine_similarity_mat[i, j]),
            "fingerprint_cosine_distance": float(fingerprint_cosine_distance_mat[i, j]),
        })

pairwise_df = (
    pd.DataFrame(pair_rows)
    .sort_values("fingerprint_euclidean_distance", ascending=False)
    .reset_index(drop=True)
)

feature_variance_df = pd.DataFrame({
    "feature_name": fingerprint_df.columns,
    "standardized_variance": np.var(fingerprint_z_matrix, axis=0),
    "mean_abs_standardized_value": np.mean(np.abs(fingerprint_z_matrix), axis=0),
}).sort_values("standardized_variance", ascending=False).reset_index(drop=True)

pca_coords, pca_explained, pca_components = pca_from_feature_matrix(fingerprint_z_matrix, n_components=2)
mds_coords = classical_mds_from_distances(fingerprint_distance_mat, n_components=2)

pca_df = pd.DataFrame({
    "channel": loaded_channels,
    "region": [channel_regions.get(ch, "Unknown") for ch in loaded_channels],
    "pc1": pca_coords[:, 0],
    "pc2": pca_coords[:, 1],
    "topology_complexity_score": complexity_score,
    "loop_strength_score": loop_strength_score,
    "topological_uniqueness_score": uniqueness_score,
    "mean_fingerprint_distance_to_others": mean_distance_to_others,
})

mds_df = pd.DataFrame({
    "channel": loaded_channels,
    "region": [channel_regions.get(ch, "Unknown") for ch in loaded_channels],
    "mds_1": mds_coords[:, 0],
    "mds_2": mds_coords[:, 1],
    "topology_complexity_score": complexity_score,
    "loop_strength_score": loop_strength_score,
    "topological_uniqueness_score": uniqueness_score,
    "mean_fingerprint_distance_to_others": mean_distance_to_others,
})

score_cols = [
    "topology_complexity_score",
    "loop_strength_score",
    "topological_activity_score",
    "topological_spread_score",
    "structure_density_score",
    "topological_uniqueness_score",
]

score_matrix_df = fingerprint_score_df.set_index("channel")[score_cols]

summary_csv_path = os.path.join(mode_out_dir, "h1_topology_channel_fingerprint_summary.csv")
fingerprint_csv_path = os.path.join(mode_out_dir, "h1_topology_channel_fingerprint_matrix.csv")
fingerprint_z_csv_path = os.path.join(mode_out_dir, "h1_topology_channel_fingerprint_matrix_standardized.csv")
score_csv_path = os.path.join(mode_out_dir, "h1_topology_channel_fingerprint_scores.csv")
feature_variance_csv_path = os.path.join(mode_out_dir, "h1_topology_fingerprint_feature_variance_ranking.csv")
distance_csv_path = os.path.join(mode_out_dir, "h1_topology_fingerprint_euclidean_distance_matrix.csv")
cosine_similarity_csv_path = os.path.join(mode_out_dir, "h1_topology_fingerprint_cosine_similarity_matrix.csv")
pairwise_csv_path = os.path.join(mode_out_dir, "h1_topology_fingerprint_pairwise_distances.csv")
pca_csv_path = os.path.join(mode_out_dir, "h1_topology_fingerprint_pca_coordinates.csv")
mds_csv_path = os.path.join(mode_out_dir, "h1_topology_fingerprint_mds_coordinates.csv")

summary_df.to_csv(summary_csv_path, index=False)
fingerprint_df.to_csv(fingerprint_csv_path)
fingerprint_z_df.to_csv(fingerprint_z_csv_path)
fingerprint_score_df.to_csv(score_csv_path, index=False)
feature_variance_df.to_csv(feature_variance_csv_path, index=False)

pd.DataFrame(fingerprint_distance_mat, index=loaded_channels, columns=loaded_channels).to_csv(distance_csv_path)
pd.DataFrame(fingerprint_cosine_similarity_mat, index=loaded_channels, columns=loaded_channels).to_csv(cosine_similarity_csv_path)
pairwise_df.to_csv(pairwise_csv_path, index=False)
pca_df.to_csv(pca_csv_path, index=False)
mds_df.to_csv(mds_csv_path, index=False)

npz_path = os.path.join(mode_out_dir, "h1_topology_channel_fingerprints_outputs.npz")

save_dict = {
    "channels": np.asarray(loaded_channels),
    "fingerprint_feature_names": np.asarray(fingerprint_df.columns),
    "fingerprint_matrix": fingerprint_matrix,
    "fingerprint_matrix_standardized": fingerprint_z_matrix,
    "fingerprint_feature_mean": fingerprint_mean,
    "fingerprint_feature_std": fingerprint_std,
    "fingerprint_distance_matrix": fingerprint_distance_mat,
    "fingerprint_cosine_similarity_matrix": fingerprint_cosine_similarity_mat,
    "fingerprint_score_names": np.asarray(score_cols),
    "fingerprint_score_matrix": score_matrix_df.to_numpy(dtype=float),
    "pca_coordinates": pca_coords,
    "pca_explained_variance_ratio": pca_explained,
    "pca_components": pca_components,
    "mds_coordinates": mds_coords,
    "betti_grid": betti_grid,
    "landscape_grid": landscape_grid,
    "birth_grid": birth_grid,
    "persistence_grid": persistence_grid,
    "sigma_birth": np.asarray([sigma_birth]),
    "sigma_persistence": np.asarray([sigma_persistence]),
    "fast_option": np.asarray([FAST_OPTION]),
    "backend": np.asarray(["ripser_presampled_fps + topology channel fingerprints"]),
}

for ch_name in loaded_channels:
    save_dict[f"{ch_name}_H1_fingerprint_diagram"] = diagram_store[ch_name]
    save_dict[f"{ch_name}_H1_birth_persistence"] = bp_store[ch_name]
    save_dict[f"{ch_name}_H1_betti_curve"] = betti_store[ch_name]
    save_dict[f"{ch_name}_H1_landscapes"] = landscape_store[ch_name]
    save_dict[f"{ch_name}_H1_persistence_image"] = image_store[ch_name]

    for key, val in raw_interval_store[ch_name].items():
        save_dict[f"{ch_name}_{key}"] = val

np.savez(npz_path, **save_dict)

print("\nH1 Topology-Based Channel Fingerprint Summary:")
if HAVE_IPYTHON:
    display(summary_df)
else:
    print(summary_df)

print("\nTopology Fingerprint Scores:")
if HAVE_IPYTHON:
    display(fingerprint_score_df.sort_values("topology_complexity_score", ascending=False))
else:
    print(fingerprint_score_df.sort_values("topology_complexity_score", ascending=False))

print("\nMost different channel pairs by topology fingerprint:")
if HAVE_IPYTHON:
    display(pairwise_df.head(12))
else:
    print(pairwise_df.head(12))

print("\nTop variable fingerprint features:")
if HAVE_IPYTHON:
    display(feature_variance_df.head(20))
else:
    print(feature_variance_df.head(20))

log(f"\nSaved summary CSV: {summary_csv_path}")
log(f"Saved fingerprint matrix CSV: {fingerprint_csv_path}")
log(f"Saved standardized fingerprint matrix CSV: {fingerprint_z_csv_path}")
log(f"Saved fingerprint scores CSV: {score_csv_path}")
log(f"Saved feature variance ranking CSV: {feature_variance_csv_path}")
log(f"Saved distance matrix CSV: {distance_csv_path}")
log(f"Saved cosine similarity matrix CSV: {cosine_similarity_csv_path}")
log(f"Saved pairwise distances CSV: {pairwise_csv_path}")
log(f"Saved PCA CSV: {pca_csv_path}")
log(f"Saved MDS CSV: {mds_csv_path}")
log(f"Saved NPZ: {npz_path}")

plot_score_df = fingerprint_score_df.sort_values("topology_complexity_score", ascending=False).reset_index(drop=True)

fig, ax = plt.subplots(figsize=(13, 5.8), facecolor=BLACK)
style_ax(
    ax,
    title=f"Topology Complexity Fingerprint Score by Channel | {FAST_OPTION.upper()}",
    xlabel="Channel",
    ylabel="Topology complexity score",
)

x_pos = np.arange(len(plot_score_df))

ax.bar(
    x_pos,
    plot_score_df["topology_complexity_score"],
    color=ACCENT,
    edgecolor=ACCENT,
    alpha=0.78,
)

ax.axhline(
    np.nanmean(plot_score_df["topology_complexity_score"]),
    color=ACCENT_SOFT,
    linestyle="--",
    linewidth=1.4,
    alpha=0.8,
    label=f"Mean={np.nanmean(plot_score_df['topology_complexity_score']):.2f}",
)

ax.set_xticks(x_pos)
ax.set_xticklabels(plot_score_df["channel"], rotation=90)
ax.set_ylim(0, 105)
ax.legend(loc="best", framealpha=1.0)

complexity_path = os.path.join(mode_out_dir, "01_topology_complexity_fingerprint_score_by_channel.png")
plt.tight_layout()
save_render_close(fig, complexity_path)

score_plot_matrix = score_matrix_df.loc[plot_score_df["channel"]].to_numpy(dtype=float)

fig, ax = plt.subplots(figsize=(10, 9), facecolor=BLACK)
style_ax(
    ax,
    title="Topology-Based Channel Fingerprint Score Matrix",
    xlabel="Fingerprint score",
    ylabel="Channel",
)

im = ax.imshow(
    score_plot_matrix,
    aspect="auto",
    interpolation="nearest",
    cmap=CYAN_SEQUENTIAL_CMAP,
    vmin=0,
    vmax=100,
)

ax.set_yticks(np.arange(len(plot_score_df)))
ax.set_yticklabels(plot_score_df["channel"])
ax.set_xticks(np.arange(len(score_cols)))
ax.set_xticklabels(
    [
        "Complexity",
        "Loop strength",
        "Activity",
        "Spread",
        "Density",
        "Uniqueness",
    ],
    rotation=45,
    ha="right",
)

cbar = plt.colorbar(im, ax=ax, shrink=0.85)
style_colorbar(cbar, "Score percentile")

score_heatmap_path = os.path.join(mode_out_dir, "02_topology_fingerprint_score_heatmap.png")
plt.tight_layout()
save_render_close(fig, score_heatmap_path, width=900)

if len(loaded_channels) >= 3 and np.any(fingerprint_distance_mat > 0):
    condensed = fingerprint_distance_mat[np.triu_indices(len(loaded_channels), k=1)]
    Z = linkage(condensed, method="average")
    order = leaves_list(Z)
else:
    order = np.arange(len(loaded_channels))

ordered_channels = [loaded_channels[i] for i in order]
ordered_distance = fingerprint_distance_mat[np.ix_(order, order)]
ordered_cosine = fingerprint_cosine_similarity_mat[np.ix_(order, order)]

dist_vmin, dist_vmax = dynamic_limits(
    offdiag_values(ordered_distance),
    lower_pct=2,
    upper_pct=98,
    pad_frac=0.10,
    lower_bound=0.0,
)

save_matrix_plot(
    ordered_distance,
    ordered_channels,
    "Topology Fingerprint Euclidean Distance Matrix",
    "Fingerprint distance",
    os.path.join(mode_out_dir, "03_topology_fingerprint_euclidean_distance_heatmap.png"),
    vmin=dist_vmin,
    vmax=dist_vmax,
)

cos_vmin, cos_vmax = dynamic_limits(
    offdiag_values(ordered_cosine),
    lower_pct=2,
    upper_pct=98,
    pad_frac=0.10,
    lower_bound=-1.0,
    upper_bound=1.0,
)

save_matrix_plot(
    ordered_cosine,
    ordered_channels,
    "Topology Fingerprint Cosine Similarity Matrix",
    "Cosine similarity",
    os.path.join(mode_out_dir, "04_topology_fingerprint_cosine_similarity_heatmap.png"),
    vmin=cos_vmin,
    vmax=cos_vmax,
)

top_n_features = min(40, len(feature_variance_df))
top_features = feature_variance_df.head(top_n_features)["feature_name"].tolist()

feature_heatmap_matrix = fingerprint_z_df.loc[ordered_channels, top_features].to_numpy(dtype=float)

fig, ax = plt.subplots(figsize=(14, 9), facecolor=BLACK)
style_ax(
    ax,
    title=f"Top {top_n_features} Variable Topology Fingerprint Features",
    xlabel="Fingerprint feature",
    ylabel="Channel",
)

vlim = np.nanpercentile(np.abs(feature_heatmap_matrix), 98) if np.any(np.isfinite(feature_heatmap_matrix)) else 1.0
vlim = max(float(vlim), 1e-6)

im = ax.imshow(
    feature_heatmap_matrix,
    aspect="auto",
    interpolation="nearest",
    cmap=CYAN_SEQUENTIAL_CMAP,
    vmin=-vlim,
    vmax=vlim,
)

ax.set_yticks(np.arange(len(ordered_channels)))
ax.set_yticklabels(ordered_channels)
ax.set_xticks(np.arange(top_n_features))
ax.set_xticklabels(top_features, rotation=90, fontsize=7)

cbar = plt.colorbar(im, ax=ax, shrink=0.85)
style_colorbar(cbar, "Standardized feature value")

feature_heatmap_path = os.path.join(mode_out_dir, "05_top_variable_topology_fingerprint_features_heatmap.png")
plt.tight_layout()
save_render_close(fig, feature_heatmap_path, width=1200)

fig, ax = plt.subplots(figsize=(8, 7), facecolor=BLACK)
style_ax(
    ax,
    title=f"PCA of Topology Channel Fingerprints | PC1={pca_explained[0]*100:.1f}%, PC2={pca_explained[1]*100:.1f}%",
    xlabel="PC1",
    ylabel="PC2",
)

pca_color = pca_df["topology_complexity_score"].to_numpy(dtype=float)

sc = ax.scatter(
    pca_df["pc1"],
    pca_df["pc2"],
    c=pca_color,
    s=150,
    cmap=CYAN_SEQUENTIAL_CMAP,
    edgecolor=ACCENT,
    linewidth=1.1,
    vmin=0,
    vmax=100,
)

for _, row in pca_df.iterrows():
    ax.text(
        row["pc1"],
        row["pc2"],
        row["channel"],
        fontsize=8,
        color=ACCENT,
        alpha=0.9,
    )

cbar = plt.colorbar(sc, ax=ax, shrink=0.85)
style_colorbar(cbar, "Complexity score")

pca_path = os.path.join(mode_out_dir, "06_topology_fingerprint_pca_embedding.png")
plt.tight_layout()
save_render_close(fig, pca_path)

fig, ax = plt.subplots(figsize=(8, 7), facecolor=BLACK)
style_ax(
    ax,
    title="Classical MDS of Topology Fingerprint Distances",
    xlabel="MDS 1",
    ylabel="MDS 2",
)

sc = ax.scatter(
    mds_df["mds_1"],
    mds_df["mds_2"],
    c=mds_df["topological_uniqueness_score"],
    s=150,
    cmap=CYAN_SEQUENTIAL_CMAP,
    edgecolor=ACCENT,
    linewidth=1.1,
    vmin=0,
    vmax=100,
)

for _, row in mds_df.iterrows():
    ax.text(
        row["mds_1"],
        row["mds_2"],
        row["channel"],
        fontsize=8,
        color=ACCENT,
        alpha=0.9,
    )

cbar = plt.colorbar(sc, ax=ax, shrink=0.85)
style_colorbar(cbar, "Uniqueness score")

mds_path = os.path.join(mode_out_dir, "07_topology_fingerprint_mds_embedding.png")
plt.tight_layout()
save_render_close(fig, mds_path)

fig, ax = plt.subplots(figsize=(12, 6), facecolor=BLACK)
style_ax(
    ax,
    title="Most Different Channel Pairs by Topology Fingerprint",
    xlabel="Channel pair",
    ylabel="Fingerprint distance",
)

top_pairs = pairwise_df.head(min(15, len(pairwise_df))).copy()
pair_labels = top_pairs["channel_a"] + "-" + top_pairs["channel_b"]

x_pair = np.arange(len(top_pairs))

ax.bar(
    x_pair,
    top_pairs["fingerprint_euclidean_distance"],
    color=ACCENT,
    edgecolor=ACCENT,
    alpha=0.78,
)

ax.set_xticks(x_pair)
ax.set_xticklabels(pair_labels, rotation=60, ha="right")

top_pairs_path = os.path.join(mode_out_dir, "08_most_different_channel_pairs_topology_fingerprint.png")
plt.tight_layout()
save_render_close(fig, top_pairs_path)

fig, ax = plt.subplots(figsize=(12, 6), facecolor=BLACK)
style_ax(
    ax,
    title="Most Similar Channel Pairs by Topology Fingerprint",
    xlabel="Channel pair",
    ylabel="Fingerprint distance",
)

bottom_pairs = pairwise_df.tail(min(15, len(pairwise_df))).copy().sort_values("fingerprint_euclidean_distance", ascending=True)
bottom_pair_labels = bottom_pairs["channel_a"] + "-" + bottom_pairs["channel_b"]

x_pair = np.arange(len(bottom_pairs))

ax.bar(
    x_pair,
    bottom_pairs["fingerprint_euclidean_distance"],
    color=ACCENT,
    edgecolor=ACCENT,
    alpha=0.78,
)

ax.set_xticks(x_pair)
ax.set_xticklabels(bottom_pair_labels, rotation=60, ha="right")

bottom_pairs_path = os.path.join(mode_out_dir, "09_most_similar_channel_pairs_topology_fingerprint.png")
plt.tight_layout()
save_render_close(fig, bottom_pairs_path)

fig, ax = plt.subplots(figsize=(8, 6), facecolor=BLACK)
style_ax(
    ax,
    title="Loop Strength vs Topological Uniqueness",
    xlabel="Loop strength score",
    ylabel="Topological uniqueness score",
)

sc = ax.scatter(
    fingerprint_score_df["loop_strength_score"],
    fingerprint_score_df["topological_uniqueness_score"],
    c=fingerprint_score_df["topology_complexity_score"],
    s=150,
    cmap=CYAN_SEQUENTIAL_CMAP,
    edgecolor=ACCENT,
    linewidth=1.1,
    vmin=0,
    vmax=100,
)

for _, row in fingerprint_score_df.iterrows():
    ax.text(
        row["loop_strength_score"],
        row["topological_uniqueness_score"],
        row["channel"],
        fontsize=8,
        color=ACCENT,
        alpha=0.9,
    )

cbar = plt.colorbar(sc, ax=ax, shrink=0.85)
style_colorbar(cbar, "Complexity score")

loop_unique_path = os.path.join(mode_out_dir, "10_loop_strength_vs_topological_uniqueness.png")
plt.tight_layout()
save_render_close(fig, loop_unique_path)

if MAKE_EXPERT_PLOTS:
    expert_dir = os.path.join(mode_out_dir, "expert_plots")
    os.makedirs(expert_dir, exist_ok=True)

    complexity_map = dict(zip(fingerprint_score_df["channel"], fingerprint_score_df["topology_complexity_score"]))
    uniqueness_map = dict(zip(fingerprint_score_df["channel"], fingerprint_score_df["topological_uniqueness_score"]))

    scalp_xy = {
        "Fp1": (-0.45, 1.00), "Fpz": (0.00, 1.08), "Fp2": (0.45, 1.00),
        "F7": (-0.95, 0.60), "F3": (-0.45, 0.55), "Fz": (0.00, 0.60),
        "F4": (0.45, 0.55), "F8": (0.95, 0.60),
        "FC5": (-0.75, 0.25), "FC1": (-0.25, 0.25),
        "FC2": (0.25, 0.25), "FC6": (0.75, 0.25),
        "M1": (-1.15, 0.05), "T7": (-1.00, 0.00), "C3": (-0.50, 0.00),
        "Cz": (0.00, 0.00), "C4": (0.50, 0.00), "T8": (1.00, 0.00),
        "M2": (1.15, 0.05),
        "CP5": (-0.75, -0.30), "CP1": (-0.25, -0.30),
        "CP2": (0.25, -0.30), "CP6": (0.75, -0.30),
        "P7": (-0.90, -0.60), "P3": (-0.45, -0.60), "Pz": (0.00, -0.65),
        "P4": (0.45, -0.60), "P8": (0.90, -0.60),
        "POz": (0.00, -0.88),
        "O1": (-0.35, -1.00), "Oz": (0.00, -1.08), "O2": (0.35, -1.00),
    }

    xs, ys, vals, labels = [], [], [], []

    for ch in eeg_channels:
        if ch in scalp_xy and ch in complexity_map and np.isfinite(complexity_map[ch]):
            xs.append(scalp_xy[ch][0])
            ys.append(scalp_xy[ch][1])
            vals.append(complexity_map[ch])
            labels.append(ch)

    fig, ax = plt.subplots(figsize=(7, 7), facecolor=BLACK)
    style_ax(
        ax,
        title=f"Scalp Map of Topology Complexity Fingerprint Score | {FAST_OPTION.upper()}",
        grid=False,
    )

    head = plt.Circle((0, 0), 1.12, fill=False, color=ACCENT, linewidth=2)
    ax.add_patch(head)
    ax.plot([0.00, -0.10, 0.10, 0.00], [1.12, 1.28, 1.28, 1.12], color=ACCENT, linewidth=2)

    sc = ax.scatter(
        xs,
        ys,
        c=vals,
        s=260,
        cmap=CYAN_SEQUENTIAL_CMAP,
        edgecolor=ACCENT,
        linewidth=1.2,
        vmin=0,
        vmax=100,
    )

    for x0, y0, label in zip(xs, ys, labels):
        ax.text(
            x0,
            y0,
            label,
            ha="center",
            va="center",
            fontsize=8,
            color=BLACK,
            fontweight="bold",
        )

    cbar = plt.colorbar(sc, ax=ax, shrink=0.75)
    style_colorbar(cbar, "Complexity score")

    ax.set_xlim(-1.35, 1.35)
    ax.set_ylim(-1.35, 1.35)
    ax.set_aspect("equal")
    ax.set_xticks([])
    ax.set_yticks([])

    scalp_complexity_path = os.path.join(expert_dir, "11_topology_complexity_score_scalp_map.png")
    plt.tight_layout()
    save_render_close(fig, scalp_complexity_path, width=760)

    xs, ys, vals, labels = [], [], [], []

    for ch in eeg_channels:
        if ch in scalp_xy and ch in uniqueness_map and np.isfinite(uniqueness_map[ch]):
            xs.append(scalp_xy[ch][0])
            ys.append(scalp_xy[ch][1])
            vals.append(uniqueness_map[ch])
            labels.append(ch)

    fig, ax = plt.subplots(figsize=(7, 7), facecolor=BLACK)
    style_ax(
        ax,
        title=f"Scalp Map of Topological Uniqueness Score | {FAST_OPTION.upper()}",
        grid=False,
    )

    head = plt.Circle((0, 0), 1.12, fill=False, color=ACCENT, linewidth=2)
    ax.add_patch(head)
    ax.plot([0.00, -0.10, 0.10, 0.00], [1.12, 1.28, 1.28, 1.12], color=ACCENT, linewidth=2)

    sc = ax.scatter(
        xs,
        ys,
        c=vals,
        s=260,
        cmap=CYAN_SEQUENTIAL_CMAP,
        edgecolor=ACCENT,
        linewidth=1.2,
        vmin=0,
        vmax=100,
    )

    for x0, y0, label in zip(xs, ys, labels):
        ax.text(
            x0,
            y0,
            label,
            ha="center",
            va="center",
            fontsize=8,
            color=BLACK,
            fontweight="bold",
        )

    cbar = plt.colorbar(sc, ax=ax, shrink=0.75)
    style_colorbar(cbar, "Uniqueness score")

    ax.set_xlim(-1.35, 1.35)
    ax.set_ylim(-1.35, 1.35)
    ax.set_aspect("equal")
    ax.set_xticks([])
    ax.set_yticks([])

    scalp_uniqueness_path = os.path.join(expert_dir, "12_topological_uniqueness_score_scalp_map.png")
    plt.tight_layout()
    save_render_close(fig, scalp_uniqueness_path, width=760)

    regions = {
        "Frontal": ["Fp1", "Fpz", "Fp2", "F7", "F3", "Fz", "F4", "F8"],
        "Frontocentral": ["FC5", "FC1", "FC2", "FC6"],
        "Central": ["T7", "C3", "Cz", "C4", "T8"],
        "Centroparietal": ["CP5", "CP1", "CP2", "CP6"],
        "Parietal": ["P7", "P3", "Pz", "P4", "P8"],
        "Occipital": ["POz", "O1", "Oz", "O2"],
        "Mastoid": ["M1", "M2"],
    }

    region_rows = []
    region_points = []

    for region_name, chans in regions.items():
        sub = fingerprint_score_df[fingerprint_score_df["channel"].isin(chans)].copy()

        if len(sub) == 0:
            continue

        region_rows.append({
            "region": region_name,
            "mean_complexity_score": float(sub["topology_complexity_score"].mean()),
            "sem_complexity_score": float(sub["topology_complexity_score"].std(ddof=1) / np.sqrt(len(sub))) if len(sub) > 1 else 0.0,
            "mean_loop_strength_score": float(sub["loop_strength_score"].mean()),
            "mean_activity_score": float(sub["topological_activity_score"].mean()),
            "mean_spread_score": float(sub["topological_spread_score"].mean()),
            "mean_density_score": float(sub["structure_density_score"].mean()),
            "mean_uniqueness_score": float(sub["topological_uniqueness_score"].mean()),
            "n_channels": int(len(sub)),
        })

        for _, row in sub.iterrows():
            region_points.append({
                "region": region_name,
                "channel": row["channel"],
                "topology_complexity_score": row["topology_complexity_score"],
                "topological_uniqueness_score": row["topological_uniqueness_score"],
            })

    region_df = pd.DataFrame(region_rows)
    region_points_df = pd.DataFrame(region_points)

    region_csv_path = os.path.join(expert_dir, "regional_topology_channel_fingerprint_summary.csv")
    region_df.to_csv(region_csv_path, index=False)

    if len(region_df) > 0:
        fig, ax = plt.subplots(figsize=(12, 6.5), facecolor=BLACK)
        style_ax(
            ax,
            title=f"Regional Topology Complexity Fingerprint Score +/- SEM | {FAST_OPTION.upper()}",
            xlabel="Region",
            ylabel="Topology complexity score",
        )

        x_region = np.arange(len(region_df))

        ax.bar(
            x_region,
            region_df["mean_complexity_score"],
            yerr=region_df["sem_complexity_score"],
            color=ACCENT,
            edgecolor=ACCENT,
            alpha=0.82,
            width=0.68,
            capsize=8,
            error_kw={
                "elinewidth": 2.2,
                "capthick": 2.2,
                "ecolor": ACCENT,
                "alpha": 1.0,
            },
            label="Mean +/- SEM",
        )

        rng = np.random.default_rng(7)

        for i, region_name in enumerate(region_df["region"]):
            vals_region_plot = region_points_df.loc[
                region_points_df["region"] == region_name,
                "topology_complexity_score",
            ].to_numpy(dtype=float)

            if len(vals_region_plot) == 0:
                continue

            jitter = rng.uniform(-0.13, 0.13, size=len(vals_region_plot))

            ax.scatter(
                np.full(len(vals_region_plot), i) + jitter,
                vals_region_plot,
                s=42,
                facecolors=BLACK,
                edgecolors=ACCENT,
                linewidths=1.2,
                alpha=0.95,
                zorder=4,
            )

        for i, row in region_df.iterrows():
            y_val = row["mean_complexity_score"]
            sem = row["sem_complexity_score"]

            ax.text(
                i,
                y_val + sem,
                f"{y_val:.1f}\n+/-{sem:.1f}\nn={int(row['n_channels'])}",
                ha="center",
                va="bottom",
                fontsize=8,
                color=ACCENT,
            )

        ax.set_xticks(x_region)
        ax.set_xticklabels(
            region_df["region"],
            rotation=35,
            ha="right",
            rotation_mode="anchor",
        )
        ax.set_ylim(0, 105)
        ax.margins(x=0.04)
        ax.legend(loc="best", framealpha=1.0)

        region_path = os.path.join(expert_dir, "13_regional_topology_complexity_fingerprint_summary.png")
        plt.tight_layout(pad=2.0)
        save_render_close(fig, region_path)

    fig, ax = plt.subplots(figsize=(13, 6), facecolor=BLACK)
    style_ax(
        ax,
        title="Regional Topology Fingerprint Score Profiles",
        xlabel="Region",
        ylabel="Mean score",
    )

    region_profile_cols = [
        "mean_complexity_score",
        "mean_loop_strength_score",
        "mean_activity_score",
        "mean_spread_score",
        "mean_density_score",
        "mean_uniqueness_score",
    ]

    region_profile_labels = [
        "Complexity",
        "Loop strength",
        "Activity",
        "Spread",
        "Density",
        "Uniqueness",
    ]

    cyan_harmony_colors = [
        "#00D5FF",
        "#38BDF8",
        "#14B8A6",
        "#818CF8",
        "#A78BFA",
        "#F472B6",
    ]

    if len(region_df) > 0:
        x = np.arange(len(region_df))
        width = 0.12

        for idx, col in enumerate(region_profile_cols):
            ax.bar(
                x + (idx - 2.5) * width,
                region_df[col],
                width=width,
                color=cyan_harmony_colors[idx],
                edgecolor=cyan_harmony_colors[idx],
                linewidth=0.9,
                alpha=0.88,
                label=region_profile_labels[idx],
            )

        ax.set_xticks(x)
        ax.set_xticklabels(region_df["region"], rotation=35, ha="right")
        ax.set_ylim(0, 105)

        ax.grid(True, color=ACCENT_DIM, alpha=0.23, linewidth=0.8)

        legend = ax.legend(
            loc="upper left",
            framealpha=1.0,
            fontsize=8,
            facecolor=BLACK,
            edgecolor=ACCENT,
        )

        for text in legend.get_texts():
            text.set_color(ACCENT)

    region_profile_path = os.path.join(expert_dir, "14_regional_topology_fingerprint_score_profiles.png")
    plt.tight_layout()
    save_render_close(fig, region_profile_path)

    fig, ax = plt.subplots(figsize=(8, 6), facecolor=BLACK)
    style_ax(
        ax,
        title="Diagram Size vs Topology Complexity Fingerprint Score",
        xlabel="H1 diagram intervals",
        ylabel="Topology complexity score",
    )

    sc = ax.scatter(
        summary_df["n_intervals"],
        summary_df["topology_complexity_score"],
        c=summary_df["lifetime_max"],
        s=140,
        cmap=CYAN_SEQUENTIAL_CMAP,
        edgecolor=ACCENT,
        linewidth=1.0,
    )

    for _, row in summary_df.iterrows():
        ax.text(
            row["n_intervals"],
            row["topology_complexity_score"],
            row["channel"],
            fontsize=8,
            color=ACCENT,
            alpha=0.85,
        )

    cbar = plt.colorbar(sc, ax=ax, shrink=0.85)
    style_colorbar(cbar, "Max lifetime")

    ax.set_ylim(0, 105)

    diagram_size_path = os.path.join(expert_dir, "15_diagram_size_vs_topology_complexity_score.png")
    plt.tight_layout()
    save_render_close(fig, diagram_size_path)

    fig, ax = plt.subplots(figsize=(8, 6), facecolor=BLACK)
    style_ax(
        ax,
        title="Threshold Triangles vs Topology Complexity Fingerprint Score",
        xlabel="Triangles at radius",
        ylabel="Topology complexity score",
    )

    sc = ax.scatter(
        summary_df["n_triangles_at_radius"],
        summary_df["topology_complexity_score"],
        c=summary_df["n_intervals"],
        s=140,
        cmap=CYAN_SEQUENTIAL_CMAP,
        edgecolor=ACCENT,
        linewidth=1.0,
    )

    for _, row in summary_df.iterrows():
        ax.text(
            row["n_triangles_at_radius"],
            row["topology_complexity_score"],
            row["channel"],
            fontsize=8,
            color=ACCENT,
            alpha=0.85,
        )

    cbar = plt.colorbar(sc, ax=ax, shrink=0.85)
    style_colorbar(cbar, "H1 intervals")

    ax.set_ylim(0, 105)

    triangle_path = os.path.join(expert_dir, "16_threshold_triangles_vs_topology_complexity_score.png")
    plt.tight_layout()
    save_render_close(fig, triangle_path)

    qc_df = summary_df.copy().sort_values("elapsed_sec", ascending=False).reset_index(drop=True)

    fig, ax = plt.subplots(figsize=(12, 5.8), facecolor=BLACK)
    style_ax(
        ax,
        title="Topology Channel Fingerprints Runtime QC",
        xlabel="Channel",
        ylabel="Elapsed seconds",
    )

    x_qc = np.arange(len(qc_df))

    ax.bar(
        x_qc,
        qc_df["elapsed_sec"],
        color=ACCENT,
        edgecolor=ACCENT,
        alpha=0.82,
    )

    ax.set_xticks(x_qc)
    ax.set_xticklabels(qc_df["channel"], rotation=90)

    qc_path = os.path.join(expert_dir, "17_runtime_qc.png")
    plt.tight_layout()
    save_render_close(fig, qc_path)

    log(f"\nExpert plots saved and rendered from: {expert_dir}")
    log(f"Regional summary CSV: {region_csv_path}")

elapsed_total = time.time() - overall_t0

log("\n" + "=" * 72)
log(f"Finished H1 Topology-Based Channel Fingerprints in {elapsed_total:.2f}s")
log(f"Mode: {FAST_OPTION}")
log(f"Backend: ripser_presampled_fps + topology channel fingerprint scores")
log(f"Channels completed: {len(loaded_channels)}")
log(f"Fingerprint matrix shape: {fingerprint_matrix.shape}")
log(f"PCA explained variance ratio: PC1={pca_explained[0]:.4f}, PC2={pca_explained[1]:.4f}")
log(f"Summary CSV: {summary_csv_path}")
log(f"Fingerprint matrix CSV: {fingerprint_csv_path}")
log(f"Standardized fingerprint matrix CSV: {fingerprint_z_csv_path}")
log(f"Fingerprint scores CSV: {score_csv_path}")
log(f"Feature ranking CSV: {feature_variance_csv_path}")
log(f"Distance matrix CSV: {distance_csv_path}")
log(f"Cosine similarity CSV: {cosine_similarity_csv_path}")
log(f"Pairwise CSV: {pairwise_csv_path}")
log(f"PCA CSV: {pca_csv_path}")
log(f"MDS CSV: {mds_csv_path}")
log(f"NPZ: {npz_path}")
log(f"Output directory: {mode_out_dir}")
log("=" * 72)
```

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patheffects as pe
from matplotlib.colors import LinearSegmentedColormap
from scipy.spatial.distance import pdist, squareform
from scipy.cluster.hierarchy import linkage, leaves_list, fcluster, dendrogram

try:
    from scipy.linalg import eigh
    HAVE_SCIPY_EIGH = True
except Exception:
    HAVE_SCIPY_EIGH = False

try:
    from IPython.display import display, Image
    HAVE_IPYTHON = True
except Exception:
    HAVE_IPYTHON = False


base_dir = "/home/a/projects/Complete-Neural-Signal-Analysis"

FAST_OPTION = "ultra"

SAVE_PLOTS_TO_DISK = True
RENDER_ALL_IMAGES_INLINE = True
INLINE_IMAGE_WIDTH = 1050
PLOT_DPI = 140

CLUSTER_LINKAGE_METHOD = "ward"
MIN_CLUSTERS = 2
MAX_CLUSTERS = 8
TOP_FEATURES_TO_PLOT = 45

BLACK = "#000000"
ACCENT = "#00FFFF"
ACCENT_SOFT = "#66FFFF"
ACCENT_DIM = "#008B8B"

CYAN_FILL = "#00D5FF"
CYAN_FILL_SOFT = "#38BDF8"
CYAN_FILL_DARK = "#006B73"
CYAN_BOUNDARY = "#00FFFF"

MARKERS = ["o", "s", "^", "D", "P", "X", "v", "*", "<", ">"]

CYAN_SEQUENTIAL_CMAP = LinearSegmentedColormap.from_list(
    "cyan_sequential",
    [BLACK, CYAN_FILL_DARK, CYAN_FILL, ACCENT_SOFT],
    N=256,
)

CYAN_DIVERGING_CMAP = LinearSegmentedColormap.from_list(
    "cyan_diverging",
    [ACCENT_SOFT, BLACK, CYAN_FILL],
    N=256,
)

plt.rcParams.update({
    "figure.facecolor": BLACK,
    "axes.facecolor": BLACK,
    "savefig.facecolor": BLACK,
    "text.color": ACCENT,
    "axes.labelcolor": ACCENT,
    "axes.edgecolor": ACCENT,
    "axes.titlecolor": ACCENT,
    "xtick.color": ACCENT,
    "ytick.color": ACCENT,
    "grid.color": ACCENT_DIM,
    "grid.alpha": 0.25,
    "legend.facecolor": BLACK,
    "legend.edgecolor": ACCENT,
    "legend.labelcolor": ACCENT,
    "font.size": 11,
})


def log(msg):
    print(msg, flush=True)


def style_ax(ax, title=None, xlabel=None, ylabel=None, grid=True):
    ax.set_facecolor(BLACK)

    for spine in ax.spines.values():
        spine.set_color(ACCENT)
        spine.set_linewidth(1.1)

    ax.tick_params(axis="x", colors=ACCENT, labelcolor=ACCENT)
    ax.tick_params(axis="y", colors=ACCENT, labelcolor=ACCENT)

    if title is not None:
        ax.set_title(title, color=ACCENT, pad=12)

    if xlabel is not None:
        ax.set_xlabel(xlabel, color=ACCENT, labelpad=10)

    if ylabel is not None:
        ax.set_ylabel(ylabel, color=ACCENT, labelpad=10)

    if grid:
        ax.grid(True, color=ACCENT_DIM, alpha=0.25, linewidth=0.8)


def force_cyan_text(ax):
    ax.title.set_color(ACCENT)
    ax.xaxis.label.set_color(ACCENT)
    ax.yaxis.label.set_color(ACCENT)

    for tick_label in ax.get_xticklabels():
        tick_label.set_color(ACCENT)

    for tick_label in ax.get_yticklabels():
        tick_label.set_color(ACCENT)


def style_colorbar(cbar, label):
    cbar.set_label(label, color=ACCENT)
    cbar.ax.yaxis.set_tick_params(color=ACCENT)
    plt.setp(cbar.ax.get_yticklabels(), color=ACCENT)
    cbar.outline.set_edgecolor(ACCENT)


def style_legend(legend):
    if legend is not None:
        legend.get_frame().set_facecolor(BLACK)
        legend.get_frame().set_edgecolor(ACCENT)

        for text in legend.get_texts():
            text.set_color(ACCENT)


def save_render_close(fig, path, width=INLINE_IMAGE_WIDTH):
    if SAVE_PLOTS_TO_DISK:
        fig.savefig(path, dpi=PLOT_DPI, bbox_inches="tight", facecolor=BLACK)

    plt.close(fig)

    if RENDER_ALL_IMAGES_INLINE and HAVE_IPYTHON and os.path.exists(path):
        display(Image(filename=path, width=width))


def dynamic_limits(values, lower_pct=2, upper_pct=98, pad_frac=0.08, lower_bound=None, upper_bound=None):
    values = np.asarray(values, dtype=float)
    values = values[np.isfinite(values)]

    if len(values) == 0:
        vmin, vmax = 0.0, 1.0
    else:
        vmin = float(np.percentile(values, lower_pct))
        vmax = float(np.percentile(values, upper_pct))

        if not np.isfinite(vmin):
            vmin = float(np.nanmin(values))

        if not np.isfinite(vmax):
            vmax = float(np.nanmax(values))

        if vmax <= vmin:
            center = float(np.nanmean(values))
            eps = max(1e-6, abs(center) * 0.01, 1e-5)
            vmin = center - eps
            vmax = center + eps

    pad = (vmax - vmin) * pad_frac
    vmin -= pad
    vmax += pad

    if lower_bound is not None:
        vmin = max(lower_bound, vmin)

    if upper_bound is not None:
        vmax = min(upper_bound, vmax)

    if vmax <= vmin:
        vmax = vmin + 1e-6

    return vmin, vmax


def offdiag_values(M):
    M = np.asarray(M, dtype=float)

    if M.shape[0] <= 1:
        return np.asarray([], dtype=float)

    mask = ~np.eye(M.shape[0], dtype=bool)
    vals = M[mask]
    vals = vals[np.isfinite(vals)]

    return vals


def silhouette_score_precomputed(D, labels):
    D = np.asarray(D, dtype=float)
    labels = np.asarray(labels)
    unique_labels = np.unique(labels)

    if len(unique_labels) < 2 or len(unique_labels) >= len(labels):
        return np.nan

    scores = []

    for i in range(len(labels)):
        own_label = labels[i]
        own_mask = labels == own_label
        other_own = own_mask.copy()
        other_own[i] = False

        if np.any(other_own):
            a = float(np.mean(D[i, other_own]))
        else:
            a = 0.0

        b_vals = []

        for other_label in unique_labels:
            if other_label == own_label:
                continue

            other_mask = labels == other_label

            if np.any(other_mask):
                b_vals.append(float(np.mean(D[i, other_mask])))

        if len(b_vals) == 0:
            scores.append(0.0)
            continue

        b = min(b_vals)
        denom = max(a, b)

        if denom <= 0:
            scores.append(0.0)
        else:
            scores.append((b - a) / denom)

    return float(np.mean(scores))


def relabel_clusters_by_dendrogram_order(labels, order):
    labels = np.asarray(labels)
    ordered_old_labels = []

    for idx in order:
        lab = labels[idx]

        if lab not in ordered_old_labels:
            ordered_old_labels.append(lab)

    label_map = {old: new_idx + 1 for new_idx, old in enumerate(ordered_old_labels)}

    return np.asarray([label_map[x] for x in labels], dtype=int), label_map


def classical_mds_from_distances(D, n_components=2):
    D = np.asarray(D, dtype=float)
    n = D.shape[0]

    if n < 2:
        return np.zeros((n, n_components), dtype=float)

    D2 = D ** 2
    J = np.eye(n) - np.ones((n, n)) / n
    B = -0.5 * J @ D2 @ J

    if HAVE_SCIPY_EIGH:
        vals, vecs = eigh(B)
    else:
        vals, vecs = np.linalg.eigh(B)

    idx = np.argsort(vals)[::-1]
    vals = vals[idx]
    vecs = vecs[:, idx]

    coords = np.zeros((n, n_components), dtype=float)

    for k in range(min(n_components, n)):
        if vals[k] > 0:
            coords[:, k] = vecs[:, k] * np.sqrt(vals[k])

    return coords


def pca_from_feature_matrix(X, n_components=2):
    X = np.asarray(X, dtype=float)

    if X.shape[0] == 0:
        return np.empty((0, n_components)), np.zeros(n_components), np.zeros((n_components, X.shape[1]))

    Xc = X - np.mean(X, axis=0, keepdims=True)

    U, S, Vt = np.linalg.svd(Xc, full_matrices=False)

    n_take = min(n_components, U.shape[1])

    coords = np.zeros((X.shape[0], n_components), dtype=float)
    coords[:, :n_take] = U[:, :n_take] * S[:n_take]

    var = S ** 2
    total = np.sum(var)

    explained = np.zeros(n_components, dtype=float)

    if total > 0:
        explained[:n_take] = var[:n_take] / total

    components = np.zeros((n_components, X.shape[1]), dtype=float)
    components[:n_take] = Vt[:n_take]

    return coords, explained, components


def safe_sem(x):
    x = np.asarray(x, dtype=float)
    x = x[np.isfinite(x)]

    if len(x) <= 1:
        return 0.0

    return float(np.std(x, ddof=1) / np.sqrt(len(x)))


def choose_existing_input_dir():
    candidate_dirs = [
        os.path.join(
            base_dir,
            "topology_channel_fingerprints",
            f"h1_topology_channel_fingerprints_{FAST_OPTION}_fps",
        ),
        os.path.join(
            base_dir,
            "topology_channel_fingerprints",
            f"h1_topology_channel_fingerprints_{FAST_OPTION}_fps_cyberpunk",
        ),
    ]

    for candidate_dir in candidate_dirs:
        feature_path = os.path.join(candidate_dir, "h1_topology_channel_fingerprint_matrix_standardized.csv")

        if os.path.exists(feature_path):
            return candidate_dir

    raise FileNotFoundError(
        "Could not find the standardized topology fingerprint matrix.\n\n"
        "Expected one of:\n"
        + "\n".join(
            os.path.join(d, "h1_topology_channel_fingerprint_matrix_standardized.csv")
            for d in candidate_dirs
        )
        + "\n\nRun the Topology-Based Channel Fingerprints cell first, then run this clustering cell."
    )


input_dir = choose_existing_input_dir()

feature_z_csv_path = os.path.join(input_dir, "h1_topology_channel_fingerprint_matrix_standardized.csv")
feature_raw_csv_path = os.path.join(input_dir, "h1_topology_channel_fingerprint_matrix.csv")
score_csv_path = os.path.join(input_dir, "h1_topology_channel_fingerprint_scores.csv")
summary_csv_path = os.path.join(input_dir, "h1_topology_channel_fingerprint_summary.csv")

cluster_out_dir = os.path.join(input_dir, "topological_clustering_of_channels")
diagnostic_out_dir = os.path.join(cluster_out_dir, "useful_cluster_diagnostics_cyan")
os.makedirs(cluster_out_dir, exist_ok=True)
os.makedirs(diagnostic_out_dir, exist_ok=True)

log("\n" + "=" * 72)
log("Running Topological Clustering of Channels | original plots plus useful diagnostics")
log(f"Input directory: {input_dir}")
log(f"Output directory: {cluster_out_dir}")
log(f"Diagnostics directory: {diagnostic_out_dir}")
log("=" * 72)

feature_z_df = pd.read_csv(feature_z_csv_path, index_col=0)
feature_z_df.index = feature_z_df.index.astype(str)
feature_z_df = feature_z_df.replace([np.inf, -np.inf], np.nan).fillna(0.0)

if os.path.exists(feature_raw_csv_path):
    feature_raw_df = pd.read_csv(feature_raw_csv_path, index_col=0)
    feature_raw_df.index = feature_raw_df.index.astype(str)
    feature_raw_df = feature_raw_df.loc[feature_z_df.index]
else:
    feature_raw_df = feature_z_df.copy()

if os.path.exists(score_csv_path):
    score_df = pd.read_csv(score_csv_path)
else:
    score_df = pd.DataFrame({"channel": feature_z_df.index})

if "channel" not in score_df.columns:
    score_df = score_df.reset_index().rename(columns={"index": "channel"})

score_df["channel"] = score_df["channel"].astype(str)

if os.path.exists(summary_csv_path):
    summary_df = pd.read_csv(summary_csv_path)

    if "channel" in summary_df.columns:
        summary_df["channel"] = summary_df["channel"].astype(str)
else:
    summary_df = pd.DataFrame({"channel": feature_z_df.index})

channels = list(feature_z_df.index)
X = feature_z_df.to_numpy(dtype=float)
X = np.nan_to_num(X, nan=0.0, posinf=0.0, neginf=0.0)

if len(channels) < 3:
    raise RuntimeError("Need at least 3 channels for clustering.")

D = squareform(pdist(X, metric="euclidean"))

if CLUSTER_LINKAGE_METHOD == "ward":
    Z = linkage(X, method="ward", optimal_ordering=True)
else:
    Z = linkage(squareform(D), method=CLUSTER_LINKAGE_METHOD, optimal_ordering=True)

dendro_order = leaves_list(Z)

k_max = min(MAX_CLUSTERS, len(channels) - 1)
k_min = max(MIN_CLUSTERS, 2)

silhouette_rows = []

for k in range(k_min, k_max + 1):
    raw_labels_k = fcluster(Z, t=k, criterion="maxclust")
    actual_k = len(np.unique(raw_labels_k))

    if actual_k < 2:
        sil = np.nan
    else:
        sil = silhouette_score_precomputed(D, raw_labels_k)

    silhouette_rows.append({
        "requested_k": int(k),
        "actual_k": int(actual_k),
        "silhouette_score": sil,
    })

silhouette_df = pd.DataFrame(silhouette_rows)

valid_silhouette_df = silhouette_df[
    np.isfinite(silhouette_df["silhouette_score"])
    & (silhouette_df["actual_k"] >= 2)
].copy()

if len(valid_silhouette_df) > 0:
    best_k = int(valid_silhouette_df.sort_values("silhouette_score", ascending=False).iloc[0]["actual_k"])
else:
    best_k = min(3, len(channels))

raw_cluster_labels = fcluster(Z, t=best_k, criterion="maxclust")
cluster_labels, cluster_label_map = relabel_clusters_by_dendrogram_order(raw_cluster_labels, dendro_order)

cluster_ids = sorted(np.unique(cluster_labels))
n_clusters = len(cluster_ids)

cluster_assignment_df = pd.DataFrame({
    "channel": channels,
    "topological_cluster": cluster_labels,
    "cluster_name": [f"Cluster {x}" for x in cluster_labels],
})

if "region" in score_df.columns:
    cluster_assignment_df = cluster_assignment_df.merge(
        score_df[["channel", "region"]],
        on="channel",
        how="left",
    )
elif "region" in summary_df.columns:
    cluster_assignment_df = cluster_assignment_df.merge(
        summary_df[["channel", "region"]],
        on="channel",
        how="left",
    )
else:
    cluster_assignment_df["region"] = "Unknown"

score_cols = [
    "topology_complexity_score",
    "loop_strength_score",
    "topological_activity_score",
    "topological_spread_score",
    "structure_density_score",
    "topological_uniqueness_score",
]

available_score_cols = [col for col in score_cols if col in score_df.columns]

if len(available_score_cols) > 0:
    cluster_assignment_df = cluster_assignment_df.merge(
        score_df[["channel"] + available_score_cols],
        on="channel",
        how="left",
    )

mds_coords = classical_mds_from_distances(D, n_components=2)
pca_coords, pca_explained, pca_components = pca_from_feature_matrix(X, n_components=2)

cluster_assignment_df["mds_1"] = mds_coords[:, 0]
cluster_assignment_df["mds_2"] = mds_coords[:, 1]
cluster_assignment_df["pc1"] = pca_coords[:, 0]
cluster_assignment_df["pc2"] = pca_coords[:, 1]

mean_distance_to_others = []
mean_distance_to_same_cluster = []
mean_distance_to_other_clusters = []

for i in range(len(channels)):
    mask_other = np.arange(len(channels)) != i
    same_mask = (cluster_labels == cluster_labels[i]) & mask_other
    other_cluster_mask = cluster_labels != cluster_labels[i]

    mean_distance_to_others.append(float(np.mean(D[i, mask_other])) if np.any(mask_other) else 0.0)
    mean_distance_to_same_cluster.append(float(np.mean(D[i, same_mask])) if np.any(same_mask) else 0.0)
    mean_distance_to_other_clusters.append(float(np.mean(D[i, other_cluster_mask])) if np.any(other_cluster_mask) else 0.0)

cluster_assignment_df["mean_distance_to_all_other_channels"] = mean_distance_to_others
cluster_assignment_df["mean_distance_to_same_cluster"] = mean_distance_to_same_cluster
cluster_assignment_df["mean_distance_to_other_clusters"] = mean_distance_to_other_clusters

cluster_summary_rows = []

for cluster_id in cluster_ids:
    cluster_channels = cluster_assignment_df.loc[
        cluster_assignment_df["topological_cluster"] == cluster_id,
        "channel",
    ].tolist()

    idx = [channels.index(ch) for ch in cluster_channels]

    within_vals = D[np.ix_(idx, idx)]
    within_offdiag = offdiag_values(within_vals)

    if len(within_offdiag) > 0:
        mean_within_distance = float(np.mean(within_offdiag))
        median_within_distance = float(np.median(within_offdiag))
    else:
        mean_within_distance = 0.0
        median_within_distance = 0.0

    other_idx = [i for i in range(len(channels)) if i not in idx]

    if len(other_idx) > 0:
        between_vals = D[np.ix_(idx, other_idx)]
        mean_between_distance = float(np.mean(between_vals))
        median_between_distance = float(np.median(between_vals))
    else:
        mean_between_distance = 0.0
        median_between_distance = 0.0

    row = {
        "topological_cluster": int(cluster_id),
        "cluster_name": f"Cluster {cluster_id}",
        "n_channels": int(len(cluster_channels)),
        "channels": ", ".join(cluster_channels),
        "mean_within_cluster_distance": mean_within_distance,
        "median_within_cluster_distance": median_within_distance,
        "mean_between_cluster_distance": mean_between_distance,
        "median_between_cluster_distance": median_between_distance,
        "separation_ratio_between_over_within": (
            mean_between_distance / (mean_within_distance + 1e-12)
            if mean_within_distance > 0 else np.nan
        ),
    }

    if len(available_score_cols) > 0:
        sub_scores = cluster_assignment_df.loc[
            cluster_assignment_df["topological_cluster"] == cluster_id,
            available_score_cols,
        ]

        for col in available_score_cols:
            row[f"mean_{col}"] = float(sub_scores[col].mean())
            row[f"sem_{col}"] = safe_sem(sub_scores[col].to_numpy(dtype=float))

    cluster_summary_rows.append(row)

cluster_summary_df = pd.DataFrame(cluster_summary_rows)

cluster_centroids = []

for cluster_id in cluster_ids:
    idx = np.where(cluster_labels == cluster_id)[0]
    cluster_centroids.append(np.mean(X[idx], axis=0))

cluster_centroids = np.asarray(cluster_centroids, dtype=float)

if len(cluster_centroids) >= 2:
    cluster_centroid_distance_mat = squareform(pdist(cluster_centroids, metric="euclidean"))
else:
    cluster_centroid_distance_mat = np.zeros((len(cluster_centroids), len(cluster_centroids)), dtype=float)

cluster_names = [f"Cluster {cluster_id}" for cluster_id in cluster_ids]

feature_cluster_mean_rows = []

for cluster_id in cluster_ids:
    idx = np.where(cluster_labels == cluster_id)[0]
    mean_vec = np.mean(X[idx], axis=0)

    row = {
        "topological_cluster": int(cluster_id),
        "cluster_name": f"Cluster {cluster_id}",
    }

    for feat_name, val in zip(feature_z_df.columns, mean_vec):
        row[feat_name] = float(val)

    feature_cluster_mean_rows.append(row)

cluster_feature_mean_df = pd.DataFrame(feature_cluster_mean_rows)

feature_mean_matrix = cluster_feature_mean_df.drop(
    columns=["topological_cluster", "cluster_name"]
).to_numpy(dtype=float)

if feature_mean_matrix.shape[0] > 1:
    feature_cluster_variance = np.var(feature_mean_matrix, axis=0)
else:
    feature_cluster_variance = np.zeros(feature_mean_matrix.shape[1], dtype=float)

feature_cluster_variance_df = pd.DataFrame({
    "feature_name": feature_z_df.columns,
    "cluster_mean_variance": feature_cluster_variance,
    "mean_abs_cluster_mean": np.mean(np.abs(feature_mean_matrix), axis=0),
}).sort_values("cluster_mean_variance", ascending=False).reset_index(drop=True)

ordered_channels = [channels[i] for i in dendro_order]
ordered_labels = cluster_labels[dendro_order]
ordered_D = D[np.ix_(dendro_order, dendro_order)]

cluster_assignment_csv_path = os.path.join(cluster_out_dir, "topological_channel_cluster_assignments.csv")
cluster_summary_csv_path = os.path.join(cluster_out_dir, "topological_channel_cluster_summary.csv")
silhouette_csv_path = os.path.join(cluster_out_dir, "topological_clustering_silhouette_by_k.csv")
distance_csv_path = os.path.join(cluster_out_dir, "topological_channel_fingerprint_distance_matrix.csv")
centroid_distance_csv_path = os.path.join(cluster_out_dir, "topological_cluster_centroid_distance_matrix.csv")
cluster_feature_mean_csv_path = os.path.join(cluster_out_dir, "topological_cluster_feature_means_standardized.csv")
cluster_feature_variance_csv_path = os.path.join(cluster_out_dir, "topological_cluster_feature_difference_ranking.csv")
linkage_npz_path = os.path.join(cluster_out_dir, "topological_channel_clustering_linkage_outputs.npz")

cluster_assignment_df.to_csv(cluster_assignment_csv_path, index=False)
cluster_summary_df.to_csv(cluster_summary_csv_path, index=False)
silhouette_df.to_csv(silhouette_csv_path, index=False)
pd.DataFrame(D, index=channels, columns=channels).to_csv(distance_csv_path)
pd.DataFrame(cluster_centroid_distance_mat, index=cluster_names, columns=cluster_names).to_csv(centroid_distance_csv_path)
cluster_feature_mean_df.to_csv(cluster_feature_mean_csv_path, index=False)
feature_cluster_variance_df.to_csv(cluster_feature_variance_csv_path, index=False)

np.savez(
    linkage_npz_path,
    channels=np.asarray(channels),
    linkage_matrix=Z,
    dendrogram_order=dendro_order,
    cluster_labels=cluster_labels,
    distance_matrix=D,
    mds_coordinates=mds_coords,
    pca_coordinates=pca_coords,
    pca_explained_variance_ratio=pca_explained,
    cluster_centroid_distance_matrix=cluster_centroid_distance_mat,
    best_k=np.asarray([best_k]),
    linkage_method=np.asarray([CLUSTER_LINKAGE_METHOD]),
    input_dir=np.asarray([input_dir]),
)

print("\nTopological Channel Cluster Assignments:")
if HAVE_IPYTHON:
    display(cluster_assignment_df.sort_values(["topological_cluster", "channel"]))
else:
    print(cluster_assignment_df.sort_values(["topological_cluster", "channel"]))

print("\nTopological Cluster Summary:")
if HAVE_IPYTHON:
    display(cluster_summary_df)
else:
    print(cluster_summary_df)

print("\nSilhouette Scores by Cluster Count:")
if HAVE_IPYTHON:
    display(silhouette_df)
else:
    print(silhouette_df)

print("\nTop Features Separating Clusters:")
if HAVE_IPYTHON:
    display(feature_cluster_variance_df.head(25))
else:
    print(feature_cluster_variance_df.head(25))


fig, ax = plt.subplots(figsize=(8, 5.6), facecolor=BLACK)
style_ax(
    ax,
    title="Topological Clustering Silhouette Score by Cluster Count",
    xlabel="Number of clusters",
    ylabel="Mean silhouette score",
)

ax.plot(
    silhouette_df["requested_k"],
    silhouette_df["silhouette_score"],
    marker="o",
    linewidth=2.0,
    color=CYAN_FILL,
)

best_silhouette_vals = silhouette_df.loc[
    silhouette_df["actual_k"] == best_k,
    "silhouette_score",
]

if len(best_silhouette_vals) and np.isfinite(best_silhouette_vals.max()):
    best_silhouette_y = float(best_silhouette_vals.max())
else:
    finite_silhouettes = silhouette_df["silhouette_score"].to_numpy(dtype=float)
    finite_silhouettes = finite_silhouettes[np.isfinite(finite_silhouettes)]
    best_silhouette_y = float(np.max(finite_silhouettes)) if len(finite_silhouettes) else 0.0

ax.scatter(
    [best_k],
    [best_silhouette_y],
    s=140,
    color=CYAN_FILL,
    edgecolor=ACCENT,
    linewidth=1.2,
    zorder=5,
    label=f"Selected k={best_k}",
)

legend = ax.legend(loc="best", framealpha=1.0)
style_legend(legend)
force_cyan_text(ax)

silhouette_plot_path = os.path.join(cluster_out_dir, "01_silhouette_score_by_cluster_count.png")
plt.tight_layout()
save_render_close(fig, silhouette_plot_path)


fig, ax = plt.subplots(figsize=(14, 7), facecolor=BLACK)
style_ax(
    ax,
    title=f"Hierarchical Topological Clustering Dendrogram | {CLUSTER_LINKAGE_METHOD.upper()} | k={best_k}",
    xlabel="Channel",
    ylabel="Linkage distance",
    grid=False,
)

dendrogram(
    Z,
    labels=channels,
    leaf_rotation=90,
    leaf_font_size=9,
    ax=ax,
    color_threshold=None,
    above_threshold_color=ACCENT,
)

for tick_label in ax.get_xmajorticklabels():
    tick_label.set_color(ACCENT)

for tick_label in ax.get_ymajorticklabels():
    tick_label.set_color(ACCENT)

for line in ax.get_lines():
    line.set_color(ACCENT)
    line.set_linewidth(1.2)

force_cyan_text(ax)

dendrogram_path = os.path.join(cluster_out_dir, "02_hierarchical_topological_clustering_dendrogram.png")
plt.tight_layout()
save_render_close(fig, dendrogram_path, width=1150)


dist_vmin, dist_vmax = dynamic_limits(
    offdiag_values(ordered_D),
    lower_pct=2,
    upper_pct=98,
    pad_frac=0.10,
    lower_bound=0.0,
)

fig, ax = plt.subplots(figsize=(9, 8), facecolor=BLACK)
style_ax(
    ax,
    title="Topological Fingerprint Distance Matrix Ordered by Clustering",
    xlabel="Channel",
    ylabel="Channel",
)

im = ax.imshow(
    ordered_D,
    aspect="equal",
    interpolation="nearest",
    cmap=CYAN_SEQUENTIAL_CMAP,
    vmin=dist_vmin,
    vmax=dist_vmax,
)

ax.set_xticks(np.arange(len(ordered_channels)))
ax.set_xticklabels(ordered_channels, rotation=90, color=ACCENT)
ax.set_yticks(np.arange(len(ordered_channels)))
ax.set_yticklabels(ordered_channels, color=ACCENT)

boundaries = []
last_label = ordered_labels[0]

for i, lab in enumerate(ordered_labels[1:], start=1):
    if lab != last_label:
        boundaries.append(i - 0.5)
        last_label = lab

for b in boundaries:
    ax.axhline(b, color=ACCENT, linewidth=1.5, alpha=0.9)
    ax.axvline(b, color=ACCENT, linewidth=1.5, alpha=0.9)

cbar = plt.colorbar(im, ax=ax, shrink=0.85)
style_colorbar(cbar, "Fingerprint distance")
force_cyan_text(ax)

distance_heatmap_path = os.path.join(cluster_out_dir, "03_clustered_topological_distance_heatmap.png")
plt.tight_layout()
save_render_close(fig, distance_heatmap_path)


score_plot_cols = [col for col in score_cols if col in cluster_assignment_df.columns]
score_plot_labels = [
    "Complexity",
    "Loop strength",
    "Activity",
    "Spread",
    "Density",
    "Uniqueness",
]
score_plot_labels = [
    label for col, label in zip(score_cols, score_plot_labels)
    if col in score_plot_cols
]

if len(score_plot_cols) > 0:
    ordered_score_df = cluster_assignment_df.set_index("channel").loc[ordered_channels, score_plot_cols]
    score_matrix = ordered_score_df.to_numpy(dtype=float)

    fig, ax = plt.subplots(figsize=(9, 9), facecolor=BLACK)
    style_ax(
        ax,
        title="Topology Score Heatmap Ordered by Channel Clusters",
        xlabel="Topology score",
        ylabel="Channel",
    )

    im = ax.imshow(
        score_matrix,
        aspect="auto",
        interpolation="nearest",
        cmap=CYAN_SEQUENTIAL_CMAP,
        vmin=0,
        vmax=100,
    )

    ax.set_yticks(np.arange(len(ordered_channels)))
    ax.set_yticklabels(ordered_channels, color=ACCENT)
    ax.set_xticks(np.arange(len(score_plot_cols)))
    ax.set_xticklabels(score_plot_labels, rotation=45, ha="right", color=ACCENT)

    for b in boundaries:
        ax.axhline(b, color=ACCENT, linewidth=1.5, alpha=0.9)

    cbar = plt.colorbar(im, ax=ax, shrink=0.85)
    style_colorbar(cbar, "Score percentile")
    force_cyan_text(ax)

    score_heatmap_path = os.path.join(cluster_out_dir, "04_clustered_topology_score_heatmap.png")
    plt.tight_layout()
    save_render_close(fig, score_heatmap_path)


top_features = feature_cluster_variance_df.head(
    min(TOP_FEATURES_TO_PLOT, len(feature_cluster_variance_df))
)["feature_name"].tolist()

feature_heatmap_matrix = feature_z_df.loc[ordered_channels, top_features].to_numpy(dtype=float)

fig, ax = plt.subplots(figsize=(14, 9), facecolor=BLACK)
style_ax(
    ax,
    title=f"Top {len(top_features)} Cluster-Separating Fingerprint Features",
    xlabel="Fingerprint feature",
    ylabel="Channel",
)

vlim = np.nanpercentile(np.abs(feature_heatmap_matrix), 98) if np.any(np.isfinite(feature_heatmap_matrix)) else 1.0
vlim = max(float(vlim), 1e-6)

im = ax.imshow(
    feature_heatmap_matrix,
    aspect="auto",
    interpolation="nearest",
    cmap=CYAN_DIVERGING_CMAP,
    vmin=-vlim,
    vmax=vlim,
)

ax.set_yticks(np.arange(len(ordered_channels)))
ax.set_yticklabels(ordered_channels, color=ACCENT)
ax.set_xticks(np.arange(len(top_features)))
ax.set_xticklabels(top_features, rotation=90, fontsize=7, color=ACCENT)

for b in boundaries:
    ax.axhline(b, color=ACCENT, linewidth=1.5, alpha=0.9)

cbar = plt.colorbar(im, ax=ax, shrink=0.85)
style_colorbar(cbar, "Standardized feature value")
force_cyan_text(ax)

feature_heatmap_path = os.path.join(cluster_out_dir, "05_cluster_separating_fingerprint_feature_heatmap.png")
plt.tight_layout()
save_render_close(fig, feature_heatmap_path, width=1200)


fig, ax = plt.subplots(figsize=(8, 7), facecolor=BLACK)
style_ax(
    ax,
    title="MDS of Topological Fingerprint Distances by Cluster",
    xlabel="MDS 1",
    ylabel="MDS 2",
)

for cluster_idx, cluster_id in enumerate(cluster_ids):
    mask = cluster_labels == cluster_id
    marker = MARKERS[cluster_idx % len(MARKERS)]

    ax.scatter(
        mds_coords[mask, 0],
        mds_coords[mask, 1],
        s=150,
        color=CYAN_FILL,
        edgecolor=ACCENT,
        marker=marker,
        linewidth=1.1,
        alpha=0.88,
        label=f"Cluster {cluster_id}",
    )

for i, ch in enumerate(channels):
    ax.text(
        mds_coords[i, 0],
        mds_coords[i, 1],
        ch,
        fontsize=8,
        color=ACCENT,
        alpha=0.9,
    )

legend = ax.legend(loc="best", framealpha=1.0)
style_legend(legend)
force_cyan_text(ax)

mds_path = os.path.join(cluster_out_dir, "06_mds_topological_clusters.png")
plt.tight_layout()
save_render_close(fig, mds_path)


fig, ax = plt.subplots(figsize=(8, 7), facecolor=BLACK)
style_ax(
    ax,
    title=f"PCA of Topological Fingerprints by Cluster | PC1={pca_explained[0]*100:.1f}%, PC2={pca_explained[1]*100:.1f}%",
    xlabel="PC1",
    ylabel="PC2",
)

for cluster_idx, cluster_id in enumerate(cluster_ids):
    mask = cluster_labels == cluster_id
    marker = MARKERS[cluster_idx % len(MARKERS)]

    ax.scatter(
        pca_coords[mask, 0],
        pca_coords[mask, 1],
        s=150,
        color=CYAN_FILL,
        edgecolor=ACCENT,
        marker=marker,
        linewidth=1.1,
        alpha=0.88,
        label=f"Cluster {cluster_id}",
    )

for i, ch in enumerate(channels):
    ax.text(
        pca_coords[i, 0],
        pca_coords[i, 1],
        ch,
        fontsize=8,
        color=ACCENT,
        alpha=0.9,
    )

legend = ax.legend(loc="best", framealpha=1.0)
style_legend(legend)
force_cyan_text(ax)

pca_path = os.path.join(cluster_out_dir, "07_pca_topological_clusters.png")
plt.tight_layout()
save_render_close(fig, pca_path)


if len(score_plot_cols) > 0:
    cluster_score_mean_rows = []

    for cluster_id in cluster_ids:
        sub = cluster_assignment_df[cluster_assignment_df["topological_cluster"] == cluster_id]

        row = {"topological_cluster": cluster_id}

        for col in score_plot_cols:
            row[col] = float(sub[col].mean())

        cluster_score_mean_rows.append(row)

    cluster_score_mean_df = pd.DataFrame(cluster_score_mean_rows)

    fig, ax = plt.subplots(figsize=(12, 6), facecolor=BLACK)
    style_ax(
        ax,
        title="Mean Topology Score Profiles by Channel Cluster",
        xlabel="Topological cluster",
        ylabel="Mean score",
    )

    x = np.arange(len(cluster_score_mean_df))
    width = 0.12

    for idx, col in enumerate(score_plot_cols):
        ax.bar(
            x + (idx - (len(score_plot_cols) - 1) / 2.0) * width,
            cluster_score_mean_df[col],
            width=width,
            color=CYAN_FILL,
            edgecolor=ACCENT,
            linewidth=0.9,
            alpha=0.82,
            label=score_plot_labels[idx],
        )

    ax.set_xticks(x)
    ax.set_xticklabels([f"Cluster {int(c)}" for c in cluster_score_mean_df["topological_cluster"]], color=ACCENT)
    ax.set_ylim(0, 105)

    legend = ax.legend(
        loc="upper left",
        framealpha=1.0,
        fontsize=8,
        facecolor=BLACK,
        edgecolor=ACCENT,
    )
    style_legend(legend)
    force_cyan_text(ax)

    cluster_score_profile_path = os.path.join(cluster_out_dir, "08_cluster_mean_topology_score_profiles.png")
    plt.tight_layout()
    save_render_close(fig, cluster_score_profile_path)


fig, ax = plt.subplots(figsize=(9, 6), facecolor=BLACK)
style_ax(
    ax,
    title="Topological Cluster Sizes",
    xlabel="Cluster",
    ylabel="Number of channels",
)

cluster_size_df = cluster_assignment_df.groupby("topological_cluster").size().reset_index(name="n_channels")
x = np.arange(len(cluster_size_df))

ax.bar(
    x,
    cluster_size_df["n_channels"],
    color=CYAN_FILL,
    edgecolor=ACCENT,
    alpha=0.88,
)

ax.set_xticks(x)
ax.set_xticklabels([f"Cluster {int(c)}" for c in cluster_size_df["topological_cluster"]], color=ACCENT)

for i, row in cluster_size_df.iterrows():
    ax.text(
        i,
        row["n_channels"],
        str(int(row["n_channels"])),
        ha="center",
        va="bottom",
        color=ACCENT,
        fontsize=10,
    )

force_cyan_text(ax)

cluster_size_path = os.path.join(cluster_out_dir, "09_topological_cluster_sizes.png")
plt.tight_layout()
save_render_close(fig, cluster_size_path)


if "region" in cluster_assignment_df.columns:
    region_cluster_counts = (
        cluster_assignment_df
        .groupby(["region", "topological_cluster"])
        .size()
        .reset_index(name="count")
    )

    region_order = region_cluster_counts.groupby("region")["count"].sum().sort_values(ascending=False).index.tolist()

    fig, ax = plt.subplots(figsize=(12, 6), facecolor=BLACK)
    style_ax(
        ax,
        title="Region Composition of Topological Channel Clusters",
        xlabel="Region",
        ylabel="Number of channels",
    )

    bottom = np.zeros(len(region_order), dtype=float)

    for cluster_id in cluster_ids:
        vals = []

        for region in region_order:
            count = region_cluster_counts.loc[
                (region_cluster_counts["region"] == region)
                & (region_cluster_counts["topological_cluster"] == cluster_id),
                "count",
            ]

            vals.append(int(count.iloc[0]) if len(count) else 0)

        ax.bar(
            np.arange(len(region_order)),
            vals,
            bottom=bottom,
            color=CYAN_FILL,
            edgecolor=ACCENT,
            alpha=0.85,
            label=f"Cluster {cluster_id}",
        )

        bottom += np.asarray(vals, dtype=float)

    ax.set_xticks(np.arange(len(region_order)))
    ax.set_xticklabels(region_order, rotation=35, ha="right", color=ACCENT)

    legend = ax.legend(loc="best", framealpha=1.0)
    style_legend(legend)
    force_cyan_text(ax)

    region_composition_path = os.path.join(cluster_out_dir, "10_region_composition_by_topological_cluster.png")
    plt.tight_layout()
    save_render_close(fig, region_composition_path)


scalp_xy = {
    "Fp1": (-0.45, 1.00), "Fpz": (0.00, 1.08), "Fp2": (0.45, 1.00),
    "F7": (-0.95, 0.60), "F3": (-0.45, 0.55), "Fz": (0.00, 0.60),
    "F4": (0.45, 0.55), "F8": (0.95, 0.60),
    "FC5": (-0.75, 0.25), "FC1": (-0.25, 0.25),
    "FC2": (0.25, 0.25), "FC6": (0.75, 0.25),
    "M1": (-1.15, 0.05), "T7": (-1.00, 0.00), "C3": (-0.50, 0.00),
    "Cz": (0.00, 0.00), "C4": (0.50, 0.00), "T8": (1.00, 0.00),
    "M2": (1.15, 0.05),
    "CP5": (-0.75, -0.30), "CP1": (-0.25, -0.30),
    "CP2": (0.25, -0.30), "CP6": (0.75, -0.30),
    "P7": (-0.90, -0.60), "P3": (-0.45, -0.60), "Pz": (0.00, -0.65),
    "P4": (0.45, -0.60), "P8": (0.90, -0.60),
    "POz": (0.00, -0.88),
    "O1": (-0.35, -1.00), "Oz": (0.00, -1.08), "O2": (0.35, -1.00),
}

fig, ax = plt.subplots(figsize=(7, 7), facecolor=BLACK)
style_ax(
    ax,
    title=f"Scalp Map of Topological Channel Clusters | k={best_k}",
    grid=False,
)

head = plt.Circle((0, 0), 1.12, fill=False, color=ACCENT, linewidth=2)
ax.add_patch(head)
ax.plot([0.00, -0.10, 0.10, 0.00], [1.12, 1.28, 1.28, 1.12], color=ACCENT, linewidth=2)

for _, row in cluster_assignment_df.iterrows():
    ch = row["channel"]

    if ch not in scalp_xy:
        continue

    x0, y0 = scalp_xy[ch]
    cluster_id = int(row["topological_cluster"])
    cluster_idx = cluster_ids.index(cluster_id)
    marker = MARKERS[cluster_idx % len(MARKERS)]

    ax.scatter(
        [x0],
        [y0],
        s=285,
        color=CYAN_FILL,
        edgecolor=ACCENT,
        linewidth=1.3,
        marker=marker,
        alpha=0.92,
    )

    txt = ax.text(
        x0,
        y0,
        ch,
        ha="center",
        va="center",
        fontsize=8,
        color=ACCENT,
        fontweight="bold",
    )
    txt.set_path_effects([pe.withStroke(linewidth=2.0, foreground=BLACK)])

handles = []

for cluster_idx, cluster_id in enumerate(cluster_ids):
    marker = MARKERS[cluster_idx % len(MARKERS)]
    handles.append(
        plt.Line2D(
            [0],
            [0],
            marker=marker,
            color="none",
            markerfacecolor=CYAN_FILL,
            markeredgecolor=ACCENT,
            markersize=9,
            label=f"Cluster {cluster_id}",
        )
    )

legend = ax.legend(handles=handles, loc="upper right", framealpha=1.0, facecolor=BLACK, edgecolor=ACCENT)
style_legend(legend)

ax.set_xlim(-1.35, 1.35)
ax.set_ylim(-1.35, 1.35)
ax.set_aspect("equal")
ax.set_xticks([])
ax.set_yticks([])
force_cyan_text(ax)

scalp_cluster_path = os.path.join(cluster_out_dir, "11_scalp_map_topological_channel_clusters.png")
plt.tight_layout()
save_render_close(fig, scalp_cluster_path, width=760)


fig, ax = plt.subplots(figsize=(8, 6.5), facecolor=BLACK)
style_ax(
    ax,
    title="Topological Cluster Centroid Distance Matrix",
    xlabel="Cluster",
    ylabel="Cluster",
)

cent_vmax = max(float(np.nanmax(cluster_centroid_distance_mat)), 1e-6)

im = ax.imshow(
    cluster_centroid_distance_mat,
    aspect="equal",
    interpolation="nearest",
    cmap=CYAN_SEQUENTIAL_CMAP,
    vmin=0.0,
    vmax=cent_vmax,
)

ax.set_xticks(np.arange(len(cluster_names)))
ax.set_xticklabels(cluster_names, rotation=45, ha="right", color=ACCENT)
ax.set_yticks(np.arange(len(cluster_names)))
ax.set_yticklabels(cluster_names, color=ACCENT)

for i in range(len(cluster_names)):
    for j in range(len(cluster_names)):
        val = cluster_centroid_distance_mat[i, j]
        ax.text(
            j,
            i,
            f"{val:.2f}",
            ha="center",
            va="center",
            color=ACCENT if val == 0 else BLACK,
            fontsize=11,
            fontweight="bold",
        )

cbar = plt.colorbar(im, ax=ax, shrink=0.85)
style_colorbar(cbar, "Centroid distance")
force_cyan_text(ax)

cluster_centroid_path = os.path.join(cluster_out_dir, "12_cluster_centroid_distance_matrix.png")
plt.tight_layout()
save_render_close(fig, cluster_centroid_path)


if len(cluster_names) >= 2:
    centroid_pair_rows = []

    for i in range(len(cluster_names)):
        for j in range(i + 1, len(cluster_names)):
            centroid_pair_rows.append({
                "cluster_pair": f"{cluster_names[i]} vs {cluster_names[j]}",
                "centroid_distance": float(cluster_centroid_distance_mat[i, j]),
            })

    centroid_pair_df = pd.DataFrame(centroid_pair_rows).sort_values("centroid_distance", ascending=False)
    centroid_pair_csv_path = os.path.join(diagnostic_out_dir, "cluster_centroid_pair_distances.csv")
    centroid_pair_df.to_csv(centroid_pair_csv_path, index=False)

    fig, ax = plt.subplots(figsize=(10, 5.8), facecolor=BLACK)
    style_ax(
        ax,
        title="Cluster Centroid Pair Distances",
        xlabel="Centroid distance",
        ylabel="Cluster pair",
    )

    y = np.arange(len(centroid_pair_df))[::-1]

    ax.barh(
        y,
        centroid_pair_df["centroid_distance"],
        color=CYAN_FILL,
        edgecolor=ACCENT,
        alpha=0.82,
    )

    ax.set_yticks(y)
    ax.set_yticklabels(centroid_pair_df["cluster_pair"])

    for yy, val in zip(y, centroid_pair_df["centroid_distance"]):
        ax.text(
            val,
            yy,
            f"  {val:.4f}",
            va="center",
            ha="left",
            color=ACCENT,
            fontsize=10,
        )

    max_val = float(centroid_pair_df["centroid_distance"].max()) if len(centroid_pair_df) else 1.0
    ax.set_xlim(0, max_val * 1.18 if max_val > 0 else 1.0)

    force_cyan_text(ax)

    centroid_bar_path = os.path.join(diagnostic_out_dir, "13_cluster_centroid_pair_distances_bar.png")
    plt.tight_layout()
    save_render_close(fig, centroid_bar_path)


fig, axes = plt.subplots(
    1,
    n_clusters,
    figsize=(6.2 * n_clusters, 6.4),
    facecolor=BLACK,
    squeeze=False,
)

axes = axes.ravel()

for ax_idx, cluster_id in enumerate(cluster_ids):
    ax = axes[ax_idx]
    style_ax(
        ax,
        title=f"Cluster {cluster_id} membership",
        grid=False,
    )

    head = plt.Circle((0, 0), 1.12, fill=False, color=ACCENT, linewidth=2)
    ax.add_patch(head)
    ax.plot([0.00, -0.10, 0.10, 0.00], [1.12, 1.28, 1.28, 1.12], color=ACCENT, linewidth=2)

    cluster_channels = cluster_assignment_df.loc[
        cluster_assignment_df["topological_cluster"] == cluster_id,
        "channel",
    ].tolist()

    for _, row in cluster_assignment_df.iterrows():
        ch = row["channel"]

        if ch not in scalp_xy:
            continue

        x0, y0 = scalp_xy[ch]
        is_member = int(row["topological_cluster"]) == cluster_id

        if is_member:
            ax.scatter(
                [x0],
                [y0],
                s=310,
                color=CYAN_FILL,
                edgecolor=ACCENT,
                linewidth=1.6,
                alpha=0.95,
                zorder=4,
            )
        else:
            ax.scatter(
                [x0],
                [y0],
                s=210,
                facecolors=BLACK,
                edgecolors=ACCENT_DIM,
                linewidth=1.0,
                alpha=0.45,
                zorder=2,
            )

        label_color = ACCENT if is_member else ACCENT_DIM
        label_alpha = 1.0 if is_member else 0.55

        txt = ax.text(
            x0,
            y0,
            ch,
            ha="center",
            va="center",
            fontsize=8,
            color=label_color,
            fontweight="bold",
            alpha=label_alpha,
            zorder=5,
        )
        txt.set_path_effects([pe.withStroke(linewidth=2.0, foreground=BLACK)])

    cluster_text = ", ".join(cluster_channels)

    ax.text(
        0,
        -1.32,
        cluster_text,
        ha="center",
        va="top",
        fontsize=9,
        color=ACCENT,
        wrap=True,
    )

    ax.set_xlim(-1.38, 1.38)
    ax.set_ylim(-1.45, 1.35)
    ax.set_aspect("equal")
    ax.set_xticks([])
    ax.set_yticks([])
    force_cyan_text(ax)

split_scalp_path = os.path.join(diagnostic_out_dir, "14_split_scalp_cluster_membership_maps.png")
plt.tight_layout()
save_render_close(fig, split_scalp_path, width=1250)


pair_rows = []

for i in range(len(channels)):
    for j in range(i + 1, len(channels)):
        ch_a = channels[i]
        ch_b = channels[j]
        cl_a = int(cluster_labels[i])
        cl_b = int(cluster_labels[j])
        dist = float(D[i, j])

        if cl_a == cl_b:
            pair_type = f"Within Cluster {cl_a}"
        else:
            pair_type = "Between Clusters"

        pair_rows.append({
            "channel_a": ch_a,
            "channel_b": ch_b,
            "cluster_a": cl_a,
            "cluster_b": cl_b,
            "pair_type": pair_type,
            "distance": dist,
        })

pair_df = pd.DataFrame(pair_rows)

pair_distance_csv_path = os.path.join(diagnostic_out_dir, "pairwise_within_between_cluster_distances.csv")
pair_df.to_csv(pair_distance_csv_path, index=False)

pair_type_order = []

for cluster_id in cluster_ids:
    label = f"Within Cluster {cluster_id}"

    if label in pair_df["pair_type"].unique():
        pair_type_order.append(label)

if "Between Clusters" in pair_df["pair_type"].unique():
    pair_type_order.append("Between Clusters")

distance_groups = [
    pair_df.loc[pair_df["pair_type"] == pair_type, "distance"].to_numpy(dtype=float)
    for pair_type in pair_type_order
]

fig, ax = plt.subplots(figsize=(12, 6.2), facecolor=BLACK)
style_ax(
    ax,
    title="Topology Distance Distributions: Within vs Between Clusters",
    xlabel="Pair type",
    ylabel="Topological fingerprint distance",
)

box = ax.boxplot(
    distance_groups,
    patch_artist=True,
    labels=pair_type_order,
    showfliers=False,
    widths=0.58,
)

for patch in box["boxes"]:
    patch.set_facecolor(CYAN_FILL)
    patch.set_edgecolor(ACCENT)
    patch.set_alpha(0.34)
    patch.set_linewidth(1.4)

for key in ["whiskers", "caps", "medians"]:
    for item in box[key]:
        item.set_color(ACCENT)
        item.set_linewidth(1.4)

rng = np.random.default_rng(7)

for idx, vals in enumerate(distance_groups, start=1):
    if len(vals) == 0:
        continue

    jitter = rng.uniform(-0.12, 0.12, size=len(vals))

    ax.scatter(
        np.full(len(vals), idx) + jitter,
        vals,
        s=28,
        facecolors=BLACK,
        edgecolors=CYAN_FILL,
        linewidths=1.0,
        alpha=0.78,
        zorder=4,
    )

    ax.text(
        idx,
        np.nanmax(vals),
        f"mean={np.nanmean(vals):.3f}\nn={len(vals)}",
        ha="center",
        va="bottom",
        color=ACCENT,
        fontsize=9,
    )

ax.set_xticklabels(pair_type_order, rotation=20, ha="right")

within_vals_all = pair_df.loc[pair_df["pair_type"].str.startswith("Within"), "distance"].to_numpy(dtype=float)
between_vals_all = pair_df.loc[pair_df["pair_type"] == "Between Clusters", "distance"].to_numpy(dtype=float)

if len(within_vals_all) > 0 and len(between_vals_all) > 0:
    separation_ratio = float(np.nanmean(between_vals_all) / (np.nanmean(within_vals_all) + 1e-12))
    ax.text(
        0.02,
        0.98,
        f"Separation ratio = mean between / mean within = {separation_ratio:.3f}",
        transform=ax.transAxes,
        ha="left",
        va="top",
        color=ACCENT,
        fontsize=10,
        bbox=dict(facecolor=BLACK, edgecolor=ACCENT, alpha=0.85),
    )

force_cyan_text(ax)

within_between_path = os.path.join(diagnostic_out_dir, "15_within_vs_between_cluster_distance_distributions.png")
plt.tight_layout()
save_render_close(fig, within_between_path, width=1100)


summary_rows = []

for cluster_id in cluster_ids:
    within = pair_df.loc[pair_df["pair_type"] == f"Within Cluster {cluster_id}", "distance"].to_numpy(dtype=float)
    cluster_channels = cluster_assignment_df.loc[
        cluster_assignment_df["topological_cluster"] == cluster_id,
        "channel",
    ].tolist()

    summary_rows.append({
        "metric": f"Within Cluster {cluster_id}",
        "mean_distance": float(np.nanmean(within)) if len(within) else 0.0,
        "median_distance": float(np.nanmedian(within)) if len(within) else 0.0,
        "n_pairs": int(len(within)),
        "channels": ", ".join(cluster_channels),
    })

summary_rows.append({
    "metric": "Between Clusters",
    "mean_distance": float(np.nanmean(between_vals_all)) if len(between_vals_all) else 0.0,
    "median_distance": float(np.nanmedian(between_vals_all)) if len(between_vals_all) else 0.0,
    "n_pairs": int(len(between_vals_all)),
    "channels": "all cross-cluster pairs",
})

distance_summary_df = pd.DataFrame(summary_rows)
distance_summary_csv_path = os.path.join(diagnostic_out_dir, "within_between_distance_summary.csv")
distance_summary_df.to_csv(distance_summary_csv_path, index=False)

fig, ax = plt.subplots(figsize=(11, 5.8), facecolor=BLACK)
style_ax(
    ax,
    title="Mean Topological Distance: Within-Cluster vs Between-Cluster",
    xlabel="Distance category",
    ylabel="Mean topological fingerprint distance",
)

x = np.arange(len(distance_summary_df))

ax.bar(
    x,
    distance_summary_df["mean_distance"],
    color=CYAN_FILL,
    edgecolor=ACCENT,
    linewidth=1.1,
    alpha=0.82,
)

for i, row in distance_summary_df.iterrows():
    ax.text(
        i,
        row["mean_distance"],
        f"{row['mean_distance']:.3f}\nn={int(row['n_pairs'])}",
        ha="center",
        va="bottom",
        color=ACCENT,
        fontsize=9,
    )

ax.set_xticks(x)
ax.set_xticklabels(distance_summary_df["metric"], rotation=20, ha="right")
force_cyan_text(ax)

distance_summary_path = os.path.join(diagnostic_out_dir, "16_mean_within_between_distance_summary.png")
plt.tight_layout()
save_render_close(fig, distance_summary_path, width=1050)


if n_clusters == 2 and not cluster_feature_mean_df.empty:
    feature_cols = [
        c for c in cluster_feature_mean_df.columns
        if c not in ["topological_cluster", "cluster_name"]
    ]

    c1 = cluster_feature_mean_df.loc[
        cluster_feature_mean_df["topological_cluster"] == cluster_ids[0],
        feature_cols,
    ].iloc[0]

    c2 = cluster_feature_mean_df.loc[
        cluster_feature_mean_df["topological_cluster"] == cluster_ids[1],
        feature_cols,
    ].iloc[0]

    contrast = c2.to_numpy(dtype=float) - c1.to_numpy(dtype=float)

    contrast_df = pd.DataFrame({
        "feature_name": feature_cols,
        "cluster_2_minus_cluster_1": contrast,
        "abs_difference": np.abs(contrast),
    }).sort_values("abs_difference", ascending=False).reset_index(drop=True)

    contrast_csv_path = os.path.join(diagnostic_out_dir, "top_cluster_feature_contrast_cluster2_minus_cluster1.csv")
    contrast_df.to_csv(contrast_csv_path, index=False)

    top_n = min(25, len(contrast_df))
    plot_df = contrast_df.head(top_n).iloc[::-1].copy()

    fig, ax = plt.subplots(figsize=(12, 8), facecolor=BLACK)
    style_ax(
        ax,
        title="Top Topological Features Separating Cluster 2 from Cluster 1",
        xlabel="Standardized cluster-mean difference: Cluster 2 minus Cluster 1",
        ylabel="Feature",
    )

    y = np.arange(len(plot_df))

    ax.barh(
        y,
        plot_df["cluster_2_minus_cluster_1"],
        color=CYAN_FILL,
        edgecolor=ACCENT,
        linewidth=1.0,
        alpha=0.82,
    )

    ax.axvline(0, color=ACCENT_SOFT, linestyle="--", linewidth=1.2)

    ax.set_yticks(y)
    ax.set_yticklabels(plot_df["feature_name"])

    for i, val in enumerate(plot_df["cluster_2_minus_cluster_1"]):
        ha = "left" if val >= 0 else "right"
        dx = 0.03 if val >= 0 else -0.03

        ax.text(
            val + dx,
            i,
            f"{val:+.2f}",
            va="center",
            ha=ha,
            color=ACCENT,
            fontsize=8,
        )

    ax.text(
        0.02,
        0.02,
        f"Positive = higher in Cluster {cluster_ids[1]}\nNegative = higher in Cluster {cluster_ids[0]}",
        transform=ax.transAxes,
        ha="left",
        va="bottom",
        color=ACCENT,
        fontsize=9,
        bbox=dict(facecolor=BLACK, edgecolor=ACCENT, alpha=0.85),
    )

    force_cyan_text(ax)

    feature_contrast_path = os.path.join(diagnostic_out_dir, "17_top_feature_contrast_cluster2_minus_cluster1.png")
    plt.tight_layout()
    save_render_close(fig, feature_contrast_path, width=1150)

elif not cluster_feature_mean_df.empty:
    feature_cols = [
        c for c in cluster_feature_mean_df.columns
        if c not in ["topological_cluster", "cluster_name"]
    ]

    if "feature_name" in feature_cluster_variance_df.columns:
        top_profile_features = feature_cluster_variance_df.head(min(30, len(feature_cluster_variance_df)))["feature_name"].tolist()
        top_profile_features = [f for f in top_profile_features if f in feature_cols]
    else:
        cluster_feature_values = cluster_feature_mean_df[feature_cols].to_numpy(dtype=float)
        feature_var = np.var(cluster_feature_values, axis=0)
        top_idx = np.argsort(feature_var)[::-1][:30]
        top_profile_features = [feature_cols[i] for i in top_idx]

    heat_df = cluster_feature_mean_df.set_index("cluster_name")[top_profile_features]

    fig, ax = plt.subplots(figsize=(14, 5.8), facecolor=BLACK)
    style_ax(
        ax,
        title="Cluster Mean Topological Feature Profiles",
        xlabel="Feature",
        ylabel="Cluster",
    )

    vmax = max(float(np.nanpercentile(np.abs(heat_df.to_numpy(dtype=float)), 98)), 1e-6)

    im = ax.imshow(
        heat_df.to_numpy(dtype=float),
        aspect="auto",
        interpolation="nearest",
        cmap=CYAN_DIVERGING_CMAP,
        vmin=-vmax,
        vmax=vmax,
    )

    ax.set_xticks(np.arange(len(top_profile_features)))
    ax.set_xticklabels(top_profile_features, rotation=90, fontsize=7)
    ax.set_yticks(np.arange(len(heat_df.index)))
    ax.set_yticklabels(heat_df.index)

    cbar = plt.colorbar(im, ax=ax, shrink=0.85)
    style_colorbar(cbar, "Standardized cluster mean")
    force_cyan_text(ax)

    feature_profile_path = os.path.join(diagnostic_out_dir, "17_cluster_mean_topological_feature_profiles.png")
    plt.tight_layout()
    save_render_close(fig, feature_profile_path, width=1200)


if len(score_plot_cols) > 0:
    cluster_score_rows = []

    for cluster_id in cluster_ids:
        sub = cluster_assignment_df[cluster_assignment_df["topological_cluster"] == cluster_id]

        row = {"cluster": f"Cluster {cluster_id}"}

        for col in score_plot_cols:
            row[col] = float(sub[col].mean())

        cluster_score_rows.append(row)

    cluster_score_df = pd.DataFrame(cluster_score_rows)

    fig, ax = plt.subplots(figsize=(12, 6), facecolor=BLACK)
    style_ax(
        ax,
        title="Mean Topology Score Profiles by Cluster",
        xlabel="Cluster",
        ylabel="Mean score",
    )

    x = np.arange(len(cluster_score_df))
    width = 0.12

    for idx, col in enumerate(score_plot_cols):
        ax.bar(
            x + (idx - (len(score_plot_cols) - 1) / 2.0) * width,
            cluster_score_df[col],
            width=width,
            color=CYAN_FILL,
            edgecolor=ACCENT,
            linewidth=0.9,
            alpha=0.82,
            label=score_plot_labels[idx],
        )

    ax.set_xticks(x)
    ax.set_xticklabels(cluster_score_df["cluster"])
    ax.set_ylim(0, 105)

    legend = ax.legend(
        loc="upper left",
        framealpha=1.0,
        fontsize=8,
        facecolor=BLACK,
        edgecolor=ACCENT,
    )
    style_legend(legend)
    force_cyan_text(ax)

    diagnostic_cluster_score_path = os.path.join(diagnostic_out_dir, "18_mean_topology_score_profiles_by_cluster.png")
    plt.tight_layout()
    save_render_close(fig, diagnostic_cluster_score_path, width=1050)


fig, ax = plt.subplots(figsize=(10, 5.5), facecolor=BLACK)
style_ax(
    ax,
    title="Cluster Membership Counts with Channel Lists",
    xlabel="Cluster",
    ylabel="Number of channels",
)

cluster_size_df = (
    cluster_assignment_df
    .groupby("topological_cluster")
    .size()
    .reset_index(name="n_channels")
)

x = np.arange(len(cluster_size_df))

ax.bar(
    x,
    cluster_size_df["n_channels"],
    color=CYAN_FILL,
    edgecolor=ACCENT,
    linewidth=1.1,
    alpha=0.82,
)

for i, row in cluster_size_df.iterrows():
    cluster_id = int(row["topological_cluster"])
    cluster_channels = cluster_assignment_df.loc[
        cluster_assignment_df["topological_cluster"] == cluster_id,
        "channel",
    ].tolist()

    ax.text(
        i,
        row["n_channels"],
        f"{int(row['n_channels'])}",
        ha="center",
        va="bottom",
        color=ACCENT,
        fontsize=11,
    )

    ax.text(
        i,
        -0.12,
        "\n".join(cluster_channels),
        transform=ax.get_xaxis_transform(),
        ha="center",
        va="top",
        color=ACCENT,
        fontsize=8,
    )

ax.set_xticks(x)
ax.set_xticklabels([f"Cluster {int(c)}" for c in cluster_size_df["topological_cluster"]])
ax.set_ylim(0, max(cluster_size_df["n_channels"]) * 1.25)
force_cyan_text(ax)

membership_count_path = os.path.join(diagnostic_out_dir, "19_cluster_membership_counts_with_channel_lists.png")
plt.tight_layout()
save_render_close(fig, membership_count_path, width=1050)


log("\n" + "=" * 72)
log("Finished Topological Clustering of Channels")
log(f"Input directory: {input_dir}")
log(f"Output directory: {cluster_out_dir}")
log(f"Useful diagnostics directory: {diagnostic_out_dir}")
log(f"Selected cluster count: k={best_k}")
log(f"Linkage method: {CLUSTER_LINKAGE_METHOD}")
log(f"Assignments CSV: {cluster_assignment_csv_path}")
log(f"Cluster summary CSV: {cluster_summary_csv_path}")
log(f"Silhouette CSV: {silhouette_csv_path}")
log(f"Distance matrix CSV: {distance_csv_path}")
log(f"Centroid distance CSV: {centroid_distance_csv_path}")
log(f"Cluster feature means CSV: {cluster_feature_mean_csv_path}")
log(f"Cluster feature difference ranking CSV: {cluster_feature_variance_csv_path}")
log(f"Linkage NPZ: {linkage_npz_path}")
log(f"Pairwise diagnostic CSV: {pair_distance_csv_path}")
log(f"Within-between summary CSV: {distance_summary_csv_path}")
log("=" * 72)
```

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patheffects as pe
from matplotlib.colors import LinearSegmentedColormap
from matplotlib.gridspec import GridSpec
from scipy.spatial.distance import pdist, squareform
from scipy.cluster.hierarchy import linkage, leaves_list

try:
    from IPython.display import display, Image
    HAVE_IPYTHON = True
except Exception:
    HAVE_IPYTHON = False


base_dir = "/home/a/projects/Complete-Neural-Signal-Analysis"
FAST_OPTION = "ultra"

SAVE_PLOTS_TO_DISK = True
RENDER_ALL_IMAGES_INLINE = True
INLINE_IMAGE_WIDTH = 1050
PLOT_DPI = 150
TOP_FEATURES_TO_PLOT = 28

BLACK = "#000000"
ACCENT = "#00FFFF"
ACCENT_SOFT = "#66FFFF"
ACCENT_DIM = "#008B8B"

CYAN_FILL = "#00D5FF"
CYAN_FILL_SOFT = "#38BDF8"
CYAN_FILL_DARK = "#005A63"
CYAN_BOUNDARY = "#00FFFF"

MARKERS = ["o", "s", "^", "D", "P", "X", "v", "*", "<", ">"]

CYAN_SEQUENTIAL_CMAP = LinearSegmentedColormap.from_list(
    "cyan_sequential",
    [BLACK, CYAN_FILL_DARK, CYAN_FILL, ACCENT_SOFT],
    N=256,
)

CYAN_DIVERGING_CMAP = LinearSegmentedColormap.from_list(
    "cyan_diverging",
    [ACCENT_SOFT, BLACK, CYAN_FILL],
    N=256,
)

plt.rcParams.update({
    "figure.facecolor": BLACK,
    "axes.facecolor": BLACK,
    "savefig.facecolor": BLACK,
    "text.color": ACCENT,
    "axes.labelcolor": ACCENT,
    "axes.edgecolor": ACCENT,
    "axes.titlecolor": ACCENT,
    "xtick.color": ACCENT,
    "ytick.color": ACCENT,
    "grid.color": ACCENT_DIM,
    "grid.alpha": 0.25,
    "legend.facecolor": BLACK,
    "legend.edgecolor": ACCENT,
    "legend.labelcolor": ACCENT,
    "font.size": 11,
})


def log(msg):
    print(msg, flush=True)


def style_ax(ax, title=None, xlabel=None, ylabel=None, grid=True):
    ax.set_facecolor(BLACK)

    for spine in ax.spines.values():
        spine.set_color(ACCENT)
        spine.set_linewidth(1.1)

    ax.tick_params(axis="x", colors=ACCENT, labelcolor=ACCENT)
    ax.tick_params(axis="y", colors=ACCENT, labelcolor=ACCENT)

    if title is not None:
        ax.set_title(title, color=ACCENT, pad=12)

    if xlabel is not None:
        ax.set_xlabel(xlabel, color=ACCENT, labelpad=10)

    if ylabel is not None:
        ax.set_ylabel(ylabel, color=ACCENT, labelpad=10)

    if grid:
        ax.grid(True, color=ACCENT_DIM, alpha=0.25, linewidth=0.8)


def force_cyan_text(ax):
    ax.title.set_color(ACCENT)
    ax.xaxis.label.set_color(ACCENT)
    ax.yaxis.label.set_color(ACCENT)

    for tick_label in ax.get_xticklabels():
        tick_label.set_color(ACCENT)

    for tick_label in ax.get_yticklabels():
        tick_label.set_color(ACCENT)


def style_colorbar(cbar, label):
    cbar.set_label(label, color=ACCENT)
    cbar.ax.yaxis.set_tick_params(color=ACCENT)
    plt.setp(cbar.ax.get_yticklabels(), color=ACCENT)
    cbar.outline.set_edgecolor(ACCENT)


def style_legend(legend):
    if legend is not None:
        legend.get_frame().set_facecolor(BLACK)
        legend.get_frame().set_edgecolor(ACCENT)

        for text in legend.get_texts():
            text.set_color(ACCENT)


def save_render_close(fig, path, width=INLINE_IMAGE_WIDTH):
    if SAVE_PLOTS_TO_DISK:
        fig.savefig(path, dpi=PLOT_DPI, bbox_inches="tight", facecolor=BLACK)

    plt.close(fig)

    if RENDER_ALL_IMAGES_INLINE and HAVE_IPYTHON and os.path.exists(path):
        display(Image(filename=path, width=width))


def offdiag_values(M):
    M = np.asarray(M, dtype=float)

    if M.shape[0] <= 1:
        return np.asarray([], dtype=float)

    mask = ~np.eye(M.shape[0], dtype=bool)
    vals = M[mask]
    vals = vals[np.isfinite(vals)]

    return vals


def dynamic_limits(values, lower_pct=2, upper_pct=98, pad_frac=0.08, lower_bound=None, upper_bound=None):
    values = np.asarray(values, dtype=float)
    values = values[np.isfinite(values)]

    if len(values) == 0:
        vmin, vmax = 0.0, 1.0
    else:
        vmin = float(np.percentile(values, lower_pct))
        vmax = float(np.percentile(values, upper_pct))

        if not np.isfinite(vmin):
            vmin = float(np.nanmin(values))

        if not np.isfinite(vmax):
            vmax = float(np.nanmax(values))

        if vmax <= vmin:
            center = float(np.nanmean(values))
            eps = max(1e-6, abs(center) * 0.01, 1e-5)
            vmin = center - eps
            vmax = center + eps

    pad = (vmax - vmin) * pad_frac
    vmin -= pad
    vmax += pad

    if lower_bound is not None:
        vmin = max(lower_bound, vmin)

    if upper_bound is not None:
        vmax = min(upper_bound, vmax)

    if vmax <= vmin:
        vmax = vmin + 1e-6

    return vmin, vmax


def pairwise_distance_correlation(D_original, coords):
    coords = np.asarray(coords, dtype=float)
    D_emb = squareform(pdist(coords[:, :2], metric="euclidean"))

    a = offdiag_values(D_original)
    b = offdiag_values(D_emb)

    if len(a) == 0 or len(b) == 0:
        return np.nan, D_emb

    if np.std(a) <= 0 or np.std(b) <= 0:
        return np.nan, D_emb

    return float(np.corrcoef(a, b)[0, 1]), D_emb


def find_topology_embedding_dir():
    candidate_dirs = [
        os.path.join(
            base_dir,
            "topology_channel_fingerprints",
            f"h1_topology_channel_fingerprints_{FAST_OPTION}_fps",
            "topology_only_channel_embedding",
        ),
        os.path.join(
            base_dir,
            "topology_channel_fingerprints",
            f"h1_topology_channel_fingerprints_{FAST_OPTION}_fps_cyberpunk",
            "topology_only_channel_embedding",
        ),
    ]

    for candidate_dir in candidate_dirs:
        coord_path = os.path.join(candidate_dir, "topology_only_channel_embedding_coordinates.csv")
        npz_path = os.path.join(candidate_dir, "topology_only_channel_embedding_outputs.npz")

        if os.path.exists(coord_path) and os.path.exists(npz_path):
            return candidate_dir

    raise FileNotFoundError(
        "Could not find topology-only embedding outputs. Expected both coordinate CSV and NPZ in:\n\n"
        + "\n".join(candidate_dirs)
        + "\n\nRun the full Topology-Only Channel Embedding cell first, then run this full plot cell."
    )


def plot_embedding(
    coords,
    channels,
    cluster_labels,
    cluster_ids,
    title,
    xlabel,
    ylabel,
    path,
    score_values=None,
    score_label=None,
    width=930,
):
    fig, ax = plt.subplots(figsize=(8.4, 7.2), facecolor=BLACK)
    style_ax(
        ax,
        title=title,
        xlabel=xlabel,
        ylabel=ylabel,
    )

    sc = None

    for cluster_idx, cluster_id in enumerate(cluster_ids):
        mask = cluster_labels == cluster_id
        marker = MARKERS[cluster_idx % len(MARKERS)]

        if score_values is None:
            ax.scatter(
                coords[mask, 0],
                coords[mask, 1],
                s=165,
                color=CYAN_FILL,
                edgecolor=ACCENT,
                marker=marker,
                linewidth=1.2,
                alpha=0.9,
                label=f"Cluster {cluster_id}",
                zorder=3,
            )
        else:
            sc = ax.scatter(
                coords[mask, 0],
                coords[mask, 1],
                c=np.asarray(score_values)[mask],
                s=165,
                cmap=CYAN_SEQUENTIAL_CMAP,
                edgecolor=ACCENT,
                marker=marker,
                linewidth=1.2,
                alpha=0.95,
                vmin=0,
                vmax=100,
                label=f"Cluster {cluster_id}",
                zorder=3,
            )

    for i, ch in enumerate(channels):
        txt = ax.text(
            coords[i, 0],
            coords[i, 1],
            ch,
            fontsize=8,
            color=ACCENT,
            alpha=0.95,
            zorder=5,
        )
        txt.set_path_effects([pe.withStroke(linewidth=2.0, foreground=BLACK)])

    legend = ax.legend(loc="best", framealpha=1.0, fontsize=8)
    style_legend(legend)

    if score_values is not None and sc is not None:
        cbar = plt.colorbar(sc, ax=ax, shrink=0.82)
        style_colorbar(cbar, score_label or "Score")

    force_cyan_text(ax)
    plt.tight_layout()
    save_render_close(fig, path, width=width)


embedding_out_dir = find_topology_embedding_dir()
input_dir = os.path.dirname(embedding_out_dir)

coords_csv_path = os.path.join(embedding_out_dir, "topology_only_channel_embedding_coordinates.csv")
npz_path = os.path.join(embedding_out_dir, "topology_only_channel_embedding_outputs.npz")
method_quality_csv_path = os.path.join(embedding_out_dir, "topology_only_embedding_method_quality.csv")
feature_driver_csv_path = os.path.join(embedding_out_dir, "topology_only_embedding_feature_driver_ranking.csv")
isolation_csv_path = os.path.join(embedding_out_dir, "topology_only_channel_isolation_ranking.csv")
distance_csv_out_path = os.path.join(embedding_out_dir, "topology_only_original_distance_matrix.csv")
consensus_distance_csv_path = os.path.join(embedding_out_dir, "topology_only_consensus_embedding_distance_matrix.csv")

embedding_table = pd.read_csv(coords_csv_path)
embedding_table["channel"] = embedding_table["channel"].astype(str)

data = np.load(npz_path, allow_pickle=True)

channels = [str(ch) for ch in data["channels"]]
D = np.asarray(data["original_topology_distance_matrix"], dtype=float)
consensus_coords = np.asarray(data["consensus_coordinates"], dtype=float)
pca_coords = np.asarray(data["pca_coordinates"], dtype=float)
mds_coords = np.asarray(data["mds_coordinates"], dtype=float)
spectral_coords = np.asarray(data["spectral_coordinates"], dtype=float)
cluster_labels = np.asarray(data["cluster_labels"], dtype=int)
topology_isolation = np.asarray(data["topology_isolation"], dtype=float)
topology_isolation_score = np.asarray(data["topology_isolation_score"], dtype=float)

pca_explained = np.asarray(data["pca_explained_variance_ratio"], dtype=float)
mds_explained = np.asarray(data["mds_explained_variance_ratio"], dtype=float)
spectral_lambdas = np.asarray(data["spectral_lambdas"], dtype=float)

cluster_ids = sorted(np.unique(cluster_labels))

if "umap_coordinates" in data.files:
    umap_coords = np.asarray(data["umap_coordinates"], dtype=float)
    HAVE_SAVED_UMAP = True
else:
    umap_coords = None
    HAVE_SAVED_UMAP = False

consensus_corr, consensus_D = pairwise_distance_correlation(D, consensus_coords)
pca_corr, pca_D = pairwise_distance_correlation(D, pca_coords[:, :2])
mds_corr, mds_D = pairwise_distance_correlation(D, mds_coords[:, :2])
spectral_corr, spectral_D = pairwise_distance_correlation(D, spectral_coords[:, :2])

if HAVE_SAVED_UMAP:
    umap_corr, umap_D = pairwise_distance_correlation(D, umap_coords)
else:
    umap_corr = np.nan
    umap_D = None

method_quality_rows = [
    {
        "embedding_method": "consensus",
        "pairwise_distance_correlation_with_original_topology_distance": consensus_corr,
    },
    {
        "embedding_method": "pca",
        "pairwise_distance_correlation_with_original_topology_distance": pca_corr,
    },
    {
        "embedding_method": "classical_mds",
        "pairwise_distance_correlation_with_original_topology_distance": mds_corr,
    },
    {
        "embedding_method": "spectral",
        "pairwise_distance_correlation_with_original_topology_distance": spectral_corr,
    },
]

if HAVE_SAVED_UMAP:
    method_quality_rows.append({
        "embedding_method": "umap",
        "pairwise_distance_correlation_with_original_topology_distance": umap_corr,
    })

method_quality_df = pd.DataFrame(method_quality_rows)
method_quality_df.to_csv(method_quality_csv_path, index=False)

pd.DataFrame(D, index=channels, columns=channels).to_csv(distance_csv_out_path)
pd.DataFrame(consensus_D, index=channels, columns=channels).to_csv(consensus_distance_csv_path)

if "topology_isolation_score" not in embedding_table.columns:
    embedding_table["topology_isolation_score"] = topology_isolation_score

if "topology_isolation" not in embedding_table.columns:
    embedding_table["topology_isolation"] = topology_isolation

embedding_table.to_csv(coords_csv_path, index=False)

print("\nTopology-Only Channel Embedding Coordinates:")
if HAVE_IPYTHON:
    display(embedding_table)
else:
    print(embedding_table)

print("\nTopology-Only Embedding Method Quality:")
if HAVE_IPYTHON:
    display(method_quality_df)
else:
    print(method_quality_df)

log("\n" + "=" * 72)
log("Regenerating ALL topology-only channel embedding plots")
log(f"Embedding directory: {embedding_out_dir}")
log("=" * 72)

plot_embedding(
    consensus_coords,
    channels,
    cluster_labels,
    cluster_ids,
    title=f"Consensus Topology-Only Channel Embedding | {FAST_OPTION.upper()}",
    xlabel="Consensus topology axis 1",
    ylabel="Consensus topology axis 2",
    path=os.path.join(embedding_out_dir, "01_consensus_topology_only_channel_embedding.png"),
    score_values=topology_isolation_score,
    score_label="Topology isolation score",
    width=930,
)

plot_embedding(
    pca_coords[:, :2],
    channels,
    cluster_labels,
    cluster_ids,
    title=f"PCA Topology-Only Channel Embedding | PC1={pca_explained[0]*100:.1f}%, PC2={pca_explained[1]*100:.1f}%",
    xlabel="Topology PC1",
    ylabel="Topology PC2",
    path=os.path.join(embedding_out_dir, "02_pca_topology_only_channel_embedding.png"),
    score_values=topology_isolation_score,
    score_label="Topology isolation score",
    width=930,
)

plot_embedding(
    mds_coords[:, :2],
    channels,
    cluster_labels,
    cluster_ids,
    title=f"Classical MDS Topology-Only Channel Embedding | distance corr={mds_corr:.3f}",
    xlabel="Topology MDS 1",
    ylabel="Topology MDS 2",
    path=os.path.join(embedding_out_dir, "03_mds_topology_only_channel_embedding.png"),
    score_values=topology_isolation_score,
    score_label="Topology isolation score",
    width=930,
)

plot_embedding(
    spectral_coords[:, :2],
    channels,
    cluster_labels,
    cluster_ids,
    title="Spectral Topology-Only Channel Embedding",
    xlabel="Topology spectral axis 1",
    ylabel="Topology spectral axis 2",
    path=os.path.join(embedding_out_dir, "04_spectral_topology_only_channel_embedding.png"),
    score_values=topology_isolation_score,
    score_label="Topology isolation score",
    width=930,
)

if HAVE_SAVED_UMAP:
    plot_embedding(
        umap_coords,
        channels,
        cluster_labels,
        cluster_ids,
        title=f"UMAP Topology-Only Channel Embedding | distance corr={umap_corr:.3f}",
        xlabel="Topology UMAP 1",
        ylabel="Topology UMAP 2",
        path=os.path.join(embedding_out_dir, "05_umap_topology_only_channel_embedding.png"),
        score_values=topology_isolation_score,
        score_label="Topology isolation score",
        width=930,
    )

fig, ax = plt.subplots(figsize=(9, 6), facecolor=BLACK)
style_ax(
    ax,
    title="Topology-Only Embedding Distance Preservation",
    xlabel="Original topology fingerprint distance",
    ylabel="Consensus embedding distance",
)

orig_vals = offdiag_values(D)
embed_vals = offdiag_values(consensus_D)

ax.scatter(
    orig_vals,
    embed_vals,
    s=34,
    facecolors=BLACK,
    edgecolors=CYAN_FILL,
    linewidths=1.0,
    alpha=0.82,
)

if len(orig_vals) > 1 and np.std(orig_vals) > 0:
    m, b = np.polyfit(orig_vals, embed_vals, deg=1)
    xs = np.linspace(np.nanmin(orig_vals), np.nanmax(orig_vals), 150)
    ys = m * xs + b

    ax.plot(
        xs,
        ys,
        color=ACCENT,
        linewidth=1.6,
        linestyle="--",
        label=f"corr={consensus_corr:.3f}",
    )

legend = ax.legend(loc="best", framealpha=1.0)
style_legend(legend)
force_cyan_text(ax)

distance_preservation_path = os.path.join(embedding_out_dir, "06_consensus_distance_preservation.png")
plt.tight_layout()
save_render_close(fig, distance_preservation_path, width=950)

Z_order = linkage(np.asarray(data["feature_matrix_standardized"], dtype=float), method="ward", optimal_ordering=True)
dendro_order = leaves_list(Z_order)
ordered_channels = [channels[i] for i in dendro_order]
ordered_D = D[np.ix_(dendro_order, dendro_order)]

dist_vmin, dist_vmax = dynamic_limits(
    offdiag_values(ordered_D),
    lower_pct=2,
    upper_pct=98,
    pad_frac=0.10,
    lower_bound=0.0,
)

fig, ax = plt.subplots(figsize=(9, 8), facecolor=BLACK)
style_ax(
    ax,
    title="Topology-Only Fingerprint Distance Matrix Ordered by Embedding Structure",
    xlabel="Channel",
    ylabel="Channel",
)

im = ax.imshow(
    ordered_D,
    aspect="equal",
    interpolation="nearest",
    cmap=CYAN_SEQUENTIAL_CMAP,
    vmin=dist_vmin,
    vmax=dist_vmax,
)

ax.set_xticks(np.arange(len(ordered_channels)))
ax.set_xticklabels(ordered_channels, rotation=90)
ax.set_yticks(np.arange(len(ordered_channels)))
ax.set_yticklabels(ordered_channels)

cbar = plt.colorbar(im, ax=ax, shrink=0.85)
style_colorbar(cbar, "Topology fingerprint distance")
force_cyan_text(ax)

ordered_distance_path = os.path.join(embedding_out_dir, "07_ordered_topology_only_distance_heatmap.png")
plt.tight_layout()
save_render_close(fig, ordered_distance_path, width=950)

feature_names = [str(x) for x in data["feature_names"]]

if "pca_components" in data.files:
    pca_components = np.asarray(data["pca_components"], dtype=float)
    topology_feature_strength = np.sqrt(pca_components[0] ** 2 + pca_components[1] ** 2)

    feature_driver_df = pd.DataFrame({
        "feature_name": feature_names,
        "topology_embedding_driver_strength": topology_feature_strength,
        "pc1_loading": pca_components[0],
        "pc2_loading": pca_components[1],
    }).sort_values("topology_embedding_driver_strength", ascending=False).reset_index(drop=True)

    feature_driver_df.to_csv(feature_driver_csv_path, index=False)
else:
    if os.path.exists(feature_driver_csv_path):
        feature_driver_df = pd.read_csv(feature_driver_csv_path)
    else:
        feature_driver_df = pd.DataFrame()

if len(feature_driver_df) > 0:
    plot_driver_df = feature_driver_df.head(min(TOP_FEATURES_TO_PLOT, len(feature_driver_df))).iloc[::-1].copy()

    fig, ax = plt.subplots(figsize=(12, 8), facecolor=BLACK)
    style_ax(
        ax,
        title="Top Features Driving the Topology-Only Channel Embedding",
        xlabel="PCA loading strength across topology axes 1 and 2",
        ylabel="Topology fingerprint feature",
    )

    y = np.arange(len(plot_driver_df))

    ax.barh(
        y,
        plot_driver_df["topology_embedding_driver_strength"],
        color=CYAN_FILL,
        edgecolor=ACCENT,
        linewidth=1.0,
        alpha=0.82,
    )

    ax.set_yticks(y)
    ax.set_yticklabels(plot_driver_df["feature_name"])

    for i, val in enumerate(plot_driver_df["topology_embedding_driver_strength"]):
        ax.text(
            val,
            i,
            f"  {val:.3f}",
            va="center",
            ha="left",
            color=ACCENT,
            fontsize=8,
        )

    force_cyan_text(ax)

    feature_driver_path = os.path.join(embedding_out_dir, "08_topology_embedding_feature_drivers.png")
    plt.tight_layout()
    save_render_close(fig, feature_driver_path, width=1150)

isolation_df = embedding_table[[
    "channel",
    "topological_cluster",
    "topology_isolation",
    "topology_isolation_score",
]].sort_values("topology_isolation_score", ascending=False).reset_index(drop=True)

isolation_df.to_csv(isolation_csv_path, index=False)

fig, ax = plt.subplots(figsize=(13, 6), facecolor=BLACK)
style_ax(
    ax,
    title="Topology-Only Channel Isolation Ranking",
    xlabel="Channel",
    ylabel="Isolation score",
)

x = np.arange(len(isolation_df))

ax.bar(
    x,
    isolation_df["topology_isolation_score"],
    color=CYAN_FILL,
    edgecolor=ACCENT,
    linewidth=0.9,
    alpha=0.82,
)

ax.set_xticks(x)
ax.set_xticklabels(isolation_df["channel"], rotation=90)
ax.set_ylim(0, 105)

for i, row in isolation_df.iterrows():
    if i < 8:
        ax.text(
            i,
            row["topology_isolation_score"],
            f"{row['topology_isolation_score']:.0f}",
            ha="center",
            va="bottom",
            color=ACCENT,
            fontsize=8,
        )

force_cyan_text(ax)

isolation_path = os.path.join(embedding_out_dir, "09_topology_only_channel_isolation_ranking.png")
plt.tight_layout()
save_render_close(fig, isolation_path, width=1150)

scalp_xy = {
    "Fp1": (-0.45, 1.00), "Fpz": (0.00, 1.08), "Fp2": (0.45, 1.00),
    "F7": (-0.95, 0.60), "F3": (-0.45, 0.55), "Fz": (0.00, 0.60),
    "F4": (0.45, 0.55), "F8": (0.95, 0.60),
    "FC5": (-0.75, 0.25), "FC1": (-0.25, 0.25),
    "FC2": (0.25, 0.25), "FC6": (0.75, 0.25),
    "M1": (-1.15, 0.05), "T7": (-1.00, 0.00), "C3": (-0.50, 0.00),
    "Cz": (0.00, 0.00), "C4": (0.50, 0.00), "T8": (1.00, 0.00),
    "M2": (1.15, 0.05),
    "CP5": (-0.75, -0.30), "CP1": (-0.25, -0.30),
    "CP2": (0.25, -0.30), "CP6": (0.75, -0.30),
    "P7": (-0.90, -0.60), "P3": (-0.45, -0.60), "Pz": (0.00, -0.65),
    "P4": (0.45, -0.60), "P8": (0.90, -0.60),
    "POz": (0.00, -0.88),
    "O1": (-0.35, -1.00), "Oz": (0.00, -1.08), "O2": (0.35, -1.00),
}

plot_df = embedding_table[embedding_table["channel"].isin(scalp_xy)].copy()
plot_df["x"] = plot_df["channel"].map(lambda ch: scalp_xy[ch][0])
plot_df["y"] = plot_df["channel"].map(lambda ch: scalp_xy[ch][1])
plot_df["rank_high_to_low"] = plot_df["topology_isolation_score"].rank(
    method="first",
    ascending=False,
).astype(int)

scores = plot_df["topology_isolation_score"].to_numpy(dtype=float)
score_min = float(np.nanmin(scores))
score_max = float(np.nanmax(scores))

if score_max <= score_min:
    normalized_scores = np.full(len(plot_df), 0.5)
else:
    normalized_scores = (scores - score_min) / (score_max - score_min + 1e-12)

plot_df["marker_size"] = 90.0 + 520.0 * normalized_scores
plot_df["marker_alpha"] = 0.35 + 0.60 * normalized_scores

top_channels = plot_df.sort_values("topology_isolation_score", ascending=False).head(8)["channel"].tolist()
low_channels = plot_df.sort_values("topology_isolation_score", ascending=True).head(5)["channel"].tolist()

fig = plt.figure(figsize=(13.8, 8.2), facecolor=BLACK)
gs = GridSpec(
    1,
    2,
    width_ratios=[1.25, 0.85],
    wspace=0.25,
    figure=fig,
)

ax = fig.add_subplot(gs[0, 0])
rank_ax = fig.add_subplot(gs[0, 1])

style_ax(
    ax,
    title="Scalp Map of Topology-Only Isolation Score",
    grid=False,
)

head = plt.Circle((0, 0), 1.12, fill=False, color=ACCENT, linewidth=2.2)
ax.add_patch(head)
ax.plot([0.00, -0.10, 0.10, 0.00], [1.12, 1.28, 1.28, 1.12], color=ACCENT, linewidth=2.2)

scatter = ax.scatter(
    plot_df["x"],
    plot_df["y"],
    c=plot_df["topology_isolation_score"],
    s=plot_df["marker_size"],
    cmap=CYAN_SEQUENTIAL_CMAP,
    vmin=0,
    vmax=100,
    edgecolor=ACCENT,
    linewidth=1.25,
    alpha=0.92,
    zorder=4,
)

ax.scatter(
    plot_df["x"],
    plot_df["y"],
    s=plot_df["marker_size"] * 1.35,
    facecolors="none",
    edgecolors=CYAN_FILL,
    linewidth=0.9,
    alpha=0.20,
    zorder=3,
)

for _, row in plot_df.iterrows():
    ch = row["channel"]
    x0 = float(row["x"])
    y0 = float(row["y"])
    score = float(row["topology_isolation_score"])

    txt = ax.text(
        x0,
        y0,
        ch,
        ha="center",
        va="center",
        fontsize=8,
        color=ACCENT,
        fontweight="bold",
        zorder=6,
    )
    txt.set_path_effects([pe.withStroke(linewidth=2.1, foreground=BLACK)])

    if ch in top_channels:
        ax.scatter(
            [x0],
            [y0],
            s=float(row["marker_size"]) * 1.75,
            facecolors="none",
            edgecolors=ACCENT_SOFT,
            linewidth=1.8,
            alpha=0.95,
            zorder=5,
        )

        dx = 0.06 if x0 <= 0 else -0.06
        ha = "left" if x0 <= 0 else "right"

        ann = ax.text(
            x0 + dx,
            y0 + 0.055,
            f"{score:.0f}",
            ha=ha,
            va="center",
            fontsize=8,
            color=ACCENT_SOFT,
            fontweight="bold",
            zorder=7,
        )
        ann.set_path_effects([pe.withStroke(linewidth=2.0, foreground=BLACK)])

    elif ch in low_channels:
        ax.scatter(
            [x0],
            [y0],
            s=float(row["marker_size"]) * 1.35,
            facecolors="none",
            edgecolors=ACCENT_DIM,
            linewidth=1.1,
            alpha=0.60,
            zorder=5,
        )

ax.text(
    0,
    -1.30,
    "Bigger/brighter = more topologically isolated from other channels | numbers mark top isolated channels",
    ha="center",
    va="top",
    color=ACCENT,
    fontsize=9,
)

ax.set_xlim(-1.38, 1.38)
ax.set_ylim(-1.40, 1.36)
ax.set_aspect("equal")
ax.set_xticks([])
ax.set_yticks([])

cbar = plt.colorbar(scatter, ax=ax, shrink=0.78, pad=0.03)
style_colorbar(cbar, "Isolation score")

rank_df = plot_df.sort_values("topology_isolation_score", ascending=True).copy()

style_ax(
    rank_ax,
    title="Channel Isolation Ranking",
    xlabel="Isolation score",
    ylabel="Channel",
)

y = np.arange(len(rank_df))

rank_ax.barh(
    y,
    rank_df["topology_isolation_score"],
    color=CYAN_FILL,
    edgecolor=ACCENT,
    linewidth=0.8,
    alpha=0.82,
)

rank_ax.set_yticks(y)
rank_ax.set_yticklabels(rank_df["channel"])
rank_ax.set_xlim(0, 105)

rank_ax.axvline(
    float(np.nanmean(rank_df["topology_isolation_score"])),
    color=ACCENT_SOFT,
    linestyle="--",
    linewidth=1.2,
    alpha=0.85,
    label=f"Mean={np.nanmean(rank_df['topology_isolation_score']):.1f}",
)

for yy, (_, row) in zip(y, rank_df.iterrows()):
    score = float(row["topology_isolation_score"])
    ch = row["channel"]

    if ch in top_channels:
        rank_ax.text(
            score + 1.2,
            yy,
            f"{score:.0f}",
            va="center",
            ha="left",
            color=ACCENT_SOFT,
            fontsize=8,
            fontweight="bold",
        )
    else:
        rank_ax.text(
            score + 1.2,
            yy,
            f"{score:.0f}",
            va="center",
            ha="left",
            color=ACCENT,
            fontsize=7,
            alpha=0.75,
        )

legend = rank_ax.legend(loc="lower right", framealpha=1.0, fontsize=8)
style_legend(legend)

for spine in rank_ax.spines.values():
    spine.set_color(ACCENT)

rank_ax.tick_params(axis="x", colors=ACCENT, labelcolor=ACCENT)
rank_ax.tick_params(axis="y", colors=ACCENT, labelcolor=ACCENT)

plot_df_sorted = plot_df.sort_values("topology_isolation_score", ascending=False)

top_text = "Top isolated: " + ", ".join(
    [
        f"{row.channel}({row.topology_isolation_score:.0f})"
        for row in plot_df_sorted.head(5).itertuples()
    ]
)

bottom_text = "Least isolated: " + ", ".join(
    [
        f"{row.channel}({row.topology_isolation_score:.0f})"
        for row in plot_df_sorted.tail(5).itertuples()
    ]
)

fig.text(
    0.5,
    0.018,
    top_text + "    |    " + bottom_text,
    ha="center",
    va="bottom",
    color=ACCENT,
    fontsize=9,
)

fixed_scalp_path = os.path.join(embedding_out_dir, "10_scalp_map_topology_only_isolation_score.png")
fixed_scalp_copy_path = os.path.join(embedding_out_dir, "10_scalp_map_topology_only_isolation_score_fixed_ranked.png")

plt.tight_layout(rect=[0.0, 0.045, 1.0, 1.0])

if SAVE_PLOTS_TO_DISK:
    fig.savefig(fixed_scalp_path, dpi=PLOT_DPI, bbox_inches="tight", facecolor=BLACK)
    fig.savefig(fixed_scalp_copy_path, dpi=PLOT_DPI, bbox_inches="tight", facecolor=BLACK)

plt.close(fig)

if RENDER_ALL_IMAGES_INLINE and HAVE_IPYTHON and os.path.exists(fixed_scalp_path):
    display(Image(filename=fixed_scalp_path, width=1150))

available_score_cols = [
    col for col in [
        "topology_complexity_score",
        "loop_strength_score",
        "topological_activity_score",
        "topological_spread_score",
        "structure_density_score",
        "topological_uniqueness_score",
    ]
    if col in embedding_table.columns
]

if len(available_score_cols) > 0:
    score_label_map = {
        "topology_complexity_score": "Complexity",
        "loop_strength_score": "Loop strength",
        "topological_activity_score": "Activity",
        "topological_spread_score": "Spread",
        "structure_density_score": "Density",
        "topological_uniqueness_score": "Uniqueness",
    }

    ordered_score_df = embedding_table.set_index("channel").loc[ordered_channels, available_score_cols]

    fig, ax = plt.subplots(figsize=(9, 9), facecolor=BLACK)
    style_ax(
        ax,
        title="Topology Score Profiles Ordered by Topology-Only Embedding Structure",
        xlabel="Topology score",
        ylabel="Channel",
    )

    im = ax.imshow(
        ordered_score_df.to_numpy(dtype=float),
        aspect="auto",
        interpolation="nearest",
        cmap=CYAN_SEQUENTIAL_CMAP,
        vmin=0,
        vmax=100,
    )

    ax.set_yticks(np.arange(len(ordered_channels)))
    ax.set_yticklabels(ordered_channels)

    labels = [score_label_map.get(col, col) for col in available_score_cols]
    ax.set_xticks(np.arange(len(labels)))
    ax.set_xticklabels(labels, rotation=45, ha="right")

    cbar = plt.colorbar(im, ax=ax, shrink=0.85)
    style_colorbar(cbar, "Score percentile")
    force_cyan_text(ax)

    score_profile_path = os.path.join(embedding_out_dir, "11_ordered_topology_score_profiles_heatmap.png")
    plt.tight_layout()
    save_render_close(fig, score_profile_path, width=950)

log("\n" + "=" * 72)
log("Finished regenerating ALL topology-only channel embedding plots")
log(f"Embedding directory: {embedding_out_dir}")
log("Saved / replaced:")
log("01_consensus_topology_only_channel_embedding.png")
log("02_pca_topology_only_channel_embedding.png")
log("03_mds_topology_only_channel_embedding.png")
log("04_spectral_topology_only_channel_embedding.png")
if HAVE_SAVED_UMAP:
    log("05_umap_topology_only_channel_embedding.png")
log("06_consensus_distance_preservation.png")
log("07_ordered_topology_only_distance_heatmap.png")
if len(feature_driver_df) > 0:
    log("08_topology_embedding_feature_drivers.png")
log("09_topology_only_channel_isolation_ranking.png")
log("10_scalp_map_topology_only_isolation_score.png  <-- fixed informative version")
log("10_scalp_map_topology_only_isolation_score_fixed_ranked.png")
if len(available_score_cols) > 0:
    log("11_ordered_topology_score_profiles_heatmap.png")
log(f"Method quality CSV: {method_quality_csv_path}")
log(f"Isolation ranking CSV: {isolation_csv_path}")
log("=" * 72)
```

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patheffects as pe
from matplotlib.colors import LinearSegmentedColormap
from matplotlib.gridspec import GridSpec
from scipy.spatial.distance import pdist, squareform
from scipy.cluster.hierarchy import linkage, leaves_list

try:
    from scipy.linalg import eigh
    HAVE_SCIPY_EIGH = True
except Exception:
    HAVE_SCIPY_EIGH = False

try:
    from IPython.display import display, Image
    HAVE_IPYTHON = True
except Exception:
    HAVE_IPYTHON = False


base_dir = "/home/a/projects/Complete-Neural-Signal-Analysis"
FAST_OPTION = "ultra"

SAVE_PLOTS_TO_DISK = True
RENDER_ALL_IMAGES_INLINE = True
INLINE_IMAGE_WIDTH = 1050
PLOT_DPI = 140

TOP_PAIRS_TO_PLOT = 20
TOP_FEATURES_TO_PLOT = 35
NETWORK_TOP_K_NEIGHBORS = 3

BLACK = "#000000"
ACCENT = "#00FFFF"
ACCENT_SOFT = "#66FFFF"
ACCENT_DIM = "#008B8B"

CYAN_FILL = "#00D5FF"
CYAN_FILL_SOFT = "#38BDF8"
CYAN_FILL_DARK = "#005A63"

MARKERS = ["o", "s", "^", "D", "P", "X", "v", "*", "<", ">"]

CYAN_SEQUENTIAL_CMAP = LinearSegmentedColormap.from_list(
    "cyan_sequential",
    [BLACK, CYAN_FILL_DARK, CYAN_FILL, ACCENT_SOFT],
    N=256,
)

CYAN_DIVERGING_CMAP = LinearSegmentedColormap.from_list(
    "cyan_diverging",
    [ACCENT_SOFT, BLACK, CYAN_FILL],
    N=256,
)

plt.rcParams.update({
    "figure.facecolor": BLACK,
    "axes.facecolor": BLACK,
    "savefig.facecolor": BLACK,
    "text.color": ACCENT,
    "axes.labelcolor": ACCENT,
    "axes.edgecolor": ACCENT,
    "axes.titlecolor": ACCENT,
    "xtick.color": ACCENT,
    "ytick.color": ACCENT,
    "grid.color": ACCENT_DIM,
    "grid.alpha": 0.25,
    "legend.facecolor": BLACK,
    "legend.edgecolor": ACCENT,
    "legend.labelcolor": ACCENT,
    "font.size": 11,
})


def log(msg):
    print(msg, flush=True)


def style_ax(ax, title=None, xlabel=None, ylabel=None, grid=True):
    ax.set_facecolor(BLACK)

    for spine in ax.spines.values():
        spine.set_color(ACCENT)
        spine.set_linewidth(1.1)

    ax.tick_params(axis="x", colors=ACCENT, labelcolor=ACCENT)
    ax.tick_params(axis="y", colors=ACCENT, labelcolor=ACCENT)

    if title is not None:
        ax.set_title(title, color=ACCENT, pad=12)

    if xlabel is not None:
        ax.set_xlabel(xlabel, color=ACCENT, labelpad=10)

    if ylabel is not None:
        ax.set_ylabel(ylabel, color=ACCENT, labelpad=10)

    if grid:
        ax.grid(True, color=ACCENT_DIM, alpha=0.25, linewidth=0.8)


def force_cyan_text(ax):
    ax.title.set_color(ACCENT)
    ax.xaxis.label.set_color(ACCENT)
    ax.yaxis.label.set_color(ACCENT)

    for tick_label in ax.get_xticklabels():
        tick_label.set_color(ACCENT)

    for tick_label in ax.get_yticklabels():
        tick_label.set_color(ACCENT)


def style_colorbar(cbar, label):
    cbar.set_label(label, color=ACCENT)
    cbar.ax.yaxis.set_tick_params(color=ACCENT)
    plt.setp(cbar.ax.get_yticklabels(), color=ACCENT)
    cbar.outline.set_edgecolor(ACCENT)


def style_legend(legend):
    if legend is not None:
        legend.get_frame().set_facecolor(BLACK)
        legend.get_frame().set_edgecolor(ACCENT)

        for text in legend.get_texts():
            text.set_color(ACCENT)


def save_render_close(fig, path, width=INLINE_IMAGE_WIDTH):
    if SAVE_PLOTS_TO_DISK:
        fig.savefig(path, dpi=PLOT_DPI, bbox_inches="tight", facecolor=BLACK)

    plt.close(fig)

    if RENDER_ALL_IMAGES_INLINE and HAVE_IPYTHON and os.path.exists(path):
        display(Image(filename=path, width=width))


def offdiag_values(M):
    M = np.asarray(M, dtype=float)

    if M.shape[0] <= 1:
        return np.asarray([], dtype=float)

    mask = ~np.eye(M.shape[0], dtype=bool)
    vals = M[mask]
    vals = vals[np.isfinite(vals)]

    return vals


def dynamic_limits(values, lower_pct=2, upper_pct=98, pad_frac=0.08, lower_bound=None, upper_bound=None):
    values = np.asarray(values, dtype=float)
    values = values[np.isfinite(values)]

    if len(values) == 0:
        vmin, vmax = 0.0, 1.0
    else:
        vmin = float(np.percentile(values, lower_pct))
        vmax = float(np.percentile(values, upper_pct))

        if not np.isfinite(vmin):
            vmin = float(np.nanmin(values))

        if not np.isfinite(vmax):
            vmax = float(np.nanmax(values))

        if vmax <= vmin:
            center = float(np.nanmean(values))
            eps = max(1e-6, abs(center) * 0.01, 1e-5)
            vmin = center - eps
            vmax = center + eps

    pad = (vmax - vmin) * pad_frac
    vmin -= pad
    vmax += pad

    if lower_bound is not None:
        vmin = max(lower_bound, vmin)

    if upper_bound is not None:
        vmax = min(upper_bound, vmax)

    if vmax <= vmin:
        vmax = vmin + 1e-6

    return vmin, vmax


def percentile_score(values, higher_is_more=True):
    s = pd.Series(values, dtype=float)

    if not higher_is_more:
        s = -s

    if s.nunique(dropna=True) <= 1:
        return np.full(len(s), 50.0, dtype=float)

    return s.rank(method="average", pct=True).to_numpy(dtype=float) * 100.0


def safe_corr(a, b):
    a = np.asarray(a, dtype=float)
    b = np.asarray(b, dtype=float)

    mask = np.isfinite(a) & np.isfinite(b)
    a = a[mask]
    b = b[mask]

    if len(a) < 2:
        return np.nan

    if np.std(a) <= 0 or np.std(b) <= 0:
        return np.nan

    return float(np.corrcoef(a, b)[0, 1])


def cosine_similarity_matrix(X):
    X = np.asarray(X, dtype=float)
    norms = np.linalg.norm(X, axis=1, keepdims=True)
    Xn = X / (norms + 1e-12)

    S = Xn @ Xn.T
    S = np.clip(S, -1.0, 1.0)
    np.fill_diagonal(S, 1.0)

    return S


def pearson_similarity_matrix(X):
    X = np.asarray(X, dtype=float)
    S = np.corrcoef(X)

    S = np.nan_to_num(S, nan=0.0, posinf=0.0, neginf=0.0)
    S = np.clip(S, -1.0, 1.0)
    np.fill_diagonal(S, 1.0)

    return S


def gaussian_similarity_from_distance(D, sigma=None):
    D = np.asarray(D, dtype=float)

    off = offdiag_values(D)

    if sigma is None:
        if len(off) == 0:
            sigma = 1.0
        else:
            sigma = float(np.nanmedian(off))

    sigma = max(float(sigma), 1e-8)

    S = np.exp(-(D ** 2) / (2.0 * sigma ** 2))
    np.fill_diagonal(S, 1.0)

    return S, sigma


def inverse_distance_similarity(D, scale=None):
    D = np.asarray(D, dtype=float)

    off = offdiag_values(D)

    if scale is None:
        if len(off) == 0:
            scale = 1.0
        else:
            scale = float(np.nanmedian(off))

    scale = max(float(scale), 1e-8)

    S = 1.0 / (1.0 + D / scale)
    np.fill_diagonal(S, 1.0)

    return S, scale


def similarity_to_distance(S):
    S = np.asarray(S, dtype=float)
    S = np.clip(S, 0.0, 1.0)

    D = 1.0 - S
    np.fill_diagonal(D, 0.0)

    return D


def classical_mds_from_distances(D, n_components=2):
    D = np.asarray(D, dtype=float)
    n = D.shape[0]

    if n < 2:
        return np.zeros((n, n_components), dtype=float)

    D2 = D ** 2
    J = np.eye(n) - np.ones((n, n)) / n
    B = -0.5 * J @ D2 @ J

    if HAVE_SCIPY_EIGH:
        vals, vecs = eigh(B)
    else:
        vals, vecs = np.linalg.eigh(B)

    idx = np.argsort(vals)[::-1]
    vals = vals[idx]
    vecs = vecs[:, idx]

    coords = np.zeros((n, n_components), dtype=float)

    for k in range(min(n_components, n)):
        if vals[k] > 0:
            coords[:, k] = vecs[:, k] * np.sqrt(vals[k])

    return coords


def choose_existing_input_dir():
    candidate_dirs = [
        os.path.join(
            base_dir,
            "topology_channel_fingerprints",
            f"h1_topology_channel_fingerprints_{FAST_OPTION}_fps",
        ),
        os.path.join(
            base_dir,
            "topology_channel_fingerprints",
            f"h1_topology_channel_fingerprints_{FAST_OPTION}_fps_cyberpunk",
        ),
    ]

    for candidate_dir in candidate_dirs:
        feature_path = os.path.join(candidate_dir, "h1_topology_channel_fingerprint_matrix_standardized.csv")

        if os.path.exists(feature_path):
            return candidate_dir

    raise FileNotFoundError(
        "Could not find the standardized topology fingerprint matrix.\n\n"
        "Expected one of:\n"
        + "\n".join(
            os.path.join(d, "h1_topology_channel_fingerprint_matrix_standardized.csv")
            for d in candidate_dirs
        )
        + "\n\nRun the Topology-Based Channel Fingerprints cell first, then run this similarity matrix cell."
    )


def plot_matrix(matrix, labels, title, colorbar_label, path, vmin=None, vmax=None, width=950):
    fig, ax = plt.subplots(figsize=(9.5, 8.5), facecolor=BLACK)
    style_ax(
        ax,
        title=title,
        xlabel="Channel",
        ylabel="Channel",
    )

    im = ax.imshow(
        matrix,
        aspect="equal",
        interpolation="nearest",
        cmap=CYAN_SEQUENTIAL_CMAP,
        vmin=vmin,
        vmax=vmax,
    )

    ax.set_xticks(np.arange(len(labels)))
    ax.set_xticklabels(labels, rotation=90)
    ax.set_yticks(np.arange(len(labels)))
    ax.set_yticklabels(labels)

    cbar = plt.colorbar(im, ax=ax, shrink=0.85)
    style_colorbar(cbar, colorbar_label)
    force_cyan_text(ax)

    plt.tight_layout()
    save_render_close(fig, path, width=width)


input_dir = choose_existing_input_dir()

feature_z_csv_path = os.path.join(input_dir, "h1_topology_channel_fingerprint_matrix_standardized.csv")
feature_raw_csv_path = os.path.join(input_dir, "h1_topology_channel_fingerprint_matrix.csv")
score_csv_path = os.path.join(input_dir, "h1_topology_channel_fingerprint_scores.csv")
summary_csv_path = os.path.join(input_dir, "h1_topology_channel_fingerprint_summary.csv")

cluster_assignment_csv_path = os.path.join(
    input_dir,
    "topological_clustering_of_channels",
    "topological_channel_cluster_assignments.csv",
)

topology_embedding_csv_path = os.path.join(
    input_dir,
    "topology_only_channel_embedding",
    "topology_only_channel_embedding_coordinates.csv",
)

similarity_out_dir = os.path.join(input_dir, "topology_derived_channel_similarity_matrix")
os.makedirs(similarity_out_dir, exist_ok=True)

log("\n" + "=" * 72)
log("Running Topology-Derived Channel Similarity Matrix")
log(f"Input directory: {input_dir}")
log(f"Output directory: {similarity_out_dir}")
log("=" * 72)

feature_z_df = pd.read_csv(feature_z_csv_path, index_col=0)
feature_z_df.index = feature_z_df.index.astype(str)
feature_z_df = feature_z_df.replace([np.inf, -np.inf], np.nan).fillna(0.0)

if os.path.exists(feature_raw_csv_path):
    feature_raw_df = pd.read_csv(feature_raw_csv_path, index_col=0)
    feature_raw_df.index = feature_raw_df.index.astype(str)
    feature_raw_df = feature_raw_df.loc[feature_z_df.index]
else:
    feature_raw_df = feature_z_df.copy()

if os.path.exists(score_csv_path):
    score_df = pd.read_csv(score_csv_path)
    score_df["channel"] = score_df["channel"].astype(str)
else:
    score_df = pd.DataFrame({"channel": feature_z_df.index})

if os.path.exists(summary_csv_path):
    summary_df = pd.read_csv(summary_csv_path)

    if "channel" in summary_df.columns:
        summary_df["channel"] = summary_df["channel"].astype(str)
else:
    summary_df = pd.DataFrame({"channel": feature_z_df.index})

channels = list(feature_z_df.index)
X = feature_z_df.to_numpy(dtype=float)
X = np.nan_to_num(X, nan=0.0, posinf=0.0, neginf=0.0)

n_channels = len(channels)

if n_channels < 3:
    raise RuntimeError("Need at least 3 channels for topology-derived similarity.")

if os.path.exists(cluster_assignment_csv_path):
    cluster_assignment_df = pd.read_csv(cluster_assignment_csv_path)
    cluster_assignment_df["channel"] = cluster_assignment_df["channel"].astype(str)
    cluster_assignment_df = cluster_assignment_df.set_index("channel").loc[channels].reset_index()
    cluster_labels = cluster_assignment_df["topological_cluster"].to_numpy(dtype=int)
else:
    cluster_labels = np.ones(n_channels, dtype=int)
    cluster_assignment_df = pd.DataFrame({
        "channel": channels,
        "topological_cluster": cluster_labels,
        "cluster_name": ["Cluster 1"] * n_channels,
    })

cluster_ids = sorted(np.unique(cluster_labels))

if "region" in score_df.columns:
    region_map = score_df.set_index("channel")["region"].to_dict()
elif "region" in summary_df.columns:
    region_map = summary_df.set_index("channel")["region"].to_dict()
else:
    region_map = {}

regions = [region_map.get(ch, "Unknown") for ch in channels]

euclidean_distance_mat = squareform(pdist(X, metric="euclidean"))
gaussian_similarity_mat, gaussian_sigma = gaussian_similarity_from_distance(euclidean_distance_mat)
inverse_similarity_mat, inverse_scale = inverse_distance_similarity(euclidean_distance_mat)
cosine_raw_similarity_mat = cosine_similarity_matrix(X)
pearson_raw_similarity_mat = pearson_similarity_matrix(X)

cosine_similarity_01_mat = (cosine_raw_similarity_mat + 1.0) / 2.0
pearson_similarity_01_mat = (pearson_raw_similarity_mat + 1.0) / 2.0

np.fill_diagonal(cosine_similarity_01_mat, 1.0)
np.fill_diagonal(pearson_similarity_01_mat, 1.0)

consensus_similarity_mat = (
    0.42 * gaussian_similarity_mat
    + 0.28 * cosine_similarity_01_mat
    + 0.20 * pearson_similarity_01_mat
    + 0.10 * inverse_similarity_mat
)

consensus_similarity_mat = np.clip(consensus_similarity_mat, 0.0, 1.0)
consensus_similarity_mat = 0.5 * (consensus_similarity_mat + consensus_similarity_mat.T)
np.fill_diagonal(consensus_similarity_mat, 1.0)

consensus_distance_mat = similarity_to_distance(consensus_similarity_mat)

if n_channels >= 3:
    Z_order = linkage(squareform(consensus_distance_mat, checks=False), method="average", optimal_ordering=True)
    ordered_idx = leaves_list(Z_order)
else:
    ordered_idx = np.arange(n_channels)

ordered_channels = [channels[i] for i in ordered_idx]
ordered_regions = [regions[i] for i in ordered_idx]
ordered_clusters = cluster_labels[ordered_idx]

ordered_consensus_similarity = consensus_similarity_mat[np.ix_(ordered_idx, ordered_idx)]
ordered_consensus_distance = consensus_distance_mat[np.ix_(ordered_idx, ordered_idx)]
ordered_gaussian_similarity = gaussian_similarity_mat[np.ix_(ordered_idx, ordered_idx)]
ordered_cosine_similarity = cosine_similarity_01_mat[np.ix_(ordered_idx, ordered_idx)]
ordered_pearson_similarity = pearson_similarity_01_mat[np.ix_(ordered_idx, ordered_idx)]
ordered_inverse_similarity = inverse_similarity_mat[np.ix_(ordered_idx, ordered_idx)]

pair_rows = []

for i in range(n_channels):
    for j in range(i + 1, n_channels):
        same_region = regions[i] == regions[j]
        same_cluster = cluster_labels[i] == cluster_labels[j]

        pair_rows.append({
            "channel_a": channels[i],
            "channel_b": channels[j],
            "region_a": regions[i],
            "region_b": regions[j],
            "cluster_a": int(cluster_labels[i]),
            "cluster_b": int(cluster_labels[j]),
            "same_region": bool(same_region),
            "same_topological_cluster": bool(same_cluster),
            "topology_consensus_similarity": float(consensus_similarity_mat[i, j]),
            "topology_consensus_distance": float(consensus_distance_mat[i, j]),
            "topology_gaussian_similarity": float(gaussian_similarity_mat[i, j]),
            "topology_cosine_similarity_01": float(cosine_similarity_01_mat[i, j]),
            "topology_pearson_similarity_01": float(pearson_similarity_01_mat[i, j]),
            "topology_inverse_distance_similarity": float(inverse_similarity_mat[i, j]),
            "topology_feature_euclidean_distance": float(euclidean_distance_mat[i, j]),
        })

pairwise_similarity_df = pd.DataFrame(pair_rows).sort_values(
    "topology_consensus_similarity",
    ascending=False,
).reset_index(drop=True)

hub_rows = []

for i, ch in enumerate(channels):
    mask = np.arange(n_channels) != i
    sims = consensus_similarity_mat[i, mask]
    dists = consensus_distance_mat[i, mask]

    other_channels = np.asarray(channels, dtype=object)[mask]

    sorted_neighbor_idx = np.argsort(sims)[::-1]
    top_neighbors = other_channels[sorted_neighbor_idx[:5]].tolist()
    top_neighbor_sims = sims[sorted_neighbor_idx[:5]].tolist()

    hub_rows.append({
        "channel": ch,
        "region": regions[i],
        "topological_cluster": int(cluster_labels[i]),
        "mean_similarity_to_other_channels": float(np.nanmean(sims)),
        "median_similarity_to_other_channels": float(np.nanmedian(sims)),
        "max_similarity_to_other_channels": float(np.nanmax(sims)),
        "min_similarity_to_other_channels": float(np.nanmin(sims)),
        "std_similarity_to_other_channels": float(np.nanstd(sims)),
        "mean_distance_to_other_channels": float(np.nanmean(dists)),
        "nearest_topology_neighbor": str(top_neighbors[0]) if len(top_neighbors) else "",
        "nearest_topology_neighbor_similarity": float(top_neighbor_sims[0]) if len(top_neighbor_sims) else np.nan,
        "top_5_topology_neighbors": ", ".join(
            [f"{a}({b:.3f})" for a, b in zip(top_neighbors, top_neighbor_sims)]
        ),
    })

hub_df = pd.DataFrame(hub_rows)
hub_df["similarity_hub_score"] = percentile_score(
    hub_df["mean_similarity_to_other_channels"],
    higher_is_more=True,
)
hub_df["topological_outlier_score"] = percentile_score(
    hub_df["mean_distance_to_other_channels"],
    higher_is_more=True,
)

hub_df = hub_df.sort_values("similarity_hub_score", ascending=False).reset_index(drop=True)

unique_regions = []
for r in regions:
    if r not in unique_regions:
        unique_regions.append(r)

region_similarity_mat = np.zeros((len(unique_regions), len(unique_regions)), dtype=float)
region_pair_count_mat = np.zeros((len(unique_regions), len(unique_regions)), dtype=int)

for i, r1 in enumerate(unique_regions):
    idx1 = np.where(np.asarray(regions) == r1)[0]

    for j, r2 in enumerate(unique_regions):
        idx2 = np.where(np.asarray(regions) == r2)[0]

        vals = []

        for a in idx1:
            for b in idx2:
                if a == b:
                    continue

                vals.append(consensus_similarity_mat[a, b])

        if len(vals) == 0:
            region_similarity_mat[i, j] = np.nan
            region_pair_count_mat[i, j] = 0
        else:
            region_similarity_mat[i, j] = float(np.nanmean(vals))
            region_pair_count_mat[i, j] = len(vals)

region_summary_rows = []

for region_name in unique_regions:
    region_channels = [ch for ch, reg in zip(channels, regions) if reg == region_name]
    idx = [channels.index(ch) for ch in region_channels]

    within_vals = []
    between_vals = []

    for a in idx:
        for b in range(n_channels):
            if a == b:
                continue

            if regions[b] == region_name:
                within_vals.append(consensus_similarity_mat[a, b])
            else:
                between_vals.append(consensus_similarity_mat[a, b])

    region_summary_rows.append({
        "region": region_name,
        "n_channels": len(region_channels),
        "channels": ", ".join(region_channels),
        "mean_within_region_similarity": float(np.nanmean(within_vals)) if len(within_vals) else np.nan,
        "mean_between_region_similarity": float(np.nanmean(between_vals)) if len(between_vals) else np.nan,
        "within_minus_between_similarity": (
            float(np.nanmean(within_vals) - np.nanmean(between_vals))
            if len(within_vals) and len(between_vals) else np.nan
        ),
    })

region_summary_df = pd.DataFrame(region_summary_rows)

method_agreement_rows = []

method_mats = {
    "Gaussian/RBF": gaussian_similarity_mat,
    "Cosine": cosine_similarity_01_mat,
    "Pearson": pearson_similarity_01_mat,
    "Inverse distance": inverse_similarity_mat,
}

consensus_offdiag = offdiag_values(consensus_similarity_mat)

for method_name, M in method_mats.items():
    method_agreement_rows.append({
        "similarity_method": method_name,
        "correlation_with_consensus_similarity": safe_corr(offdiag_values(M), consensus_offdiag),
        "mean_offdiag_similarity": float(np.nanmean(offdiag_values(M))),
        "std_offdiag_similarity": float(np.nanstd(offdiag_values(M))),
    })

method_agreement_df = pd.DataFrame(method_agreement_rows)

consensus_similarity_csv_path = os.path.join(similarity_out_dir, "topology_consensus_similarity_matrix.csv")
consensus_distance_csv_path = os.path.join(similarity_out_dir, "topology_consensus_distance_matrix.csv")
gaussian_similarity_csv_path = os.path.join(similarity_out_dir, "topology_gaussian_rbf_similarity_matrix.csv")
cosine_similarity_csv_path = os.path.join(similarity_out_dir, "topology_cosine_similarity_01_matrix.csv")
pearson_similarity_csv_path = os.path.join(similarity_out_dir, "topology_pearson_similarity_01_matrix.csv")
inverse_similarity_csv_path = os.path.join(similarity_out_dir, "topology_inverse_distance_similarity_matrix.csv")
pairwise_similarity_csv_path = os.path.join(similarity_out_dir, "topology_pairwise_channel_similarities.csv")
hub_csv_path = os.path.join(similarity_out_dir, "topology_similarity_hub_scores.csv")
region_similarity_csv_path = os.path.join(similarity_out_dir, "topology_region_similarity_matrix.csv")
region_summary_csv_path = os.path.join(similarity_out_dir, "topology_region_similarity_summary.csv")
method_agreement_csv_path = os.path.join(similarity_out_dir, "topology_similarity_method_agreement.csv")
npz_path = os.path.join(similarity_out_dir, "topology_derived_channel_similarity_outputs.npz")

pd.DataFrame(consensus_similarity_mat, index=channels, columns=channels).to_csv(consensus_similarity_csv_path)
pd.DataFrame(consensus_distance_mat, index=channels, columns=channels).to_csv(consensus_distance_csv_path)
pd.DataFrame(gaussian_similarity_mat, index=channels, columns=channels).to_csv(gaussian_similarity_csv_path)
pd.DataFrame(cosine_similarity_01_mat, index=channels, columns=channels).to_csv(cosine_similarity_csv_path)
pd.DataFrame(pearson_similarity_01_mat, index=channels, columns=channels).to_csv(pearson_similarity_csv_path)
pd.DataFrame(inverse_similarity_mat, index=channels, columns=channels).to_csv(inverse_similarity_csv_path)
pairwise_similarity_df.to_csv(pairwise_similarity_csv_path, index=False)
hub_df.to_csv(hub_csv_path, index=False)
pd.DataFrame(region_similarity_mat, index=unique_regions, columns=unique_regions).to_csv(region_similarity_csv_path)
region_summary_df.to_csv(region_summary_csv_path, index=False)
method_agreement_df.to_csv(method_agreement_csv_path, index=False)

np.savez(
    npz_path,
    channels=np.asarray(channels),
    regions=np.asarray(regions),
    cluster_labels=cluster_labels,
    feature_names=np.asarray(feature_z_df.columns),
    feature_matrix_standardized=X,
    euclidean_distance_matrix=euclidean_distance_mat,
    consensus_similarity_matrix=consensus_similarity_mat,
    consensus_distance_matrix=consensus_distance_mat,
    gaussian_similarity_matrix=gaussian_similarity_mat,
    cosine_similarity_01_matrix=cosine_similarity_01_mat,
    pearson_similarity_01_matrix=pearson_similarity_01_mat,
    inverse_distance_similarity_matrix=inverse_similarity_mat,
    gaussian_sigma=np.asarray([gaussian_sigma]),
    inverse_distance_scale=np.asarray([inverse_scale]),
    ordered_indices=ordered_idx,
    ordered_channels=np.asarray(ordered_channels),
    region_names=np.asarray(unique_regions),
    region_similarity_matrix=region_similarity_mat,
)

print("\nTopology-Derived Pairwise Channel Similarities:")
if HAVE_IPYTHON:
    display(pairwise_similarity_df)
else:
    print(pairwise_similarity_df)

print("\nTopology Similarity Hub Scores:")
if HAVE_IPYTHON:
    display(hub_df)
else:
    print(hub_df)

print("\nTopology Similarity Method Agreement:")
if HAVE_IPYTHON:
    display(method_agreement_df)
else:
    print(method_agreement_df)

print("\nTopology Region Similarity Summary:")
if HAVE_IPYTHON:
    display(region_summary_df)
else:
    print(region_summary_df)

sim_vmin, sim_vmax = dynamic_limits(
    offdiag_values(ordered_consensus_similarity),
    lower_pct=2,
    upper_pct=98,
    pad_frac=0.10,
    lower_bound=0.0,
    upper_bound=1.0,
)

plot_matrix(
    ordered_consensus_similarity,
    ordered_channels,
    "Topology-Derived Consensus Channel Similarity Matrix",
    "Consensus similarity",
    os.path.join(similarity_out_dir, "01_topology_consensus_similarity_heatmap.png"),
    vmin=sim_vmin,
    vmax=sim_vmax,
    width=950,
)

plot_matrix(
    ordered_gaussian_similarity,
    ordered_channels,
    f"Topology Gaussian/RBF Similarity Matrix | sigma={gaussian_sigma:.3f}",
    "Gaussian similarity",
    os.path.join(similarity_out_dir, "02_topology_gaussian_rbf_similarity_heatmap.png"),
    vmin=0.0,
    vmax=1.0,
    width=950,
)

plot_matrix(
    ordered_cosine_similarity,
    ordered_channels,
    "Topology Cosine Similarity Matrix",
    "Cosine similarity, rescaled",
    os.path.join(similarity_out_dir, "03_topology_cosine_similarity_heatmap.png"),
    vmin=0.0,
    vmax=1.0,
    width=950,
)

plot_matrix(
    ordered_pearson_similarity,
    ordered_channels,
    "Topology Pearson Similarity Matrix",
    "Pearson similarity, rescaled",
    os.path.join(similarity_out_dir, "04_topology_pearson_similarity_heatmap.png"),
    vmin=0.0,
    vmax=1.0,
    width=950,
)

dist_vmin, dist_vmax = dynamic_limits(
    offdiag_values(ordered_consensus_distance),
    lower_pct=2,
    upper_pct=98,
    pad_frac=0.10,
    lower_bound=0.0,
)

plot_matrix(
    ordered_consensus_distance,
    ordered_channels,
    "Topology-Derived Consensus Channel Distance Matrix",
    "Consensus distance",
    os.path.join(similarity_out_dir, "05_topology_consensus_distance_heatmap.png"),
    vmin=dist_vmin,
    vmax=dist_vmax,
    width=950,
)

top_similar_pairs = pairwise_similarity_df.head(min(TOP_PAIRS_TO_PLOT, len(pairwise_similarity_df))).copy()
top_similar_pairs["pair"] = top_similar_pairs["channel_a"] + "-" + top_similar_pairs["channel_b"]

fig, ax = plt.subplots(figsize=(13, 6.2), facecolor=BLACK)
style_ax(
    ax,
    title="Most Topologically Similar Channel Pairs",
    xlabel="Channel pair",
    ylabel="Consensus similarity",
)

x = np.arange(len(top_similar_pairs))

ax.bar(
    x,
    top_similar_pairs["topology_consensus_similarity"],
    color=CYAN_FILL,
    edgecolor=ACCENT,
    linewidth=0.9,
    alpha=0.82,
)

ax.set_xticks(x)
ax.set_xticklabels(top_similar_pairs["pair"], rotation=60, ha="right")
ax.set_ylim(
    max(0.0, float(top_similar_pairs["topology_consensus_similarity"].min()) - 0.04),
    min(1.02, float(top_similar_pairs["topology_consensus_similarity"].max()) + 0.04),
)

for i, row in top_similar_pairs.iterrows():
    ax.text(
        i,
        row["topology_consensus_similarity"],
        f"{row['topology_consensus_similarity']:.3f}",
        ha="center",
        va="bottom",
        color=ACCENT,
        fontsize=8,
        rotation=90,
    )

force_cyan_text(ax)

top_pairs_path = os.path.join(similarity_out_dir, "06_most_topologically_similar_channel_pairs.png")
plt.tight_layout()
save_render_close(fig, top_pairs_path, width=1150)

least_similar_pairs = (
    pairwise_similarity_df
    .sort_values("topology_consensus_similarity", ascending=True)
    .head(min(TOP_PAIRS_TO_PLOT, len(pairwise_similarity_df)))
    .copy()
)
least_similar_pairs["pair"] = least_similar_pairs["channel_a"] + "-" + least_similar_pairs["channel_b"]

fig, ax = plt.subplots(figsize=(13, 6.2), facecolor=BLACK)
style_ax(
    ax,
    title="Least Topologically Similar Channel Pairs",
    xlabel="Channel pair",
    ylabel="Consensus similarity",
)

x = np.arange(len(least_similar_pairs))

ax.bar(
    x,
    least_similar_pairs["topology_consensus_similarity"],
    color=CYAN_FILL,
    edgecolor=ACCENT,
    linewidth=0.9,
    alpha=0.82,
)

ax.set_xticks(x)
ax.set_xticklabels(least_similar_pairs["pair"], rotation=60, ha="right")
ax.set_ylim(
    max(0.0, float(least_similar_pairs["topology_consensus_similarity"].min()) - 0.04),
    min(1.02, float(least_similar_pairs["topology_consensus_similarity"].max()) + 0.04),
)

for i, row in least_similar_pairs.iterrows():
    ax.text(
        i,
        row["topology_consensus_similarity"],
        f"{row['topology_consensus_similarity']:.3f}",
        ha="center",
        va="bottom",
        color=ACCENT,
        fontsize=8,
        rotation=90,
    )

force_cyan_text(ax)

least_pairs_path = os.path.join(similarity_out_dir, "07_least_topologically_similar_channel_pairs.png")
plt.tight_layout()
save_render_close(fig, least_pairs_path, width=1150)

hub_plot_df = hub_df.sort_values("similarity_hub_score", ascending=False).reset_index(drop=True)

fig, ax = plt.subplots(figsize=(13, 6.2), facecolor=BLACK)
style_ax(
    ax,
    title="Topology Similarity Hub Score by Channel",
    xlabel="Channel",
    ylabel="Similarity hub score",
)

x = np.arange(len(hub_plot_df))

ax.bar(
    x,
    hub_plot_df["similarity_hub_score"],
    color=CYAN_FILL,
    edgecolor=ACCENT,
    linewidth=0.9,
    alpha=0.82,
)

ax.set_xticks(x)
ax.set_xticklabels(hub_plot_df["channel"], rotation=90)
ax.set_ylim(0, 105)

for i, row in hub_plot_df.iterrows():
    if i < 8:
        ax.text(
            i,
            row["similarity_hub_score"],
            f"{row['similarity_hub_score']:.0f}",
            ha="center",
            va="bottom",
            color=ACCENT,
            fontsize=8,
        )

force_cyan_text(ax)

hub_score_path = os.path.join(similarity_out_dir, "08_topology_similarity_hub_score_ranking.png")
plt.tight_layout()
save_render_close(fig, hub_score_path, width=1150)

if os.path.exists(topology_embedding_csv_path):
    topo_embed_df = pd.read_csv(topology_embedding_csv_path)
    topo_embed_df["channel"] = topo_embed_df["channel"].astype(str)

    if {"consensus_topology_x", "consensus_topology_y"}.issubset(topo_embed_df.columns):
        coord_df = topo_embed_df.set_index("channel").loc[channels]
        network_coords = coord_df[["consensus_topology_x", "consensus_topology_y"]].to_numpy(dtype=float)
    else:
        network_coords = classical_mds_from_distances(consensus_distance_mat, n_components=2)
else:
    network_coords = classical_mds_from_distances(consensus_distance_mat, n_components=2)

edge_set = set()

for i in range(n_channels):
    sims = consensus_similarity_mat[i].copy()
    sims[i] = -np.inf
    nn_idx = np.argsort(sims)[::-1][:min(NETWORK_TOP_K_NEIGHBORS, n_channels - 1)]

    for j in nn_idx:
        a = min(i, j)
        b = max(i, j)
        edge_set.add((a, b))

edge_rows = []

for i, j in sorted(edge_set):
    edge_rows.append({
        "channel_a": channels[i],
        "channel_b": channels[j],
        "consensus_similarity": float(consensus_similarity_mat[i, j]),
    })

network_edge_df = pd.DataFrame(edge_rows).sort_values(
    "consensus_similarity",
    ascending=False,
).reset_index(drop=True)

network_edge_csv_path = os.path.join(similarity_out_dir, "topology_similarity_network_edges_top_neighbors.csv")
network_edge_df.to_csv(network_edge_csv_path, index=False)

fig, ax = plt.subplots(figsize=(8.5, 7.5), facecolor=BLACK)
style_ax(
    ax,
    title=f"Topology Similarity Network | top {NETWORK_TOP_K_NEIGHBORS} neighbors/channel",
    xlabel="Topology embedding axis 1",
    ylabel="Topology embedding axis 2",
)

edge_sims = network_edge_df["consensus_similarity"].to_numpy(dtype=float) if len(network_edge_df) else np.asarray([])

if len(edge_sims):
    edge_min = float(np.nanmin(edge_sims))
    edge_max = float(np.nanmax(edge_sims))
else:
    edge_min = 0.0
    edge_max = 1.0

for _, row in network_edge_df.iterrows():
    i = channels.index(row["channel_a"])
    j = channels.index(row["channel_b"])
    sim = float(row["consensus_similarity"])

    if edge_max > edge_min:
        edge_norm = (sim - edge_min) / (edge_max - edge_min + 1e-12)
    else:
        edge_norm = 0.5

    ax.plot(
        [network_coords[i, 0], network_coords[j, 0]],
        [network_coords[i, 1], network_coords[j, 1]],
        color=ACCENT,
        linewidth=0.7 + 2.2 * edge_norm,
        alpha=0.18 + 0.60 * edge_norm,
        zorder=1,
    )

hub_map = hub_df.set_index("channel")["similarity_hub_score"].to_dict()
hub_scores_ordered = np.asarray([hub_map.get(ch, 50.0) for ch in channels], dtype=float)

for cluster_idx, cluster_id in enumerate(cluster_ids):
    mask = cluster_labels == cluster_id
    marker = MARKERS[cluster_idx % len(MARKERS)]

    sc = ax.scatter(
        network_coords[mask, 0],
        network_coords[mask, 1],
        c=hub_scores_ordered[mask],
        s=165,
        cmap=CYAN_SEQUENTIAL_CMAP,
        vmin=0,
        vmax=100,
        edgecolor=ACCENT,
        marker=marker,
        linewidth=1.2,
        alpha=0.95,
        label=f"Cluster {cluster_id}",
        zorder=4,
    )

for i, ch in enumerate(channels):
    txt = ax.text(
        network_coords[i, 0],
        network_coords[i, 1],
        ch,
        fontsize=8,
        color=ACCENT,
        alpha=0.95,
        zorder=6,
    )
    txt.set_path_effects([pe.withStroke(linewidth=2.0, foreground=BLACK)])

legend = ax.legend(loc="best", framealpha=1.0, fontsize=8)
style_legend(legend)

cbar = plt.colorbar(sc, ax=ax, shrink=0.82)
style_colorbar(cbar, "Similarity hub score")
force_cyan_text(ax)

network_path = os.path.join(similarity_out_dir, "09_topology_similarity_network_top_neighbors.png")
plt.tight_layout()
save_render_close(fig, network_path, width=950)

scalp_xy = {
    "Fp1": (-0.45, 1.00), "Fpz": (0.00, 1.08), "Fp2": (0.45, 1.00),
    "F7": (-0.95, 0.60), "F3": (-0.45, 0.55), "Fz": (0.00, 0.60),
    "F4": (0.45, 0.55), "F8": (0.95, 0.60),
    "FC5": (-0.75, 0.25), "FC1": (-0.25, 0.25),
    "FC2": (0.25, 0.25), "FC6": (0.75, 0.25),
    "M1": (-1.15, 0.05), "T7": (-1.00, 0.00), "C3": (-0.50, 0.00),
    "Cz": (0.00, 0.00), "C4": (0.50, 0.00), "T8": (1.00, 0.00),
    "M2": (1.15, 0.05),
    "CP5": (-0.75, -0.30), "CP1": (-0.25, -0.30),
    "CP2": (0.25, -0.30), "CP6": (0.75, -0.30),
    "P7": (-0.90, -0.60), "P3": (-0.45, -0.60), "Pz": (0.00, -0.65),
    "P4": (0.45, -0.60), "P8": (0.90, -0.60),
    "POz": (0.00, -0.88),
    "O1": (-0.35, -1.00), "Oz": (0.00, -1.08), "O2": (0.35, -1.00),
}

scalp_df = hub_df[hub_df["channel"].isin(scalp_xy)].copy()
scalp_df["x"] = scalp_df["channel"].map(lambda ch: scalp_xy[ch][0])
scalp_df["y"] = scalp_df["channel"].map(lambda ch: scalp_xy[ch][1])

fig = plt.figure(figsize=(13.5, 8.0), facecolor=BLACK)
gs = GridSpec(
    1,
    2,
    width_ratios=[1.25, 0.85],
    wspace=0.25,
    figure=fig,
)

ax = fig.add_subplot(gs[0, 0])
rank_ax = fig.add_subplot(gs[0, 1])

style_ax(
    ax,
    title="Scalp Map of Topology Similarity Hub Score",
    grid=False,
)

head = plt.Circle((0, 0), 1.12, fill=False, color=ACCENT, linewidth=2.2)
ax.add_patch(head)
ax.plot([0.00, -0.10, 0.10, 0.00], [1.12, 1.28, 1.28, 1.12], color=ACCENT, linewidth=2.2)

top_hub_channels = scalp_df.sort_values("similarity_hub_score", ascending=False).head(8)["channel"].tolist()
hub_scores = scalp_df["similarity_hub_score"].to_numpy(dtype=float)

if np.nanmax(hub_scores) > np.nanmin(hub_scores):
    hub_norm = (hub_scores - np.nanmin(hub_scores)) / (np.nanmax(hub_scores) - np.nanmin(hub_scores) + 1e-12)
else:
    hub_norm = np.full(len(scalp_df), 0.5)

scalp_df["marker_size"] = 95.0 + 520.0 * hub_norm

scatter = ax.scatter(
    scalp_df["x"],
    scalp_df["y"],
    c=scalp_df["similarity_hub_score"],
    s=scalp_df["marker_size"],
    cmap=CYAN_SEQUENTIAL_CMAP,
    vmin=0,
    vmax=100,
    edgecolor=ACCENT,
    linewidth=1.25,
    alpha=0.92,
    zorder=4,
)

for _, row in scalp_df.iterrows():
    ch = row["channel"]
    x0 = float(row["x"])
    y0 = float(row["y"])
    score = float(row["similarity_hub_score"])

    txt = ax.text(
        x0,
        y0,
        ch,
        ha="center",
        va="center",
        fontsize=8,
        color=ACCENT,
        fontweight="bold",
        zorder=6,
    )
    txt.set_path_effects([pe.withStroke(linewidth=2.1, foreground=BLACK)])

    if ch in top_hub_channels:
        ax.scatter(
            [x0],
            [y0],
            s=float(row["marker_size"]) * 1.65,
            facecolors="none",
            edgecolors=ACCENT_SOFT,
            linewidth=1.8,
            alpha=0.95,
            zorder=5,
        )

        dx = 0.06 if x0 <= 0 else -0.06
        ha = "left" if x0 <= 0 else "right"

        ann = ax.text(
            x0 + dx,
            y0 + 0.055,
            f"{score:.0f}",
            ha=ha,
            va="center",
            fontsize=8,
            color=ACCENT_SOFT,
            fontweight="bold",
            zorder=7,
        )
        ann.set_path_effects([pe.withStroke(linewidth=2.0, foreground=BLACK)])

ax.text(
    0,
    -1.30,
    "Bigger/brighter = more similar to many other channels | numbers mark top hubs",
    ha="center",
    va="top",
    color=ACCENT,
    fontsize=9,
)

ax.set_xlim(-1.38, 1.38)
ax.set_ylim(-1.40, 1.36)
ax.set_aspect("equal")
ax.set_xticks([])
ax.set_yticks([])

cbar = plt.colorbar(scatter, ax=ax, shrink=0.78, pad=0.03)
style_colorbar(cbar, "Hub score")

rank_df = scalp_df.sort_values("similarity_hub_score", ascending=True).copy()

style_ax(
    rank_ax,
    title="Similarity Hub Ranking",
    xlabel="Hub score",
    ylabel="Channel",
)

y = np.arange(len(rank_df))

rank_ax.barh(
    y,
    rank_df["similarity_hub_score"],
    color=CYAN_FILL,
    edgecolor=ACCENT,
    linewidth=0.8,
    alpha=0.82,
)

rank_ax.set_yticks(y)
rank_ax.set_yticklabels(rank_df["channel"])
rank_ax.set_xlim(0, 105)

rank_ax.axvline(
    float(np.nanmean(rank_df["similarity_hub_score"])),
    color=ACCENT_SOFT,
    linestyle="--",
    linewidth=1.2,
    alpha=0.85,
    label=f"Mean={np.nanmean(rank_df['similarity_hub_score']):.1f}",
)

for yy, (_, row) in zip(y, rank_df.iterrows()):
    score = float(row["similarity_hub_score"])

    rank_ax.text(
        score + 1.2,
        yy,
        f"{score:.0f}",
        va="center",
        ha="left",
        color=ACCENT,
        fontsize=7,
        alpha=0.82,
    )

legend = rank_ax.legend(loc="lower right", framealpha=1.0, fontsize=8)
style_legend(legend)
force_cyan_text(rank_ax)

top_text = "Top hubs: " + ", ".join(
    [
        f"{row.channel}({row.similarity_hub_score:.0f})"
        for row in scalp_df.sort_values("similarity_hub_score", ascending=False).head(5).itertuples()
    ]
)

bottom_text = "Least hub-like: " + ", ".join(
    [
        f"{row.channel}({row.similarity_hub_score:.0f})"
        for row in scalp_df.sort_values("similarity_hub_score", ascending=False).tail(5).itertuples()
    ]
)

fig.text(
    0.5,
    0.018,
    top_text + "    |    " + bottom_text,
    ha="center",
    va="bottom",
    color=ACCENT,
    fontsize=9,
)

scalp_hub_path = os.path.join(similarity_out_dir, "10_scalp_map_topology_similarity_hub_score.png")
plt.tight_layout(rect=[0.0, 0.045, 1.0, 1.0])
save_render_close(fig, scalp_hub_path, width=1150)

if len(unique_regions) > 1:
    region_plot_mat = region_similarity_mat.copy()
    region_plot_mat = np.nan_to_num(region_plot_mat, nan=0.0)

    fig, ax = plt.subplots(figsize=(8.8, 7.4), facecolor=BLACK)
    style_ax(
        ax,
        title="Regional Topology-Derived Similarity Matrix",
        xlabel="Region",
        ylabel="Region",
    )

    region_vals = region_plot_mat[np.isfinite(region_plot_mat)]

    im = ax.imshow(
        region_plot_mat,
        aspect="equal",
        interpolation="nearest",
        cmap=CYAN_SEQUENTIAL_CMAP,
        vmin=max(0.0, float(np.nanmin(region_vals)) - 0.02),
        vmax=min(1.0, float(np.nanmax(region_vals)) + 0.02),
    )

    ax.set_xticks(np.arange(len(unique_regions)))
    ax.set_xticklabels(unique_regions, rotation=45, ha="right")
    ax.set_yticks(np.arange(len(unique_regions)))
    ax.set_yticklabels(unique_regions)

    for i in range(len(unique_regions)):
        for j in range(len(unique_regions)):
            val = region_plot_mat[i, j]

            ax.text(
                j,
                i,
                f"{val:.2f}",
                ha="center",
                va="center",
                color=BLACK if val > np.nanmean(region_plot_mat) else ACCENT,
                fontsize=9,
                fontweight="bold",
            )

    cbar = plt.colorbar(im, ax=ax, shrink=0.82)
    style_colorbar(cbar, "Mean similarity")
    force_cyan_text(ax)

    region_matrix_path = os.path.join(similarity_out_dir, "11_regional_topology_similarity_matrix.png")
    plt.tight_layout()
    save_render_close(fig, region_matrix_path, width=900)

    region_pair_rows = []

    for _, row in pairwise_similarity_df.iterrows():
        if row["same_region"]:
            pair_type = "Within same region"
        else:
            pair_type = "Between regions"

        region_pair_rows.append({
            "pair_type": pair_type,
            "similarity": float(row["topology_consensus_similarity"]),
        })

    region_pair_df = pd.DataFrame(region_pair_rows)

    group_order = ["Within same region", "Between regions"]
    distance_groups = [
        region_pair_df.loc[region_pair_df["pair_type"] == group, "similarity"].to_numpy(dtype=float)
        for group in group_order
    ]

    fig, ax = plt.subplots(figsize=(9.5, 6), facecolor=BLACK)
    style_ax(
        ax,
        title="Topology Similarity Distribution: Within Region vs Between Regions",
        xlabel="Pair type",
        ylabel="Consensus similarity",
    )

    box = ax.boxplot(
        distance_groups,
        patch_artist=True,
        labels=group_order,
        showfliers=False,
        widths=0.58,
    )

    for patch in box["boxes"]:
        patch.set_facecolor(CYAN_FILL)
        patch.set_edgecolor(ACCENT)
        patch.set_alpha(0.34)
        patch.set_linewidth(1.4)

    for key in ["whiskers", "caps", "medians"]:
        for item in box[key]:
            item.set_color(ACCENT)
            item.set_linewidth(1.4)

    rng = np.random.default_rng(7)

    for idx, vals in enumerate(distance_groups, start=1):
        if len(vals) == 0:
            continue

        jitter = rng.uniform(-0.12, 0.12, size=len(vals))

        ax.scatter(
            np.full(len(vals), idx) + jitter,
            vals,
            s=26,
            facecolors=BLACK,
            edgecolors=CYAN_FILL,
            linewidths=1.0,
            alpha=0.75,
            zorder=4,
        )

        ax.text(
            idx,
            np.nanmax(vals),
            f"mean={np.nanmean(vals):.3f}\nn={len(vals)}",
            ha="center",
            va="bottom",
            color=ACCENT,
            fontsize=9,
        )

    force_cyan_text(ax)

    region_distribution_path = os.path.join(similarity_out_dir, "12_within_vs_between_region_similarity_distribution.png")
    plt.tight_layout()
    save_render_close(fig, region_distribution_path, width=950)

fig, ax = plt.subplots(figsize=(9.2, 5.8), facecolor=BLACK)
style_ax(
    ax,
    title="Similarity Method Agreement With Consensus",
    xlabel="Similarity method",
    ylabel="Correlation with consensus similarity",
)

x = np.arange(len(method_agreement_df))

ax.bar(
    x,
    method_agreement_df["correlation_with_consensus_similarity"],
    color=CYAN_FILL,
    edgecolor=ACCENT,
    linewidth=0.9,
    alpha=0.82,
)

ax.set_xticks(x)
ax.set_xticklabels(method_agreement_df["similarity_method"], rotation=25, ha="right")
ax.set_ylim(0, 1.05)

for i, row in method_agreement_df.iterrows():
    val = row["correlation_with_consensus_similarity"]

    ax.text(
        i,
        val,
        f"{val:.3f}",
        ha="center",
        va="bottom",
        color=ACCENT,
        fontsize=9,
    )

force_cyan_text(ax)

method_agreement_path = os.path.join(similarity_out_dir, "13_similarity_method_agreement_with_consensus.png")
plt.tight_layout()
save_render_close(fig, method_agreement_path, width=950)

fig, ax = plt.subplots(figsize=(8.5, 6.5), facecolor=BLACK)
style_ax(
    ax,
    title="Gaussian Similarity vs Cosine Similarity",
    xlabel="Gaussian/RBF topology similarity",
    ylabel="Cosine topology similarity",
)

rbf_vals = offdiag_values(gaussian_similarity_mat)
cos_vals = offdiag_values(cosine_similarity_01_mat)

ax.scatter(
    rbf_vals,
    cos_vals,
    s=32,
    facecolors=BLACK,
    edgecolors=CYAN_FILL,
    linewidths=1.0,
    alpha=0.78,
)

corr_rbf_cos = safe_corr(rbf_vals, cos_vals)

if np.isfinite(corr_rbf_cos):
    m, b = np.polyfit(rbf_vals, cos_vals, deg=1)
    xs = np.linspace(np.nanmin(rbf_vals), np.nanmax(rbf_vals), 150)
    ys = m * xs + b

    ax.plot(
        xs,
        ys,
        color=ACCENT,
        linewidth=1.5,
        linestyle="--",
        label=f"corr={corr_rbf_cos:.3f}",
    )

legend = ax.legend(loc="best", framealpha=1.0)
style_legend(legend)
force_cyan_text(ax)

method_scatter_path = os.path.join(similarity_out_dir, "14_gaussian_vs_cosine_similarity_scatter.png")
plt.tight_layout()
save_render_close(fig, method_scatter_path, width=900)

if len(cluster_ids) > 1:
    cluster_pair_rows = []

    for _, row in pairwise_similarity_df.iterrows():
        if row["same_topological_cluster"]:
            pair_type = f"Within Cluster {int(row['cluster_a'])}"
        else:
            pair_type = "Between Clusters"

        cluster_pair_rows.append({
            "pair_type": pair_type,
            "similarity": float(row["topology_consensus_similarity"]),
        })

    cluster_pair_df = pd.DataFrame(cluster_pair_rows)

    cluster_pair_order = []

    for cluster_id in cluster_ids:
        lab = f"Within Cluster {cluster_id}"

        if lab in cluster_pair_df["pair_type"].unique():
            cluster_pair_order.append(lab)

    if "Between Clusters" in cluster_pair_df["pair_type"].unique():
        cluster_pair_order.append("Between Clusters")

    groups = [
        cluster_pair_df.loc[cluster_pair_df["pair_type"] == group, "similarity"].to_numpy(dtype=float)
        for group in cluster_pair_order
    ]

    fig, ax = plt.subplots(figsize=(11.5, 6.2), facecolor=BLACK)
    style_ax(
        ax,
        title="Topology Similarity Distribution: Within vs Between Clusters",
        xlabel="Pair type",
        ylabel="Consensus similarity",
    )

    box = ax.boxplot(
        groups,
        patch_artist=True,
        labels=cluster_pair_order,
        showfliers=False,
        widths=0.58,
    )

    for patch in box["boxes"]:
        patch.set_facecolor(CYAN_FILL)
        patch.set_edgecolor(ACCENT)
        patch.set_alpha(0.34)
        patch.set_linewidth(1.4)

    for key in ["whiskers", "caps", "medians"]:
        for item in box[key]:
            item.set_color(ACCENT)
            item.set_linewidth(1.4)

    rng = np.random.default_rng(8)

    for idx, vals in enumerate(groups, start=1):
        if len(vals) == 0:
            continue

        jitter = rng.uniform(-0.12, 0.12, size=len(vals))

        ax.scatter(
            np.full(len(vals), idx) + jitter,
            vals,
            s=26,
            facecolors=BLACK,
            edgecolors=CYAN_FILL,
            linewidths=1.0,
            alpha=0.75,
            zorder=4,
        )

        ax.text(
            idx,
            np.nanmax(vals),
            f"mean={np.nanmean(vals):.3f}\nn={len(vals)}",
            ha="center",
            va="bottom",
            color=ACCENT,
            fontsize=9,
        )

    ax.set_xticklabels(cluster_pair_order, rotation=20, ha="right")
    force_cyan_text(ax)

    cluster_distribution_path = os.path.join(similarity_out_dir, "15_within_vs_between_cluster_similarity_distribution.png")
    plt.tight_layout()
    save_render_close(fig, cluster_distribution_path, width=1050)

log("\n" + "=" * 72)
log("Finished Topology-Derived Channel Similarity Matrix")
log(f"Input directory: {input_dir}")
log(f"Output directory: {similarity_out_dir}")
log(f"Gaussian sigma: {gaussian_sigma:.6f}")
log(f"Inverse-distance scale: {inverse_scale:.6f}")
log(f"Consensus similarity CSV: {consensus_similarity_csv_path}")
log(f"Consensus distance CSV: {consensus_distance_csv_path}")
log(f"Gaussian similarity CSV: {gaussian_similarity_csv_path}")
log(f"Cosine similarity CSV: {cosine_similarity_csv_path}")
log(f"Pearson similarity CSV: {pearson_similarity_csv_path}")
log(f"Inverse-distance similarity CSV: {inverse_similarity_csv_path}")
log(f"Pairwise similarity CSV: {pairwise_similarity_csv_path}")
log(f"Hub scores CSV: {hub_csv_path}")
log(f"Region similarity CSV: {region_similarity_csv_path}")
log(f"Region summary CSV: {region_summary_csv_path}")
log(f"Method agreement CSV: {method_agreement_csv_path}")
log(f"Network edge CSV: {network_edge_csv_path}")
log(f"NPZ: {npz_path}")
log("=" * 72)
```

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patheffects as pe
from matplotlib.colors import LinearSegmentedColormap
from matplotlib.gridspec import GridSpec
from scipy.spatial.distance import pdist, squareform
from scipy.cluster.hierarchy import linkage, leaves_list

try:
    from scipy.linalg import eigh
    HAVE_SCIPY_EIGH = True
except Exception:
    HAVE_SCIPY_EIGH = False

try:
    from IPython.display import display, Image
    HAVE_IPYTHON = True
except Exception:
    HAVE_IPYTHON = False


base_dir = "/home/a/projects/Complete-Neural-Signal-Analysis"

FAST_OPTION = "ultra"
SHUFFLE_SURROGATE_TYPE = "axis_shuffle"

SAVE_PLOTS_TO_DISK = True
RENDER_ALL_IMAGES_INLINE = True
INLINE_IMAGE_WIDTH = 1050
PLOT_DPI = 150

BLACK = "#000000"
ACCENT = "#00FFFF"
ACCENT_SOFT = "#66FFFF"
ACCENT_DIM = "#008B8B"

CYAN_FILL = "#00D5FF"
CYAN_FILL_SOFT = "#38BDF8"
CYAN_FILL_DARK = "#005A63"

CYAN_SEQUENTIAL_CMAP = LinearSegmentedColormap.from_list(
    "cyan_sequential",
    [BLACK, CYAN_FILL_DARK, CYAN_FILL, ACCENT_SOFT],
    N=256,
)

CYAN_DIVERGING_CMAP = LinearSegmentedColormap.from_list(
    "cyan_diverging",
    [ACCENT_SOFT, BLACK, CYAN_FILL],
    N=256,
)

plt.rcParams.update({
    "figure.facecolor": BLACK,
    "axes.facecolor": BLACK,
    "savefig.facecolor": BLACK,
    "text.color": ACCENT,
    "axes.labelcolor": ACCENT,
    "axes.edgecolor": ACCENT,
    "axes.titlecolor": ACCENT,
    "xtick.color": ACCENT,
    "ytick.color": ACCENT,
    "grid.color": ACCENT_DIM,
    "grid.alpha": 0.25,
    "legend.facecolor": BLACK,
    "legend.edgecolor": ACCENT,
    "legend.labelcolor": ACCENT,
    "font.size": 11,
})


def log(msg):
    print(msg, flush=True)


def style_ax(ax, title=None, xlabel=None, ylabel=None, grid=True):
    ax.set_facecolor(BLACK)

    for spine in ax.spines.values():
        spine.set_color(ACCENT)
        spine.set_linewidth(1.1)

    ax.tick_params(axis="x", colors=ACCENT, labelcolor=ACCENT)
    ax.tick_params(axis="y", colors=ACCENT, labelcolor=ACCENT)

    if title is not None:
        ax.set_title(title, color=ACCENT, pad=12)

    if xlabel is not None:
        ax.set_xlabel(xlabel, color=ACCENT, labelpad=10)

    if ylabel is not None:
        ax.set_ylabel(ylabel, color=ACCENT, labelpad=10)

    if grid:
        ax.grid(True, color=ACCENT_DIM, alpha=0.25, linewidth=0.8)


def force_cyan_text(ax):
    ax.title.set_color(ACCENT)
    ax.xaxis.label.set_color(ACCENT)
    ax.yaxis.label.set_color(ACCENT)

    for tick_label in ax.get_xticklabels():
        tick_label.set_color(ACCENT)

    for tick_label in ax.get_yticklabels():
        tick_label.set_color(ACCENT)


def style_colorbar(cbar, label):
    cbar.set_label(label, color=ACCENT)
    cbar.ax.yaxis.set_tick_params(color=ACCENT)
    plt.setp(cbar.ax.get_yticklabels(), color=ACCENT)
    cbar.outline.set_edgecolor(ACCENT)


def style_legend(legend):
    if legend is not None:
        legend.get_frame().set_facecolor(BLACK)
        legend.get_frame().set_edgecolor(ACCENT)

        for text in legend.get_texts():
            text.set_color(ACCENT)


def save_render_close(fig, path, width=INLINE_IMAGE_WIDTH):
    if SAVE_PLOTS_TO_DISK:
        fig.savefig(path, dpi=PLOT_DPI, bbox_inches="tight", facecolor=BLACK)

    plt.close(fig)

    if RENDER_ALL_IMAGES_INLINE and HAVE_IPYTHON and os.path.exists(path):
        display(Image(filename=path, width=width))


def offdiag_values(M):
    M = np.asarray(M, dtype=float)

    if M.shape[0] <= 1:
        return np.asarray([], dtype=float)

    mask = ~np.eye(M.shape[0], dtype=bool)
    vals = M[mask]
    vals = vals[np.isfinite(vals)]

    return vals


def dynamic_limits(values, lower_pct=2, upper_pct=98, pad_frac=0.08, lower_bound=None, upper_bound=None):
    values = np.asarray(values, dtype=float)
    values = values[np.isfinite(values)]

    if len(values) == 0:
        vmin, vmax = 0.0, 1.0
    else:
        vmin = float(np.percentile(values, lower_pct))
        vmax = float(np.percentile(values, upper_pct))

        if not np.isfinite(vmin):
            vmin = float(np.nanmin(values))

        if not np.isfinite(vmax):
            vmax = float(np.nanmax(values))

        if vmax <= vmin:
            center = float(np.nanmean(values))
            eps = max(1e-6, abs(center) * 0.01, 1e-5)
            vmin = center - eps
            vmax = center + eps

    pad = (vmax - vmin) * pad_frac
    vmin -= pad
    vmax += pad

    if lower_bound is not None:
        vmin = max(lower_bound, vmin)

    if upper_bound is not None:
        vmax = min(upper_bound, vmax)

    if vmax <= vmin:
        vmax = vmin + 1e-6

    return vmin, vmax


def classical_mds_from_distances(D, n_components=2):
    D = np.asarray(D, dtype=float)
    n = D.shape[0]

    if n < 2:
        return np.zeros((n, n_components), dtype=float)

    D2 = D ** 2
    J = np.eye(n) - np.ones((n, n)) / n
    B = -0.5 * J @ D2 @ J

    if HAVE_SCIPY_EIGH:
        vals, vecs = eigh(B)
    else:
        vals, vecs = np.linalg.eigh(B)

    idx = np.argsort(vals)[::-1]
    vals = vals[idx]
    vecs = vecs[:, idx]

    coords = np.zeros((n, n_components), dtype=float)

    for k in range(min(n_components, n)):
        if vals[k] > 0:
            coords[:, k] = vecs[:, k] * np.sqrt(vals[k])

    return coords


def gaussian_similarity_from_distance(D, sigma=None):
    D = np.asarray(D, dtype=float)
    off = offdiag_values(D)

    if sigma is None:
        sigma = float(np.nanmedian(off)) if len(off) else 1.0

    sigma = max(float(sigma), 1e-8)

    S = np.exp(-(D ** 2) / (2.0 * sigma ** 2))
    np.fill_diagonal(S, 1.0)

    return S, sigma


mode_out_dir = os.path.join(
    base_dir,
    "topology_shuffle_surrogate_persistent_homology",
    f"h1_shuffle_surrogate_ph_{FAST_OPTION}_fps_{SHUFFLE_SURROGATE_TYPE}",
)

summary_csv_path = os.path.join(mode_out_dir, "h1_shuffle_surrogate_ph_summary.csv")
shuffle_long_csv_path = os.path.join(mode_out_dir, "h1_shuffle_surrogate_ph_shuffle_long_table.csv")
diagnostic_csv_path = os.path.join(mode_out_dir, "h1_shuffle_surrogate_ph_runtime_diagnostics.csv")
signature_csv_path = os.path.join(mode_out_dir, "h1_shuffle_surrogate_ph_z_signature_matrix.csv")
signature_distance_csv_path = os.path.join(mode_out_dir, "h1_shuffle_surrogate_ph_signature_distance_matrix.csv")
signature_similarity_csv_path = os.path.join(mode_out_dir, "h1_shuffle_surrogate_ph_signature_similarity_matrix.csv")
mds_csv_path = os.path.join(mode_out_dir, "h1_shuffle_surrogate_ph_mds_coordinates.csv")
region_summary_csv_path = os.path.join(mode_out_dir, "h1_shuffle_surrogate_ph_region_summary.csv")
npz_path = os.path.join(mode_out_dir, "h1_shuffle_surrogate_ph_outputs.npz")

if not os.path.exists(summary_csv_path):
    raise FileNotFoundError(
        "Could not find shuffle-surrogate PH summary CSV:\n"
        f"{summary_csv_path}\n\n"
        "Run the full Shuffle-Surrogate Persistent Homology cell first."
    )

summary_df = pd.read_csv(summary_csv_path)

if os.path.exists(shuffle_long_csv_path):
    shuffle_long_df = pd.read_csv(shuffle_long_csv_path)
else:
    shuffle_long_df = pd.DataFrame()

if os.path.exists(diagnostic_csv_path):
    diagnostic_df = pd.read_csv(diagnostic_csv_path)
else:
    diagnostic_df = summary_df.copy()

primary_metric_names = [
    "total_lifetime",
    "max_lifetime",
    "q90_lifetime",
    "betti_auc",
    "betti_l2",
    "landscape_l2",
    "landscape_max",
]

metric_labels = [
    "Total lifetime",
    "Max lifetime",
    "Q90 lifetime",
    "Betti AUC",
    "Betti L2",
    "Landscape L2",
    "Landscape max",
]

primary_z_cols = [f"shuffle_z_{metric}" for metric in primary_metric_names]
primary_q_cols = [f"shuffle_q_greater_{metric}" for metric in primary_metric_names]
primary_p_cols = [f"shuffle_p_greater_{metric}" for metric in primary_metric_names]
primary_neglogq_cols = [f"shuffle_neglog10_q_{metric}" for metric in primary_metric_names]

for col in primary_z_cols:
    if col not in summary_df.columns:
        raise ValueError(f"Missing required column in summary CSV: {col}")

for metric in primary_metric_names:
    q_col = f"shuffle_q_greater_{metric}"
    p_col = f"shuffle_p_greater_{metric}"

    if q_col not in summary_df.columns and p_col not in summary_df.columns:
        raise ValueError(f"Missing both q and p columns for metric: {metric}")

    if q_col not in summary_df.columns and p_col in summary_df.columns:
        summary_df[q_col] = summary_df[p_col]

    neg_col = f"shuffle_neglog10_q_{metric}"

    if neg_col not in summary_df.columns:
        summary_df[neg_col] = -np.log10(summary_df[q_col].to_numpy(dtype=float) + 1e-12)

score_col = "shuffle_corrected_ph_score"

if score_col not in summary_df.columns:
    summary_df[score_col] = (
        pd.Series(summary_df["shuffle_corrected_mean_primary_z"].to_numpy(dtype=float))
        .rank(method="average", pct=True)
        .to_numpy(dtype=float)
        * 100.0
    )

if "shuffle_rejection_score" not in summary_df.columns:
    summary_df["shuffle_rejection_score"] = (
        pd.Series(summary_df["shuffle_corrected_mean_primary_neglog10_q"].to_numpy(dtype=float))
        .rank(method="average", pct=True)
        .to_numpy(dtype=float)
        * 100.0
    )

if "region" not in summary_df.columns:
    channel_regions = {
        "Fp1": "Frontal", "Fpz": "Frontal", "Fp2": "Frontal",
        "F7": "Frontal", "F3": "Frontal", "Fz": "Frontal", "F4": "Frontal", "F8": "Frontal",
        "FC5": "Frontocentral", "FC1": "Frontocentral", "FC2": "Frontocentral", "FC6": "Frontocentral",
        "M1": "Mastoid", "M2": "Mastoid",
        "T7": "Central", "C3": "Central", "Cz": "Central", "C4": "Central", "T8": "Central",
        "CP5": "Centroparietal", "CP1": "Centroparietal", "CP2": "Centroparietal", "CP6": "Centroparietal",
        "P7": "Parietal", "P3": "Parietal", "Pz": "Parietal", "P4": "Parietal", "P8": "Parietal",
        "POz": "Occipital", "O1": "Occipital", "Oz": "Occipital", "O2": "Occipital",
    }
    summary_df["region"] = summary_df["channel"].map(lambda ch: channel_regions.get(str(ch), "Unknown"))

if os.path.exists(signature_csv_path):
    corrected_signature_df = pd.read_csv(signature_csv_path, index_col=0)
else:
    corrected_signature_df = summary_df.set_index("channel")[primary_z_cols].copy()
    corrected_signature_df.to_csv(signature_csv_path)

corrected_signature_df = corrected_signature_df.replace([np.inf, -np.inf], np.nan).fillna(0.0)
signature_channels = list(corrected_signature_df.index.astype(str))
signature_X = corrected_signature_df.to_numpy(dtype=float)

if os.path.exists(signature_distance_csv_path):
    corrected_signature_distance = pd.read_csv(signature_distance_csv_path, index_col=0).loc[signature_channels, signature_channels].to_numpy(dtype=float)
else:
    corrected_signature_distance = squareform(pdist(signature_X, metric="euclidean")) if len(signature_channels) >= 2 else np.zeros((len(signature_channels), len(signature_channels)))
    pd.DataFrame(corrected_signature_distance, index=signature_channels, columns=signature_channels).to_csv(signature_distance_csv_path)

if os.path.exists(signature_similarity_csv_path):
    corrected_signature_similarity = pd.read_csv(signature_similarity_csv_path, index_col=0).loc[signature_channels, signature_channels].to_numpy(dtype=float)
else:
    corrected_signature_similarity, corrected_similarity_sigma = gaussian_similarity_from_distance(corrected_signature_distance)
    pd.DataFrame(corrected_signature_similarity, index=signature_channels, columns=signature_channels).to_csv(signature_similarity_csv_path)

if len(signature_channels) >= 3 and np.any(corrected_signature_distance > 0):
    Z_signature = linkage(squareform(corrected_signature_distance, checks=False), method="average", optimal_ordering=True)
    order = leaves_list(Z_signature)
else:
    order = np.arange(len(signature_channels))

ordered_channels = [signature_channels[i] for i in order]
ordered_signature_distance = corrected_signature_distance[np.ix_(order, order)]
ordered_signature_similarity = corrected_signature_similarity[np.ix_(order, order)]
ordered_signature_z = corrected_signature_df.loc[ordered_channels].to_numpy(dtype=float)

if os.path.exists(mds_csv_path):
    mds_df = pd.read_csv(mds_csv_path)
else:
    mds_coords = classical_mds_from_distances(corrected_signature_distance, n_components=2)
    mds_df = pd.DataFrame({
        "channel": signature_channels,
        "shuffle_corrected_mds_1": mds_coords[:, 0],
        "shuffle_corrected_mds_2": mds_coords[:, 1],
    })
    mds_df["region"] = [
        summary_df.set_index("channel").loc[ch, "region"] if ch in summary_df.set_index("channel").index else "Unknown"
        for ch in signature_channels
    ]
    score_map = summary_df.set_index("channel")[score_col].to_dict()
    reject_map = summary_df.set_index("channel")["shuffle_rejection_score"].to_dict()
    mds_df[score_col] = [score_map.get(ch, np.nan) for ch in signature_channels]
    mds_df["shuffle_rejection_score"] = [reject_map.get(ch, np.nan) for ch in signature_channels]
    mds_df.to_csv(mds_csv_path, index=False)

if os.path.exists(region_summary_csv_path):
    region_summary_df = pd.read_csv(region_summary_csv_path)
else:
    region_summary_df = (
        summary_df
        .groupby("region")
        .agg(
            n_channels=("channel", "count"),
            mean_shuffle_corrected_ph_score=(score_col, "mean"),
            sem_shuffle_corrected_ph_score=(score_col, lambda x: float(np.std(x, ddof=1) / np.sqrt(len(x))) if len(x) > 1 else 0.0),
            mean_primary_z=("shuffle_corrected_mean_primary_z", "mean"),
            mean_primary_neglog10_q=("shuffle_corrected_mean_primary_neglog10_q", "mean"),
            mean_n_primary_q_lt_0_05=("shuffle_corrected_n_primary_q_lt_0_05", "mean"),
        )
        .reset_index()
        .sort_values("mean_shuffle_corrected_ph_score", ascending=False)
    )
    region_summary_df.to_csv(region_summary_csv_path, index=False)

print("\nH1 Shuffle-Surrogate Persistent Homology Summary:")
if HAVE_IPYTHON:
    display(summary_df)
else:
    print(summary_df)

print("\nShuffle-Corrected H1 Z-Signature Matrix:")
if HAVE_IPYTHON:
    display(corrected_signature_df)
else:
    print(corrected_signature_df)

print("\nRegional Shuffle-Corrected PH Summary:")
if HAVE_IPYTHON:
    display(region_summary_df)
else:
    print(region_summary_df)

log("\n" + "=" * 72)
log("Regenerating ALL shuffle-surrogate PH plots with fixed plot 03")
log(f"Output directory: {mode_out_dir}")
log("=" * 72)

fig, ax = plt.subplots(figsize=(13, 6.2), facecolor=BLACK)
style_ax(
    ax,
    title=f"H1 Shuffle-Surrogate Persistent Homology Score | {FAST_OPTION.upper()}",
    xlabel="Channel",
    ylabel="Shuffle-corrected PH score",
)

rank_df = summary_df.sort_values(score_col, ascending=False).reset_index(drop=True)
x = np.arange(len(rank_df))

ax.bar(
    x,
    rank_df[score_col],
    color=CYAN_FILL,
    edgecolor=ACCENT,
    linewidth=0.9,
    alpha=0.82,
)

ax.axhline(
    float(np.nanmean(rank_df[score_col])),
    color=ACCENT_SOFT,
    linestyle="--",
    linewidth=1.3,
    alpha=0.85,
    label=f"Mean={np.nanmean(rank_df[score_col]):.1f}",
)

for i, row in rank_df.iterrows():
    if i < 8:
        ax.text(
            i,
            row[score_col],
            f"{row[score_col]:.0f}",
            ha="center",
            va="bottom",
            color=ACCENT,
            fontsize=8,
        )

ax.set_xticks(x)
ax.set_xticklabels(rank_df["channel"], rotation=90)
ax.set_ylim(0, 105)

legend = ax.legend(loc="best", framealpha=1.0)
style_legend(legend)
force_cyan_text(ax)

score_path = os.path.join(mode_out_dir, "01_shuffle_corrected_ph_score_ranking.png")
plt.tight_layout()
save_render_close(fig, score_path, width=1150)

fig, ax = plt.subplots(figsize=(12, 8.8), facecolor=BLACK)
style_ax(
    ax,
    title="Shuffle-Corrected H1 Primary Metric Z-Score Heatmap",
    xlabel="H1 metric",
    ylabel="Channel",
)

z_vlim = np.nanpercentile(np.abs(ordered_signature_z), 98) if np.any(np.isfinite(ordered_signature_z)) else 1.0
z_vlim = max(float(z_vlim), 1e-6)

im = ax.imshow(
    ordered_signature_z,
    aspect="auto",
    interpolation="nearest",
    cmap=CYAN_DIVERGING_CMAP,
    vmin=-z_vlim,
    vmax=z_vlim,
)

ax.set_yticks(np.arange(len(ordered_channels)))
ax.set_yticklabels(ordered_channels)
ax.set_xticks(np.arange(len(primary_z_cols)))
ax.set_xticklabels(metric_labels, rotation=45, ha="right")

cbar = plt.colorbar(im, ax=ax, shrink=0.85)
style_colorbar(cbar, "Observed vs shuffled z-score")
force_cyan_text(ax)

z_heatmap_path = os.path.join(mode_out_dir, "02_shuffle_corrected_primary_metric_z_heatmap.png")
plt.tight_layout()
save_render_close(fig, z_heatmap_path, width=1050)

order_df = summary_df.set_index("channel").loc[ordered_channels].reset_index()

z_matrix = order_df[primary_z_cols].to_numpy(dtype=float)
q_matrix = order_df[primary_q_cols].to_numpy(dtype=float)
obs_matrix = order_df[[f"observed_{metric}" for metric in primary_metric_names]].to_numpy(dtype=float)
null_matrix = order_df[[f"shuffle_mean_{metric}" for metric in primary_metric_names]].to_numpy(dtype=float)

sig_005 = q_matrix < 0.05
sig_010 = (q_matrix >= 0.05) & (q_matrix < 0.10)

fig = plt.figure(figsize=(15.5, 9.2), facecolor=BLACK)

ax = fig.add_axes([0.08, 0.18, 0.56, 0.72])
rank_ax = fig.add_axes([0.70, 0.18, 0.20, 0.72])
cbar_ax = fig.add_axes([0.655, 0.18, 0.018, 0.72])

for this_ax in [ax, rank_ax]:
    this_ax.set_facecolor(BLACK)

    for spine in this_ax.spines.values():
        spine.set_color(ACCENT)
        spine.set_linewidth(1.1)

    this_ax.tick_params(axis="x", colors=ACCENT, labelcolor=ACCENT)
    this_ax.tick_params(axis="y", colors=ACCENT, labelcolor=ACCENT)

z_abs = np.abs(z_matrix[np.isfinite(z_matrix)])

if len(z_abs) == 0:
    fixed_z_vlim = 1.0
else:
    fixed_z_vlim = max(float(np.nanpercentile(z_abs, 98)), 1.0)

im = ax.imshow(
    z_matrix,
    aspect="auto",
    interpolation="nearest",
    cmap=CYAN_DIVERGING_CMAP,
    vmin=-fixed_z_vlim,
    vmax=fixed_z_vlim,
)

ax.set_title(
    "Shuffle-Corrected H1 Significance + Effect Direction\n"
    "Color = observed-vs-shuffle z-score, markers = FDR significance",
    color=ACCENT,
    pad=14,
)

ax.set_xlabel("H1 metric", color=ACCENT, labelpad=12)
ax.set_ylabel("Channel", color=ACCENT, labelpad=12)

ax.set_xticks(np.arange(len(metric_labels)))
ax.set_xticklabels(metric_labels, rotation=45, ha="right", color=ACCENT)
ax.set_yticks(np.arange(len(ordered_channels)))
ax.set_yticklabels(ordered_channels, color=ACCENT)

ax.set_xticks(np.arange(-0.5, len(metric_labels), 1), minor=True)
ax.set_yticks(np.arange(-0.5, len(ordered_channels), 1), minor=True)
ax.grid(which="minor", color=ACCENT_DIM, linewidth=0.55, alpha=0.35)
ax.tick_params(which="minor", bottom=False, left=False)

for i in range(z_matrix.shape[0]):
    for j in range(z_matrix.shape[1]):
        z_val = z_matrix[i, j]

        if not np.isfinite(z_val):
            continue

        if sig_005[i, j]:
            marker = "●"
            marker_size = 11
            marker_color = ACCENT_SOFT
        elif sig_010[i, j]:
            marker = "○"
            marker_size = 10
            marker_color = ACCENT_SOFT
        else:
            marker = ""

        if marker:
            ax.text(
                j,
                i,
                marker,
                ha="center",
                va="center",
                color=marker_color,
                fontsize=marker_size,
                fontweight="bold",
            )

        if abs(z_val) >= 1.0:
            ax.text(
                j,
                i + 0.30,
                f"{z_val:+.1f}",
                ha="center",
                va="center",
                color=ACCENT,
                fontsize=6.5,
                alpha=0.85,
            )

cbar = plt.colorbar(im, cax=cbar_ax)
cbar.set_label("Observed vs shuffle z-score", color=ACCENT)
cbar.ax.yaxis.set_tick_params(color=ACCENT)
plt.setp(cbar.ax.get_yticklabels(), color=ACCENT)
cbar.outline.set_edgecolor(ACCENT)

score_values = order_df[score_col].to_numpy(dtype=float)
y = np.arange(len(ordered_channels))

rank_ax.barh(
    y,
    score_values,
    color=CYAN_FILL,
    edgecolor=ACCENT,
    linewidth=0.7,
    alpha=0.82,
)

rank_ax.set_title("Shuffle-corrected\nPH score", color=ACCENT, pad=14)
rank_ax.set_xlabel("Score", color=ACCENT, labelpad=10)
rank_ax.set_yticks(y)
rank_ax.set_yticklabels([])
rank_ax.invert_yaxis()
rank_ax.set_xlim(0, 105)

for yy, val in zip(y, score_values):
    if np.isfinite(val):
        rank_ax.text(
            val + 1.0,
            yy,
            f"{val:.1f}",
            va="center",
            ha="left",
            color=ACCENT,
            fontsize=7,
            alpha=0.9,
        )

rank_ax.grid(True, color=ACCENT_DIM, alpha=0.25, linewidth=0.8)

n_sig_005_by_metric = sig_005.sum(axis=0)

footer_lines = []
footer_lines.append("● = FDR q < 0.05    ○ = FDR q < 0.10")
footer_lines.append(
    "q<0.05 counts by metric: "
    + ", ".join([f"{label}: {count}" for label, count in zip(metric_labels, n_sig_005_by_metric)])
)
footer_lines.append(
    "Fixed view: q-only values saturated at the empirical shuffle limit, so this version shows z-score effect direction and q-value markers."
)

fig.text(
    0.08,
    0.055,
    "\n".join(footer_lines),
    ha="left",
    va="bottom",
    color=ACCENT,
    fontsize=9,
)

q_heatmap_path = os.path.join(mode_out_dir, "03_shuffle_corrected_primary_metric_q_heatmap.png")
q_heatmap_fixed_copy_path = os.path.join(mode_out_dir, "03_shuffle_corrected_primary_metric_significance_effect_heatmap_fixed.png")

if SAVE_PLOTS_TO_DISK:
    fig.savefig(q_heatmap_path, dpi=PLOT_DPI, bbox_inches="tight", facecolor=BLACK)
    fig.savefig(q_heatmap_fixed_copy_path, dpi=PLOT_DPI, bbox_inches="tight", facecolor=BLACK)

plt.close(fig)

if RENDER_ALL_IMAGES_INLINE and HAVE_IPYTHON and os.path.exists(q_heatmap_path):
    display(Image(filename=q_heatmap_path, width=1150))

sim_vmin, sim_vmax = dynamic_limits(
    offdiag_values(ordered_signature_similarity),
    lower_pct=2,
    upper_pct=98,
    pad_frac=0.10,
    lower_bound=0.0,
    upper_bound=1.0,
)

fig, ax = plt.subplots(figsize=(9.5, 8.5), facecolor=BLACK)
style_ax(
    ax,
    title="Shuffle-Corrected PH Signature Similarity Matrix",
    xlabel="Channel",
    ylabel="Channel",
)

im = ax.imshow(
    ordered_signature_similarity,
    aspect="equal",
    interpolation="nearest",
    cmap=CYAN_SEQUENTIAL_CMAP,
    vmin=sim_vmin,
    vmax=sim_vmax,
)

ax.set_xticks(np.arange(len(ordered_channels)))
ax.set_xticklabels(ordered_channels, rotation=90)
ax.set_yticks(np.arange(len(ordered_channels)))
ax.set_yticklabels(ordered_channels)

cbar = plt.colorbar(im, ax=ax, shrink=0.85)
style_colorbar(cbar, "Shuffle-corrected signature similarity")
force_cyan_text(ax)

signature_similarity_path = os.path.join(mode_out_dir, "04_shuffle_corrected_ph_signature_similarity_heatmap.png")
plt.tight_layout()
save_render_close(fig, signature_similarity_path, width=950)

dist_vmin, dist_vmax = dynamic_limits(
    offdiag_values(ordered_signature_distance),
    lower_pct=2,
    upper_pct=98,
    pad_frac=0.10,
    lower_bound=0.0,
)

fig, ax = plt.subplots(figsize=(9.5, 8.5), facecolor=BLACK)
style_ax(
    ax,
    title="Shuffle-Corrected PH Signature Distance Matrix",
    xlabel="Channel",
    ylabel="Channel",
)

im = ax.imshow(
    ordered_signature_distance,
    aspect="equal",
    interpolation="nearest",
    cmap=CYAN_SEQUENTIAL_CMAP,
    vmin=dist_vmin,
    vmax=dist_vmax,
)

ax.set_xticks(np.arange(len(ordered_channels)))
ax.set_xticklabels(ordered_channels, rotation=90)
ax.set_yticks(np.arange(len(ordered_channels)))
ax.set_yticklabels(ordered_channels)

cbar = plt.colorbar(im, ax=ax, shrink=0.85)
style_colorbar(cbar, "Shuffle-corrected signature distance")
force_cyan_text(ax)

signature_distance_path = os.path.join(mode_out_dir, "05_shuffle_corrected_ph_signature_distance_heatmap.png")
plt.tight_layout()
save_render_close(fig, signature_distance_path, width=950)

fig, ax = plt.subplots(figsize=(8.5, 7.2), facecolor=BLACK)
style_ax(
    ax,
    title="MDS of Shuffle-Corrected PH Signatures",
    xlabel="Shuffle-corrected PH MDS 1",
    ylabel="Shuffle-corrected PH MDS 2",
)

sc = ax.scatter(
    mds_df["shuffle_corrected_mds_1"],
    mds_df["shuffle_corrected_mds_2"],
    c=mds_df[score_col],
    s=165,
    cmap=CYAN_SEQUENTIAL_CMAP,
    vmin=0,
    vmax=100,
    edgecolor=ACCENT,
    linewidth=1.2,
    alpha=0.95,
)

for _, row in mds_df.iterrows():
    txt = ax.text(
        row["shuffle_corrected_mds_1"],
        row["shuffle_corrected_mds_2"],
        row["channel"],
        fontsize=8,
        color=ACCENT,
        alpha=0.95,
        zorder=5,
    )
    txt.set_path_effects([pe.withStroke(linewidth=2.0, foreground=BLACK)])

cbar = plt.colorbar(sc, ax=ax, shrink=0.82)
style_colorbar(cbar, "Shuffle-corrected PH score")
force_cyan_text(ax)

mds_path = os.path.join(mode_out_dir, "06_mds_shuffle_corrected_ph_signatures.png")
plt.tight_layout()
save_render_close(fig, mds_path, width=930)

plot_metric = "total_lifetime"

fig, ax = plt.subplots(figsize=(8.2, 7.0), facecolor=BLACK)
style_ax(
    ax,
    title="Observed vs Shuffle Mean H1 Total Lifetime",
    xlabel="Shuffle mean total lifetime",
    ylabel="Observed total lifetime",
)

obs_vals = summary_df[f"observed_{plot_metric}"].to_numpy(dtype=float)
null_vals = summary_df[f"shuffle_mean_{plot_metric}"].to_numpy(dtype=float)
score_vals = summary_df[score_col].to_numpy(dtype=float)

sc = ax.scatter(
    null_vals,
    obs_vals,
    c=score_vals,
    s=150,
    cmap=CYAN_SEQUENTIAL_CMAP,
    vmin=0,
    vmax=100,
    edgecolor=ACCENT,
    linewidth=1.1,
    alpha=0.95,
)

lim_max = max(float(np.nanmax(obs_vals)), float(np.nanmax(null_vals)), 1e-6)
ax.plot(
    [0, lim_max],
    [0, lim_max],
    color=ACCENT_SOFT,
    linestyle="--",
    linewidth=1.3,
    alpha=0.85,
    label="Observed = shuffle mean",
)

for _, row in summary_df.iterrows():
    txt = ax.text(
        row[f"shuffle_mean_{plot_metric}"],
        row[f"observed_{plot_metric}"],
        row["channel"],
        fontsize=8,
        color=ACCENT,
        alpha=0.9,
    )
    txt.set_path_effects([pe.withStroke(linewidth=2.0, foreground=BLACK)])

legend = ax.legend(loc="best", framealpha=1.0)
style_legend(legend)

cbar = plt.colorbar(sc, ax=ax, shrink=0.82)
style_colorbar(cbar, "Shuffle-corrected PH score")
force_cyan_text(ax)

obs_null_path = os.path.join(mode_out_dir, "07_observed_vs_shuffle_mean_total_lifetime.png")
plt.tight_layout()
save_render_close(fig, obs_null_path, width=930)

if len(shuffle_long_df) > 0 and "total_lifetime" in shuffle_long_df.columns:
    top_channels = summary_df.sort_values(score_col, ascending=False).head(min(8, len(summary_df)))["channel"].tolist()

    fig, ax = plt.subplots(figsize=(13, 6.4), facecolor=BLACK)
    style_ax(
        ax,
        title="Shuffle Null Distributions for Top Corrected Channels",
        xlabel="Channel",
        ylabel="H1 total lifetime",
    )

    box_data = []
    box_labels = []
    obs_for_labels = []

    for ch in top_channels:
        vals = shuffle_long_df.loc[
            shuffle_long_df["channel"] == ch,
            "total_lifetime",
        ].to_numpy(dtype=float)

        obs = float(summary_df.loc[summary_df["channel"] == ch, "observed_total_lifetime"].iloc[0])

        box_data.append(vals)
        box_labels.append(ch)
        obs_for_labels.append(obs)

    box = ax.boxplot(
        box_data,
        patch_artist=True,
        labels=box_labels,
        showfliers=False,
        widths=0.55,
    )

    for patch in box["boxes"]:
        patch.set_facecolor(CYAN_FILL)
        patch.set_edgecolor(ACCENT)
        patch.set_alpha(0.32)
        patch.set_linewidth(1.4)

    for key in ["whiskers", "caps", "medians"]:
        for item in box[key]:
            item.set_color(ACCENT)
            item.set_linewidth(1.4)

    for idx, obs in enumerate(obs_for_labels, start=1):
        ax.scatter(
            [idx],
            [obs],
            s=95,
            color=CYAN_FILL,
            edgecolor=ACCENT,
            linewidth=1.2,
            zorder=5,
            label="Observed" if idx == 1 else None,
        )

        ax.text(
            idx,
            obs,
            f"  obs={obs:.2f}",
            color=ACCENT,
            fontsize=8,
            va="center",
            ha="left",
        )

    legend = ax.legend(loc="best", framealpha=1.0)
    style_legend(legend)
    force_cyan_text(ax)

    null_box_path = os.path.join(mode_out_dir, "08_shuffle_null_distributions_top_channels.png")
    plt.tight_layout()
    save_render_close(fig, null_box_path, width=1150)

scalp_xy = {
    "Fp1": (-0.45, 1.00), "Fpz": (0.00, 1.08), "Fp2": (0.45, 1.00),
    "F7": (-0.95, 0.60), "F3": (-0.45, 0.55), "Fz": (0.00, 0.60),
    "F4": (0.45, 0.55), "F8": (0.95, 0.60),
    "FC5": (-0.75, 0.25), "FC1": (-0.25, 0.25),
    "FC2": (0.25, 0.25), "FC6": (0.75, 0.25),
    "M1": (-1.15, 0.05), "T7": (-1.00, 0.00), "C3": (-0.50, 0.00),
    "Cz": (0.00, 0.00), "C4": (0.50, 0.00), "T8": (1.00, 0.00),
    "M2": (1.15, 0.05),
    "CP5": (-0.75, -0.30), "CP1": (-0.25, -0.30),
    "CP2": (0.25, -0.30), "CP6": (0.75, -0.30),
    "P7": (-0.90, -0.60), "P3": (-0.45, -0.60), "Pz": (0.00, -0.65),
    "P4": (0.45, -0.60), "P8": (0.90, -0.60),
    "POz": (0.00, -0.88),
    "O1": (-0.35, -1.00), "Oz": (0.00, -1.08), "O2": (0.35, -1.00),
}

fig = plt.figure(figsize=(13.8, 8.2), facecolor=BLACK)
gs = GridSpec(
    1,
    2,
    width_ratios=[1.25, 0.85],
    wspace=0.25,
    figure=fig,
)

ax = fig.add_subplot(gs[0, 0])
rank_ax = fig.add_subplot(gs[0, 1])

style_ax(
    ax,
    title="Scalp Map of Shuffle-Corrected PH Score",
    grid=False,
)

head = plt.Circle((0, 0), 1.12, fill=False, color=ACCENT, linewidth=2.2)
ax.add_patch(head)
ax.plot([0.00, -0.10, 0.10, 0.00], [1.12, 1.28, 1.28, 1.12], color=ACCENT, linewidth=2.2)

scalp_df = summary_df[summary_df["channel"].isin(scalp_xy)].copy()
scalp_df["x"] = scalp_df["channel"].map(lambda ch: scalp_xy[ch][0])
scalp_df["y"] = scalp_df["channel"].map(lambda ch: scalp_xy[ch][1])

scores = scalp_df[score_col].to_numpy(dtype=float)

if np.nanmax(scores) > np.nanmin(scores):
    norm_scores = (scores - np.nanmin(scores)) / (np.nanmax(scores) - np.nanmin(scores) + 1e-12)
else:
    norm_scores = np.full(len(scores), 0.5)

scalp_df["marker_size"] = 90.0 + 520.0 * norm_scores

scatter = ax.scatter(
    scalp_df["x"],
    scalp_df["y"],
    c=scalp_df[score_col],
    s=scalp_df["marker_size"],
    cmap=CYAN_SEQUENTIAL_CMAP,
    vmin=0,
    vmax=100,
    edgecolor=ACCENT,
    linewidth=1.25,
    alpha=0.92,
    zorder=4,
)

top_score_channels = scalp_df.sort_values(score_col, ascending=False).head(8)["channel"].tolist()

for _, row in scalp_df.iterrows():
    ch = row["channel"]
    x0 = float(row["x"])
    y0 = float(row["y"])
    score = float(row[score_col])

    txt = ax.text(
        x0,
        y0,
        ch,
        ha="center",
        va="center",
        fontsize=8,
        color=ACCENT,
        fontweight="bold",
        zorder=6,
    )
    txt.set_path_effects([pe.withStroke(linewidth=2.1, foreground=BLACK)])

    if ch in top_score_channels:
        ax.scatter(
            [x0],
            [y0],
            s=float(row["marker_size"]) * 1.75,
            facecolors="none",
            edgecolors=ACCENT_SOFT,
            linewidth=1.8,
            alpha=0.95,
            zorder=5,
        )

        dx = 0.06 if x0 <= 0 else -0.06
        ha = "left" if x0 <= 0 else "right"

        ann = ax.text(
            x0 + dx,
            y0 + 0.055,
            f"{score:.0f}",
            ha=ha,
            va="center",
            fontsize=8,
            color=ACCENT_SOFT,
            fontweight="bold",
            zorder=7,
        )
        ann.set_path_effects([pe.withStroke(linewidth=2.0, foreground=BLACK)])

ax.text(
    0,
    -1.30,
    "Bigger/brighter = stronger H1 structure after axis-shuffle correction | numbers mark top corrected channels",
    ha="center",
    va="top",
    color=ACCENT,
    fontsize=9,
)

ax.set_xlim(-1.38, 1.38)
ax.set_ylim(-1.40, 1.36)
ax.set_aspect("equal")
ax.set_xticks([])
ax.set_yticks([])

cbar = plt.colorbar(scatter, ax=ax, shrink=0.78, pad=0.03)
style_colorbar(cbar, "Shuffle-corrected PH score")

rank_side_df = scalp_df.sort_values(score_col, ascending=True).copy()

style_ax(
    rank_ax,
    title="Shuffle-Corrected PH Ranking",
    xlabel="Corrected PH score",
    ylabel="Channel",
)

y = np.arange(len(rank_side_df))

rank_ax.barh(
    y,
    rank_side_df[score_col],
    color=CYAN_FILL,
    edgecolor=ACCENT,
    linewidth=0.8,
    alpha=0.82,
)

rank_ax.set_yticks(y)
rank_ax.set_yticklabels(rank_side_df["channel"])
rank_ax.set_xlim(0, 105)

rank_ax.axvline(
    float(np.nanmean(rank_side_df[score_col])),
    color=ACCENT_SOFT,
    linestyle="--",
    linewidth=1.2,
    alpha=0.85,
    label=f"Mean={np.nanmean(rank_side_df[score_col]):.1f}",
)

for yy, (_, row) in zip(y, rank_side_df.iterrows()):
    score = float(row[score_col])

    rank_ax.text(
        score + 1.2,
        yy,
        f"{score:.0f}",
        va="center",
        ha="left",
        color=ACCENT,
        fontsize=7,
        alpha=0.82,
    )

legend = rank_ax.legend(loc="lower right", framealpha=1.0, fontsize=8)
style_legend(legend)
force_cyan_text(rank_ax)

top_text = "Top shuffle-corrected: " + ", ".join(
    [
        f"{row.channel}({getattr(row, score_col):.0f})"
        for row in scalp_df.sort_values(score_col, ascending=False).head(5).itertuples()
    ]
)

bottom_text = "Lowest shuffle-corrected: " + ", ".join(
    [
        f"{row.channel}({getattr(row, score_col):.0f})"
        for row in scalp_df.sort_values(score_col, ascending=False).tail(5).itertuples()
    ]
)

fig.text(
    0.5,
    0.018,
    top_text + "    |    " + bottom_text,
    ha="center",
    va="bottom",
    color=ACCENT,
    fontsize=9,
)

scalp_path = os.path.join(mode_out_dir, "09_scalp_map_shuffle_corrected_ph_score.png")
plt.tight_layout(rect=[0.0, 0.045, 1.0, 1.0])
save_render_close(fig, scalp_path, width=1150)

fig, ax = plt.subplots(figsize=(12, 6.2), facecolor=BLACK)
style_ax(
    ax,
    title="Regional Shuffle-Corrected PH Score",
    xlabel="Region",
    ylabel="Mean shuffle-corrected PH score",
)

x = np.arange(len(region_summary_df))

ax.bar(
    x,
    region_summary_df["mean_shuffle_corrected_ph_score"],
    yerr=region_summary_df["sem_shuffle_corrected_ph_score"],
    color=CYAN_FILL,
    edgecolor=ACCENT,
    linewidth=0.9,
    alpha=0.82,
    capsize=7,
    error_kw={"elinewidth": 1.5, "capthick": 1.5, "ecolor": ACCENT},
)

for i, row in region_summary_df.iterrows():
    ax.text(
        i,
        row["mean_shuffle_corrected_ph_score"] + row["sem_shuffle_corrected_ph_score"],
        f"{row['mean_shuffle_corrected_ph_score']:.1f}\nn={int(row['n_channels'])}",
        ha="center",
        va="bottom",
        color=ACCENT,
        fontsize=9,
    )

ax.set_xticks(x)
ax.set_xticklabels(region_summary_df["region"], rotation=35, ha="right")
ax.set_ylim(0, 105)
force_cyan_text(ax)

region_path = os.path.join(mode_out_dir, "10_regional_shuffle_corrected_ph_score.png")
plt.tight_layout()
save_render_close(fig, region_path, width=1050)

fig, ax = plt.subplots(figsize=(13, 6.2), facecolor=BLACK)
style_ax(
    ax,
    title="Shuffle-Corrected Primary Metric Z-Profiles by Channel",
    xlabel="Channel",
    ylabel="Mean primary z-score",
)

profile_df = summary_df.sort_values("shuffle_corrected_mean_primary_z", ascending=False).reset_index(drop=True)
x = np.arange(len(profile_df))

ax.bar(
    x,
    profile_df["shuffle_corrected_mean_primary_z"],
    color=CYAN_FILL,
    edgecolor=ACCENT,
    linewidth=0.9,
    alpha=0.82,
)

ax.axhline(0, color=ACCENT_SOFT, linestyle="--", linewidth=1.2)

for i, row in profile_df.iterrows():
    if i < 8:
        ax.text(
            i,
            row["shuffle_corrected_mean_primary_z"],
            f"{row['shuffle_corrected_mean_primary_z']:.1f}",
            ha="center",
            va="bottom" if row["shuffle_corrected_mean_primary_z"] >= 0 else "top",
            color=ACCENT,
            fontsize=8,
        )

ax.set_xticks(x)
ax.set_xticklabels(profile_df["channel"], rotation=90)
force_cyan_text(ax)

mean_z_path = os.path.join(mode_out_dir, "11_mean_primary_shuffle_z_ranking.png")
plt.tight_layout()
save_render_close(fig, mean_z_path, width=1150)

qc_df = summary_df.sort_values("elapsed_sec", ascending=False).reset_index(drop=True)

fig, ax = plt.subplots(figsize=(13, 5.8), facecolor=BLACK)
style_ax(
    ax,
    title="Shuffle-Surrogate PH Runtime QC",
    xlabel="Channel",
    ylabel="Elapsed seconds",
)

x = np.arange(len(qc_df))

ax.bar(
    x,
    qc_df["elapsed_sec"],
    color=CYAN_FILL,
    edgecolor=ACCENT,
    linewidth=0.9,
    alpha=0.82,
)

ax.set_xticks(x)
ax.set_xticklabels(qc_df["channel"], rotation=90)

for i, row in qc_df.iterrows():
    if i < 6:
        ax.text(
            i,
            row["elapsed_sec"],
            f"{row['elapsed_sec']:.0f}s",
            ha="center",
            va="bottom",
            color=ACCENT,
            fontsize=8,
        )

force_cyan_text(ax)

runtime_path = os.path.join(mode_out_dir, "12_runtime_qc.png")
plt.tight_layout()
save_render_close(fig, runtime_path, width=1150)

log("\n" + "=" * 72)
log("Finished regenerating ALL shuffle-surrogate PH plots")
log("Saved / replaced:")
log("01_shuffle_corrected_ph_score_ranking.png")
log("02_shuffle_corrected_primary_metric_z_heatmap.png")
log("03_shuffle_corrected_primary_metric_q_heatmap.png  <-- fixed informative version")
log("03_shuffle_corrected_primary_metric_significance_effect_heatmap_fixed.png")
log("04_shuffle_corrected_ph_signature_similarity_heatmap.png")
log("05_shuffle_corrected_ph_signature_distance_heatmap.png")
log("06_mds_shuffle_corrected_ph_signatures.png")
log("07_observed_vs_shuffle_mean_total_lifetime.png")
if len(shuffle_long_df) > 0 and "total_lifetime" in shuffle_long_df.columns:
    log("08_shuffle_null_distributions_top_channels.png")
log("09_scalp_map_shuffle_corrected_ph_score.png")
log("10_regional_shuffle_corrected_ph_score.png")
log("11_mean_primary_shuffle_z_ranking.png")
log("12_runtime_qc.png")
log(f"Output directory: {mode_out_dir}")
log("=" * 72)
```
